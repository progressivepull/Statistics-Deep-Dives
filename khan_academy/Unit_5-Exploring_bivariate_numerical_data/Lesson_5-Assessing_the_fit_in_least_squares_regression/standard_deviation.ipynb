# Standard Deviation of Residuals (RMSE) — Explanation and Code Example

## Overview

A common way to measure how well a linear model fits data is to examine **how far the actual data points are from the model’s predictions**.  
This measure is known by several names:

- **Standard deviation of the residuals**
- **Root Mean Square Error (RMSE)**
- **Residual standard error**

All of these describe the *typical size of the residuals*—the differences between observed values and predicted values.

A **smaller value** indicates a **better model fit**.

---

## Residuals

For a linear model, the *i-th residual* is defined as:

$$
r_i = y_i - \hat{y}_i
$$

Where:
- $ y_i $ = actual observed value  
- $ \hat{y}_i $ = value predicted by the regression model  

- If the point is **above** the regression line → residual is **positive**
- If the point is **below** the regression line → residual is **negative**

---

## Example Data and Model

Assume the linear regression model:

$$
\hat{y} = 2.5x - 2
$$

| x | Actual y | Predicted y ($\hat{y}$) | Residual |
|---|---------|---------------------------|----------|
| 1 | 1 | 0.5 | $1 - 0.5 = 0.5$ |
| 2 | 3 | 3 | $3 - 3 = 0$ |
| 2 | 2 | 3 | $2 - 3 = -1$ |
| 3 | 6 | 5.5 | $6 - 5.5 = 0.5$ |

Residuals:
$$
[0.5,\; 0,\; -1,\; 0.5]
$$

---

## Calculating the Standard Deviation of Residuals (RMSE)

### Step 1: Square each residual

$$
0.5^2 = 0.25,\quad 0^2 = 0,\quad (-1)^2 = 1,\quad 0.5^2 = 0.25
$$

### Step 2: Add the squared residuals

$$
0.25 + 0 + 1 + 0.25 = 1.5
$$

### Step 3: Divide by $n - 1$

There are 4 residuals, so:

$$
\frac{1.5}{4 - 1} = \frac{1.5}{3} = 0.5
$$

### Step 4: Take the square root

$$
\sqrt{0.5} \approx 0.707
$$

---

## Final Result

$$
\boxed{\text{Standard deviation of residuals} \approx 0.707}
$$

This value represents the **typical distance** between the data points and the regression line.

---

## Interpretation

- This is **not** the simple average of residuals (those would cancel out).
- Squaring the residuals gives **larger errors more influence**, making RMSE sensitive to outliers.
- A **lower RMSE** means the model predictions are closer to the actual data.



```python
import numpy as np

# Actual x and y values
x = np.array([1, 2, 2, 3])
y_actual = np.array([1, 3, 2, 6])

# Linear model: y_hat = 2.5x - 2
y_predicted = 2.5 * x - 2

# Residuals
residuals = y_actual - y_predicted

# Standard deviation of residuals (sample RMSE)
rmse = np.sqrt(np.sum(residuals**2) / (len(residuals) - 1))

print("Residuals:", residuals.tolist()) 
print(f"RMSE ≈ {rmse:.3f}")
```
