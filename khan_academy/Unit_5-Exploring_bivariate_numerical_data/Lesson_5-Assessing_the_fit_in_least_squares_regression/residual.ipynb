# Residual Plots in Regression Analysis

## Context
We have a simple **least squares regression line** fitted to four data points.  
The goal is to understand **residuals** and how to visualize them.

---

## What is a Residual?
A **residual** is defined as:



$$
\text{Residual} = \text{Actual Value} - \text{Expected Value (from regression line)}
$$



### Examples
- At $x = 1$:  
  - Actual $y = 1$  
  - Expected $y = 2.5(1) - 2 = 0.5$  
  - Residual = $1 - 0.5 = +0.5$

- At $x = 2$:  
  - Actual $y = 2$  
  - Expected $y = 3$  
  - Residual = $2 - 3 = -1$

- At $x = 2$ (another point):  
  - Actual $y = 3$  
  - Expected $y = 3$  
  - Residual = $0$

- At $x = 3$:  
  - Actual $y = 6$  
  - Expected $y = 5.5$  
  - Residual = $6 - 5.5 = +0.5$

---

## Plotting Residuals
- **X-axis**: Input values (e.g., study time, salary, etc.)  
- **Y-axis**: Residuals (positive above, negative below)  
- Residuals are plotted at each $x$ to show how far actual points deviate from the regression line.

This visualization is called a **residual plot**.

---

## Why Use Residual Plots?
Residual plots help evaluate the **fit of a regression model**:
- If residuals are **randomly scattered** above and below the x-axis → linear model is appropriate.  
- If residuals show a **trend** (upward, downward, or curved) → linear model may not be suitable; consider a non-linear model.  
- Large residuals far from the x-axis → poor fit, low correlation (R value not close to 1).

---

## Examples
- **Good Fit**: Residuals evenly scattered, no discernible trend → linear regression works well.  
- **Poor Fit**: Residuals curve or trend upward/downward → suggests a non-linear relationship.  
- **Weak Fit**: Residuals far from the x-axis → regression line does not explain the data well.

---

## Key Takeaway
Residual plots provide a **diagnostic tool** for regression:
- They reveal whether a linear model is appropriate.  
- They highlight patterns that suggest the need for more complex models.  
- They measure how well the regression line explains the variability in the data.


```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data points
x = np.array([1, 2, 2, 3])
y = np.array([1, 3, 2, 6])

# Fit a regression line: y = slope * x + intercept
slope, intercept = np.polyfit(x, y, 1)
y_pred = slope * x + intercept

# Calculate residuals: actual - expected
residuals = y - y_pred

# --- Plot regression line with data points ---
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, y_pred, color='red', label=f'Regression Line: y = {slope:.2f}x + {intercept:.2f}')
plt.title('Regression Line Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

# --- Plot residuals ---
plt.subplot(1, 2, 2)
plt.scatter(x, residuals, color='purple')
plt.axhline(0, color='black', linestyle='--')
plt.title('Residual Plot')
plt.xlabel('x')
plt.ylabel('Residual (Actual - Expected)')
plt.grid(True)

plt.tight_layout()
plt.show()

# Print residuals for clarity
for xi, yi, ri in zip(x, y, residuals):
    print(f"x={xi}, actual y={yi}, residual={ri:.2f}")
```

# What this adds:

* R² value is computed and displayed in the regression plot title.
* R² close to 1 → strong fit; R² near 0 → weak fit.
* You still see the residual plot to visually check scatter patterns.


# What this does:

* Fits a regression line to the sample data.
* Computes residuals as **actual − expected**.
* Shows two plots:
    - Left: regression line with data points.
    - Right: residual plot (points above/below the zero line).
* Prints residual values for each data point.

This way you can visually check whether residuals are **randomly scattered**(good fit) or show a pattern (suggesting a non‑linear model).


# Understanding the R² Value in Regression

## Definition
The **R² value**, also called the *coefficient of determination*, measures how well a regression line fits the data.



$$
R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}}
$$



Where:
- $SS_{\text{res}}$ = Sum of squared residuals (errors)  
- $SS_{\text{tot}}$ = Total sum of squares (variation in the data)

---

## Interpretation
- **R² = 1.0** → Perfect fit. The regression line explains all variability in the data.  
- **R² = 0.0** → No fit. The regression line explains none of the variability.  
- **0 < R² < 1** → Partial fit. The closer to 1, the stronger the model explains the data.  

---

## Example
Suppose we fit a regression line to study time vs. test scores:
- Residuals are calculated as **actual − predicted**.  
- If the residuals are small and scattered randomly, R² will be **high**.  
- If residuals are large or show a pattern, R² will be **low**.

---

## Why It Matters
- R² helps evaluate the **quality of a regression model**.  
- It tells us the proportion of variance in the dependent variable explained by the independent variable(s).  
- A high R² does not guarantee the model is appropriate — residual plots should also be checked for trends.

---

## Key Takeaway
The **R² value** quantifies how well a regression line explains the variability in the data.  
It is a **diagnostic tool**: use it alongside residual plots to judge whether a linear model is suitable.
