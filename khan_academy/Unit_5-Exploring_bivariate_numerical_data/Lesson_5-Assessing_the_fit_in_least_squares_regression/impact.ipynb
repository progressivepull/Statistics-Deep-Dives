# Effects of Removing an Outlier from a Least-Squares Regression Line

## Example 1: Removing the Outlier (95, 1)

The scatterplot shows positively correlated bivariate data with an upward-sloping least-squares regression line. The point **(95, 1)** is a clear outlier with a large **negative residual**, meaning it lies far below the regression line.

### Key Observations
- The correlation coefficient **r** is already **greater than 0** and **less than 1**.
- The outlier pulls the regression line downward and weakens the strength of the linear relationship.
- Removing this outlier allows the regression line to better fit the remaining data.

### Effects of Removing the Outlier
- **Correlation coefficient (r) increases**
- **Slope of the regression line increases**
- **Coefficient of determination (r²) increases**

### Correct Conclusions
- ✅ The coefficient of determination **r² would increase**
- ✅ The slope of the least-squares regression line **would increase**
- ❌ The correlation coefficient **would not get closer to 0** (it gets closer to 1)

---

## Example 2: Removing the Outlier (10, −18)

In this scatterplot, the data show a **negative correlation**, and the regression line has a negative slope. The point **(10, −18)** is an outlier that pulls the left side of the regression line downward.

### Key Observations
- The correlation coefficient **r** is **less than 0** and **greater than −1**.
- The outlier weakens the negative linear relationship.
- Removing the outlier strengthens the negative correlation.

### Effects of Removing the Outlier
- **Correlation coefficient (r) becomes more negative (closer to −1)**
- **Slope of the regression line decreases (more negative)**
- **Coefficient of determination (r²) increases**
- **Y-intercept increases**

> Note: Even though **r decreases** (becomes more negative), **r² increases** because squaring a value with larger magnitude increases the result.

### Correct Conclusions
- ❌ The coefficient of determination **r² would not decrease**
- ❌ The slope of the regression line **would not increase**
- ✅ The **y-intercept would increase**

---

## Summary

Removing an outlier generally:
- Strengthens the linear relationship
- Increases **r²**
- Makes the slope steeper in the direction of the correlation
- Adjusts the y-intercept based on pivoting around the mean of the data

This demonstrates how influential outliers can significantly affect regression results.


```python
# Least-Squares Regression: Effect of Removing an Outlier
# Example inspired by the scatterplot with outlier (95, 1)

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

# -----------------------------
# Original dataset (includes outlier)
# -----------------------------
x = np.array([5, 15, 25, 35, 45, 55, 65, 75, 85, 95])
y = np.array([3, 4, 3, 5, 6, 4, 6, 5, 7, 1])  # (95, 1) is the outlier

# Regression with outlier
slope1, intercept1, r1, _, _ = linregress(x, y)
r2_1 = r1**2

# -----------------------------
# Dataset without outlier
# -----------------------------
x_no = x[:-1]
y_no = y[:-1]

# Regression without outlier
slope2, intercept2, r2, _, _ = linregress(x_no, y_no)
r2_2 = r2**2

# -----------------------------
# Print results
# -----------------------------
print("WITH outlier (95, 1)")
print(f"Slope: {slope1:.4f}")
print(f"Correlation r: {r1:.4f}")
print(f"Coefficient of determination r²: {r2_1:.4f}\n")

print("WITHOUT outlier (95, 1)")
print(f"Slope: {slope2:.4f}")
print(f"Correlation r: {r2:.4f}")
print(f"Coefficient of determination r²: {r2_2:.4f}")

# -----------------------------
# Visualization
# -----------------------------
plt.scatter(x, y)
plt.plot(x, slope1 * x + intercept1, label="With outlier")

plt.scatter(x_no, y_no)
plt.plot(x_no, slope2 * x_no + intercept2, linestyle="--", label="Without outlier")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Effect of Removing an Outlier on Least-Squares Regression")
plt.legend()
plt.show()
```

# ✅ What This Code Demonstrates
After running it, you will observe:

* **Slope increases** after removing the outlier
* **Correlation coefficient (r) increases** (moves closer to 1)
* **Coefficient of determination (r²)** increases
* The regression line fits the remaining data much better

These results confirm the correct answers:
* ✅ A. The coefficient of determination (r²) would increase
* ❌ B. The correlation coefficient (r) would not get closer to 0
* ✅ C. The slope of the regression line would increase
