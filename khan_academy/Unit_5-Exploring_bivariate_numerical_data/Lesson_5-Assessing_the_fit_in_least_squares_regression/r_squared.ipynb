# Understanding Regression, Squared Error, and R²

## Regression Line
Given **n points** with coordinates $(x_i, y_i)$, we can fit a line:



$$
y = mx + b
$$



This line minimizes the **squared distance** between the data points and the line.

---

## Squared Error
For each point $(x_i, y_i)$, the error is:



$$
\text{error}_i = y_i - (mx_i + b)
$$



The **total squared error** is:



$$
SS_{\text{res}} = \sum_{i=1}^n \left(y_i - (mx_i + b)\right)^2
$$



This measures how far the points are from the regression line.

---

## Total Variation in $y$
The total variation in $y$ is measured relative to the mean of $y$:



$$
SS_{\text{tot}} = \sum_{i=1}^n \left(y_i - \bar{y}\right)^2
$$



- $\bar{y}$ = mean of all $y$-values  
- This represents the overall spread of the data around its mean.

---

## Coefficient of Determination (R²)
We want to know: **What percentage of the variation in $y$ is explained by variation in $x$?**



$$
R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}}
$$



- $SS_{\text{res}}$: variation **not explained** by the regression line  
- $SS_{\text{tot}}$: total variation in $y$  

Thus, $R^2$ measures the proportion of variation in $y$ explained by the regression line.

---

## Interpretation
- **R² close to 1** → Line is a good fit; most variation in $y$ is explained by $x$.  
- **R² close to 0** → Line is a poor fit; very little of the variation in $y$ is explained by $x$.  

---

## Key Takeaway
- The regression line minimizes squared error.  
- Total variation in $y$ is compared against unexplained variation.  
- **R² quantifies how well the regression line explains the data.**


```python
# Linear Regression and R² from scratch
import numpy as np

# Example dataset: x values and y values
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

# Step 1: Compute slope (m) and intercept (b) using formulas
x_mean = np.mean(x)
y_mean = np.mean(y)

m = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean)**2)
b = y_mean - m * x_mean

print(f"Slope (m): {m:.3f}, Intercept (b): {b:.3f}")

# Step 2: Predicted y values from regression line
y_pred = m * x + b

# Step 3: Compute squared errors
SS_res = np.sum((y - y_pred)**2)       # Residual sum of squares
SS_tot = np.sum((y - y_mean)**2)       # Total sum of squares

# Step 4: Compute R²
R2 = 1 - (SS_res / SS_tot)

print(f"R² (coefficient of determination): {R2:.3f}")

# Optional: visualize the fit
import matplotlib.pyplot as plt

plt.scatter(x, y, color="blue", label="Data points")
plt.plot(x, y_pred, color="red", label="Regression line")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression Fit")
plt.legend()
plt.show()
```

# What this does

* **Compute slope and intercept** using the least squares formula.
* **Predict y values** from the regression line.
* **Calculate squared errors:** residual vs. total variation.
* **Derive R²** to measure how well the line explains the data.
* **Visualize the fit** with a scatter plot and regression line.
