# Summary: Interpreting Computer Output for Linear Regression

## Purpose

Although linear regression can be done by hand, in practice it is usually computed using a **computer or calculator**. The goal is to understand typical computer output, not be intimidated by it, and learn how to extract the **regression equation** and key information.

---

## Study Scenario

- **Research question:** Do students who consume more caffeine tend to study more?
- **Sample:** 20 randomly selected students
- **Variables:**
  - **X (Predictor):** Caffeine consumed (milligrams)
  - **Y (Response):** Hours spent studying
- A scatterplot indicates a **linear relationship**.
- A **least-squares regression analysis** is performed by a computer.

---

## Regression Equation Form

A linear regression model has the form:

$$
\hat{Y} = MX + B
$$

Where:
- $ \hat{Y} $ = predicted hours studying  
- $ X $ = caffeine intake (mg)  
- $ M $ = slope  
- $ B $ = y-intercept  

---

## Using the Computer Output

### Predictors and Coefficients

The regression output includes a table with:

- **Predictor: Constant**
  - Coefficient = **2.544**
  - This is the **y-intercept (B)**

- **Predictor: Caffeine**
  - Coefficient = **0.164**
  - This is the **slope (M)**

---

## Regression Line

From the coefficients:

$$
\boxed{\hat{Y} = 0.164X + 2.544}
$$

**Interpretation:**
- For each additional milligram of caffeine consumed, predicted study time increases by about **0.164 hours**.
- When caffeine intake is zero, predicted study time is about **2.544 hours**.

---

## Other Important Output Values

### R-squared ($R^2$)
- Measures how much of the variation in study hours is explained by caffeine intake.
- The correlation coefficient $R$ is the **square root of $R^2$**.
- Since the slope is positive, **R is positive**.

---

### Adjusted R-squared
- Mainly useful when **multiple predictors** are used.
- Not very important for **simple (one-variable) regression**.

---

### S (Standard Deviation of Residuals)
- Represents the **typical prediction error**.
- Measures how well the regression line fits the data.
- Smaller values indicate a **better fit**.

---

## Key Takeaways

- Computer output makes finding the regression equation **quick and accurate**.
- The most important values to identify are:
  - **Slope**
  - **Intercept**
  - **R²**
  - **Standard deviation of residuals**
- While regression output includes many statistics, only a few are needed to understand and use the model effectively.
- Knowing what to look for makes regression results **straightforward and useful**.


```python
"""
Linear Regression Example: Caffeine Intake vs. Study Hours

This program demonstrates how a computer produces regression output
similar to the table shown (coefficients, standard errors, t-values,
p-values, R-squared, adjusted R-squared, and residual standard deviation).
"""

import numpy as np
import pandas as pd
import statsmodels.api as sm

# ---------------------------------------------------
# Example data (replace with actual collected data)
# ---------------------------------------------------
# X = caffeine intake in mg
# Y = hours spent studying
caffeine_mg = np.array([
    50, 80, 100, 120, 150, 180, 200, 220, 250, 275,
    300, 320, 350, 380, 400, 420, 450, 480, 500, 520
])

hours_studying = np.array([
    3.1, 3.4, 3.7, 4.0, 4.3, 4.6, 4.9, 5.1, 5.5, 5.7,
    6.0, 6.2, 6.6, 6.9, 7.1, 7.3, 7.7, 8.0, 8.2, 8.4
])

# ---------------------------------------------------
# Build regression model
# ---------------------------------------------------
X = sm.add_constant(caffeine_mg)  # adds intercept (constant)
model = sm.OLS(hours_studying, X)
results = model.fit()

# ---------------------------------------------------
# Display full regression summary
# ---------------------------------------------------
print(results.summary())

# ---------------------------------------------------
# Extract key values (similar to the computer output)
# ---------------------------------------------------
coefficients = results.params
std_errors = results.bse
t_values = results.tvalues
p_values = results.pvalues

r_squared = results.rsquared
adj_r_squared = results.rsquared_adj

# Standard deviation of residuals (S)
residual_std = np.sqrt(results.mse_resid)

print("\nKey Regression Output")
print("---------------------")
print(f"Intercept (Constant): {coefficients[0]:.3f}")
print(f"Slope (Caffeine): {coefficients[1]:.3f}")
print(f"R-squared: {r_squared:.3%}")
print(f"Adjusted R-squared: {adj_r_squared:.3%}")
print(f"S (Std. Dev. of Residuals): {residual_std:.3f}")

# ---------------------------------------------------
# Regression equation
# ---------------------------------------------------
print("\nRegression Equation:")
print(f"Y_hat = {coefficients[1]:.3f} * X + {coefficients[0]:.3f}")
```
