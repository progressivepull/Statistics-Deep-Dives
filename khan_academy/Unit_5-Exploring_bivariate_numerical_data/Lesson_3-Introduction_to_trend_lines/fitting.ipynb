# Fitting Data to a Line (Linear Regression) — Conceptual Overview

This example introduces the idea of **fitting data to a line**, also known as **linear regression**, using a spreadsheet tool like **Excel**. The focus is on **conceptual understanding**, not on the underlying mathematics.

---

## The Problem

We are given data from the **U.S. Census Bureau** showing the **median California family income** from **1995 to 2002**.

We are asked to:

1. Draw a **scatter plot**
2. Find the **equation of the best-fit line**
3. Predict the **median income in 2010**
4. Interpret the **slope** and **y-intercept**

---

## Organizing the Data

To make the data usable in a spreadsheet, it is organized into two columns:

- **Years since 1995**
- **Median family income (USD)**

Using “years since 1995” instead of calendar years simplifies interpretation and calculation.

Example structure:

| Years Since 1995 | Median Income ($) |
|------------------|------------------|
| 0 | 53,807 |
| 1 | 55,217 |
| 2 | 55,209 |
| 3 | 55,415 |
| 4 | 63,100 |
| 5 | 63,206 |
| 6 | 63,761 |
| 7 | 65,766 |

---

## Creating the Scatter Plot

In Excel (or similar spreadsheet software):

1. Select both columns of data
2. Insert a **scatter plot**
3. Use “Years since 1995” on the x-axis
4. Use “Median Income” on the y-axis

This visualizes how income changes over time.

---

## Fitting a Line (Linear Regression)

Excel can automatically fit a line to the data:

1. Select the scatter plot
2. Add a **trendline**
3. Choose a **linear trendline**
4. Display the **equation of the line**

Excel produces the regression equation:

$$
y = 1882.3x + 52,\!847
$$

---

## Interpreting the Equation

### Y-Intercept (52,847)

- Represents the **predicted median income in 1995** (when $ x = 0 $)
- The actual 1995 value was slightly higher, but the line represents a **best overall fit**

### Slope (1,882.3)

- Indicates that median income increases by approximately **$1,882 per year**
- This is the **average annual change** according to the linear model

---

## Making a Prediction (Extrapolation)

To predict income in **2010**:

- 2010 is **15 years after 1995**
- Substitute $ x = 15 $ into the equation:

$$
y = 1882.3(15) + 52,\!847 \approx 81,\!081.50
$$

### Prediction:
**Median California family income in 2010 ≈ $81,000**

This is an example of **extrapolation**, since we are predicting beyond the observed data range.

---

## Key Takeaways

- Linear regression finds a line that best represents the overall trend
- The line minimizes the **squared distances** between data points and the line
- Spreadsheet tools make regression fast and accessible
- The **slope** shows rate of change
- The **y-intercept** anchors the model at the starting point
- Linear models can be used for **prediction**, but extrapolation should be done cautiously

---

## Why This Matters

This example shows why **linear models** are powerful:

- They summarize trends in real-world data
- They help interpret relationships
- They allow informed predictions
- They provide intuition for more advanced statistical tools

Spreadsheets are practical tools for exploring data and building intuition before diving into the mathematics behind regression.


```python
# Linear regression example: fitting data to a line and making a prediction
# This mirrors the Excel-based conceptual example using Python.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# -----------------------------
# Given data
# -----------------------------
# Years since 1995
x = np.array([0, 1, 2, 3, 4, 5, 6, 7]).reshape(-1, 1)

# Median California family income (USD)
y = np.array([
    53807,
    55217,
    55209,
    55415,
    63100,
    63206,
    63761,
    65766
])

# -----------------------------
# Fit linear regression model
# -----------------------------
model = LinearRegression()
model.fit(x, y)

slope = model.coef_[0]
intercept = model.intercept_

print(f"Regression equation:")
print(f"y = {slope:.1f}x + {intercept:.0f}")

# -----------------------------
# Predict income for 2010
# -----------------------------
# 2010 is 15 years after 1995
x_2010 = np.array([[15]])
income_2010 = model.predict(x_2010)[0]

print(f"\nPredicted median income in 2010: ${income_2010:,.2f}")

# -----------------------------
# Create scatter plot and regression line
# -----------------------------
x_line = np.linspace(0, 15, 100).reshape(-1, 1)
y_line = model.predict(x_line)

plt.figure()
plt.scatter(x, y, label="Observed Data")
plt.plot(x_line, y_line, label="Best-Fit Line")
plt.xlabel("Years Since 1995")
plt.ylabel("Median Family Income ($)")
plt.title("Linear Regression: California Median Income")
plt.legend()
plt.show()
```
