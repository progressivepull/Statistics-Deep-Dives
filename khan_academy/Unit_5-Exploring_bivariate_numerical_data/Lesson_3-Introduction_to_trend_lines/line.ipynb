# Estimating Smoking Rates Using a Linear Trend

This problem asks us to **estimate the percentage of American adults who smoked in 1945**, based on a graph showing smoking rates over time and the assumption that the **trend has been consistent since 1945**.

---

## Understanding the Graph

### Axes Interpretation
- **Horizontal axis:** *Years since 1965*
  - 0 → year **1965**
  - Positive values → years *after* 1965
  - Negative values → years *before* 1965
- **Vertical axis:** Percentage of American adults who smoke

---

## Observations from the Data

- At **0 years since 1965 (the year 1965)**:
  - The graph shows approximately **41–42%** of American adults smoked.
- As time increases:
  - The percentage of smokers **decreases steadily**
  - This indicates a **negative linear relationship**

This means:
> As years go by, the percentage of adults who smoke goes down.

Because the points follow a fairly straight downward pattern, it is reasonable to **fit a line** to the data.

---

## Identifying the Trend

By eyeballing the graph:

- Over a span of **20 years**, the smoking rate appears to decrease by about **10–11 percentage points**
- This decline looks **consistent** across the graph

So, the approximate rate of change is:
- **−10% per 20 years**  
  (or about −0.5% per year)

---

## Estimating the Smoking Rate in 1945

### Step 1: Locate 1945 on the Graph
- 1945 is **20 years before 1965**
- That corresponds to **−20 years since 1965**

### Step 2: Extend the Trend Backward
- If smoking rates decrease by ~10% every 20 years going forward,
- Then they must **increase by ~10%** every 20 years going backward

### Step 3: Apply the Estimate
- Smoking rate in 1965 ≈ **41–42%**
- Add ~10% for going back 20 years:

$$
41\% + 10\% \approx 51\% \quad \text{to} \quad 52\%
$$

---

## Final Estimate

**Estimated percentage of American adults who smoked in 1945:**  
$$
\boxed{51\% \text{ to } 52\%}
$$

---

## Important Notes on Interpretation

- This is an **extrapolation**, not a direct observation
- The estimate depends on the assumption that:
  - The **linear trend remained consistent before 1965**
- Real-world behavior may not follow a perfect linear pattern

---

## Key Takeaways

- Linear trends can be used to estimate past or future values
- Negative slope indicates a decreasing trend over time
- Extrapolation should always be treated as an **approximation**
- Visual estimation (“eyeballing”) can be reasonable when data are clearly linear

This example highlights how graphs and linear reasoning can be used to make informed estimates—even when exact data are unavailable.


```python
# Estimating historical smoking rates using a linear trend
# This example mirrors the graph-based reasoning used in the problem.

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# Approximate data from the graph
# (years since 1965, percent smoking)
# -----------------------------------
years_since_1965 = np.array([0, 5, 10, 15, 20, 25, 30])
percent_smokers = np.array([42, 38, 34, 30, 26, 23, 20])

# -----------------------------------
# Fit a linear regression line
# -----------------------------------
slope, intercept = np.polyfit(years_since_1965, percent_smokers, 1)

print(f"Best-fit line:")
print(f"Percent smoking = {slope:.3f}(years since 1965) + {intercept:.2f}")

# -----------------------------------
# Estimate smoking rate in 1945
# 1945 is 20 years before 1965 → x = -20
# -----------------------------------
year_1945 = -20
estimate_1945 = slope * year_1945 + intercept

print(f"\nEstimated percent of smokers in 1945: {estimate_1945:.1f}%")

# -----------------------------------
# Plot data, regression line, and estimate
# -----------------------------------
x_line = np.linspace(-25, 35, 200)
y_line = slope * x_line + intercept

plt.figure(figsize=(7, 4))
plt.scatter(years_since_1965, percent_smokers, label="Observed Data")
plt.plot(x_line, y_line, label="Linear Trend")
plt.scatter(year_1945, estimate_1945, color="red", label="1945 Estimate")

plt.xlabel("Years Since 1965")
plt.ylabel("Percent of Adults Who Smoke")
plt.title("Estimating Smoking Rates Using Linear Extrapolation")
plt.axvline(0, linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
```

# What this code demonstrates

* How to **model a trend** using a straight line
* How to **extrapolate backward in time**
* Why negative x-values represent years before 1965
* How graph-based intuition translates into computation

# Key takeaway

This Python example reproduces the same reasoning used in the problem:                                          
a **negative linear relationship** allows us to estimate that a**bout 50–52%** of American adults smoked in **1945**, assuming the trend remained consistent.
