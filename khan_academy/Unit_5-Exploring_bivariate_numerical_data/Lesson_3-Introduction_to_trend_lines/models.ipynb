# Modeling Test Scores vs. Study Time with a Line

This example examines a **scatter plot** showing the relationship between how many hours students reported studying for a math test and the score they earned. Each point represents a student, and a **line has been fit to the data** to model this relationship.

The task is to:
1. Identify which **linear equation** best represents the fitted line
2. Use that equation to **estimate a test score**
3. Reflect on the **limitations of linear models**

---

## Understanding the Scatter Plot

From the plot:

- A student who studied **a little over 0.5 hours** scored around **43–44**
- A student who studied about **2 hours** scored around **64–65**
- A student who studied **over 4 hours** scored around **95–96**

Overall, the points show a **clear positive linear trend**:
- More study time → higher test score
- The fitted line appears to describe the data reasonably well

---

## Identifying the Equation of the Line

### Step 1: Find the Y-Intercept

- The line crosses the y-axis at **20**
- This suggests that a student who studied **0 hours** would be predicted to score **20**

All answer choices share this same y-intercept, so this alone does not determine the correct equation.

---

### Step 2: Determine the Slope

To estimate the slope, observe how the score changes as study time increases:

- When study time increases by **1 hour**
- The test score increases by approximately **20 points**

So the slope is:

$$
\text{slope} = \frac{\Delta y}{\Delta x} = \frac{20}{1} = 20
$$

---

### Best-Fit Linear Equation

The equation that best matches the model is:

$$
y = 20x + 20
$$

Where:
- $ x $ = hours studied
- $ y $ = predicted test score

---

## Estimating a Score Using the Model

### Given:
A student studied **3.8 hours**

### Using the equation:
$$
y = 20(3.8) + 20 = 76 + 20 = 96
$$

### Visual Estimate from the Graph:
- The intersection appears close to **97**

### Final Estimate:
**Approximately 97**

---

## Interpreting the Model

- The model suggests that **each additional hour of studying increases the score by about 20 points**
- The y-intercept reflects a **baseline score** with no studying
- The line provides a **reasonable estimate**, not a guarantee

---

## Important Cautions About Linear Models

- This is only a **model**, not a rule
- Extrapolating too far can lead to impossible predictions  
  (e.g., studying 9 hours does *not* mean scoring 200)
- Real-world outcomes depend on many factors besides study time

---

## Key Takeaways

- Linear models help summarize trends in data
- Slope represents **rate of change**
- Y-intercept represents a **baseline prediction**
- Predictions should be used carefully and interpreted with context

Linear regression is a powerful tool—but it should always be used with **critical thinking and common sense**.


```python
# Linear model example: test score vs. hours studied
# This mirrors the conceptual problem and fitted line from the scatter plot.

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# Given (approximate) observed data
# -----------------------------------
# Hours studied (x) and test score (y)
x = np.array([0.5, 1.2, 2.0, 2.5, 3.1, 3.8, 4.2])
y = np.array([44, 55, 65, 72, 82, 97, 96])

# -----------------------------------
# Define the fitted linear model
# From the problem: y = 20x + 20
# -----------------------------------
def model(x):
    return 20 * x + 20

# -----------------------------------
# Estimate score for 3.8 hours
# -----------------------------------
hours = 3.8
predicted_score = model(hours)

print(f"Predicted score for {hours} hours studied: {predicted_score:.0f}")

# -----------------------------------
# Plot scatter data and fitted line
# -----------------------------------
x_line = np.linspace(0, 5, 100)
y_line = model(x_line)

plt.figure(figsize=(6, 4))
plt.scatter(x, y, label="Student Data")
plt.plot(x_line, y_line, label="Model: y = 20x + 20")

plt.xlabel("Hours Studied")
plt.ylabel("Test Score")
plt.title("Test Score vs. Hours Studied")
plt.legend()
plt.grid(True)
plt.show()
```

# What this code illustrates

* How a **linear equation** models a relationship between two variables
* How to use a model to **estimate outcomes**
* The meaning of:

    - **Slope (20)** → points gained per hour studied
    - **Y-intercept (20)** → baseline score with no studying

* Why predictions are estimates, not guarantees
