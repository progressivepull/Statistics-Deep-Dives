```python
# Identifying a Line of Best Fit (or No Linear Correlation)

This exercise focuses on **deciding whether a linear model is appropriate** for a given set of data points. The task is not to compute an exact equation, but to visually determine whether a **line of best fit** reasonably captures the overall trend—or whether **no linear correlation** exists.

---

## What Does “Line of Best Fit” Mean?

A **line of best fit** is a line that:
- Captures the **overall trend** in the data
- Balances the data points above and below the line
- Does **not ignore clusters of points** to favor others

If no such line reasonably represents the data, we conclude that **no linear correlation** exists.

---

## Example 1: No Good Line of Best Fit

- The data points are scattered in such a way that:
  - A line could be drawn to fit *some* points
  - But doing so would require **ignoring several other points**
- Different lines could be justified depending on which points are ignored

### Conclusion:
**No good line of best fit exists**  
There is **no clear linear trend** in the data.

---

## Example 2: Still No Linear Correlation

- Similar situation as the first example
- Any proposed line:
  - Fits one group of points
  - But fails badly for another group
- The data does not consistently increase or decrease

### Conclusion:
**No good best-fit line exists**  
The data shows **no meaningful linear correlation**.

---

## Example 3: Clear Linear Trend

- The data points follow a recognizable upward trend
- A single line:
  - Fits most of the points reasonably well
  - Does not require ignoring major portions of the data
- Small adjustments to the line improve the fit without bias

### Conclusion:
**A good line of best fit exists**  
The data shows a **clear linear relationship**.

---

## Key Takeaways

- A line of best fit should represent **all the data**, not just a subset
- If fitting a line requires ignoring many points, the data likely has:
  - **Weak or no linear correlation**
- Visual inspection is a powerful first step before calculations
- Not all datasets are well-modeled by a line—and that’s an important result

---

## Big Picture

Understanding when **not** to use a linear model is just as important as knowing how to compute one.  
This skill builds intuition for:
- Correlation
- Regression
- Model selection in real-world data analysis
```

```python
# Determining whether a line of best fit exists

import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Approximate data points taken from the scatterplot
# (visually estimated from the image)
# --------------------------------------------------
x = np.array([-7, -5,  2,  3,  6,  7,  0, -1, -2])
y = np.array([ 7,  8,  6,  6,  9,  6, -3, -5, -7])

# --------------------------------------------------
# Compute correlation coefficient
# --------------------------------------------------
r = np.corrcoef(x, y)[0, 1]
print(f"Correlation coefficient r = {r:.3f}")

# --------------------------------------------------
# Fit a linear regression line
# --------------------------------------------------
slope, intercept = np.polyfit(x, y, 1)
print(f"Best-fit line: y = {slope:.2f}x + {intercept:.2f}")

# --------------------------------------------------
# Plot the data and the fitted line
# --------------------------------------------------
x_line = np.linspace(min(x) - 1, max(x) + 1, 100)
y_line = slope * x_line + intercept

plt.figure(figsize=(6, 6))
plt.scatter(x, y, s=60, label="Data points")
plt.plot(x_line, y_line, linestyle="--", label="Best-fit line")

plt.axhline(0, linewidth=0.5)
plt.axvline(0, linewidth=0.5)

plt.xlabel("x")
plt.ylabel("y")
plt.title("No Clear Linear Correlation")
plt.legend()
plt.grid(True)
plt.show()
```

# What this code shows

* The **correlation coefficient** is close to **0**
* The fitted line does **not** represent the data well
* Points are scattered in different directions
* Any line drawn would ignore significant groups of points

# Conclusion

✔ **No meaningful line of best fit exists**                               
✔ The dataset shows **no linear correlation**
