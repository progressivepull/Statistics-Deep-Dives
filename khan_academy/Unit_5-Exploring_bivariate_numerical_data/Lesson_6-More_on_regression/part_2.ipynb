# Simplifying the Sum of Squared Errors for Linear Regression

Our goal is to simplify the expression for the total squared error between *n* data points $(x_i, y_i)$ and the line y = mx + b.

The total squared error (SSE) is:

$$
SSE = \sum_{i=1}^n (y_i - mx_i - b)^2
$$

After expanding this expression (as done in previous steps), we arrive at a more complex algebraic form. To simplify it, we introduce statistical notation using means:

- $\bar{y^2}$ = mean of the squared y-values = $\frac{1}{n} \sum y_i^2$
- $\bar{xy}$ = mean of the products = $\frac{1}{n} \sum x_i y_i$
- $\bar{y}$ = mean of the y-values = $\frac{1}{n} \sum y_i$
- $\bar{x^2}$ = mean of the squared x-values = $\frac{1}{n} \sum x_i^2$
- $\bar{x}$ = mean of the x-values = $\frac{1}{n} \sum x_i$

Multiplying each mean by *n* recovers the original sums:

- $\sum y_i^2 = n \bar{y^2}$
- $\sum x_i y_i = n \bar{xy}$
- $\sum y_i = n \bar{y}$
- $\sum x_i^2 = n \bar{x^2}$
- $\sum x_i = n \bar{x}$

Substituting these into the expanded SSE gives a much cleaner expression:

$$
SSE = n \bar{y^2} - 2m \cdot n \bar{xy} - 2b \cdot n \bar{y} + m^2 \cdot n \bar{x^2} + 2mb \cdot n \bar{x} + n b^2
$$

This simplified form uses only the computed means, making the algebra far more manageable.

## Next Step: Minimizing the Squared Error

With the data points fixed, all means $bar{y^2}, \bar{xy}, \bar{y}, \bar{x^2}, \bar{x}$ are constants. The SSE now becomes a function of the two variables *m* and *b*:

$$
SSE(m, b) = n \bar{y^2} - 2m n \bar{xy} - 2b n \bar{y} + m^2 n \bar{x^2} + 2mb n \bar{x} + n b^2
$$

This expression represents a surface in 3D space:

- Horizontal axes: *m* and *b*
- Vertical axis: SSE (total squared error)

The surface forms a bowl-shaped quadratic (a rotated and possibly distorted 3D paraboloid) with a single minimum point.

To find the values of *m* and *b* that minimize the SSE (i.e., the best-fitting line), we locate the bottom of this bowl. This minimum occurs where both partial derivatives are zero:

$$
\frac{\partial SSE}{\partial m} = 0 \quad \text{and} \quad \frac{\partial SSE}{\partial b} = 0
$$

In the next step, we will compute these partial derivatives, set them equal to zero, and solve the resulting system of equations for *m* and *b*. This yields the closed-form solution for ordinary least squares regression.
