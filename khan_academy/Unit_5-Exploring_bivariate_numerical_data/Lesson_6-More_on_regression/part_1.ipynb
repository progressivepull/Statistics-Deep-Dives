# Algebraic Expansion of Squared Error (Preparation for Optimization)

In the previous discussion, we showed that the **squared error** between a line  
$$
y = mx + b
$$
and each of the $ n $ data points can be written as a sum of squared residuals.

In this section, the goal is **purely algebraic**: to rewrite and expand this expression so it is ready for the **calculus stage**, where we will later optimize it to find the values of $ m $ and $ b $ that minimize the total error.

---

## Starting Point: Squared Error Expression

The total squared error between the line and the data points is

$$
\sum_{i=1}^{n} \left( y_i - (m x_i + b) \right)^2
$$

This entire discussion focuses on **rewriting this same expression repeatedly**, using algebraic expansion and simplification.

---

## Expanding Each Squared Term

Consider the first term:

$$
(y_1 - (m x_1 + b))^2
$$

Expanding the binomial gives:

$$
y_1^2 - 2 y_1 (m x_1 + b) + (m x_1 + b)^2
$$

Expanding further:

$$
y_1^2
- 2 m x_1 y_1
- 2 b y_1
+ m^2 x_1^2
+ 2 m b x_1
+ b^2
$$

The same expansion applies to every data point. The only difference is the index:

- Second point: $ x_2, y_2 $
- Third point: $ x_3, y_3 $
- …
- $ n $-th point: $ x_n, y_n $

---

## Writing Out the Full Sum

After expanding all $ n $ terms, the total squared error becomes the sum of all corresponding pieces:

$$
\begin{aligned}
& (y_1^2 + y_2^2 + \cdots + y_n^2) \\
& - 2m (x_1 y_1 + x_2 y_2 + \cdots + x_n y_n) \\
& - 2b (y_1 + y_2 + \cdots + y_n) \\
& + m^2 (x_1^2 + x_2^2 + \cdots + x_n^2) \\
& + 2mb (x_1 + x_2 + \cdots + x_n) \\
& + n b^2
\end{aligned}
$$

Each group comes from **factoring common terms** across all $ n $ expanded expressions.

---

## Key Observations

- Every term arises directly from expanding the squared residuals.
- Like terms naturally group together across all data points.
- No calculus has been used yet—this is **entirely algebraic manipulation**.
- The final result is a cleaner expression that depends only on:
  - Sums of $ x_i $, $ y_i $
  - Sums of squares $ x_i^2 $, $ y_i^2 $
  - Cross-products $ x_i y_i $

---

## What Comes Next

This expanded form is the **starting point for optimization**.  
In the next step, we will continue simplifying and then apply calculus to find the values of $ m $ and $ b $ that minimize the squared error.

At this stage, the hard work is algebra—setting the problem up so calculus can be applied cleanly.


# 🧮 Python Example: Expanding the Squared Error for Linear Regression
Below is a Python script that:

* Computes the squared error expression
* Expands it term‑by‑term exactly like the algebra you walked through
* Produces the fully expanded symbolic form
* Lets you plug in numerical data to evaluate the expression
* This uses SymPy, which is perfect for symbolic algebra.


# 📌 Symbolic Expansion of the Squared Error

```python
# Sample data
data_x = [1, 2, 3]
data_y = [2, 4, 5]
slope = 1.5
intercept = 0.5

# Calculate SSE manually using a loop (standard approach)
sse = sum((data_y[j] - (slope * data_x[j] + intercept))**2 for j in range(len(data_x)))
print(f"Numerical Sum of Squared Errors: {sse}")
```

# Linear Regression: Squared Error Derivation

### 🔍 What This Code Does

---

### 1. Builds the squared error
It constructs the sum of squared residuals, $S(m, b)$, exactly like your starting point:
$$S(m, b) = \sum_{i=1}^{n}(y_i - (mx_i + b))^2$$

---

### 2. Expands each term
The logic follows your algebra by expanding the outer square first:
$$(y_i - (mx_i + b))^2 \implies y_i^2 - 2y_i(mx_i + b) + (mx_i + b)^2$$

---

### 3. Expands again
It then expands the remaining squared binomial:
$$(mx_i + b)^2 = m^2x_i^2 + 2mbx_i + b^2$$

---

### 4. Collects like terms
Finally, the summation is distributed and like terms are grouped to produce the final form:
$$S(m, b) = m^2\sum x_i^2 - 2m\sum x_iy_i + 2mb\sum x_i + nb^2 - 2b\sum y_i + \sum y_i^2$$

**Note:** In the term $nb^2$, the $n$ appears because we are summing the constant $b^2$ over $n$ observations. This structure is exactly what you derived algebraically.
