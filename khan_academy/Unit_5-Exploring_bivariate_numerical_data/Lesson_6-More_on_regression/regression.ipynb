# Least Squares Regression: Final Formula and Worked Example

## Big Picture

In the previous videos, we worked through some fairly heavy mathematics to arrive at a very clean and useful result:

> **A closed-form formula for the slope and y-intercept of the best-fitting regression line**,  
when error is measured using **squared distances** (least squares).

Even if you skipped some of the derivations, this result is the key takeaway.

---

## Final Formulas for the Regression Line

We are fitting a line of the form:

$$
y = mx + b
$$

### Slope $ m $

One common form of the slope formula is:

$$
m
=
\frac{
\overline{x}\,\overline{y} - \overline{xy}
}{
(\overline{x})^2 - \overline{x^2}
}
$$

You may also see this written as:

$$
m
=
\frac{
\overline{xy} - \overline{x}\,\overline{y}
}{
\overline{x^2} - (\overline{x})^2
}
$$

These two expressions are **identical**—the second is obtained by multiplying the numerator and denominator of the first by $-1$.

---

### Intercept $ b $

Once $ m $ is known, the y-intercept is:

$$
b = \overline{y} - m\,\overline{x}
$$

That’s it—these two formulas fully define the optimal regression line.

---

## Worked Example

### Given Data Points

We are given three non-colinear points:

$$
(1, 2), \quad (2, 1), \quad (4, 3)
$$

Our goal is to find the **least squares regression line** for these points.

---

## Step 1: Compute the Required Means

### Mean of $ x $

$$
\overline{x}
=
\frac{1 + 2 + 4}{3}
=
\frac{7}{3}
$$

---

### Mean of $ y $

$$
\overline{y}
=
\frac{2 + 1 + 3}{3}
=
2
$$

---

### Mean of $ xy $

$$
\overline{xy}
=
\frac{(1)(2) + (2)(1) + (4)(3)}{3}
=
\frac{2 + 2 + 12}{3}
=
\frac{16}{3}
$$

---

### Mean of $ x^2 $

$$
\overline{x^2}
=
\frac{1^2 + 2^2 + 4^2}{3}
=
\frac{1 + 4 + 16}{3}
=
7
$$

---

## Step 2: Compute the Slope $ m $

Substitute into the slope formula:

$$
m
=
\frac{
\left(\frac{7}{3}\right)(2) - \frac{16}{3}
}{
\left(\frac{7}{3}\right)^2 - 7
}
$$

### Simplify

Numerator:
$$
\frac{14}{3} - \frac{16}{3} = -\frac{2}{3}
$$

Denominator:
$$
\frac{49}{9} - \frac{63}{9} = -\frac{14}{9}
$$

$$
m
=
\frac{-\frac{2}{3}}{-\frac{14}{9}}
=
\frac{2}{3} \cdot \frac{9}{14}
=
\frac{3}{7}
$$

---

## Step 3: Compute the Intercept $ b $

$$
b
=
\overline{y} - m\,\overline{x}
=
2 - \left(\frac{3}{7}\right)\left(\frac{7}{3}\right)
=
2 - 1
=
1
$$

---

## Final Regression Line

$$
\boxed{
y = \frac{3}{7}x + 1
}
$$

---

## Interpretation

- The line does **not** pass exactly through all points
- Instead, it **minimizes the total squared distance** from the points to the line
- This is the defining property of **least squares regression**

Graphically:
- The y-intercept is $ 1 $
- The slope is $ \frac{3}{7} $ (rise 3 for every run of 7)

---

## Key Takeaway

Despite the heavy algebra needed to derive it, the final result is elegant:

- Compute four means:  
  $ \overline{x}, \overline{y}, \overline{xy}, \overline{x^2} $
- Plug them into two simple formulas
- You get the **best-fitting line** under squared-error minimization

At least mathematically, that’s a pretty neat result.


# Least Squares Regression Line
It computes the least squares regression line manually (using the derived formulas) and also verifies the result using NumPy.

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Example Data

```python
# Given data points
x = np.array([1, 2, 4])
y = np.array([2, 1, 3])
```

### Step 1: Compute the Required Means

```python
x_mean = np.mean(x)
y_mean = np.mean(y)
xy_mean = np.mean(x * y)
x2_mean = np.mean(x ** 2)

x_mean, y_mean, xy_mean, x2_mean
```

###  Step 2: Compute Slope (m) Using the Derived Formula

$$
m = \frac{\overline{x} \cdot \overline{y} - \overline{xy}}{(\overline{x})^2 - \overline{x^2}}
$$

```python
m = (x_mean * y_mean - xy_mean) / (x_mean**2 - x2_mean)
m
```

### Step 3: Compute Intercept (b)

$$
b = \overline{y} - m \cdot \overline{x}
$$


```python
b = y_mean - m * x_mean
b
```

### Final Regression Line

```python
print(f"Regression line: y = {m:.4f}x + {b:.4f}")
```

### Step 4: Verify Using NumPy (Optional Check)

```python
m_np, b_np = np.polyfit(x, y, 1)
m_np, b_np
```

✅ The values should match (up to floating-point rounding).

### Step 5: Plot the Data and Regression Line

```python
# Generate x values for the regression line
x_line = np.linspace(min(x) - 0.5, max(x) + 0.5, 100)
y_line = m * x_line + b

plt.figure()
plt.scatter(x, y)
plt.plot(x_line, y_line)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Least Squares Regression Line")
plt.show()
```

### Summary

* This code implements the least squares formulas directly
* Confirms results with numpy.polyfit
* Visualizes the best-fitting regression line
* Matches the derived result:

$$
y = -\frac{3}{7}x + 1
$$
