# Deriving the Least-Squares Regression Line

In this lesson, we begin a process that ultimately leads to a **simple and widely used formula** in statistics: the equation of the least-squares regression line. Although most statistics courses present only the final formula, this approach focuses on **understanding how that formula is derived**.

> ⚠️ **Warning:**  
> The derivation involves **heavy algebra** and a small amount of **calculus** (partial derivatives). If that feels overwhelming, it’s perfectly acceptable to skip ahead and use the final formula. However, seeing the derivation can be very satisfying and deepen understanding.

---

## Setup: A Collection of Data Points

Assume we are given **n data points** on a coordinate plane:

$$
(x_1, y_1), (x_2, y_2), \dots, (x_n, y_n)
$$

For simplicity, we visualize these points in the first quadrant, though this assumption is not required.

---

## Goal: Find the Best-Fitting Line

We want to find a line that best represents the relationship between the data points. The general form of a line is:

$$
y = mx + b
$$

Where:
- $m$ is the **slope**
- $b$ is the **y-intercept**

Our objective is to find the values of $m$ and $b$ that make this line the **best fit** for the given data.

---

## Defining the Error

For each data point, the **error** is defined as the **vertical distance** between the actual y-value and the y-value predicted by the line.

For the first point:
$$
\text{Error}_1 = y_1 - (mx_1 + b)
$$

Similarly:
$$
\text{Error}_2 = y_2 - (mx_2 + b)
$$
$$
\vdots
$$
$$
\text{Error}_n = y_n - (mx_n + b)
$$

---

## Squared Error

Instead of summing the raw errors (which could cancel out), we square each error. This leads to the **sum of squared errors (SSE)**:

$$
\text{SSE} = (y_1 - (mx_1 + b))^2
+ (y_2 - (mx_2 + b))^2
+ \cdots
+ (y_n - (mx_n + b))^2
$$

This quantity measures how well the line fits the data.

---

## Next Step

The goal of least-squares regression is to **find the values of $m$ and $b$** that **minimize** this sum of squared errors.

In the next part, we will:
- Treat the SSE as a function of $m$ and $b$
- Use **calculus (partial derivatives)** to find the values of $m$ and $b$ that minimize it

This step-by-step approach helps ensure clarity and reduces the chance of mistakes when working through the mathematics.

---


Below is a Python code example suitable for a Jupyter Notebook that demonstrates the idea described in your summary:
defining errors, squaring them, and finding the least-squares regression line.

This example shows:

1. How the sum of squared errors (SSE) is defined
2. How calculus (via optimization) finds the best 𝑚 and 𝑏
3. The resulting regression line

# Least-Squares Regression: Python Example

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
```

# Step 1: Sample Data (n points)

```python
# Example dataset
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 4, 6])

n = len(x)
```

# Step 2: Define the Sum of Squared Errors (SSE)

This matches the mathematical expression:

$$
\sum_{i=1}^{n} (y_i - (m x_i + b))^2
$$


```python
def squared_error(params, x, y):
    m, b = params
    predictions = m * x + b
    errors = y - predictions
    return np.sum(errors**2)
```

# Step 3: Minimize the Squared Error

This step conceptually replaces taking partial derivatives and setting them equal to zero.

```python
# Initial guess for m and b
initial_guess = [0, 0]

# Minimize the SSE
result = minimize(squared_error, initial_guess, args=(x, y))

m_opt, b_opt = result.x
```

# Step 4: Display the Results

```python
print(f"Optimal slope (m): {m_opt:.4f}")
print(f"Optimal intercept (b): {b_opt:.4f}")
```

# Step 5: Visualize the Fit

```python
# Generate regression line
x_line = np.linspace(min(x), max(x), 100)
y_line = m_opt * x_line + b_opt

plt.scatter(x, y)
plt.plot(x_line, y_line)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Least-Squares Regression Line")
plt.show()
```

# What This Code Demonstrates

* Each data point contributes a **vertical error**
* Errors are **squared and summed**
* The best-fitting line minimizes this total squared error
* The resulting line is the **least-squares regression line**


## Key Takeaway

This code directly mirrors the mathematical derivation of linear regression by least squares.

1. **Define the errors (residuals):**

   $$
   y_i - (m x_i + b)
   $$

2. **Square the errors and sum them:**

   $$
   \sum_{i=1}^{n} \left( y_i - (m x_i + b) \right)^2
   $$

3. **Optimize the parameters:**

   Find the values of $ m $ (slope) and $ b $ (intercept) that minimize this sum.
