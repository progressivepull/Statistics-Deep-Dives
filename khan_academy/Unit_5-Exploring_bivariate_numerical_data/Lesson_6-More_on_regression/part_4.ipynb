# Solving for the Optimal Line (Least Squares Regression)

## Goal

After several steps of algebra and calculus, we now arrive at the **final objective**:

> **Find the optimal line**  
$$
y = mx + b
$$
that **minimizes the squared distance** (squared error) to a set of data points.

This is the classic **least squares regression line**.

---

## Two Equivalent Approaches

From previous results, we know there are **two mathematically equivalent ways** to find the optimal values of $ m $ and $ b $:

1. **Geometric approach**:  
   Use the fact that we know **two points** that lie on the optimal line and compute the slope and intercept directly.

2. **Algebraic approach**:  
   Solve the **system of equations** obtained from setting the partial derivatives of the squared error to zero.

Here, we proceed with the **algebraic approach**, while noting its equivalence to the geometric interpretation.

---

## System of Equations

From earlier derivations, the optimal $ m $ and $ b $ satisfy:

$$
\begin{aligned}
m \, \overline{x^2} + b \, \overline{x} &= \overline{xy} \\
m \, \overline{x} + b &= \overline{y}
\end{aligned}
$$

---

## Solving for the Slope $ m $

To eliminate $ b $, subtract the second equation (after appropriate scaling) from the first.

This yields:

$$
m \left( \overline{x} - \frac{\overline{x^2}}{\overline{x}} \right)
=
\overline{y} - \frac{\overline{xy}}{\overline{x}}
$$

Solving for $ m $:

$$
m
=
\frac{
\overline{y} - \frac{\overline{xy}}{\overline{x}}
}{
\overline{x} - \frac{\overline{x^2}}{\overline{x}}
}
$$

---

## Simplifying the Expression for $ m $

To remove fractions within fractions, multiply **both numerator and denominator** by $ \overline{x} $:

$$
m
=
\frac{
\overline{x}\,\overline{y} - \overline{xy}
}{
(\overline{x})^2 - \overline{x^2}
}
$$

This is the **closed-form solution for the slope** of the least-squares regression line.

---

## Geometric Interpretation

This formula for $ m $ is exactly the same as computing the **slope between the two known points** on the optimal line:

- $ (\overline{x}, \overline{y}) $
- $ \left( \dfrac{\overline{x^2}}{\overline{x}}, \dfrac{\overline{xy}}{\overline{x}} \right) $

Thus:
$$
m = \frac{\Delta y}{\Delta x}
$$

This confirms the equivalence between the **algebraic** and **geometric** approaches.

---

## Solving for the Intercept $ b $

Once $ m $ is known, substitute it into the simpler equation:

$$
m \, \overline{x} + b = \overline{y}
$$

Solving for $ b $:

$$
b = \overline{y} - m \, \overline{x}
$$

---

## Final Procedure (Algorithm)

Given a dataset:

1. Compute:
   - $ \overline{x} $
   - $ \overline{y} $
   - $ \overline{xy} $
   - $ \overline{x^2} $

2. Compute the slope:
$$
m
=
\frac{
\overline{x}\,\overline{y} - \overline{xy}
}{
(\overline{x})^2 - \overline{x^2}
}
$$

3. Compute the intercept:
$$
b = \overline{y} - m \, \overline{x}
$$

4. Write the optimal line:
$$
y = mx + b
$$

---

## Key Takeaways

- These formulas define the **best-fitting line** under the **squared error** criterion.
- The optimal line:
  - Minimizes total squared error
  - Passes through $ (\overline{x}, \overline{y}) $
- This is the mathematical foundation of **simple linear regression**.
