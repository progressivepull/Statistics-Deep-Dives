# Covariance and Its Connection to Least Squares Regression

## Introduction

In this section, we introduce the idea of **covariance** between two random variables. Covariance measures how two variables vary **together**—whether they tend to increase and decrease simultaneously, or move in opposite directions.

Formally, the covariance between two random variables $X$ and $Y$ is defined as the expected value of the product of their deviations from their respective means.

---

## Definition of Covariance

$$
\operatorname{Cov}(X, Y) = \mathbb{E}\left[(X - \mathbb{E}[X])(Y - \mathbb{E}[Y])\right]
$$

- $\mathbb{E}[X]$ is the expected value (population mean) of $X$
- $\mathbb{E}[Y]$ is the expected value (population mean) of $Y$

This expression measures how much $X$ and $Y$ deviate from their means **together**.

---

## Intuition Behind Covariance

Consider a single observation from the population:

- Suppose $X = 1$ and $Y = 3$
- Suppose $\mathbb{E}[X] = 0$ and $\mathbb{E}[Y] = 4$

Then the product inside the expectation is:

$$
(1 - 0)(3 - 4) = 1 \times (-1) = -1
$$

This negative value indicates that, for this observation:
- $X$ is **above** its mean
- $Y$ is **below** its mean

If this pattern occurred consistently across the population, the covariance would be **negative**, meaning the variables tend to move in opposite directions.

- **Positive covariance**: both variables tend to increase or decrease together
- **Negative covariance**: one variable tends to increase while the other decreases
- **Magnitude**: indicates the strength of this relationship

---

## Algebraic Expansion of Covariance

Starting from the definition:

$$
\operatorname{Cov}(X, Y) = \mathbb{E}\left[(X - \mathbb{E}[X])(Y - \mathbb{E}[Y])\right]
$$

Expand the product:

$$
= \mathbb{E}[XY - X\mathbb{E}[Y] - \mathbb{E}[X]Y + \mathbb{E}[X]\mathbb{E}[Y]]
$$

Using linearity of expectation:

$$
= \mathbb{E}[XY]
- \mathbb{E}[X]\mathbb{E}[Y]
- \mathbb{E}[X]\mathbb{E}[Y]
+ \mathbb{E}[X]\mathbb{E}[Y]
$$

After cancellation:

$$
\boxed{
\operatorname{Cov}(X, Y) = \mathbb{E}[XY] - \mathbb{E}[X]\mathbb{E}[Y]
}
$$

This is the most commonly used and practical formula for covariance.

---

## Estimating Covariance from Sample Data

When working with a **sample** rather than the full population:

- $\mathbb{E}[XY]$ is approximated by the **sample mean of the products** $x_i y_i$
- $\mathbb{E}[X]$ is approximated by the sample mean of $x$
- $\mathbb{E}[Y]$ is approximated by the sample mean of $y$

Thus, sample covariance can be approximated as:

$$
\operatorname{Cov}(X, Y) \approx \overline{XY} - \bar{X}\bar{Y}
$$

---

## Connection to Least Squares Regression

Recall the formula for the slope $m$ of the least squares regression line:

$$
m = \frac{\overline{XY} - \bar{X}\bar{Y}}{\overline{X^2} - \bar{X}^2}
$$

Notice:

- The **numerator** is the **covariance between $X$ and $Y$**
- The **denominator** is the **variance of $X$**

Why?

- $\overline{X^2} - \bar{X}^2 = \operatorname{Var}(X)$
- The covariance of a variable with itself is its variance

---

## Final Interpretation

The slope of the regression line can be written as:

$$
\boxed{
m = \frac{\operatorname{Cov}(X, Y)}{\operatorname{Var}(X)}
}
$$

This shows that:

- Regression slope measures how strongly $X$ and $Y$ vary together
- Covariance is the key quantity connecting probability theory and regression
- Many ideas in statistics are deeply interconnected

---

## Summary

- Covariance measures **joint variability**
- It is positive when variables move together, negative when they move oppositely
- Algebraically:
  $$
  \operatorname{Cov}(X, Y) = \mathbb{E}[XY] - \mathbb{E}[X]\mathbb{E}[Y]
  $$
- Least squares regression slope is **covariance divided by variance**
- This connection explains why covariance plays a central role in regression analysis


# Covariance

Below is a clear Python code example suitable for a Jupyter Notebook that demonstrates:

1. Covariance (manual and NumPy)
2. Variance of $X$
3. Regression slope using
$$
m = \frac{\text{Cov}(X, Y)}{\text{Var}(X)}
$$
4. Verification with NumPy’s regression
5. Visualization


```python
import numpy as np
import matplotlib.pyplot as plt
```

###  Example Data

```python
# Sample data points
x = np.array([-2, -1, 1, 4])
y = np.array([-3, -1, 2, 3])
```

### Step 1: Compute Sample Means

```python
x_mean = np.mean(x)
y_mean = np.mean(y)

x_mean, y_mean
```

### Step 2: Compute Covariance Manually

Using the formula:

$$
\text{Cov}(X, Y) = \overline{XY} - \overline{X} \cdot \overline{Y}
$$


```python
xy_mean = np.mean(x * y)
cov_xy = xy_mean - x_mean * y_mean

cov_xy
```

### Step 3: Compute Variance of X

$$
\text{Var}(X) = \overline{X^2} - \left( \overline{X} \right)^2
$$


```python
x2_mean = np.mean(x ** 2)
var_x = x2_mean - x_mean ** 2

var_x
```

### Step 4: Compute Regression Slope Using Covariance

$$
m = \frac{\text{Cov}(X, Y)}{\text{Var}(X)}
$$


```python
m = cov_xy / var_x
m
```

### Step 5: Compute Intercept

$$
b = \bar{Y} - m \cdot \bar{X}
$$


```python
b = y_mean - m * x_mean
b
```

### Regression Line

```python
print(f"Regression line: y = {m:.4f}x + {b:.4f}")
```

### Step 6: Verify Using NumPy

```python
m_np, b_np = np.polyfit(x, y, 1)
m_np, b_np
```

### Step 7: Visualize the Result

```python
x_line = np.linspace(min(x) - 1, max(x) + 1, 100)
y_line = m * x_line + b

plt.figure()
plt.scatter(x, y)
plt.plot(x_line, y_line)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Regression Line via Covariance")
plt.show()
```

# Key Takeaway

* Covariance measures how two variables vary together
* Variance measures spread of a single variable
* Regression slope is:

$$
m = \frac{\text{Cov}(X, Y)}{\text{Var}(X)}
$$

This code shows exactly how covariance naturally leads to the least-squares regression line.
