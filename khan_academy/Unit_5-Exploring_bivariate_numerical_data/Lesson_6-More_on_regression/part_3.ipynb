# Minimizing Squared Error Using Partial Derivatives

## Overview

We previously simplified the algebraic expression for the **squared error** between a line  
$$
y = mx + b
$$
and a set of $ n $ data points. This squared error can be visualized as a **surface in three dimensions**, where:

- Each pair $(m, b)$ corresponds to a point on the surface
- The height of the surface represents the total squared error for that line

Our objective is to find the values of $ m $ and $ b $ that **minimize** this squared error.

---

## Geometric Intuition

The minimum squared error occurs at a point on this surface where:

- The partial derivative with respect to $ m $ is zero
- The partial derivative with respect to $ b $ is also zero

At this point, the surface is **flat in both directions**, indicating a minimum.

---

## Strategy

To find this minimum:

1. Take the **partial derivative** of the squared error with respect to $ m $
2. Take the **partial derivative** with respect to $ b $
3. Set both derivatives equal to zero
4. Solve the resulting system of equations for $ m $ and $ b $

Recall:  
When taking a partial derivative, all variables except the one being differentiated are treated as constants.

---

## Partial Derivative with Respect to $ m $

- Terms without $ m $ are constants → derivative is 0
- Terms involving $ m $ contribute according to standard differentiation rules

After computing and simplifying, we set the result equal to zero.

---

## Partial Derivative with Respect to $ b $

- Terms without $ b $ are constants → derivative is 0
- Linear and quadratic terms in $ b $ contribute their coefficients

This derivative is also set equal to zero.

---

## Simplifying the System

Both resulting equations contain a common factor of $ 2n $. Dividing each equation by $ 2n $ simplifies the system without changing the solution:

### Equation (1)
$$
- \overline{xy} + m\,\overline{x^2} + b\,\overline{x} = 0
$$

### Equation (2)
$$
- \overline{y} + m\,\overline{x} + b = 0
$$

Here, the overline notation (e.g. $ \overline{x} $) denotes the **mean** of the corresponding values.

---

## Rewriting the Equations

### From Equation (1)
Add $ \overline{xy} $ to both sides:
$$
m\,\overline{x^2} + b\,\overline{x} = \overline{xy}
$$

### From Equation (2)
Add $ \overline{y} $ to both sides:
$$
m\,\overline{x} + b = \overline{y}
$$

---

## Key Insight: A Point on the Best-Fit Line

The second equation has the familiar form:
$$
y = mx + b
$$

This tells us something important:

> **The best-fitting line always passes through the point**
$$
(\overline{x}, \overline{y})
$$

That is, the mean of the $ x $-values and the mean of the $ y $-values lies on the optimal regression line.

This result is both mathematically and intuitively satisfying.

---

## Finding Another Point on the Line

To gain further insight, divide Equation (1) by $ \overline{x} $:

$$
m \frac{\overline{x^2}}{\overline{x}} + b
= \frac{\overline{xy}}{\overline{x}}
$$

This reveals another point on the best-fitting line:

- **x-coordinate:** $ \displaystyle \frac{\overline{x^2}}{\overline{x}} $
- **y-coordinate:** $ \displaystyle \frac{\overline{xy}}{\overline{x}} $

---

## Interpretation

We now know **two points** that lie on the optimal line (under squared error minimization):

1. $ (\overline{x}, \overline{y}) $
2. $ \left( \frac{\overline{x^2}}{\overline{x}}, \frac{\overline{xy}}{\overline{x}} \right) $

Since a line is uniquely determined by two points, this gives us a powerful geometric and algebraic understanding of linear regression.

---

## What’s Next

In the next step, we can:

- Solve the system directly for $ m $ and $ b $, **or**
- Use these two points to derive the equation of the best-fitting line

Either approach leads to the same result, and both offer valuable intuition about why linear regression works the way it does.
