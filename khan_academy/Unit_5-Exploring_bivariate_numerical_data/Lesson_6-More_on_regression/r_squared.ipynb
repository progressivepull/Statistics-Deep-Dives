# Computing $ R^2 $: Measuring the Goodness of Fit

## Objective

In the previous step, we found the **least-squares regression line** for four data points.  
Now, we want to answer a key question:

> **How good is this line as a fit to the data?**

We measure this using **$ R^2 $** (R-squared), which tells us:

- The **percentage of variation in $ y $**  
- That is **explained by variation in $ x $** through the regression line

---

## Reminder: The Regression Line

From the previous result, the regression line is:

$$
y = \frac{41}{42}x - \frac{5}{21}
$$

This line predicts a value $ \hat{y} $ (sometimes written as $ y^* $) for each $ x $.

---

## Conceptual Breakdown of $ R^2 $

To compute $ R^2 $, we compare two quantities:

### 1. Total Variation in $ y $

This measures how much the $ y $-values vary **around their mean**:

$$
\text{Total Variation} = \sum (y - \overline{y})^2
$$

This represents the error if we **only used the mean of $ y $** as a predictor.

---

### 2. Unexplained Variation (Residual Error)

This measures how far the data points are from the **regression line**:

$$
\text{Unexplained Variation} = \sum (y - \hat{y})^2
$$

This is the **total squared error** of the regression line.

---

## Organizing the Computation (Spreadsheet Logic)

To compute these values cleanly, we organize the data into columns:

| Column | Description |
|------|-------------|
| $ x $ | Original x-values |
| $ y $ | Original y-values |
| $ \hat{y} $ | Predicted y-values from the regression line |
| $ (y - \hat{y})^2 $ | Squared error from the line |
| $ (y - \overline{y})^2 $ | Squared deviation from the mean |

---

## Given Data Points

$$
(-2, -3), \quad (-1, -1), \quad (1, 2), \quad (4, 3)
$$

The mean of the $ y $-values is:

$$
\overline{y} = \frac{1}{4}
$$

---

## Key Numerical Results

After computing all values and summing:

### Total Squared Error (Unexplained Variation)

$$
\sum (y - \hat{y})^2 \approx 2.74
$$

This represents how much variation the **line fails to explain**.

---

### Total Squared Variation from the Mean

$$
\sum (y - \overline{y})^2 = 22.75
$$

This represents the **entire variation** in the $ y $-values.

---

## Computing $ R^2 $

First, compute the fraction **not explained** by the line:

$$
\frac{2.74}{22.75} \approx 0.12
$$

This means **12% of the variation** in $ y $ is **not explained** by the regression line.

Now subtract from 1:

$$
R^2 = 1 - 0.12 = 0.88
$$

---

## Final Result

$$
\boxed{R^2 = 0.88}
$$

---

## Interpretation

- **88% of the variation in $ y $** is explained by variation in $ x $
- Only **12% remains unexplained**
- Since $ R^2 $ is close to 1, this indicates a **strong linear fit**

Visually, each data point lies:
- Much closer to the **regression line**
- Than to the horizontal line at $ y = \overline{y} $

---

## Key Takeaway

- $ R^2 $ quantifies **how useful the regression line is**
- A higher $ R^2 $ means:
  - Better explanatory power
  - Stronger linear relationship
- In this case, the regression line provides a **very good fit** to the data


# Compute $R^2$ for the Regression Line (From Scratch)

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Step 1: Define the Data

```python
# Data points
x = np.array([-2, -1, 1, 4])
y = np.array([-3, -1, 2, 3])
```

### Step 2: Define the Regression Line

From the previous derivation:

$$
y = \frac{41}{42}x - \frac{5}{21}
$$

```python
m = 41 / 42
b = -5 / 21
```

### Step 3: Predicted Values ( $\hat{y}$ )

```python
y_hat = m * x + b
y_hat
```

### Step 4: Compute the Mean of $y$

```python
y_mean = np.mean(y)
y_mean
```

### Step 5: Compute Squared Errors
Squared Error from the Regression Line (Unexplained Variation)

$$
\sum (y - \hat{y})^2
$$


```python
squared_error_line = (y - y_hat) ** 2
SSE = np.sum(squared_error_line)
SSE
```

###  Squared Deviation from the Mean (Total Variation)

$$
\sum (y - \bar{y})^2
$$

```python
squared_error_mean = (y - y_mean) ** 2
SST = np.sum(squared_error_mean)
SST
```

### Step 6: Compute $R^2$

$$
R^2 = 1 - \frac{\text{SSE}}{\text{SST}}
$$



```python
r_squared = 1 - SSE / SST
r_squared
```

###  Final Result

```python
print(f"R^2 = {r_squared:.4f}")
```

### Step 7: Visualize the Fit

```python
x_line = np.linspace(min(x) - 1, max(x) + 1, 200)
y_line = m * x_line + b

plt.figure()
plt.scatter(x, y)
plt.plot(x_line, y_line)
plt.axhline(y=y_mean, linestyle='--')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Regression Line and Mean of y")
plt.show()
```

# Interpretation

* **SST** → total variation in $y$
* **SSE** → variation not explained by the line $R^2=0.88$ means:
    - **88% of the variation in $y$** is explained by the regression line
    - The linear fit is strong
