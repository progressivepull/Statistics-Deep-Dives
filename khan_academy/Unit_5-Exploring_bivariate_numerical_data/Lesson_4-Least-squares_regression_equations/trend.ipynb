# Study Time vs. Test Score: Interpreting the Slope

## Context
Shira's math test included a survey asking students how many hours they studied.  
She plotted a **scatter graph** showing:
- **X-axis**: Hours spent studying  
- **Y-axis**: Test scores  

Each blue dot represents a student’s study time and corresponding test score.  
Shira drew a **trend line** to model the relationship.

---

## Observations
- A student who studied **~0.6 hours** scored **≈43–44**.  
- A student who studied **~4.5 hours** scored **≈94–95**.  
- The trend line shows a **positive linear relationship** between study time and test score.

---

## Interpreting the Slope
The line’s **slope = 15**.  



$$
\text{Slope} = \frac{\Delta \text{Score}}{\Delta \text{Hours}} = 15
$$



This means:
- For every **additional hour studied**, the model predicts an **average increase of 15 points** on the test.  
- It reflects a **general trend**, not a guarantee.  
- Some students score **above or below** the trend line due to individual variation.

---

## Evaluating Statements
- "Students who didn’t study got scores of about 15." → ❌ Incorrect.  
- "Studying one hour more guarantees exactly 15 more points." → ❌ Incorrect (trend ≠ certainty).  
- "Studying one extra hour was associated with a 15-point improvement." → ✅ Correct interpretation.  
- "Studying 15 extra hours leads to a 1-point improvement." → ❌ Incorrect (inverse slope).

---

## Key Takeaway
The **slope of 15** represents the **average rate of score improvement per hour studied**, according to the trend line.  
It helps visualize how study time correlates with performance, but **individual results may vary**.


```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data: study time in hours and corresponding test scores
study_time = np.array([0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5])
test_scores = np.array([40, 55, 60, 65, 70, 75, 80, 85, 95])

# Fit a linear regression line
slope, intercept = np.polyfit(study_time, test_scores, 1)
trend_line = slope * study_time + intercept

# Plotting
plt.figure(figsize=(8, 6))
plt.scatter(study_time, test_scores, color='blue', label='Student Data')
plt.plot(study_time, trend_line, color='deeppink', label=f'Trend Line: y = {slope:.1f}x + {intercept:.1f}')
plt.title('Study Time vs. Test Score')
plt.xlabel('Study Time (hours)')
plt.ylabel('Test Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

# Sample data
study_time = np.array([0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5])
test_scores = np.array([40, 55, 60, 65, 70, 75, 80, 85, 95])

# Fit a line: y = slope * x + intercept
slope, intercept = np.polyfit(study_time, test_scores, 1)

print(f"Slope: {slope:.2f}")
print(f"Intercept: {intercept:.2f}")
```


* **Slope (15)** → On average, each extra hour of studying is associated with a 15‑point increase in test score.

* **Intercept (≈32.5)** → If a student studied 0 hours, the model predicts a score of about 32.5 (though extrapolation at 0 hours is less reliable).
