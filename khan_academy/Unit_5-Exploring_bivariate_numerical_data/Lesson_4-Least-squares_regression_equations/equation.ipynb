# From Correlation to the Least Squares Regression Line

In previous work, we calculated the **correlation coefficient** for a bivariate dataset and found:

$$
r \approx 0.946
$$

This value indicates a **strong positive linear correlation** between the variables.  
In this section, we build on that result to **derive the equation of the least squares regression line** and understand *why* the formula works.

---

## Review: Correlation Coefficient

Recall:
- The correlation coefficient $ r $ is the **average of the products of paired Z-scores**
- Its value tells us about the **direction and strength** of a linear relationship:
  - $ r = 1 $: perfect positive correlation
  - $ r = -1 $: perfect negative correlation
  - $ r = 0 $: no linear correlation

For our dataset, $ r = 0.946 $, which means a **very strong positive linear relationship**.

---

## Visualizing the Data and Statistics

We begin by plotting:
- The **four data points**
- The **sample mean of x**:
  $$
  \bar{x} = \frac{1 + 2 + 2 + 3}{4} = 2
  $$
- One **sample standard deviation above and below** the mean of x
- The **sample mean of y**:
  $$
  \bar{y} = 3
  $$
- One **sample standard deviation above and below** the mean of y

These reference lines help visualize how spread in x relates to spread in y and how the regression line should behave.

---

## The Regression Line Form

Any line can be written as:

$$
\hat{y} = mx + b
$$

Where:
- $ m $ is the **slope**
- $ b $ is the **y-intercept**
- The “hat” indicates this is a **predicted value** from a regression model

---

## Formula for the Slope of the Least Squares Line

For the **least squares regression line**, the slope is:

$$
m = r \cdot \frac{s_y}{s_x}
$$

Where:
- $ r $ = correlation coefficient
- $ s_y $ = sample standard deviation of y
- $ s_x $ = sample standard deviation of x

This formula has intuitive meaning:
- $ \frac{s_y}{s_x} $ reflects the **relative spread** of y compared to x
- $ r $ scales this spread based on how strong the linear relationship is

### Special Cases
- If $ r = 1 $: slope = $ \frac{s_y}{s_x} $
- If $ r = -1 $: slope = $ -\frac{s_y}{s_x} $
- If $ r = 0 $: slope = 0 (horizontal line at $ \bar{y} $)

---

## Intercept of the Regression Line

A key property of the least squares regression line:

> **It always passes through the point**  
> $$
> (\bar{x}, \bar{y})
> $$

Once the slope is known, the intercept is found by substituting the means:

$$
\bar{y} = m\bar{x} + b
$$

---

## Computing the Regression Line for This Dataset

### Given:
- $ r = 0.946 $
- $ s_y = 2.160 $
- $ s_x = 0.816 $

### Step 1: Compute the slope

$$
m = 0.946 \cdot \frac{2.160}{0.816}
\approx 2.50
$$

### Step 2: Compute the intercept

Using $ (\bar{x}, \bar{y}) = (2, 3) $:

$$
3 = 2.50(2) + b
$$

$$
b = -2
$$

---

## Final Regression Equation

$$
\boxed{
\hat{y} = 2.50x - 2
}
$$

---

## Interpretation

- The line passes through the mean point $ (2, 3) $
- The slope reflects:
  - Strong positive correlation
  - Greater spread in y than in x
- Because $ r $ is close to 1, the line is close to the “perfect correlation” line but not exactly on it

---

## Key Takeaways

- The regression slope depends on **both correlation and variability**
- Least squares regression is not arbitrary—it is tied directly to:
  - Z-scores
  - Standard deviations
  - The correlation coefficient
- The regression line:
  - Minimizes the sum of squared residuals
  - Always passes through $ (\bar{x}, \bar{y}) $

This completes the transition from **correlation** to **regression** with both mathematical structure and intuitive meaning.


```python
# From correlation to the least squares regression line
# This example reproduces the logic and calculations step by step.

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# Original bivariate data
# -----------------------------------
x = np.array([1, 2, 2, 3])
y = np.array([1, 2, 3, 6])

# -----------------------------------
# Compute sample statistics
# -----------------------------------
x_mean = np.mean(x)
y_mean = np.mean(y)

sx = np.std(x, ddof=1)  # sample standard deviation
sy = np.std(y, ddof=1)

r = np.corrcoef(x, y)[0, 1]

print(f"x̄ = {x_mean}")
print(f"ȳ = {y_mean}")
print(f"s_x = {sx:.3f}")
print(f"s_y = {sy:.3f}")
print(f"r = {r:.3f}")

# -----------------------------------
# Least squares regression formulas
# -----------------------------------
m = r * (sy / sx)              # slope
b = y_mean - m * x_mean        # intercept

print("\nLeast squares regression line:")
print(f"ŷ = {m:.2f}x + {b:.2f}")

# -----------------------------------
# Verify using NumPy's built-in fit
# -----------------------------------
m_np, b_np = np.polyfit(x, y, 1)

print("\nNumPy polyfit result:")
print(f"ŷ = {m_np:.2f}x + {b_np:.2f}")

# -----------------------------------
# Plot data, means, and regression line
# -----------------------------------
x_line = np.linspace(0.5, 3.5, 100)
y_line = m * x_line + b

plt.figure(figsize=(6, 5))
plt.scatter(x, y, label="Data points")
plt.plot(x_line, y_line, label="Least Squares Line")

# Plot means
plt.axvline(x_mean, linestyle="--", label="x mean")
plt.axhline(y_mean, linestyle="--", label="y mean")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Least Squares Regression from Correlation")
plt.legend()
plt.grid(True)
plt.show()
```

# What this code demonstrates

* How the correlation coefficient connects directly to regression

* How the slope depends on:

* correlation (r)

* relative spread (s_y / s_x)

* Why the regression line always passes through $ (\bar{x}, \bar{y}) $
  
* That the manually derived line matches NumPy’s least-squares solution
