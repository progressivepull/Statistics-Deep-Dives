# Least Squares Regression and Residuals: Bicycle Frame Sizes

This example explores how **least squares regression** can be used to model and predict relationships between two variables, and how to compute a **residual** for a specific data point.

---

## The Scenario

Vera rents bicycles to tourists. For each customer, she recorded:

- **Customer height** (in centimeters)
- **Bicycle frame size** rented (in centimeters)

After plotting the data, Vera observed that the relationship between height and frame size was **fairly linear**. Based on this, she computed a **least squares regression equation** to predict bicycle frame size from customer height.

---

## Interpreting the Regression Model

The regression equation is:

$$
\hat{y} = \frac{1}{3} + \frac{1}{3}x
$$

Where:
- $ x $ = customer height (cm)
- $ \hat{y} $ = predicted bicycle frame size (cm)

This line represents the **best-fitting linear model** that minimizes the **sum of the squared residuals**—the standard goal of least squares regression.

---

## Conceptual View of the Data

- Height is plotted on the **x-axis**
- Frame size is plotted on the **y-axis**
- Each point represents one customer
- The regression line models the overall trend:
  - Taller customers generally rent larger frames

This line can be used to **predict frame size** for new customers based on their height.

---

## Understanding Residuals

A **residual** measures how far an actual data point is from the regression line.

$$
\text{Residual} = \text{Actual value} - \text{Predicted value}
$$

- **Positive residual:** actual value is above the line
- **Negative residual:** actual value is below the line

Residuals help us understand how well the model fits individual data points.

---

## Calculating the Residual for a Specific Customer

### Given:
- Customer height: **155 cm**
- Actual frame size rented: **51 cm**

### Step 1: Compute the Predicted Frame Size

Using the regression equation:

$$
\hat{y} = \frac{1}{3} + \frac{1}{3}(155)
= \frac{156}{3}
= 52
$$

### Step 2: Compute the Residual

$$
\text{Residual} = 51 - 52 = -1
$$

---

## Interpreting the Result

- The residual is **−1 cm**
- This means the customer rented a frame that was **1 cm smaller** than the model predicted
- The data point lies **slightly below** the regression line

---

## Key Takeaways

- Least squares regression provides a line that best fits the overall data
- Residuals measure individual deviations from that line
- Negative residuals indicate actual values below the predicted values
- Residuals are central to evaluating model accuracy and fit

This example shows how regression models are used not only for prediction, but also for understanding how individual observations compare to expected outcomes.


```python
# Least squares regression and residual calculation
# Bicycle frame size (cm) predicted from customer height (cm)

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# Regression model given in the problem
# y_hat = (1/3) + (1/3)x
# -----------------------------------
def predict_frame_size(height_cm):
    return (1/3) + (1/3) * height_cm

# -----------------------------------
# Given customer data
# -----------------------------------
height = 155        # cm
actual_frame = 51  # cm

# -----------------------------------
# Compute prediction and residual
# -----------------------------------
predicted_frame = predict_frame_size(height)
residual = actual_frame - predicted_frame

print(f"Customer height: {height} cm")
print(f"Actual frame size: {actual_frame} cm")
print(f"Predicted frame size: {predicted_frame:.1f} cm")
print(f"Residual (actual - predicted): {residual:.1f} cm")

# -----------------------------------
# Visualize the regression line and point
# -----------------------------------
x_vals = np.linspace(120, 180, 100)
y_vals = predict_frame_size(x_vals)

plt.figure(figsize=(7, 5))
plt.plot(x_vals, y_vals, label="Regression Line: y = (1/3)x + 1/3")
plt.scatter(height, actual_frame, label="Observed Data Point")

# Draw residual
plt.plot([height, height],
         [predicted_frame, actual_frame],
         linestyle="--",
         label="Residual")

plt.xlabel("Customer Height (cm)")
plt.ylabel("Bicycle Frame Size (cm)")
plt.title("Least Squares Regression and Residual")
plt.legend()
plt.grid(True)
plt.show()
```

# What this code demonstrates

* How to **use a regression equation** for prediction

* How to compute a **residual**

* Why a **negative residual** means the actual value is below the regression line

* How residuals are represented visually as **vertical distances**
