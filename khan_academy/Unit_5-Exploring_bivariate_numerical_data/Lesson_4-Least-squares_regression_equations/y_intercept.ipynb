# Coaches' Salaries vs. Winning Percentages (2000–2011)

## Overview
Adrianna collected data on:
- **Winning percentages** of different schools  
- **Average yearly salary** of their head coaches (in millions of dollars)  

She plotted the data as a **scatter plot** with a **trend line**.

---

## Observations
- A coach earning **over $4 million** had a winning percentage **above 80%**.  
- Another coach earning **about $1.5 million** had a winning percentage **above 85%**.  
- Each data point represents a coach, showing their salary vs. winning percentage.  
- The trend line suggests a **positive linear correlation**, but:
  - The correlation is **not very tight**.  
  - There are **outliers** far from the model.  
  - Coaches with **lower salaries** show wide variation (≈20% to ≈60% winning).

---

## Interpreting the Y‑Intercept
The trend line’s **y‑intercept = 39**.  



$$
\text{Winning Percentage at Salary = 0} \approx 39\%
$$



This means:
- If a coach earned **$0**, the model predicts they would win **39% of their games**.  
- This is **unrealistic**, since coaches are always paid something.  
- It highlights why models should be used cautiously, especially at extreme values.

---

## Evaluating Choices
- "The average salary was 39 million dollars." → ❌ Incorrect.  
- "Each million dollar increase in salary was associated with a 39% increase in winning percentage." → ❌ Incorrect (that describes slope, not intercept).  
- "The average winning percentage was 39%." → ❌ Incorrect.  
- "Teams with coaches who had a salary of $0 will average a winning percentage of ≈39%." → ✅ Correct interpretation of the intercept.

---

## Key Takeaway
The **y‑intercept of 39** reflects the model’s prediction at a salary of zero, not an actual observed value.  
It underscores the importance of being **skeptical of models**, especially when extrapolating beyond realistic ranges.


```python
# Scatter plot and trend line for coach salary vs. winning percentage

import matplotlib.pyplot as plt
import numpy as np

# Sample data: (salary in millions, winning percentage)
salaries = np.array([1.5, 2.0, 2.5, 3.0, 3.5, 4.0])
winning_percentages = np.array([85, 60, 65, 70, 75, 82])

# Fit a linear regression line
coefficients = np.polyfit(salaries, winning_percentages, 1)
trend_line = np.poly1d(coefficients)

# Generate x values for the trend line
x_vals = np.linspace(min(salaries), max(salaries), 100)
y_vals = trend_line(x_vals)

# Plotting
plt.figure(figsize=(8, 6))
plt.scatter(salaries, winning_percentages, color='blue', label='Coach Data')
plt.plot(x_vals, y_vals, color='red', label=f'Trend Line: y = {coefficients[0]:.2f}x + {coefficients[1]:.2f}')
plt.title('Coach Salary vs. Winning Percentage')
plt.xlabel('Salary (Millions of Dollars)')
plt.ylabel('Winning Percentage')
plt.legend()
plt.grid(True)
plt.show()
```
