# Height vs. Weight: Linear Regression and Residuals

This example introduces the **concept of linear regression** using a real-world scenario: the relationship between a person’s **height (in inches)** and **weight (in pounds)**. The focus is on **intuition**, not formal calculation.

---

## Collecting and Plotting the Data

- A random sample of people is taken
- For each person:
  - Height is measured (in inches)
  - Weight is measured (in pounds)
- Each person is represented by a **point** on a scatter plot

### Axes
- **X-axis:** Height (in inches)
- **Y-axis:** Weight (in pounds)

Example:
- A person who is **60 inches tall** and weighs **100 pounds**
- Represented by the point **(60, 100)**

After plotting several people (nine in this example), a pattern begins to emerge.

---

## Observing the Relationship

From the scatter plot:
- The relationship appears **roughly linear**
- It is **positive**:
  - As height increases, weight generally increases
- The points do not lie perfectly on a single line

This suggests that a **linear model** may reasonably describe the trend.

---

## Drawing a Regression Line (Eyeballing)

By visually inspecting the data:
- Some candidate lines fit poorly:
  - Too many points above the line
  - Too many points below the line
- A better line balances the points above and below

This visually chosen line can be thought of as a **regression line**, written as:

$$
y = mx + b
$$

Where:
- $ x $ = height
- $ y $ = weight
- $ m $ = slope (rate of change of weight with respect to height)
- $ b $ = y-intercept (predicted weight when height is 0, not meaningful here but required mathematically)

---

## Residuals: Measuring Error

Because the line cannot pass through all points, there is error.

### Definition of a Residual
A **residual** is:

$$
\text{residual} = \text{actual value} - \text{predicted value}
$$

---

### Example Residual

For the point **(60, 100)**:
- Predicted weight from the model ≈ **150 pounds**
- Actual weight = **100 pounds**

$$
\text{residual} = 100 - 150 = -50
$$

- This is a **negative residual**
- Negative residual → actual value is **below** the predicted value

---

### Positive vs. Negative Residuals

- **Positive residual:**  
  Actual value is **greater** than predicted
- **Negative residual:**  
  Actual value is **less** than predicted

Residuals tell us **how well the model fits each individual data point**.

---

## Measuring Overall Fit

To judge how well a line fits *all* the data, we must combine residuals.

### Why Not Just Add Residuals?
- Positive and negative residuals can cancel out
- A poor fit could misleadingly appear good

---

## Least Squares Regression

To avoid cancellation:
- Square each residual
- Add them together

This gives the **sum of squared residuals**.

### Why Squaring Matters
- Squaring removes negative signs
- Larger residuals are penalized more heavily
- Outliers have greater influence

This leads to **least squares regression**:

> The regression line is the one that **minimizes the sum of the squares of the residuals**.

---

## Key Takeaways

- Linear regression models relationships with a straight line
- Residuals measure how far predictions are from actual values
- A good model balances residuals above and below the line
- Least squares regression is the standard method because:
  - It handles positive/negative errors
  - It emphasizes large deviations
- This example builds intuition before formal calculation

---

```python
# Height vs. Weight: Linear regression and residuals
# This example demonstrates residuals and least squares regression conceptually.

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# Sample height (inches) and weight (pounds) data
# -----------------------------------
height = np.array([60, 62, 64, 65, 67, 68, 70, 72, 74])
weight = np.array([100, 120, 135, 140, 155, 165, 180, 190, 210])

# -----------------------------------
# Fit a least-squares regression line
# y = mx + b
# -----------------------------------
m, b = np.polyfit(height, weight, 1)

print(f"Regression line:")
print(f"Weight = {m:.2f} * Height + {b:.2f}")

# -----------------------------------
# Predicted values and residuals
# -----------------------------------
predicted_weight = m * height + b
residuals = weight - predicted_weight

print("\nResiduals (actual - predicted):")
for h, w, r in zip(height, weight, residuals):
    print(f"Height {h} in: residual = {r:.1f} lbs")

# -----------------------------------
# Sum of squared residuals
# -----------------------------------
ssr = np.sum(residuals ** 2)
print(f"\nSum of squared residuals: {ssr:.2f}")

# -----------------------------------
# Plot data, regression line, and residuals
# -----------------------------------
x_line = np.linspace(min(height) - 1, max(height) + 1, 100)
y_line = m * x_line + b

plt.figure(figsize=(7, 5))
plt.scatter(height, weight, label="Observed Data")
plt.plot(x_line, y_line, label="Regression Line")

# Draw residuals as vertical lines
for h, w, p in zip(height, weight, predicted_weight):
    plt.plot([h, h], [p, w], linestyle="--")

plt.xlabel("Height (inches)")
plt.ylabel("Weight (pounds)")
plt.title("Height vs. Weight with Regression Line and Residuals")
plt.legend()
plt.grid(True)
plt.show()
```

# What this code demonstrates

* How to **fit a regression line** using least squares

* How to compute **predicted values**

* How to calculate **residuals** for each data point

* Why residuals can be **positive or negative**

* How the **sum of squared residuals** measures overall fit
