# Interpreting the Slope of a Regression Line

This example focuses on **interpreting the slope** of a linear regression model relating **hours spent studying** to **test scores**.

---

## The Scenario

Liz’s math test included a survey asking students how many hours they spent studying.  
The results were plotted on a scatter plot, with:

- **X-axis:** Hours spent studying  
- **Y-axis:** Test score  

A **trend line** was fitted to the data, and the slope of this line is **15**.

---

## Understanding the Scatter Plot

From the plotted data:

- A student who studied **between 0.5 and 1 hour** scored just under **45**
- A student who studied **just under 2 hours** scored a little over **60**
- A student who studied **over 4 hours** scored around **94–95**

The scatter plot shows a **clear positive linear trend**:
- As study time increases, test score increases

The fitted line appears to represent this trend well.

---

## What Does a Slope of 15 Mean?

The **slope** of a regression line represents:

$$
\text{Change in score} \div \text{Change in study time}
$$

A slope of **15** means:

> **For each additional hour of studying, the model predicts an average increase of 15 points in test score.**

Visually:
- Moving **1 unit to the right** on the x-axis (1 more hour)
- Corresponds to moving **15 units up** on the y-axis (15 more points)

This matches what we see in the graph—for example, scores rising from about 45 to about 60 as study time increases by roughly one hour.

---

## Evaluating Possible Interpretations

### ❌ Incorrect Interpretations

- *“The model predicts that a student who scored zero studied for 15 hours.”*  
  → This confuses the roles of variables and does not describe slope.

- *“The model predicts that students who did not study will score 15 points.”*  
  → This would relate to the **y-intercept**, not the slope.  
  The graph suggests a score closer to the high 30s or low 40s with zero study time.

- *“The model predicts study time increases by 15 hours for each additional point scored.”*  
  → This reverses the dependent and independent variables.

---

### ✅ Correct Interpretation

- **“The model predicts the score will increase by 15 points for each additional hour of study time.”**

This statement correctly describes the meaning of the slope.

---

## Key Takeaways

- The **slope** of a regression line tells us how the dependent variable changes when the independent variable increases by one unit
- Here:
  - Independent variable → hours studied
  - Dependent variable → test score
- A slope of 15 means **15 points per hour**
- Always interpret slope in context and with correct variable direction

Understanding slope interpretation is essential for making meaningful conclusions from linear models.


```python
# Interpreting the slope of a regression line: study time vs. test score

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# Approximate data from the scatter plot
# -----------------------------------
hours_studied = np.array([0.6, 1.8, 2.5, 3.2, 4.1])
test_scores = np.array([44, 62, 70, 82, 95])

# -----------------------------------
# Fit a linear regression line
# -----------------------------------
slope, intercept = np.polyfit(hours_studied, test_scores, 1)

print(f"Regression equation:")
print(f"Score = {slope:.1f} * (hours studied) + {intercept:.1f}")

# -----------------------------------
# Interpret the slope
# -----------------------------------
print("\nInterpretation:")
print(
    f"For each additional hour of studying, "
    f"the model predicts an average increase of {slope:.0f} points."
)

# -----------------------------------
# Visualize the data and regression line
# -----------------------------------
x_line = np.linspace(0, 5, 100)
y_line = slope * x_line + intercept

plt.figure(figsize=(6, 4))
plt.scatter(hours_studied, test_scores, label="Student Data")
plt.plot(x_line, y_line, label="Best-Fit Line")

plt.xlabel("Hours Studied")
plt.ylabel("Test Score")
plt.title("Study Time vs. Test Score")
plt.legend()
plt.grid(True)
plt.show()
```

# What this code demonstrates

* How to **fit a linear regression line** from data

* How to compute and extract the **slope**

* How slope directly answers the question:

    - How much does the score change for each additional hour of studying?

* Why a slope of about **15** means **~15 points per hour**
