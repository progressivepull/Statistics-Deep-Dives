# Calculating the Correlation Coefficient by Hand

This lesson walks through **computing the correlation coefficient (r) by hand** for a small set of **bivariate data**.  
*Bivariate* simply means that **each X value has a corresponding Y value**.

The purpose is not efficiency (software usually does this), but to build **intuution** for what correlation is measuring and *why* the formula works.

---

## Given Data and Summary Statistics

We assume the data are **samples** from a larger population.

### X Data
- Values: 1, 2, 2, 3
- Sample mean:  
  $$
  \bar{x} = \frac{1 + 2 + 2 + 3}{4} = 2
  $$
- Sample standard deviation:
  $$
  s_x = \sqrt{\frac{(1-2)^2 + (2-2)^2 + (2-2)^2 + (3-2)^2}{3}}
       = \sqrt{\frac{2}{3}} \approx 0.816
  $$

### Y Data
- Values: 1, 2, 3, 6
- Sample mean:  
  $$
  \bar{y} = \frac{1 + 2 + 3 + 6}{4} = 3
  $$
- Sample standard deviation:  
  $$
  s_y \approx 2.160
  $$

---

## Correlation Coefficient Formula

The sample correlation coefficient is defined as:

$$
r = \frac{1}{n-1} \sum_{i=1}^{n}
\left( \frac{x_i - \bar{x}}{s_x} \right)
\left( \frac{y_i - \bar{y}}{s_y} \right)
$$

Key ideas:
- Each term is a **product of Z-scores**
- Z-scores measure how many **standard deviations** a point is from the mean
- The sum averages how often X and Y move together

---

## Computing Z-Score Products

We compute Z-scores for each pair and multiply them:

| (x, y) | Zₓ | Zᵧ | Product |
|------|----|----|---------|
| (1, 1) | negative | negative | positive |
| (2, 2) | 0 | negative | 0 |
| (2, 3) | 0 | 0 | 0 |
| (3, 6) | positive | positive | positive |

Observations:
- When **both Z-scores have the same sign**, the product is positive
- When one Z-score is zero, the product contributes nothing
- Opposite signs would produce a **negative contribution**

---

## Final Calculation

After simplifying:

$$
r = \frac{1}{3} \cdot \frac{5}{0.816 \times 2.160}
\approx 0.946
$$

---

## Interpretation of the Result

- **r ≈ 0.946**, which is **close to 1**
- This indicates a **strong positive linear relationship**
- A straight line does a very good (but not perfect) job describing the data

### What the Value of r Means
- **r = 1** → perfect positive linear relationship  
- **r = -1** → perfect negative linear relationship  
- **r = 0** → no linear relationship  

Since our value is close to 1, the relationship between X and Y is **strongly linear and positive**.

---

## Geometric and Intuitive Insight

- The **least-squares regression line always passes through**:
  $$
  (\bar{x}, \bar{y}) = (2, 3)
  $$
- Points where X and Y are both:
  - above their means → increase r
  - below their means → increase r
- Points where one is above and the other below:
  - decrease r

### Why Z-Score Products Work
- Same-sign Z-scores → positive contribution
- Opposite-sign Z-scores → negative contribution
- Larger magnitudes → stronger influence

This explains how the formula naturally captures:
- **direction** (sign of r)
- **strength** (magnitude of r)

---

## Key Takeaways

- Correlation measures **how well a line describes the relationship**
- It is based on **standardized distances from the mean**
- Products of Z-scores encode whether variables move together or apart
- Computing it by hand builds intuition for later regression concepts

This foundation will be essential for understanding **least-squares regression** and predictive modeling.


```python
# Calculating the correlation coefficient by hand (step-by-step)
# This example mirrors the conceptual walkthrough from the lesson.

import numpy as np

# -----------------------------
# Given bivariate sample data
# -----------------------------
x = np.array([1, 2, 2, 3])
y = np.array([1, 2, 3, 6])

n = len(x)

# -----------------------------
# Step 1: Compute sample means
# -----------------------------
x_mean = np.mean(x)
y_mean = np.mean(y)

print(f"x̄ (mean of x) = {x_mean}")
print(f"ȳ (mean of y) = {y_mean}")

# ---------------------------------------
# Step 2: Compute sample standard deviations
# (ddof=1 gives sample standard deviation)
# ---------------------------------------
x_std = np.std(x, ddof=1)
y_std = np.std(y, ddof=1)

print(f"s_x (std of x) = {x_std:.3f}")
print(f"s_y (std of y) = {y_std:.3f}")

# -----------------------------
# Step 3: Compute Z-scores
# -----------------------------
z_x = (x - x_mean) / x_std
z_y = (y - y_mean) / y_std

print("\nZ-scores:")
for i in range(n):
    print(f"Point {i+1}: Zx = {z_x[i]:.3f}, Zy = {z_y[i]:.3f}")

# -----------------------------------------
# Step 4: Multiply corresponding Z-scores
# -----------------------------------------
z_products = z_x * z_y

print("\nProducts of corresponding Z-scores:")
for i in range(n):
    print(f"Point {i+1}: Zx * Zy = {z_products[i]:.3f}")

# -----------------------------------------
# Step 5: Compute correlation coefficient
# -----------------------------------------
r_manual = (1 / (n - 1)) * np.sum(z_products)

print(f"\nCorrelation coefficient (by hand) r = {r_manual:.3f}")

# -----------------------------------------
# Step 6: Verify using NumPy's built-in method
# -----------------------------------------
r_numpy = np.corrcoef(x, y)[0, 1]

print(f"Correlation coefficient (NumPy)   r = {r_numpy:.3f}")
```
