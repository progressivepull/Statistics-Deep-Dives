# Correlation Coefficient Intuition — Summary

This exercise is based on screen captures from a Khan Academy activity focused on **building intuition for correlation coefficients**. The task is to match given correlation values to different scatterplots by visually assessing how well a **linear model** describes the relationship between two variables.

The goal is **not** to compute correlation coefficients yet, but to understand **what correlation is trying to measure**.

---

## What Is a Correlation Coefficient?

A correlation coefficient (denoted by **r**) measures **how well a linear model describes the relationship between two variables**.

- It ranges from **-1 to 1**
- The sign indicates **direction**
- The magnitude indicates **strength**

---

## Key Reference Cases

### Perfect Positive Correlation (r = 1)

- As **x increases, y increases**
- All points lie exactly on a straight line with a positive slope
- A linear model fits the data perfectly

**Interpretation:**  
When one variable gets larger, the other also gets larger (and vice versa).

---

### Perfect Negative Correlation (r = -1)

- As **x increases, y decreases**
- All points lie exactly on a straight line with a negative slope
- A linear model fits the data perfectly

**Interpretation:**  
The variables move in opposite directions, but in a perfectly linear way.

---

### No Linear Correlation (r ≈ 0)

- Points are scattered with **no clear linear trend**
- Many different lines could reasonably be drawn
- A linear model does **not** describe the data well

**Interpretation:**  
Knowing one variable gives little to no information about the other (linearly).

---

## Matching Correlations to Scatterplots

### Scatterplot A

- Clear **negative trend**
- When one variable is high, the other tends to be low
- Pattern is strong and visually apparent

**Best match:**  
**r = -0.72**

> This choice is based on intuition, not exact calculation. The relationship is clearly negative and fairly strong.

---

### Scatterplot B

- Clear **positive trend**
- Points cluster reasonably well around a line
- A linear model fits better here than in some other plots

**Best match:**  
**r = 0.84**

> Among the positive correlations, this scatterplot shows the strongest linear relationship.

---

### Scatterplot C

- Points are scattered with **no obvious trend**
- No clear direction (positive or negative)
- A line fits very poorly

**Best match:**  
**r = -0.02**

> This value is very close to zero, which matches the lack of linear structure.

---

### Scatterplot D

- Some **positive trend**, but weaker than Scatterplot B
- Several points lie far from the best-fit line
- Linear model fits, but not especially well

**Best match:**  
**r = 0.65**

> The relationship is positive, but less tightly clustered than in Scatterplot B.

---

## Final Matches Summary

---

## Key Takeaway

Correlation coefficients capture **how well a straight line can describe the relationship between two variables**:

- Stronger correlation → points closer to a line
- Sign indicates direction
- Real-world data is rarely perfect, so intuition and comparison matter

This visual intuition lays the groundwork for later learning how correlation is **calculated mathematically**.


```python
# Correlation coefficient intuition — Python example
#
# This example shows how different datasets produce different
# correlation coefficients and scatterplots.

import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# ---- Create datasets ----

# Perfect positive correlation (r = 1)
x_pos = np.arange(0, 10)
y_pos = x_pos

# Strong negative correlation (approximately r = -0.7)
x_neg = np.linspace(0, 10, 50)
y_neg = -x_neg + np.random.normal(0, 2, size=len(x_neg))

# Near-zero correlation (r ≈ 0)
x_zero = np.random.uniform(0, 10, 50)
y_zero = np.random.uniform(0, 10, 50)

# Moderate positive correlation (approximately r = 0.6–0.7)
x_mod = np.linspace(0, 10, 50)
y_mod = 0.6 * x_mod + np.random.normal(0, 3, size=len(x_mod))

# ---- Compute correlations ----
def corr(x, y):
    return np.corrcoef(x, y)[0, 1]

print("Correlation coefficients:")
print(f"Perfect positive: {corr(x_pos, y_pos):.2f}")
print(f"Strong negative:  {corr(x_neg, y_neg):.2f}")
print(f"Near zero:       {corr(x_zero, y_zero):.2f}")
print(f"Moderate positive:{corr(x_mod, y_mod):.2f}")

# ---- Plot scatterplots ----
datasets = [
    (x_pos, y_pos, "Perfect Positive Correlation"),
    (x_neg, y_neg, "Strong Negative Correlation"),
    (x_zero, y_zero, "Near-Zero Correlation"),
    (x_mod, y_mod, "Moderate Positive Correlation"),
]

plt.figure(figsize=(10, 8))

for i, (x, y, title) in enumerate(datasets, start=1):
    plt.subplot(2, 2, i)
    plt.scatter(x, y)
    plt.title(f"{title}\nr = {corr(x, y):.2f}")
    plt.xlabel("x")
    plt.ylabel("y")

plt.tight_layout()
plt.show()
```

# What this demonstrates

* **Correlation measures linear fit**, not just “pattern”
* Strong correlations cluster tightly around a line
* Near-zero correlation looks scattered with no clear direction
* Real data includes noise, so correlations are rarely exactly ±1
