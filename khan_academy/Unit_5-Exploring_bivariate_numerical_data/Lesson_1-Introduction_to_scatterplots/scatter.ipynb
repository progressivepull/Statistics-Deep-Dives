# Summary: Identifying the Appropriate Plot

The percentage of adults who smoke, recorded every few years since 1967, suggests a **negative linear association with no outliers**. On average, the percentage of smokers **decreases by about 0.5 percentage points per year**.

## Step-by-Step Reasoning

- A **negative linear association** means that as the years increase, the percentage of smokers decreases.
- A **positive correlation** can be ruled out immediately, so **Graph 4** is eliminated.
- The description specifies **no outliers**.  
  - Although **Graph 3** shows a general downward trend, it contains a noticeable outlier around 12–13 years after 1967 (approximately 1980).  
  - Because of this outlier, **Graph 3** is also ruled out.

## Comparing the Remaining Options

This leaves **Graph 1** and **Graph 2**.

### Graph 1
- Starts around **55%** in 1967.
- Drops to just under **45%** after 10 years.
- This represents a decline of about **10 percentage points in 10 years**, or roughly **1% per year**.
- This rate of decrease is **too steep** compared to the given average of 0.5% per year.

### Graph 2
- Starts around **42%**.
- Drops to about **37%** after 10 years.
- This is a decline of **5 percentage points in 10 years**, which equals **0.5% per year**.
- This matches the given information and shows a smooth negative linear trend with no outliers.

## Conclusion

**Graph 2** best fits the description:  
- Negative linear association  
- No outliers  
- Average decrease of **0.5 percentage points per year**

✅ **Correct choice: Graph 2**


```python
import matplotlib.pyplot as plt
```

```python
# --------------------
# Graph 1
# --------------------
x1 = [5, 10, 15, 20, 23, 26, 30, 33, 36, 38]
y1 = [52, 45, 40, 33, 30, 27, 18, 14, 10, 8]

plt.figure()
plt.scatter(x1, y1)
plt.xlabel("Years since 1967")
plt.ylabel("% of smokers")
plt.title("Graph 1")
plt.show()


```

```python
# --------------------
# Graph 2
# --------------------
x2 = [5, 10, 15, 18, 21, 24, 28, 32, 35, 38]
y2 = [40, 37, 35, 33, 30, 27, 25, 23, 21, 20]

plt.figure()
plt.scatter(x2, y2)
plt.xlabel("Years since 1967")
plt.ylabel("% of smokers")
plt.title("Graph 2")
plt.show()


```
