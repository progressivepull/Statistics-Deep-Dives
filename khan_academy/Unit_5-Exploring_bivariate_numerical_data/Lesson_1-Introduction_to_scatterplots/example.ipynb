# Summary: Relationships Between Test Grades, Study Time, and Shoe Size

The graphs show test grades for students in Dexter’s class and compare grades with two different variables: **study time** and **shoe size**.

## Graph 1: Study Time vs. Test Grade
- The horizontal axis represents **study time**.
- The vertical axis represents **test grade**.
- The points show a **positive linear relationship**.
- As study time increases, test grades generally increase.
- While individual results vary, a line could reasonably be drawn through the data to represent this upward trend.
- This supports the idea that spending more time studying tends to improve test performance.

## Graph 2: Shoe Size vs. Test Grade
- The horizontal axis represents **shoe size**.
- The vertical axis represents **test grade**.
- The data points are scattered with **no clear pattern or trend**.
- For the same shoe size, some students score high while others score low.
- It is difficult, if not impossible, to fit a meaningful line to this data.

## Conclusion
- There is a **positive linear relationship** between **study time and test score**.
- There is **no apparent relationship** between **shoe size and test score**.

**Best description:**  
There is a positive linear relationship between study time and score, and no relationship between shoe size and score.


```python
import matplotlib.pyplot as plt

# Sample data approximating the graphs
study_time = [1, 1.5, 2, 2.5, 3, 3, 3.5, 4, 2, 3.5]
scores_study = [55, 60, 65, 70, 75, 80, 85, 80, 68, 78]

shoe_size = [6, 7, 8, 9, 10, 7, 8.5, 9.5, 6.5, 8]
scores_shoe = [60, 70, 75, 80, 65, 78, 82, 85, 62, 77]

# First scatter plot: Study Time vs Score
plt.figure()
plt.scatter(study_time, scores_study)
plt.xlabel("Study Time (hours)")
plt.ylabel("Score")
plt.title("Test Score vs Study Time")
plt.show()

# Second scatter plot: Shoe Size vs Score
plt.figure()
plt.scatter(shoe_size, scores_shoe)
plt.xlabel("Shoe Size")
plt.ylabel("Score")
plt.title("Test Score vs Shoe Size")
plt.show()
```

# Interpretation

* **Study time vs. score** shows a **positive correlation**: as study time increases, scores generally increase.
* **Shoe size vs. score** shows **no clear relationship**: scores are scattered randomly.
