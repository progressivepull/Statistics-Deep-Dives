# Exam Period vs. Average Score — Summary

Aubrey wanted to determine whether there is a relationship between the time of day an exam takes place and the average score on that exam. To investigate this, she collected data from exams given during the previous year.

The data include three key pieces of information:
- **Class**
- **Period of the day** when the exam occurred
- **Average exam score**

To analyze the data, a **scatter plot** was created.

## Scatter Plot Setup
- The **horizontal axis (x-axis)** represents the **Period**.  
  This is the independent variable because Aubrey is investigating whether the time of day influences performance.
- The **vertical axis (y-axis)** represents the **Average Score**, which is the dependent variable.

Each class is plotted as a single point using its corresponding period and average score.

## Observations
After plotting all the data points, there does not appear to be a clear or consistent trend between exam period and average score. Scores vary across periods, and no strong positive or negative relationship is evident based on this dataset alone.

## Conclusion
Based on Aubrey’s data, there is **no obvious connection** between the time an exam is given and the average score. While other factors (such as subject difficulty or student preparedness) may influence scores, the scatter plot suggests that exam period by itself does not strongly predict performance.


```python
import matplotlib.pyplot as plt

# Data from the table
classes = [
    "English", "Math", "History", "Computer Science",
    "Arts", "Biology", "Physics", "Chemistry", "Spanish"
]

period = [1, 6, 2, 4, 4, 1, 3, 1, 3]
average_score = [93, 87, 70, 62, 86, 73, 73, 80, 96]

# Create scatter plot
plt.scatter(period, average_score)

# Label points (optional but helpful)
for i, cls in enumerate(classes):
    plt.annotate(cls, (period[i], average_score[i]), fontsize=9, xytext=(5,5), textcoords="offset points")

# Labels and title
plt.xlabel("Class Period")
plt.ylabel("Average Exam Score")
plt.title("Exam Period vs. Average Score")

plt.show()
```
