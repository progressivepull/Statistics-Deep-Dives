## Summary: The Handshake Problem (Combinations)

**Scenario:**  
There are 4 people in a room: A, B, C, and D.  
Each person must shake hands **exactly once** with every other person.

The question is: **How many handshakes occur in total?**

---

## Why This Is Not a Permutations Problem

A handshake always involves **two people**.  
So each handshake is really just a **pair** of people.

If we try to count directly:
- Choose the first person: 4 choices  
- Choose the second person: 3 choices  

This gives:
$$
4 \times 3 = 12
$$

However, this **overcounts**.

- A shaking hands with B  
- B shaking hands with A  

These describe the **same handshake**, not two different ones.

So counting this way counts **each handshake twice**.

---

## Correcting for Double Counting

Since each pair of people can be arranged in:
$$
2! = 2
$$
ways, we divide by 2:

$$
\frac{4 \times 3}{2} = 6
$$

---

## Combinations Interpretation

Each handshake is simply:
> Choosing **2 people** from a group of **4**

This is a combinations problem:

$$
\binom{4}{2}
$$

Using the formula:
$$
\binom{n}{k} = \frac{n!}{k!(n-k)!}
$$

Apply it here:
$$
\binom{4}{2} = \frac{4!}{2! \cdot 2!} = 6
$$

---

## Listing the Handshakes

The unique handshakes are:
- A–B  
- A–C  
- A–D  
- B–C  
- B–D  
- C–D  

Total:
$$
\boxed{6 \text{ handshakes}}
$$

---

## Key Takeaways

- Handshakes involve **unordered pairs**
- Permutations overcount because order doesn’t matter
- Divide by $2!$ to correct for double counting
- The problem is exactly **“4 choose 2”**
- **Combinations** are the right tool when order doesn’t matter



```python
import math
import itertools

# People in the room
people = ['A', 'B', 'C', 'D']

# --------------------------------
# Method 1: Using combinations
# --------------------------------
# Each handshake is a unique pair of people
handshakes = list(itertools.combinations(people, 2))

print("Handshakes (combinations):")
for h in handshakes:
    print(h)

print("Total handshakes:", len(handshakes))


# --------------------------------
# Method 2: Using the formula n choose 2
# --------------------------------
n = len(people)

# C(n, 2) = n! / (2! * (n - 2)!)
handshakes_formula = math.factorial(n) // (
    math.factorial(2) * math.factorial(n - 2)
)

print("\nTotal handshakes using formula:", handshakes_formula)


# --------------------------------
# Method 3: Show double counting with permutations
# --------------------------------
permutations = list(itertools.permutations(people, 2))

print("\nPermutations (double-counted handshakes):")
for p in permutations:
    print(p)

print("Total permutations:", len(permutations))
print("Corrected by dividing by 2:", len(permutations) // 2)
```
