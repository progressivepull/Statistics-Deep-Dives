## Summary: Permutations vs. Combinations (6 People, 3 Chairs)

We start with **six people**: A, B, C, D, E, and F, and **three distinct chairs** (Chair 1, Chair 2, Chair 3).

---

### Permutations: Order Matters
A **permutation** counts arrangements where **who sits in which chair matters**.

- Chair 1: 6 choices  
- Chair 2: 5 remaining choices  
- Chair 3: 4 remaining choices  

$$
6 \times 5 \times 4 = 120
$$

So, there are **120 permutations** of seating six people into three chairs.

Even if the same three people are used, **different seating orders count as different permutations**.  
For example:
- (A, B, C)
- (A, C, B)
- (B, A, C)
- (B, C, A)
- (C, A, B)
- (C, B, A)

These are **6 different permutations** of the same three people.

---

### Combinations: Order Does Not Matter
A **combination** counts selections where **only the group matters**, not the order or chairs.

- Each group of 3 people can be arranged in **6 different ways**
- These 6 permutations all represent **the same combination**

To find combinations:
$$
\text{Combinations} = \frac{\text{Permutations}}{\text{Arrangements of 3 people}}
$$
$$
\frac{120}{6} = 20
$$

So, there are **20 combinations** of choosing 3 people from 6.

---

### Key Takeaway
- **Permutations (120):** Who sits where matters  
- **Combinations (20):** Only which people are chosen matters  

In short:
- *Arranging* 3 of 6 people → **120 permutations**
- *Choosing* 3 of 6 people → **20 combinations**


```python
import itertools
import math

# Define the people
people = ["A", "B", "C", "D", "E", "F"]

# -------------------------
# Permutations: order matters
# -------------------------
permutations = list(itertools.permutations(people, 3))
print("Number of permutations:", len(permutations))  # 120

# Example permutations
print("Sample permutations:")
for p in permutations[:6]:
    print(p)

# -------------------------
# Combinations: order does NOT matter
# -------------------------
combinations = list(itertools.combinations(people, 3))
print("\nNumber of combinations:", len(combinations))  # 20

# Example combinations
print("Sample combinations:")
for c in combinations[:6]:
    print(c)

# -------------------------
# Manual calculation check
# -------------------------
# Permutations: 6 * 5 * 4
perm_count = math.perm(6, 3)
print("\nPermutations using formula:", perm_count)

# Combinations: permutations / 3!
comb_count = perm_count // math.factorial(3)
print("Combinations using formula:", comb_count)
```
