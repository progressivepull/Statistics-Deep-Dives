A card game using 36 unique cards: four suits (diamonds, hearts, clubs, and spades) with cards numbered from 1 to 9 in each suit. A hand is a collection of 9 cards, which can be sorted however the player chooses. What is the probability of getting all four of the 1s?

## Summary: Probability of Getting All Four 1’s in a 9-Card Hand

### Problem Setup
- There are **36 unique cards**:
  - 4 suits (diamonds, hearts, clubs, spades)
  - Cards numbered **1 through 9** in each suit
- A **hand consists of 9 cards**
- **Order does not matter**, so we use combinations
- We want the probability that a 9-card hand contains **all four of the 1’s**

---

### Key Idea: Probability as a Ratio
$$
\text{Probability} = \frac{\text{Number of hands with all four 1’s}}{\text{Total number of possible hands}}
$$

---

### Total Number of Possible Hands
- We are choosing 9 cards from 36, where order does not matter:
$$
\binom{36}{9}
$$

---

### Number of Hands Containing All Four 1’s
- The four 1’s **must** be included
- That leaves **5 additional cards** to choose
- After removing the four 1’s, there are **32 remaining cards**
- We choose any 5 of those:
$$
\binom{32}{5}
$$

---

### Probability Calculation
$$
\frac{\binom{32}{5}}{\binom{36}{9}}
$$

After simplifying the expression algebraically, this reduces to:
$$
\frac{2}{935}
$$

---

### Final Answer
- **Exact probability:**  
  $$
  \boxed{\frac{2}{935}}
  $$

- **Interpretation:**  
  This is approximately a **1 in 500 chance** of getting all four 1’s in a 9-card hand drawn from the 36-card deck.

---

### Takeaway
Even though the deck is relatively small, requiring **all four specific cards** makes the event quite rare. The key strategy is fixing the required cards first and then counting the remaining choices using combinations.


```python
import math
from fractions import Fraction

def probability_all_four_ones():
    """
    Probability of getting all four 1s in a 9-card hand
    drawn from a 36-card deck (order does not matter).
    """
    # Total number of possible 9-card hands
    total_hands = math.comb(36, 9)
    
    # Number of hands that contain all four 1s
    # Choose remaining 5 cards from the other 32 cards
    favorable_hands = math.comb(32, 5)
    
    probability = favorable_hands / total_hands
    probability_fraction = Fraction(favorable_hands, total_hands)
    
    return probability, probability_fraction


prob_decimal, prob_fraction = probability_all_four_ones()

print(f"Probability (fraction): {prob_fraction}")
print(f"Probability (decimal): {prob_decimal:.6f}")
print(f"Approximate odds: 1 in {1/prob_decimal:.0f}")
```
