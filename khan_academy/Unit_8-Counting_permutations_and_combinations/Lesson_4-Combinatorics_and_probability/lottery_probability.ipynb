To win a particular lottery game, a player chooses 4 numbers from 1 to 60. Each number can only be chosen once. If all 4 numbers match the 4 winning numbers, regardless of order, the player wins. What is the probability that the winning numbers are 3, 15, 46, and 49?

## Summary: Probability of a Specific Lottery Outcome

In this lottery game:
- A player selects **4 distinct numbers** from **1 to 60**
- **Order does not matter**
- The player wins only if **all 4 numbers match** the winning numbers exactly

We are asked for the probability that the winning numbers are **3, 15, 46, and 49**.

---

### Step 1: Total Possible Outcomes
Because order does not matter, this is a **combinations** problem.

The total number of possible sets of 4 numbers chosen from 60 is:
$$
\binom{60}{4}
$$

Using the combinations formula:
$$
\binom{60}{4} = \frac{60!}{4!\,56!}
$$

This simplifies to:
$$
\frac{60 \times 59 \times 58 \times 57}{4 \times 3 \times 2 \times 1}
$$

After simplifying and calculating:
$$
\binom{60}{4} = 487,\!635
$$

So, there are **487,635 equally likely combinations**.

---

### Step 2: Favorable Outcomes
- The set **{3, 15, 46, 49}** is **one specific combination**
- There is **exactly one favorable outcome**

---

### Step 3: Probability Calculation
$$
\text{Probability} = \frac{1}{487,\!635}
$$

---

### Final Answer
- **Probability of winning with numbers 3, 15, 46, and 49**:
$$
\boxed{\frac{1}{487,\!635}}
$$

---

### Key Takeaways
- Lottery problems almost always involve **combinations**, not permutations
- When order does not matter, divide by factorials to avoid overcounting
- Every specific 4-number set has the **same extremely small probability**
- This explains why lottery wins are so rare


```python
import math
from fractions import Fraction

def lottery_probability(n_numbers=60, picks=4):
    """
    Probability of winning a lottery by matching
    a specific set of numbers (order does not matter).
    """
    total_combinations = math.comb(n_numbers, picks)
    probability = 1 / total_combinations
    return probability, Fraction(1, total_combinations)


# Given lottery parameters
prob_decimal, prob_fraction = lottery_probability(60, 4)

print(f"Total possible combinations: {math.comb(60, 4)}")
print(f"Probability (fraction): {prob_fraction}")
print(f"Probability (decimal): {prob_decimal:.10f}")
print(f"Odds: 1 in {math.comb(60, 4)}")
```
