A club of 9 people wants to choose a board of three officers: President, Vice President, and Secretary. Assuming the officers are chosen at random, what is the probability that the officers are Marsha for President, Sabita for Vice President, and Robert for Secretary?

## Summary: Probability of a Specific Board of Officers

A club with **9 members** is choosing **three distinct officer positions**:  
- President  
- Vice President  
- Secretary  

All selections are made **at random**, and each position is filled by a different person.

---

### Step 1: Understand the Event
The event of interest is **one specific outcome**:
- Marcia is President  
- Sabita is Vice President  
- Robert is Secretary  

This exact arrangement counts as **one single outcome**.

---

### Step 2: Count Total Possible Outcomes
Because the roles are **distinct**, order matters. We count permutations:

- **President**: 9 choices  
- **Vice President**: 8 remaining choices  
- **Secretary**: 7 remaining choices  

$$
\text{Total outcomes} = 9 \times 8 \times 7 = 504
$$

---

### Step 3: Compute the Probability
Since all outcomes are equally likely and only **one** outcome matches the given scenario:

$$
\text{Probability} = \frac{1}{504}
$$

---

### Final Answer
- **Probability = $ \frac{1}{504} $**

This represents the chance that Marcia, Sabita, and Robert are selected for those exact roles in that exact order.


```python
import math

def probability_specific_officers(total_people):
    """
    Probability of selecting one specific ordered assignment
    (President, Vice President, Secretary) from total_people.
    """
    total_outcomes = total_people * (total_people - 1) * (total_people - 2)
    return 1 / total_outcomes


# Given
total_people = 9

probability = probability_specific_officers(total_people)

print(f"Total possible outcomes: {total_people * 8 * 7}")
print(f"Probability: {probability}")
print(f"Probability as fraction: 1/{total_people * 8 * 7}")
print(f"Probability as decimal: {probability:.6f}")
```
