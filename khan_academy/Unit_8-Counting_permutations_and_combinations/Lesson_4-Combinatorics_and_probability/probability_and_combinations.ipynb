## Summary: Binomial Probability with an 80% Free-Throw Shooter

This lesson extends probability concepts from fair coins to **unequal probabilities**, using a basketball free-throw example.

### Problem Setup
- A player makes a free throw with probability **0.8 (80%)**
- The probability of a miss is **0.2 (20%)**
- The player takes **5 independent free throws**
- We want to find probabilities for:
  - **Exactly 3 made shots**
  - **At least 3 made shots**

---

### Probability of Exactly 3 Made Shots (Out of 5)

1. **Probability of one specific sequence** with 3 makes and 2 misses:
   $$
   0.8^3 \times 0.2^2
   $$

2. **Number of such sequences**:
   - Choose which 3 of the 5 shots are makes
   $$
   \binom{5}{3} = 10
   $$

3. **Total probability**:
   $$
   10 \times 0.8^3 \times 0.2^2 = 0.2048
   $$

✅ **Result:**  
**20.48% chance** of making exactly 3 out of 5 free throws.

---

### Probability of Exactly 4 Made Shots

1. Probability of one sequence:
   $$
   0.8^4 \times 0.2
   $$

2. Number of sequences:
   $$
   \binom{5}{4} = 5
   $$

3. Total probability:
   $$
   5 \times 0.8^4 \times 0.2 = 0.4096
   $$

✅ **Result:**  
**40.96% chance** of making exactly 4 out of 5.

---

### Probability of Exactly 5 Made Shots

- Only one possible sequence:
  $$
  0.8^5 = 0.32768
  $$

✅ **Result:**  
**32.77% chance** of making all 5 shots.

---

### Probability of At Least 3 Made Shots

“At least 3” means **3, 4, or 5 makes**:

$$
0.2048 + 0.4096 + 0.32768 = 0.94208
$$

✅ **Final Answer:**  
**≈ 94.21% chance** of making at least 3 out of 5 free throws.

---

### Key Takeaways
- This is a classic **binomial probability** problem.
- Each outcome probability is:
  $$
  \binom{n}{k} p^k (1 - p)^{n-k}
  $$
- To find “at least” probabilities, **sum the relevant exact cases**.
- With a high success rate (80%), getting at least 3 successes out of 5 is very likely.


```python
import math

def binomial_probability(n, k, p):
    """
    Probability of exactly k successes in n trials
    with success probability p.
    """
    combinations = math.comb(n, k)
    return combinations * (p ** k) * ((1 - p) ** (n - k))


def probability_at_least_k(n, k_min, p):
    """
    Probability of getting at least k_min successes
    in n trials with success probability p.
    """
    return sum(binomial_probability(n, k, p) for k in range(k_min, n + 1))


# Given values
n_shots = 5
p_make = 0.8

# Exact probabilities
prob_3 = binomial_probability(n_shots, 3, p_make)
prob_4 = binomial_probability(n_shots, 4, p_make)
prob_5 = binomial_probability(n_shots, 5, p_make)

# At least 3 makes
prob_at_least_3 = probability_at_least_k(n_shots, 3, p_make)

# Display results
print(f"Exactly 3 makes: {prob_3:.5f} ({prob_3*100:.2f}%)")
print(f"Exactly 4 makes: {prob_4:.5f} ({prob_4*100:.2f}%)")
print(f"Exactly 5 makes: {prob_5:.5f} ({prob_5*100:.2f}%)")
print(f"At least 3 makes: {prob_at_least_3:.5f} ({prob_at_least_3*100:.2f}%)")
```
