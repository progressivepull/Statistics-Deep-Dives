## Summary: Probability of Exactly Three Heads in Five Coin Flips

We analyze a **fair coin** flipped **five times** and compute the probability of getting **exactly three heads**.

---

### Step 1: Total Possible Outcomes
- Each flip has 2 outcomes: heads or tails.
- With 5 independent flips:
$$
2^5 = 32
$$
There are **32 equally likely outcomes**.

---

### Step 2: Count Outcomes with Exactly Three Heads

We want to count how many of these outcomes contain **exactly three heads**.

#### Combinatorial Reasoning
- There are 5 flip positions.
- We must choose **3 of the 5 positions** to be heads.
- Although we may label the heads as A, B, and C for counting, they are **indistinguishable**, so we must avoid overcounting.

1. If we temporarily treated the three heads as distinct:
$$
5 \times 4 \times 3 = 60
$$

2. Since the three heads are identical, divide by the number of ways to rearrange three items:
$$
3 \times 2 \times 1 = 6
$$

3. Total distinct outcomes with exactly three heads:
$$
\frac{5 \times 4 \times 3}{3!} = \frac{60}{6} = 10
$$

This matches the binomial coefficient:
$$
\binom{5}{3} = 10
$$

---

### Step 3: Compute the Probability
$$
\text{Probability} = \frac{10}{32} = \frac{5}{16}
$$

---

### Final Answer
- **Probability of exactly three heads in five flips:**  
$$
\boxed{\frac{5}{16}}
$$

---

### Key Takeaways
- Total outcomes for $ n $ coin flips: $ 2^n $
- Outcomes with exactly $ k $ heads: $ \binom{n}{k} $
- Combinations naturally account for identical outcomes and avoid overcounting.
- This method scales efficiently to many flips without listing all possibilities.


```python
import math

def probability_exact_heads(n_flips, k_heads):
    """
    Calculate the probability of getting exactly k_heads
    when flipping a fair coin n_flips times.
    """
    favorable_outcomes = math.comb(n_flips, k_heads)
    total_outcomes = 2 ** n_flips
    return favorable_outcomes / total_outcomes


# Given: 5 coin flips
n = 5
k = 3

probability = probability_exact_heads(n, k)

print(f"Probability of exactly {k} heads in {n} flips: {probability}")
print(f"As a fraction: {math.comb(n, k)}/{2**n}")
print(f"As a percentage: {probability * 100:.2f}%")
```
