# Conditional probability and combinations

## Summary: Using Bayes’ Theorem to Identify a Coin

### Problem Setup
- A bag contains:
  - **5 fair coins** (50% heads, 50% tails)
  - **10 unfair coins** (80% heads, 20% tails)
- One coin is drawn at random.
- The coin is flipped **6 times**, resulting in **4 heads**.
- We want to find:

$$
P(\text{Fair coin} \mid \text{4 heads out of 6})
$$

---

### Step 1: Apply Bayes’ Theorem

Bayes’ Theorem states:

$$
P(A \mid B) = \frac{P(B \mid A)P(A)}{P(B)}
$$

Where:
- $A$: the coin is **fair**
- $B$: getting **4 heads out of 6 flips**

---

### Step 2: Prior Probabilities
- Probability of picking a fair coin:
$$
P(A) = \frac{5}{15} = \frac{1}{3}
$$
- Probability of picking an unfair coin:
$$
\frac{2}{3}
$$

---

### Step 3: Likelihoods

#### Given a Fair Coin
- Probability of exactly 4 heads out of 6:
$$
\binom{6}{4}\left(\frac{1}{2}\right)^6 = \frac{15}{64}
$$

#### Given an Unfair Coin
- Each head has probability 0.8, tail 0.2:
$$
\binom{6}{4}(0.8)^4(0.2)^2 = 15 \times 0.8^4 \times 0.2^2
$$

---

### Step 4: Total Probability of 4 Heads
Using the law of total probability:

$$
P(B) = \frac{1}{3}\cdot\frac{15}{64}
+ \frac{2}{3}\cdot\left(15 \cdot 0.8^4 \cdot 0.2^2\right)
\approx 0.241965
$$

---

### Step 5: Posterior Probability
Applying Bayes’ Theorem:

$$
P(\text{Fair} \mid 4 \text{ heads}) =
\frac{\frac{15}{64} \cdot \frac{1}{3}}{0.241965}
\approx 0.323
$$

---

### Final Answer
- **Probability the coin was fair given 4 heads out of 6:**
$$
\boxed{32.3\%}
$$

---

### Key Insights
- Before flipping, the chance of picking a fair coin was **33.3%**.
- Observing **more heads than tails** slightly favors the unfair coin.
- However, 4 heads out of 6 is **not rare** for a fair coin, so the probability only decreases slightly.
- Bayes’ Theorem updates beliefs by balancing prior probabilities with observed evidence.

This example brings together **combinations**, **binomial probability**, **conditional probability**, and **Bayesian reasoning** in a single problem.


```python
import math

def binomial_probability(n, k, p):
    """
    Probability of exactly k heads in n flips
    with head probability p.
    """
    return math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))


def probability_fair_given_four_heads():
    # Coin counts
    fair_coins = 5
    unfair_coins = 10
    total_coins = fair_coins + unfair_coins

    # Prior probabilities
    p_fair = fair_coins / total_coins
    p_unfair = unfair_coins / total_coins

    # Likelihoods
    p_four_given_fair = binomial_probability(6, 4, 0.5)
    p_four_given_unfair = binomial_probability(6, 4, 0.8)

    # Total probability of 4 heads
    p_four = (
        p_fair * p_four_given_fair +
        p_unfair * p_four_given_unfair
    )

    # Bayes' Theorem
    p_fair_given_four = (p_four_given_fair * p_fair) / p_four

    return {
        "P(4 heads | fair)": p_four_given_fair,
        "P(4 heads | unfair)": p_four_given_unfair,
        "P(4 heads)": p_four,
        "P(fair | 4 heads)": p_fair_given_four
    }


results = probability_fair_given_four_heads()

for key, value in results.items():
    print(f"{key}: {value:.6f}")
```
