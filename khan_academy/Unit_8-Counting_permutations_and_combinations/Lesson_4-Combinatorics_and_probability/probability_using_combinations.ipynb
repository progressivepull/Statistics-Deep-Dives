## Summary: Probability of Exactly 3 Heads in 8 Coin Flips

This lesson explains how **permutations and combinations** are used to calculate probabilities, using the example of flipping a **fair coin 8 times** and finding the probability of getting **exactly 3 heads**.

### Step 1: Total Possible Outcomes
- Each coin flip has 2 possible outcomes: heads or tails.
- For 8 independent flips, the total number of possible outcomes is:
  $$
  2^8 = 256
  $$

### Step 2: Favorable Outcomes (Exactly 3 Heads)
- We only care about outcomes where **exactly 3 of the 8 flips are heads**.
- The order of flips does **not** matter—only which 3 flips are heads.
- This is a **combination problem**: choosing 3 items out of 8.

Using the combination formula:
$$
\binom{8}{3} = \frac{8!}{3! \cdot 5!} = 56
$$

So, there are **56 different ways** to get exactly 3 heads.

### Step 3: Calculate the Probability
$$
\text{Probability} = \frac{\text{Favorable outcomes}}{\text{Total outcomes}} = \frac{56}{256}
$$

Simplifying:
$$
\frac{56}{256} = \frac{7}{32}
$$

### Step 4: Decimal Approximation
$$
\frac{7}{32} = 0.21875 \approx 21.9\%
$$

### Final Answer
- The probability of getting **exactly 3 heads in 8 coin flips** is:
  - **7/32**
  - **≈ 21.9%**

### Key Takeaway
This method generalizes to many problems:
- Any number of coin flips
- Any exact number of desired outcomes  
As long as outcomes are equally likely, combinations help count favorable cases efficiently.


```python
import math

def probability_exact_heads(n_flips, k_heads):
    """
    Returns the probability of getting exactly k_heads heads
    when flipping a fair coin n_flips times.
    """
    # number of favorable outcomes (combinations)
    favorable = math.comb(n_flips, k_heads)
    
    # total possible outcomes
    total = 2 ** n_flips
    
    return favorable / total


# Example: exactly 3 heads out of 8 flips
n = 8
k = 3

prob = probability_exact_heads(n, k)
print(f"Probability: {prob}")
print(f"Percentage: {prob * 100:.2f}%")
```
