## Summary: General Formula for Getting *k* Heads in *n* Coin Flips

This lesson generalizes earlier coin-flip problems to find the probability of getting **exactly *k* heads in *n* flips** of a **fair coin**.

---

### Step 1: Total Number of Possible Outcomes
- Each coin flip has 2 equally likely outcomes.
- With *n* independent flips:
$$
2^n
$$
There are **$2^n$ equally likely outcomes**.

---

### Step 2: Count Outcomes with Exactly *k* Heads

We want to count how many outcomes contain **exactly *k* heads**.

#### Counting the Placements of Heads
- The first head can go in any of the *n* flip positions.
- The second head can go in any of the remaining *n − 1* positions.
- Continue until all *k* heads are placed.

This gives:
$$
n \times (n-1) \times (n-2) \times \dots \times (n-k+1)
$$

#### Correcting for Overcounting
- The *k* heads are **indistinguishable**, so we must divide by the number of ways to rearrange *k* items:
$$
k! = k \times (k-1) \times \dots \times 1
$$

---

### Step 3: Express Using Factorials

The product above can be written compactly using factorials:

$$
\frac{n!}{(n-k)!}
$$

Dividing by $k!$ gives the total number of distinct outcomes with exactly *k* heads:

$$
\binom{n}{k} = \frac{n!}{k!(n-k)!}
$$

This is called the **binomial coefficient**.

---

### Step 4: Probability Formula

Since all outcomes are equally likely:

$$
\text{Probability of exactly } k \text{ heads} =
\frac{\binom{n}{k}}{2^n}
$$

---

### Final Formula

$$
\boxed{
P(\text{exactly } k \text{ heads in } n \text{ flips})
= \frac{n!}{k!(n-k)!} \cdot \frac{1}{2^n}
}
$$

---

### Key Takeaways
- Binomial coefficients count how many ways *k* successes can occur in *n* trials.
- Factorials naturally arise from counting arrangements while avoiding overcounting.
- Reasoning through the placement of heads provides intuition and avoids memorization.
- This framework generalizes easily to **biased coins** and other probability problems.


```python
import math

def probability_k_heads(n, k):
    """
    Probability of getting exactly k heads
    in n flips of a fair coin.
    """
    # Number of ways to choose k heads from n flips
    favorable_outcomes = math.comb(n, k)
    
    # Total number of equally likely outcomes
    total_outcomes = 2 ** n
    
    return favorable_outcomes / total_outcomes


# Example usage
n = 5   # number of flips
k = 3   # number of heads

probability = probability_k_heads(n, k)

print(f"Probability of exactly {k} heads in {n} flips: {probability}")
print(f"As a fraction: {math.comb(n, k)}/{2**n}")
print(f"As a percentage: {probability * 100:.2f}%")
```
