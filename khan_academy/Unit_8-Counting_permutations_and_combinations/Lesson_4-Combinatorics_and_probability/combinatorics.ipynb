## Summary: Probabilities with Four Coin Flips

We consider a **fair coin** flipped **four times** and compute probabilities for getting **exactly one head** and **exactly two heads**.

---

### Total Possible Outcomes
- Each flip has 2 outcomes: heads or tails.
- With 4 independent flips:
$$
2^4 = 16
$$
There are **16 equally likely outcomes**, each with probability **1/16**.

---

## Probability of Exactly One Head

- Exactly one head can occur in **4 distinct ways**:
  - Head on flip 1, tails elsewhere
  - Head on flip 2, tails elsewhere
  - Head on flip 3, tails elsewhere
  - Head on flip 4, tails elsewhere

Each outcome has probability **1/16**, and these outcomes are **mutually exclusive**, so we add them:

$$
\frac{4}{16} = \frac{1}{4}
$$

**Result:**  
- **Probability of exactly one head = 1/4**

---

## Probability of Exactly Two Heads

### Method 1: Counting Outcomes
- Out of the 16 total outcomes, **6** contain exactly two heads.
$$
\frac{6}{16} = \frac{3}{8}
$$

---

### Method 2: Combinatorial Reasoning
- We choose **2 of the 4 flip positions** to be heads.
- Number of ways to choose these positions:
$$
\binom{4}{2} = 6
$$

Each of these 6 outcomes has probability **1/16**, so:

$$
6 \times \frac{1}{16} = \frac{3}{8}
$$

---

### Final Results
- **Exactly 1 head:** $ \frac{1}{4} $
- **Exactly 2 heads:** $ \frac{3}{8} $

---

### Key Takeaways
- When outcomes are equally likely, probability is:
$$
\frac{\text{favorable outcomes}}{\text{total outcomes}}
$$
- Combinations help count outcomes efficiently without listing all possibilities.
- This approach scales to larger numbers of flips where listing outcomes is impractical.


```python
import math

def probability_exact_heads(n_flips, k_heads):
    """
    Returns the probability of getting exactly k_heads
    when flipping a fair coin n_flips times.
    """
    total_outcomes = 2 ** n_flips
    favorable_outcomes = math.comb(n_flips, k_heads)
    return favorable_outcomes / total_outcomes


# Given: 4 coin flips
n = 4

# Exactly 1 head
prob_1_head = probability_exact_heads(n, 1)

# Exactly 2 heads
prob_2_heads = probability_exact_heads(n, 2)

print(f"Probability of exactly 1 head: {prob_1_head} ({prob_1_head:.2%})")
print(f"Probability of exactly 2 heads: {prob_2_heads} ({prob_2_heads:.2%})")
```
