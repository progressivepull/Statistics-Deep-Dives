## Summary: The Birthday Problem (30 People)

This classic probability problem asks:

> **In a room of 30 randomly selected people, what is the probability that at least two people share the same birthday?**

At first glance, this seems complicated because many different scenarios can cause a shared birthday. However, the problem becomes much simpler by using a **complementary probability** approach.

---

### Key Insight: Use the Complement
Instead of directly computing the probability that *at least two people share a birthday*, we compute:

- **Probability that everyone has a different birthday**
- Then subtract this from 1

$$
P(\text{at least one shared birthday}) = 1 - P(\text{all birthdays are different})
$$

---

### Step 1: Probability That All Birthdays Are Different

Assume:
- 365 possible birthdays
- Birthdays are uniformly distributed
- No leap years

Consider people one by one:

- Person 1: any birthday → $ \frac{365}{365} $
- Person 2: different from person 1 → $ \frac{364}{365} $
- Person 3: different from first two → $ \frac{363}{365} $
- ...
- Person 30: different from previous 29 → $ \frac{336}{365} $

Multiply all probabilities:

$$
P(\text{all different}) =
\frac{365 \times 364 \times 363 \times \cdots \times 336}{365^{30}}
$$

---

### Step 2: Factorial Representation

The numerator can be compactly written using factorials:

$$
\frac{365!}{(365 - 30)!}
$$

So the full expression becomes:

$$
P(\text{all different}) =
\frac{365!}{335! \cdot 365^{30}}
$$

---

### Step 3: Numerical Evaluation

Evaluating this expression gives:

$$
P(\text{all birthdays different}) \approx 0.2937
$$

---

### Step 4: Final Probability

$$
P(\text{at least one shared birthday})
= 1 - 0.2937
= 0.7063
$$

---

### Final Answer
- **Probability that at least two people share a birthday in a group of 30**:

$$
\boxed{70.6\%}
$$

---

### Key Takeaways
- Complementary probabilities can greatly simplify complex problems
- Even with only 30 people, shared birthdays are **more likely than not**
- This result is often surprising and is known as the **Birthday Paradox**
- The probability grows rapidly as the group size increases

In short:  
> With 30 people in a room, there is about a **70% chance** that at least two people share the same birthday.


```python
import math

def birthday_problem_probability(n_people, days_in_year=365):
    """
    Returns:
    - Probability that all birthdays are different
    - Probability that at least two people share a birthday
    """
    # Probability all birthdays are different
    prob_all_different = 1.0
    for i in range(n_people):
        prob_all_different *= (days_in_year - i) / days_in_year

    # Complement probability
    prob_shared = 1 - prob_all_different

    return prob_all_different, prob_shared


# Given: 30 people
n = 30

prob_distinct, prob_shared = birthday_problem_probability(n)

print(f"People: {n}")
print(f"Probability all birthdays are different: {prob_distinct:.6f}")
print(f"Probability at least two share a birthday: {prob_shared:.6f}")
print(f"Percentage chance of shared birthday: {prob_shared * 100:.2f}%")
```
