## Summary: Why $0!$ Is Defined as 1

- **Factorial Review**  
  The factorial of a positive integer $n$, written $n!$, is the product of all positive integers from $n$ down to 1:  
  $$
  n! = n \times (n-1) \times (n-2) \times \cdots \times 1
  $$  
  Examples:
  - $3! = 3 \times 2 \times 1 = 6$
  - $2! = 2 \times 1 = 2$
  - $1! = 1$

- **The Question of $0!$**  
  At first glance, one might think $0! = 0$, but factorials are a **human-defined mathematical operation**, and mathematicians choose definitions that make formulas consistent and useful.

- **Why Define $0! = 1$?**  
  Factorials appear most often in **permutations and combinations**.  
  The number of ways to arrange $n$ objects into $k$ positions is:
  $$
  \frac{n!}{(n-k)!}
  $$

- **Consistency with Permutations**  
  If we permute $n$ objects into $n$ positions, logic tells us the answer should be $n!$.  
  Using the formula:
  $$
  \frac{n!}{(n-n)!} = \frac{n!}{0!}
  $$
  For this to equal $n!$, it must be true that:
  $$
  0! = 1
  $$

- **Conclusion**  
  Defining $0! = 1$ ensures:
  - Permutation formulas work correctly in all cases
  - Mathematical definitions remain consistent
  - Factorials behave logically within combinatorics

  Although it may seem unintuitive at first, defining $0! = 1$ is both **practical and essential** in mathematics.


```python
def factorial(n: int) -> int:
    """
    Compute n! (n factorial) for a non-negative integer n.
    By definition, 0! = 1.
    """
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")

    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


def permutations(n: int, k: int) -> int:
    """
    Compute the number of permutations of n items taken k at a time:
    P(n, k) = n! / (n - k)!
    """
    if k < 0 or k > n:
        raise ValueError("k must be between 0 and n")

    return factorial(n) // factorial(n - k)


# Examples
print("Factorials:")
for i in range(6):
    print(f"{i}! = {factorial(i)}")

print("\nPermutations:")
print("P(5, 5) =", permutations(5, 5))  # equals 5!
print("P(5, 3) =", permutations(5, 3))
```
