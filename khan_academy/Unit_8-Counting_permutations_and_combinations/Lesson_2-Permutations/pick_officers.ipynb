A club of 9 people wants to choose a board of three officers: President, Vice President, and Secretary. How many ways are there to choose the board from the 9 people?

## Summary: Choosing Officers from a Club of Nine People

- **The Problem**  
  A club with **9 people** wants to choose **3 officers**:
  - President  
  - Vice President  
  - Secretary  

  Each person can hold **only one office**, and the offices are **distinct**, so order matters.

- **Step-by-Step Reasoning**
  - **President**: 9 possible choices  
  - **Vice President**: After choosing the President, 8 people remain  
  - **Secretary**: After choosing the first two officers, 7 people remain  

- **Counting the Arrangements**
  Since each choice affects the next, we multiply the number of options:
  $$
  9 \times 8 \times 7
  $$

- **Calculation**
  $$
  9 \times 8 = 72,\quad 72 \times 7 = 504
  $$

- **Key Idea**
  This is a **permutation** problem because:
  - Order matters (each office is different)
  - No repetition is allowed

- **Final Answer**  
  There are **504 different ways** to choose the board of officers from the nine people.


```python
def count_boards(total_people: int, positions: int) -> int:
    """
    Count the number of ways to assign distinct officer positions
    when one person can hold only one position.
    """
    if positions > total_people:
        raise ValueError("Number of positions cannot exceed number of people")

    result = 1
    for i in range(positions):
        result *= (total_people - i)
    return result


# Given problem values
people = 9
officer_positions = 3  # President, Vice President, Secretary

# Calculate result
possible_boards = count_boards(people, officer_positions)

print("Number of possible boards:", possible_boards)
```

This code models the reasoning directly:

$$
9×8×7=504
$$
