## Summary: Introduction to Permutations

- **What Are Permutations?**  
  A *permutation* is the number of different ways objects can be **arranged** when order matters.

- **Simple Example (3 People, 3 Seats)**  
  Imagine a sofa with three distinct seats (left, middle, right) and three people: A, B, and C.  
  Each unique seating arrangement counts as a different permutation.

- **Counting by Listing**  
  By systematically placing each person in the first seat and arranging the remaining people, we find:
  - A in seat 1 → 2 possible arrangements
  - B in seat 1 → 2 possible arrangements
  - C in seat 1 → 2 possible arrangements  
  This gives a total of **6 permutations**.

- **Counting Mathematically**  
  Instead of listing every case, we count choices step by step:
  - Seat 1: 3 choices
  - Seat 2: 2 remaining choices
  - Seat 3: 1 remaining choice  
  Multiply the choices:
  $$
  3 \times 2 \times 1 = 6
  $$

- **Factorial Notation**  
  This multiplication is called a **factorial**:
  $$
  3! = 3 \times 2 \times 1 = 6
  $$

- **Larger Example (5 People, 5 Seats)**  
  Using the same logic:
  - Seat 1: 5 choices
  - Seat 2: 4 choices
  - Seat 3: 3 choices
  - Seat 4: 2 choices
  - Seat 5: 1 choice  

  $$
  5! = 5 \times 4 \times 3 \times 2 \times 1 = 120
  $$

- **Key Idea**  
  When arranging $n$ distinct people into $n$ distinct seats, the total number of permutations is:
  $$
  n!
  $$

  Factorials provide a powerful and efficient way to count permutations without listing every possibility.


```python
def factorial(n: int) -> int:
    """
    Compute n! (n factorial) for a non-negative integer n.
    """
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")

    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


def permutations_same_size(n: int) -> int:
    """
    Compute the number of permutations of n people sitting in n distinct seats.
    This is simply n!.
    """
    return factorial(n)


# Examples
print("3 people, 3 seats:")
print("Number of permutations:", permutations_same_size(3))  # 6

print("\n5 people, 5 seats:")
print("Number of permutations:", permutations_same_size(5))  # 120
```
