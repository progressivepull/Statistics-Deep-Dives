In one game, a code made using different colors is created by one player (the codemaker), and the other player (the codebreaker) tries to guess the code. The codemaker gives hints about whether the colors are correct and in the right position.

The possible colors are Blue, Yellow, White, Red, Orange, and Green. How many 4-color codes can be made if the colors cannot be repeated?

## Summary: Counting 4-Color Codes Without Repetition

- **Game Setup**  
  One player (the *codemaker*) creates a color code, and another player (the *codebreaker*) tries to guess it.  
  The codemaker provides hints about whether the guessed colors are correct and in the correct positions.

- **Available Colors**  
  There are **6 distinct colors**:
  - Blue  
  - Yellow  
  - White  
  - Red  
  - Orange  
  - Green  

- **The Question**  
  How many **4-color codes** can be made if:
  - No color can be repeated  
  - The **order matters** (e.g., Blue–Red–Yellow–Green is different from Green–Red–Yellow–Blue)

- **Reasoning Step by Step**
  - First position: 6 possible choices  
  - Second position: 5 remaining choices  
  - Third position: 4 remaining choices  
  - Fourth position: 3 remaining choices  

- **Using Permutations**  
  Since order matters and repetition is not allowed, this is a permutation problem:
  $$
  6 \times 5 \times 4 \times 3
  $$

- **Final Calculation**
  $$
  6 \times 5 = 30,\quad 4 \times 3 = 12,\quad 30 \times 12 = 360
  $$

- **Final Answer**  
  There are **360 different possible 4-color codes**.


```python
def count_color_codes(total_colors: int, code_length: int) -> int:
    """
    Count the number of possible color codes when:
    - order matters
    - colors cannot be repeated
    """
    if code_length > total_colors:
        raise ValueError("Code length cannot exceed the number of available colors")

    result = 1
    for i in range(code_length):
        result *= (total_colors - i)
    return result


# Given problem values
colors = 6        # Blue, Yellow, White, Red, Orange, Green
code_length = 4   # 4-color code

# Calculate result
possible_codes = count_color_codes(colors, code_length)

print("Number of possible 4-color codes:", possible_codes)
```

This code follows the same step-by-step reasoning as the explanation:

$$
6×5×4×3=360.
$$
