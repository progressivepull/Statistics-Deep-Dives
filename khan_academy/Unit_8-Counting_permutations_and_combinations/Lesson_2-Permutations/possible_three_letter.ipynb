## Summary: Counting Three-Letter Words Using the English Alphabet

- **The Alphabet**  
  The English alphabet has **26 letters** (A through Z).

- **Question 1: How Many Possible Three-Letter Words (Repetition Allowed)?**  
  We are not concerned with meaning, pronunciation, or letter repetition.  
  Examples like **ZGT**, **SKJ**, or **HHH** are all allowed.

  - First letter: 26 choices  
  - Second letter: 26 choices  
  - Third letter: 26 choices  

  Total number of possible three-letter words:
  $$
  26 \times 26 \times 26 = 26^3
  $$

- **Key Idea**  
  Because letters **can be repeated**, the number of choices stays the same for each position.

- **Question 2: How Many Three-Letter Words with All Different Letters?**  
  Now, each letter must be unique—no repetition is allowed.

  - First letter: 26 choices  
  - Second letter: 25 remaining choices  
  - Third letter: 24 remaining choices  

  Total number of such words:
  $$
  26 \times 25 \times 24
  $$

- **Connection to Permutations**  
  This situation can be described using permutations:
  $$
  P(26, 3) = \frac{26!}{(26 - 3)!} = \frac{26!}{23!}
  $$
  Which simplifies to:
  $$
  26 \times 25 \times 24
  $$

- **Important Takeaways**
  - Do **not** blindly apply permutation or combination formulas.
  - Always think carefully about whether repetition is allowed.
  - Reasoning step-by-step often provides more clarity than memorizing formulas.

Understanding *why* the counting works is more important than remembering which formula to use.


```python
def three_letter_words_with_repetition():
    """
    Number of possible three-letter 'words' using the English alphabet
    when letter repetition is allowed.
    """
    letters = 26
    return letters ** 3


def three_letter_words_unique():
    """
    Number of possible three-letter 'words' using the English alphabet
    when all letters must be different.
    """
    letters = 26
    return letters * (letters - 1) * (letters - 2)


# Examples
print("Three-letter words (repetition allowed):")
print(three_letter_words_with_repetition())  # 26^3

print("\nThree-letter words (all letters different):")
print(three_letter_words_unique())  # 26 * 25 * 24
```
