# Counting outcomes: flower pots

[Counting pot and flower scenarios | Statistics and probability | 7th grade | Khan Academy | YouTube](https://www.youtube.com/watch?v=B_vAlneziHo&t=1s)

### Summary

At the florist store, there are:

- **4 types of flowers**: roses, tulips, sunflowers, and lilies  
- **3 types of pots**: brown, yellow, and green  

To buy a potted flower, you must choose **one flower** and **one pot**.

**Total possible combinations:**
- Each pot can be paired with any of the 4 flowers
- Number of combinations = $ 3 \times 4 = 12 $

**Examples include:**
- Rose in a brown pot  
- Tulip in a yellow pot  
- Sunflower in a green pot  
- Lily in a yellow pot  

**Final Answer:**
There are **12 different flower-and-pot combinations** that you could walk out of the florist store with.


```python
# Flower and pot combinations

flowers = ["rose", "tulip", "sunflower", "lily"]
pots = ["brown", "yellow", "green"]

# Generate all possible combinations
combinations = [(flower, pot) for pot in pots for flower in flowers]

# Total number of combinations
total_combinations = len(combinations)

# Output results
print("Possible flower and pot combinations:")
for combo in combinations:
    print(f"{combo[0].capitalize()} in a {combo[1]} pot")

print(f"\nTotal combinations: {total_combinations}")
```
