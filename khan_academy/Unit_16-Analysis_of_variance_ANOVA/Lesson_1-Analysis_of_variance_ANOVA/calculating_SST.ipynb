# ANOVA 1: Calculating SST (total sum of squares)

[ANOVA 1: Calculating SST (total sum of squares) | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=EFdlFoHI_0I)

The **total sum of squares (SST)** measures how far each observation
lies from the *grand mean*, squared and summed. It is the foundation of
ANOVA because it later gets partitioned into **within‑group** and
**between‑group** variation.

**🔹 Bullet Summary**

- The **grand mean** is the mean of all observations or equivalently the
  mean of the group means.

- For the dataset Group 1: 3, 2, 1 Group 2: 5, 3, 4 Group 3: 5, 6, 7 the
  grand mean is **4**.

- The **total sum of squares (SST)** is

$$
\text{SST} = \sum(x_{i} - {\overset{ˉ}{x}}_{\text{grand}})^{2}
$$

- Computing each squared deviation and summing gives **SST = 30**.

- If you wanted the variance of all 9 points, you would divide by the
  degrees of freedom $9 - 1 = 8$.

- ANOVA later decomposes this total variation into:

  - **SSW** (within‑group variation)

  - **SSB** (between‑group variation) and

$$
\text{SST} = \text{SSW} + \text{SSB}
$$

**🔹 Explanation of the Key Ideas**

**Grand Mean**

You can compute it two equivalent ways:

1.  **Mean of all 9 values**

$$
\frac{3 + 2 + 1 + 5 + 3 + 4 + 5 + 6 + 7}{9} = 4
$$

2.  **Mean of group means**

    - Group means: 2, 4, 6

    - Mean of these:

$$
\frac{2 + 4 + 6}{3} = 4
$$

**Total Sum of Squares**

For each observation $x_{i}$, compute
$\left( x_{i}-4)^{2} \right.\ $and sum them:

- Group 1: $1^{2} + ( - 2)^{2} + ( - 3)^{2} = 1 + 4 + 9 = 14$

- Group 2: $1^{2} + ( - 1)^{2} + 0^{2} = 1 + 1 + 0 = 2$

- Group 3: $1^{2} + 2^{2} + 3^{2} = 1 + 4 + 9 = 14$

Total:

$$
14 + 2 + 14 = 30
$$

This is the **total variability** in the dataset before splitting it
into within‑ and between‑group components.

**🔹 Python Code Example**

A clean, reproducible calculation of the grand mean and SST:

```python
import numpy as np

# Data grouped
groups = [
    [3, 2, 1],
    [5, 3, 4],
    [5, 6, 7]
]

# Flatten into one list
data = np.concatenate(groups)

# Grand mean
grand_mean = np.mean(data)

# Total Sum of Squares (SST)
sst = np.sum((data - grand_mean)**2)

grand_mean, sst
```
