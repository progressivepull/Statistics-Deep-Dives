# ANOVA 2: Calculating SSW and SSB (total sum of squares within and between)

[ANOVA 2: Calculating SSW and SSB (total sum of squares within and between) | Khan Academy | YouTube](https://www.youtube.com/watch?v=j9ZPMlVHJVs&t=12s)

The **sum of squares within (SSW)** and **sum of squares between (SSB)**
decompose the total variation (SST = 30) into two meaningful parts. The
attached text walks through each calculation step-by-step, showing that
**SSW = 6** and **SSB = 24**, and that their degrees of freedom add up
exactly to the total degrees of freedom. The document states that SSW is
computed by “the distance between each data point and that group’s mean”
and that SSB is computed by “how much of the variation is due to the
variation between the means” .

**🔹 Bullet Summary**

- The dataset has **3 groups**, each with **3 observations**.

- The **grand mean** is 4.

- **Total Sum of Squares (SST)** = 30.

- **Sum of Squares Within (SSW)** measures variation *inside* each
  group:

  - Computed by summing
    $\left( x_{ij}-{\overset{ˉ}{x}}_{j})^{2} \right.\ $for all points.

  - The document shows this equals **6** (“this is going to be equal to
    our sum of squares within… is 6” ).

  - Degrees of freedom: $m(n - 1) = 3(3 - 1) = 6$.

- **Sum of Squares Between (SSB)** measures variation *between* group
  means:

  - Computed by summing
    $n({\overset{ˉ}{x}}_{j} - \overset{ˉ}{x})^{2}$across groups.

  - The document shows this equals **24** (“we get… is equal to 24” ).

  - Degrees of freedom: $m - 1 = 2$.

- **ANOVA identity holds:**

$$
\text{SST} = \text{SSW} + \text{SSB} = 6 + 24 = 30
$$

- Degrees of freedom also add:

$$
6 + 2 = 8 = 9 - 1
$$

**🔹 Explanation**

The attached document emphasizes that total variation can be partitioned
into two components:

**1. Variation Within Groups (SSW)**

This captures how spread out each group is around its own mean. The text
explains that we compute

$$
\left( x-{\overset{ˉ}{x}}_{\text{group}})^{2} \right.\ 
$$

for each observation and sum them. For example, in Group 1, the document
shows: “3 minus the mean here is 2 squared, plus 2 minus 2 squared, plus
1 minus 2 squared” .

Adding all groups gives **SSW = 6**.

**2. Variation Between Groups (SSB)**

This captures how far each group mean is from the grand mean. The
document computes terms like

$$
n({\overset{ˉ}{x}}_{j} - \overset{ˉ}{x})^{2}
$$

and shows that for Group 1, “this is equal to 3 times 4… is equal to 12”
and similarly for Group 3, giving another 12, while Group 2 contributes
0 because its mean equals the grand mean .

Thus **SSB = 24**.

**3. Why This Matters**

ANOVA works because:

- Total variation = within-group variation + between-group variation.

- If **between-group variation is large relative to within-group
  variation**, the group means differ more than random noise would
  explain.

Here, SSB (24) is much larger than SSW (6), suggesting strong
differences between group means.

**🔹 Python Code Example (Reproducing All Calculations)**

```python
import numpy as np

# Data grouped
groups = [
    np.array([3, 2, 1]),
    np.array([5, 3, 4]),
    np.array([5, 6, 7])
]

# Flatten data
data = np.concatenate(groups)

# Grand mean
grand_mean = data.mean()

# SST: total sum of squares
sst = np.sum((data - grand_mean)**2)

# SSW: sum of squares within
ssw = sum(np.sum((g - g.mean())**2) for g in groups)

# SSB: sum of squares between
ssb = sum(len(g) * (g.mean() - grand_mean)**2 for g in groups)

sst, ssw, ssb
```
