# ANOVA 3: Hypothesis test with F-statistic

[ANOVA 3: Hypothesis test with F-statistic | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=Xg8_iSkJpAE&t=4s)

**The key idea is that ANOVA compares *between‑group* variation to
*within‑group* variation using an F‑statistic. A large F means the
groups differ more than random noise would explain, so we reject the
hypothesis that all population means are equal.**

**🔹 Bullet Summary**

- Total variation in the 9 data points: **SST = 30**.

- Variation *within* groups: **SSW = 6**.

- Variation *between* groups: **SSB = 24**.

- Goal: test whether the **true population means** for Food 1, Food 2,
  and Food 3 are all equal.

- Null hypothesis $H_{0}$: $\mu_{1} = \mu_{2} = \mu_{3}$(food has no
  effect).

- Alternative hypothesis $H_{a}$: At least one mean differs (food
  affects performance).

- F‑statistic formula:

$$
F = \frac{SSB/df_{B}}{SSW/df_{W}}
$$

- Degrees of freedom: $df_{B} = 2$, $df_{W} = 6$.

- Computed F:

$$
F = \frac{24/2}{6/6} = \frac{12}{1} = 12
$$

- Critical value at $\alpha = 0.10$, $df_{B} = 2$, $df_{W} = 6$:
  **3.46**.

- Since **12 \> 3.46**, reject $H_{0}$.

- Interpretation: The differences between group means are too large to
  be explained by random chance alone.

- Conclusion: **Food type likely affects test performance.**

**🔹 Explanation in Plain Language**

ANOVA splits total variation into two sources:

1.  **Between‑group variation** — how far the group means are from each
    other.

2.  **Within‑group variation** — how spread out scores are inside each
    group.

If food has *no* effect, then the group means should be close together,
and most variation should come from randomness *within* groups.

But in your data:

- The between‑group variation (24) is **four times larger** than the
  within‑group variation (6).

- After adjusting for degrees of freedom, the F‑statistic becomes
  **12**, which is very large for an F distribution with (2,6) degrees
  of freedom.

A large F means:

“The groups differ more than we would expect from random noise.”

Since 12 is far above the critical value 3.46, the probability of
getting such an extreme F under the null hypothesis is very small.

So we reject the idea that all population means are equal.

This suggests that **the type of food truly influences test
performance**.

**🔹 Python Code Example (Simulating the ANOVA)**

Below is a clean, reproducible Python example using scipy.stats and
manual calculations to mirror the video.

You can use the exact values from the Khan Academy example, which produce:

* Food 1: 5, 7, 6
* Food 2: 8, 9, 7
* Food 3: 13, 14, 15

These three groups give:

* Total SS = 30
* SS Within = 6
* SS Between = 24
* F = 12

```python
import numpy as np
from scipy.stats import f


food1 = np.array([5, 7, 6])
food2 = np.array([8, 9, 7])
food3 = np.array([13, 14, 15])

groups = [food1, food2, food3]

# --- Manual ANOVA calculations ---

# Grand mean
all_data = np.concatenate(groups)
grand_mean = np.mean(all_data)

# SSB: Sum of Squares Between
SSB = sum(len(g) * (np.mean(g) - grand_mean)**2 for g in groups)

# SSW: Sum of Squares Within
SSW = sum(sum((g - np.mean(g))**2) for g in groups)

# Degrees of freedom
df_between = len(groups) - 1
df_within = len(all_data) - len(groups)

# Mean squares
MSB = SSB / df_between
MSW = SSW / df_within

# F statistic
F_stat = MSB / MSW

# p-value (right tail)
p_value = 1 - f.cdf(F_stat, df_between, df_within)

print("SSB:", SSB)
print("SSW:", SSW)
print("F statistic:", F_stat)
print("p-value:", p_value)
```

**What this code does**

- Computes SSB, SSW, MSB, MSW exactly like the video.

- Computes the F‑statistic.

- Computes the p‑value using the F distribution.

- Lets you plug in any three groups of data.
