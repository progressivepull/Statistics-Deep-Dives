A 95 percent confidence interval for the difference in two population
proportions comes from treating the observed difference, 0.051, as one
draw from the sampling distribution of
${\widehat{p}}_{1} - {\widehat{p}}_{2}$, then finding how far that
draw typically falls from the true difference. The width of the interval
is set by 1.96 standard errors, because 95 percent of a normal
distribution lies within 1.96 standard deviations of its mean.

**Bullet summary**

- You observed sample proportions ${\widehat{p}}_{1} = 0.642$(men) and
  ${\widehat{p}}_{2} = 0.591$(women), giving a sample difference of
  0.051.

- The sampling distribution of
  ${\widehat{p}}_{1} - {\widehat{p}}_{2}$is approximately normal
  because both samples are large and the proportions are not near 0 or
  1.

- Its mean is the true difference $p_{1} - p_{2}$, and its standard
  deviation (standard error) is

$$
\sqrt{\frac{p_{1}(1 - p_{1})}{n_{1}} + \frac{p_{2}(1 - p_{2})}{n_{2}}}.
$$

- Because $p_{1}$and $p_{2}$are unknown, you substitute the sample
  proportions.

- The estimated standard error is about 0.022.

- A 95 percent interval uses $1.96 \times \text{SE}$, giving a margin
  of error of about 0.043.

- The resulting confidence interval for $p_{1} - p_{2}$is

$$
0.051 \pm 0.043 = (0.008,\text{\:\,}0.094).
$$

- Since the entire interval is positive, the data suggest men are more
  likely to vote for the candidate.

**Explanation**

The logic of the interval comes from symmetry: if a random sample
difference has a 95 percent chance of falling within $d$of the true
difference, then the true difference has a 95 percent chance of falling
within $d$of the observed sample difference. The value of $d$is
determined by the spread of the sampling distribution. For proportions,
that spread depends on the Bernoulli variance $p(1 - p)$scaled by
sample size. With large samples, the normal approximation is accurate,
so the central 95 percent of the distribution lies between
$\pm 1.96$standard errors.

Replacing the unknown population proportions with their sample estimates
gives a workable estimate of the standard error. Multiplying by 1.96
gives the margin of error. Adding and subtracting that margin from the
observed difference yields the confidence interval.

**Code example**




```python
import math

# Sample sizes and counts
n1, x1 = 1000, 642
n2, x2 = 1000, 591

# Sample proportions
p1_hat = x1 / n1
p2_hat = x2 / n2

# Observed difference
diff_hat = p1_hat - p2_hat

# Estimated standard error
se = math.sqrt(p1_hat*(1 - p1_hat)/n1 + p2_hat*(1 - p2_hat)/n2)

# 95% critical value
z = 1.96

# Margin of error
moe = z * se

# Confidence interval
lower = diff_hat - moe
upper = diff_hat + moe

diff_hat, lower, upper
```

This produces a difference of about 0.051 and a 95 percent confidence
interval of roughly (0.008, 0.094).

If you want to extend this into a hypothesis test for whether the
proportions differ, that builds directly on the same ingredients.
