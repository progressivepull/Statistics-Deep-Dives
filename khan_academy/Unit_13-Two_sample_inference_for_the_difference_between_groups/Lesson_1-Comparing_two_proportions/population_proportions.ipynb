# Hypothesis test comparing population proportions

[Hypothesis test comparing population proportions | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=dvSa_tx04hw&t=4s)

A hypothesis test for the difference in two population proportions asks
whether the observed gap between men and women (0.051) is plausibly
explained by random sampling **if the true proportions were actually
equal**. The test statistic is a Z‑score computed under the assumption
$p_{1} = p_{2}$, and the pooled proportion is used to estimate the
standard error. Because the resulting Z‑score (≈2.35) exceeds the 5
percent two‑tailed critical value (1.96), the data provide evidence
against equality of proportions.

**🔹 Bullet summary**

- The **null hypothesis**
  $\lbrack H_{0}\rbrack(copilot - action://composer - send? = What$:
  $p_{1} = p_{2}$, meaning no difference between men and women.

- The **alternative hypothesis**
  $\lbrack H_{1}\rbrack(copilot - action://composer - send? = What$:
  $p_{1} \neq p_{2}$.

- Observed sample proportions:

  - Men: ${\widehat{p}}_{1} = 0.642$

  - Women: ${\widehat{p}}_{2} = 0.591$

  - Difference: $0.051$.

- Under $H_{0}$, both groups share a common true proportion $p$.

- The **pooled estimate**
  $\lbrack pooled\text{ 					}proportion\rbrack(copilot - action://composer - send?text = Explain$:

$$
\widehat{p} = \frac{642 + 591}{2000} = 0.6165.
$$

- The **standard error** under the null:

$$
SE = \sqrt{\frac{2\widehat{p}(1 - \widehat{p})}{1000}} \approx 0.0217.
$$

- The **Z‑score**:

$$
Z = \frac{0.051}{0.0217} \approx 2.35.
$$

- The **critical value** for a 5 percent two‑tailed test is ±1.96.

- Since $2.35 > 1.96$, the result is statistically significant.

- Conclusion: reject $H_{0}$; evidence suggests a real difference
  between men and women.

**🔹 Explanation**

A two‑proportion hypothesis test compares two independent groups by
asking: *If the true proportions were equal, how likely is it to observe
a difference as large as the one we saw?*

Under the assumption that men and women have the same true voting
proportion, the sampling distribution of
${\widehat{p}}_{1} - {\widehat{p}}_{2}$is centered at 0. Its spread
depends on the pooled proportion, because under the null both samples
estimate the same underlying probability. The pooled estimate uses all
2,000 observations to approximate that shared value.

Once the standard error is computed, the Z‑score measures how many
standard deviations the observed difference (0.051) lies from the null
value (0). A Z of 2.35 falls in the extreme 5 percent of the
distribution, meaning such a difference would be unlikely if the true
proportions were equal. That improbability leads to rejecting the null
hypothesis.

This test complements the earlier confidence interval: both methods
point to the same conclusion that men appear more likely to support the
candidate.

**🔹 Code example (Python)**




```python
import math

# Sample data
n1, x1 = 1000, 642
n2, x2 = 1000, 591

# Sample proportions
p1_hat = x1 / n1
p2_hat = x2 / n2
diff_hat = p1_hat - p2_hat

# Pooled proportion under H0
p_pool = (x1 + x2) / (n1 + n2)

# Standard error under H0
se = math.sqrt(2 * p_pool * (1 - p_pool) / n1)

# Z statistic
z = diff_hat / se

# Two-tailed p-value (optional)
from math import erf, sqrt
p_value = 2 * (1 - 0.5 * (1 + erf(abs(z) / sqrt(2))))

z, p_value
```

This produces a Z‑score around **2.35** and a p‑value below **0.02**,
matching the rejection of the null hypothesis.
