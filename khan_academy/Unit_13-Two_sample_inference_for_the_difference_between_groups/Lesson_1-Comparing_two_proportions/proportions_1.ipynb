A 95% confidence interval for the difference in two population
proportions is built by treating each group as a Bernoulli process,
estimating each proportion from the sample, and then using the sampling
distribution of the **difference** of sample proportions.

**Bullet Summary**

- You have two populations: **men** with true proportion $p_{1}$and
  **women** with true proportion $p_{2}$who support a candidate.

- Each person’s response is a **Bernoulli** random variable (1 = vote
  for candidate, 0 = not).

- For a Bernoulli variable:

  - Mean = population proportion

  - Variance = $p(1 - p)$

- You take two independent random samples of size
  $n_{1} = n_{2} = 1000$.

  - Men: 642 supporters → ${\widehat{p}}_{1} = 0.642$

  - Women: 591 supporters → ${\widehat{p}}_{2} = 0.591$

- The sampling distribution of each sample proportion is approximately
  **normal** because $n$is large.

- The sampling distribution of the **difference**
  ${\widehat{p}}_{1} - {\widehat{p}}_{2}$is also normal with:

  - Mean: $p_{1} - p_{2}$

  - Variance:
    $\frac{p_{1}(1 - p_{1})}{n_{1}} + \frac{p_{2}(1 - p_{2})}{n_{2}}$

- For a 95% confidence interval, use:

$$
({\widehat{p}}_{1} - {\widehat{p}}_{2}) \pm z^* \sqrt{\frac{{\widehat{p}}_{1}(1 - {\widehat{p}}_{1})}{n_{1}} + \frac{{\widehat{p}}_{2}(1 - {\widehat{p}}_{2})}{n_{2}}}
$$

- With $z^* = 1.96$, you can compute the
  interval numerically.

**Explanation (Intuition + Mechanics)**

**1. Why Bernoulli matters**

Each person either votes for the candidate (1) or not (0). That’s
exactly a Bernoulli trial. The mean of a Bernoulli distribution is the
probability of success, so the population mean equals the population
proportion.

**2. Why sample proportions are approximately normal**

With large samples (1000 each), the Central Limit Theorem ensures the
distribution of sample proportions is close to normal. This lets us use
a z‑based confidence interval.

**3. Why variances add**

The samples of men and women are independent. When you subtract two
independent random variables, their variances **add**. That’s why the
standard error of the difference is the square root of the sum of the
two individual variances.

**4. What the confidence interval tells you**

The CI estimates the plausible range of the true difference
$p_{1} - p_{2}$.

- If the interval **does not include 0**, the difference is
  statistically meaningful.

- If it **includes 0**, the data are consistent with no difference.

**Code Example (Python)**

This computes the 95% CI for the difference in proportions using your
sample data.




```python
import math

# Sample data
n1 = 1000
x1 = 642
n2 = 1000
x2 = 591

# Sample proportions
p1_hat = x1 / n1
p2_hat = x2 / n2

# Difference in sample proportions
diff_hat = p1_hat - p2_hat

# Standard error of the difference
se = math.sqrt(p1_hat*(1 - p1_hat)/n1 + p2_hat*(1 - p2_hat)/n2)

# 95% z critical value
z_star = 1.96

# Confidence interval
lower = diff_hat - z_star * se
upper = diff_hat + z_star * se

diff_hat, lower, upper
```

**Expected Output (approx.)**

- Estimated difference: **0.051**

- 95% CI: roughly **(0.017, 0.085)**

This interval does **not** include 0, suggesting a meaningful difference
in support rates.
