# Confidence interval of difference of means

[Confidence interval of difference of means | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=hxZ6uooEJOk&t=3s)

We're trying to test whether a new, low-fat diet actually helps obese people lose weight. 100 randomly assigned obese people are assigned to group 1 and put on the low fat diet. Another 100 randomly assigned obese people are assigned to group 2 and put on a diet of approximately the same amount of food, but not as low in fat. After 4 months, the mean weight loss was 9.31 lbs. for group 1 (s=4.67) and 7.40 lbs. (s=4.04) for group 2.

A clear 95 percent confidence interval for the difference in mean weight
loss between the low‑fat diet and the control diet is built by combining
the sample difference (1.91 pounds) with the standard error of the
difference (0.617), then multiplying by the critical value 1.96. The
document states: “the 95 percent confidence interval is going to be the
difference of our means, 1.91, plus or minus this number, 1.21” and
earlier shows how the standard error was computed: “we calculated the
standard deviation of this distribution to be 0.617.”

**🔹 Bullet Summary**

- Two groups of 100 obese participants:

  - **Low‑fat diet**: mean loss 9.31 lb, SD 4.67

  - **Control diet**: mean loss 7.40 lb, SD 4.04

- Observed difference in sample means:

$$
{\overset{ˉ}{x}}_{1} - {\overset{ˉ}{x}}_{2} = 1.91\text{ lb}
$$

- Standard error of the difference:

$$
SE = \sqrt{\frac{{4.67}^{2}}{100} + \frac{{4.04}^{2}}{100}} = 0.617
$$

(Document: “we calculated the standard deviation of this distribution to
be 0.617.”)

- Critical value for 95 percent CI: **1.96**

- Margin of error:

$$
1.96 \times 0.617 = 1.21
$$

- 95 percent confidence interval:

$$
1.91 \pm 1.21 = (0.70,\text{\:\,}3.12)
$$

- Interpretation: Even the lower bound is positive, suggesting the
  low‑fat diet likely produces more weight loss than the control diet.

**🔹 Explanation (Intuition First)**

The goal is to estimate the *true* difference in average weight loss
between the two diets. We only observe sample means, so we build a
confidence interval around the observed difference.

1.  **Compute the observed difference** The low‑fat group lost 1.91 lb
    more on average.

2.  **Quantify sampling variability** Each sample mean has its own
    standard error. When comparing two means, the variability adds in
    quadrature:

$$
SE = \sqrt{SE_{1}^{2} + SE_{2}^{2}}
$$

3.  **Use the normal approximation** With sample sizes of 100 each, the
    sampling distribution of the difference is approximately normal.

4.  **Construct the interval** Multiply the standard error by 1.96 to
    capture the middle 95 percent of the distribution.

5.  **Interpretation** The entire interval is above zero, meaning the
    low‑fat diet appears to produce more weight loss.

**Python Code Example (Reproducible)**

```python
import numpy as np
from math import sqrt

# Given data
mean1, sd1, n1 = 9.31, 4.67, 100   # Low-fat group
mean2, sd2, n2 = 7.40, 4.04, 100   # Control group

# Difference in sample means
diff = mean1 - mean2

# Standard error of the difference
SE = sqrt(sd1**2 / n1 + sd2**2 / n2)

# Critical value for 95% CI
z = 1.96

# Confidence interval
lower = diff - z * SE
upper = diff + z * SE

diff, SE, (lower, upper)
```
