# Hypothesis test for difference of means

[Hypothesis test for difference of means | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=N984XGLjQfs&t=3s)

We’re trying to test whether a new, low-fat diet actually helps obese people lose weight. 100 randomly assigned obese people are assigned to group 1 and put on the low fat diet. Another 100 randomly assigned obese people are assigned to group 2 and put on a diet of approximately the same amount of food, but not as low in fat. After 4 months, the mean weight loss was 9.31 lbs. for group 1 (s=4.67) and 7.40 lbs. (s=4.04) for group 2.

**Bullet summary**

- The goal is to test whether the **low‑fat diet increases mean weight
  loss** compared to the control diet.

- **Null hypothesis:** $\mu_{1} - \mu_{2} = 0$, meaning the diet has
  *no effect*.

“Our null hypothesis should be that this low-fat diet does nothing… the
population mean on our low-fat diet minus the population mean on our
control should be equal to zero.”

- **Alternative hypothesis:** $\mu_{1} - \mu_{2} > 0$, meaning the
  low‑fat diet leads to *greater* weight loss.

- **Significance level:** $\alpha = 0.05$(one‑tailed test).

- **Standard error:** previously computed as $SE = 0.617$.

“It’s this quantity right up here… so it’s 0.617.”

- **Critical z‑value for one‑tailed 5%:** $z = 1.65$.

- **Critical threshold:**

$$
1.65 \cdot 0.617 = 1.02
$$

- **Observed difference:** $1.91$, which is **greater** than the
  critical value $1.02$.

- **Conclusion:** Reject $H_{0}$. The probability of observing a
  difference as large as 1.91 under the null is \< 5%, so the data
  supports that the low‑fat diet increases weight loss.

“The probability of getting this… is less than our significance level…
we will reject the null hypothesis.”

**Explanation**

The hypothesis test asks whether the low‑fat diet truly improves weight
loss, not just in the sample but in the population.

**1. Hypotheses**

- **Null:** $\mu_{1} - \mu_{2} = 0$. The diet has no effect.

- **Alternative:** $\mu_{1} - \mu_{2} > 0$. The diet increases weight
  loss.

This is a **right‑tailed** test because we are only looking for
improvement.

**2. Sampling distribution under the null**

If the null is true, the mean difference should be 0. The sampling
distribution of ${\overset{ˉ}{x}}_{1} - {\overset{ˉ}{x}}_{2}$is
approximately normal with standard error:

$$
SE = \sqrt{\frac{s_{1}^{2}}{n_{1}} + \frac{s_{2}^{2}}{n_{2}}} = 0.617
$$

**3. Critical region**

For a one‑tailed test at $\alpha = 0.05$, the critical z‑value is:

$$
z_{0.95} = 1.65
$$

Convert this to a critical difference:

$$
\text{Critical difference} = 1.65 \cdot 0.617 = 1.02
$$

Any observed difference **greater than 1.02** is statistically
significant.

**4. Compare observed difference**

The sample difference was:

$$
{\overset{ˉ}{x}}_{1} - {\overset{ˉ}{x}}_{2} = 1.91
$$

Since $1.91 > 1.02$, the result lies in the rejection region.

**5. Conclusion**

The probability of observing a difference as large as 1.91 if the diet
truly had no effect is less than 5%. Therefore, we **reject the null
hypothesis** and conclude that the low‑fat diet likely increases weight
loss.

**Code example (Python)**




```python
import math
from scipy.stats import norm

# Sample statistics
mean1, sd1, n1 = 9.31, 4.67, 100
mean2, sd2, n2 = 7.40, 4.04, 100

# Observed difference
diff = mean1 - mean2

# Standard error
SE = math.sqrt(sd1**2 / n1 + sd2**2 / n2)

# One-tailed critical z-value for alpha = 0.05
z_crit = norm.ppf(0.95)

# Critical threshold for the difference
critical_value = z_crit * SE

# Test statistic
z_stat = diff / SE

print(f"Observed difference: {diff:.2f}")
print(f"Standard error: {SE:.3f}")
print(f"Critical difference: {critical_value:.2f}")
print(f"Z-statistic: {z_stat:.2f}")

if diff > critical_value:
    print("Reject H0: Evidence suggests the low-fat diet increases weight loss.")
else:
    print("Fail to reject H0: Not enough evidence to conclude improvement.")
```
