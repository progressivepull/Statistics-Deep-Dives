# Clarification of confidence interval of difference of means

[Confidence interval of difference of means | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=hxZ6uooEJOk&t=3s)

We’re trying to test whether a new, low-fat diet actually helps obese people lose weight. 100 randomly assigned obese people are assigned to group 1 and put on the low fat diet. Another 100 randomly assigned obese people are assigned to group 2 and put on a diet of approximately the same amount of food, but not as low in fat. After 4 months, the mean weight loss was 9.31 lbs. for group 1 (s=4.67) and 7.40 lbs. (s=4.04) for group 2.

**Bullet summary**

- **Design:** 200 obese participants, randomly assigned: 100 to
  **low‑fat diet** (group 1), 100 to **control diet** (group 2).

“100 randomly assigned people are assigned to group one and put on the
low-fat diet. Another 100 randomly assigned obese people are assigned to
group two…”

- **Sample statistics:**

  - Group 1: ${\overset{ˉ}{x}}_{1} = 9.31$lb, $s_{1} = 4.67$

  - Group 2: ${\overset{ˉ}{x}}_{2} = 7.40$lb, $s_{2} = 4.04$

“So our sample mean for group one is 9.31 pounds, with a sample standard
deviation of 4.67… the sample mean, 7.40 pounds for group two. With a
sample standard deviation of 4.04 pounds.”

- **Observed difference in means:**

$$
\widehat{\Delta} = {\overset{ˉ}{x}}_{1} - {\overset{ˉ}{x}}_{2} = 1.91\text{ lb}
$$

- **Standard error of the difference:**

$$
SE = \sqrt{\frac{s_{1}^{2}}{n_{1}} + \frac{s_{2}^{2}}{n_{2}}} = \sqrt{\frac{{4.67}^{2}}{100} + \frac{{4.04}^{2}}{100}} \approx 0.617
$$

- **95% critical value:** $z = 1.96$.

- **Margin of error:**

$$
ME = 1.96 \times 0.617 \approx 1.21
$$

- **95% confidence interval for** $\mu_{1} - \mu_{2}$**:**

$$
1.91 \pm 1.21 = (0.70,\text{\:\,}3.12)
$$

- **Interpretation:** The entire CI is above 0, suggesting the low‑fat
  diet yields more weight loss than the control diet in the population.

**Explanation**

**1. Parameter of interest**

- **Goal:** Estimate the true difference in mean weight loss between
  low‑fat and control diets:

$$
\mu_{1} - \mu_{2}
$$

- **Estimator:** The difference in sample means
  $\widehat{\Delta} = {\overset{ˉ}{x}}_{1} - {\overset{ˉ}{x}}_{2}$.

**2. Sampling distribution**

- For large samples, ${\overset{ˉ}{x}}_{1}$and
  ${\overset{ˉ}{x}}_{2}$are approximately normal.

- The difference ${\overset{ˉ}{x}}_{1} - {\overset{ˉ}{x}}_{2}$is also
  approximately normal with mean $\mu_{1} - \mu_{2}$and standard
  error:

$$
SE({\overset{ˉ}{x}}_{1} - {\overset{ˉ}{x}}_{2}) = \sqrt{\frac{\sigma_{1}^{2}}{n_{1}} + \frac{\sigma_{2}^{2}}{n_{2}}}
$$

- Replace unknown $\sigma_{1},\sigma_{2}$with sample SDs
  $s_{1},s_{2}$.

**3. Compute the standard error**

$$
SE = \sqrt{\frac{{4.67}^{2}}{100} + \frac{{4.04}^{2}}{100}} \approx 0.617
$$

This quantifies how much the observed difference $1.91$would vary from
sample to sample.

**4. Build the 95% confidence interval**

- Use $z = 1.96$for a two‑sided 95% CI.

- Margin of error:

$$
ME = 1.96 \times 0.617 \approx 1.21
$$

- CI:

$$
\widehat{\Delta} \pm ME = 1.91 \pm 1.21 = (0.70,\text{\:\,}3.12)
$$

**5. Interpretation**

If we repeated this experiment many times and built a CI each time using
this method, about 95% of those intervals would contain the true
$\mu_{1} - \mu_{2}$. In this particular study, the plausible range for
the true extra weight loss from the low‑fat diet is roughly 0.7 to 3.1
pounds over four months.

**Python code example**




```python
import math

# Sample statistics
mean1, sd1, n1 = 9.31, 4.67, 100   # Low-fat group
mean2, sd2, n2 = 7.40, 4.04, 100   # Control group

# Difference in sample means
diff = mean1 - mean2

# Standard error of the difference
SE = math.sqrt(sd1**2 / n1 + sd2**2 / n2)

# 95% z critical value
z = 1.96

# Margin of error
ME = z * SE

# Confidence interval
lower = diff - ME
upper = diff + ME

print(f"Difference in means: {diff:.2f} lb")
print(f"Standard error: {SE:.3f}")
print(f"95% CI for (mu1 - mu2): ({lower:.2f}, {upper:.2f}) lb")
```

empirically.
