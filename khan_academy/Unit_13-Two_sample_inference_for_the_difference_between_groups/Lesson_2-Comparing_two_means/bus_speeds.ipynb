# Statistical significance on bus speeds

[Statistical significance on bus speeds | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=W3C07uH-b9o)

Giovanna usually takes bus B to work, but now she thinks that bus A gets her to work faster.

She randomized 50 workdays between a treatment group and a control group. For each day from the treatment group, she took bus A; and for each day from the control group, she took bus B. Each day she timed the length of her drive.

The results of the experiment showed that the median travel duration for bus A is 8 minutes less than the median travel duration for bus B. To test whether the results could be explained by random chance, she created the table below, which summarizes the results of 1000 re-randomizations of the data (with differences between medians rounded to the nearest 5 minutes).

According to the simulations, what is the probability of the treatment group’s median being lower than the control group’s median by 8 minutes or more?

![chart.png](./IMAGES/chart.png)

**Bullet Summary**

- Giovanna wants to know whether **Bus A is actually faster** than Bus
  B.

- She **randomized 50 workdays** into two groups (25 A, 25 B) to avoid
  human bias and confounding factors.

- Her observed result: **Median(A) − Median(B) = −8 minutes**, meaning
  Bus A was 8 minutes faster.

- To test whether this difference could be due to **random chance**, she
  performed **1000 re‑randomizations** of the same 50 commute times.

- For each shuffle, she computed the **difference in medians** again.

- She counted how many shuffled datasets produced a difference of **−8
  minutes or more extreme** (i.e., ≤ −8).

- She found **93 out of 1000** such cases → **p ≈ 0.093** (9.3%).

- Since 9.3% is **greater than the common 5% significance threshold**,
  the result is **not statistically significant**.

- Interpretation: Bus A *might* be faster, but the evidence is not
  strong enough to rule out chance.

**Explanation**

The key idea is **randomization inference**. Giovanna wants to know
whether the observed 8‑minute advantage for Bus A is real or just a
fluke caused by random day‑to‑day variation.

Because she randomized the assignment of days to buses, she can simulate
what would happen **if there were actually no difference** between
buses. She does this by:

1.  Keeping the 50 commute times fixed.

2.  Randomly shuffling which 25 belong to Bus A and which 25 belong to
    Bus B.

3.  Computing the difference in medians.

4.  Repeating this **1000 times** to build a **null distribution**.

If the observed difference (−8 minutes) is very rare under random
shuffling, then it is unlikely to be due to chance.

But she finds that **93 out of 1000** shuffles produce a difference of
−8 or more extreme. That’s **9.3%**, meaning such a result is not
particularly rare.

So the evidence is **suggestive** but not **strong**.

**Python Code Example**

Below is a clean simulation showing how you would perform the same
re‑randomization test in Python.




```python
import numpy as np

# Example: Giovanna's 50 commute times (fake data for demonstration)
# Replace with actual data if available
np.random.seed(42)
commute_times = np.random.randint(15, 45, size=50)

# Observed assignment: first 25 = Bus A, last 25 = Bus B
busA = commute_times[:25]
busB = commute_times[25:]

observed_diff = np.median(busA) - np.median(busB)
print("Observed difference:", observed_diff)

# Re-randomization test
num_simulations = 1000
diffs = []

for _ in range(num_simulations):
    shuffled = np.random.permutation(commute_times)
    A = shuffled[:25]
    B = shuffled[25:]
    diffs.append(np.median(A) - np.median(B))

diffs = np.array(diffs)

# Compute p-value: probability of <= observed difference
p_value = np.mean(diffs <= observed_diff)
print("Estimated p-value:", p_value)
```

**What this code does**

- Generates or loads the 50 commute times.

- Computes the **observed** difference in medians.

- Performs **1000 random shuffles** of the data.

- Computes the distribution of median differences under the null
  hypothesis.

- Estimates the **p‑value** as the proportion of shuffled differences
  that are ≤ the observed difference.
