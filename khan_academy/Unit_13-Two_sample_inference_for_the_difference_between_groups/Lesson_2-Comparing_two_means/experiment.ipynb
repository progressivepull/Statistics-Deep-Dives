# Statistical significance of experiment

[Statistical significance of experiment | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=jLFeqQxGtOc&t=2s)

In an experiment aimed at studying the effect of advertising on eating behavior in children, a group of 500 children, 7-11 years old, were randomly assigned to two different groups. After randomization, each child was asked to watch a cartoon in a private room, containing a large bowl of goldfish crackers. The cartoon included 2 commercial breaks.

The first group watched food commercials (mostly snacks), while the second group watched non-food commercials (games and entertainment products). Once the child finished watching the cartoon, the conductors of the experiment weighed the crackers bowl to measure how many grams of crackers the child ate. They found that the mean amount of crackers eaten by the children who watched food commercials is 10 grams greater than the mean amount of crackers eaten by the children who watched non-food commercials.

Using a simulator, they re-randomized the results into two new groups and measured the difference between the means of the new groups. They repeated this simulation 150 times, and plotted the resulting differences, as given below. 

**According to the simulations, is the result of the experiment significant?**

![graph.png](./IMAGES/graph.png)

A randomization‑based hypothesis test asks whether the observed 10‑gram
difference in cracker consumption could plausibly arise **just by
chance** if food commercials had *no real effect*. The experimenters
simulate this “no‑effect world” by repeatedly shuffling the 500 children
into two fake groups and recomputing the difference in means. Because a
10‑gram gap appears in fewer than 1–2 percent of these simulations, the
observed result is unlikely to be due to random assignment alone,
providing strong evidence that food commercials increased eating.

**🟦 Bullet summary**

- **Experiment setup**: 500 children randomly assigned to a **treatment
  group** (food commercials) and a **control group** (non‑food
  commercials).

- **Outcome measured**: grams of crackers eaten while watching a
  cartoon.

- **Observed result**: treatment group ate **10 grams more** on average
  than the control group.

- **Key question**: Could a 10‑gram difference arise **purely by
  chance** even if commercials had no effect?

- **Randomization test**:

  - Keep all 500 cracker‑eating amounts fixed.

  - Randomly shuffle children into two new groups of 250 each.

  - Compute the difference in means.

  - Repeat this **150 times** to approximate the “null distribution.”

- **Simulation results**:

  - Most shuffled differences cluster near **0 grams**.

  - A difference as large as **10 grams** appears only **1–2 times out
    of 150** (\< 2%).

- **Interpretation**:

  - Probability of seeing a 10‑gram difference by chance is **\< 2%**,
    well below the common 5% significance threshold.

  - Conclude the experiment provides **statistically significant
    evidence** that food commercials increase eating.

**🟦 Explanation**

The logic mirrors a formal hypothesis test:

**1. Null hypothesis**

The commercials have **no effect**, so both groups come from the same
population distribution of cracker consumption.

**2. What the simulation represents**

If the null is true, then the only reason one group might eat more than
the other is **random assignment**. By shuffling the 500 observed
outcomes into two new groups repeatedly, the simulation builds the
distribution of differences we would expect **if the null were true**.

**3. Comparing observed difference to null distribution**

The actual experiment produced a **10‑gram** difference. In the
simulated “no‑effect world,” such a large difference almost never
occurs—only about **1 in 150** trials.

**4. Statistical conclusion**

A result that rare under the null suggests the null is unlikely. Thus,
the data support the conclusion that **food commercials caused children
to eat more**.

This is a classic example of a **randomization test** (also called a
permutation test), especially useful when the sampling distribution is
unknown or when the experiment is randomized.

**🟦 Code example (Python simulation)**

Below is a simple Python simulation that mimics the randomization test.
It assumes you have a list of cracker amounts for all 500 children and a
vector indicating which children originally saw food commercials.




```python
import numpy as np

# ---------------------------------------------------------
# 1. Create the arrays (replace these with your real data!)
# ---------------------------------------------------------

# Example: 500 cracker-eating amounts (in grams)
# Replace this with your actual 500-length array
crackers = np.random.normal(loc=30, scale=10, size=500)

# Example: original assignment (250 treatment, 250 control)
# Replace with your real 0/1 assignment
group = np.array([1]*250 + [0]*250)

# ---------------------------------------------------------
# 2. Observed difference
# ---------------------------------------------------------
obs_diff = crackers[group == 1].mean() - crackers[group == 0].mean()

# ---------------------------------------------------------
# 3. Randomization test
# ---------------------------------------------------------
n_sim = 150
sim_diffs = []

for _ in range(n_sim):
    shuffled = np.random.permutation(crackers)
    group1 = shuffled[:250]
    group2 = shuffled[250:]
    sim_diffs.append(group1.mean() - group2.mean())

sim_diffs = np.array(sim_diffs)

# ---------------------------------------------------------
# 4. p-value
# ---------------------------------------------------------
p_value = np.mean(np.abs(sim_diffs) >= abs(obs_diff))

obs_diff, p_value

```

**Expected behavior**

- obs_diff ≈ **10 grams**

- p_value ≈ **0.01–0.02**, matching the experiment’s conclusion.
