# Using a confidence interval to test slope

[Using a confidence interval to test slope | More on regression | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=rL9MPL3wqsk&t=2s)

Hashem obtained a random sample of students and noticed a positive linear relationship between their ages and their backpack weights. A 95% confidence interval for the slope of the regression line was 0.39 ±0.23.                           

Hashem wants to use this interval to test H₀: β = 0 vs. Hₐ: β ≠ 0 at α = 0.05 level of significance. Assume that all conditions for inference have been met.

A 95 percent confidence interval for the slope that stays entirely above
zero already tells you the sample provides strong evidence of a positive
relationship. That interval runs from 0.16 to 0.62, so zero is not a
plausible value for the true slope under the sampling model. Because a
two‑sided hypothesis test at the 5 percent level rejects the null
exactly when zero is outside the 95 percent confidence interval, the
correct conclusion is to reject the null hypothesis and infer that the
true slope is nonzero.

**📌 Bullet Summary**

- Hashem is studying the relationship between **age** and **backpack
  weight** using linear regression.

- The 95 percent confidence interval for the population slope is

$$
0.39 \pm 0.23 = (0.16,\text{\:\,}0.62).
$$

- The null hypothesis claims the true slope is zero, meaning no linear
  relationship.

- A confidence interval that **does not include zero** means zero is not
  a plausible value for the slope.

- At significance level $\alpha = 0.05$, a two‑sided test rejects the
  null exactly when zero lies outside the 95 percent interval.

- Since zero is below the entire interval, the sample result is too
  extreme to occur under the null.

- Conclusion: **Reject the null hypothesis** and infer that the true
  slope is not zero.

- Interpretation: There is evidence of a **nonzero linear relationship**
  between age and backpack weight in the population.

**🧠 Explanation**

A confidence interval gives a range of plausible values for the true
slope $\beta$. If the null hypothesis value $\beta = 0$were true,
then 95 percent of intervals constructed this way would contain zero.
But Hashem’s interval does not contain zero; it is entirely positive.
That places this sample in the rare 5 percent of cases where the
interval misses the true value if the null were correct. Because this
event is unlikely under the null, the hypothesis test rejects it.

This logic mirrors the equivalence between confidence intervals and
two‑sided hypothesis tests: a 95 percent interval corresponds to a 5
percent significance level. When the null value lies outside the
interval, the test rejects.

**💻 Code Example (Reproducing the Logic)**

The following Python snippet reconstructs the interval, checks whether
zero lies inside it, and performs the equivalent hypothesis test using
the interval:

```python
import numpy as np

# Given confidence interval center and margin
center = 0.39
margin = 0.23

# Compute interval
lower = center - margin
upper = center + margin

print("95% CI for slope:", (lower, upper))

# Hypothesis test using CI logic
beta_null = 0
reject_null = not (lower <= beta_null <= upper)

if reject_null:
    print("Conclusion: Reject H0 (slope is not zero).")
else:
    print("Conclusion: Fail to reject H0.")
```

This code checks whether the null value lies inside the interval.
Because it does not, the program concludes that the null hypothesis
should be rejected.
