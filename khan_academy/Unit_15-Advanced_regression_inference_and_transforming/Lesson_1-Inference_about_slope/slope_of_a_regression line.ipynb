# Confidence interval for the slope of a regression line

[Confidence interval for the slope of a regression line | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=8w6EPyEqE9M&t=1s)

Musa is interested in the relationship between hours spent studying and caffeine consumption among students at his school. He randomly selects 20 students at his school and records their caffeine intake (mg) and the amount of time spent studying in a given week. Here is computer output from a least-squares regression analysis on his sample:

| Predictor | Coef  | SE    | t     | p     |
|-----------|-------|-------|-------|-------|
| Constant  | 2.544 | 0.134 | 18.955 | 0.000 |
| Caffeine  | 0.164 | 0.057 | 2.862  | 0.010 |

S = 1.532R-sq = 60.0%


Assume that all conditions for inference have been met.                  

**What is the 95% confidence interval for the slope of the least squares regression line?**











**Bullet summary**

- Musa studies whether **caffeine intake (x)** predicts **hours spent
  studying (y)** using a sample of 20 students.

- The regression output gives a **slope estimate** of $0.164$and a
  **standard error of the slope** of $0.057$.

- To build a **95% confidence interval**, use

$$
b \pm t^{\text{*}} \cdot SE_{b}.
$$

- Degrees of freedom for regression slope: $n - 2 = 20 - 2 = 18$.

- The **critical t‑value** for 95% confidence and df = 18 is **2.101**.

- Margin of error:

$$
2.101 \cdot 0.057 \approx 0.1198.
$$

- Confidence interval:

$$
0.164 \pm 0.1198 = (0.0442,\text{\:\,}0.2838).
$$

- Interpretation: the true slope is likely between **0.044 and 0.284**,
  suggesting a positive relationship between caffeine intake and study
  time.

**Explanation**

The slope of the regression line represents how much study time changes,
on average, for each additional milligram of caffeine consumed. Because
the slope is estimated from a sample of 20 students, it varies from
sample to sample. A confidence interval quantifies the uncertainty
around this estimate.

The formula for a confidence interval for the slope is:

$$
b \pm t^* \cdot SE_{b},
$$

where:

- $b$is the sample slope (0.164),

- $SE_{b}$is the standard error of the slope (0.057),

- $t^*$ is the critical value from the
  t‑distribution with $n - 2$degrees of freedom.

With 20 students, the degrees of freedom are $18$. For a 95%
confidence level, the critical value is
$t^* = 2.101$. Multiplying this by the standard
error gives the margin of error:

$$
2.101 \cdot 0.057 \approx 0.1198.
$$

Adding and subtracting this from the slope gives the interval:

$$
(0.164 - 0.1198,\text{\:\,}0.164 + 0.1198) = (0.0442,\text{\:\,}0.2838).
$$

This interval does not include zero, which suggests the relationship
between caffeine intake and study time is likely positive in the
population.

**Code example**

```python
import scipy.stats as stats

# Given values
b = 0.164          # slope estimate
SE_b = 0.057       # standard error of slope
n = 20             # sample size

# Degrees of freedom
df = n - 2

# Critical t-value for 95% CI
t_crit = stats.t.ppf(0.975, df)

# Margin of error
ME = t_crit * SE_b

# Confidence interval
lower = b - ME
upper = b + ME

print(f"95% CI for slope: ({lower:.4f}, {upper:.4f})")
```

This code reproduces the confidence interval
$\left( 0.0442,\text{\:\,}0.2838 \right)$.
