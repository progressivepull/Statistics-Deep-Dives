# Using a P-value to make conclusions in a test about slope

[Using a P-value to make conclusions in a test about slope | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=Mpd83AuDTrU&t=8s)

Alicia took a random sample of mobile phones and found a positive linear relationship between their processor speeds and their prices. Here is computer output from a least-squares regression analysis on her sample:          

**Regression: Price vs. speed**


| Predictor | Coef     | SE Coef | t       | p      |
|-----------|----------|---------|---------|--------|
| Constant  | 127.092  | 57.507  | 2.210   | 0.032  |
| Speed     | 6.084    | 2.029   | 2.999   | 0.004  |


Alicia wants to test H₀: β = 0 vs. Hₐ: β > 0. Assume that all conditions for inference have been met.          

**At the α = 0.01 level of significance, is there sufficient evidence to conclude a positive linear relationship between these variables for all mobile phones? Why?**



**📌 Bullet Summary**

- Alicia sampled mobile phones and ran a simple linear regression with
  **price** as the response and **processor speed** as the predictor.

- The sample regression output shows:

  - **Slope (Speed)**: $b_{1} = 6.084$

  - **Standard error of slope**: $SE(b_{1}) = 2.029$

  - **t‑statistic**: $t = 2.999$

  - **Two‑sided p‑value**: $0.004$

- The hypotheses are:

  - $H_{0}:\beta = 0$(no linear relationship)

  - $H_{a}:\beta > 0$(positive linear relationship)

- Because the computer output gives a **two‑sided** p‑value, the correct
  **one‑sided** p‑value is half of it:

$$
p = 0.004/2 = 0.002
$$

- At significance level $\alpha = 0.01$, the p‑value is smaller:

$$
0.002 < 0.01
$$

- This means the observed slope is unlikely under the null, so we
  **reject** $H_{0}$.

- Conclusion: There is **sufficient evidence** of a positive linear
  relationship between processor speed and price for all mobile phones.

**📘 Explanation**

Alicia’s regression estimates how much phone price tends to increase for
each additional unit of processor speed. The slope
$b_{1} = 6.084$suggests faster processors are associated with higher
prices in her sample.

But to generalize to all mobile phones, she performs a hypothesis test.
The null hypothesis assumes **no relationship** in the population
($\beta = 0$). The t‑statistic measures how many standard errors the
observed slope is above zero:

$$
t = \frac{b_{1} - 0}{SE(b_{1})} = \frac{6.084}{2.029} = 2.999
$$

The computer output gives a two‑sided p‑value of 0.004, but Alicia’s
alternative hypothesis is **one‑sided** ($\beta > 0$), so the correct
p‑value is half of that: 0.002.

Since 0.002 is smaller than the significance level 0.01, the sample
provides strong evidence that the true slope is positive. In practical
terms, this means processor speed is genuinely associated with higher
price across the population of mobile phones.

**💻 Code Example (Compute t‑statistic and p‑value)**

```python
import numpy as np
from scipy import stats

# Regression output
b1 = 6.084        # sample slope
SE_b1 = 2.029     # standard error of slope
t_stat = b1 / SE_b1

print("t-statistic:", t_stat)

# Suppose sample size was n (not given in problem)
# For demonstration, assume n = 30
n = 30
df = n - 2

# Two-sided p-value from output
p_two_sided = 0.004

# One-sided p-value for Ha: beta > 0
p_one_sided = p_two_sided / 2

print("One-sided p-value:", p_one_sided)

# Alternatively compute from t-distribution
p_from_t = 1 - stats.t.cdf(t_stat, df=df)
print("One-sided p-value (computed):", p_from_t)
```

This code:

- Computes the t‑statistic from the slope and its standard error.

- Converts the two‑sided p‑value to a one‑sided p‑value.

- Optionally computes the p‑value directly from the t‑distribution.
