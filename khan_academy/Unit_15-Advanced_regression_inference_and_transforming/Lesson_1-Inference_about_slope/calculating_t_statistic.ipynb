**Calculate the test statistic that should be used for testing a null hypothesis that the population slope is actually 0?**



A clear test statistic for Jian’s regression slope comes from comparing
the **sample slope** to the **null hypothesis slope** and scaling by the
**standard error of the slope**. The value is:

$$
t = \frac{b_{1} - 0}{SE(b_{1})} = \frac{14.686}{13.329} \approx 1.10
$$

**🔹 Bullet Summary**

- Jian sampled **24 students** and ran a simple linear regression:
  reaction time $x$predicting memory time $y$.

- He wants to test whether the **true population slope** is zero:

$$
H_{0}:\beta = 0\text{vs.}H_{a}:\beta > 0
$$

- From the regression output:

  - Sample slope $b_{1} = 14.686$

  - Standard error $SE(b_{1}) = 13.329$

- The test statistic for a slope uses a **t‑ratio**:

$$
t = \frac{b_{1} - \beta_{0}}{SE(b_{1})}
$$

- Under the null, $\beta_{0} = 0$, so:

$$
t = \frac{14.686}{13.329} \approx 1.10
$$

- Degrees of freedom: $n - 2 = 22$. This t‑value would then be used to
  compute a one‑sided p‑value.

**🔹 Explanation in Plain Terms**

The slope $b_{1}$tells us how much memory time tends to increase for
each extra unit of reaction time. If the true slope were really zero,
then any nonzero slope we observe is just sampling noise. The
t‑statistic measures how many standard errors the observed slope is
above zero. A value near 1 means the slope is only about one standard
error away from zero, which is not strong evidence against the null.

**🔹 Python Code Example (Simulation + Test Statistic)**

This snippet computes the test statistic directly and shows how you
could simulate the sampling distribution of slopes under the null.

```python
import numpy as np
from scipy import stats

# Given regression output
b1 = 14.686
SE_b1 = 13.329

# Test statistic
t_stat = b1 / SE_b1
print("t statistic:", t_stat)

# Degrees of freedom for simple linear regression
df = 24 - 2

# One-sided p-value (beta > 0)
p_value = 1 - stats.t.cdf(t_stat, df=df)
print("One-sided p-value:", p_value)

# Optional: simulate slopes under H0 for intuition
np.random.seed(0)
n = 24
num_sims = 10000
sim_slopes = np.random.normal(loc=0, scale=SE_b1, size=num_sims)
sim_t = sim_slopes / SE_b1

print("Simulated proportion exceeding observed t:", np.mean(sim_t >= t_stat))
```

This code:

- Computes the t‑statistic.

- Computes the one‑sided p‑value.

- Simulates many slopes under the null to visualize how unusual the
  observed slope is.
