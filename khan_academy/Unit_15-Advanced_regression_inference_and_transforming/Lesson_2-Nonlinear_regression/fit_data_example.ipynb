# Comparing models to fit data example

[Comparing models to fit data | Regression | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=2EwTHdg-xgw&t=2s)

Kristine works at the movie store in her hometown. Using the store's total selection, she documented the price of each movie title and how many years it has been since it was featured in movie theaters. She plotted the points below.    

She is looking for a function that models her data. Since the trend of the data is decreasing and convex, she found a decreasing convex exponential model and a decreasing convex quadratic model.        

Which of the following functions better fits the data?                    
○ Function A: y = 14 · 0.45ˣ         
○ Function B: y = 0.5x² − 5x + 13       

Use the function of best fit to predict the price of a movie that was featured in theaters 5.5 years ago. Round your answer to the nearest cent.      

**📌 Bullet Summary**

- Christine collected data on **movie price** versus **years since
  theatrical release** and plotted the points.

- The pattern is **decreasing** and **convex**, meaning prices drop over
  time but at a slowing rate.

- She compared two candidate models:

  - **Function A (exponential)** — consistently **underestimates** most
    data points.

  - **Function B (quadratic)** — has a **balanced mix** of small
    overestimates and underestimates.

- Because Function B stays closer to the data across the full range, it
  is the **better fit**.

- To predict the price of a movie released **5.5 years ago**, substitute
  $x = 5.5$into Function B:

$$
y = 0.5x^{2} - 5x + 13
$$

- The computed value is **\$0.625**, which rounds to **\$0.63**.

**🎬 Explanation**

Christine wants a model that captures how movie prices decline as they
age. The scatterplot shows a downward‑sloping, convex curve: prices fall
quickly at first, then level off. An exponential model can capture
decreasing behavior, but in this case the exponential curve sits **below
almost every data point**, meaning it systematically underpredicts
prices. A good model should not be biased in one direction.

The quadratic model bends downward and then flattens, matching the
convex shape of the data. It slightly overestimates in some regions and
slightly underestimates in others, but the deviations are small and
balanced. That makes it a more faithful representation of the underlying
trend.

Using the quadratic model:

$$
y = 0.5x^{2} - 5x + 13
$$

Plugging in $x = 5.5$:

$$
y = 0.5(5.5)^{2} - 5(5.5) + 13 = 0.625
$$

Rounded to the nearest cent, the predicted price is **\$0.63**.

**💻 Code Example (Quadratic Prediction)**

```python
def movie_price(x):
    # Quadratic model: y = 0.5x^2 - 5x + 13
    return 0.5 * x**2 - 5 * x + 13

x = 5.5
price = movie_price(x)

print("Predicted price:", round(price, 2))
```

This code evaluates the quadratic model at 5.5 years and rounds to the
nearest cent.
