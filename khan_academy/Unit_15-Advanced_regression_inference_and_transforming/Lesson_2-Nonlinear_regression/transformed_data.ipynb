A conservation group with a long-term goal of preserving species believes that all at-risk species will disappear whenever land inhabited by those species is developed. It has an opportunity to purchase land in an area about to be developed. The group has a choice of creating one large nature preserve with an area of 45 square kilometers and containing 70 at-risk species, or 5 small nature preserves, each with an area of 3 square kilometers and each containing 16 at-risk species unique to that preserve. Which choice would you recommend and why?

![island.png](./IMAGES/island.png)

The **large single preserve** is the better choice because the
regression model predicts **far fewer extinctions** for a 45 km²
preserve with 70 species than for five 3 km² preserves totaling 80
species. The log–area relationship shows that **small preserves suffer
disproportionately higher extinction rates**, even when they
collectively hold more species.

**🟦 Bullet Summary**

- The regression model relates **proportion extinct** to **ln(area)**:

$$
\widehat{p} = 0.28996 - 0.05323 \cdot \ln(\text{area})
$$

- Larger areas have **lower predicted extinction proportions** because
  the slope is negative.

- **Scenario 1: One large preserve (45 km², 70 species)**

  - Predicted proportion extinct ≈ 0.087

  - Predicted extinctions ≈ 6

  - Predicted species saved ≈ 64

- **Scenario 2: Five small preserves (3 km² each, 16 species each → 80
  total)**

  - Predicted proportion extinct ≈ 0.231

  - Predicted extinctions ≈ 19

  - Predicted species saved ≈ 61

- **Recommendation:** Choose the **large preserve** because it saves
  more species and loses fewer.

**🟩 Why the Large Preserve Wins**

The key insight is the **logarithmic transformation** of area.
Extinction risk drops quickly as area increases, but rises sharply as
area becomes very small. This creates a **nonlinear penalty** for small
preserves:

- Going from 45 km² to 3 km² is not a 15× increase in risk; it’s much
  worse because

$$
\ln(3) \ll \ln(45)
$$

- Even though the five small preserves collectively hold more species
  (80 vs. 70), the **much higher extinction proportion** overwhelms that
  advantage.

This is a classic application of the **species–area relationship** and
**island biogeography theory**: small isolated patches lose species
faster.

**🟧 Code Example (Python)**

A clean, reproducible version of the calculations:

```python
import numpy as np

# Regression coefficients
intercept = 0.28996
slope = -0.05323

def predicted_extinctions(area, species):
    proportion = intercept + slope * np.log(area)
    return proportion, proportion * species

# Scenario 1: One large preserve
area_large = 45
species_large = 70
p_large, extinct_large = predicted_extinctions(area_large, species_large)

# Scenario 2: Five small preserves
area_small = 3
species_small_total = 5 * 16
p_small, extinct_small = predicted_extinctions(area_small, species_small_total)

p_large, extinct_large, p_small, extinct_small
```

**🟦 Final Recommendation**

Choose the **single 45 km² preserve**. It is expected to **save more
species** and **lose fewer**, based on the regression model and the
ecological principle that **larger continuous habitats reduce extinction
risk**.
