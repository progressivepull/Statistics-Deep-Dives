# Summary: Robustness in Statistics (Salaries Example)

This report evaluates how a single outlier can skew statistical results and why different measures of central tendency and spread are preferred depending on the data distribution.

### 1. The Data Set (Salaries in $1,000s)
*   **Population:** 9 recent graduates.
*   **Values:** 35, 50, 50, 50, **56**, 60, 60, 75, 250.
*   **Observation:** The value **250** is an outlier, sitting significantly higher than the rest of the group.

---

### 2. Central Tendency: Mean vs. Median
The goal is to find the "typical" salary.

*   **Arithmetic Mean ($\approx 76.2$):** Calculated by summing all values and dividing by 9. This value is misleading because it is higher than 8 out of the 9 salaries. It is "pulled" upward by the outlier.
*   **Median (56):** The middle value of the ordered list. It is **robust**, meaning it remains at 56 even if the highest salary increases to 250 million. 

**Conclusion:** The **Median** is a better measure for skewed data.

---

### 3. Measures of Spread: Standard Deviation vs. IQR
The goal is to understand the "dispersion" or variation of salaries.

*   **Standard Deviation ($\approx 61.5$):** Since this is calculated using the mean, it is also heavily skewed by the outlier, giving the impression of a much wider spread than exists for the majority of the group.
*   **Interquartile Range (IQR = 17.5):** This measures the spread of the middle 50% of the data.
    *   **Q1 (25th percentile):** 50
    *   **Q3 (75th percentile):** 67.5
    *   **IQR calculation:** $67.5 - 50 = 17.5$

**Conclusion:** The **IQR** is a better measure of spread for skewed data as it ignores extreme outliers.

---

### Summary Table
When analyzing data with extreme outliers (like the $250,000 salary in this group), **robust measures** provide a more accurate representation of the typical experience than **sensitive measures**.

| Metric Type          | Sensitive Measure (Skewed) | Robust Measure (Preferred) |
|----------------------|----------------------------|-----------------------------|
| Central Tendency     | Mean (76.2)                | Median (56)                 |
| Spread / Dispersion  | Standard Deviation (61.5)  | IQR (17.5)                  |

### Key Insights:
* **Central Tendency:** The **Mean** is pulled upward by the outlier, making the "average" higher than 88% of the students' salaries. The **Median** remains at the center, unaffected by how high the top salary goes.
* **Spread:** The **Standard Deviation** suggests a massive variation ($61,500) because it calculates distance from the skewed mean. The **Interquartile Range (IQR)** focuses on the middle 50% of the data, providing a more realistic sense of the salary spread for the majority of the class.


```python
import statistics

# Salaries in thousands (as provided)
salaries = [35, 50, 50, 50, 56, 60, 60, 75, 250]

def calculate_statistics(data):
    # Sort data for median and IQR calculations
    data.sort()
    n = len(data)

    # 1. Central Tendency
    mean_val = sum(data) / n
    median_val = statistics.median(data)

    # 2. Spread (Standard Deviation)
    # Using population standard deviation (pstd) as this is the whole "class"
    std_dev = statistics.pstdev(data)

    # 3. Spread (Interquartile Range - IQR)
    # Lower half (excluding median of the 9 points)
    lower_half = data[:4]
    # Upper half (excluding median of the 9 points)
    upper_half = data[5:]
    
    q1 = statistics.median(lower_half)
    q3 = statistics.median(upper_half)
    iqr = q3 - q1

    return {
        "Mean": round(mean_val, 2),
        "Median": median_val,
        "Std Dev": round(std_dev, 2),
        "IQR": iqr,
        "Q1": q1,
        "Q3": q3
    }

# Execute and Print Results
stats = calculate_statistics(salaries)

print(f"--- Salary Analysis (in $1,000s) ---")
print(f"Mean:            {stats['Mean']}")
print(f"Median:          {stats['Median']}")
print(f"Std Deviation:   {stats['Std Dev']}")
print(f"IQR:             {stats['IQR']} (from Q1: {stats['Q1']} to Q3: {stats['Q3']})")

print("\n--- Analysis ---")
if stats['Mean'] > stats['Median']:
    print(f"The data is right-skewed by the $250k outlier.")
    print(f"The Median ({stats['Median']}k) is a better measure of central tendency.")
    print(f"The IQR ({stats['IQR']}k) is a more robust measure of spread.")
```

### Why this code matters for your data:
1. **Mean vs. Median:** The mean (~76.2) is higher than 8 out of the 9 salaries. It gives a "false impression" of a typical salary because it is pulled up by the single 250k earner. The median (56) accurately reflects the middle of the group.
2. **Standard Deviation vs. IQR:** The standard deviation is heavily inflated by the 250k outlier. The IQR focuses on the middle 50% of the class, showing that most graduates earn within a tight range of \\$17,500 of each other (\\$50k to \\$67.5k).
