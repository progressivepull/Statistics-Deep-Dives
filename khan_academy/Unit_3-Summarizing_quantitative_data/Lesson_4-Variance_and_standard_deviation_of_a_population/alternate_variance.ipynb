# Population Variance Formulas

In statistics, the population variance ($\sigma^2$) measures the spread of a data set. While both formulas below yield the same result, they serve different purposes during calculation.

## 1. Definitional Formula
This formula is used to define the concept of variance. it shows the average squared distance of each data point ($x$) from the population mean ($\mu$).

$$\sigma^2 = \frac{\sum (x - \mu)^2}{N}$$

## 2. Computational (Alternate) Formula
This version is mathematically equivalent but is often preferred for manual calculations. It is colloquially known as **"the mean of the squares minus the square of the mean."**

$$\sigma^2 = \frac{\sum x^2}{N} - \mu^2$$

---

### Variable Key
*   **$\sigma^2$**: Population Variance
*   **$\sum$**: Summation operator
*   **$x$**: Each individual value in the data set
*   **$\mu$**: Population Mean
*   **$N$**: Total number of values in the population

### Why use the Alternate Formula?
When calculating by hand, the computational formula is more efficient because:
1.  **Reduced Subtraction**: You only subtract once at the very end.
2.  **Cleaner Numbers**: You avoid dealing with the decimals that often result from $(x - \mu)$ until the final step, reducing the risk of rounding errors.
3.  **Efficiency**: You only need to track two running totals: the sum of the values ($\sum x$) and the sum of the squared values ($\sum x^2$).


```python
# variance_calculation.py

def calculate_variance(data):
    """
    Calculates population variance using both the definitional 
    and computational formulas to demonstrate their equivalence.
    """
    N = len(data)
    if N == 0:
        return None

    # 1. Definitional Formula: σ² = Σ(x - μ)² / N
    mu = sum(data) / N
    squared_diffs = [(x - mu) ** 2 for x in data]
    definitional_variance = sum(squared_diffs) / N

    # 2. Computational Formula: σ² = (Σx² / N) - μ²
    sum_of_squares = sum([x ** 2 for x in data])
    computational_variance = (sum_of_squares / N) - (mu ** 2)

    return definitional_variance, computational_variance, mu

# Example Data
population_data = [10, 12, 23, 23, 16, 23, 21, 16]

# Execution
def_var, comp_var, mean = calculate_variance(population_data)

print(f"Population Mean (μ): {mean}")
print(f"Variance (Definitional): {def_var:.4f}")
print(f"Variance (Computational): {comp_var:.4f}")

# Verification with Python's built-in library
import statistics
print(f"Statistics module (pvariance): {statistics.pvariance(population_data):.4f}")
```

### Key Python Functions Used

*   **`sum(data)`**: Adds all elements in a list to find $\sum x$.
*   **`len(data)`**: Returns the total number of observations, $N$.
*   **`** 2`**: The operator used for squaring values in Python.
*   **`statistics.pvariance()`**: A built-in function specifically for calculating population variance ($N$ in the denominator) rather than sample variance ($n-1$).
