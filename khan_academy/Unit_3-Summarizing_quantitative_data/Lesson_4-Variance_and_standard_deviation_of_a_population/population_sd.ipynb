# Summary: Population Mean, Variance, and Standard Deviation

This report summarizes the statistical measurements for a population of five cars based on their lengths in meters.

### 1. Population Mean ($\mu$)
The arithmetic mean represents the central tendency of the data.
*   **Data (meters):** 4, 4.2, 5, 4.3, 5.5
*   **Calculation:**
    $$\mu = \frac{4 + 4.2 + 5 + 4.3 + 5.5}{5} = \frac{23}{5} = 4.6$$
*   **Result:** The mean car length is **4.6 meters**.

---

### 2. Population Variance ($\sigma^2$)
Variance measures the dispersion of the data. Because the differences from the mean are squared, the units become **square meters ($m^2$)**.
*   **Formula:** $\sigma^2 = \frac{\sum (x_i - \mu)^2}{N}$
*   **Calculation:**
    1. $(4 - 4.6)^2 = 0.36$
    2. $(4.2 - 4.6)^2 = 0.16$
    3. $(5 - 4.6)^2 = 0.16$
    4. $(4.3 - 4.6)^2 = 0.09$
    5. $(5.5 - 4.6)^2 = 0.81$
*   **Sum of Squares:** $1.58$
*   **Average (Variance):** $1.58 / 5 = 0.316$
*   **Result:** The population variance is **0.316 $m^2$**.

---

### 3. Population Standard Deviation ($\sigma$)
To return the measure of dispersion to a relatable unit (meters), we take the square root of the variance. This is the most common way to represent how much data varies from the mean.
*   **Formula:** $\sigma = \sqrt{\sigma^2}$
*   **Calculation:**
    $$\sigma = \sqrt{0.316} \approx 0.562$$
*   **Result:** The population standard deviation is approximately **0.562 meters**.

---

### Summary of Key Terms


```python
import math

def calculate_stats(data):
    # Calculate population mean
    n = len(data)
    mean = sum(data) / n
    
    # Calculate population variance
    variance = sum((x - mean)**2 for x in data) / n
    
    # Calculate population standard deviation
    std_dev = math.sqrt(variance)
    
    return mean, variance, std_dev

# Car length data in meters
car_lengths = [4, 4.2, 5, 4.3, 5.5]

mu, sigma_sq, sigma = calculate_stats(car_lengths)

print(f"Population Mean: {mu} m")
print(f"Population Variance: {sigma_sq:.3f} m^2")
print(f"Population Standard Deviation: {sigma:.3f} m")
```
