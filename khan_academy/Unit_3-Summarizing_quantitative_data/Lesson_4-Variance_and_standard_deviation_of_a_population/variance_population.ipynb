# Summary: Population Mean and Variance Calculation

This summary outlines the process for calculating the arithmetic mean and population variance using a sample population of five employees at Khan Academy.

### 1. The Population Mean ($\mu$)
The population mean is the average years of experience for the entire group.

*   **Data Set ($x$):** 1, 3, 5, 7, 14
*   **Number of Observations ($N$):** 5
*   **Calculation:** 
    $$\mu = \frac{\sum_{i=1}^{N} x_i}{N} = \frac{1 + 3 + 5 + 7 + 14}{5} = \frac{30}{5} = 6$$
*   **Result:** The mean years of experience is **6 years**.

---

### 2. The Population Variance ($\sigma^2$)
Variance measures the spread of data by determining the average squared distance of each data point from the mean. Squaring the differences ensures the result is positive and highlights larger deviations.

#### Step-by-Step Calculation:
1.  **Find the difference from the mean ($x - \mu$) and square it:**
    *   $(1 - 6)^2 = (-5)^2 = \mathbf{25}$
    *   $(3 - 6)^2 = (-3)^2 = \mathbf{9}$
    *   $(5 - 6)^2 = (-1)^2 = \mathbf{1}$
    *   $(7 - 6)^2 = (1)^2 = \mathbf{1}$
    *   $(14 - 6)^2 = (8)^2 = \mathbf{64}$

2.  **Sum the squared distances:**
    $$25 + 9 + 1 + 1 + 64 = 100$$

3.  **Divide by the total number of data points ($N$):**
    $$\sigma^2 = \frac{100}{5} = 20$$

*   **Result:** The population variance is **20**.

---

### Mathematical Notation
The process used above is represented by the following formula for **Population Variance**:

$$\sigma^2 = \frac{\sum_{i=1}^{N} (x_i - \mu)^2}{N}$$



```python
def calculate_population_metrics(data):
    """
    Calculates the arithmetic mean and population variance.
    """
    # Number of data points (N)
    n = len(data)
    
    # 1. Calculate the Population Mean (mu)
    # Sum of all points divided by N
    population_mean = sum(data) / n
    
    # 2. Calculate the Population Variance (sigma squared)
    # The average of the squared differences from the mean
    sum_squared_diff = sum((x - population_mean) ** 2 for x in data)
    population_variance = sum_squared_diff / n
    
    return population_mean, population_variance

# Data points: Years of experience for 5 people
years_of_experience = [1, 3, 5, 7, 14]

# Perform calculations
mu, sigma_sq = calculate_population_metrics(years_of_experience)

# Output results
print(f"Data Points: {years_of_experience}")
print(f"Population Mean (mu): {mu}")
print(f"Population Variance (sigma squared): {sigma_sq}")

# Verification of the manual steps from the summary
print("\nStep-by-step Variance Calculation:")
for x in years_of_experience:
    diff = x - mu
    sq_diff = diff ** 2
    print(f"Point: {x:2} | Diff: {diff:2} | Squared Diff: {sq_diff:2}")
```
