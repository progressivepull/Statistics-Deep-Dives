# Measures of Dispersion: Range, Variance, and Standard Deviation

This summary explains how to measure how spread out data values are, going beyond simply calculating the average (mean).

## 1. Two Data Sets With the Same Mean
- **Data Set A:** −10, 0, 10, 20, 30  
- **Data Set B:** 8, 9, 10, 11, 12  

Both have a **mean of 10**, even though Data Set A is much more spread out.

---

## 2. Range
The simplest measure of dispersion:

**Range = largest value − smallest value**

- **Data Set A:** 30 − (−10) = **40**  
- **Data Set B:** 12 − 8 = **4**

Range shows A is more dispersed, but range alone can be misleading.

---

## 3. Variance (Population Variance)

Variance measures the *average squared distance* from the mean.

**Formula:**  
σ² = ( Σ(xᵢ − μ)² ) / N

### Variance of Data Set A
- Squared distances: 400, 100, 0, 100, 400  
- Sum = 1000  
- Variance = 1000 / 5 = **200**

### Variance of Data Set B
- Squared distances: 4, 1, 0, 1, 4  
- Sum = 10  
- Variance = 10 / 5 = **2**

Data Set A has a much larger variance, indicating greater spread.

---

## 4. Standard Deviation

Standard deviation is the *square root of the variance*, giving a dispersion measure in the same units as the data.

- **Data Set A:** √200 = **10√2**  
- **Data Set B:** √2  

Data Set A’s standard deviation is **10 times** that of Data Set B.

---

# Key Takeaways
- Data sets can have the same mean but very different spreads.  
- **Range** is simple but incomplete.  
- **Variance** measures average squared deviation from the mean.  
- **Standard deviation** is widely used because it is easy to interpret.  
- Larger standard deviation = more dispersed data.



```python
import math

def calculate_stats(data):
    n = len(data)
    
    # Mean
    mean = sum(data) / n
    
    # Range
    data_range = max(data) - min(data)
    
    # Variance (population variance)
    variance = sum((x - mean) ** 2 for x in data) / n
    
    # Standard Deviation
    std_dev = math.sqrt(variance)
    
    return mean, data_range, variance, std_dev


# Example datasets
data_set_A = [-10, 0, 10, 20, 30]
data_set_B = [8, 9, 10, 11, 12]

# Calculate statistics
stats_A = calculate_stats(data_set_A)
stats_B = calculate_stats(data_set_B)

# Display results
print("Data Set A:", data_set_A)
print(f"Mean: {stats_A[0]}, Range: {stats_A[1]}, Variance: {stats_A[2]}, Std Dev: {stats_A[3]}\n")

print("Data Set B:", data_set_B)
print(f"Mean: {stats_B[0]}, Range: {stats_B[1]}, Variance: {stats_B[2]}, Std Dev: {stats_B[3]}")
```
