# Watermelon Seed Density Study: Statistical Summary

This summary outlines the process of using small samples to estimate the population parameters (seed density) of an entire watermelon farm without destroying the crop.

## 1. The Sample Data
A random sample of $n = 8$ cubic-inch chunks was taken, yielding the following seed counts:
**4, 3, 5, 7, 2, 9, 11, 7**

## 2. Measures of Central Tendency
To estimate the population mean ($\mu$), we calculate the **sample mean** ($\bar{x}$):

$$\bar{x} = \frac{\sum x_i}{n} = \frac{48}{8} = 6$$

*   **Result:** The estimated average seed density is **6 seeds per cubic inch**.

## 3. Measures of Spread
To estimate how much the seed density varies across the population, we calculate the variance and standard deviation.

### Unbiased Sample Variance ($s^2$)
To ensure the estimate is unbiased for the population, we divide the sum of squared deviations by $n-1$ (7) rather than $n$ (8):

1.  **Sum of squares:** $(4-6)^2 + (3-6)^2 + (5-6)^2 + (7-6)^2 + (2-6)^2 + (9-6)^2 + (11-6)^2 + (7-6)^2 = 66$
2.  **Calculation:** $s^2 = \frac{66}{7} \approx 9.43$

### Sample Standard Deviation ($s$)
The standard deviation is the square root of the unbiased sample variance:

$$s = \sqrt{9.43} \approx 3.07$$

## 4. Key Statistical Insight: The Bias Paradox
While $s^2$ is an **unbiased** estimator of population variance (because we divided by $n-1$), the sample standard deviation ($s$) is technically a **biased** estimator of the population standard deviation. 

*   **Why?** The square root function is non-linear, which introduces bias during the transformation.
*   **Application:** Despite this bias, the square root of the unbiased variance is the standard tool used in statistics because calculating a truly unbiased standard deviation depends heavily on the specific distribution of the population, making it too complex for general use.


```python
import math

def analyze_watermelon_seeds(data):
    n = len(data)
    
    # 1. Calculate Sample Mean
    sample_mean = sum(data) / n
    
    # 2. Calculate Unbiased Sample Variance
    # We use (n - 1) for the denominator to get an unbiased estimate
    sum_sq_diff = sum((x - sample_mean) ** 2 for x in data)
    unbiased_variance = sum_sq_diff / (n - 1)
    
    # 3. Calculate Sample Standard Deviation
    # Note: As discussed, this is the square root of the unbiased variance
    sample_std_dev = math.sqrt(unbiased_variance)
    
    return {
        "mean": sample_mean,
        "variance": unbiased_variance,
        "std_dev": sample_std_dev,
        "n": n
    }

# Data from the watermelon chunks
seed_counts = [4, 3, 5, 7, 2, 9, 11, 7]

# Perform analysis
results = analyze_watermelon_seeds(seed_counts)

# Output results
print(f"--- Watermelon Seed Analysis (n={results['n']}) ---")
print(f"Sample Mean: {results['mean']:.2f} seeds/cu in")
print(f"Unbiased Sample Variance: {results['variance']:.2f}")
print(f"Sample Standard Deviation: {results['std_dev']:.2f}")
print("-" * 40)
print("Note: The variance is unbiased because we divided by n-1.")
print("Note: The standard deviation is the square root of the unbiased variance.")
```
