# Summary: Mean and Mean Absolute Deviation (MAD) from a Bar Graph

## Scenario

A bar graph shows the number of bubbles blown by four different gum-chewers:

- **Manuela:** 4 bubbles  
- **Sophia:** 5 bubbles  
- **Jada:** 6 bubbles  
- **Tara:** 1 bubble  

The goals are to:
1. Find the **mean** number of bubbles blown.
2. Measure how spread out the data is using the **Mean Absolute Deviation (MAD)**.

---

## Step 1: Calculate the Mean

The **mean** is the sum of all values divided by the number of data points.

$$
\text{Mean} = \frac{4 + 5 + 6 + 1}{4} = \frac{16}{4} = 4
$$

So, the **mean number of bubbles blown is 4**.

---

## Step 2: Calculate Absolute Deviations from the Mean

The **absolute deviation** is the distance of each data point from the mean, ignoring direction.

| Gum-Chewer | Bubbles Blown | Absolute Deviation from Mean (4) |
|-----------|---------------|----------------------------------|
| Manuela  | 4             | $|4 - 4| = 0$                  |
| Sophia   | 5             | $|5 - 4| = 1$                  |
| Jada     | 6             | $|6 - 4| = 2$                  |
| Tara     | 1             | $|1 - 4| = 3$                  |

---

## Step 3: Calculate the Mean Absolute Deviation (MAD)

The **Mean Absolute Deviation** is the average of the absolute deviations.

$$
\text{MAD} = \frac{0 + 1 + 2 + 3}{4} = \frac{6}{4} = 1.5
$$

---

## Final Results

- **Mean:** 4  
- **Mean Absolute Deviation (MAD):** 1.5  

---

## Key Takeaways

- The mean gives a **measure of center**.
- The Mean Absolute Deviation gives a **measure of variability**.
- A MAD of 1.5 means that, on average, each gum-chewer’s bubble count differs from the mean by **1.5 bubbles**.
- Absolute value is used so deviations are measured as distances, not directions.

📌 **Conclusion:**  
While the mean shows the typical number of bubbles blown, the MAD explains how much individual values vary around that mean.


```python
import numpy as np

# Bubble data for each gum-chewer
# Manuela, Sophia, Jada, Tara
bubbles = np.array([4, 5, 6, 1])

# Calculate the mean
mean_bubbles = np.mean(bubbles)

# Calculate Mean Absolute Deviation (MAD)
mad_bubbles = np.mean(np.abs(bubbles - mean_bubbles))

# Display results
print("Bubble counts:", bubbles)
print("Mean number of bubbles:", mean_bubbles)
print("Mean Absolute Deviation (MAD):", mad_bubbles)
```

# Explanation

* **np.mean(bubbles)** computes the **mean**
* **np.abs(bubbles - mean_bubbles)** computes each **absolute deviation**
* Taking the mean of those deviations gives the MAD
