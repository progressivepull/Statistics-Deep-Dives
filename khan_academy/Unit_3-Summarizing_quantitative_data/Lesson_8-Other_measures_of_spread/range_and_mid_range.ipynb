# Range and Mid-Range

## Range

The **range** measures how spread out a set of numbers is.

- **Largest value:** 94  
- **Smallest value:** 65  

### Calculation

$$
\text{Range} = 94 - 65 = 29
$$

A larger range means the data is more spread out.  
A smaller range means the data is more tightly clustered.

---

## Mid-Range

The **mid-range** is a simple measure of central tendency.  
It is the average of the largest and smallest values.

### Calculation

$$
\text{Mid-range} = \frac{94 + 65}{2}
$$

$$
94 + 65 = 159
$$

$$
\frac{159}{2} = 79.5
$$

So, the **mid-range is 79.5**.

---

## Notes

Other measures of central tendency include:

- **Mean** – arithmetic average of all values  
- **Median** – middle value when the data is sorted  
- **Mode** – most frequently occurring value


```python
# Example: Calculating range and mid-range in Python

numbers = [94, 65, 72, 88, 79, 91]

# Range = max - min
data_range = max(numbers) - min(numbers)

# Mid-range = (max + min) / 2
mid_range = (max(numbers) + min(numbers)) / 2

print("Numbers:", numbers)
print("Range:", data_range)
print("Mid-range:", mid_range)
```
