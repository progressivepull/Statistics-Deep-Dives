# Summary: Mean and Mean Absolute Deviation (MAD)

## Comparing Two Data Sets

We examine two different data sets:

- **Data Set A:** 2, 2, 4, 4  
- **Data Set B:** 1, 1, 6, 4  

Both data sets contain four values.

---

## Mean (Measure of Center)

The **mean** is found by summing all values and dividing by the number of values.

- **Mean of Data Set A:**  
  $$
  \frac{2 + 2 + 4 + 4}{4} = \frac{12}{4} = 3
  $$

- **Mean of Data Set B:**  
  $$
  \frac{1 + 1 + 6 + 4}{4} = \frac{12}{4} = 3
  $$

✅ Both data sets have the **same mean (3)**.

---

## Why the Data Sets Feel Different

Although the means are the same, the data sets look different when visualized.  
This is because they differ in how **spread out** the values are around the mean.

To describe this spread, we use a measure of variability.

---

## Mean Absolute Deviation (MAD)

The **Mean Absolute Deviation (MAD)** measures how far, on average, data points are from the mean.

### Steps to Calculate MAD
1. Find the mean.
2. Compute the absolute deviation of each data point from the mean.
3. Take the average of those absolute deviations.

---

## MAD for Each Data Set

### Data Set A: 2, 2, 4, 4 (Mean = 3)

Absolute deviations:
- |2 − 3| = 1  
- |2 − 3| = 1  
- |4 − 3| = 1  
- |4 − 3| = 1  

$$
\text{MAD} = \frac{1 + 1 + 1 + 1}{4} = 1
$$

### Data Set B: 1, 1, 6, 4 (Mean = 3)

Absolute deviations:
- |1 − 3| = 2  
- |1 − 3| = 2  
- |6 − 3| = 3  
- |4 − 3| = 1  

$$
\text{MAD} = \frac{2 + 2 + 3 + 1}{4} = 2
$$

---

## Key Takeaways

- Both data sets have the **same mean (3)**.
- **Data Set B** is more spread out than **Data Set A**.
- This difference is captured by **Mean Absolute Deviation**:
  - MAD of Data Set A = **1**
  - MAD of Data Set B = **2**

📌 **Conclusion:**  
The mean describes the center of the data, but the **Mean Absolute Deviation** describes how spread out the data is around that center.


```python
import numpy as np

# Define the data sets
data_set_a = np.array([2, 2, 4, 4])
data_set_b = np.array([1, 1, 6, 4])

# Function to calculate Mean Absolute Deviation (MAD)
def mean_absolute_deviation(data):
    mean = np.mean(data)
    mad = np.mean(np.abs(data - mean))
    return mean, mad

# Calculate mean and MAD for each data set
mean_a, mad_a = mean_absolute_deviation(data_set_a)
mean_b, mad_b = mean_absolute_deviation(data_set_b)

# Display results
print("Data Set A:", data_set_a)
print("Mean:", mean_a)
print("Mean Absolute Deviation (MAD):", mad_a)

print("\nData Set B:", data_set_b)
print("Mean:", mean_b)
print("Mean Absolute Deviation (MAD):", mad_b)
```

# Key Insight

* Both data sets have the same mean (3)
* Data Set B has a larger MAD, showing it is more spread out
