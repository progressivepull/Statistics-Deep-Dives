Below is a clean **bullet summary**, a clear **explanation**, and a
**Python code example** that mirrors the logic of sampling distributions
for a **sample proportion**. I’ll also embed helpful links to concepts
like the Bernoulli variable, the binomial distribution, and the sampling
distribution.

**⭐ Bullet Summary**

- A gumball machine contains multiple colors; we focus on the **yellow**
  ones.

- The **population proportion** of yellow gumballs is $P = 0.6$.

- Define a **Bernoulli random variable** $Y$:

  - $Y = 1$if a randomly selected gumball is yellow

  - $Y = 0$otherwise

- For this Bernoulli variable:

  - Mean = $P = 0.6$

  - Standard deviation = $\sqrt{P(1 - P)}$

- Define a **binomial random variable** $X$as the sum of 10
  independent Bernoulli trials.

- For this binomial variable:

  - Mean = $nP = 10 \cdot 0.6 = 6$

  - Standard deviation = $\sqrt{nP(1 - P)}$

- The **sample proportion** $\widehat{p}$is defined as $X/n$.

- The **sampling distribution** of $\widehat{p}$:

  - Mean = $P$

  - Standard deviation = $\sqrt{\frac{P(1 - P)}{n}}$

- With $P = 0.6$and $n = 10$:

  - $\text{SD}(\widehat{p}) \approx 0.15$

- This shows that the sample proportion is an **unbiased estimator** of
  the population proportion.

**⭐ Explanation (Intuitive and Clear)**

The machine has many gumballs, and 60% of them are yellow. That 60% is
the **population proportion**, a fixed truth about the entire machine.

When you take **one** gumball, the event “yellow or not yellow” is
modeled by a Bernoulli variable. Its mean is simply the probability of
success, which is the proportion of yellow gumballs.

When you take **10** gumballs (with replacement or using the 10% rule),
the number of yellows you get is a binomial random variable. Its mean is
$nP$, and its standard deviation is $\sqrt{nP(1 - P)}$.

But we’re not just counting yellows—we’re computing the **sample
proportion**:

$$
\widehat{p} = \frac{X}{n}
$$

This is a statistic used to estimate the true population proportion
$P$.

If you repeatedly take samples of size 10 and compute
$\widehat{p}$each time, the distribution of all those
$\widehat{p}$values is the **sampling distribution** of the sample
proportion.

Key facts:

- The mean of this sampling distribution is exactly $P$. → This makes
  $\widehat{p}$an **unbiased estimator**.

- The standard deviation shrinks as $n$grows:

$$
\sqrt{\frac{P(1 - P)}{n}}
$$

For $P = 0.6$and $n = 10$, this becomes:

$$
\sqrt{\frac{0.6 \cdot 0.4}{10}} \approx 0.15
$$

So most sample proportions will fall within about 0.15 of the true
value.

If you want to explore how this behaves for larger $n$, you might
enjoy looking at the Central Limit Theorem.


# ⭐ Python Code Example (Simulating the Sampling Distribution)
This simulation:

* Creates a population with 60% yellow
* Repeatedly draws samples of size 10
* Computes the sample proportion each time
* Plots the resulting sampling distribution


```python
import numpy as np
import matplotlib.pyplot as plt

# Population proportion
P = 0.6

# Simulation settings
n = 10          # sample size
trials = 10000  # number of simulated samples

sample_props = []

for _ in range(trials):
    # Bernoulli trials: 1 = yellow, 0 = not yellow
    sample = np.random.choice([1, 0], size=n, p=[P, 1-P])
    sample_props.append(np.mean(sample))

# Plot histogram
plt.hist(sample_props, bins=np.linspace(0,1,22), density=True, alpha=0.7)
plt.axvline(P, color='red', linestyle='--', label='True P = 0.6')
plt.xlabel("Sample Proportion")
plt.ylabel("Density")
plt.title("Sampling Distribution of Sample Proportion (n=10)")
plt.legend()
plt.show()

# Print theoretical SD
sd_hat_p = np.sqrt(P*(1-P)/n)
print("Theoretical SD of sample proportion:", sd_hat_p)
```

```python
This will show:

* The distribution is centered at 0.6
* The spread matches the theoretical standard deviation (~0.15)
* The sample proportion is indeed an unbiased estimator

---

If you want, I can also simulate the binomial distribution, compare mean vs. proportion, or show how the sampling distribution tightens as 
𝑛
 increases.
```
