**⭐ Bullet Summary**

- The **sampling distribution** of a sample proportion can be
  **normal**, **right‑skewed**, or **left‑skewed**.

- A common rule of thumb says the distribution is **approximately
  normal** when both

$$
np \geq 10  \text{ and }  n(1 - p) \geq 10
$$

where $n$is sample size and $p$is the population proportion.

- If $np < 10$, the distribution tends to be **right‑skewed** (mean
  near 0).

- If $n(1 - p) < 10$, the distribution tends to be **left‑skewed**
  (mean near 1).

- Larger sample sizes shrink the standard deviation

$$
\sqrt{\frac{p(1 - p)}{n}}
$$

making the distribution tighter and more normal‑looking.

- Example 1 (tangerines): $n = 50$, $p = 0.12$.

  - $np = 6 < 10$→ **right‑skewed**.

- Example 2 (radio reach): $n = 125$, $p = 0.88$.

  - $np \geq 10$and $n(1 - p) \geq 10$→ **approximately normal**.

You can explore related ideas like normal approximation, sample
proportion, or skewness.

**⭐ Explanation (Intuitive and Clear)**

The shape of the **sampling distribution of the sample proportion**
depends on how far the true proportion $p$is from 0.5 and how large
the sample size $n$is.

**✔ When both** $\mathbf{np}$**and** $\mathbf{n(1 - p)}$**are at
least 10**

The distribution is **approximately normal**. This happens because there
are enough expected “successes” and “failures” for the binomial
distribution to behave symmetrically.

**✔ When** $\mathbf{np < 10}$

The mean is near **0**, so the distribution is **right‑skewed**. There
is a long tail to the right because proportions cannot go below 0 but
can vary upward.

**✔ When** $\mathbf{n(1 - p) < 10}$

The mean is near **1**, so the distribution is **left‑skewed**. There is
a long tail to the left because proportions cannot exceed 1 but can vary
downward.

**Example 1 — Tangerines**

- $n = 50$

- $p = 0.12$

- $np = 6 < 10$→ fails

- $n(1 - p) = 44 \geq 10$→ passes

Since $np < 10$, the distribution is **right‑skewed**, centered at
$p = 0.12$.

**Example 2 — Radio Reach**

- $n = 125$

- $p = 0.88$

- $np = 110 \geq 10$

- $n(1 - p) = 15 \geq 10$

Both conditions pass → **approximately normal**.

Even though $p$is close to 1, the large sample size makes the

# ⭐ Python Code Example

This script checks the rule‑of‑thumb conditions and simulates the sampling distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_sampling_distribution(p, n, num_samples=5000):
    # Generate sample proportions
    props = np.mean(
        np.random.binomial(1, p, size=(num_samples, n)),
        axis=1
    )
    return props

def check_conditions(p, n):
    np_val = n * p
    nq_val = n * (1 - p)
    shape = "approximately normal" if np_val >= 10 and nq_val >= 10 else \
            "right-skewed" if np_val < 10 else \
            "left-skewed"
    return np_val, nq_val, shape

# Example: Tangerines
p = 0.12
n = 50

np_val, nq_val, shape = check_conditions(p, n)
print("np =", np_val)
print("n(1-p) =", nq_val)
print("Predicted shape:", shape)

# Simulate and plot
props = simulate_sampling_distribution(p, n)
plt.hist(props, bins=20, density=True, alpha=0.7, color='orange')
plt.title(f"Sampling Distribution (p={p}, n={n})")
plt.xlabel("Sample Proportion")
plt.ylabel("Density")
plt.show()
```
