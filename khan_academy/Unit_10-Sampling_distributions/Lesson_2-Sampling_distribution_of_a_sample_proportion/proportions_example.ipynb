# Probability of sample proportions example

[Probability of sample proportions example | Sampling distributions | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=BkRJJsHoRdw&t=5s)

Suppose that 15% of the 1750 students at a school have experienced
extreme levels of stress during the past month. A high school newspaper
doesn't know this figure, but they are curious what it is, so they
decide to ask a simple random sample of 160 students if they have
experienced extreme levels of stress during the past month.
Subsequently, they find that 10% of the sample replied “yes” to the
question.

Assuming the true proportion is 15%, what is the approximate probability
that more than 10% of the sample would report that they experienced
extreme levels of stress during the past month?

Among the choices:

- \(A\) P(p\>0.10) ≈ 0.92

- \(B\) P(p\>0.10) ≈ 0.94

- \(C\) P(p\>0.10) ≈ 0.96

- \(D\) P(p\>0.10) ≈ 0.98

**Bullet summary**

- **Population proportion:** $p = 0.15$, **sample size:** $n = 160$.

- **Check normal approximation:**

$$
np = 160(0.15) = 24 \geq 10,n(1 - p) = 160(0.85) = 136 \geq 10
$$

so $\widehat{p}$is approximately normal.

- **Sampling distribution:**

$$
\mu_{\widehat{p}} = p = 0.15,\sigma_{\widehat{p}} = \sqrt{\frac{p(1 - p)}{n}} \approx 0.028
$$

- **Target probability:** $P(\widehat{p} > 0.10)$.

- **Z-score:**

$$
z = \frac{0.10 - 0.15}{0.028} \approx - 1.79
$$

- **Probability:**

$$
P(\widehat{p} > 0.10) = P(Z > - 1.79) = P(Z < 1.79) \approx 0.96
$$

- **Answer:** Approximate probability $\approx 0.96$(96%).

**Step-by-step explanation**

We treat the sample proportion $\widehat{p}$as a random variable.
Under simple random sampling with a large enough $n$, the sampling
distribution of $\widehat{p}$is approximately normal with:

$$
\mu_{\widehat{p}} = p,\sigma_{\widehat{p}} = \sqrt{\frac{p(1 - p)}{n}}
$$

Given $p = 0.15$and $n = 160$:

$$
\sigma_{\widehat{p}} = \sqrt{\frac{0.15 \cdot 0.85}{160}} \approx \sqrt{0.000796875} \approx 0.028
$$

We want:

$$
P(\widehat{p} > 0.10)
$$

Convert 0.10 to a z-score:

$$
z = \frac{0.10 - 0.15}{0.028} \approx \frac{- 0.05}{0.028} \approx - 1.79
$$

So:

$$
P(\widehat{p} > 0.10) = P(Z > - 1.79) = P(Z < 1.79) \approx 0.963
$$

Rounded, that’s about $0.96$, or 96%.


# Python code example (theory + simulation)

```python
import math
from mpmath import quad, exp, sqrt, pi
import random

# Parameters
p = 0.15
n = 160
threshold = 0.10

# Theoretical normal approximation
mu = p
sigma = math.sqrt(p * (1 - p) / n)

z = (threshold - mu) / sigma

# Standard normal CDF using numerical integration
def phi(x):
    return (1 / math.sqrt(2 * math.pi)) * math.exp(-x**2 / 2)

def normal_cdf(z):
    return 0.5 * (1 + math.erf(z / math.sqrt(2)))

prob_theoretical = 1 - normal_cdf(z)  # P(hat p > threshold)

print("Mean (mu):", mu)
print("SD (sigma):", sigma)
print("z-score:", z)
print("Theoretical P(hat p > 0.10):", prob_theoretical)

# Monte Carlo simulation to validate
trials = 100_000
count = 0

for _ in range(trials):
    # Simulate number of stressed students in sample of size n
    stressed = sum(1 for _ in range(n) if random.random() < p)
    phat = stressed / n
    if phat > threshold:
        count += 1

prob_sim = count / trials
print("Simulated P(hat p > 0.10):", prob_sim)
```

This will print the theoretical probability (≈0.96) and a simulated estimate that should be very close, reinforcing the normal approximation intuition.
