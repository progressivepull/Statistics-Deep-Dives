# Sample statistic bias worked example

Alejandro was curious if sample median was an unbiased estimator of population median. He placed ping pong balls numbered from 0 to 32 in a drum and mixed them well. Note that the median of the population is 16.

He then took a random sample of 5 balls and calculated the median of the sample. He replaced the balls and repeated this process for a total of 50 trials. His results are summarized in the dotplot below, where each dot represents the sample median from a sample of 5 balls.

![plot.png](./../IMAGES/Lesson_1-What_is_a_sampling_distribution/plot.png)

Based on these results, does sample median appear to be a biased or unbiased estimator of population median?

Here’s a clean, structured breakdown of Alejandro’s experiment — a
**bullet summary**, a clear **explanation**, and a **Python simulation**
that mirrors what he did. I’ll also weave in helpful links so you can
explore ideas like the sampling distribution, the sample median, and
bias in estimators.

**⭐ Bullet Summary**

- Alejandro has a population of 33 ping‑pong balls numbered 0–32.

- The **population median** is 16 — a fixed population parameter.

- He repeatedly draws a **sample of 5 balls**, with replacement, and
  computes the **sample median** each time.

- He repeats this process 50 times, producing 50 sample medians.

- These medians are plotted in a dot plot, forming an approximation of
  the **sampling distribution of the sample median**.

- To judge whether the sample median is an **unbiased estimator**, we
  compare the center of this sampling distribution to the true
  population median.

- The dot plot shows medians roughly balanced around 16 — similar number
  above and below.

- Therefore, the sample median appears to be an **unbiased estimator**
  of the population median in this context.

**⭐ Explanation (Intuitive and Clear)**

Alejandro wants to know whether the **sample median** tends to hit the
true **population median** on average.

He knows:

- The population is the set of numbers 0 through 32.

- The true median of this population is 16.

- If the sample median is an **unbiased estimator**, then across many
  samples, the sample medians should cluster around 16.

He draws 5 balls, computes the median, replaces them, and repeats this
50 times. Each dot on the plot is one of those medians.

How do we judge bias?

- If the sampling distribution were **skewed** — mostly above or mostly
  below 16 — that would suggest bias.

- But Alejandro’s dot plot shows a roughly symmetric spread around 16.

- About as many sample medians fall below 16 as above it.

This symmetry suggests that the **expected value** of the sample median
equals the population median — the definition of an **unbiased
estimator**.

So based on the experiment, the sample median behaves like an unbiased
estimator of the population median.

If you want to explore deeper, you might enjoy looking at the sampling

```python
# ⭐ Python Code Example (Simulating Alejandro’s Experiment)

Below is a simulation that:

* Creates the population 0–32
* Repeatedly samples 5 values with replacement
* Computes the sample median
* Plots the resulting sampling distribution
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Population: numbers 0 through 32
population = np.arange(33)
population_median = np.median(population)

# Simulation parameters
n = 5          # sample size
trials = 50    # number of samples (Alejandro's experiment)

sample_medians = []

for _ in range(trials):
    sample = np.random.choice(population, size=n, replace=True)
    sample_medians.append(np.median(sample))

# Display results
print("Population median:", population_median)
print("Sample medians:", sample_medians)

plt.figure(figsize=(10, 4))
plt.scatter(sample_medians, np.zeros_like(sample_medians), alpha=0.7)
plt.yticks([])
plt.xlabel("Sample Median")
plt.title("Sampling Distribution Approximation (50 Trials)")
plt.show()
```

This code gives you a dot plot similar to Alejandro’s.
If you increase trials to something like 10,000, you’ll get a much smoother approximation of the true sampling distribution.
