# Inferring population mean from sample mean

[Inferring population mean from sample mean | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=k5EbijWu-Ss&t=11s)

**Bullet summary**

- **Goal:** Estimate the **population mean height** of men in the U.S.
  without measuring everyone.

- **Population vs sample:**

  - **Population:** All men in the U.S. (~150 million).

  - **Sample:** A smaller subset (e.g., 5 men, 100 men, 1,000 men)
    chosen to represent the population.

- **Sample mean (**$\overset{ˉ}{x}$**):**

$$
\overset{ˉ}{x} = \frac{x_{1} + x_{2} + \cdots + x_{n}}{n}
$$

It’s a **statistic**—a number computed from sample data.

- **Population mean (**$\mu$**):**

$$
\mu = \frac{x_{1} + x_{2} + \cdots + x_{N}}{N}
$$

It’s a **parameter**—a (usually unknown) number describing the entire
population.

- **Key idea in statistics:** Use **sample statistics** (like
  $\overset{ˉ}{x}$) to **infer** or estimate **population parameters**
  (like $\mu$).

- **Notation:**

  - **Sample mean:**
    $\overset{ˉ}{x} = \frac{1}{n}\sum_{i = 1}^{n}x_{i}$

  - **Population mean:** $\mu = \frac{1}{N}\sum_{i = 1}^{N}x_{i}$

- **Practical constraint:** Measuring everyone is usually impossible, so
  we rely on **random samples** to get reasonable estimates.

**Explanation in plain language**

You want to design a product that depends on men’s height in the U.S.
Ideally, you’d like to know the **true average height** of all men in
the country—that’s the **population mean**, denoted by $\mu$. In
theory, you’d measure every man, add all their heights, and divide by
the total number of men. But with around 150 million men, that’s not
realistic.

So instead, you take a **sample**—a smaller group of men that you hope
is **representative** of the population. The key is that the sample
should be **random**, not biased toward certain groups (like only
basketball players or only people at a gym).

From that sample, you compute the **sample mean**, denoted
$\overset{ˉ}{x}$. For example, with 5 sampled heights:

- $x_{1} = 6.2$

- $x_{2} = 5.5$

- $x_{3} = 5.75$

- $x_{4} = 6.3$

- $x_{5} = 5.9$

You compute:

$$
\overset{ˉ}{x} = \frac{6.2 + 5.5 + 5.75 + 6.3 + 5.9}{5} = 5.93\text{ feet}
$$

This **5.93 feet** is your **sample mean**—a **statistic**. You use it
as an estimate of the **population mean** $\mu$, which is a
**parameter**. Most of statistics is about this: using **sample
statistics** to say something about **population parameters** we can’t
directly observe.

The notation with $\sum$(sigma) is just a compact way of saying “add
all the values”:

- Sample mean:

$$
\overset{ˉ}{x} = \frac{1}{n}\sum_{i = 1}^{n}x_{i}
$$

- Population mean:

$$
\mu = \frac{1}{N}\sum_{i = 1}^{N}x_{i}
$$

Same operation—add and divide—just over different sizes: $n$for
sample, $N$for population.

**Python code example**

Let’s mirror the story: compute the sample mean from a small sample, and
conceptually compare it to a (simulated) population mean.



```python
import random
import statistics as stats

# --- Simulate a "true" population of men's heights (in feet) ---
# For illustration, assume population heights are roughly normal around 5.9 ft with some variation.
random.seed(0)

N = 150_000  # pretend this is our "population" size (much smaller than reality, but enough to illustrate)
population_heights = [random.gauss(5.9, 0.3) for _ in range(N)]

# True population mean (parameter mu)
mu = stats.mean(population_heights)
print(f"True population mean (mu): {mu:.3f} ft")

# --- Take a random sample from this population ---
n = 5  # small sample, like in the example
sample = random.sample(population_heights, n)

print("Sample heights:", [round(h, 3) for h in sample])

# Sample mean (statistic x̄)
x_bar = stats.mean(sample)
print(f"Sample mean (x̄): {x_bar:.3f} ft")

# --- Increase sample size to see stability ---
for n_large in [30, 100, 1000]:
    sample_large = random.sample(population_heights, n_large)
    x_bar_large = stats.mean(sample_large)
    print(f"Sample size {n_large:4d} -> sample mean: {x_bar_large:.3f} ft")
```

**What this code shows:**

- We create a **synthetic population** of heights and compute its **true
  mean** $\mu$.

- We then take a **small sample** (size 5) and compute
  $\overset{ˉ}{x}$. It may be somewhat off from $\mu$.

- As we increase the **sample size** (30, 100, 1000), the sample mean
  tends to get **closer** to the true population mean—illustrating why
  larger, random samples give more reliable estimates of $\mu$.
