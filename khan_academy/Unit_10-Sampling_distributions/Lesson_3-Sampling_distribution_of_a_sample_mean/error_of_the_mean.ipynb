**⭐ Bullet Summary (from the attached document)**

- Start with **any distribution**, even a wildly non‑normal one.

- Take repeated samples of size $n$, compute the **sample mean**, and
  plot those means.

- The resulting distribution is the **sampling distribution** of the
  sample mean.

- As **sample size** $n$increases:

  - The sampling distribution becomes **more normal** (Central Limit
    Theorem).

  - The sampling distribution becomes **narrower** (smaller standard
    deviation).

- The **mean** of the sampling distribution is always the **population
  mean** $\mu$.

- The **variance** of the sampling distribution is:

$$
\frac{\sigma^{2}}{n}
$$

where $\sigma^{2}$is the variance of the original distribution.

- The **standard deviation** of the sampling distribution (the
  **standard error**) is:

$$
\frac{\sigma}{\sqrt{n}}
$$

- Larger $n$→ smaller standard error → sample means cluster tightly
  around $\mu$.

- Simulations confirm the formula:

  - Example: original SD = 9.3

  - For $n = 16$: predicted SE = $9.3/4 = 2.32$(matches simulation ≈
    2.33)

  - For $n = 25$: predicted SE = $9.3/5 = 1.86$(matches simulation ≈
    1.87)

**⭐ Explanation (intuitive)**

The attached document builds on the **Central Limit Theorem** and shows
how the **spread** of the sampling distribution changes with sample
size.

**1. Shape becomes more normal**

Even if the original distribution is jagged, discrete, or multimodal,
the sampling distribution of the mean becomes smoother and more
bell‑shaped as $n$increases.

**2. Spread becomes smaller**

Averaging reduces variability. If you average 100 values, it’s extremely
unlikely that all 100 are extreme outliers. So the sample mean stays
close to the true mean.

This shrinking spread is captured by:

$$
\text{SE} = \frac{\sigma}{\sqrt{n}}
$$

**3. Why the formula makes sense**

- If $n$doubles, the denominator $\sqrt{n}$increases → SE decreases.

- If $n$becomes huge, SE becomes tiny → sample means cluster tightly
  around $\mu$.

- If $n = 1$, SE = $\sigma$→ sampling distribution = original
  distribution.

**4. Simulation confirms the math**

The document shows a simulation where:

- Original SD = 9.3

- For $n = 16$: SE ≈ 2.33

- For $n = 25$: SE ≈ 1.87

Both match the theoretical formula almost exactly.

This is the heart of statistical inference:

**Larger samples give more precise estimates of the population mean.**

**⭐ Python Code Example — Demonstrating the Formula and the CLT**

This code simulates sampling distributions for different sample sizes
and compares the empirical standard error to the theoretical value.




```python
import numpy as np
import matplotlib.pyplot as plt

# --- Create a weird, non-normal distribution ---
np.random.seed(0)
values = np.array([0, 1, 2, 5, 9])
probs  = np.array([0.30, 0.10, 0.25, 0.05, 0.30])

# True population mean and SD
mu = np.sum(values * probs)
sigma = np.sqrt(np.sum((values - mu)**2 * probs))

print("Population mean:", mu)
print("Population SD:", sigma)

# --- Function to generate sampling distribution ---
def sampling_distribution(n, num_samples=10000):
    means = []
    for _ in range(num_samples):
        sample = np.random.choice(values, size=n, p=probs)
        means.append(np.mean(sample))
    return np.array(means)

# --- Compare different sample sizes ---
sample_sizes = [5, 16, 25, 50]

for n in sample_sizes:
    means = sampling_distribution(n)
    empirical_se = np.std(means)
    theoretical_se = sigma / np.sqrt(n)

    print(f"\nn = {n}")
    print("Empirical SE:", round(empirical_se, 3))
    print("Theoretical SE:", round(theoretical_se, 3))

    plt.hist(means, bins=30, density=True, alpha=0.6)
    plt.title(f"Sampling Distribution (n={n})")
    plt.xlabel("Sample Mean")
    plt.ylabel("Density")
    plt.show()
```

**What you’ll see:**

- As $n$increases, the histogram becomes **more normal**.

- The spread shrinks exactly according to $\sigma/\sqrt{n}$.

- Empirical and theoretical standard errors match closely.
