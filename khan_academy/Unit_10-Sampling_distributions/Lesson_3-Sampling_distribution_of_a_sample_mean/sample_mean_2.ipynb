**⭐ Bullet Summary (from the attached document)**

- The **sampling distribution** of the **sample mean** becomes more
  normal as the **sample size** $n$increases.

- When $n = 1$, the sampling distribution is just the original
  distribution — **not normal at all**.

- When $n = 2$, the distribution begins to smooth out but still has
  gaps and is **not close to normal**.

- When $n \geq 10$, the sampling distribution becomes **very close to
  a normal distribution**, even if the original distribution is
  extremely non‑normal.

- As $n$increases:

  - The sampling distribution becomes **more normal** (CLT).

  - The **standard deviation decreases**, meaning the distribution
    becomes **more concentrated around the mean**.

  - The mean of the sampling distribution **equals the population
    mean**.

- The standard deviation of the sampling distribution (the **standard
  error**) is:

$$
\frac{\sigma}{\sqrt{n}}
$$

- Larger $n$→ lower standard error → sample means cluster tightly
  around the true mean.

- Simulations show:

  - Small $n$→ skew and kurtosis far from normal.

  - Large $n$→ skew and kurtosis close to 0 (normal‑like).

- Even bizarre, multimodal, discrete distributions produce nearly normal
  sampling distributions when $n$is moderately large.

**⭐ Explanation (intuitive)**

The attached document walks through a powerful idea:

**The shape of the sampling distribution depends heavily on the sample
size** $n$**.**

**When** $\mathbf{n = 1}$

You’re not averaging anything — you’re just sampling the original
distribution. So the sampling distribution **looks exactly like the
original**, no matter how weird it is.

**When** $\mathbf{n = 2}$

You average two values. The distribution becomes a bit smoother, but
still has gaps and is not normal.

**When** $\mathbf{n = 4}$**,** $\mathbf{n = 5}$

You start to see the **bell‑curve shape** emerging.

**When** $\mathbf{n = 10}$**,** $\mathbf{n = 16}$**,**
$\mathbf{n = 25}$

The sampling distribution becomes:

- **More normal**

- **More symmetric**

- **Less skewed**

- **Less kurtotic**

- **More tightly concentrated around the mean**

This happens **even if the original distribution is wildly non‑normal**.

**Why does the spread shrink?**

Because averaging reduces variability. If you average **25** values,
it’s extremely unlikely that all 25 are extreme outliers. So the sample
mean stays close to the true mean.

This is why the standard deviation of the sampling distribution is:

$$
\frac{\sigma}{\sqrt{n}}
$$

As $n$grows, the denominator grows → the spread shrinks.

This is the heart of the **Central Limit Theorem**.

**⭐ Python Code Example — Showing How **$\mathbf{n}$** Changes the
Sampling Distribution**

This simulation uses a **very non‑normal distribution** and compares
different sample sizes.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Create a bizarre, non-normal distribution ---
np.random.seed(0)
values = np.array([0, 1, 2, 5, 9])          # weird discrete values
probs  = np.array([0.30, 0.10, 0.25, 0.05, 0.30])  # uneven probabilities

# True population mean
mu = np.sum(values * probs)
print("Population mean:", mu)

# --- Function to generate sampling distribution ---
def sampling_distribution(n, num_samples=10000):
    means = []
    for _ in range(num_samples):
        sample = np.random.choice(values, size=n, p=probs)
        means.append(np.mean(sample))
    return np.array(means)

# --- Compare different sample sizes ---
sample_sizes = [1, 2, 5, 10, 25]

plt.figure(figsize=(12, 8))

for i, n in enumerate(sample_sizes, 1):
    plt.subplot(2, 3, i)
    means = sampling_distribution(n)
    plt.hist(means, bins=30, density=True, alpha=0.7)
    plt.title(f"n = {n}")
    plt.xlabel("Sample Mean")
    plt.ylabel("Density")

plt.tight_layout()
plt.show()
```

**What you’ll see:**

- **n = 1** → looks exactly like the original distribution

- **n = 2** → still jagged, not normal

- **n = 5** → noticeably smoother

- **n = 10** → close to normal

- **n = 25** → very close to a perfect bell curve, tightly centered
  around the mean

This is the Central Limit Theorem in action.
