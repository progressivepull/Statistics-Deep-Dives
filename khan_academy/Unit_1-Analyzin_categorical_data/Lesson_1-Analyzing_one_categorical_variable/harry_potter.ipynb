# Hogwarts House Points Summary

## Overview
The bar chart displays the number of house points earned today by the four Hogwarts houses:
- Gryffindor  
- Hufflepuff  
- Ravenclaw  
- Slytherin  

## Hufflepuff's Points
By matching each bar to its corresponding house name, the bar beginning with **H** represents **Hufflepuff**.  
Hufflepuff earned **3 points** today.

## Slytherin's Points
The bar labeled for **Slytherin** reaches **4 points**.  
Slytherin earned **4 points** today.


```python
# Example: Working with Hogwarts house points in Python

# Store today's house points in a dictionary
house_points = {
    "Gryffindor": 5,
    "Hufflepuff": 3,
    "Ravenclaw": 2,
    "Slytherin": 4
}

# Access specific values
hufflepuff_points = house_points["Hufflepuff"]
slytherin_points = house_points["Slytherin"]

print("Hufflepuff earned:", hufflepuff_points)
print("Slytherin earned:", slytherin_points)

# Optional: visualize with a simple bar chart
import matplotlib.pyplot as plt

houses = list(house_points.keys())
points = list(house_points.values())

plt.bar(houses, points, color=["red", "yellow", "blue", "green"])
plt.title("Hogwarts House Points Today")
plt.ylabel("Points")
plt.show()
```
