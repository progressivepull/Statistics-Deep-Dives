# Conditional probability explained visually

[Conditional Probability | YouTube](https://www.youtube.com/watch?v=KqCKZwh5WY8&t=3s)

### Summary

The story illustrates how conditional probability can be understood using **probability trees** and updating beliefs when new evidence is observed.

**Two-coin scenario:**
Bob has one fair coin and one double-sided (always heads) coin. He picks one at random and flips it.
- After one flip resulting in **heads**, the probability Bob chose the fair coin is **1/3**.
- After two consecutive **heads**, the probability the coin is fair drops to **1/5**.
As more heads occur, confidence that the coin is unfair increases, but it never reaches 100%.

**Key idea:**  
Each new flip grows the probability tree. When evidence is observed (like “heads”), branches inconsistent with that evidence are removed, and probabilities are recalculated using the remaining outcomes.

**Three-coin scenario:**
Bob now has three coins: two fair coins and one biased coin (heads with probability 2/3).
- After observing **heads**, the probability Bob chose the biased coin is **4/10 = 40%**.

**Main takeaway:**  
All conditional probability problems can be solved by:
1. Building a balanced probability tree  
2. Trimming branches that contradict observed evidence  
3. Recomputing probabilities from the remaining outcomes  

This method is equivalent to using **Bayes’ Theorem**, but trees provide an intuitive, visual approach.


```python
# Conditional probability examples using simple calculations

# -------------------------------
# Example 1: One fair coin, one double-sided coin
# -------------------------------

# Prior probabilities
P_fair = 0.5
P_double = 0.5

# Likelihoods
P_heads_given_fair = 0.5
P_heads_given_double = 1.0

# After one head
P_heads = (P_heads_given_fair * P_fair) + (P_heads_given_double * P_double)
P_fair_given_heads = (P_heads_given_fair * P_fair) / P_heads

print("After one head:")
print("P(fair | heads) =", round(P_fair_given_heads, 3))

# After two heads
P_two_heads_given_fair = 0.5 ** 2
P_two_heads_given_double = 1.0 ** 2

P_two_heads = (
    P_two_heads_given_fair * P_fair
    + P_two_heads_given_double * P_double
)

P_fair_given_two_heads = (
    P_two_heads_given_fair * P_fair
) / P_two_heads

print("\nAfter two heads:")
print("P(fair | two heads) =", round(P_fair_given_two_heads, 3))


# -------------------------------
# Example 2: Two fair coins, one biased coin
# -------------------------------

# Prior probabilities
P_fair = 2 / 3
P_biased = 1 / 3

# Likelihoods
P_heads_given_fair = 0.5
P_heads_given_biased = 2 / 3

# Total probability of heads
P_heads = (
    P_heads_given_fair * P_fair
    + P_heads_given_biased * P_biased
)

# Posterior probability
P_biased_given_heads = (
    P_heads_given_biased * P_biased
) / P_heads

print("\nTwo fair coins, one biased coin:")
print("P(biased | heads) =", round(P_biased_given_heads, 3))
```
