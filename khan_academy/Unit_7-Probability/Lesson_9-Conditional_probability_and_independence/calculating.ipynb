# Calculating conditional probability

[Calculating conditional probability | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=6xPkG2pA-TU&t=1s)

Rahul’s two favorite foods are bagels and pizza. Let *A* represent the event that he eats a bagel for breakfast and *B* represent the event that he eats pizza for lunch.

On a randomly selected day, the probability that Rahul will eat a bagel for breakfast, *P(A)*, is 0.6, the probability that he will eat pizza for lunch, *P(B)*, is equal to 0.5, and the conditional probability that he eats a bagel for breakfast, given that he eats pizza for lunch, *P(A | B)*, is equal to 0.7.

**Based on this information, what is *P(B | A)*, the conditional probability that Rahul eats pizza for lunch, given that he eats a bagel for breakfast, rounded to the nearest hundredth?**


### Summary

Rahul likes bagels for breakfast (event **A**) and pizza for lunch (event **B**).  
Given information:
- $ P(A) = 0.6 $
- $ P(B) = 0.5 $
- $ P(A \mid B) = 0.7 $

Since $ P(A) \neq P(A \mid B) $, events **A** and **B** are dependent.

To find $ P(B \mid A) $, first compute the probability that both events occur:
$$
P(A \cap B) = P(A \mid B) \times P(B) = 0.7 \times 0.5 = 0.35
$$

Then use the conditional probability formula:
$$
P(B \mid A) = \frac{P(A \cap B)}{P(A)} = \frac{0.35}{0.6} \approx 0.58
$$

**Final Answer:**  
The probability that Rahul eats pizza for lunch given that he eats a bagel for breakfast is **0.58** (rounded to the nearest hundredth).


```python
# Given probabilities
P_A = 0.6          # P(A)
P_B = 0.5          # P(B)
P_A_given_B = 0.7  # P(A | B)

# Compute joint probability P(A and B)
P_A_and_B = P_A_given_B * P_B

# Compute conditional probability P(B | A)
P_B_given_A = P_A_and_B / P_A

# Round to nearest hundredth
P_B_given_A_rounded = round(P_B_given_A, 2)

print("P(B | A) =", P_B_given_A_rounded)
```
