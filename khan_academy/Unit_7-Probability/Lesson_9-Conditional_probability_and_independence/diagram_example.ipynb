A company screens job applicants for illegal drug use at a certain stage in their hiring process. The specific test they use has a false positive rate of 2% and a false negative rate of 1%. Suppose that 5% of all their applicants are actually using illegal drugs, and we randomly select an applicant.

Given the applicant tests positive, what is the probability that they are actually on drugs?


### Summary

This problem uses conditional probability to determine how reliable a positive drug test result is.

**Given information:**
- 5% of applicants use illegal drugs
- False positive rate = 2%
- False negative rate = 1%

To simplify calculations, imagine **10,000 applicants**:
- **500** are actually on drugs
- **9,500** are not on drugs

**Test results for drug users:**
- 99% test positive → **495 true positives**
- 1% test negative → **5 false negatives**

**Test results for non-drug users:**
- 2% test positive → **190 false positives**
- 98% test negative → **9,310 true negatives**

**Total positive tests:**
- 495 (true positives) + 190 (false positives) = **685**

**Probability an applicant is actually on drugs given a positive test:**
$$
\frac{495}{685} \approx 0.72
$$

**Final Result:**
- Given a positive test, there is about a **72% chance** the applicant is actually using illegal drugs.
- Conversely, there is still a **28% chance** the positive result is false.

**Key takeaway:**
Even with low false positive and false negative rates, a positive test result does not guarantee guilt. When the condition being tested for is relatively rare, false positives can make up a significant portion of positive results. This reasoning is essential in areas like medical testing, law, and policy decisions.


```python
# Drug test conditional probability calculation

# Total applicants
total_applicants = 10_000

# Given rates
drug_use_rate = 0.05
false_positive_rate = 0.02
false_negative_rate = 0.01

# Actual counts
on_drugs = total_applicants * drug_use_rate
not_on_drugs = total_applicants * (1 - drug_use_rate)

# Test outcomes
true_positives = on_drugs * (1 - false_negative_rate)
false_negatives = on_drugs * false_negative_rate

false_positives = not_on_drugs * false_positive_rate
true_negatives = not_on_drugs * (1 - false_positive_rate)

# Total positives
total_positives = true_positives + false_positives

# Conditional probability
prob_on_drugs_given_positive = true_positives / total_positives

# Output results
print(f"True positives: {true_positives}")
print(f"False positives: {false_positives}")
print(f"Total positives: {total_positives}")
print(f"P(on drugs | positive test) = {prob_on_drugs_given_positive:.2%}")
```
