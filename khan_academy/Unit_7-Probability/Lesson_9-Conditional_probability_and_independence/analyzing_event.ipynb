# Analyzing event probability for independence

Tomas’s favorite colors are blue and green.

He has 1 blue shirt, 1 green shirt, 1 blue hat, 1 green scarf, 1 blue pair of pants, and 1 green pair of pants.

Tomas selects one of these garments at random. Let *A* be the event that he selects a blue garment and let *B* be the event that he chooses a shirt.

Which of the following statements are true?

Select all that apply.

- *P(A | B) = P(A)*, the probability that Tomas selects a blue garment given that he has chosen a shirt is equal to the probability that Tomas selects a blue garment.

- *P(B | A) = P(B)*, the probability that Tomas selects a shirt given that he has chosen a blue garment is equal to the probability that Tomas selects a shirt.

- Events *A* and *B* are independent events.

- The outcome of events *A* and *B* are dependent on each other.

- *P(A and B) = P(A) · P(B)*, the probability that Tomas selects a blue garment that is a shirt is equal to the probability that Tomas selects a blue garment multiplied by the probability that he selects a shirt.


### Summary

Tomas has six garments total: three blue (shirt, hat, pants) and three green (shirt, scarf, pants).  
He randomly selects one garment.

- Event **A**: selecting a blue garment  
- Event **B**: selecting a shirt  

**Step 1: Compute basic probabilities**
- $ P(A) = \frac{3}{6} = \frac{1}{2} $
- $ P(B) = \frac{2}{6} = \frac{1}{3} $

**Step 2: Conditional probabilities**
- $ P(A \mid B) = \frac{1}{2} $  
  (Given a shirt, one of the two shirts is blue)
- $ P(B \mid A) = \frac{1}{3} $  
  (Given a blue garment, only one of the three is a shirt)

**Step 3: Joint probability**
- $ P(A \cap B) = \frac{1}{6} $  
  (Only one blue shirt out of six garments)

**Step 4: Check independence**
- $ P(A \mid B) = P(A) $
- $ P(B \mid A) = P(B) $
- $ P(A \cap B) = P(A) \cdot P(B) $

All conditions for independence are satisfied.

**Conclusion:**
- Events **A** and **B** are **independent**
- The true statements are:
  - $ P(A \mid B) = P(A) $
  - $ P(B \mid A) = P(B) $
  - Events **A** and **B** are independent
  - $ P(A \cap B) = P(A)\cdot P(B) $

The only false statement is:
- *The outcome of events A and B are dependent on each other.*


```python
# Tomas's garments probability calculations

# Total garments
total = 6

# Counts
blue_garments = 3
shirts = 2
blue_shirts = 1

# Probabilities
P_A = blue_garments / total          # P(A)
P_B = shirts / total                 # P(B)
P_A_given_B = blue_shirts / shirts   # P(A | B)
P_B_given_A = blue_shirts / blue_garments  # P(B | A)
P_A_and_B = blue_shirts / total      # P(A and B)

# Output results
print(f"P(A) = {P_A}")
print(f"P(B) = {P_B}")
print(f"P(A | B) = {P_A_given_B}")
print(f"P(B | A) = {P_B_given_A}")
print(f"P(A and B) = {P_A_and_B}")

# Independence check
if P_A_and_B == P_A * P_B:
    print("Events A and B are independent.")
else:
    print("Events A and B are dependent.")
```
