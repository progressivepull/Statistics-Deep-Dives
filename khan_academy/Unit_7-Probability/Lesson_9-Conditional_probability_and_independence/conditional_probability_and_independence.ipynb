# Conditional probability and independence

James is interested in the weather conditions and whether the downtown train he sometimes takes runs on time. For a year, James records whether each day is sunny, cloudy, rainy, or snowy, as well as whether this train arrives on time or is delayed. His results are displayed in the table below.

|           | On time | Delayed | Total |
|-----------|---------|---------|-------|
| Sunny     | 167     | 3       | 170   |
| Cloudy   | 115     | 5       | 120   |
| Rainy    | 40      | 15      | 55    |
| Snowy    | 8       | 12      | 20    |
| **Total** | **330** | **35**  | **365** |

For these days, are the events "delayed" and "snowy" independent?


### Summary

James records weather conditions and whether his downtown train is delayed over the course of a year, producing experimental data for 365 days.

To determine whether the events **“delayed”** and **“snowy”** are independent, we compare:
- The overall probability that the train is delayed
- The probability that the train is delayed given that it is snowy

**Overall probability of delay:**
- The train is delayed on 35 out of 365 days
- $ P(\text{Delayed}) = \frac{35}{365} \approx 0.096 $ (about 9.6%)

**Probability of delay given snowy weather:**
- There are 20 snowy days total
- The train is delayed on 12 of those days
- $ P(\text{Delayed} \mid \text{Snowy}) = \frac{12}{20} = 0.6 $ (60%)

**Conclusion:**
Because the probability of a delay on snowy days (60%) is much higher than the overall probability of a delay (about 9.6%), the events **“delayed”** and **“snowy”** are **not independent** based on this experimental data.

This large difference suggests that snowy weather significantly affects whether the train is delayed.


```python
# Train delay and weather independence check

# Given data
total_days = 365
total_delayed = 35

snowy_days = 20
snowy_delayed = 12

# Experimental probabilities
P_delayed = total_delayed / total_days
P_delayed_given_snowy = snowy_delayed / snowy_days

# Output results
print(f"P(Delayed) = {P_delayed:.4f}")
print(f"P(Delayed | Snowy) = {P_delayed_given_snowy:.4f}")

# Independence check
if abs(P_delayed - P_delayed_given_snowy) < 1e-6:
    print("Events 'delayed' and 'snowy' appear independent.")
else:
    print("Events 'delayed' and 'snowy' are not independent.")
```
