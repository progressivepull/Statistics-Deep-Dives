# Probability with Playing Cards — Summary

## Standard Deck
- A standard deck (no jokers) has **52 cards**.
- There are **4 suits**: spades, diamonds, clubs, hearts.
- Each suit has **13 ranks**: Ace, 2–10, Jack, Queen, King.
- Total cards:  
  $$
  4 \times 13 = 52
  $$

## Basic Probability Principle
$$
\text{Probability} = \frac{\text{Number of favorable outcomes}}{\text{Total number of outcomes}}
$$

All cards are assumed to be **equally likely** when the deck is well shuffled.

---

## Example Probabilities

### 1. Probability of Drawing a Jack
- Number of Jacks: 4
- Total cards: 52  
$$
P(\text{Jack}) = \frac{4}{52} = \frac{1}{13}
$$

---

### 2. Probability of Drawing a Heart
- Number of hearts: 13
- Total cards: 52  
$$
P(\text{Heart}) = \frac{13}{52} = \frac{1}{4}
$$

---

### 3. Probability of Drawing the Jack of Hearts
- Only **one** card satisfies both conditions  
$$
P(\text{Jack of Hearts}) = \frac{1}{52}
$$

---

## Combined Events (OR Probability)

### Probability of Drawing a Jack **or** a Heart
- Number of Jacks: 4
- Number of Hearts: 13
- Overlap (Jack of Hearts): 1

Using the inclusion–exclusion principle:
$$
P(J \text{ or } H) = \frac{4 + 13 - 1}{52} = \frac{16}{52} = \frac{4}{13}
$$

---

## Key Takeaway
When calculating probabilities involving **“or”**, be careful not to double-count outcomes that satisfy **both** conditions. Subtract the overlap to correct for this.


```python
# Probability calculations with a standard 52-card deck (no jokers)

total_cards = 52

# Counts
num_jacks = 4
num_hearts = 13
num_jack_of_hearts = 1

# Probabilities
p_jack = num_jacks / total_cards
p_heart = num_hearts / total_cards
p_jack_of_hearts = num_jack_of_hearts / total_cards

# Probability of Jack OR Heart (inclusion–exclusion principle)
p_jack_or_heart = (num_jacks + num_hearts - num_jack_of_hearts) / total_cards

# Display results
print(f"Probability of drawing a Jack: {p_jack} ({num_jacks}/{total_cards})")
print(f"Probability of drawing a Heart: {p_heart} ({num_hearts}/{total_cards})")
print(f"Probability of drawing the Jack of Hearts: {p_jack_of_hearts} (1/{total_cards})")
print(f"Probability of drawing a Jack or a Heart: {p_jack_or_heart} (16/{total_cards})")
```
