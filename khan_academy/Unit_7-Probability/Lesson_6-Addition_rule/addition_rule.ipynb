# Probability with Objects in a Bag — Summary

## Objects in the Bag
The bag contains a total of **29 objects**, broken down as follows:

- **Green cubes:** 8  
- **Green spheres:** 9  
- **Yellow cubes:** 5  
- **Yellow spheres:** 7  

$$
8 + 9 + 5 + 7 = 29
$$

Each object is assumed to be **equally likely** to come out first.

---

## Basic Probability Rule
$$
\text{Probability} = \frac{\text{Number of favorable outcomes}}{\text{Total number of outcomes}}
$$

---

## Individual Probabilities

### 1. Probability of Getting a Cube (Any Color)
- Total cubes: $8 + 5 = 13$

$$
P(\text{Cube}) = \frac{13}{29}
$$

---

### 2. Probability of Getting a Yellow Object
- Total yellow objects: $5 + 7 = 12$

$$
P(\text{Yellow}) = \frac{12}{29}
$$

---

### 3. Probability of Getting a Yellow Cube
- Yellow cubes: 5

$$
P(\text{Yellow and Cube}) = \frac{5}{29}
$$

This corresponds to the **overlap** between the sets “Yellow” and “Cube.”

---

## Combined Events (OR Probability)

### Probability of Getting a Yellow Object **or** a Cube
- Yellow objects: 12  
- Cubes: 13  
- Yellow cubes (overlap): 5  

Using the **addition rule of probability**:

$$
P(\text{Yellow or Cube}) = \frac{12 + 13 - 5}{29} = \frac{20}{29}
$$

---

## General Addition Rule of Probability

For any two events $A$ and $B$:

$$
P(A \text{ or } B) = P(A) + P(B) - P(A \text{ and } B)
$$

- The subtraction prevents **double counting** outcomes that satisfy both conditions.

---

## Mutually Exclusive Events
- If two events **cannot occur at the same time**, they are *mutually exclusive*.
- In that case:
$$
P(A \text{ and } B) = 0
$$

So the rule simplifies to:
$$
P(A \text{ or } B) = P(A) + P(B)
$$

---

## Key Takeaway
Always account for **overlap** when combining probabilities.  
If events overlap, subtract the intersection. If they do not, simple addition works.


```python
# Probability calculations for objects in a bag

# Object counts
green_cubes = 8
green_spheres = 9
yellow_cubes = 5
yellow_spheres = 7

# Total objects
total_objects = green_cubes + green_spheres + yellow_cubes + yellow_spheres

# Category counts
total_cubes = green_cubes + yellow_cubes
total_yellow = yellow_cubes + yellow_spheres
yellow_and_cube = yellow_cubes

# Probabilities
p_cube = total_cubes / total_objects
p_yellow = total_yellow / total_objects
p_yellow_and_cube = yellow_and_cube / total_objects

# Addition rule (Yellow OR Cube)
p_yellow_or_cube = p_yellow + p_cube - p_yellow_and_cube

# Display results
print(f"Total objects: {total_objects}")
print(f"Probability of a cube: {p_cube} ({total_cubes}/{total_objects})")
print(f"Probability of a yellow object: {p_yellow} ({total_yellow}/{total_objects})")
print(f"Probability of a yellow cube: {p_yellow_and_cube} ({yellow_and_cube}/{total_objects})")
print(f"Probability of yellow OR cube: {p_yellow_or_cube} (20/{total_objects})")
```
