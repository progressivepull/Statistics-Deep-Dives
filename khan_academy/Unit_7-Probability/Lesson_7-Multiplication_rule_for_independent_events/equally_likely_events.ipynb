## Summary: Two Ways to Think About Probability

### 1. Classical (Equally Likely Outcomes) Probability
- Probability can be defined as:
  > **Number of favorable outcomes ÷ total number of equally likely outcomes**
- Examples:
  - **Fair coin**:  
    - Heads = 1 favorable outcome out of 2 → probability = **1/2**
    - Tails = **1/2**
  - **Fair die**:  
    - Even numbers (2, 4, 6) = 3 outcomes out of 6 → probability = **1/2**
- This approach works **only when all outcomes are equally likely**.

---

### 2. Frequentist (Long-Run Frequency) Probability
- Used when outcomes are **not equally likely**.
- Example: **Unfair coin**
  - Probability of heads = **60% = 0.6 = 3/5**
  - Probability of tails = **40% = 0.4 = 2/5**
- Interpretation:
  - If the coin is flipped a very large number of times, about **60%** of flips will be heads and **40%** will be tails.
- Probabilities of all mutually exclusive outcomes must sum to **1 (100%)**.

---

### 3. Independent Events
- Coin flips are **independent**:
  - The result of one flip does **not** affect the next.
- For independent events, probabilities are **multiplied**.

---

### 4. Examples with an Unfair Coin

#### Two Heads in a Row
- Probability =  
  $$
  0.6 \times 0.6 = 0.36
  $$
- Interpretation: **36% chance** of getting two heads in a row  
  (higher than 25% with a fair coin).

#### Sequence: Tails → Heads → Tails
- Probability =  
  $$
  0.4 \times 0.6 \times 0.4 = 0.096
  $$
- Interpretation: **9.6% chance** of getting exactly that sequence.

---

### 5. Key Takeaways
- Use **counting** when outcomes are equally likely.
- Use **long-run frequency** when outcomes are not equally likely.
- For independent events:
  - **Multiply probabilities** to find the chance of a specific sequence.
- Probabilities describe what happens **on average over many trials**, not guarantees for short runs.


```python
# Probability calculations for an unfair coin

def sequence_probability(probabilities):
    """
    probabilities: list of probabilities for each event in sequence
    returns: probability of the full sequence
    """
    result = 1.0
    for p in probabilities:
        result *= p
    return result


# Unfair coin probabilities
P_HEADS = 0.6
P_TAILS = 0.4


# Example 1: Probability of two heads in a row
two_heads = sequence_probability([P_HEADS, P_HEADS])
print(f"Probability of two heads in a row: {two_heads:.3f} ({two_heads*100:.1f}%)")


# Example 2: Probability of Tails -> Heads -> Tails
t_h_t = sequence_probability([P_TAILS, P_HEADS, P_TAILS])
print(f"Probability of Tails-Heads-Tails: {t_h_t:.3f} ({t_h_t*100:.1f}%)")


# General example: n heads in a row
def n_heads_in_a_row(p_heads, n):
    return p_heads ** n


# Example: 5 heads in a row with unfair coin
five_heads = n_heads_in_a_row(P_HEADS, 5)
print(f"Probability of 5 heads in a row: {five_heads:.4f} ({five_heads*100:.2f}%)")
```
