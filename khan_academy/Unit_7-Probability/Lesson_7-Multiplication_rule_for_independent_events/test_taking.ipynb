On a muliple choice test, problem 1 has 4 choices and problem 2 has 3 choice. Each problem has only one correct answer. What is the probability of randomly guessing the correct answer to both problems ?

## Summary: Probability of Guessing Correct Answers

We are given two independent multiple-choice problems:

- **Problem 1** has **4 choices**, with **1 correct answer**.
- **Problem 2** has **3 choices**, with **1 correct answer**.

### Step 1: Probability of Each Correct Answer
- Probability of guessing correctly on **Problem 1**:
  $$
  P(\text{correct on 1}) = \frac{1}{4}
  $$
- Probability of guessing correctly on **Problem 2**:
  $$
  P(\text{correct on 2}) = \frac{1}{3}
  $$

### Step 2: Independence of Events
The two events are **independent**, meaning that the outcome of one guess does not affect the outcome of the other.

For independent events, the probability of both occurring is the **product** of their individual probabilities.

### Step 3: Probability of Getting Both Correct
$$
P(\text{correct on both}) = \frac{1}{4} \times \frac{1}{3} = \frac{1}{12}
$$

### Visual Interpretation (Conceptual)
- There are **4 possible outcomes** for Problem 1 and **3 possible outcomes** for Problem 2.
- This creates a total of:
  $$
  4 \times 3 = 12 \text{ possible outcome pairs}
  $$
- Only **1 of these 12 outcomes** corresponds to getting both answers correct.

### ✅ Final Answer
The probability of randomly guessing the correct answer on **both problems** is:

$$
\boxed{\frac{1}{12}}
$$


```python
# Number of choices for each problem
choices_problem_1 = 4
choices_problem_2 = 3

# Probability of guessing correctly on each problem
p_correct_1 = 1 / choices_problem_1
p_correct_2 = 1 / choices_problem_2

# Since the events are independent, multiply the probabilities
p_both_correct = p_correct_1 * p_correct_2

print("Probability of correct on problem 1:", p_correct_1)
print("Probability of correct on problem 2:", p_correct_2)
print("Probability of correct on both problems:", p_both_correct)
```
