# Probability of At Least One Head in Coin Flips

## Problem Setup
We flip a **fair coin** multiple times and want to find the probability of getting **at least one head**.

---

## Case 1: Three Coin Flips

### Total Possible Outcomes
Each flip has 2 outcomes (H or T).  
With 3 flips:

$$
2 \times 2 \times 2 = 8 \text{ equally likely outcomes}
$$

### Counting Favorable Outcomes
Out of these 8 outcomes:
- **7 outcomes** contain **at least one head**
- **1 outcome** (TTT) contains **no heads**

So:

$$
P(\text{at least one head}) = \frac{7}{8}
$$

---

## A Smarter Approach (Complement Rule)

Listing outcomes works for small cases, but becomes impractical for many flips.  
Instead, use the **complement**.

### Key Idea
> Getting **at least one head** is the same as **not getting all tails**.

These two events are:
- **Mutually exclusive**
- **Exhaustive**

So:

$$
P(\text{at least one head}) = 1 - P(\text{all tails})
$$

---

## Applying the Method (3 Flips)

- Probability of tails on one flip: $ \frac{1}{2} $
- Probability of all tails in 3 flips:

$$
\left(\frac{1}{2}\right)^3 = \frac{1}{8}
$$

Therefore:

$$
P(\text{at least one head}) = 1 - \frac{1}{8} = \frac{7}{8}
$$

Same result, much less work.

---

## Case 2: Ten Coin Flips

### Step 1: Probability of All Tails
$$
P(\text{all tails in 10 flips}) = \left(\frac{1}{2}\right)^{10} = \frac{1}{1024}
$$

### Step 2: Use the Complement
$$
P(\text{at least one head}) = 1 - \frac{1}{1024} = \frac{1023}{1024}
$$

### As a Percentage
$$
\frac{1023}{1024} \approx 99.9\%
$$

---

## Final Takeaway
- Counting outcomes works for small problems, but not for large ones.
- Using the **complement rule** makes complex probability problems simple.
- For many repeated trials, it’s often easier to calculate the probability of the **one thing you don’t want**, then subtract from 1.

This approach is especially powerful for exams and real-world probability problems.


```python
def probability_at_least_one_head(num_flips):
    """
    Calculates the probability of getting at least one head
    when flipping a fair coin num_flips times.
    """
    # Probability of getting all tails
    prob_all_tails = (1 / 2) ** num_flips

    # Complement: at least one head
    prob_at_least_one_head = 1 - prob_all_tails

    return prob_at_least_one_head


# Examples
flips_3 = probability_at_least_one_head(3)
flips_10 = probability_at_least_one_head(10)

print(f"Probability of at least one head in 3 flips: {flips_3} ({flips_3:.2%})")
print(f"Probability of at least one head in 10 flips: {flips_10} ({flips_10:.2%})")
```
