# Probability Game Show Vacation — Summary

## Scenario
You are on a game show and have earned the chance to win a **vacation package**.  
Each vacation is defined by **two attributes**:

- **Type of vacation**
  - Island beach
  - Ski
  - Camping
- **Length of stay**
  - 1 day
  - 2 days
  - **3 days**

The game randomly selects **one vacation package**, and all possible packages are **equally likely**.

---

## Sample Space
To find probabilities, we first list **all possible outcomes**.

There are:
- 3 vacation types  
- 3 possible lengths  

This creates a **3 × 3 grid** with **9 total outcomes**:

### All Possible Vacation Packages
- **Island trips**
  - 1-day island
  - 2-day island
  - 3-day island
- **Ski trips**
  - 1-day ski
  - 2-day ski
  - 3-day ski
- **Camping trips**
  - 1-day camping
  - 2-day camping
  - 3-day camping

Since each outcome is equally likely:
- **Total outcomes = 9**

---

## Probability Question
**What is the probability of winning a vacation that lasts _at least two days_ and has _no snow_?**

### Interpreting the conditions:
- **At least two days** → 2-day or 3-day trips
- **No snow** → excludes ski trips  
  (Island and camping trips are assumed snow-free)

---

## Favorable Outcomes
The qualifying vacation packages are:

- 2-day island
- 3-day island
- 2-day camping
- 3-day camping

That gives:
- **4 favorable outcomes**

---

## Final Probability
$$
\text{Probability} = \frac{\text{Favorable outcomes}}{\text{Total outcomes}} = \frac{4}{9}
$$

---

## Conclusion
You have a **4/9 chance** of winning a vacation that:
- Keeps you away from snow  
- Lasts **at least two days**

A solid outcome for anyone dreaming of warmth instead of ski slopes!


```python
# Define vacation types and lengths
vacation_types = ["Island", "Ski", "Camping"]
lengths = [1, 2, 3]  # days

# Construct the sample space
sample_space = [(v, d) for v in vacation_types for d in lengths]

# Total number of outcomes
total_outcomes = len(sample_space)

# Define the event: at least 2 days and no snow (no Ski trips)
favorable_outcomes = [
    (v, d)
    for (v, d) in sample_space
    if d >= 2 and v != "Ski"
]

# Compute probability
probability = len(favorable_outcomes) / total_outcomes

# Display results
print("Sample Space:")
for outcome in sample_space:
    print(outcome)

print("\nFavorable Outcomes (≥ 2 days, no snow):")
for outcome in favorable_outcomes:
    print(outcome)

print(f"\nTotal outcomes: {total_outcomes}")
print(f"Favorable outcomes: {len(favorable_outcomes)}")
print(f"Probability: {probability}  (which is {len(favorable_outcomes)}/{total_outcomes})")
```
