Find the probability of rolling even numbers three times, using a six-sided die numbered from 1 to 6.

## Summary: Probability of Rolling Even Numbers Three Times

We are asked to find the probability of rolling **even numbers three times in a row** using a fair six-sided die numbered from 1 to 6.

### Step 1: Probability of Rolling an Even Number Once
A six-sided die has the following possible outcomes:
$$
\{1, 2, 3, 4, 5, 6\}
$$

The even numbers are:
$$
\{2, 4, 6\}
$$

- Number of favorable outcomes (even numbers): 3  
- Total possible outcomes: 6  

$$
P(\text{even}) = \frac{3}{6} = \frac{1}{2}
$$

### Step 2: Independence of Rolls
Each roll of the die is an **independent event**. The outcome of one roll does not affect the outcome of the next roll.

### Step 3: Probability of Rolling Even Three Times
To roll an even number **three consecutive times**, we multiply the probabilities:

$$
P(\text{even on all three rolls}) = \frac{1}{2} \times \frac{1}{2} \times \frac{1}{2}
$$

$$
= \frac{1}{8}
$$

### ✅ Final Answer
The probability of rolling even numbers on **all three rolls** is:

$$
\boxed{\frac{1}{8}}
$$


```python
# Probability of rolling an even number on a six-sided die
# Even numbers are 2, 4, and 6 → 3 out of 6 outcomes
p_even = 3 / 6

# Number of rolls
rolls = 3

# Since rolls are independent, multiply the probability for each roll
p_even_three_times = p_even ** rolls

print("Probability of rolling an even number once:", p_even)
print("Probability of rolling even numbers three times:", p_even_three_times)
```
