# What Are the Odds of Making 10 Free Throws in a Row?

Sal explains that if a player has a consistent free throw percentage, the probability of making multiple free throws **in a row** is found by repeatedly multiplying that percentage.

## Key Idea
- Each free throw is assumed to be **independent** of the previous one.
- The probability of making a single free throw stays the same each time.
- To make *n* free throws in a row, you multiply the free throw probability by itself *n* times.

Mathematically:

$$
\text{Probability of } n \text{ in a row} = (\text{Free Throw Percentage})^n
$$

## LeBron James Example
- Career free throw percentage: **75% (0.75)**
- Number of consecutive free throws: **10**

$$
0.75^{10} \approx 0.056 \approx 6\%
$$

**Result:**  
LeBron has about a **6% chance** (roughly **1 in 20**) of making **10 free throws in a row**, even with a strong free throw percentage.

## General Rule
For **any player**:
- If your free throw percentage is $ p $
- And you want to know the odds of making $ n $ in a row

$$
\text{Probability} = p^n
$$

### Example
- Free throw percentage: **60% (0.6)**
- Free throws in a row: **5**

$$
0.6^5 \approx 0.08
$$

That’s about an **8% chance**.

## Takeaway
Even high-percentage shooters face surprisingly low odds when aiming for long streaks. As the number of required consecutive shots increases, the probability drops quickly.


```python
def probability_of_consecutive_free_throws(free_throw_percentage, n):
    """
    Calculate the probability of making n free throws in a row.

    Parameters:
        free_throw_percentage (float): Free throw percentage as a decimal (e.g., 0.75 for 75%)
        n (int): Number of consecutive free throws

    Returns:
        float: Probability of making n free throws in a row
    """
    return free_throw_percentage ** n


# Example usage:

# LeBron James example
lebron_ft_pct = 0.75
shots_in_a_row = 10
lebron_probability = probability_of_consecutive_free_throws(lebron_ft_pct, shots_in_a_row)

print(f"Probability of making {shots_in_a_row} free throws in a row: {lebron_probability:.4f} ({lebron_probability*100:.2f}%)")

# Another example
my_ft_pct = 0.60
shots_in_a_row = 5
my_probability = probability_of_consecutive_free_throws(my_ft_pct, shots_in_a_row)

print(f"Probability of making {shots_in_a_row} free throws in a row: {my_probability:.4f} ({my_probability*100:.2f}%)")
```
