# Do You Have Better Odds: 3 Free Throws in a Row or 1 Three-Pointer?

Sal compares LeBron’s chances of making **three consecutive free throws** versus making **one three-pointer**, using career shooting percentages and basic probability.

## Shooting Percentages (Career Averages)
- **Free throw percentage:** ~75% (0.75)
- **Three-point percentage:** ~33% (0.33)

At first glance, a single free throw is much easier than a three-pointer—but the real question is about **three in a row**.

## Probability of 3 Free Throws in a Row
Assuming each free throw is independent:

$$
P(\text{3 in a row}) = 0.75^3
$$

$$
0.75 \times 0.75 \times 0.75 \approx 0.42
$$

**Result:**  
- Probability of making **3 free throws in a row ≈ 42%**

## Probability of 1 Three-Pointer
- Probability of making **one three-pointer ≈ 33%**

## Final Comparison
| Shot Type | Probability |
|----------|-------------|
| 3 free throws in a row | **42%** |
| 1 three-pointer | **33%** |

## Conclusion
Yes — **you have better odds of making 3 free throws in a row than making a single three-pointer**.  

Even though consecutive shots reduce probability, LeBron’s high free throw percentage still makes three in a row more likely than hitting a three from beyond the arc.

## Takeaway
This comparison depends entirely on shooting percentages. For some players with exceptional three-point accuracy or poor free throw shooting, the result could be reversed.


# Exact probability (math approach)

If a player makes a free throw with probability p, then the probability of making n free throws in a row is: 

$$
    P^n
$$

```python
def probability_n_in_a_row(p, n):
    """
    p: free throw percentage as a decimal (e.g., 0.75)
    n: number of consecutive free throws
    """
    return p ** n


# Example: LeBron (~75%) making 10 in a row
p = 0.75
n = 10

prob = probability_n_in_a_row(p, n)
print(f"Probability of making {n} free throws in a row: {prob:.4f} ({prob*100:.2f}%)")
```

# Monte Carlo simulation (to verify intuitively)

This simulates many “LeBrons” shooting 10 free throws.

```python
import random

def simulate_n_in_a_row(p, n, trials=1_000_000):
    success = 0

    for _ in range(trials):
        if all(random.random() < p for _ in range(n)):
            success += 1

    return success / trials


sim_prob = simulate_n_in_a_row(0.75, 10)
print(f"Simulated probability: {sim_prob:.4f} ({sim_prob*100:.2f}%)")
```

# General takeaway (Sal’s rule)

```python
print(0.60 ** 5)   # 60% shooter, 5 in a row
print(0.90 ** 10)  # 90% shooter, 10 in a row
```
