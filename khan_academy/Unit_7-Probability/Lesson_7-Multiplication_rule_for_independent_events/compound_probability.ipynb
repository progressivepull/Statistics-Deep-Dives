## Summary: Probabilities with a Fair Coin

### Single Coin Flip
- A fair coin has **two equally likely outcomes**: Heads (H) and Tails (T).
- **Probability of Heads** = 1/2  
- **Probability of Tails** = 1/2  
- The **sum of probabilities of all possible outcomes equals 1**, which is always true in probability.

---

### Two Coin Flips
- Each flip is **independent** (one flip does not affect another).
- Possible outcomes when flipping twice:
  - HH, HT, TH, TT  
- There are **4 equally likely outcomes**.
- **Probability of getting Heads then Heads (HH)**:
  - By counting outcomes: 1 out of 4 → **1/4**
  - By independence:  
    $$
    P(HH) = \frac{1}{2} \times \frac{1}{2} = \frac{1}{4}
    $$

---

### Independence and the Gambler’s Fallacy
- Coin flips are **independent events**.
- Past results do **not** change future probabilities.
- The belief that a tail is “due” after many heads is called the **gambler’s fallacy** and is incorrect.

---

### Three Coin Flips (Specific Order)
- Example: **Tails → Heads → Tails (THT)**
- Since flips are independent:
  $$
  P(THT) = \frac{1}{2} \times \frac{1}{2} \times \frac{1}{2} = \frac{1}{8}
  $$

- All possible outcomes for three flips:
  - HHH, HHT, HTH, HTT, THH, **THT**, TTH, TTT  
- There are **8 equally likely outcomes**, and THT is **1 of them**, confirming the probability of **1/8**.

---

### Key Takeaways
- Probabilities of independent events **multiply**.
- The number of possible outcomes doubles with each additional coin flip.
- Exact sequences become **less likely** as the number of flips increases.


```python
import itertools

def single_flip_probabilities():
    outcomes = ["H", "T"]
    prob = 1 / len(outcomes)
    return {outcome: prob for outcome in outcomes}

def n_flip_outcomes(n):
    return list(itertools.product(["H", "T"], repeat=n))

def probability_of_sequence(sequence):
    # For a fair coin, each flip has probability 1/2
    return (1 / 2) ** len(sequence)

# Examples

# 1 flip
print("Single flip probabilities:")
print(single_flip_probabilities())

# 2 flips
two_flip_outcomes = n_flip_outcomes(2)
print("\nTwo-flip outcomes:", two_flip_outcomes)
print("Probability of HH:", probability_of_sequence(["H", "H"]))

# 3 flips
three_flip_outcomes = n_flip_outcomes(3)
print("\nThree-flip outcomes:", three_flip_outcomes)
print("Probability of THT:", probability_of_sequence(["T", "H", "T"]))

# Verify by counting equally likely outcomes
def probability_by_counting(sequence, outcomes):
    return outcomes.count(tuple(sequence)) / len(outcomes)

print("\nVerification by counting:")
print("Probability of HH:", probability_by_counting(["H", "H"], two_flip_outcomes))
print("Probability of THT:", probability_by_counting(["T", "H", "T"], three_flip_outcomes))
```
