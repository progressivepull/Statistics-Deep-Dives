# Sample Spaces — Summary

## What Is a Sample Space?
A **sample space** is the set of **all possible outcomes** of a probabilistic experiment or trial.

- It does **not** assign probabilities by itself.
- It simply lists **what can happen**.
- Once the sample space is known, probabilities can be assigned—especially easily when outcomes are equally likely.

---

## Simple Sample Space Example: Coin Flip
- Possible outcomes: **Heads (H)** or **Tails (T)**
$$
\text{Sample Space} = \{H, T\}
$$

If the coin is fair:
- Each outcome has probability $ \frac{1}{2} $

---

## More Complex Example: Cupcakes

### Cupcake Characteristics
Cupcakes vary in **two different ways**:

**Flavors (3):**
- Chocolate
- Strawberry
- Vanilla

**Sizes (3):**
- Small
- Medium
- Large

Each flavor comes in each size.

---

## Constructing the Sample Space
Each outcome consists of **both a flavor and a size**.

### Total Outcomes
$$
3 \text{ flavors} \times 3 \text{ sizes} = 9 \text{ outcomes}
$$

Examples:
- Small chocolate
- Medium strawberry
- Large vanilla

---

## Ways to Represent the Sample Space

### 1. Tree Diagram
- First branch by flavor, then by size (or vice versa)
- Each path represents one unique outcome

### 2. Grid (Matrix)
- Rows represent sizes
- Columns represent flavors
- Each cell corresponds to one outcome

### 3. Table or List
| Flavor      | Size   |
|-------------|--------|
| Chocolate   | Small  |
| Chocolate   | Medium |
| Chocolate   | Large  |
| Strawberry  | Small  |
| Strawberry  | Medium |
| Strawberry  | Large  |
| Vanilla     | Small  |
| Vanilla     | Medium |
| Vanilla     | Large  |

All three methods represent the **same sample space**.

---

## Compound Sample Spaces
When outcomes depend on **more than one factor**, the sample space is called a **compound sample space**.

- Coin flip → simple sample space
- Cupcake flavor + size → compound sample space
- Can extend to more than two variables

---

## Why Sample Spaces Matter
- They define all possible outcomes
- They are essential for calculating probabilities
- Especially powerful when outcomes are equally likely

---

## Key Takeaway
A sample space is the foundation of probability—it tells you **what can happen** before asking **how likely** each outcome is.


```python
# Sample space examples in Python

# 1. Simple sample space: coin flip
coin_sample_space = ["Heads", "Tails"]

print("Coin flip sample space:")
print(coin_sample_space)
print(f"Total outcomes: {len(coin_sample_space)}\n")


# 2. Compound sample space: cupcakes

flavors = ["Chocolate", "Strawberry", "Vanilla"]
sizes = ["Small", "Medium", "Large"]

# Construct compound sample space using list comprehension
cupcake_sample_space = [(flavor, size) for flavor in flavors for size in sizes]

print("Cupcake sample space (Flavor, Size):")
for outcome in cupcake_sample_space:
    print(outcome)

print(f"\nTotal cupcake outcomes: {len(cupcake_sample_space)}")


# 3. Example probability calculation (assuming equally likely outcomes)
# Probability of getting a small cupcake OR a chocolate cupcake

small_or_chocolate = [
    (flavor, size)
    for (flavor, size) in cupcake_sample_space
    if size == "Small" or flavor == "Chocolate"
]

probability = len(small_or_chocolate) / len(cupcake_sample_space)

print("\nOutcomes that are Small OR Chocolate:")
for outcome in small_or_chocolate:
    print(outcome)

print(f"\nProbability of Small OR Chocolate: {probability}")
```
