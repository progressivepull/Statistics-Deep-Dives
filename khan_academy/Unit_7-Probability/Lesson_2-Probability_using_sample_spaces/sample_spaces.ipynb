# Describing Subsets of Sample Spaces – Summary

This section reviews how to describe **sets and subsets** within a **sample space** using two probability examples.

---

## Example 1: Ollivanders Wand Shop

### Problem Setup

Harry Potter visits Ollivanders Wand Shop, where the wand selection is treated as a **random process**.

- **Wood types**: holly, elm, maple, wenge (4 options)
- **Core materials**: phoenix feather, unicorn hair, dragon scale, raven feather, thestral tail (5 options)

Each wand consists of **one wood** and **one core**.

---

### Sample Space

Since each wood can be combined with each core:

$$
4 \times 5 = 20
$$

There are **20 equally likely outcomes** in the sample space.

---

### Event 1: “Holly or Unicorn Hair”

This event includes:
- All **holly** wands (5 outcomes)
- All **unicorn hair** wands that are **not already holly**:
  - Elm Unicorn
  - Maple Unicorn
  - Wenge Unicorn

Total favorable outcomes:

$$
5 + 3 = 8
$$

---

### Probability

$$
P(\text{Holly or Unicorn Hair}) = \frac{8}{20} = \frac{4}{10} = 40\%
$$

---

### Event 2: “Holly and Unicorn Hair”

This event includes **only one outcome**:

- Holly Unicorn

---

### Probability

$$
P(\text{Holly and Unicorn Hair}) = \frac{1}{20}
$$

---

### Comparison

- “Holly **or** Unicorn Hair” includes **8 outcomes**
- “Holly **and** Unicorn Hair” includes **1 outcome**

Since the first event includes the second event **plus seven additional outcomes**, it is clearly **more likely**.

---

## Example 2: Fire–Water–Sponge Game

This game is similar to Rock–Paper–Scissors.

### Rules

- Fire beats Sponge
- Sponge beats Water
- Water beats Fire
- Same choices result in a **tie**

Each player chooses one of:
- Fire
- Water
- Sponge

---

### Sample Space

Each player has 3 choices:

$$
3 \times 3 = 9
$$

There are **9 total outcomes**.

---

### Given Subset

The subset includes outcomes:
- 1, 3, 4, 5, 7, and 8

These exclude:
- Fire–Fire
- Water–Water
- Sponge–Sponge

---

### Pattern Identified

The excluded outcomes are exactly the **ties**.

The included outcomes are all situations where:
- **One player wins**
- **There is no tie**

---

### Correct Description of the Subset

> **The subset consists of all outcomes where there is not a tie.**

---

## Key Takeaways

- A **sample space** includes all possible outcomes
- A **subset** includes only outcomes that meet specific criteria
- “OR” events usually include **more outcomes** than “AND” events
- Identifying patterns in outcomes helps correctly describe subsets
- Carefully checking examples prevents incorrect assumptions


```python
# ---------------------------------------
# Example 1: Ollivanders Wand Shop
# ---------------------------------------

import itertools

# Woods and cores
woods = ["Holly", "Elm", "Maple", "Wenge"]
cores = ["Phoenix Feather", "Unicorn Hair", "Dragon Scale", "Raven Feather", "Thestral Tail"]

# Sample space (all possible wands)
sample_space = list(itertools.product(woods, cores))

print("Total outcomes:", len(sample_space))
print()

# Event 1: Holly OR Unicorn Hair
holly_or_unicorn = [
    (wood, core) for wood, core in sample_space
    if wood == "Holly" or core == "Unicorn Hair"
]

# Event 2: Holly AND Unicorn Hair
holly_and_unicorn = [
    (wood, core) for wood, core in sample_space
    if wood == "Holly" and core == "Unicorn Hair"
]

print("Holly OR Unicorn Hair outcomes:")
print(holly_or_unicorn)
print("Count:", len(holly_or_unicorn))
print("Probability:", len(holly_or_unicorn) / len(sample_space))
print()

print("Holly AND Unicorn Hair outcomes:")
print(holly_and_unicorn)
print("Count:", len(holly_and_unicorn))
print("Probability:", len(holly_and_unicorn) / len(sample_space))
print()


# ---------------------------------------
# Example 2: Fire–Water–Sponge Game
# ---------------------------------------

choices = ["Fire", "Water", "Sponge"]

# Sample space (player, friend)
game_outcomes = list(itertools.product(choices, choices))

print("Fire–Water–Sponge sample space:")
print(game_outcomes)
print("Total outcomes:", len(game_outcomes))
print()

# Subset: outcomes where there is NOT a tie
not_tie = [
    (player, friend) for player, friend in game_outcomes
    if player != friend
]

print("Outcomes where there is not a tie:")
print(not_tie)
print("Count:", len(not_tie))
print("Probability:", len(not_tie) / len(game_outcomes))
```
