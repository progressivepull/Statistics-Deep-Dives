# Probability of Getting Exactly Two Heads

This example finds the probability of getting **exactly two heads** when flipping **three fair coins**.

---

## Step 1: Define the Experiment

- Each coin flip has two possible outcomes: **Heads (H)** or **Tails (T)**
- We flip **three coins**
- All outcomes are equally likely

---

## Step 2: List the Sample Space

The **sample space** consists of all possible outcomes of three coin flips:

$$
\begin{aligned}
&HHH \\
&HHT \\
&HTH \\
&HTT \\
&THH \\
&THT \\
&TTH \\
&TTT
\end{aligned}
$$

There are **8 total possible outcomes**.

---

## Step 3: Identify Favorable Outcomes

We are interested in outcomes with **exactly two heads**.

These outcomes are:

- HHT  
- HTH  
- THH  

There are **3 favorable outcomes**.

---

## Step 4: Calculate the Probability

$$
P(\text{Exactly 2 Heads}) =
\frac{\text{number of favorable outcomes}}{\text{total number of outcomes}}
= \frac{3}{8}
$$

---

## Final Answer

$$
\boxed{P(\text{Exactly 2 Heads}) = \frac{3}{8}}
$$

---

## Key Takeaways

- A **sample space** lists all possible outcomes
- A **compound event** can include multiple outcomes
- Probability is found by counting favorable outcomes and dividing by total outcomes
- When flipping three coins, getting exactly two heads occurs in **3 out of 8** possible cases


```python
# -----------------------------
# Probability of Exactly Two Heads
# -----------------------------

import itertools

# Possible outcomes for a single coin flip
coin = ["H", "T"]

# Generate all possible outcomes for three coin flips
sample_space = list(itertools.product(coin, repeat=3))

# Convert tuples to strings for readability
outcomes = ["".join(outcome) for outcome in sample_space]

# Find outcomes with exactly two heads
exactly_two_heads = [
    outcome for outcome in outcomes if outcome.count("H") == 2
]

# Probability calculation
probability = len(exactly_two_heads) / len(outcomes)

print("Sample Space:")
print(outcomes)
print()

print("Outcomes with exactly two heads:")
print(exactly_two_heads)
print()

print("Probability of exactly two heads:", probability)


# -----------------------------
# Optional Simulation
# -----------------------------

import random

def flip_three_coins(trials=100000):
    count = 0
    for _ in range(trials):
        flips = [random.choice(["H", "T"]) for _ in range(3)]
        if flips.count("H") == 2:
            count += 1
    return count / trials

simulated_probability = flip_three_coins()
print("Simulated probability (approx):", simulated_probability)
```
