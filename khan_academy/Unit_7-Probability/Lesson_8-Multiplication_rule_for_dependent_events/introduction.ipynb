## Summary: Probability and Expected Value in a Marble Game (Without Replacement)

### Game Setup
- A bag contains **5 marbles**:
  - **3 green**
  - **2 orange**
- You draw **two marbles one at a time**, **without replacement**.
- You **win $1** if **both marbles are green**.
- It **costs $0.35** to play the game.

The question is whether this game makes **economic sense** to play.

---

### Step 1: Probability the First Marble Is Green
- Total marbles: 5  
- Green marbles: 3  

$$
P(\text{first green}) = \frac{3}{5}
$$

---

### Step 2: Dependence Between Draws
The marbles are **not replaced**, so the second draw **depends on the first**.

If the **first marble is green**:
- Remaining marbles: 4  
- Remaining green marbles: 2  

$$
P(\text{second green} \mid \text{first green}) = \frac{2}{4} = \frac{1}{2}
$$

---

### Step 3: Probability of Winning (Two Green Marbles)
To win, **both events must occur**:

$$
P(\text{two greens}) = P(\text{first green}) \times P(\text{second green} \mid \text{first green})
$$

$$
= \frac{3}{5} \times \frac{1}{2} = \frac{3}{10}
$$

$$
= 0.30 \quad \text{(30% chance of winning)}
$$

---

### Step 4: Expected Value
- Probability of winning: **30%**
- Prize: **$1**

$$
\text{Expected value} = 0.30 \times \$1 = \$0.30
$$

---

### Final Decision
- **Cost to play:** $0.35  
- **Expected return:** $0.30  

Since you are paying **more ($0.35)** than you expect to win **on average ($0.30)**:

### ❌ It does **not** make economic sense to play this game.

---

### Thought Question
Would the game become fair or profitable **if the first marble were replaced before the second draw**?  
(This would make the events independent and change the probability.)


```python
# Number of marbles
green = 3
orange = 2
total = green + orange

# Cost and reward
cost_to_play = 0.35
reward = 1.00

# Probability first marble is green
p_first_green = green / total

# After drawing one green marble (no replacement)
remaining_green = green - 1
remaining_total = total - 1

# Probability second marble is green given first was green
p_second_green_given_first = remaining_green / remaining_total

# Probability of winning (two green marbles)
p_win = p_first_green * p_second_green_given_first

# Expected value of the game
expected_value = p_win * reward

print("Probability first marble is green:", p_first_green)
print("Probability second marble is green given first was green:", p_second_green_given_first)
print("Probability of winning (two greens):", p_win)
print("Expected value of the game: $", expected_value)
print("Cost to play: $", cost_to_play)
print("Net expected gain/loss: $", expected_value - cost_to_play)
```
