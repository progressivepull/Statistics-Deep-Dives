# Dependent probability example

[Dependent probability example 2 | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=7BkcNLOf56w)

You have 4 coins in a bag. 3 of them are unfair in that they have a 45% chance of coming up tails when flipped (the rest are fair coins). You randomly choose one coin from the bag and flip it 4 times. What is the percent probability of getting 4 heads?

## Summary: Probability of Getting Four Heads

We have **4 coins in a bag**:
- **3 unfair coins** with a **45% chance of tails** → **55% chance of heads**
- **1 fair coin** with a **50% chance of heads**

A coin is chosen **at random** and flipped **4 times**. We want the **overall probability of getting 4 heads**.

---

### Step 1: Consider Each Possible Coin

#### Case 1: Unfair Coin
- Probability of choosing an unfair coin:  
  $$
  \frac{3}{4}
  $$
- Probability of getting 4 heads with an unfair coin:  
  $$
  (0.55)^4 = 0.09150625
  $$
- Combined probability:
  $$
  \frac{3}{4} \times 0.55^4 = 0.0686296875 \approx 6.86\%
  $$

#### Case 2: Fair Coin
- Probability of choosing the fair coin:  
  $$
  \frac{1}{4}
  $$
- Probability of getting 4 heads with a fair coin:  
  $$
  (0.5)^4 = 0.0625
  $$
- Combined probability:
  $$
  \frac{1}{4} \times 0.5^4 = 0.015625 = 1.56\%
  $$

---

### Step 2: Add the Probabilities

Since either case results in 4 heads:
$$
0.0686296875 + 0.015625 = 0.0842546875
$$

---

### ✅ Final Answer

**The probability of getting four heads is approximately:**

$$
\boxed{8.43\%}
$$

This is slightly higher than using a fair coin alone, because there is a **75% chance** of selecting a coin that is **biased toward heads**.



```python
# Probability setup
p_unfair_coin = 3 / 4
p_fair_coin = 1 / 4

# Head probabilities
p_heads_unfair = 0.55
p_heads_fair = 0.50

# Probability of 4 heads in each case
p_4_heads_unfair = p_heads_unfair ** 4
p_4_heads_fair = p_heads_fair ** 4

# Total probability using the law of total probability
total_probability = (
    p_unfair_coin * p_4_heads_unfair +
    p_fair_coin * p_4_heads_fair
)

# Convert to percentage
percentage = total_probability * 100

print(f"Probability of getting 4 heads: {percentage:.2f}%")
```
