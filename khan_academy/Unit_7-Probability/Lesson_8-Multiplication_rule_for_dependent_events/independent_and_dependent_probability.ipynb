The marching band is holding a raffle at a football game, with two prizes. After the first ticket is pulled out and the winner determined, the ticket os taped to the price. The next is pulled out to determine the winner of the second price.

Are the two events independent? Explain.

## Summary: Are the Two Raffle Events Independent?

Two events are **independent** if the outcome of one event does **not** affect the outcome of the other.

---

### Description of the Raffle
- A marching band raffle has **two prizes**.
- A ticket is drawn to determine the winner of the **first prize**.
- That ticket is **not replaced**; it is taped to the prize.
- A second ticket is then drawn to determine the **second prize winner**.

---

### Why the Events Are *Not* Independent

Because the first ticket is **removed from the pool**, the possible outcomes for the second drawing depend on which ticket was drawn first.

**Example:**
- Suppose the tickets are **A, B, and C**.
- If ticket **A** is drawn first, only **B or C** can be drawn second.
- If ticket **B** is drawn first, only **A or C** can be drawn second.

This shows that the **set of possible outcomes for the second prize changes** based on the outcome of the first prize.

---

### Conclusion

✅ **The two events are not independent.**  
The outcome of the first ticket draw directly affects the possible outcomes of the second draw.

---

### How They *Could* Be Independent

If the first ticket were:
- Recorded (winner written down), and then
- **Placed back into the bag** before drawing the second ticket,

then the second draw would have the same possible outcomes regardless of the first draw.

🔁 With replacement → **Independent events**  
❌ Without replacement → **Dependent events**


```python
import random

# Simulate the raffle WITHOUT replacement (dependent events)
def raffle_without_replacement(tickets):
    first_winner = random.choice(tickets)
    remaining_tickets = tickets.copy()
    remaining_tickets.remove(first_winner)
    second_winner = random.choice(remaining_tickets)
    return first_winner, second_winner


# Simulate the raffle WITH replacement (independent events)
def raffle_with_replacement(tickets):
    first_winner = random.choice(tickets)
    second_winner = random.choice(tickets)
    return first_winner, second_winner


# Example tickets
tickets = ["A", "B", "C"]

# Run one example of each
print("Without replacement (dependent):")
print(raffle_without_replacement(tickets))

print("\nWith replacement (independent):")
print(raffle_with_replacement(tickets))
```

# This code demonstrates that: 

* Without replacement, the second winner depends on the first.
* With replacement, the second winner is unaffected by the first.
