## Summary: Probability with a 6-Sided Die and a 4-Sided Die

Erika rolls **two dice simultaneously**:
- One **6-sided die**
- One **4-sided die**

Define the events:
- **Event A**: Erika rolls **doubles**
- **Event B**: The **4-sided die shows a 4**

---

### Sample Space
- The total number of equally likely outcomes is:
  $$
  6 \times 4 = 24
  $$

---

## Individual Probabilities

### Probability of A (Rolling Doubles)
Possible doubles:
- (1,1), (2,2), (3,3), (4,4)

There are **4 favorable outcomes** out of 24.

$$
P(A) = \frac{4}{24} = \frac{1}{6}
$$

---

### Probability of B (4-Sided Die Is 4)
For each value of the 6-sided die, the 4-sided die can be 4:
- (1,4), (2,4), (3,4), (4,4), (5,4), (6,4)

There are **6 favorable outcomes** out of 24.

$$
P(B) = \frac{6}{24} = \frac{1}{4}
$$

---

## Conditional Probabilities

### Probability of A Given B
Given the 4-sided die is 4, the possible outcomes are:
- (1,4), (2,4), (3,4), (4,4), (5,4), (6,4)

Only **(4,4)** is a double.

$$
P(A \mid B) = \frac{1}{6}
$$

---

### Probability of B Given A
Given doubles occurred:
- (1,1), (2,2), (3,3), (4,4)

Only **(4,4)** has the 4-sided die equal to 4.

$$
P(B \mid A) = \frac{1}{4}
$$

---

## Joint Probability

### Probability of A and B
Both A and B occur only when the outcome is **(4,4)**.

$$
P(A \cap B) = \frac{1}{24}
$$

---

## Verifying Probability Rules

Using conditional probability formulas:

$$
P(A \cap B) = P(A) \cdot P(B \mid A) = \frac{1}{6} \cdot \frac{1}{4} = \frac{1}{24}
$$

$$
P(A \cap B) = P(B) \cdot P(A \mid B) = \frac{1}{4} \cdot \frac{1}{6} = \frac{1}{24}
$$

✔ Both methods give the same result, confirming the rules of probability.

---

## Key Takeaway

- Conditional probability restricts the sample space.
- Joint probability can be computed using either event as the starting point.
- These relationships always hold:
  $$
  P(A \cap B) = P(A)P(B \mid A) = P(B)P(A \mid B)
  $$


```python
# Define the sample space
# (six-sided die, four-sided die)
sample_space = [(d6, d4) for d6 in range(1, 7) for d4 in range(1, 5)]

# Total number of outcomes
total_outcomes = len(sample_space)

# Define events
event_A = [outcome for outcome in sample_space if outcome[0] == outcome[1]]  # doubles
event_B = [outcome for outcome in sample_space if outcome[1] == 4]            # 4-sided die is 4

# Probabilities
P_A = len(event_A) / total_outcomes
P_B = len(event_B) / total_outcomes

# Conditional probabilities
P_A_given_B = len([o for o in event_B if o in event_A]) / len(event_B)
P_B_given_A = len([o for o in event_A if o in event_B]) / len(event_A)

# Joint probability
P_A_and_B = len([o for o in sample_space if o in event_A and o in event_B]) / total_outcomes

# Display results
print(f"Total outcomes: {total_outcomes}")
print(f"P(A) = {P_A}")
print(f"P(B) = {P_B}")
print(f"P(A | B) = {P_A_given_B}")
print(f"P(B | A) = {P_B_given_A}")
print(f"P(A and B) = {P_A_and_B}")

# Verify probability identities
print("\nVerification:")
print(f"P(A) * P(B | A) = {P_A * P_B_given_A}")
print(f"P(B) * P(A | B) = {P_B * P_A_given_B}")
```
