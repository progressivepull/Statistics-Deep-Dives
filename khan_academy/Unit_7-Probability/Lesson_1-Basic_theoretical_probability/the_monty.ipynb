# The Monty Hall Problem – Probability Summary

The **Monty Hall problem** is a classic probability thought experiment based on the game show *Let’s Make a Deal*, hosted by Monty Hall. It explores whether switching choices affects your chances of winning.

---

## Problem Setup

- You are presented with **three doors**:
  - Behind **one door** is a valuable prize (e.g., a car).
  - Behind the **other two doors** are undesirable prizes (e.g., goats).
- You choose **one door**.
- The host, who **knows where the prize is**, opens **one of the remaining two doors** that does **not** contain the prize.
- You are then asked whether you want to:
  - **Stay** with your original choice, or
  - **Switch** to the remaining unopened door.

The key question is:  
**Does switching improve your chances of winning?**

---

## Strategy 1: Do Not Switch (Stick With the Original Choice)

When you first choose a door:

- There are **3 equally likely possibilities**.
- Only **1** of them has the prize.

### Probability of Winning

$$
P(\text{Win | Do not switch}) = \frac{1}{3}
$$

### Probability of Losing

$$
P(\text{Lose | Do not switch}) = \frac{2}{3}
$$

If you never switch, your chance of winning stays at **1/3**.

---

## Strategy 2: Always Switch

To understand this strategy, consider how switching leads to a win:

- If your **initial choice is wrong** (which happens 2 out of 3 times), the host removes the other wrong door.
- Switching then **guarantees** you get the prize.
- If your **initial choice is correct** (1 out of 3 times), switching makes you lose.

---

### Probability of Winning When Switching

The probability of winning by switching is the probability that your **initial choice was wrong**:

$$
P(\text{Win | Switch}) = \frac{2}{3}
$$

### Probability of Losing When Switching

$$
P(\text{Lose | Switch}) = \frac{1}{3}
$$

---

## Why Switching Works

- Initially:
  - Your chosen door has a **1/3 chance** of hiding the prize.
  - The other two doors together have a **2/3 chance**.
- When the host opens a losing door, that **2/3 probability collapses** onto the single remaining unopened door.
- By switching, you capture that **2/3 chance**.

---

## Final Comparison

| Strategy        | Probability of Winning |
|-----------------|------------------------|
| Do not switch   | $ \frac{1}{3} $      |
| Always switch   | $ \frac{2}{3} $      |

---

## Key Takeaways

- Switching doors **doubles your chance of winning**.
- The result is counterintuitive but mathematically sound.
- The host’s knowledge and deliberate choice of a losing door are crucial.
- **Always switching is the optimal strategy**.

The Monty Hall problem shows how human intuition can be misleading—and how probability clarifies what’s really happening.


```python
# -----------------------------
# Monty Hall Problem Simulation
# -----------------------------

import random

def monty_hall_trial(switch: bool) -> bool:
    """
    Run a single Monty Hall trial.
    Returns True if the contestant wins, False otherwise.
    """
    doors = [0, 1, 2]                 # Three doors
    prize_door = random.choice(doors)
    initial_choice = random.choice(doors)

    # Host opens a door that is not the prize and not the player's choice
    possible_doors_to_open = [
        d for d in doors
        if d != prize_door and d != initial_choice
    ]
    opened_door = random.choice(possible_doors_to_open)

    # If switching, choose the remaining unopened door
    if switch:
        final_choice = next(
            d for d in doors
            if d not in (initial_choice, opened_door)
        )
    else:
        final_choice = initial_choice

    return final_choice == prize_door


def run_simulation(trials=100000):
    stay_wins = 0
    switch_wins = 0

    for _ in range(trials):
        if monty_hall_trial(switch=False):
            stay_wins += 1
        if monty_hall_trial(switch=True):
            switch_wins += 1

    print("Monty Hall Simulation Results")
    print(f"Trials: {trials}")
    print(f"P(Win | Stay)   ≈ {stay_wins / trials:.4f}")
    print(f"P(Win | Switch) ≈ {switch_wins / trials:.4f}")


# Run the simulation
run_simulation()
```
