We have a bag with 9 red marbles, 2 blue marbles, and 3 green marbles in it. What is the probability of randomly selecting a non-Blue marble from the bag?

# Probability Exercises – Summary

This section summarizes several probability exercises that apply the same core idea:  
**probability is the fraction of all possible outcomes that satisfy a given condition.**

---

## Example 1: Selecting a Non-Blue Marble

### Problem Setup
A bag contains:
- 9 red marbles  
- 2 blue marbles  
- 3 green marbles  

All marbles are equally likely to be selected.

---

### Total Possible Outcomes
The total number of marbles is:

$$
9 + 2 + 3 = 14
$$

So, there are **14 possible outcomes**.

---

### Favorable Outcomes (Non-Blue Marbles)
Non-blue marbles include:
- Red marbles (9)
- Green marbles (3)

$$
9 + 3 = 12 \text{ non-blue marbles}
$$

---

### Probability Calculation

$$
P(\text{Non-blue}) = \frac{12}{14} = \frac{6}{7}
$$

---

### Final Answer

$$
P(\text{Non-blue}) = \frac{6}{7}
$$

---

## Example 2: Selecting a Multiple of 5

### Problem Setup
A number is randomly chosen from a list of **12 numbers**, all equally likely.

---

### Total Possible Outcomes

$$
12
$$

---

### Favorable Outcomes
A number is a multiple of 5 if it ends in **0 or 5**.  
From the list, there are **7 numbers** that meet this condition.

---

### Probability Calculation

$$
P(\text{Multiple of 5}) = \frac{7}{12}
$$

---

### Final Answer

$$
P(\text{Multiple of 5}) = \frac{7}{12}
$$

---

## Example 3: Random Point Inside Two Circles

### Problem Setup
- The **circumference** of a large circle is $36\pi$
- A smaller circle inside it has an **area** of $16\pi$
- A point is chosen at random from the larger circle

We want the probability that the point also lies inside the smaller circle.

---

### Step 1: Find the Radius of the Larger Circle

Circumference formula:

$$
C = 2\pi r
$$

$$
36\pi = 2\pi r \Rightarrow r = 18
$$

---

### Step 2: Find the Area of the Larger Circle

$$
A = \pi r^2 = \pi(18)^2 = 324\pi
$$

---

### Step 3: Use Area Ratio for Probability

Since points are chosen uniformly, the probability is the **ratio of areas**:

$$
P = \frac{\text{Area of smaller circle}}{\text{Area of larger circle}}
= \frac{16\pi}{324\pi}
$$

Cancel $\pi$ and simplify:

$$
P = \frac{16}{324} = \frac{4}{81}
$$

---

### Final Answer

$$
P = \frac{4}{81}
$$

---

## Key Takeaways

- Probability is calculated as:

$$
P(\text{event}) = \frac{\text{number of favorable outcomes}}{\text{total outcomes}}
$$

- For **discrete outcomes** (marbles, numbers), count outcomes directly  
- For **continuous outcomes** (points in a region), use **area ratios**
- Complex wording often hides a simple counting or ratio problem


```python
# -----------------------------
# Example 1: Non-Blue Marble
# -----------------------------

# Marble counts
red = 9
blue = 2
green = 3

# Total marbles
total_marbles = red + blue + green

# Non-blue marbles
non_blue = red + green

# Probability calculation
prob_non_blue = non_blue / total_marbles

print("Example 1: Non-Blue Marble")
print("Total marbles:", total_marbles)
print("Non-blue marbles:", non_blue)
print("Probability of selecting a non-blue marble:", prob_non_blue)
print()


# -----------------------------
# Example 2: Multiple of 5
# -----------------------------

# List of numbers
numbers = [32, 49, 55, 30, 56, 28, 50, 40, 40, 45, 3, 25]

# Count totals
total_numbers = len(numbers)
multiples_of_5 = [n for n in numbers if n % 5 == 0]

# Probability calculation
prob_multiple_5 = len(multiples_of_5) / total_numbers

print("Example 2: Multiple of 5")
print("Total numbers:", total_numbers)
print("Multiples of 5:", multiples_of_5)
print("Probability of selecting a multiple of 5:", prob_multiple_5)
print()


# -----------------------------
# Example 3: Point Inside Two Circles
# -----------------------------

import math

# Given values
circumference = 36 * math.pi
small_circle_area = 16 * math.pi

# Find radius of large circle
radius_large = circumference / (2 * math.pi)

# Area of large circle
large_circle_area = math.pi * radius_large**2

# Probability calculation using area ratio
prob_point_inside = small_circle_area / large_circle_area

print("Example 3: Point in Smaller Circle")
print("Radius of large circle:", radius_large)
print("Area of large circle:", large_circle_area)
print("Probability point lies in smaller circle:", prob_point_inside)
```
