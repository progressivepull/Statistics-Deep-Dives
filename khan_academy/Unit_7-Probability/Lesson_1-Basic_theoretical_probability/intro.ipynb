# Introduction to Probability

This notebook provides a basic overview of **probability**, focusing on intuitive explanations and simple examples such as coin flips and dice rolls.

---

## What Is Probability?

Probability is a way to describe the likelihood of an event that is fundamentally random.  
Even though we cannot predict the exact outcome of a random experiment, we can quantify how likely each outcome is.

Two common ways to think about probability are:

- **Equally Likely Outcomes**
- **Long-Run Frequency (Repeated Experiments)**

---

## Example 1: Flipping a Fair Coin

A **fair coin** has two sides:

- Heads  
- Tails  

Each side is equally likely to occur.

---

### Probability Using Equally Likely Outcomes

- Total possible outcomes: 2  
- Outcomes that satisfy the condition (Heads): 1  

$$
P(\text{Heads}) = \frac{1}{2} = 50\%
$$

---

### Probability Using Repeated Experiments

Another way to understand probability is to imagine repeating the experiment many times:

- Flip the coin thousands or millions of times  
- Count how often Heads occurs  
- As the number of trials increases, the proportion of Heads will approach **50%**

This idea illustrates that probability reflects **long-term behavior**, not guaranteed short-term outcomes.

---

## Example 2: Rolling a Fair Die

A standard fair die has six sides:

$$
\{1, 2, 3, 4, 5, 6\}
$$

Each outcome is equally likely.

---

### Probability of Rolling a 1

- Total possible outcomes: 6  
- Favorable outcomes: 1  

$$
P(1) = \frac{1}{6}
$$

---

### Probability of Rolling a 1 or a 6

- Favorable outcomes: \{1, 6\} → 2 outcomes  

$$
P(1 \text{ or } 6) = \frac{2}{6} = \frac{1}{3}
$$

---

### Probability of Rolling a 2 and a 3 (Single Roll)

This event is **impossible** in one roll.

- You cannot roll two numbers at the same time in a single trial  
- These events are **mutually exclusive**

$$
P(2 \text{ and } 3) = 0
$$

---

### Probability of Rolling an Even Number

Even numbers on a die:

- 2, 4, 6  

- Favorable outcomes: 3  
- Total outcomes: 6  

$$
P(\text{Even}) = \frac{3}{6} = \frac{1}{2}
$$

---

## Key Takeaways

- Probability measures how likely an event is to occur  
- For equally likely outcomes:

$$
P(\text{event}) = \frac{\text{number of favorable outcomes}}{\text{total outcomes}}
$$

- Over many repetitions, experimental results tend to match theoretical probabilities  
- Some events are impossible (probability 0), while others are certain (probability 1)


```python
import random

# -----------------------------
# Coin Flip Simulations
# -----------------------------

def flip_coin(num_trials):
    """
    Simulate flipping a fair coin num_trials times.
    Returns the proportion of heads.
    """
    heads_count = 0
    for _ in range(num_trials):
        if random.choice(["H", "T"]) == "H":
            heads_count += 1
    return heads_count / num_trials


# Run coin flip experiments
coin_trials = [10, 100, 1000, 10000]

print("Coin Flip Simulation (Proportion of Heads):")
for trials in coin_trials:
    print(f"Trials: {trials}, Heads Proportion: {flip_coin(trials):.4f}")


# -----------------------------
# Die Roll Simulations
# -----------------------------

def roll_die(num_trials):
    """
    Simulate rolling a fair six-sided die num_trials times.
    Returns a list of outcomes.
    """
    return [random.randint(1, 6) for _ in range(num_trials)]


# Run die roll experiment
die_trials = 10000
rolls = roll_die(die_trials)

# Probability calculations
prob_one = rolls.count(1) / die_trials
prob_one_or_six = sum(1 for r in rolls if r in (1, 6)) / die_trials
prob_even = sum(1 for r in rolls if r % 2 == 0) / die_trials

print("\nDie Roll Simulation:")
print(f"P(1) ≈ {prob_one:.4f}")
print(f"P(1 or 6) ≈ {prob_one_or_six:.4f}")
print(f"P(Even) ≈ {prob_even:.4f}")


# -----------------------------
# Impossible Event Check
# -----------------------------

print("\nImpossible Event Example:")
print("P(2 and 3 in a single roll) = 0 (cannot occur)")
```
