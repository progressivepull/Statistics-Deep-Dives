# Understanding Probability: An Intuitive Perspective

This section builds intuition about what probabilities mean, how to compare them, and what their possible values are.

---

## Comparing Probabilities with a Die Roll

Consider rolling a **fair six-sided die**.  
The possible outcomes are:

$$
\{1, 2, 3, 4, 5, 6\}
$$

Each outcome is equally likely.

---

### Probability of Rolling a Number ≤ 2

Numbers less than or equal to 2 are:

- 1
- 2

That gives **2 favorable outcomes** out of **6 total outcomes**.

$$
P(\leq 2) = \frac{2}{6} = \frac{1}{3}
$$

---

### Probability of Rolling a Number ≥ 3

Numbers greater than or equal to 3 are:

- 3, 4, 5, 6

That gives **4 favorable outcomes** out of **6 total outcomes**.

$$
P(\geq 3) = \frac{4}{6} = \frac{2}{3}
$$

---

### Which Event Is More Likely?

- $ P(\leq 2) = \frac{1}{3} $
- $ P(\geq 3) = \frac{2}{3} $

Rolling a number **greater than or equal to 3** is **more likely**.  
In fact, it is **twice as likely**, since:

$$
\frac{2}{3} = 2 \times \frac{1}{3}
$$

This matches intuition: there are twice as many favorable outcomes (4 vs. 2).

---

## The Range of Possible Probabilities

### Minimum Probability: 0

Consider the probability of rolling a **7** on a six-sided die.

- There are no outcomes that meet this condition.

$$
P(7) = 0
$$

A probability of **0** means the event is **impossible**.

---

### Maximum Probability: 1

Now consider the probability of rolling **any number from 1 to 6**.

- All possible outcomes satisfy this condition.

$$
P(1 \text{ to } 6) = 1
$$

A probability of **1** means the event is **certain**.

---

## Probabilities Are Always Between 0 and 1

- Probabilities **cannot be negative**
- Probabilities **cannot be greater than 1**
- As a percentage:
  - $1 = 100\%$
  - Values above 100% (or above 1) are not valid probabilities

---

## Events That Are “Almost Certain” or “Almost Impossible”

In real life, many events feel certain or impossible but are not truly so.

- The sun rising tomorrow has a probability **very close to 1**, but not exactly 1
- A gopher writing a great novel has a probability **very close to 0**, but not exactly 0

Probability describes **likelihood**, not absolute guarantees.

---

## Key Takeaways

- **Higher probability → more likely event**
- **Probability = 0** → impossible  
- **Probability = 1** → certain  
- **Probability = 1/2** → equally likely to happen or not happen  
- **Probability > 1/2** → more likely to happen than not  
- **Probability < 1/2** → less likely to happen than not  

Probability always lies between **0 and 1**, inclusive.


```python
# -----------------------------
# Probability Intuition Examples
# -----------------------------

import random

# -----------------------------
# Die Roll Setup
# -----------------------------

def roll_die():
    """Simulate rolling a fair six-sided die."""
    return random.randint(1, 6)


# -----------------------------
# Example 1: P(X ≤ 2) vs P(X ≥ 3)
# -----------------------------

trials = 100000
count_le_2 = 0
count_ge_3 = 0

for _ in range(trials):
    roll = roll_die()
    if roll <= 2:
        count_le_2 += 1
    if roll >= 3:
        count_ge_3 += 1

prob_le_2 = count_le_2 / trials
prob_ge_3 = count_ge_3 / trials

print("Die Roll Probabilities (Simulation)")
print(f"P(X ≤ 2) ≈ {prob_le_2}")
print(f"P(X ≥ 3) ≈ {prob_ge_3}")
print()


# -----------------------------
# Example 2: Impossible Event
# -----------------------------

# Probability of rolling a 7 on a six-sided die
prob_roll_7 = 0

print("Impossible Event:")
print("P(rolling a 7) =", prob_roll_7)
print()


# -----------------------------
# Example 3: Certain Event
# -----------------------------

# Probability of rolling a number from 1 to 6
prob_roll_1_to_6 = 1

print("Certain Event:")
print("P(rolling a number from 1 to 6) =", prob_roll_1_to_6)
print()


# -----------------------------
# Example 4: Comparing to 1/2
# -----------------------------

print("Comparison to 1/2:")
print("If probability > 0.5 → event is more likely to happen")
print("If probability < 0.5 → event is less likely to happen")
print("If probability = 0.5 → event is equally likely to happen or not")
print()


# -----------------------------
# Example 5: Relative Likelihood
# -----------------------------

ratio = prob_ge_3 / prob_le_2
print("Relative Likelihood:")
print(f"Rolling ≥ 3 is about {ratio:.2f} times as likely as rolling ≤ 2")
```
