# Probability Example: Picking a Marble from a Bag

This example illustrates how to calculate probability using a simple and intuitive approach.

---

## Problem Setup

A bag contains the following marbles:

- 3 yellow marbles  
- 2 red marbles  
- 2 green marbles  
- 1 blue marble  

We want to find the **probability of picking a yellow marble** from the bag.

---

## Understanding Probability

Probability measures the likelihood that a specific event will occur.  
In this case, the event is **picking a yellow marble**.

To find the probability, we compare:

- The number of outcomes that satisfy the event  
- The total number of possible outcomes  

---

## Sample Space

The **sample space** is the set of all possible outcomes of the experiment (also called a **trial**).

Since each marble is a possible outcome, the total number of outcomes is:

$$
3 + 2 + 2 + 1 = 8
$$

So, there are **8 possible outcomes** when one marble is drawn.

---

## Favorable Outcomes

The outcomes that satisfy the event (picking a yellow marble) are:

- 3 yellow marbles  

So, there are **3 favorable outcomes**.

---

## Calculating the Probability

$$
P(\text{Yellow}) = \frac{\text{number of favorable outcomes}}{\text{total number of outcomes}}
= \frac{3}{8}
$$

---

## Final Answer

$$
P(\text{Yellow}) = \frac{3}{8}
$$

This means there is a **3 out of 8 chance** of picking a yellow marble from the bag.

---

## Key Idea

Probability is often simpler than it sounds:

> **Count how many outcomes satisfy the event, and divide by the total number of possible outcomes.**

Fancy terms like *sample space* and *trial* just describe this basic counting process.


```python
# Probability of drawing a yellow marble from a bag

# Number of marbles by color
yellow = 3
red = 2
green = 2
blue = 1

# Total number of marbles
total_marbles = yellow + red + green + blue

# Probability calculation
prob_yellow = yellow / total_marbles

print("Total marbles:", total_marbles)
print("Yellow marbles:", yellow)
print("Probability of drawing a yellow marble:", prob_yellow)


# -----------------------------
# Optional simulation
# -----------------------------

import random

def draw_marble_simulation(trials=10000):
    marbles = (
        ["yellow"] * yellow +
        ["red"] * red +
        ["green"] * green +
        ["blue"] * blue
    )
    
    yellow_count = 0
    for _ in range(trials):
        if random.choice(marbles) == "yellow":
            yellow_count += 1
    
    return yellow_count / trials


simulated_probability = draw_marble_simulation()
print("Simulated probability (approx):", simulated_probability)
```
