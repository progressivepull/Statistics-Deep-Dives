# Theoretical vs. Experimental Probability

This section contrasts **theoretical probability**, which can be calculated exactly in simple, well-defined situations, with **experimental probability**, which is estimated from real-world data when systems are too complex to model precisely.

---

## Theoretical Probability

**Theoretical probability** is used when:
- All possible outcomes are known
- Outcomes are equally likely
- The system is simple and well-defined

### Example 1: Flipping a Fair Coin

- Possible outcomes: Heads, Tails
- Both outcomes are equally likely

$$
P(\text{Heads}) = \frac{1}{2}
$$

---

### Example 2: Rolling a Fair Die

A fair six-sided die has outcomes:

$$
\{1, 2, 3, 4, 5, 6\}
$$

To find the probability of rolling a number **greater than or equal to 3**:

- Favorable outcomes: 3, 4, 5, 6 (4 outcomes)
- Total outcomes: 6

$$
P(\geq 3) = \frac{4}{6} = \frac{2}{3}
$$

These probabilities are exact *in theory*, assuming the die is fair and rolled fairly.

---

## When Theoretical Probability Breaks Down

Many real-world situations are too complex for exact theoretical modeling.  
Examples include:
- Sports games
- Human behavior
- Weather-dependent events
- Systems involving many interacting variables

In such cases, **theoretical probability is impractical or impossible to compute exactly**.

---

## Experimental Probability

**Experimental probability** estimates likelihoods based on **observed data** and past experience.

It is calculated as:

$$
P(\text{event}) \approx \frac{\text{number of times the event occurred}}{\text{total number of trials}}
$$

---

## Example: Football Scoring Data

Suppose a football team has played **16 games**, with the following scoring distribution:

| Points Scored | Number of Games |
|---------------|-----------------|
| 0–9           | 2               |
| 10–19         | 4               |
| 20–29         | 5               |
| 30–39         | 3               |
| 40–49         | 2               |

---

### Estimating a Probability

We want to estimate the probability that the team scores **30 or more points** in the next game.

- Games with ≥ 30 points: $3 + 2 = 5$
- Total games: 16

$$
P(\text{Score} \geq 30) \approx \frac{5}{16}
$$

---

## Important Cautions

- This value is **not exact**
- It is based only on **past data**
- Future games may differ due to:
  - Opponent strength
  - Weather
  - Injuries or illness
  - Coaching changes
  - Player development

Experimental probability should be treated as an **estimate**, not a certainty.

---

## Key Takeaways

- **Theoretical probability** applies to simple, idealized systems
- **Experimental probability** applies to complex, real-world situations
- Experimental probabilities depend on data and can change over time
- Past performance is informative—but never guarantees future results
- Always interpret experimental probabilities **with caution**


```python
# ---------------------------------------
# Theoretical vs Experimental Probability
# ---------------------------------------

import random

# -----------------------------
# Theoretical Probability Examples
# -----------------------------

# Coin flip
theoretical_coin_heads = 1 / 2
print("Theoretical P(Heads on a fair coin):", theoretical_coin_heads)

# Die roll: P(X >= 3)
theoretical_die_ge_3 = 4 / 6
print("Theoretical P(Roll ≥ 3 on a fair die):", theoretical_die_ge_3)
print()


# -----------------------------
# Experimental Probability
# -----------------------------

def flip_coin_experiment(trials=100000):
    heads = 0
    for _ in range(trials):
        if random.choice(["H", "T"]) == "H":
            heads += 1
    return heads / trials

def roll_die_experiment(trials=100000):
    count = 0
    for _ in range(trials):
        if random.randint(1, 6) >= 3:
            count += 1
    return count / trials

print("Experimental P(Heads):", flip_coin_experiment())
print("Experimental P(Roll ≥ 3):", roll_die_experiment())
print()


# -----------------------------
# Football Scoring Example
# -----------------------------

# Past scoring data (games played)
games_scored = {
    "0-9": 2,
    "10-19": 4,
    "20-29": 5,
    "30-39": 3,
    "40-49": 2
}

# Total games played
total_games = sum(games_scored.values())

# Games with scores >= 30
games_ge_30 = games_scored["30-39"] + games_scored["40-49"]

# Experimental probability estimate
experimental_prob_ge_30 = games_ge_30 / total_games

print("Football Scoring Data:")
print("Total games played:", total_games)
print("Games with ≥ 30 points:", games_ge_30)
print("Experimental P(Score ≥ 30):", experimental_prob_ge_30)
print()


# -----------------------------
# Interpretation Helper
# -----------------------------

if experimental_prob_ge_30 > 0.5:
    print("Based on past data, scoring ≥ 30 points is more likely than not.")
else:
    print("Based on past data, scoring ≥ 30 points is less likely than not.")

print("\nNOTE: This is an estimate based on past experience, not a guaranteed probability.")
```
