# Predicting Outcomes Using Probability: Spinner Example

This example shows how **theoretical probability** can be used to make a **reasonable prediction** about the outcome of many repeated trials.

---

## Spinner Setup

A fair spinner is divided into **7 equal sections**:

- **4 sections** show an **elephant**
- **2 sections** show a **mouse**
- **1 section** shows a **monkey**

Because the spinner is fair, each section is **equally likely** to occur on any spin.

---

## Probability of Getting an Elephant (One Spin)

There are:
- 7 total possible outcomes
- 4 favorable outcomes (elephant)

$$
P(\text{Elephant}) = \frac{4}{7}
$$

---

## Making a Prediction Over Many Spins

Suppose the spinner is spun **210 times**.

A reasonable prediction is that the elephant will appear in approximately **4/7 of the spins**.

### Expected Number of Elephants

$$
210 \times \frac{4}{7} = 30 \times 4 = 120
$$

So, a reasonable prediction is:

> **The spinner will land on an elephant about 120 times out of 210 spins.**

---

## Important Clarifications

- This is **not a guarantee**
- It is possible (though unlikely) to get:
  - More than 120 elephants (e.g., 123 or 128)
  - Fewer than 120 elephants (e.g., 110 or 90)
- In theory, it is even possible to get:
  - 0 elephants
  - 210 elephants  
  (both are extremely unlikely but possible)

---

## What This Prediction Means

- The prediction is based on **theoretical probability**
- It represents the **most reasonable expectation** before running the experiment
- It does **not** claim certainty

The idea is:

> If an event has probability $ \frac{4}{7} $, then over many trials, we expect it to occur about $ \frac{4}{7} $ of the time.

---

## Key Takeaways

- Probability helps guide **expectations**, not guarantees
- Predictions improve with **larger numbers of trials**
- Actual results will vary due to randomness
- Theoretical probability informs what is **most reasonable**, not what is **certain**
- Expectation ≠ exact outcome

This reasoning is foundational for understanding **long-run behavior** in probability and statistics.


```python
# ---------------------------------------
# Spinner Prediction and Simulation
# ---------------------------------------

import random

# Spinner setup
# 4 elephants, 2 mice, 1 monkey (7 equally likely outcomes)
spinner = ["Elephant"] * 4 + ["Mouse"] * 2 + ["Monkey"]

# -----------------------------
# Theoretical prediction
# -----------------------------

total_spins = 210
prob_elephant = 4 / 7
expected_elephants = total_spins * prob_elephant

print("Theoretical probability of Elephant:", prob_elephant)
print("Expected number of Elephants in 210 spins:", expected_elephants)
print()


# -----------------------------
# Experimental simulation
# -----------------------------

def spin_spinner(trials):
    elephant_count = 0
    for _ in range(trials):
        if random.choice(spinner) == "Elephant":
            elephant_count += 1
    return elephant_count


# Run a single simulation
elephants_observed = spin_spinner(total_spins)

print("Simulation result:")
print("Elephants observed:", elephants_observed)
print("Experimental proportion:", elephants_observed / total_spins)
print()


# -----------------------------
# Multiple simulations to see variation
# -----------------------------

simulations = 1000
results = [spin_spinner(total_spins) for _ in range(simulations)]

average_elephants = sum(results) / simulations
min_elephants = min(results)
max_elephants = max(results)

print("After", simulations, "simulations:")
print("Average elephants:", average_elephants)
print("Minimum elephants:", min_elephants)
print("Maximum elephants:", max_elephants)
print()

print("Note:")
print("• The expected value is 120 elephants")
print("• Individual experiments vary due to randomness")
print("• Over many simulations, the average approaches the theoretical prediction")
```
