# Comparing Theoretical and Experimental Probability with Marbles

This example illustrates how **experimental results** can either support or challenge a **theoretical probability**, especially as the number of trials increases.

---

## Theoretical Probability Setup

Suppose a bag contains:
- 50 **magenta** marbles
- 50 **blue** marbles

There are **100 marbles total**, and the bag is thoroughly mixed.  
You pick one marble at random while blindfolded.

### Theoretical Probability

Since all marbles are assumed to be equally likely:

$$
P(\text{Magenta}) = \frac{50}{100} = \frac{1}{2}
$$

So, **theoretical probability** predicts a 50% chance of picking a magenta marble.

---

## Experimental Probability: Small Number of Trials

Now suppose you actually run the experiment.

- After **10 trials**:
  - 7 magenta
  - 3 blue

This result differs from the expected 5–5 split, but this is **not surprising**.

### Interpretation

- With only 10 trials, random variation is large
- Getting 7 magenta out of 10 is **well within reason**
- This result alone does **not** contradict the theoretical probability

---

## Experimental Probability: Large Number of Trials

Now consider a much larger experiment.

- After **10,000 trials**:
  - 8,000 magenta
  - 2,000 blue

### Experimental Probability

$$
P_{\text{experimental}}(\text{Magenta}) = \frac{8000}{10000} = 0.8
$$

This is **very different** from the theoretical probability of 0.5.

---

## Why This Raises Concern

With a large number of trials:
- Random variation tends to average out
- Results should be close to the theoretical probability if assumptions are correct

Getting **80% magenta** after 10,000 trials is **extremely unlikely** if the true probability were 50%.

This suggests that something may be wrong with the original assumptions.

---

## Possible Explanations

If the marbles were truly equally likely, this result would be very rare. Possible reasons for the discrepancy include:

- Blue marbles are heavier and settle to the bottom
- Magenta marbles are easier to grip
- Differences in texture affect selection
- The mixing process is imperfect

In short, the experiment may **not** be as fair as initially assumed.

---

## Key Takeaways

- **Theoretical probability** assumes ideal, perfectly fair conditions
- **Experimental probability** is based on observed data
- Small sample sizes can vary widely by chance
- Large sample sizes should closely match theory
- Persistent discrepancies suggest hidden bias or flawed assumptions

When experimental results strongly disagree with theory—especially after many trials—it’s a signal to **re-examine the system**, not ignore the data.


```python
# ---------------------------------------
# Theoretical vs Experimental Probability
# Marble Experiment Simulation
# ---------------------------------------

import random

# -----------------------------
# Theoretical Probability
# -----------------------------

magenta = 50
blue = 50
total = magenta + blue

theoretical_prob_magenta = magenta / total
print("Theoretical P(Magenta):", theoretical_prob_magenta)
print()


# -----------------------------
# Experimental Probability Function
# -----------------------------

def marble_experiment(trials):
    """
    Simulate drawing marbles with replacement.
    Returns number of magenta and blue marbles drawn.
    """
    marbles = ["Magenta"] * magenta + ["Blue"] * blue
    magenta_count = 0
    blue_count = 0

    for _ in range(trials):
        draw = random.choice(marbles)
        if draw == "Magenta":
            magenta_count += 1
        else:
            blue_count += 1

    return magenta_count, blue_count


# -----------------------------
# Small Number of Trials
# -----------------------------

trials_small = 10
m_small, b_small = marble_experiment(trials_small)

print(f"After {trials_small} trials:")
print("Magenta:", m_small)
print("Blue:", b_small)
print("Experimental P(Magenta):", m_small / trials_small)
print()


# -----------------------------
# Large Number of Trials
# -----------------------------

trials_large = 10000
m_large, b_large = marble_experiment(trials_large)

print(f"After {trials_large} trials:")
print("Magenta:", m_large)
print("Blue:", b_large)
print("Experimental P(Magenta):", m_large / trials_large)
print()


# -----------------------------
# Interpretation Helper
# -----------------------------

if abs((m_large / trials_large) - theoretical_prob_magenta) > 0.1:
    print("⚠️ Large deviation detected.")
    print("This suggests the marbles may not be equally likely.")
else:
    print("✅ Experimental results align with theoretical probability.")
```
