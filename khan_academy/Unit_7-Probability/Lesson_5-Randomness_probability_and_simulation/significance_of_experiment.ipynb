In an experiment aimed at studying the effect of advertising on eating behavior in children, a group of 500 children, 7-11 years old, were randomly assigned to two different groups. After randomization, each child was asked to watch a cartoon in a private room, containing a large bowl of goldfish crackers. The cartoon included 2 commercial breaks.

The first group watched food commercials (mostly snacks), while the second group watched non-food commercials (games and entertainment products). Once the child finished watching the cartoon, the conductors of the experiment weighed the crackers bowl to measure how many grams of crackers the child ate. They found that the mean amount of crackers eaten by the children who watched food commercials is 10 grams greater than the mean amount of crackers eaten by the children who watched non-food commercials.

Using a simulator, they re-randomized the results into two new groups and measured the difference between the means of the new groups. They repeated this simulation 150 times, and plotted the resulting differences, as given below.

# Assessing Statistical Significance with Simulation

This example explains how **randomization (permutation) testing** can be used to determine whether an observed result is **statistically significant** or could have occurred by chance.

---

## Experiment Overview

- **Participants**: 500 children, ages 7–11
- **Design**: Randomized experiment with two groups
- **Activity**:
  - Each child watched a cartoon with two commercial breaks
  - A large bowl of goldfish crackers was provided
- **Measurement**: Grams of crackers eaten

---

## Groups

- **Treatment Group**: Watched **food commercials** (mostly snacks)
- **Control Group**: Watched **non-food commercials** (games and entertainment)

After the cartoon:
- The crackers were weighed for each child
- The **mean amount eaten** was calculated for each group

---

## Observed Result

- Children who watched **food commercials** ate, on average, **10 grams more** crackers than those who watched non-food commercials.

At first glance, this suggests that food advertising may increase eating behavior.

---

## Key Question

> **Could this 10-gram difference have happened by chance**, even if commercials had no real effect?

Random assignment alone can sometimes produce large differences between groups.

---

## Simulation (Randomization Test)

To answer this question, researchers performed a **re-randomization simulation**:

1. Keep each child’s measured cracker consumption fixed
2. Randomly reassign the 500 children into two new groups
3. Compute the difference in means between the two new groups
4. Repeat this process **150 times**
5. Plot the resulting differences

This simulates what differences would look like **if commercials had no effect**.

---

## Simulation Results

- Most simulations showed **little or no difference** in means
- A **10-gram difference** occurred **only about 1–2 times out of 150**
- This corresponds to a probability of **less than 2%**, and likely **under 1%**

In other words:
> When group assignments are random, differences as large as 10 grams are **very rare**.

---

## Interpretation

Because the observed 10-gram difference:
- Occurs **less than 1% of the time** under random assignment
- Is far outside what we typically expect by chance

We conclude that the result is **very unlikely to be due to random variation alone**.

---

## Statistical Significance

- A common threshold for significance is **5%**
- This result is well below that threshold

### Conclusion:
> **Yes — the experiment’s result is statistically significant.**

There is strong evidence that watching food commercials caused children to eat more crackers.

---

## Key Takeaways

- Randomization tests help assess whether results could occur by chance
- Large observed differences must be compared to a null distribution
- Rare outcomes under the null hypothesis indicate significance
- A result occurring <5% of the time by chance is considered statistically significant
- This experiment shows convincing evidence of an advertising effect


```python
import random
import matplotlib.pyplot as plt

# -----------------------------
# Step 1: Create observed data
# -----------------------------

random.seed(0)

num_children = 500
group_size = num_children // 2

# Control group: baseline cracker consumption (grams)
control = [random.gauss(30, 10) for _ in range(group_size)]

# Treatment group: +10 grams on average
treatment = [x + 10 for x in control]

# Observed difference in means
observed_diff = (sum(treatment) / group_size) - (sum(control) / group_size)

# Combine all measurements
all_data = treatment + control

# -----------------------------
# Step 2: Randomization simulation
# -----------------------------

num_simulations = 150
differences = []

for _ in range(num_simulations):
    random.shuffle(all_data)
    new_group1 = all_data[:group_size]
    new_group2 = all_data[group_size:]
    
    diff = (sum(new_group1) / group_size) - (sum(new_group2) / group_size)
    differences.append(diff)

# -----------------------------
# Step 3: Plot results
# -----------------------------

plt.figure()
plt.scatter(differences, [0]*len(differences))
plt.axvline(observed_diff)
plt.xlabel("Difference in Means (grams)")
plt.yticks([])
plt.title("Randomization Test: Difference in Means")
plt.show()

# -----------------------------
# Step 4: Significance check
# -----------------------------

extreme_count = sum(1 for d in differences if abs(d) >= abs(observed_diff))
p_value = extreme_count / num_simulations

print("Observed difference in means:", observed_diff)
print("Extreme simulations:", extreme_count)
print("Estimated p-value:", p_value)
```

# 📊 What This Code Shows

* The **blue dots** represent differences in means under random assignment
* The **vertical line at ~10 grams** is the observed experimental result
* Almost all simulated differences cluster near 0
* The observed difference lies **far outside** the randomization distribution

# 📌 Interpretation

* **Estimated p-value ≈ 0**
* The probability of seeing a 10-gram difference **by chance alone**is extremely small
* This meets (and exceeds) the standard **5% significance threshold**

# ✅ Final Conclusion

* **Yes — according to the simulations, the experiment’s result is statistically significant.**
