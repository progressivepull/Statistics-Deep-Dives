# Estimating the Average Number of Boxes to Collect All 6 Prizes

This problem explores how to estimate the **average number of cereal boxes** needed to collect all six prizes when each prize is **equally likely** to appear in any box. This is a classic example of using **simulation** to approximate an expected value.

---

## Problem Setup

- There are **6 distinct prizes**, labeled: 1, 2, 3, 4, 5, 6
- Each cereal box contains **one prize**
- Each prize is **equally likely**
- Amanda wants to know how many boxes it takes **on average** to collect all six prizes

---

## Two Possible Approaches

1. **Theoretical (mathematical)** approach  
   - Derive an exact expected value using probability theory

2. **Experimental (simulation)** approach  
   - Simulate opening many boxes using random numbers
   - Record how many boxes it takes each time to get all six prizes
   - Average the results over many trials

This example focuses on the **simulation approach**.

---

## Simulation Method

- Assign each prize a number: **1 through 6**
- Generate a long sequence of random digits
- Read the digits from left to right:
  - Digits **1–6** represent valid prizes
  - Digits **0, 7, 8, 9** are ignored
- Each valid digit represents opening one cereal box
- Continue until all six prizes have been collected
- Count how many valid boxes were needed

Each run through the numbers is one **experiment**.

---

## Experiment 1

Collected prizes in this order (valid numbers only):

1 → 5 → 6 → 6 → 2 → 4 → 6 → 3

- All six prizes collected after **8 boxes**

**Result:**  
$$
\text{Experiment 1} = 8 \text{ boxes}
$$

---

## Experiment 2

Collected prizes in this order (valid numbers only):

2 → 1 → 2 → 5 → 4 → 3 → 6

- All six prizes collected after **17 boxes**

**Result:**  
$$
\text{Experiment 2} = 17 \text{ boxes}
$$

---

## Experiment 3

Collected prizes in this order (valid numbers only):

4 → 5 → 2 → 6 → 1 → 3

- All six prizes collected after **10 boxes**

**Result:**  
$$
\text{Experiment 3} = 10 \text{ boxes}
$$

---

## Calculating the Experimental Average

Using the three experiments:

$$
\text{Average} = \frac{8 + 17 + 10}{3} = \frac{35}{3} = 11\tfrac{2}{3}
$$

---

## Interpretation

- The average of **11⅔ boxes** is **not guaranteed** to be the true expected value
- It is only based on **three experiments**
- To get a better estimate:
  - Run **many more simulations**
  - Larger sample sizes lead to more reliable averages

---

## Key Takeaways

- Simulation is useful when theoretical solutions are difficult or unknown
- Small numbers of trials can vary widely
- As the number of experiments increases, the average:
  - Becomes more stable
  - Gets closer to the true expected value
- This approach helps estimate real-world expectations through data

This process gives Amanda a practical way to estimate how many cereal boxes she might need—on average—to collect all six prizes.


```python
# ---------------------------------------
# Amanda Young – Prize Collection Simulation
# Using the GIVEN random number sequence
# ---------------------------------------

# Given sequence of digits (as a string)
digits = """
1566 2476 0387 8218 2959 4313 3220 3113 0564 9995 5278 7671 1929
1372 3911 0248 9395 9001 2359 8782 6728 3759 3838 6466 0889 3346
4068 7735 1337 4847 8008 1149 1086 4899 4058 3307 5597 3057 8670
9793 4667 7787 5361 2207 6590 3005 7976 4606 1149 1769 9307 3866
"""

# Clean the digit string into a list of integers
digits = digits.replace(" ", "").replace("\n", "")
numbers = [int(d) for d in digits]

PRIZES = {1, 2, 3, 4, 5, 6}

def run_experiments(numbers):
    """
    Runs prize-collection experiments using the given digit stream.
    Digits 1–6 are valid prizes; others are ignored.
    Each experiment stops once all 6 prizes are collected.
    """
    experiments = []
    collected = set()
    box_count = 0

    for n in numbers:
        if n in PRIZES:
            box_count += 1
            collected.add(n)

            if collected == PRIZES:
                experiments.append(box_count)
                collected = set()
                box_count = 0

    return experiments


# -----------------------------
# Run the simulation
# -----------------------------

results = run_experiments(numbers)

# -----------------------------
# Output results
# -----------------------------

print("Amanda Young – Prize Collection Simulation")
print("Number of completed experiments:", len(results))
print("Boxes needed per experiment:", results)

if results:
    average_boxes = sum(results) / len(results)
    print("Average boxes needed:", average_boxes)
```

# What this code does

* Uses **only the provided random digits**
* Treats digits **1–6** as valid prizes
* Ignores **0, 7, 8, 9**
* Simulates **ultiple experiments**
* Stops each experiment when all 6 prizes are collected
* Computes the **average number of boxes needed**
