# The Law of Large Numbers: Experimental vs. Theoretical Probability

This section explains how **experimental probability** tends to get closer to **theoretical probability** as the number of trials increases—a principle known as the **Law of Large Numbers**.

---

## The Law of Large Numbers

The **Law of Large Numbers** states that:

> As the number of experiments or trials increases, the experimental probability of an event is likely to get closer to its theoretical probability.

When only a small number of trials are conducted, experimental results can differ significantly from what theory predicts. As the number of trials grows into the hundreds, thousands, or more, large discrepancies become increasingly unlikely.

---

## Coin Flip Simulation Example

Consider a coin with a **50% probability of landing heads**.

A computer simulation allows us to:
- Flip a (fair) coin many times
- Track the **proportion of heads** after each flip
- Observe how this proportion changes over time

---

## Behavior with Few Trials

With a small number of flips:
- After 1 flip, the experimental probability might be **100% heads**
- After 2 flips, it could drop to **50%**
- After 3 flips, it might fall to **33%**
- After 5 flips, it could rise to **60%**

### Key Insight
With only a handful of trials, **large swings** in experimental probability are completely normal and expected.

---

## Increasing the Number of Trials

As the simulation continues to:
- 10 flips
- 200 flips
- 800 flips
- 1,000+ flips

Several patterns emerge:
- There are runs of many heads in a row
- There are runs of many tails in a row
- The experimental probability may still differ from 50% after hundreds of flips

However, these fluctuations begin to **average out** over time.

---

## Convergence Toward Theoretical Probability

As the number of flips grows into the thousands:
- The proportion of heads moves closer to **50%**
- Differences between experimental and theoretical probability shrink
- Large deviations become less likely

For example:
- At around 1,200 flips, the proportion might be around **50.6%**
- With even more flips, it would likely move even closer to **50%**

---

## Important Clarifications

- The Law of Large Numbers does **not** guarantee exact equality
- Long runs of heads or tails are still possible
- What changes is the **likelihood of large divergence**, which decreases as trials increase

---

## Key Takeaways

- Small sample sizes can be misleading
- Random variation is normal in short runs
- Larger samples provide more reliable estimates
- Experimental probability **tends toward** theoretical probability over time
- This principle underpins much of statistics and data science

The Law of Large Numbers helps explain why probability theory is useful for predicting long-term behavior—even when short-term outcomes appear chaotic.


```python
# ---------------------------------------
# Law of Large Numbers: Coin Flip Simulation
# ---------------------------------------

import random
import matplotlib.pyplot as plt

# Number of coin flips
trials = 5000

heads_count = 0
proportions = []

# Simulate coin flips and track experimental probability
for i in range(1, trials + 1):
    if random.choice(["H", "T"]) == "H":
        heads_count += 1
    proportions.append(heads_count / i)

# ---------------------------------------
# Visualization
# ---------------------------------------

plt.figure()
plt.plot(proportions)
plt.axhline(0.5)   # theoretical probability
plt.xlabel("Number of Flips")
plt.ylabel("Proportion of Heads")
plt.title("Law of Large Numbers: Experimental Probability Converging to 0.5")
plt.show()
```
