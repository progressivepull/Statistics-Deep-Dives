# Experimental Probability: *Three Rolls to 10*

This example explores how to estimate the probability of winning a game using **experimental probability** rather than a direct theoretical calculation.

---

## Game Description

**Three Rolls to 10** is a game invented by Pascale Ricketts:

- You roll a **fair six-sided die three times**
- If the **sum of the three rolls is 10 or greater**, you win
- If the sum is **less than 10**, you lose

The question is:

> **What is the probability of winning Three Rolls to 10?**

---

## Approach: Experimental Probability

Instead of listing every possible outcome or calculating the probability theoretically, the strategy here is to:

- Run **many experiments**
- Observe how often the game is won
- Estimate the probability based on the **proportion of wins**

The more experiments we perform, the more accurate our estimate becomes.

---

## Using Random Digits to Simulate Rolls

A computer generates a long list of random digits from **0 to 9**.

To simulate rolling a fair six-sided die:
- Digits **1–6** are treated as valid rolls
- Digits **0, 7, 8, and 9** are ignored (as if the die roll didn’t count)

Each experiment:
- Uses the **first three valid digits** as the three die rolls
- Adds the rolls to find the sum
- Determines whether the game is won or lost

---

## Sample Experiments

A table is created with:
- **Experiment number**
- **Sum of the three rolls**
- **Win or loss**

### Results from the First 10 Experiments

- Experiments 1–3: wins (sums ≥ 10)
- Experiments 4–8: losses (sums < 10)
- Experiments 9–10: wins

Out of **10 experiments**:
- **5 wins**
- **5 losses**

---

## Experimental Probability After 10 Trials

$$
P(\text{win}) \approx \frac{5}{10} = 0.50
$$

So based on these 10 experiments, the experimental probability of winning appears to be **50%**.

---

## Important Considerations

- This does **not** mean the true probability is exactly 50%
- Ten trials is a **small sample size**
- Experimental probability can fluctuate significantly with few trials

To get a better estimate:
- Run **many more experiments** (e.g., thousands or tens of thousands)
- Use a computer program to automate the simulation

---

## Key Takeaways

- Experimental probability estimates likelihood based on observed outcomes
- Small samples can be misleading
- Larger numbers of trials produce more reliable estimates
- Simulations are especially useful when theoretical calculations are complex

This method provides a practical way to approximate the probability of winning *Three Rolls to 10* and motivates further experimentation to approach the true theoretical value.


```python
# ---------------------------------------
# Pascale Ricketts – "3 Rolls to Ten"
# Experimental Probability using GIVEN digits
# ---------------------------------------

# Given random digit stream
digits = """
1566 2476 0387 8218 2959 4313 3220 3113 0564 9995 5278 7671 1929
1372 3911 0248 9395 9001 2359 8782 6728 3759 3838 6466 0889 3346
4068 7735 1337 4847 8008 1149 1086 4899 4058 3307 5597 3057 8670
9793 4667 7787 5361 2207 6590 3005 7976 4606 1149 1769 9307 3866
"""

# Convert to a list of individual digits
digits = digits.replace(" ", "").replace("\n", "")
numbers = [int(d) for d in digits]

VALID_ROLLS = {1, 2, 3, 4, 5, 6}

def run_experiments(numbers):
    """
    Simulates the game '3 Rolls to Ten' using the given digits.
    Digits 1–6 are valid rolls; others are ignored.
    Each experiment consists of 3 valid rolls.
    """
    experiments = []
    current_rolls = []

    for n in numbers:
        if n in VALID_ROLLS:
            current_rolls.append(n)

            if len(current_rolls) == 3:
                experiments.append(sum(current_rolls))
                current_rolls = []

    return experiments


# -----------------------------
# Run the simulation
# -----------------------------

sums = run_experiments(numbers)

wins = [s for s in sums if s >= 10]
losses = [s for s in sums if s < 10]

# -----------------------------
# Results
# -----------------------------

print("3 Rolls to Ten – Experimental Simulation")
print("Total experiments:", len(sums))
print("Wins (sum ≥ 10):", len(wins))
print("Losses (sum < 10):", len(losses))

if sums:
    experimental_probability = len(wins) / len(sums)
    print("Experimental probability of winning:", experimental_probability)
```

# What this code does

* Uses **only the digits shown in the image**
* Treats **1–6** as valid die rolls
* Ignores **0, 7, 8, 9**
* Groups rolls into **sets of three**
* Determines wins based on **sum ≥ 10**
* Computes the **experimental probability of winning**
