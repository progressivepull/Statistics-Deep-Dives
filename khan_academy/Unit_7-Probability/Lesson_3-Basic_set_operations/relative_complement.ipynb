# Set Difference and Relative Complements

This section introduces the idea of **finding differences between sets**, also known as **relative complements**, and explains how to interpret and compute them.

---

## Set Difference: $ A - B $

The **difference of set A and set B** consists of all elements that are:

- In **set A**
- **Not** in **set B**

This is often read as:
- “A minus B”
- “B subtracted from A”
- “The relative complement of B in A”

---

### Example Sets

Let:
- $ A = \{5, 3, 12, 17, 19\} $
- $ B = \{17, 19, 6\} $

---

### Computing $ A - B $

Start with all elements in $A$, then remove anything that also appears in $B$:

- 17 → remove (in B)
- 19 → remove (in B)
- 6 → ignore (not in A)

Remaining elements:

$$
A - B = \{5, 3, 12\}
$$

---

### Interpretation

- Elements not in $B$: includes many things (numbers, colors, animals, etc.)
- **Relative complement** restricts this idea to elements that are **also in A**

So:

> $ A - B $ = elements that are **not in B but are in A**

---

## Important Clarification

If an element in $B$ is **not present** in $A$, removing it has **no effect**.

Example:
- Removing 6 from $A$ does nothing since 6 was never in $A$

---

## Alternative Notation

Set difference can also be written as:

$$
A \setminus B
$$

This notation means the same thing as $ A - B $.

---

## Reversing the Difference: $ B - A $

Now consider the difference in the opposite direction.

### Computing $ B - A $

Start with set $B$, then remove elements that appear in $A$:

- 17 → remove (in A)
- 19 → remove (in A)
- 6 → keep (not in A)

$$
B - A = \{6\}
$$

---

## Difference of a Set with Itself

Consider:

$$
A - A
$$

This means removing **all elements of A from A**.

- Every element is removed
- Nothing remains

---

### The Empty (Null) Set

$$
A - A = \varnothing
$$

- The **empty set** (or **null set**) contains **no elements**
- It represents a set with absolutely nothing in it

---

## Key Takeaways

- **Set difference** keeps elements from one set while removing those found in another
- $ A - B \neq B - A $ in general
- Removing elements that were never present has no effect
- A set minus itself results in the **empty set**
- Set difference is also called the **relative complement**


```python
# ---------------------------------------
# Set Difference and Relative Complements
# ---------------------------------------

# Define sets A and B
A = {5, 3, 12, 17, 19}
B = {17, 19, 6}

print("Set A:", A)
print("Set B:", B)
print()

# -----------------------------
# A - B (Relative complement of B in A)
# -----------------------------

A_minus_B = A - B

print("A - B (Elements in A but not in B):", A_minus_B)
print()

# -----------------------------
# B - A (Relative complement of A in B)
# -----------------------------

B_minus_A = B - A

print("B - A (Elements in B but not in A):", B_minus_A)
print()

# -----------------------------
# A - A (Empty Set)
# -----------------------------

A_minus_A = A - A

print("A - A (Empty set):", A_minus_A)
print()

# -----------------------------
# Alternative Method Using .difference()
# -----------------------------

print("Using .difference() method:")
print("A.difference(B):", A.difference(B))
print("B.difference(A):", B.difference(A))
```
