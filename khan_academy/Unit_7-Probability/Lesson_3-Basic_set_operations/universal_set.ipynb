# Universal Sets and Complements

This section introduces the ideas of a **universal set** and the **complement (absolute complement)** of a set.

---

## The Universal Set

The **universal set** (often denoted by a capital **U**) represents **all elements under consideration** in a given context.

- It is usually drawn as a **rectangle** in a Venn diagram
- Every set being discussed is a **subset** of this universe
- The universe depends on the problem:
  - All people
  - All real numbers
  - All countries
  - All objects relevant to the discussion

Although a universe could theoretically include *everything*, in practice it is restricted to what matters for the problem.

---

## Subsets of the Universal Set

Let **A** be a subset of the universal set **U**.

- Set **A** contains some elements of **U**
- Everything in **U** that is not in **A** lies outside the region representing **A**

---

## The Complement of a Set

The **complement** (or **absolute complement**) of a set **A** is:

> The set of all elements in the universe that are **not** in **A**

---

### Notation for Complements

All of the following mean the same thing:

- $ A^c $ (A complement)
- $ U - A $
- $ U \setminus A $

Each represents **all elements in the universe that are not in A**.

---

## Example with Numbers

### Define the Universe

Let the universal set be the **set of integers**:

$$
U = \mathbb{Z}
$$

> The symbol $ \mathbb{Z} $ (blackboard bold Z) comes from the German word *Zahlen*, meaning integers.

Other common number sets include:
- $ \mathbb{R} $: real numbers
- $ \mathbb{Q} $: rational numbers (Q for *quotient*)

---

### Define a Subset

Let:

$$
C = \{-5, 0, 7\}
$$

This is a **finite subset** of the infinite set of integers.

---

### Membership Notation

- $ -5 \in C $
- $ 0 \in C $
- $ 7 \in C $

The symbol $ \in $ means **“is an element of”**.

Non-members:
- $ -8 \notin C $
- $ 42 \notin C $
- $ 53 \notin C $

---

## The Complement of Set C

The complement of $ C $, denoted $ C^c $, is:

$$
C^c = U - C
$$

This includes **all integers except** $-5$, $0$, and $7$.

---

### Membership in the Complement

- $ -5 \notin C^c $
- $ 0 \notin C^c $
- $ 7 \notin C^c $
- $ 42 \in C^c $
- $ 53 \in C^c $

---

## Visual Interpretation

In a Venn diagram:
- The rectangle represents **U**
- The circle represents **C**
- Everything **outside the circle but inside the rectangle** is $ C^c $

---

## Key Takeaways

- The **universal set** defines the scope of the discussion
- The **complement** includes everything in the universe **not in the set**
- Complements depend on the chosen universe
- Membership symbols ($ \in $, $ \notin $) clarify whether elements belong to a set
- Complements are also called **absolute complements** when taken relative to the universal set


```python
# ---------------------------------------
# Universal Set and Complements
# ---------------------------------------

# Define the universal set (integers in a limited range for practicality)
U = set(range(-10, 11))   # Integers from -10 to 10

# Define subset C
C = {-5, 0, 7}

print("Universal set U:", U)
print("Subset C:", C)
print()

# -----------------------------
# Complement of C
# -----------------------------

C_complement = U - C

print("Complement of C (U - C):", C_complement)
print()

# -----------------------------
# Membership Checks
# -----------------------------

elements = [-5, 0, 7, 42, -8]

for elem in elements:
    if elem in C:
        print(f"{elem} ∈ C")
    else:
        print(f"{elem} ∉ C")

print()

for elem in elements:
    if elem in C_complement:
        print(f"{elem} ∈ Cᶜ")
    else:
        print(f"{elem} ∉ Cᶜ")

# -----------------------------
# Alternative Notation
# -----------------------------

# Using set difference method
C_complement_alt = U.difference(C)

print("\nC complement using .difference():", C_complement_alt)
```
