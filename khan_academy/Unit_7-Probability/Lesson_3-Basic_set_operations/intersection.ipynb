# Introduction to Sets and Set Operations

This section introduces the concept of a **set** and two fundamental operations on sets: **intersection** and **union**.

---

## What Is a Set?

A **set** is a collection of **distinct objects**.  
The objects in a set can be anything—numbers, colors, people, or even other sets.

Sets are usually written using **curly braces** `{ }`, with elements separated by commas.

### Example Sets

Let:

$$
X = \{3, 12, 5, 13\}
$$

$$
Y = \{14, 15, 6, 3\}
$$

Each element appears **only once**, since sets contain distinct objects.

---

## Intersection of Sets

The **intersection** of two sets contains elements that are in **both sets**.

- Notation:  
  $$
  X \cap Y
  $$
- Think of intersection as **“AND”**

### Example

Comparing sets $X$ and $Y$:

- The number **3** appears in both sets
- No other elements are shared

$$
X \cap Y = \{3\}
$$

---

## Union of Sets

The **union** of two sets contains all elements that are in **either set**.

- Notation:  
  $$
  X \cup Y
  $$
- Think of union as **“OR”**
- Duplicate elements are included **only once**

### Example

Combining all elements from $X$ and $Y$:

$$
X \cup Y = \{3, 12, 5, 13, 14, 15, 6\}
$$

The order does **not** matter.

---

## Visualizing Sets with Venn Diagrams

- Each circle represents a set
- The overlapping region represents the **intersection**
- The total area covered by both circles represents the **union**

In the example:
- The overlap contains only **3**
- The union contains all elements from both sets

---

## Second Example

Let:

$$
A = \{11, 4, 12, 7\}
$$

$$
B = \{13, 4, 12, 10, 3\}
$$

---

### Intersection

Elements common to both sets:

- 4
- 12

$$
A \cap B = \{4, 12\}
$$

This intersection can also be labeled as a new set, for example:

$$
C = A \cap B = \{4, 12\}
$$

---

### Union

All elements in either set $A$ or $B$:

$$
A \cup B = \{11, 4, 12, 7, 13, 10, 3\}
$$

---

## Key Takeaways

- A **set** is a collection of distinct objects
- **Intersection (∩)** includes elements common to both sets
- **Union (∪)** includes all elements from both sets
- Order does not matter in sets
- Venn diagrams help visualize set relationships


```python
# ---------------------------------------
# Introduction to Sets and Set Operations
# ---------------------------------------

# Define sets X and Y
X = {3, 12, 5, 13}
Y = {14, 15, 6, 3}

print("Set X:", X)
print("Set Y:", Y)
print()

# -----------------------------
# Intersection
# -----------------------------

intersection_XY = X.intersection(Y)

print("X ∩ Y (Intersection):", intersection_XY)
print()

# -----------------------------
# Union
# -----------------------------

union_XY = X.union(Y)

print("X ∪ Y (Union):", union_XY)
print()


# ---------------------------------------
# Second Example with Sets A and B
# ---------------------------------------

A = {11, 4, 12, 7}
B = {13, 4, 12, 10, 3}

print("Set A:", A)
print("Set B:", B)
print()

# Intersection of A and B
intersection_AB = A & B   # Using operator form

print("A ∩ B (Intersection):", intersection_AB)
print()

# Union of A and B
union_AB = A | B           # Using operator form

print("A ∪ B (Union):", union_AB)
print()


# ---------------------------------------
# Optional: Labeling Intersection as a New Set
# ---------------------------------------

C = intersection_AB
print("Set C (A ∩ B):", C)
```
