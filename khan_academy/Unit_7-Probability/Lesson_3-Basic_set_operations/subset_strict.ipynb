# Subsets, Strict Subsets, and Supersets

This section introduces the concepts of **subsets**, **strict (proper) subsets**, and **supersets**, using simple numerical examples.

---

## Defining the Sets

Let:

$$
A = \{1, 3, 5, 7, 18\}
$$

$$
B = \{1, 7, 18\}
$$

$$
C = \{18, 7, 1, 19\}
$$

---

## What Is a Subset?

A set is a **subset** of another set if **every element of the first set is also an element of the second set**.

- Notation:  
  $$
  B \subseteq A
  $$

---

### Is $ B \subseteq A $?

- Every element of $B$ (1, 7, 18) appears in $A$

$$
B \subseteq A \quad \text{(True)}
$$

---

## Strict (Proper) Subsets

A **strict** (or **proper**) subset means:

- All elements of one set are contained in the other **and**
- The two sets are **not equal**

- Notation:
  $$
  B \subset A
  $$

Since $A$ contains elements (3 and 5) that are not in $B$:

$$
B \subset A \quad \text{(True)}
$$

---

### A Set Compared to Itself

- Every set is a subset of itself:

$$
A \subseteq A \quad \text{(True)}
$$

- No set is a strict subset of itself:

$$
A \subset A \quad \text{(False)}
$$

---

## Checking Other Subset Relationships

### Is $ B \subseteq C $?

- $C$ contains 1, 7, and 18

$$
B \subseteq C \quad \text{(True)}
$$

- Since $C$ also contains 19, which is not in $B$:

$$
B \subset C \quad \text{(True)}
$$

---

### Is $ C \subseteq A $?

- $C$ contains 19
- $A$ does **not** contain 19

$$
C \subseteq A \quad \text{(False)}
$$

---

## Supersets

A **superset** relationship reverses the subset notation.

- If $ B \subseteq A $, then:
  $$
  A \supseteq B
  $$

This means **A contains all elements of B**.

---

### Strict Superset

If $A$ contains all elements of $B$ **and more**, then:

$$
A \supset B
$$

This is equivalent to saying $B$ is a strict subset of $A$.

---

## Key Takeaways

- **Subset (⊆)**: every element of one set appears in the other
- **Strict subset (⊂)**: subset, but not equal
- Every set is a subset of itself
- No set is a strict subset of itself
- **Supersets (⊇, ⊃)** reverse subset relationships
- Subset and superset relationships help compare and organize sets


```python
# ---------------------------------------
# Subsets, Strict Subsets, and Supersets
# ---------------------------------------

# Define sets
A = {1, 3, 5, 7, 18}
B = {1, 7, 18}
C = {18, 7, 1, 19}

print("Set A:", A)
print("Set B:", B)
print("Set C:", C)
print()

# -----------------------------
# Subset checks
# -----------------------------

print("Is B a subset of A?        ", B.issubset(A))
print("Is B a strict subset of A?", B < A)
print()

print("Is A a subset of A?        ", A.issubset(A))
print("Is A a strict subset of A?", A < A)
print()

print("Is B a subset of C?        ", B.issubset(C))
print("Is B a strict subset of C?", B < C)
print()

print("Is C a subset of A?        ", C.issubset(A))
print()

# -----------------------------
# Superset checks
# -----------------------------

print("Is A a superset of B?        ", A.issuperset(B))
print("Is A a strict superset of B?", A > B)
print()

print("Is C a superset of B?        ", C.issuperset(B))
print("Is C a strict superset of B?", C > B)
```
