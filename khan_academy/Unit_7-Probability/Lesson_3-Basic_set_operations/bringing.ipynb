# Combining Set Operations: A Step-by-Step Breakdown

This example applies multiple **set operations**—relative complements, absolute complements, intersections, and unions—to simplify a complex set expression. The key strategy is to **work from the inside out**, just like with algebraic expressions.

---

## Given Sets

The sets contain a mix of elements (numbers, colors, symbols):

- **Set A**:  
  $$
  A = \{3, 7, -5, 0, 13\}
  $$

- **Set B**:  
  $$
  B = \{0, 17, 3, \text{Blue}, ⭐\}
  $$

- **Set C**:  
  $$
  C = \{17, 3, ⭐\}
  $$

The **universal set** includes all elements shown (numbers, colors, and symbols).

---

## Step 1: Relative Complement of C in B ( $ B - C $ )

This means: *elements in B that are not in C*.

- 0 → keep (not in C)  
- 17 → remove (in C)  
- 3 → remove (in C)  
- Blue → keep (not in C)  
- ⭐ → remove (in C)  

$$
B - C = \{0, \text{Blue}\}
$$

---

## Step 2: Absolute Complement of $ B - C $

This is the set of **all elements in the universe** that are **not** 0 or Blue:

$$
(B - C)^c = \{\text{everything except } 0 \text{ and Blue}\}
$$

---

## Step 3: Intersection with A

Now intersect set A with the complement found above:

$$
A \cap (B - C)^c
$$

Check each element of A:

- 3 → keep  
- 7 → keep  
- -5 → keep  
- 0 → remove (excluded by complement)  
- 13 → keep  

$$
A \cap (B - C)^c = \{3, 7, -5, 13\}
$$

---

## Step 4: Relative Complement of This Set in A

Now subtract this result from A:

$$
A - \{3, 7, -5, 13\}
$$

Only one element remains:

$$
= \{0\}
$$

---

## Step 5: Intersection of B and C

Find elements common to both B and C:

- 17 → common  
- 3 → common  
- ⭐ → common  

$$
B \cap C = \{17, 3, ⭐\}
$$

---

## Step 6: Final Union

Take the union of the results from Step 4 and Step 5:

$$
\{0\} \cup \{17, 3, ⭐\}
$$

---

## ✅ Final Result

$$
\boxed{\{0, 17, 3, ⭐\}}
$$

---

## Key Takeaways

- Always simplify **inside parentheses first**
- Relative complements remove elements
- Absolute complements depend on the universe
- Intersections apply an **AND** condition
- Unions combine results
- Even complex expressions can reduce to simple sets when handled step by step


```python
# ---------------------------------------
# Complex Set Operations: Step-by-Step
# ---------------------------------------

# Define the sets
A = {3, 7, -5, 0, 13}
B = {0, 17, 3, "Blue", "⭐"}
C = {17, 3, "⭐"}

# Define the universe (must include all possible elements involved)
U = A.union(B).union(C)

print("Set A:", A)
print("Set B:", B)
print("Set C:", C)
print("Universe U:", U)
print()

# -----------------------------
# Step 1: Relative Complement of C in B (B - C)
# -----------------------------

B_minus_C = B - C
print("B - C:", B_minus_C)
print()

# -----------------------------
# Step 2: Absolute Complement of (B - C)
# -----------------------------

complement_B_minus_C = U - B_minus_C
print("(B - C)^c:", complement_B_minus_C)
print()

# -----------------------------
# Step 3: Intersection with A
# -----------------------------

intersection_with_A = A & complement_B_minus_C
print("A ∩ (B - C)^c:", intersection_with_A)
print()

# -----------------------------
# Step 4: Relative Complement in A
# -----------------------------

result_left = A - intersection_with_A
print("A - [A ∩ (B - C)^c]:", result_left)
print()

# -----------------------------
# Step 5: Intersection of B and C
# -----------------------------

B_intersect_C = B & C
print("B ∩ C:", B_intersect_C)
print()

# -----------------------------
# Step 6: Final Union
# -----------------------------

final_result = result_left | B_intersect_C
print("Final Result:", final_result)
```
