# Finding the mean and standard deviation of a binomial random variable

[Finding the mean and standard deviation of a binomial random variable | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=gro5b3PZ-Bw)

A company produces processing chips for cell phones. At one of its large factories, 2% of the chips produced are defective. A quality check involves randomly selecting and testing 500 chips.                                               

What are the mean and standard deviation of the number of defective processing chips in these samples? 


**📌 Bullet Summary: Mean & Standard Deviation of Defective Chips**

- We define a random variable $X$= number of defective chips in a
  sample of 500.

- Each chip is either:

  - **Defective (1)** with probability $p = 0.02$

  - **Not defective (0)** with probability $1 - p = 0.98$

- This is a **binomial distribution** because:

  - Fixed number of trials: $n = 500$

  - Two outcomes per trial

  - Constant probability of success (defect)

  - Trials are independent

- **Mean of a binomial distribution**:

$$
\mu = np
$$

- **Variance of a binomial distribution**:

$$
\sigma^{2} = np(1 - p)
$$

- **Standard deviation**:

$$
\sigma = \sqrt{np(1 - p)}
$$

**Final Answers**

- **Mean**: $500 \times 0.02 = 10$

- **Standard deviation**:
  $\sqrt{500 \times 0.02 \times 0.98} \approx 3.13$

**📘 Explanation (Step‑by‑Step)**

**1. Identify the distribution**

We’re counting how many defective chips appear in a sample of 500. Each
chip is defective with probability $0.02$. This fits the definition of
a **Binomial(n, p)** random variable.

**2. Compute the mean**

$$
E\lbrack X\rbrack = np = 500 \cdot 0.02 = 10
$$

So on average, we expect **10 defective chips** in each sample of 500.

**3. Compute the standard deviation**

Start with the variance:

$$
Var(X) = np(1 - p) = 500 \cdot 0.02 \cdot 0.98 = 9.8
$$

Then take the square root:

$$
SD(X) = \sqrt{9.8} \approx 3.13
$$

So the number of defective chips typically varies by about **3.13**
around the mean.


# 🧪 Python Code Example

```python
import numpy as np

# Parameters
n = 500      # number of chips sampled
p = 0.02     # probability a chip is defective

# Mean
mean = n * p

# Variance and standard deviation
variance = n * p * (1 - p)
std_dev = np.sqrt(variance)

print("Mean number of defective chips:", mean)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```
