# Bernoulli distribution mean and variance formulas

[Bernoulli distribution mean and variance formulas | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=ry81_iSHt6E)

**📌 Bullet Summary: General Bernoulli Distribution (Parameter** $\mathbf{p}$**)**

**Definition**

- A **Bernoulli random variable** takes only two values:

  - **1** (success) with probability $p$

  - **0** (failure) with probability $1 - p$

- This is the simplest possible probability distribution.

**Mean (Expected Value)**

$$
E\lbrack X\rbrack = p
$$

**Why?**

- Expected value = probability‑weighted sum:

$$
E\lbrack X\rbrack = (1 - p) \cdot 0 + p \cdot 1 = p
$$

- The mean is the **long‑run proportion of successes**, not a value the
  variable can actually take.

**Variance**

$$
Var(X) = p(1 - p)
$$

**Why?**

Start with the definition:

$$
Var(X) = (1 - p)(0 - p)^{2} + p(1 - p)^{2}
$$

Expand:

- First term: $(1 - p)p^{2}$

- Second term: $p(1 - 2p + p^{2})$

Combine and simplify:

$$
Var(X) = p - p^{2} = p(1 - p)
$$

**Standard Deviation**

$$
SD(X) = \sqrt{p(1 - p)}
$$

**Key Takeaways**

- Bernoulli distributions are fully determined by **one number**: $p$.

- Mean = probability of success.

- Variance = success probability × failure probability.

- Standard deviation = square root of variance.

- These formulas match the earlier example where $p = 0.6$.

# 🧪 Python Code Example (General Bernoulli Formulas)

```python
import numpy as np

def bernoulli_stats(p):
    # Mean
    mean = p
    
    # Variance
    variance = p * (1 - p)
    
    # Standard deviation
    std_dev = np.sqrt(variance)
    
    return mean, variance, std_dev

# Example: p = 0.6
p = 0.6
mean, variance, std_dev = bernoulli_stats(p)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```
