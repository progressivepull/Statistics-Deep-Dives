# Expected value of a binomial variable

[Expected value of a binomial variable | Random variables | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=fjjhAm5trSM&t=2s)

**📌 Bullet Summary: Expected Value of a Binomial Random Variable**

- A **binomial random variable** $X$counts the number of successes in
  $n$independent trials.

- Each trial:

  - Has only two outcomes: **success (1)** or **failure (0)**.

  - Has a constant probability of success $p$.

- You can think of each trial as a **Bernoulli random variable**
  $Y_{i}$.

- The binomial variable is the **sum of Bernoulli trials**:

$$
X = Y_{1} + Y_{2} + \cdots + Y_{n}
$$

- Expected value of a sum = sum of expected values:

$$
E\lbrack X\rbrack = E\lbrack Y_{1}\rbrack + E\lbrack Y_{2}\rbrack + \cdots + E\lbrack Y_{n}\rbrack
$$

- Each Bernoulli trial has:

$$
E\lbrack Y_{i}\rbrack = p
$$

- Therefore:

$$
E\lbrack X\rbrack = np
$$

- Interpretation: If you take $n$shots with success probability $p$,
  you expect $np$successes.

**📘 Explanation (Intuitive + Mathematical)**

**1. Define the binomial variable**

- $X$= number of successes in $n$trials.

- Each trial succeeds with probability $p$.

**2. Represent each trial as a Bernoulli variable**

Let:

$$
Y_{i} = \left\{ \begin{matrix}
1 & \text{if trial }i\text{ is a success} \\
0 & \text{if trial }i\text{ is a failure}
\end{matrix} \right.\ 
$$

Then:

$$
X = \sum_{i = 1}^{n}Y_{i}
$$

**3. Use linearity of expectation**

Expectation distributes over sums, even if variables are dependent (but
here they’re independent anyway):

$$
E\lbrack X\rbrack = \sum_{i = 1}^{n}{E\lbrack}Y_{i}\rbrack
$$

**4. Compute the expectation of a Bernoulli variable**

$$
E\lbrack Y_{i}\rbrack = 1 \cdot p + 0 \cdot (1 - p) = p
$$

**5. Combine the results**

$$
E\lbrack X\rbrack = \sum_{i = 1}^{n}{p = np}
$$

**Interpretation**

If you shoot 10 free throws with a 30% success rate:

$$
E\lbrack X\rbrack = 10 \cdot 0.3 = 3
$$

You *expect* to make 3 shots.


# 🧪 Python Code Example

```python
import numpy as np

def binomial_expected_value(n, p):
    # Expected value of a Binomial(n, p)
    return n * p

# Example: 10 free throws, 30% success rate
n = 10
p = 0.3

expected = binomial_expected_value(n, p)
print("Expected number of successes:", expected)

# Optional: simulate to verify
trials = 100000
simulated = np.random.binomial(n, p, size=trials).mean()
print("Simulated average successes:", simulated)
```
