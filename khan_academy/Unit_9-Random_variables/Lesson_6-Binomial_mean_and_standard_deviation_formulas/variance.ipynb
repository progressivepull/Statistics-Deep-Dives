# Variance of a binomial variable

**📌 Bullet Summary: Variance of a Binomial Random Variable**

**Definition**

- A binomial random variable $X$counts the number of successes in:

  - $n$independent trials

  - Each trial has success probability $p$

- Each trial can be represented as a **Bernoulli random variable**
  $Y_{i}$.

**Key Relationships**

- $X = Y_{1} + Y_{2} + \cdots + Y_{n}$

- Each $Y_{i}$takes values:

  - $1$with probability $p$

  - $0$with probability $1 - p$

**Expected Value**

$$
E\lbrack X\rbrack = np
$$

**Variance**

- Variance of a sum of independent variables = sum of variances:

$$
Var(X) = Var(Y_{1}) + \cdots + Var(Y_{n})
$$

- Each Bernoulli variable has variance:

$$
Var(Y_{i}) = p(1 - p)
$$

- Therefore:

$$
Var(X) = np(1 - p)
$$

**Standard Deviation**

$$
SD(X) = \sqrt{np(1 - p)}
$$

**Example**

- $n = 10$free throws

- $p = 0.3$success probability

- Variance:

$$
Var(X) = 10 \cdot 0.3 \cdot 0.7 = 2.1
$$

- Standard deviation:

$$
SD(X) = \sqrt{2.1}
$$

**📘 Explanation (Step‑by‑Step)**

**1. Represent the binomial variable as a sum**

Let:

$$
X = \sum_{i = 1}^{n}Y_{i}
$$

where each $Y_{i}$is a Bernoulli trial.

**2. Use variance additivity**

Because the trials are independent:

$$
Var(X) = \sum_{i = 1}^{n}{Var(}Y_{i})
$$

**3. Compute variance of a Bernoulli variable**

A Bernoulli variable takes:

- $1$with probability $p$

- $0$with probability $1 - p$

Variance formula:

$$
Var(Y) = p(1 - p)
$$

**4. Combine results**

$$
Var(X) = n \cdot p(1 - p)
$$

This is the classic binomial variance formula.


# 🧪 Python Code Example

```python
import numpy as np

def binomial_variance(n, p):
    # Variance of a Binomial(n, p)
    return n * p * (1 - p)

def binomial_std(n, p):
    return np.sqrt(binomial_variance(n, p))

# Example: 10 free throws, 30% success rate
n = 10
p = 0.3

variance = binomial_variance(n, p)
std_dev = binomial_std(n, p)

print("Variance:", variance)
print("Standard Deviation:", std_dev)

# Optional: simulate to verify
trials = 100000
simulated = np.random.binomial(n, p, size=trials).var()
print("Simulated Variance:", simulated)
```
