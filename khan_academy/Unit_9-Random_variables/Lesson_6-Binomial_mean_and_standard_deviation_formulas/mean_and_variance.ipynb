# Mean and variance of Bernoulli distribution example

[Mean and variance of Bernoulli distribution example | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=O8vB1eInP_8)

**📌 Bullet Summary: Bernoulli Distribution Example (Favorable vs. Unfavorable Rating)**

**Setup**

- You survey *every member* of a population (a census).

- Each person answers either:

  - **Unfavorable (U)** → assign value $0$

  - **Favorable (F)** → assign value $1$

- Population proportions:

  - $P(U) = 0.4$

  - $P(F) = 0.6$

**Probability Distribution**

| **Outcome** | **Assigned Value** | **Probability** |
|-------------|--------------------|-----------------|
| Unfavorable | 	0              |      0.4        | 
| Favorable   | 	1              |      0.6        | 


This is a **Bernoulli distribution** with parameter $p = 0.6$.

**Expected Value (Mean)**

The expected value of a discrete distribution is:

$$
E\lbrack X\rbrack = \sum p_{i}x_{i}
$$

Here:

$$
E\lbrack X\rbrack = 0.4(0) + 0.6(1) = 0.6
$$

**Interpretation**

- The mean $0.6$is **not an actual possible response** (nobody is “0.6
  favorable”).

- It represents the **long‑run proportion** of favorable responses.

**Variance**

Variance of a discrete distribution:

$$
Var(X) = \sum p_{i}(x_{i} - \mu)^{2}
$$

Compute each term:

- For $x = 0$:

$$
0.4(0 - 0.6)^{2} = 0.4(0.36) = 0.144
$$

- For $x = 1$:

$$
0.6(1 - 0.6)^{2} = 0.6(0.16) = 0.096
$$

Add them:

$$
Var(X) = 0.144 + 0.096 = 0.24
$$

**Standard Deviation**

$$
SD = \sqrt{0.24} \approx 0.49
$$

**Key Takeaways**

- This is a **Bernoulli distribution** with $p = 0.6$.

- **Mean** = $p = 0.6$

- **Variance** = $p(1 - p) = 0.24$

- **Standard deviation** = $\sqrt{p(1 - p)} \approx 0.49$

- The mean is a *long‑run proportion*, not a possible individual
  outcome.

- This example is the simplest case of the **binomial distribution**.


### 🧪 Python Code Example

```python
import numpy as np

# Probabilities
p_unfavorable = 0.4
p_favorable = 0.6

# Assigned values
values = np.array([0, 1])
probs = np.array([p_unfavorable, p_favorable])

# Expected value
mean = np.sum(values * probs)

# Variance
variance = np.sum(probs * (values - mean)**2)

# Standard deviation
std_dev = np.sqrt(variance)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```
