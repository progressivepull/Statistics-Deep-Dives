# Law of large numbers

[Law of large numbers | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=VpuN8vCQ--M&t=8s)

Here’s a clean, structured breakdown of the **Law of Large Numbers
(LLN)** — **bullet summary**, **intuitive explanation**, and a **Python
code example** that *demonstrates* the law in action.

**🔹 Bullet Summary**

- The **Law of Large Numbers (LLN)** says that the **sample mean** of
  repeated observations of a random variable will **approach the true
  expected value** as the number of observations $n$becomes large.

- Let:

  - $X$= a random variable

  - $\mu = E\lbrack X\rbrack$= population mean

  - ${\overset{ˉ}{X}}_{n} = \frac{1}{n}\sum_{i = 1}^{n}X_{i}$= sample
    mean of $n$observations

- LLN states:

$$
{\overset{ˉ}{X}}_{n} \rightarrow \mu\text{as }n \rightarrow \infty
$$

- Example:

  - $X$= number of heads in 100 fair coin flips

  - Expected value: $E\lbrack X\rbrack = 100 \cdot 0.5 = 50$

  - If you repeat this experiment many times and average the results,
    the average will approach **50**.

- LLN **does NOT** mean outcomes “correct themselves.”

  - Getting many heads early does **not** make tails more likely later.

  - That misconception is the **gambler’s fallacy**.

- Instead, LLN says:

  - Even if early results are extreme, the **infinite future trials**
    dominate the average and pull it toward the true mean.

**🔹 Intuitive Explanation**

**✔ What LLN *really* says**

If you repeat an experiment many times, the **average** of your results
will get closer and closer to the true expected value.

**✔ What LLN does *not* say**

It does **not** say that if you flip 10 heads in a row, tails are “due.”
The probability of heads is always **50%**, no matter what happened
before.

**✔ Why the average stabilizes**

Even if your first few trials are extreme (e.g., 70 heads out of 100),
those early results become **less important** as you collect more data.

Example:

- First 100‑flip experiment: 70 heads

- Next 10,000 experiments: average 50 heads

The overall average will be pulled toward **50**, because the later
trials vastly outnumber the early ones.

This is the heart of LLN.


# 🔹 Python Code Example

A simulation showing the sample mean converging to the true mean (0.5 for a fair coin).
                                                                  

```python
import numpy as np
import matplotlib.pyplot as plt

# Number of coin flips
n = 100000

# Simulate fair coin flips (1 = heads, 0 = tails)
flips = np.random.randint(0, 2, size=n)

# Compute running sample mean
running_mean = np.cumsum(flips) / np.arange(1, n+1)

# Plot
plt.figure(figsize=(10,5))
plt.plot(running_mean, label="Sample Mean")
plt.axhline(0.5, color='red', linestyle='--', label="True Mean (0.5)")
plt.xlabel("Number of Flips")
plt.ylabel("Sample Mean")
plt.title("Law of Large Numbers Demonstration")
plt.legend()
plt.show()
```

```python
# What you’ll see

* The blue line (sample mean) wiggles early on.
* As the number of flips increases, the line stabilizes and approaches 0.5.
* This visually demonstrates the Law of Large Numbers.

--- 

If you want, I can also show:

* A simulation of the coin‑box example from your description
* A comparison of weak vs. strong law of large numbers
* How LLN is used in casinos, insurance, and polling

Just tell me where you want to go next.
```
