# Term life insurance and death probability

[Term life insurance and death probability | Finance & Capital Markets | Khan Academy | YouTube](https://www.youtube.com/watch?v=NSSoMafbBqQ&t=1s)

Here’s a clean, structured breakdown of the idea behind the insurance
company’s pricing — plus an intuitive explanation and a small Python
simulation to make the math feel concrete.

**🔹 Bullet Summary**

- You’re considering a **20‑year term life insurance policy** with:

  - Annual premium: **\$500**

  - Coverage amount: **\$1,000,000**

- Over 20 years, you would pay:

$$
500 \cdot 20 = 10,000
$$

- From the insurer’s perspective, they collect **\$10,000** to insure
  **\$1,000,000** of risk.

- That means they collect **\$1 in premium for every \$100 of insurance
  payout**.

- If they insured **100 people just like you**, they would collect:

  - Premiums: **100 × \$10,000 = \$1,000,000**

  - They must pay **\$1,000,000** if **one person dies**.

- So to **break even**, at most **1 out of 100** people can die during
  the 20‑year term.

- Therefore, the **maximum probability of death** over 20 years that
  still lets them break even is:

$$
\frac{1}{100} = 0.01 = 1\%
$$

- In reality, insurers want profit, so they likely believe your true
  probability is **lower** (maybe 1 in 200 or 1 in 300).

- This back‑of‑the‑envelope calculation gives a rough estimate of how
  insurers think about risk.

**🔹 Explanation (Intuitive)**

Insurance companies pool risk. They don’t care which specific person
dies — they care about the **average** across many similar people.

Your premium structure implies:

- They get **\$10,000** from you.

- They might have to pay **\$1,000,000**.

If they insured **100 people**, they’d collect **\$1,000,000** total. If
**one** of those 100 dies, they pay **\$1,000,000** and break even.

So the insurer is implicitly assuming:

- **Probability of death ≤ 1% over 20 years**.

If the true probability were higher, they’d expect to lose money.

This is why term life insurance is cheap for young, healthy people — the
probability of death over 20 years is extremely low.


# 🔹 Python Code Example
A small simulation to illustrate the break‑even logic.

```python
import numpy as np

# Parameters
num_people = 100_000        # simulate 100k people like you
premium_per_year = 500
years = 20
coverage = 1_000_000

# Implied break-even probability
p = 1/100  # 1% chance of death over 20 years

# Simulate who dies (1 = death, 0 = survives)
deaths = np.random.binomial(1, p, size=num_people)

# Total premiums collected
total_premiums = num_people * premium_per_year * years

# Total payouts
total_payouts = deaths.sum() * coverage

print("Total premiums collected:", total_premiums)
print("Total payouts:", total_payouts)
print("Profit (positive = insurer wins):", total_premiums - total_payouts)
```

# What this code shows

* If the true probability is **1%**, the insurer roughly **breaks even**.
* If the true probability is **less than 1%**, the insurer **makes money**.
* If the true probability is **greater than 1%**, the insurer **loses money**.

This matches the back‑of‑the‑envelope reasoning.
