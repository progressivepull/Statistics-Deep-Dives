# Expected profit from lottery ticket

[Expected profit from lottery ticket | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=6vlBOHckmzU)

Ahmed is playing a lottery game where he must pick 2 numbers from 0 to 9 and then one letter out of the 26-letter English alphabet. (He may choose the same number both times.)

If his ticket matches the 2 numbers and 1 letter drawn in order, he wins the grand prize and receives 10,405 dollar. If just his letter matches but one or both of his numbers do not match, he wins the small prize of 100 dollar. Under any other outcome, he loses and receives nothing. The game costs him $5 to play.

He has chosen the ticket 04R.

Here’s a clean, structured breakdown of the lottery problem — **bullet
summary**, **clear explanation**, and a **Python code example** that
computes the expected net profit exactly the way the reasoning
describes.

**🔹 Bullet Summary**

- Ahmed plays a lottery game:

  - Picks **two digits** (0–9) and **one letter** (A–Z).

  - His ticket is **04R**.

- Outcomes:

  - **Grand prize**: all three match in order → wins **\$10,405**.

  - **Small prize**: only the **letter** matches → wins **\$100**.

  - **Otherwise**: wins **\$0**.

- Cost to play: **\$5**.

- Define random variable $X$= **net profit** from playing.

- Net profits:

  - Grand prize: $10,405 - 5 = 10,400$

  - Small prize: $100 - 5 = 95$

  - Lose: $- 5$

- Probabilities:

  - Grand prize:

$$
\frac{1}{10} \cdot \frac{1}{10} \cdot \frac{1}{26} = \frac{1}{2600}
$$

- Letter match (small prize):

$$
\frac{1}{26} - \frac{1}{2600}
$$

- Lose:

$$
1 - \frac{1}{26}
$$

- Expected net profit:

$$
E\lbrack X\rbrack \approx 2.81
$$

- **Surprisingly, this game has a positive expected value for the
  player.**

**🔹 Explanation (Step‑by‑Step)**

**1. Compute probabilities**

- **Grand prize**: must match digit–digit–letter exactly

$$
P(\text{grand}) = \frac{1}{10} \cdot \frac{1}{10} \cdot \frac{1}{26} = \frac{1}{2600}
$$

- **Small prize**: letter matches, but at least one digit is wrong

$$
P(\text{small}) = \frac{1}{26} - \frac{1}{2600}
$$

(subtract the grand‑prize case)

- **Lose**: letter does not match

$$
P(\text{lose}) = 1 - \frac{1}{26} = \frac{25}{26}
$$

**2. Net profits**

- Grand prize:

$$
10,405 - 5 = 10,400
$$

- Small prize:

$$
100 - 5 = 95
$$

- Lose:

$$
- 5
$$

**3. Expected value**

$$
E\lbrack X\rbrack = P(\text{grand}) \cdot 10,400 + P(\text{small}) \cdot 95 + P(\text{lose}) \cdot ( - 5)
$$

Plugging in the numbers gives:

$$
E\lbrack X\rbrack \approx 2.81
$$

So on average, Ahmed **expects to gain \$2.81 per play**.

This is extremely unusual — most lottery games have a **negative**
expected value for the player.


# 🔹 Python Code Example

Here’s a small script that computes the expected net profit exactly:


```python
# Probabilities
p_grand = 1 / 2600
p_small = (1 / 26) - p_grand
p_lose = 1 - (1 / 26)

# Net profits
net_grand = 10400
net_small = 95
net_lose = -5

# Expected value
expected_value = (
    p_grand * net_grand +
    p_small * net_small +
    p_lose * net_lose
)

expected_value
```
