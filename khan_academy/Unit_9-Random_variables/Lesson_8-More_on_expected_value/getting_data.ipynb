# Getting data from expected value

[Getting data from expected value | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=gHpl5UcgRkI&t=6s)

Jamie’s dad gave her a die for her birthday. She wanted to make sure it was fair so she took her die to school and rolled it 500 times and kept track of how many times the die rolled each number. Afterwards she calculated the expected value of the sum of 20 rolls to be 67.4.

On her way home from school it was raining and 2 values were washed away from her data table.

Find the 2 missing absolute frequencies from Jamie’s data table.




| Die value	|Absolute frequency|
| --------- | ---------------- |
|   1	    |      missing     |
|   2	    |       110        |
|   3	    |       95         |
|   4	    |       70         | 
|   5	    |       75         |
|   6	    |      missing     |
|  Total	|       500        |




Here’s a clean, structured breakdown of the entire problem — **bullet
summary**, **clear explanation**, and a **Python code example** that
solves for the missing frequencies.

**🔹 Bullet Summary**

- Jamie rolled a die **500 times** and recorded frequencies for outcomes
  2–5.

- Frequencies for **1** and **6** were washed away.

- She computed the **expected value of the sum of 20 rolls = 67.4**.

- Therefore, the **expected value of one roll** is:

$$
\frac{67.4}{20} = 3.37
$$

- Expected value formula using frequencies:

$$
E\lbrack X\rbrack = \sum_{i = 1}^{6}{i \cdot}\frac{f_{i}}{500}
$$

- Let:

  - $A = f_{1}$

  - $B = f_{6}$

- Two equations come from:

  1.  Expected value equation

$$
A + 6B = 525
$$

2.  Total frequency equation

$$
A + B = 150
$$

- Solving the system gives:

  - $A = 75$

  - $B = 75$

- **Final answer:**

  - **1 was rolled 75 times**

  - **6 was rolled 75 times**

**🔹 Step‑by‑Step Explanation**

**1. Expected value of one roll**

Jamie computed:

$$
E\lbrack\text{sum of 20 rolls}\rbrack = 67.4
$$

So:

$$
E\lbrack\text{one roll}\rbrack = \frac{67.4}{20} = 3.37
$$

**2. Build the expected value equation**

Using frequencies:

$$
E\lbrack X\rbrack = \frac{1 \cdot A + 2 \cdot 110 + 3 \cdot 95 + 4 \cdot 70 + 5 \cdot 75 + 6 \cdot B}{500}
$$

Multiply both sides by 500:

$$
A + 220 + 285 + 280 + 375 + 6B = 3.37 \cdot 500 = 1685
$$

Combine constants:

$$
A + 6B + 1160 = 1685
$$

Subtract 1160:

$$
A + 6B = 525
$$

**3. Use the total frequency**

All frequencies must sum to 500:

$$
A + 110 + 95 + 70 + 75 + B = 500
$$

Combine constants:

$$
A + B = 150
$$

**4. Solve the system**

$$
\left\{ \begin{array}{r}
A + 6B = 525 \\
A + B = 150
\end{array} \right.\ 
$$

Subtract the second from the first:

$$
5B = 375 \Rightarrow B = 75
$$

Then:

$$
A + 75 = 150 \Rightarrow A = 75
$$


# 🔹 Python Code Example

Here’s a small script that sets up the equations and solves them:

```python
import sympy as sp

# Unknown frequencies
A, B = sp.symbols('A B')

# Given frequencies
f2, f3, f4, f5 = 110, 95, 70, 75

# Expected value of one roll
E = 3.37

# Equation 1: expected value
eq1 = A*1 + f2*2 + f3*3 + f4*4 + f5*5 + B*6 - 500*E

# Equation 2: total frequencies
eq2 = A + f2 + f3 + f4 + f5 + B - 500

solution = sp.solve((eq1, eq2), (A, B))
solution
```
