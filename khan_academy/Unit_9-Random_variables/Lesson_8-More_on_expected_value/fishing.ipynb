# Expected value while fishing

[Expected value while fishing | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=86nb02Bx_5w&t=12s)

You and your friend, Jeremy, are fishing in a pond that contains 10 trout and 10 sunfish. Each time one of you catches a fish, you release it back into the water. Jeremy offers you the choice of 2 different bets.                

Bet 1: If the next 3 fish he catches are all sunfish, you will pay him 100 dollar, otherwise he will pay you 20 dollar.                

Bet 2: If you catch at least 2 sunfish of the next 3 fish that you catch, he will pay you 50 dollar, otherwise you will pay him 25 dollar .           

What is your expected value from bet 1? Round your answer to the nearest cent.                          
$ [ ]            

What is your expected value from bet 2? Round your answer to the nearest cent.                  
$ [ ]              

Your friend says he is willing to take both bets a combined total of 50 times. If you want to maximize your expected value, what should you do?       

( ) Take bet 1 all 50 times.                                   
( ) Take bet 2 all 50 times.

Here’s a clean, structured breakdown of the fishing‑bets problem —
**bullet summary**, **clear explanation**, and a **Python code example**
that computes both expected values exactly.

**🎣 Bullet Summary**

**Setup**

- Pond has **10 trout** and **10 sunfish** → probability of sunfish on
  any catch = $\frac{10}{20} = \frac{1}{2}$.

- Fish are released → trials are **independent**.

**Bet 1**

- Jeremy catches 3 fish.

- If **all 3 are sunfish**, **you lose \$100**.

- Otherwise, **you win \$20**.

- Probability of 3 sunfish in a row:

$$
\left( \frac{1}{2} \right)^{3} = \frac{1}{8}
$$

- Expected value:

$$
E_{1} = \frac{1}{8}( - 100) + \frac{7}{8}(20) = 5
$$

- **Expected profit from Bet 1 = \$5.00**

**Bet 2**

- YOU catch 3 fish.

- If you catch **at least 2 sunfish**, you **win \$50**.

- Otherwise, you **lose \$25**.

- Total possible outcomes of 3 catches: $2^{3} = 8$.

- Outcomes with ≥2 sunfish: 4 of them.

$$
P(\text{≥2 sunfish}) = \frac{4}{8} = \frac{1}{2}
$$

- Expected value:

$$
E_{2} = \frac{1}{2}(50) + \frac{1}{2}( - 25) = 12.50
$$

- **Expected profit from Bet 2 = \$12.50**

**Which bet should you take 50 times?**

- Bet 1 EV = **\$5**

- Bet 2 EV = **\$12.50**

- **Bet 2 is better. Take Bet 2 all 50 times.**

**📘 Explanation (Step‑by‑Step)**

**Bet 1: Jeremy catches 3 fish**

Probability all 3 are sunfish:

$$
\left( \frac{1}{2} \right)^{3} = \frac{1}{8}
$$

Expected value:

$$
E_{1} = \frac{1}{8}( - 100) + \frac{7}{8}(20)
$$

$$
E_{1} = - 12.5 + 17.5 = 5
$$

So on average, you **gain \$5** per play.

**Bet 2: You catch 3 fish**

List all 8 equally likely sequences (S = sunfish, T = trout):

1.  S S S

2.  S S T

3.  S T S

4.  T S S

5.  S T T

6.  T S T

7.  T T S

8.  T T T

Sequences with ≥2 sunfish: **1, 2, 3, 4 → 4 outcomes**

Probability:

$$
\frac{4}{8} = \frac{1}{2}
$$

Expected value:

$$
E_{2} = \frac{1}{2}(50) + \frac{1}{2}( - 25)
$$

$$
E_{2} = 25 - 12.5 = 12.5
$$

So on average, you **gain \$12.50** per play.


# 🧪 Python Code Example

```python
# Probabilities
p_sun = 0.5

# --- Bet 1 ---
p_three_sun = p_sun**3
p_not_three_sun = 1 - p_three_sun

ev_bet1 = p_three_sun * (-100) + p_not_three_sun * 20

# --- Bet 2 ---
p_at_least_two = 4/8  # from enumeration
p_less_than_two = 1 - p_at_least_two

ev_bet2 = p_at_least_two * 50 + p_less_than_two * (-25)

ev_bet1, ev_bet2
```

# ⭐ Final Answers

* Expected value of Bet 1: 5.00 dollar
* Expected value of Bet 2: 12.50 dollar
* Best strategy over 50 rounds: Take Bet 2 every time.
