# Comparing insurance with expected value

Paul has the option between a high deductible or a low deductible health insurance plan.                 

If Paul chooses the low deductible plan, he will have to pay the first 1000 dollar of any medical costs. The low deductible plan costs 8000  dollar  for a year.                

If Paul chooses the high deductible plan, he will have to pay the first 2500  dollar  of any medical costs. The high deductible plan costs 7500  dollar for a year.                

To help himself choose a plan, Paul found some statistics about common health problems for people similar to him. Assume that the table below correctly shows the probabilities and costs of total medical incidents within the next year.                

| Medical costs    | Probability  | 
| ---------------- | ------------ |
| 0 dollar         | 30 percent   |
| 1000  dollar     | 25 percent   | 
| 4000  dollar     | 20 percent   | 
| 7000  dollar     | 20 percent   | 
| 15,000  dollar   | 5 percent    | 







1. Including the cost of insurance, what are Paul's expected total medical costs with the low deductible plan? Round your answer to the nearest cent.

   _______ dollar                

3. Including the cost of insurance, what are Paul's expected total medical costs with the high deductible plan? Round your answer to the nearest cent.

    _______ dollar                

5. If Paul wants the least payoff in the long run and must buy one of the two insurance plans, he should purchase the _______ plan.                





Here’s a clean, structured breakdown of Paul’s insurance‑choice problem
— **bullet summary**, **clear explanation**, and a **Python code
example** that computes both expected costs exactly.

**🔹 Bullet Summary**

**Insurance Options**

- **Low‑deductible plan**

  - Annual premium: **\$8,000**

  - Deductible: **\$1,000**

  - Paul pays **only the first \$1,000** of any medical bill.

- **High‑deductible plan**

  - Annual premium: **\$7,500**

  - Deductible: **\$2,500**

  - Paul pays **only the first \$2,500** of any medical bill.

**Medical Cost Distribution**

| **Medical Cost** | **Probability** |
|------------------|-----------------|
| \$0              | 30%             |
| \$1,000          | 25%             |
| \$4,000          | 20%             |
| \$7,000          | 20%             |
| \$15,000         | 5%              |

**Out‑of‑Pocket Rules**

- Under **low deductible**, Paul pays:

  - \$0 → \$0

  - \$1,000 → \$1,000

  - \$4,000 → \$1,000

  - \$7,000 → \$1,000

  - \$15,000 → \$1,000

- Under **high deductible**, Paul pays:

  - \$0 → \$0

  - \$1,000 → \$1,000

  - \$4,000 → \$2,500

  - \$7,000 → \$2,500

  - \$15,000 → \$2,500

**Expected Total Cost**

- **Low‑deductible expected cost:**

$$
8000 + 0.25(1000) + 0.20(1000) + 0.20(1000) + 0.05(1000) = 8700
$$

- **High‑deductible expected cost:**

$$
7500 + 0.25(1000) + 0.20(2500) + 0.20(2500) + 0.05(2500) = 8875
$$

**Conclusion**

- **Low‑deductible plan has the lower expected total cost.**

- Paul should choose the **low‑deductible plan** based on expected
  value.

**🔹 Explanation (Step‑by‑Step)**

**1. Add the guaranteed insurance premium**

- Low deductible: always pays **\$8,000**

- High deductible: always pays **\$7,500**

**2. Compute expected out‑of‑pocket medical spending**

Use:

$$
E\lbrack\text{OOP}\rbrack = \sum(\text{probability}) \times (\text{amount Paul pays})
$$

**Low‑deductible OOP**

- 30% → \$0

- 25% → \$1,000

- 20% → \$1,000

- 20% → \$1,000

- 5% → \$1,000

$$
E\lbrack\text{OOP}\rbrack = 0.25(1000) + 0.20(1000) + 0.20(1000) + 0.05(1000) = 700
$$

Total:

$$
8000 + 700 = 8700
$$

**High‑deductible OOP**

- 30% → \$0

- 25% → \$1,000

- 20% → \$2,500

- 20% → \$2,500

- 5% → \$2,500

$$
E\lbrack\text{OOP}\rbrack = 0.25(1000) + 0.20(2500) + 0.20(2500) + 0.05(2500) = 1375
$$

Total:

$$
7500 + 1375 = 8875
$$

**Final Decision**

- Low deductible: **\$8,700 expected cost**

- High deductible: **\$8,875 expected cost**

**→ Low‑deductible plan is better in expectation.**


# 🔹 Python Code Example

```python
# Probabilities
p0 = 0.30
p1k = 0.25
p4k = 0.20
p7k = 0.20
p15k = 0.05

# Deductibles
low_deductible = 1000
high_deductible = 2500

# Premiums
low_premium = 8000
high_premium = 7500

# Out-of-pocket under each plan
def oop(cost, deductible):
    return min(cost, deductible)

# Expected cost for each plan
low_expected = (
    low_premium +
    p0 * oop(0, low_deductible) +
    p1k * oop(1000, low_deductible) +
    p4k * oop(4000, low_deductible) +
    p7k * oop(7000, low_deductible) +
    p15k * oop(15000, low_deductible)
)

high_expected = (
    high_premium +
    p0 * oop(0, high_deductible) +
    p1k * oop(1000, high_deductible) +
    p4k * oop(4000, high_deductible) +
    p7k * oop(7000, high_deductible) +
    p15k * oop(15000, high_deductible)
)

low_expected, high_expected
```
