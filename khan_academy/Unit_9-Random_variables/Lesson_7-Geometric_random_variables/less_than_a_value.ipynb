# Cumulative geometric probability (less than a value)

[Cumulative geometric probability (less than a value) | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=nxUP4lJ_hbU&t=10s)

Lilyana runs a cake decorating business, for which 10% of her orders come over the telephone. Let C be the number of cake orders Lilyana receives in a month until she first gets an order over the telephone. Assume the method of placing each cake order is independent.

Find the probability that it takes fewer than 5 orders for Lilyana to get her first telephone order of the month.

**📌 Bullet summary**

- **Setup:** 10% of Lilyana’s cake orders come over the telephone →
  $p = 0.10$.

- Let $C$= number of cake orders until the **first telephone order**.

- Each order is independent, with constant probability of “telephone” →
  **geometric random variable**.

- We want:

$$
P(C < 5) = P(C = 1) + P(C = 2) + P(C = 3) + P(C = 4)
$$

- Easier approach:

$$
P(C < 5) = 1 - P(\text{no telephone order in first 4})
$$

- Probability an order is **not** by telephone: $1 - p = 0.9$.

- So:

$$
P(\text{no telephone order in first 4}) = {0.9}^{4}
$$

$$
P(C < 5) = 1 - {0.9}^{4} \approx 0.3439
$$

- **Final answer:** $P(C < 5) \approx 0.34$(about **34.39%**).

**📘 Explanation**

We’re told 10% of Lilyana’s orders come over the telephone. We define:

- $C$= number of orders until the **first telephone order**.

This is a **geometric random variable** because:

- Each order is a trial.

- Success = “telephone order”.

- Probability of success is constant: $p = 0.10$.

- Trials are independent.

- We’re counting **how many trials until the first success**.

We want the probability that it takes **fewer than 5 orders** to get the
first telephone order:

$$
P(C < 5) = P(C = 1) + P(C = 2) + P(C = 3) + P(C = 4)
$$

You *could* compute each term directly, but there’s a cleaner way:

- “Fewer than 5 orders” means: **at least one telephone order in the
  first 4 orders**.

- The complement event is: **no telephone orders in the first 4
  orders**.

So:

$$
P(C < 5) = 1 - P(\text{no telephone orders in first 4})
$$

The probability a single order is **not** by telephone is $0.9$. For
the first 4 orders all to be non-telephone:

$$
P(\text{no telephone orders in first 4}) = {0.9}^{4}
$$

Thus:

$$
P(C < 5) = 1 - {0.9}^{4} \approx 1 - 0.6561 = 0.3439
$$

So there’s about a **34.39% chance** that Lilyana gets her first
telephone order within the first 4 orders.


# 🧪 Python code example

```python
import numpy as np

p = 0.10          # Probability an order is by telephone
q = 1 - p         # Probability an order is NOT by telephone

# Theoretical probability: C < 5
prob_theoretical = 1 - q**4
print("Theoretical P(C < 5):", prob_theoretical)

# Optional: simulate to verify
def simulate_C(trials=100000):
    results = []
    for _ in range(trials):
        count = 0
        while True:
            count += 1
            # Telephone order with probability p
            if np.random.rand() < p:
                break
        results.append(count)
    return np.array(results)

samples = simulate_C()
prob_empirical = np.mean(samples < 5)
print("Empirical P(C < 5):", prob_empirical)
```
