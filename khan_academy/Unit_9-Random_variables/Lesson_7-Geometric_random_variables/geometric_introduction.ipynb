**📌 Bullet Summary: Binomial vs. Geometric Random Variables**

**Random Variable X: “Number of sixes after 12 rolls”**

- Each roll = a trial.

- Success = rolling a 6.

- Probability of success = $p = \frac{1}{6}$.

- Fixed number of trials = 12.

- Trials are independent.

- Probability of success stays constant.

- **Meets all four binomial conditions.**

- ✔️ **X is a Binomial random variable**:

$$
X \sim \text{Binomial}(n = 12,p = \frac{1}{6})
$$

**Random Variable Y: “Number of rolls until we get a six”**

- Each roll = a trial.

- Success = rolling a 6.

- Probability of success = $p = \frac{1}{6}$.

- Trials are independent.

- Probability stays constant.

- ❌ **No fixed number of trials** — could take 1 roll, 10 rolls, 100
  rolls, or theoretically forever.

- Minimum value of $Y$= 1.

- Maximum value of $Y$= ∞ (unbounded).

- ✔️ **Y is a Geometric random variable**:

$$
Y \sim \text{Geometric}(p = \frac{1}{6})
$$

**📘 Explanation (Conceptual)**

**Binomial Random Variable (X)**

A binomial variable counts **how many successes occur in a fixed number
of trials**.

For X:

- 12 rolls = fixed number of trials.

- Each roll is independent.

- Probability of rolling a 6 is constant.

- Success/failure is clearly defined.

So X fits the binomial pattern:

$$
\text{“How many successes in }n\text{ trials?”}
$$

**Geometric Random Variable (Y)**

A geometric variable counts **how many trials it takes until the first
success**.

For Y:

- We keep rolling until we get a 6.

- No fixed number of trials.

- Could theoretically take infinitely many rolls.

- Still independent with constant probability.

So Y fits the geometric pattern:

$$
\text{“How many trials until the first success?”}
$$


# 🧪 Python Code Example
Below is a small simulation to illustrate the difference between binomial and geometric random variables.

```python
import numpy as np

# Parameters
p = 1/6       # probability of rolling a 6
n = 12        # number of rolls for the binomial variable

# Simulate X: number of sixes in 12 rolls
def simulate_binomial(n, p, trials=100000):
    return np.random.binomial(n, p, size=trials)

# Simulate Y: number of rolls until first six
def simulate_geometric(p, trials=100000):
    return np.random.geometric(p, size=trials)

# Run simulations
X_samples = simulate_binomial(n, p)
Y_samples = simulate_geometric(p)

print("Binomial X: mean =", X_samples.mean(), "std =", X_samples.std())
print("Geometric Y: mean =", Y_samples.mean(), "std =", Y_samples.std())
```
