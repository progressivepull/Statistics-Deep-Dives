I keep picking cards from a standard deck until I get a king (I replace the cards if they are not a king).      

What is the probability that I need to pick 5 cards?                

What is the probability that I need to pick less than 10 cards?                     

What is the probability that I need to pick more than 12 cards?                           

Here’s a clean, structured breakdown of the three geometric‑distribution
questions—**bullet summary**, **explanation**, and a **Python code
example** that mirrors what the TI‑84 does with geometpdf and geometcdf.

**📌 Bullet Summary: Picking Cards Until a King Appears**

- You draw cards **with replacement** until you get a **king**.

- Probability of success (drawing a king):

$$
p = \frac{4}{52} = \frac{1}{13}
$$

- Let $X$= number of draws until the first king → **geometric random
  variable**.

- Geometric formulas:

  - **PMF:**

$$
P(X = k) = (1 - p)^{k - 1}p
$$

- **CDF:**

$$
P(X \leq k) = 1 - (1 - p)^{k}
$$

- **Complement:**

$$
P(X > k) = (1 - p)^{k}
$$

**Final Answers**

1.  **Probability you need 5 cards:**

$$
P(X = 5) = \left( \frac{12}{13} \right)^{4} \cdot \frac{1}{13} \approx 0.056
$$

2.  **Probability you need fewer than 10 cards:**

$$
P(X < 10) = P(X \leq 9) = 1 - \left( \frac{12}{13} \right)^{9} \approx 0.513
$$

3.  **Probability you need more than 12 cards:**

$$
P(X > 12) = \left( \frac{12}{13} \right)^{12} \approx 0.383
$$

**📘 Explanation (Step‑by‑Step)**

**1. Identify the distribution**

- You draw until the first king.

- Each draw is independent because you replace the card.

- Probability of success is constant: $p = 1/13$.

- This is exactly the definition of a **geometric random variable**.

**2. Probability you need exactly 5 cards**

You must:

- Miss the first 4 draws (draw non‑kings),

- Then draw a king on the 5th.

$$
P(X = 5) = (1 - p)^{4}p = \left( \frac{12}{13} \right)^{4} \cdot \frac{1}{13}
$$

**3. Probability you need fewer than 10 cards**

“Fewer than 10” means $X \leq 9$.

Use the geometric CDF:

$$
P(X \leq 9) = 1 - (1 - p)^{9}
$$

**4. Probability you need more than 12 cards**

“More than 12” means you fail 12 times in a row:

$$
P(X > 12) = (1 - p)^{12}
$$


# 🧪 Python Code Example (Geometric PMF & CDF)

```python
import numpy as np

p = 1/13
q = 1 - p

# 1. Probability X = 5
prob_5 = q**4 * p

# 2. Probability X < 10  (i.e., X <= 9)
prob_less_10 = 1 - q**9

# 3. Probability X > 12
prob_more_12 = q**12

print("P(X = 5):", prob_5)
print("P(X < 10):", prob_less_10)
print("P(X > 12):", prob_more_12)
```
