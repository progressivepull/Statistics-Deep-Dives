# Probability for a geometric random variable

[Probability for a geometric random variable | Random variables | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=jKlgEvRFItE&t=17s)

Jeremiah makes 25% of the three-point shots he attempts. For a warm-up, Jeremiah shoots until he successfully makes one. Let M represent the number of shots it takes for Jeremiah to successfully make his first three-point shot. Assume each shot is independent.

Find the probability that Jeremiah’s first successful shot occurs on his 3rd attempt.
You may round your answer to the nearest hundredth.

**📌 Bullet Summary: Probability First Success Occurs on 3rd Shot**

- Jeremiah makes **25%** of his three‑point shots → $p = 0.25$.

- He keeps shooting **until he makes his first shot**.

- Let $M$= number of shots until first success.

- This is a **geometric random variable** because:

  - Each trial has two outcomes (make/miss).

  - Probability of success is constant.

  - Trials are independent.

  - We are counting **how many trials until the first success**.

- Probability formula for geometric distribution:

$$
P(M = k) = (1 - p)^{k - 1}p
$$

- For $k = 3$:

$$
P(M = 3) = (0.75)^{2} \cdot 0.25 = \frac{9}{64} \approx 0.14
$$

**Final Answer**

- **Probability first make is on 3rd shot:**

0.14 or 14%

**📘 Explanation (Step‑by‑Step)**

**1. Identify the distribution**

Jeremiah shoots until he makes one. This matches the definition of a
**geometric random variable**.

**2. Interpret the event** $\mathbf{M = 3}$

For the first make to occur on the 3rd attempt:

- He must **miss** shot 1.

- He must **miss** shot 2.

- He must **make** shot 3.

**3. Compute the probability**

- Probability of a miss: $1 - p = 0.75$

- Probability of a make: $p = 0.25$

So:

$$
P(M = 3) = 0.75 \times 0.75 \times 0.25 = \frac{9}{64} \approx 0.14
$$

This means there’s about a **14% chance** his first make happens on the
third shot.


# 🧪 Python Code Example

```python
import numpy as np

p = 0.25  # probability of making a shot

# Geometric probability: first success on k-th trial
def geometric_pmf(k, p):
    return (1 - p)**(k - 1) * p

# Probability first make is on 3rd shot
prob = geometric_pmf(3, p)
print("P(M = 3):", prob)

# Optional: simulate to verify
trials = 100000
simulated = np.random.geometric(p, size=trials)
estimate = np.mean(simulated == 3)

print("Simulated probability:", estimate)
```
