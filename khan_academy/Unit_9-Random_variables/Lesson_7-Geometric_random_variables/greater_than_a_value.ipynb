# Cumulative geometric probability (greater than a value)

[Cumulative geometric probability (greater than a value) | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=7TYMKKQK8s8&t=6s)

Emelia registers vehicles for the Department of Transportation. Sports utility vehicles (SUVs) make up 12% of the vehicles she registers. Let V represent the number of vehicles Emelia registers in a day until she first registers an SUV. Assume the type of each vehicle is independent.

Find the probability that Emelia registers more than 4 vehicles before she registers an SUV.

**📌 Bullet summary: Geometric setup and answer**

- **Scenario:** Emelia registers vehicles; **12%** are SUVs.

- Let $V$= number of vehicles she registers **until she first
  registers an SUV**.

- Each vehicle is independently an SUV with probability $p = 0.12$.

- This is a **geometric random variable**: “number of trials until first
  success.”

- We want: $P(V > 4)$= probability she registers **more than 4
  vehicles** before the first SUV.

- “More than 4 before an SUV” means: **first 4 vehicles are not SUVs**.

- Probability a vehicle is **not** an SUV: $1 - p = 0.88$.

- So:

$$
P(V > 4) = (0.88)^{4} \approx 0.5997
$$

- **Final answer:** $P(V > 4) \approx 0.60$(about **60%**).

**📘 Explanation (step-by-step)**

1.  **Identify the distribution**

    - Each vehicle is either:

      - SUV (success) with probability $0.12$,

      - Not SUV (failure) with probability $0.88$.

    - Trials are independent.

    - We count how many vehicles until the **first SUV**.

    - This matches a **geometric random variable**.

2.  **Interpret** $P(V > 4)$

    - $V > 4$means: she has **not** seen an SUV in the first 4
      vehicles.

    - That is: vehicle 1, 2, 3, and 4 are all **not SUVs**.

3.  **Compute the probability**

    - Probability a single vehicle is not an SUV: $0.88$.

    - Probability first 4 are all not SUVs:

$$
P(V > 4) = {0.88}^{4}
$$

- Numerically:

$$
{0.88}^{4} \approx 0.5997
$$

- So there’s about a **60% chance** she registers more than 4 vehicles
  before seeing an SUV.


# 🧪 Python code example

```python
import numpy as np

p = 0.12          # Probability a vehicle is an SUV
q = 1 - p         # Probability a vehicle is NOT an SUV

# Theoretical probability: V > 4
prob_theoretical = q**4
print("Theoretical P(V > 4):", prob_theoretical)

# Optional: simulate to verify
def simulate_V(trials=100000):
    results = []
    for _ in range(trials):
        count = 0
        while True:
            count += 1
            # SUV with probability p
            if np.random.rand() < p:
                break
        results.append(count)
    return np.array(results)

samples = simulate_V()
prob_empirical = np.mean(samples > 4)
print("Empirical P(V > 4):", prob_empirical)
```
