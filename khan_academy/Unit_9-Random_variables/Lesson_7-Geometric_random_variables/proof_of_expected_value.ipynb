# Proof of expected value of geometric random variable

Here’s a clear, structured breakdown of the **expected value of a
geometric random variable**—a **bullet summary**, an **intuitive
explanation**, and a **Python code example** that mirrors the math you
walked through.

**📌 Bullet Summary: Expected Value of a Geometric Random Variable**

- A geometric random variable $X$counts the number of independent
  trials until the **first success**.

- Each trial has:

  - Success probability $p$

  - Failure probability $1 - p$

- Possible values of $X$: $1,2,3,\ldots$

- Probability mass function:

$$
P(X = k) = (1 - p)^{k - 1}p
$$

- Expected value (mean):

$$
E\lbrack X\rbrack = \frac{1}{p}
$$

- This formula comes from summing an infinite geometric series.

**📘 Explanation (Step‑by‑Step)**

**1. Start with the definition of expected value**

$$
E\lbrack X\rbrack = \sum_{k = 1}^{\infty}{k \cdot P(X = k)}
$$

For a geometric random variable:

$$
P(X = k) = (1 - p)^{k - 1}p
$$

So:

$$
E\lbrack X\rbrack = p(1 + 2(1 - p) + 3(1 - p)^{2} + 4(1 - p)^{3} + \cdots\text{ })
$$

**2. Use a clever trick: multiply by** $\mathbf{1 - p}$

Let:

$$
S = E\lbrack X\rbrack = p(1 + 2q + 3q^{2} + 4q^{3} + \cdots\text{ })
$$

where $q = 1 - p$.

Compute:

$$
qS = p(q + 2q^{2} + 3q^{3} + \cdots\text{ })
$$

Subtract:

$$
S - qS = p(1 + q + q^{2} + q^{3} + \cdots\text{ })
$$

Left side:

$$
S(1 - q) = Sp
$$

Right side is a geometric series:

$$
1 + q + q^{2} + q^{3} + \cdots = \frac{1}{1 - q} = \frac{1}{p}
$$

So:

$$
Sp = p \cdot \frac{1}{p} = 1
$$

Thus:

$$
S = \frac{1}{p}
$$

**3. Final result**

$$
\boxed{E\lbrack X\rbrack = \frac{1}{p}}
$$

This means:

- If success happens 20% of the time, you expect 5 trials.

- If success happens 5% of the time, you expect 20 trials.

- If success happens 50% of the time, you expect 2 trials.


# 🧪 Python Code Example
This simulation verifies that the expected value approaches 1/𝑝.

```python
import numpy as np

def simulate_geometric(p, trials=1_000_00):
    # Simulate geometric random variable: number of trials until first success
    return np.random.geometric(p, size=trials)

p = 0.25  # example probability of success
samples = simulate_geometric(p)

print("Theoretical E[X] =", 1/p)
print("Simulated E[X]  =", samples.mean())
```
