# Deriving the variance of the difference of random variables

[Variance of differences of random variables | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=rLdoKZ7w0xI)

## Summary: Tools for Sums and Differences of Random Variables

We consider two **independent random variables**, $X$ and $Y$, and review key tools for working with their **sums and differences**.

---

## Review of Notation

- The **expected value** of a random variable is the same as its **mean**:
  $$
  E[X] = \mu_X \quad \text{and} \quad E[Y] = \mu_Y
  $$

- The **variance** measures average squared distance from the mean:
  $$
  \operatorname{Var}(X) = E[(X - \mu_X)^2] = \sigma_X^2
  $$
  $$
  \operatorname{Var}(Y) = E[(Y - \mu_Y)^2] = \sigma_Y^2
  $$

---

## Expected Value of a Sum

Define a new random variable:
$$
Z = X + Y
$$

A fundamental property of expected values is:
$$
E[X + Y] = E[X] + E[Y]
$$

So the mean of $Z$ is:
$$
\mu_Z = \mu_X + \mu_Y
$$

This makes intuitive sense: if the variables are independent, their average contributions simply add.

---

## Expected Value of a Difference

Define another random variable:
$$
A = X - Y
$$

The expected value follows the same linear rule:
$$
E[X - Y] = E[X] - E[Y]
$$

So the mean of $A$ is:
$$
\mu_A = \mu_X - \mu_Y
$$

---

## Variance of a Sum (Independence Required)

For **independent** random variables, variances add:
$$
\operatorname{Var}(X + Y) = \operatorname{Var}(X) + \operatorname{Var}(Y)
$$

This reflects the idea that independent sources of variation combine to increase overall spread.

---

## Variance of a Difference

Now consider the variance of:
$$
A = X - Y
$$

Rewrite this as:
$$
A = X + (-Y)
$$

Using the variance rule for sums:
$$
\operatorname{Var}(X - Y) = \operatorname{Var}(X) + \operatorname{Var}(-Y)
$$

But note:
$$
\operatorname{Var}(-Y) = \operatorname{Var}(Y)
$$

This is because variance depends on **squared distance from the mean**, and squaring removes the sign.

Therefore:
$$
\operatorname{Var}(X - Y) = \operatorname{Var}(X) + \operatorname{Var}(Y)
$$

---

## Key Takeaways

- **Linearity of expectation**:
  - Mean of a sum = sum of means
  - Mean of a difference = difference of means

- **Variance rules (for independent variables)**:
  - Variance of a sum = sum of variances
  - Variance of a difference = sum of variances

- Variance does **not** subtract, even when variables are subtracted
- Independence is a crucial assumption for these variance formulas

These results are powerful tools that will be used repeatedly when analyzing distributions formed from sums and differences of random variables.


```python
import numpy as np

# Simulate two independent random variables X and Y
np.random.seed(0)

# X and Y are independent
X = np.random.normal(loc=5, scale=2, size=10000)   # mean=5, std=2
Y = np.random.normal(loc=7, scale=3, size=10000)   # mean=7, std=3

# Define new random variables
Z = X + Y      # sum
A = X - Y      # difference

# Compute empirical means
mean_X = np.mean(X)
mean_Y = np.mean(Y)
mean_Z = np.mean(Z)
mean_A = np.mean(A)

# Compute empirical variances
var_X = np.var(X)
var_Y = np.var(Y)
var_Z = np.var(Z)
var_A = np.var(A)

print("Means:")
print(f"E[X] = {mean_X:.2f}")
print(f"E[Y] = {mean_Y:.2f}")
print(f"E[X + Y] = {mean_Z:.2f}  (Expected: {mean_X + mean_Y:.2f})")
print(f"E[X - Y] = {mean_A:.2f}  (Expected: {mean_X - mean_Y:.2f})")

print("\nVariances:")
print(f"Var(X) = {var_X:.2f}")
print(f"Var(Y) = {var_Y:.2f}")
print(f"Var(X + Y) = {var_Z:.2f}  (Expected: {var_X + var_Y:.2f})")
print(f"Var(X - Y) = {var_A:.2f}  (Expected: {var_X + var_Y:.2f})")
```

# What this code demonstrates:

* The **mean of a sum** equals the **sum of the means**
* The **mean of a difference** equals the **difference of the means**
* For **independent random variables**, the **variance of both the sum and the difference** equals the sum of the variances
* The simulation numerically confirms the theoretical rules discussed in the summary
