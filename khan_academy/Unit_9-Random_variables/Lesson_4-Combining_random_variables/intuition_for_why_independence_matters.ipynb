## Summary: Why Independence Matters for Variance

In earlier discussions, we established an important result:

> If two random variables $X$ and $Y$ are **independent**, then  
> $$
> \operatorname{Var}(X + Y) = \operatorname{Var}(X) + \operatorname{Var}(Y)
> $$
> and the same holds for $X - Y$.

This implies that **variation increases** when we add or subtract independent random variables.

What this discussion focuses on is **why independence is essential** for that result to hold.

---

## A Counterexample: Dependent Random Variables

Consider two random variables that are clearly **not independent**:

- $X$: the number of hours a randomly selected person slept yesterday  
- $Y$: the number of hours that same person was awake yesterday  

These are both valid random variables, but they are **completely dependent**.

Why?

Because:
$$
X + Y = 24
$$

If you know how long the person slept, you immediately know how long they were awake. One variable **fully determines** the other.

---

## Why Independence Fails Here

Because $X$ and $Y$ are tied together:
- The probability of one variable depends entirely on the value of the other
- They are **not independent**

Suppose, for the sake of argument:
$$
\operatorname{Var}(X) = 4
$$
$$
\operatorname{Var}(Y) = 4
$$

If we blindly applied the variance formula for independent variables, we might (incorrectly) claim:
$$
\operatorname{Var}(X + Y) = 4 + 4 = 8
$$

But this makes no sense.

---

## The Reality

Since:
$$
X + Y = 24
$$
for every single outcome, the sum **never changes**.

A random variable that is always equal to a constant has **no variability**, so:
$$
\operatorname{Var}(X + Y) = 0
$$

---

## Key Insight

This example shows clearly why **independence matters**:

- When variables are dependent, variation can cancel out
- The rule $\operatorname{Var}(X \pm Y) = \operatorname{Var}(X) + \operatorname{Var}(Y)$ **does not apply**
- Knowing one variable gives complete information about the other

---

## Takeaways

- Independence is a **crucial assumption** when adding variances
- Without independence, variance can decrease or even drop to zero
- Dependent variables can move in perfectly opposite or complementary ways
- Always check independence before applying variance formulas

This intuition explains why variance formulas behave the way they do and sets the stage for understanding covariance in future topics.


```python
import numpy as np

# Simulate dependent random variables:
# X = hours slept, Y = hours awake, with X + Y = 24

np.random.seed(0)

# Simulate hours slept (X) with some variation
X = np.random.normal(loc=8, scale=2, size=10000)  # mean 8, std 2

# Y is completely determined by X
Y = 24 - X

# Compute variances
var_X = np.var(X)
var_Y = np.var(Y)
var_sum = np.var(X + Y)

print("Dependent random variables example:")
print(f"Variance of X (hours slept): {var_X:.2f}")
print(f"Variance of Y (hours awake): {var_Y:.2f}")
print(f"Variance of X + Y: {var_sum:.2f}")

# Show what would happen if we (incorrectly) assumed independence
print("\nIf we incorrectly assumed independence:")
print(f"Var(X) + Var(Y) = {var_X + var_Y:.2f}")
```

# This code clearly demonstrates that:

* Even though both variables vary individually,
* Their **sum has zero variance** because they are perfectly dependent,
* Showing exactly why **independence is required** when adding variances.
