## Summary: Variance of Sums and Differences of Independent Random Variables

We define two random variables:

- $X$: the **weight of cereal** in a randomly selected closed box of Mathies cereal  
- $Y$: the **weight of cereal** in a randomly filled bowl of Mathies cereal  

---

## Given Information

### Random Variable $X$ (Box of Cereal)
- Mean (expected value):  
  $$
  \mu_X = 16 \text{ ounces}
  $$
- Standard deviation:  
  $$
  \sigma_X = 0.8 \text{ ounces}
  $$
- Constrained range:  
  $$
  15 \le X \le 17
  $$

This reflects real-world variation due to packing differences, flake shapes, and density.

---

### Random Variable $Y$ (Bowl of Cereal)
- Mean (expected value):  
  $$
  \mu_Y = 4 \text{ ounces}
  $$
- Standard deviation:  
  $$
  \sigma_Y = 0.6 \text{ ounces}
  $$
- Constrained range:  
  $$
  3 \le Y \le 5
  $$

Variation arises from how the bowl is filled.

---

## Expected Value of the Sum

We consider the combined weight:

$$
X + Y
$$

The expected value of a sum of random variables is the sum of their expected values:

$$
E[X + Y] = E[X] + E[Y] = 16 + 4 = 20 \text{ ounces}
$$

---

## Variance of the Sum

A key rule (assuming **independence**) is:

$$
\operatorname{Var}(X + Y) = \operatorname{Var}(X) + \operatorname{Var}(Y)
$$

This also applies to differences:

$$
\operatorname{Var}(X - Y) = \operatorname{Var}(X) + \operatorname{Var}(Y)
$$

> **Important:** This rule only holds when $X$ and $Y$ are **independent**.

---

### Calculating the Variance

- Variance of $X$:  
  $$
  \sigma_X^2 = (0.8)^2 = 0.64
  $$
- Variance of $Y$:  
  $$
  \sigma_Y^2 = (0.6)^2 = 0.36
  $$

So:

$$
\operatorname{Var}(X + Y) = 0.64 + 0.36 = 1
$$

---

## Standard Deviation of the Sum

The standard deviation is the square root of the variance:

$$
\sigma_{X+Y} = \sqrt{1} = 1 \text{ ounce}
$$

---

## Intuition Using Ranges

- Individual ranges:
  - $X$: width of 2 ounces
  - $Y$: width of 2 ounces
- Combined range:
  $$
  18 \le X + Y \le 22
  $$
  (a width of 4 ounces)

The spread of the sum is larger than the spread of either variable alone, which supports why variances add.

The same intuition applies to differences:

$$
10 \le X - Y \le 14
$$

Again, the variability increases.

---

## Key Takeaways

- Expected values **add and subtract directly**
- **Standard deviations do not add**, but **variances do**
- For independent random variables:
  $$
  \operatorname{Var}(X \pm Y) = \operatorname{Var}(X) + \operatorname{Var}(Y)
  $$
- Standard deviation of the sum or difference is the square root of that total variance
- Independence is crucial for these rules to hold

In the next discussion, we’ll build deeper intuition for **why independence matters** in these variance formulas.


```python
import math

# Given statistics
mean_X = 16.0
std_X = 0.8

mean_Y = 4.0
std_Y = 0.6

# Expected value of sum
mean_sum = mean_X + mean_Y

# Variances
var_X = std_X ** 2
var_Y = std_Y ** 2

# Variance and standard deviation of the sum (assuming independence)
var_sum = var_X + var_Y
std_sum = math.sqrt(var_sum)

# Expected value of difference
mean_diff = mean_X - mean_Y

# Variance and standard deviation of the difference (also assuming independence)
var_diff = var_X + var_Y
std_diff = math.sqrt(var_diff)

print("Random variable X (box of cereal):")
print(f"Mean = {mean_X}, Standard Deviation = {std_X}")

print("\nRandom variable Y (bowl of cereal):")
print(f"Mean = {mean_Y}, Standard Deviation = {std_Y}")

print("\nSum X + Y:")
print(f"Mean = {mean_sum}")
print(f"Variance = {var_sum}")
print(f"Standard Deviation = {std_sum}")

print("\nDifference X - Y:")
print(f"Mean = {mean_diff}")
print(f"Variance = {var_diff}")
print(f"Standard Deviation = {std_diff}")
```
