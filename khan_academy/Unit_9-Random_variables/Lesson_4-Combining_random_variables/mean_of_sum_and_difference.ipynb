## Summary: Expected Value of Sums and Differences of Random Variables

Let $X$ be a random variable representing the **number of dogs seen in a day**, and let $Y$ be a random variable representing the **number of cats seen in a day**.

Suppose we know the expected values (means) of these random variables:

$$
\mu_X = E[X] = 3
$$

$$
\mu_Y = E[Y] = 4
$$

---

## Expected Value of a Sum

We are interested in the expected value of the sum $X + Y$, which represents the **total number of dogs and cats seen in a day**.

A fundamental result in probability is:

$$
E[X + Y] = E[X] + E[Y]
$$

So in this case:

$$
E[X + Y] = 3 + 4 = 7
$$

**Interpretation:**  
On average, you would expect to see **7 animals (dogs and cats combined)** per day.

---

## Expected Value of a Difference

Now consider the difference $Y - X$, which represents **how many more cats than dogs** you expect to see in a day.

Another key result is:

$$
E[Y - X] = E[Y] - E[X]
$$

So:

$$
E[Y - X] = 4 - 3 = 1
$$

**Interpretation:**  
On average, you would expect to see **1 more cat than dog per day**.

---

## Important Notes

- These results apply to **discrete random variables**, like counting animals.
- The **expected value itself does not need to be an integer**, even if the random variable only takes integer values.
- Expected values describe **long-run averages**, not exact outcomes on a single day.
- The rules:
  - Mean of a sum = sum of means
  - Mean of a difference = difference of means  
  hold **very generally**, without requiring additional assumptions.

A formal proof of these results will be explored in a future discussion.


```python
# Expected values of the random variables
E_X = 3  # expected number of dogs per day
E_Y = 4  # expected number of cats per day

# Expected value of the sum
E_sum = E_X + E_Y

# Expected value of the difference
E_diff = E_Y - E_X

print("Expected values:")
print(f"E[X] = {E_X}")
print(f"E[Y] = {E_Y}")

print("\nExpected value of the sum:")
print(f"E[X + Y] = {E_sum}")

print("\nExpected value of the difference:")
print(f"E[Y - X] = {E_diff}")
```
