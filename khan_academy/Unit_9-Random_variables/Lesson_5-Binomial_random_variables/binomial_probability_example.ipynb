# Binomial Probability Example --- Free Throws (Summary)

## Given Information

-   Probability of scoring a free throw:

    $ p = 0.7  $ (70%)

-   Probability of missing a free throw:

    $ 1 - p = 0.3  $ (30%)

-   Only two outcomes:

    -   Score
    -   Miss

-   Probabilities must sum to 1.

------------------------------------------------------------------------

## Problem

-   6 free throw attempts.
-   Find the probability of exactly 2 scores.

------------------------------------------------------------------------

## Step 1: Probability of One Specific Arrangement

Example arrangement:

Score, Score, Miss, Miss, Miss, Miss

Probability of this exact sequence:

$ 0.7^2 \times 0.3^4 $

Any specific sequence with: - 2 scores - 4 misses

has probability:

$ 0.7^2 \times 0.3^4 $

------------------------------------------------------------------------

## Step 2: Number of Ways to Get 2 Scores

We must count how many ways we can choose 2 scoring attempts out of 6.

$ \binom{6}{2} = \frac{6!}{2!(6-2)!} $

Compute:

$ \binom{6}{2} = 15 $

So there are 15 different ways to get exactly 2 scores.

------------------------------------------------------------------------

## Step 3: Final Probability

$ P(X = 2) = \binom{6}{2} \times 0.7^2 \times 0.3^4 $

Substitute values:

$ P(X = 2) = 15 \times 0.49 \times 0.0081 $

$ P(X = 2) = 0.059535 $

------------------------------------------------------------------------

## Final Answer

$ P(X = 2) \approx 0.0595 $

Approximately:

$ 6% $

There is about a 6% probability of scoring exactly 2 times in 6 attempts
when the success rate is 70%.

------------------------------------------------------------------------

## Key Insight

-   This is a Binomial Distribution problem.
-   General formula:

$ P(X = k) = \binom{n}{k} p^k (1-p)^{n-k} $

Where: - $n = 6$ - $k = 2$ - $p = 0.7$


### Python Code Example (Binomial Probability)
Here is a clean Python example that calculates the probability of exactly 2 scores in 6 free throw attempts when the probability of scoring is $ p = 0.7$ .

```python
import math

# Given values
n = 6          # number of attempts
k = 2          # number of scores we want
p = 0.7        # probability of scoring
q = 1 - p      # probability of missing

# Combination function (n choose k)
def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Binomial probability formula
probability = combination(n, k) * (p ** k) * (q ** (n - k))

print("Probability of exactly 2 scores in 6 attempts:")
print(f"P(X = 2) = {probability:.6f}")
print(f"Approximately {probability * 100:.2f}%")
```

## What This Code Uses

-   This is a Binomial Distribution problem.
-   General formula:

$ P(X = k) = \binom{n}{k} p^k (1-p)^{n-k} $

Where: - $n = 6$ - $k = 2$ - $p = 0.7$
