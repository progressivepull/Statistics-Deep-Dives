# Visualizing the Binomial Distribution --- Summary

## Recap of the Random Variable

-   Let $x$ = number of heads from flipping a **fair coin** 5 times.
-   Possible values:
    -   0, 1, 2, 3, 4, 5
-   Previously computed probabilities:

  |  $x$  |    Probability |
  | ----- | -------------- |
  | 0     |  1/32          |
  | 1     |  5/32          |
  | 2     |  10/32         |
  | 3     |  10/32         |
  | 4     |  5/32          |
  | 5     |  1/32          |

------------------------------------------------------------------------

## Plotting the Probability Distribution

-   Horizontal axis: possible values of $x$
-   Vertical axis: probabilities
-   The graph is a **histogram (bar chart)**.
-   Each bar represents $P(x = k)$.

------------------------------------------------------------------------

## Key Observations

-   The distribution:
    -   Starts at $1/32$
    -   Rises to $10/32$
    -   Falls symmetrically back to $1/32$
-   The pattern is symmetric:
    -   1, 5, 10, 10, 5, 1
-   This symmetry occurs because:
    -   $P(5 \text{ heads}) = P(0 \text{ heads})$
    -   $P(4 \text{ heads}) = P(1 \text{ head})$

------------------------------------------------------------------------

## Type of Distribution

-   This is a **discrete probability distribution**.
-   It is finite:
    -   Only 6 possible values.
-   It is specifically a **Binomial Distribution**.

The probabilities come from binomial coefficients:

$$
P(X = k) = \frac{\binom{5}{k}}{32}
$$

------------------------------------------------------------------------

## Relationship to the Normal Distribution

-   If the number of flips increases:
    -   The bars become narrower.
    -   The histogram begins to resemble a **bell curve**.
-   With many trials:

$$
\text{Binomial Distribution}
$$

$$
\rightarrow \text{Normal Distribution}
$$

-   The normal distribution is:
    -   Continuous
    -   A probability density function
    -   Extremely important in statistics and science

------------------------------------------------------------------------

## Big Picture Insight

-   The binomial distribution models discrete repeated trials.
-   As the number of trials becomes very large:
    -   It approximates the normal distribution.
-   Understanding this connection is fundamental in statistics.


```python
import math
import matplotlib.pyplot as plt

# Parameters
n = 5          # number of trials (coin flips)
p = 0.5        # probability of heads

# Combination function (n choose k)
def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Compute probabilities
x_values = list(range(n + 1))
probabilities = [
    combination(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in x_values
]

# Print probability table
print("Binomial Distribution (n=5, p=0.5)\n")
print("x   P(X = x)")
print("-" * 15)

for x, prob in zip(x_values, probabilities):
    print(f"{x}   {prob:.5f}")

# Plot the distribution
plt.figure()
plt.bar(x_values, probabilities)

plt.xlabel("Number of Heads (x)")
plt.ylabel("P(X = x)")
plt.title("Binomial Distribution (n=5, p=0.5)")

plt.show()
```
