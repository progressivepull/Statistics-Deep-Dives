# Random Variable Explanation — Selecting Employees (Hypergeometric Setting)

## 🔹 Bullet Summary
- Total employees: 20 (11 female, 9 male)
- A manager randomly selects 3 employees from a hat
- Random variable $ X $ = number of female employees selected
- Selection is **without replacement**
- Trials are **dependent** (probabilities change after each pick)
- This is **NOT a binomial distribution**
- It follows a **hypergeometric distribution**
- Possible values of $ X $: $ 0, 1, 2, 3 $

---

## 🔹 Explanation

The manager is choosing 3 employees randomly from a group of 20 employees (11 female and 9 male).  
We define the random variable:

$$
X = \text{number of female employees chosen}
$$

Each selection is made **without replacement**, meaning once a name is drawn, it is not returned to the hat.

### Why This Is NOT Binomial
For a distribution to be binomial, it must satisfy:
- Independent trials  
- Constant probability of success  
- Fixed number of trials  
- Two outcomes (success/failure)  

Although we have a fixed number of selections (3) and two outcomes (female or male), the key issue is:

- The probability of selecting a female changes after each draw
- Example:
  - First draw: $ P(\text{female}) = \frac{11}{20} $
  - Second draw (if a female was already picked): $ \frac{10}{19} $
- Because probabilities change, the trials are **not independent**

Thus, the distribution is **not binomial**.

---

## 🔹 Correct Distribution Type: Hypergeometric

Since:
- We are sampling without replacement
- From a finite population
- Counting number of successes (female employees)

The random variable $ X $ follows a **hypergeometric distribution**.

Notation:
$$
X \sim \text{Hypergeometric}(N, K, n)
$$
Where:
- $ N = 20 $ (total employees)
- $ K = 11 $ (number of female employees)
- $ n = 3 $ (number selected)

---

## 🔹 Probability Formula

The probability of selecting exactly $ k $ female employees is:

$$
P(X = k) = \frac{\binom{11}{k} \binom{9}{3-k}}{\binom{20}{3}}
$$

Where:
- $ \binom{11}{k} $ = ways to choose $ k $ females  
- $ \binom{9}{3-k} $ = ways to choose remaining males  
- $ \binom{20}{3} $ = total ways to choose any 3 employees  

---

## 🔹 Possible Values of X
$$
X \in \{0, 1, 2, 3\}
$$
- 0 females (all males)
- 1 female
- 2 females
- 3 females (all females)

---

## 🔹 Key Insight
Even though the situation looks similar to repeated success/failure trials, the lack of replacement makes the probability change after each draw.  
That dependence between selections is what makes this a **hypergeometric model instead of a binomial model**.


```python
import random

# Create list of employees
employees = ["F"] * 11 + ["M"] * 9

# Randomly choose 3 employees without replacement
selected = random.sample(employees, 3)

# Count number of females chosen
X = selected.count("F")

print("Selected employees:", selected)
print("Number of female employees chosen (X):", X)
```

# Example 2: Drawing Cards Without Replacement

In a game involving a standard deck of 52 playing cards, an individual randomly draws 7 cards without replacement. Let Y = the number of aces drawn.

The conditions for a binomial random variable are:      
• The outcome of each trial can be classified as either success or failure.    
• Each trial is independent of the others.    
• There is a fixed number of trials.     
• The probability p of success on each trial remains constant.       


# Random Variable Explanation — Number of Aces in 7 Draws (Without Replacement)

## 🔹 Bullet Summary
- A standard deck has 52 cards total
- There are 4 aces in the deck
- A player draws 7 cards randomly **without replacement**
- Random variable $ Y $ = number of aces drawn
- Sampling without replacement → probabilities change each draw
- Trials are dependent (not independent)
- This is **not a binomial distribution**
- Correct model: **Hypergeometric distribution**
- Possible values of $ Y $: $ 0, 1, 2, 3, 4 $

---

## 🔹 Explanation

In this game, 7 cards are drawn from a standard 52-card deck without putting any cards back.  
We define the random variable:

$$
Y = \text{number of aces drawn in 7 cards}
$$

Since there are only 4 aces in the deck, the maximum number of aces that can be drawn is 4.

---

## 🔹 Why This Is NOT a Binomial Variable

A binomial distribution requires:
- Independent trials  
- Constant probability of success  
- Fixed number of trials  
- Two outcomes (success/failure)  

Although:
- The number of draws is fixed (7)
- Each draw can be classified as ace (success) or not ace (failure)

It still fails the binomial conditions because:

### 1. Trials Are Not Independent
- After drawing one card, the deck composition changes
- Example:
  - First draw: $ P(\text{Ace}) = \frac{4}{52} $
  - If an ace is drawn, next probability becomes $ \frac{3}{51} $

### 2. Probability of Success Is Not Constant
- The probability of drawing an ace changes after every card drawn
- This happens because cards are **not replaced**

Therefore, the trials are dependent and the success probability is not constant.

---

## 🔹 Correct Distribution: Hypergeometric Distribution

This situation follows a hypergeometric model because:
- Finite population (52 cards)
- Fixed number of draws (7)
- Sampling without replacement
- Counting number of successes (aces)

Notation:
$$
Y \sim \text{Hypergeometric}(N, K, n)
$$

Where:
- $ N = 52 $ (total cards)
- $ K = 4 $ (number of aces in the deck)
- $ n = 7 $ (cards drawn)

---

## 🔹 Probability Formula

The probability of drawing exactly $ k $ aces is:

$$
P(Y = k) = \frac{\binom{4}{k} \binom{48}{7-k}}{\binom{52}{7}}
$$

Where:
- $ \binom{4}{k} $ = ways to choose $ k $ aces  
- $ \binom{48}{7-k} $ = ways to choose the remaining non-aces  
- $ \binom{52}{7} $ = total ways to choose any 7 cards  

---

## 🔹 Possible Values of Y
$$
Y \in \{0, 1, 2, 3, 4\}
$$

- 0 aces drawn  
- 1 ace drawn  
- 2 aces drawn  
- 3 aces drawn  
- 4 aces drawn (maximum possible)

---

## 🔹 Key Insight
Because the cards are drawn **without replacement**, each draw changes the deck’s composition.  
This makes the probability of drawing an ace different at each step, which creates dependent trials and leads to a **hypergeometric distribution instead of a binomial distribution**.


#### 🔹 Manual Formula Implementation (Pure Python)

The probability of drawing exactly \( k \) aces when 7 cards are drawn from a 52-card deck without replacement is:

$$
P(Y = k) = \frac{\binom{4}{k}\binom{48}{7-k}}{\binom{52}{7}}
$$

```python
import math

def hypergeometric_probability(N, K, n, k):
    numerator = math.comb(K, k) * math.comb(N - K, n - k)
    denominator = math.comb(N, n)
    return numerator / denominator

# Parameters
N = 52  # total cards
K = 4   # aces
n = 7   # cards drawn

# Example: probability of exactly 2 aces
k = 2
prob = hypergeometric_probability(N, K, n, k)

print(f"P(Y = {k}) =", prob)
```

### 🔹 Simulation Code (Monte Carlo Approach)

This simulates many 7-card draws to estimate the distribution of aces.

```python
import random

# Create deck (4 aces = 1, other cards = 0)
deck = [1]*4 + [0]*48

def simulate_draws(trials=100000):
    results = []
    
    for _ in range(trials):
        hand = random.sample(deck, 7)  # draw 7 cards without replacement
        aces = sum(hand)               # count number of aces
        results.append(aces)
    
    return results

# Run simulation
results = simulate_draws()

# Estimate probability of getting exactly 2 aces
prob_2_aces = results.count(2) / len(results)

print("Estimated P(Y = 2) =", prob_2_aces)
```

$$
