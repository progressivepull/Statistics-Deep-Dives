# Example 1: Selecting Employees for a Business Trip

A manager oversees 11 female employees and 9 male employees. They need to pick 3 of these employees to go on a business trip, so the manager places all 20 names in a hat and chooses at random. Let X = the number of female employees chosen. 

## Is X a binomial variable? Why or why not?

Choose 1 answer:
* **A** Each trial isn't being classified as a success or failure, so X is not a bionomial variables.
* **B** There is no fixed number of trials, so X is not a binomial variable.
* **C** The trials are not independent, so X is not a binomial variable.
* **D** This situation satisfies each of the conditions for a binomial variables, so X has a binomial distribution. 

# Random Variable Explanation — Selecting Employees (Hypergeometric Setting)

## 🔹 Bullet Summary
- Total employees: 20 (11 female, 9 male)
- A manager randomly selects 3 employees from a hat
- Random variable $ X $ = number of female employees selected
- Selection is **without replacement**
- Trials are **dependent** (probabilities change after each pick)
- This is **NOT a binomial distribution**
- It follows a **hypergeometric distribution**
- Possible values of $ X $: $ 0, 1, 2, 3 $

---

## 🔹 Explanation

The manager is choosing 3 employees randomly from a group of 20 employees (11 female and 9 male).  
We define the random variable:

$$
X = \text{number of female employees chosen}
$$

Each selection is made **without replacement**, meaning once a name is drawn, it is not returned to the hat.

### Why This Is NOT Binomial
For a distribution to be binomial, it must satisfy:
- Independent trials  
- Constant probability of success  
- Fixed number of trials  
- Two outcomes (success/failure)  

Although we have a fixed number of selections (3) and two outcomes (female or male), the key issue is:

- The probability of selecting a female changes after each draw
- Example:
  - First draw: $ P(\text{female}) = \frac{11}{20} $
  - Second draw (if a female was already picked): $ \frac{10}{19} $
- Because probabilities change, the trials are **not independent**

Thus, the distribution is **not binomial**.

---

## 🔹 Correct Distribution Type: Hypergeometric

Since:
- We are sampling without replacement
- From a finite population
- Counting number of successes (female employees)

The random variable $ X $ follows a **hypergeometric distribution**.

Notation:
$$
X \sim \text{Hypergeometric}(N, K, n)
$$
Where:
- $ N = 20 $ (total employees)
- $ K = 11 $ (number of female employees)
- $ n = 3 $ (number selected)

---

## 🔹 Probability Formula

The probability of selecting exactly $ k $ female employees is:

$$
P(X = k) = \frac{\binom{11}{k} \binom{9}{3-k}}{\binom{20}{3}}
$$

Where:
- $ \binom{11}{k} $ = ways to choose $ k $ females  
- $ \binom{9}{3-k} $ = ways to choose remaining males  
- $ \binom{20}{3} $ = total ways to choose any 3 employees  

---

## 🔹 Possible Values of X
$$
X \in \{0, 1, 2, 3\}
$$
- 0 females (all males)
- 1 female
- 2 females
- 3 females (all females)

---

## 🔹 Key Insight
Even though the situation looks similar to repeated success/failure trials, the lack of replacement makes the probability change after each draw.  
That dependence between selections is what makes this a **hypergeometric model instead of a binomial model**.


```python
import random

# Create list of employees
employees = ["F"] * 11 + ["M"] * 9

# Randomly choose 3 employees without replacement
selected = random.sample(employees, 3)

# Count number of females chosen
X = selected.count("F")

print("Selected employees:", selected)
print("Number of female employees chosen (X):", X)
```

# Example 2: Drawing Cards Without Replacement

In a game involving a standard deck of 52 playing cards, an individual randomly draws 7 cards without replacement. Let Y = the number of aces drawn.

The conditions for a binomial random variable are:      
• The outcome of each trial can be classified as either success or failure.    
• Each trial is independent of the others.    
• There is a fixed number of trials.     
• The probability p of success on each trial remains constant.       


# Random Variable Explanation — Number of Aces in 7 Draws (Without Replacement)

## 🔹 Bullet Summary
- A standard deck has 52 cards total
- There are 4 aces in the deck
- A player draws 7 cards randomly **without replacement**
- Random variable $ Y $ = number of aces drawn
- Sampling without replacement → probabilities change each draw
- Trials are dependent (not independent)
- This is **not a binomial distribution**
- Correct model: **Hypergeometric distribution**
- Possible values of $ Y $: $ 0, 1, 2, 3, 4 $

---

## 🔹 Explanation

In this game, 7 cards are drawn from a standard 52-card deck without putting any cards back.  
We define the random variable:

$$
Y = \text{number of aces drawn in 7 cards}
$$

Since there are only 4 aces in the deck, the maximum number of aces that can be drawn is 4.

---

## 🔹 Why This Is NOT a Binomial Variable

A binomial distribution requires:
- Independent trials  
- Constant probability of success  
- Fixed number of trials  
- Two outcomes (success/failure)  

Although:
- The number of draws is fixed (7)
- Each draw can be classified as ace (success) or not ace (failure)

It still fails the binomial conditions because:

### 1. Trials Are Not Independent
- After drawing one card, the deck composition changes
- Example:
  - First draw: $ P(\text{Ace}) = \frac{4}{52} $
  - If an ace is drawn, next probability becomes $ \frac{3}{51} $

### 2. Probability of Success Is Not Constant
- The probability of drawing an ace changes after every card drawn
- This happens because cards are **not replaced**

Therefore, the trials are dependent and the success probability is not constant.

---

## 🔹 Correct Distribution: Hypergeometric Distribution

This situation follows a hypergeometric model because:
- Finite population (52 cards)
- Fixed number of draws (7)
- Sampling without replacement
- Counting number of successes (aces)

Notation:
$$
Y \sim \text{Hypergeometric}(N, K, n)
$$

Where:
- $ N = 52 $ (total cards)
- $ K = 4 $ (number of aces in the deck)
- $ n = 7 $ (cards drawn)

---

## 🔹 Probability Formula

The probability of drawing exactly $ k $ aces is:

$$
P(Y = k) = \frac{\binom{4}{k} \binom{48}{7-k}}{\binom{52}{7}}
$$

Where:
- $ \binom{4}{k} $ = ways to choose $ k $ aces  
- $ \binom{48}{7-k} $ = ways to choose the remaining non-aces  
- $ \binom{52}{7} $ = total ways to choose any 7 cards  

---

## 🔹 Possible Values of Y
$$
Y \in \{0, 1, 2, 3, 4\}
$$

- 0 aces drawn  
- 1 ace drawn  
- 2 aces drawn  
- 3 aces drawn  
- 4 aces drawn (maximum possible)

---

## 🔹 Key Insight
Because the cards are drawn **without replacement**, each draw changes the deck’s composition.  
This makes the probability of drawing an ace different at each step, which creates dependent trials and leads to a **hypergeometric distribution instead of a binomial distribution**.


#### 🔹 Manual Formula Implementation (Pure Python)

The probability of drawing exactly \( k \) aces when 7 cards are drawn from a 52-card deck without replacement is:

$$
P(Y = k) = \frac{\binom{4}{k}\binom{48}{7-k}}{\binom{52}{7}}
$$

```python
import math

def hypergeometric_probability(N, K, n, k):
    numerator = math.comb(K, k) * math.comb(N - K, n - k)
    denominator = math.comb(N, n)
    return numerator / denominator

# Parameters
N = 52  # total cards
K = 4   # aces
n = 7   # cards drawn

# Example: probability of exactly 2 aces
k = 2
prob = hypergeometric_probability(N, K, n, k)

print(f"P(Y = {k}) =", prob)
```

### 🔹 Simulation Code (Monte Carlo Approach)

This simulates many 7-card draws to estimate the distribution of aces.

```python
import random

# Create deck (4 aces = 1, other cards = 0)
deck = [1]*4 + [0]*48

def simulate_draws(trials=100000):
    results = []
    
    for _ in range(trials):
        hand = random.sample(deck, 7)  # draw 7 cards without replacement
        aces = sum(hand)               # count number of aces
        results.append(aces)
    
    return results

# Run simulation
results = simulate_draws()

# Estimate probability of getting exactly 2 aces
prob_2_aces = results.count(2) / len(results)

print("Estimated P(Y = 2) =", prob_2_aces)
```

$$

# Example 3: Tomato Plant Survival

60% of a certain species of tomato live after transplanting from pot to garden. Eli transplants 16 of these tomato plants. Assume that the plants live independently of each other. Let T = the number of tomato plants that live.

The conditions for a binomial random variable are:      
• The outcome of each trial can be classified as either success or failure.    
• Each trial is independent of the others.    
• There is a fixed number of trials.     
• The probability p of success on each trial remains constant.    


# Random Variable Explanation — Number of Tomato Plants That Live

## 🔹 Bullet Summary
- 60% of tomato plants survive after transplanting
- Eli transplants 16 tomato plants
- Each plant lives or dies independently
- Random variable $ T $ = number of plants that live
- Fixed number of trials: $ n = 16 $
- Two outcomes per trial: live (success) or die (failure)
- Constant probability of success: $ p = 0.60 $
- Trials are independent
- Therefore, $ T $ follows a **Binomial distribution**
- Possible values of $ T $: $ 0, 1, 2, \dots, 16 $

---

## 🔹 Explanation

Eli transplants 16 tomato plants into the garden.  
Each plant has:
- Probability of living = $ 0.60 $
- Probability of dying = $ 0.40 $

We define the random variable:
$$
T = \text{number of tomato plants that live}
$$

Each plant represents one trial. For every trial:
- Success = plant lives
- Failure = plant dies

---

## 🔹 Why This Is a Binomial Random Variable

This situation satisfies all binomial conditions:

### 1. Fixed Number of Trials
- Eli transplants exactly 16 plants  
$$
n = 16
$$

### 2. Independent Trials
- The problem states plants live **independently**
- One plant surviving does not affect another

### 3. Two Possible Outcomes
- Live (success)
- Die (failure)

### 4. Constant Probability of Success
- Each plant has the same survival probability  
$$
p = 0.60
$$

Since all four conditions are met, the random variable follows a binomial model.

---

## 🔹 Distribution Notation
$$
T \sim \text{Binomial}(n = 16, \; p = 0.60)
$$

---

## 🔹 Probability Formula

The probability that exactly $ k $ plants live is:

$$
P(T = k) = \binom{16}{k} (0.60)^k (0.40)^{16-k}
$$

Where:
- $ k $ = number of plants that survive
- $ \binom{16}{k} $ = number of ways to choose $ k $ surviving plants

---

## 🔹 Expected Value (Mean)

The expected number of plants that live is:
$$
E[T] = n \cdot p = 16 \cdot 0.60 = 9.6
$$

This means, on average, about 9 to 10 plants are expected to survive.


# Python Code Example — Binomial Distribution (Tomato Plants Survival)

## 🔹 Problem Setup
- Probability a plant lives: $ p = 0.60 $
- Number of plants transplanted: $ n = 16 $
- Random variable:
  $$
  T = \text{number of tomato plants that live}
  $$
- Distribution:
  $$
  T \sim \text{Binomial}(n=16, \; p=0.60)
  $$

## 🔹 Method 1: Using SciPy (Recommended)

```python
from scipy.stats import binom

# Parameters
n = 16      # number of trials (plants)
p = 0.60    # probability a plant lives
k = 10      # number of plants that live

# Probability P(T = k)
probability = binom.pmf(k, n, p)

print(f"P(T = {k}) =", probability)
```

# Example 4: Rolling Dice Until a Double
In a game of luck, a turn consists of a player continuing to roll a pair of six-sided dice until they roll a double (two of the same face values). Let X = the number of rolls in one turn.

The conditions for a binomial random variable are:      
• The outcome of each trial can be classified as either success or failure.    
• Each trial is independent of the others.    
• There is a fixed number of trials.     
• The probability p of success on each trial remains constant.    


# Random Variable Explanation — Number of Rolls Until First Double

## 🔹 Bullet Summary
- A player rolls two six-sided dice repeatedly
- A turn ends when a double (same numbers) is rolled
- Random variable $ X $ = number of rolls until the first double
- Each roll is independent of previous rolls
- Probability of rolling a double stays constant
- This is a **Geometric distribution**
- Success = rolling a double
- Failure = not rolling a double
- Possible values of $ X $: $ 1, 2, 3, \dots $ (no fixed maximum)

---

## 🔹 Explanation

In this game, the player keeps rolling a pair of dice until they get a double (like 1-1, 2-2, ..., 6-6).  
We define:
$$
X = \text{number of rolls needed to get the first double}
$$

Each roll of the dice is one trial.

---

## 🔹 Identifying the Type of Distribution

This situation follows a **geometric distribution** because:

### 1. Repeated Independent Trials
- Each roll of the dice does not depend on previous rolls
- The dice outcomes are independent each time

### 2. Constant Probability of Success
- A double occurs when both dice show the same number
- There are 6 doubles: (1,1), (2,2), ..., (6,6)
- Total possible outcomes when rolling two dice = 36
$$
P(\text{double}) = \frac{6}{36} = \frac{1}{6}
$$

### 3. Trials Continue Until First Success
- The player keeps rolling until the first double appears
- This matches the definition of a geometric random variable

---

## 🔹 Distribution Notation
$$
X \sim \text{Geometric}(p = \tfrac{1}{6})
$$

Where:
- $ p = \frac{1}{6} $ is the probability of rolling a double on any roll

---

## 🔹 Probability Formula

The probability that the first double occurs on the $ k $-th roll is:

$$
P(X = k) = (1 - p)^{k-1} \cdot p
$$

Substituting $ p = \frac{1}{6} $:

$$
P(X = k) = \left(\frac{5}{6}\right)^{k-1} \cdot \frac{1}{6}
$$

---

## 🔹 Expected Value (Mean)

The expected number of rolls until the first double is:
$$
E[X] = \frac{1}{p} = \frac{1}{1/6} = 6
$$

On average, a player will need about 6 rolls to get their first double.

---

## 🔹 Key Insight
Because each roll has the same probability of success (rolling a double) and the rolls continue until the first success occurs, the random variable $ X $ is best modeled by a **geometric distribution**, not a binomial or hypergeometric distribution.


# Python Program — Geometric Distribution (Roll Until First Double)

## 🔹 Problem Setup
- Roll two six-sided dice repeatedly
- Stop when a double (same face values) appears
- Probability of success (double on one roll):
  $$
  p = \frac{6}{36} = \frac{1}{6}
  $$
- Random variable:
  $$
  X = \text{number of rolls until the first double}
  $$
- Distribution:
  $$
  X \sim \text{Geometric}(p = \tfrac{1}{6})
  $$

```python
import random

def rolls_until_double():
    rolls = 0
    
    while True:
        die1 = random.randint(1, 6)
        die2 = random.randint(1, 6)
        rolls += 1
        
        if die1 == die2:  # success: rolled a double
            return rolls

# Run one game
result = rolls_until_double()
print("Number of rolls until first double:", result)
```

#### 🔹 Program 2: Run Multiple Simulations (Monte Carlo)

This estimates the average number of rolls needed.

```python
import random

def rolls_until_double():
    rolls = 0
    while True:
        rolls += 1
        if random.randint(1, 6) == random.randint(1, 6):
            return rolls

# Number of simulations
trials = 10000

results = [rolls_until_double() for _ in range(trials)]

print("First 10 results:", results[:10])
print("Average rolls until first double:", sum(results) / trials)
```

#### 🔹 Program 3: Using NumPy (Geometric Distribution)

Since the probability of a double is p=1/6, we can model directly.

```python
import numpy as np

p = 1/6       # probability of rolling a double
trials = 10000

# Generate geometric random variables
samples = np.random.geometric(p, size=trials)

print("First 10 simulated values:", samples[:10])
print("Estimated mean number of rolls:", samples.mean())
```

#### 🔹 Program 4: Probability of First Double on k-th Roll

Using the geometric formula:

$$
P(X = k) = (1 − p)^{k−1} ⋅ p
$$

```python
def geometric_probability(k, p):
    return (1 - p)**(k - 1) * p

p = 1/6
k = 4  # example: probability first double occurs on 4th roll

prob = geometric_probability(k, p)
print(f"P(X = {k}) =", prob)
```
