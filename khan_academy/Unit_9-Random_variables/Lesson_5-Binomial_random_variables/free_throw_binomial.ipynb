# Free throw binomial probability distribution

# Binomial Random Variable --- Free Throws (Summary)

## Definition of the Random Variable

-   Let $X$ = number of made free throws out of 6 attempts.

-   Assume:

    $ p = 0.7 $

    (70% free throw probability)

-   Then:

    $ 1 - p = 0.3 $

------------------------------------------------------------------------

## Possible Values of $X$

The random variable can only take:

$ 0, 1, 2, 3, 4, 5, 6 $

It cannot take negative values or decimals.

------------------------------------------------------------------------

## Probability Formula

For exactly $k$ made shots:

$ P(X = k) = \binom{6}{k} (0.7)^k (0.3)^{6-k} $

------------------------------------------------------------------------

## Approximate Probability Distribution

  | $k$    | Probability    | Percent  |
  | ------ | -------------- | ---------|
  |  0     |  0.0007        |  0.1%    |
  |  1     |  0.010         |  1.0%    |
  |  2     |  0.060         |  6.0%    |
  |  3     |  0.185         |  18.5%   |
  |  4     |  0.324         |  32.4%   |
  |  5     |  0.303         |  30.3%   |
  |  6     |  0.118         |  11.8%   |

------------------------------------------------------------------------

## Key Observations

-   The distribution is not symmetric.

-   Why?

    -   Because $p = 0.7$ (not 0.5).
    -   Making a shot is more likely than missing.

-   However, the binomial coefficients are symmetric:

    -   $\binom{6}{0} = \binom{6}{6}$
    -   $\binom{6}{1} = \binom{6}{5}$
    -   $\binom{6}{2} = \binom{6}{4}$

-   The asymmetry comes from the weighting:

    $ (0.7)^k (0.3)^{6-k} $

------------------------------------------------------------------------

## Big Picture

-   This is a Binomial Distribution.
-   Requirements satisfied:
    -   Fixed number of trials (6)
    -   Two outcomes (make/miss)
    -   Constant probability (0.7)
    -   Independent trials
-   The table above gives the full probability distribution of the
    random variable $X$.


# Python Code Example
Here is a clean Python example that computes and displays the full binomial probability distribution for the free throw random variable:

* 𝑛 = 6 attempts
* 𝑝 = 0.7 probability of making a shot


```python
import math

# Parameters
n = 6          # number of attempts
p = 0.7        # probability of making a shot

# Combination function (n choose k)
def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Binomial probability function
def binomial_probability(n, k, p):
    return combination(n, k) * (p ** k) * ((1 - p) ** (n - k))

print("Binomial Distribution (n=6, p=0.7)\n")
print("k   P(X = k)      Percent")
print("-" * 30)

# Compute and display full distribution
for k in range(n + 1):
    prob = binomial_probability(n, k, p)
    print(f"{k}   {prob:.6f}     {prob*100:.2f}%")
```

# What This Code Does

* Implements the binomial formula:

$ P(X = k) = \binom{n}{k} p^k (1-p)^{n-k} $

* Computes probabilities for:
    - 𝑘 = 0,1,2,3,4,5,6
* Prints:
    - Decimal probability
    - Percentage form
