# Binomial and binomial functions

I have a 0.35 probability of making a free throw.       
What is the probability of making 4 out of 7 free throws?     


What is the probability of making less than 5 free throws ?    


# Using a Graphing Calculator for Binomial Random Variables --- Summary

## Scenario

-   Probability of making a free throw:

    $ p = 0.35 $

-   Number of trials:

    $ n = 7 $

-   Random variable:

    $ X = \text{number of made free throws out of 7} $

------------------------------------------------------------------------

## Question 1

### What is the probability of making exactly 4 out of 7 free throws?

This means:

$ P(X = 4) $

Use the calculator function:

binompdf(n, p, x)

Substitute:

-   $n = 7$
-   $p = 0.35$
-   $x = 4$

Result:

$ P(X = 4) \approx 0.14 $

------------------------------------------------------------------------

## Question 2

### What is the probability of making fewer than 5 free throws?

This means:

$ P(X < 5) $

Which is equivalent to:

$ P(X \le 4) $

Use the cumulative function:

binomcdf(n, p, x)

Substitute:

-   $n = 7$
-   $p = 0.35$
-   $x = 4$

This calculates:

$ P(X = 0,1,2,3,4) $

Result:

$ P(X \le 4) \approx 0.94 $

------------------------------------------------------------------------

## Key Calculator Functions

-   binompdf → Probability of exactly x successes
-   binomcdf → Probability of x or fewer successes

------------------------------------------------------------------------

## Key Takeaways

-   Clearly define:
    -   $n$ (number of trials)
    -   $p$ (probability of success)
    -   $x$ (number of successes)
-   Use binompdf for exact probability.
-   Use binomcdf for cumulative probability.
-   Graphing calculators can save significant time on exams like AP
    Statistics.


### Python Code Example

Here is a clean Python example that replicates what the graphing calculator does using Python.

It computes:
* $ 𝑃(𝑋 = 4 ) $ using the binomial probability formula (like **binompdf** )
* $ P(𝑋 \le 4) $ using cumulative probability (like **binomcdf** )

```python
import math

# Given values
n = 7          # number of trials
p = 0.35       # probability of success
x = 4          # number of successes

# Combination function (n choose k)
def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Binomial probability function (like binompdf)
def binomial_probability(n, k, p):
    return combination(n, k) * (p ** k) * ((1 - p) ** (n - k))

# Compute P(X = 4)
prob_exact = binomial_probability(n, x, p)

# Compute cumulative probability P(X <= 4) (like binomcdf)
prob_cumulative = sum(binomial_probability(n, k, p) for k in range(x + 1))

print("Binomial Distribution Example (n=7, p=0.35)\n")

print(f"P(X = {x}) = {prob_exact:.6f}")
print(f"Approximately {prob_exact:.2f}")

print(f"\nP(X <= {x}) = {prob_cumulative:.6f}")
print(f"Approximately {prob_cumulative:.2f}")
```

# What This Code Does

* Implements the binomial formula:

$ P(X = k) = \binom{n}{k} p^k (1-p)^{n-k} $

* Computes:
    - Exact probability (like **binompdf** )
    - Cumulative probability (like **binomcdf** )
