# 10% Rule and Approximate Independence: Summary

In statistics, we often assume distributions are **Normal** or **Binomial**. These assumptions allow us to make useful inferences and apply powerful statistical tools to our data.

---

## Independence Requirement
Normal and binomial models typically assume that observations are the sum of **independent trials**.

* **True Independence:** Easily achieved when sampling **with replacement**.
* **Real-World Reality:** In scenarios like surveying people leaving a mall, sampling is done **without replacement**. This means trials are technically not independent.

## The Practical Question
If the population is very large relative to the sample, does the lack of true independence actually matter?
* **The Intuition:** If you are only sampling a tiny fraction of a population, the removal of one individual doesn't significantly change the probabilities for the next. This is called **approximate independence**.

---

## The 10% Rule
To determine when this approximation is safe to use, statisticians use a practical guideline:

> If the **sample size $n \le$ 10% of the population $N$**, it is reasonable to assume approximate independence and treat the distribution as binomial.

While the 10% threshold is a somewhat arbitrary guideline, it is the industry standard because it works well in practice.

---

## Example: Selecting 3 Boys from a Class
Let $x$ = the number of boys selected in 3 trials. Assume the population is 50% boys and 50% girls.

### 1. With Replacement
Trials are independent. The probability of selecting 3 boys is:
$$\frac{1}{2} \times \frac{1}{2} \times \frac{1}{2} = 12.5\%$$
In this case, $x$ is **truly binomial**.

### 2. Without Replacement
Trials are not independent. As the population size increases, notice how the probability approaches the 12.5% "ideal":


| Population Size | Sample % of Population | $P(\text{3 boys})$ Without Replacement | Comparison to 12.5% |
| :--- | :--- | :--- | :--- |
| 20 | 15% | 10.5% | Noticeably different |
| 30 | 10% | Closer | More similar |
| 100 | 3% | Very close | Nearly same |
| 10,000 | 0.03% | 12.49% | Essentially equal |

---

## Key Insights
* **The Line:** If the sample $\le$ 10% of the population, binomial modeling is acceptable. If the sample is larger (e.g., 15%), the approximation may be poor.
* **The Balance:** While a smaller *percentage* of the population is better for the independence assumption, a larger *absolute sample size* is generally better for statistical accuracy.
* **The Ideal:** A large sample size (for power) drawn from an even larger population (for independence).

## Final Takeaways
The **10% Rule** justifies approximate independence when sampling without replacement. This allows us to treat variables as binomial and apply standard statistical inference tools with confidence.


### Replacement (binomial) vs without replacement (hypergeometric).
Here is a simple Python example demonstrating the 10% rule by comparing probabilities

```python
import math
from scipy.stats import binom, hypergeom

# Parameters
sample_size = 3          # number of trials
p_boy = 0.5              # 50% boys
k = 3                    # probability of selecting 3 boys

# Population sizes to compare
population_sizes = [20, 30, 100, 10000]

print("Comparing With vs Without Replacement\n")

for N in population_sizes:
    num_boys = N // 2  # 50% boys
    
    # With replacement (Binomial)
    prob_binomial = binom.pmf(k, sample_size, p_boy)
    
    # Without replacement (Hypergeometric)
    prob_hypergeom = hypergeom.pmf(k, N, num_boys, sample_size)
    
    sample_percent = (sample_size / N) * 100
    
    print(f"Population Size: {N}")
    print(f"Sample is {sample_percent:.2f}% of population")
    print(f"P(3 boys) With Replacement (Binomial): {prob_binomial:.5f}")
    print(f"P(3 boys) Without Replacement (Hypergeometric): {prob_hypergeom:.5f}")
    print("-" * 50)
```

# 10% Rule and Approximate Independence

This document outlines the relationship between population size and sampling methods, specifically comparing Binomial and Hypergeometric models.

## What This Code Demonstrates

*   **`binom.pmf()`**: Models **independent** trials (sampling with replacement).
*   **`hypergeom.pmf()`**: Models sampling **without replacement** (where trials are technically dependent).

---

## Key Observations

### 1. Population Scaling
As the population size increases while the sample size stays the same, the **hypergeometric probability approaches the binomial probability**. This happens because the impact of removing one individual from a massive group becomes negligible.

### 2. The 10% Threshold
When the sample size is **$\le$ 10% of the population**:
*   The two probabilities become very close.
*   The error introduced by assuming independence is small enough to ignore for most statistical applications.

### 3. Practical Application
The **10% Rule** justifies treating a real-world situation (without replacement) as **approximately binomial**. This allows us to use simpler formulas and standard Normal approximations.

---

## Summary Table: Approximation Accuracy


| Condition | Sampling Method | Independence | Statistical Model |
| :--- | :--- | :--- | :--- |
| With Replacement | Independent | True | Binomial |
| Without Replacement (Sample > 10%) | Dependent | No | Hypergeometric |
| Without Replacement (Sample $\le$ 10%) | Approx. Independent | Yes | **Approx. Binomial** |

> **Conclusion:** The smaller the sampling fraction, the better the approximation.
