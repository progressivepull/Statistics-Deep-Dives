# Binomial Random Variable: 5 Coin Flips --- Summary

## Definition of the Random Variable

-   Let $x$ = number of heads (H) obtained from flipping a **fair coin**
    5 times.
-   Possible values of $x$:
    -   0, 1, 2, 3, 4, 5

------------------------------------------------------------------------

## Total Possible Outcomes

-   Each flip has 2 possible outcomes: H or T
-   Total outcomes from 5 flips:

$$ 
2^5 = 32 
$$

-   All 32 outcomes are **equally likely**.

------------------------------------------------------------------------

## Strategy for Finding Probabilities

-   To find $P(x = k)$:
    -   Count how many of the 32 outcomes contain exactly $k$ heads.
-   Use combinations:

$$ 
\binom{5}{k} = \frac{5!}{k!(5-k)!}
$$

-   Then:

$$ 
P(x = k) = \frac{\binom{5}{k}}{32} 
$$

------------------------------------------------------------------------

## Probability Calculations

### $P(x = 0)$

-   Only 1 outcome (TTTTT)
-   $\binom{5}{0} = 1$

$$ 
P(x = 0) = \frac{1}{32} 
$$

------------------------------------------------------------------------

### $P(x = 1)$

-   5 possible positions for one head
-   $\binom{5}{1} = 5$

$$ 
P(x = 1) = \frac{5}{32}
$$

------------------------------------------------------------------------

### $P(x = 2)$

-   Choose 2 of 5 flips to be heads
-   $\binom{5}{2} = 10$

$$ 
P(x = 2) = \frac{10}{32}
$$

------------------------------------------------------------------------

### $P(x = 3)$

-   Choose 3 of 5 flips to be heads
-   $\binom{5}{3} = 10$

$$ 
P(x = 3) = \frac{10}{32}
$$

------------------------------------------------------------------------

### $P(x = 4)$

-   Choose 4 of 5 flips to be heads
-   $\binom{5}{4} = 5$

$$ 
P(x = 4) = \frac{5}{32}
$$

------------------------------------------------------------------------

### $P(x = 5)$

-   Only 1 outcome (HHHHH)
-   $\binom{5}{5} = 1$

$$
P(x = 5) = \frac{1}{32}
$$

------------------------------------------------------------------------

## Final Probability Distribution

| $x$   |   Probability | 
| ----- |  -------------| 
|  0    |   1/32        | 
|  1    |   5/32        | 
|  2    |   10/32       | 
|  3    |   10/32       | 
|  4    |   5/32        | 
|  5    |   1/32        | 

------------------------------------------------------------------------

## Key Observations

-   The distribution is **symmetric**.
-   Coefficients follow Pascal's Triangle: 1, 5, 10, 10, 5, 1.
-   This is a **Binomial Distribution** with:

$$ 
n = 5, \quad p = 0.5 
$$


## Python Example: 5 Coin Flips (Binomial Distribution)

```python
import math

# Parameters
n = 5          # number of coin flips
p = 0.5        # probability of heads

def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

print("Binomial Distribution for 5 Fair Coin Flips\n")
print("x  P(X = x)")
print("-" * 15)

for k in range(n + 1):
    prob = combination(n, k) * (p ** k) * ((1 - p) ** (n - k))
    print(f"{k}  {prob:.5f}")
```

# What This Code Does

## Uses the Binomial Formula

$$  
P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k} 
$$

------------------------------------------------------------------------

## Computes Probabilities For

\$ k = 0, 1, 2, 3, 4, 5 \$

------------------------------------------------------------------------

## Prints the Probability Distribution For

\$ n = 5 \$

\$ p = 0.5 \$

```python
import math
import matplotlib.pyplot as plt

# Parameters
n = 5
p = 0.5

# Combination function
def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Compute probabilities
x_values = list(range(n + 1))
probabilities = [
    combination(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in x_values
]

# Create bar plot
plt.figure()
plt.bar(x_values, probabilities)

plt.xlabel("Number of Heads (k)")
plt.ylabel("P(X = k)")
plt.title("Binomial Distribution (n=5, p=0.5)")

plt.show()
```
