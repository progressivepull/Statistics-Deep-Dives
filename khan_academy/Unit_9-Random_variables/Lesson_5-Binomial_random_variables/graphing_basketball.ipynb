### Python Code Example (Free Throw Binomial Distribution)

```python
import math
import matplotlib.pyplot as plt

# Parameters
n = 6          # number of attempts
p = 0.7        # probability of making a shot

# Combination function
def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Binomial probability function
def binomial_probability(n, k, p):
    return combination(n, k) * (p ** k) * ((1 - p) ** (n - k))

# Compute distribution
x_values = list(range(n + 1))
probabilities = [binomial_probability(n, k, p) for k in x_values]

# Print table
print("Binomial Distribution (n=6, p=0.7)\n")
print("k   P(X = k)      Percent")
print("-" * 30)

for k, prob in zip(x_values, probabilities):
    print(f"{k}   {prob:.6f}     {prob*100:.2f}%")

# Plot histogram
plt.figure()
plt.bar(x_values, probabilities)

plt.xlabel("Number of Made Free Throws (X)")
plt.ylabel("Probability")
plt.title("Binomial Distribution (n=6, p=0.7)")

plt.show()
```

# What This Code Does

* Implements the binomial formula:

$ P(X = k) = \binom{n}{k} p^k (1-p)^{n-k} $

* Computes probabilities for:
    - 𝑘 = 0,1,2,3,4,5,6
* Prints:
    - Decimal probability
    - Percentage form
