# Binomial Random Variables — Bullet Summary, Explanation, Formula, and Code

## 🔹 Bullet Summary
- A **binomial random variable** counts the number of successes in a fixed number of trials.
- Example: Number of heads in 10 coin flips where $ P(\text{Heads}) = 0.6 $.
- Key conditions for a binomial variable:
  - Finite number of trials $ n $
  - Independent trials
  - Two possible outcomes (success/failure)
  - Constant probability of success $ p $
- Each trial must be identical in structure.
- If probability changes between trials, the variable is **not binomial**.
- Example of NOT binomial: Drawing cards *without replacement* (probabilities change).
- With replacement → probabilities stay constant → becomes binomial.

---

## 🔹 Explanation

A **binomial random variable** represents the number of successes in repeated experiments where:
- Each experiment (trial) has only two outcomes (success or failure)
- The probability of success remains the same for every trial
- Trials are independent of each other
- The number of trials is fixed in advance

### Example (Binomial)
Let:
- $ X = $ number of heads in 10 coin flips
- Probability of heads $ p = 0.6 $
- Number of trials $ n = 10 $

Why this is binomial:
- Each flip = independent trial  
- Two outcomes: heads (success) or tails (failure)  
- Fixed number of trials (10 flips)  
- Constant probability $ p = 0.6 $

### Non-Binomial Example
Let:
- $ Y = $ number of kings drawn from 2 cards **without replacement**

Why NOT binomial:
- Probability changes after the first draw
- Trials are dependent
- Success probability is not constant

If cards were drawn **with replacement**, then:
- Probability stays $ \frac{4}{52} $
- Trials become independent
- The variable becomes binomial

---

## 🔹 Mathematical Formula (Binomial Distribution)

Probability of exactly $ k $ successes in $ n $ trials:

$$
P(X = k) = \binom{n}{k} p^k (1 - p)^{n-k}
$$

Where:
- $ n $ = number of trials
- $ k $ = number of successes
- $ p $ = probability of success
- $ \binom{n}{k} = \frac{n!}{k!(n-k)!} $

Alternative notation:
$$
X \sim \text{Binomial}(n, p)
$$

## 🔹 Code Example (Python)

### Using Python (Manual Calculation)

```python
import math

def binomial_probability(n, k, p):
    combination = math.comb(n, k)
    return combination * (p ** k) * ((1 - p) ** (n - k))

# Example: Probability of getting exactly 6 heads in 10 flips (p = 0.6)
n = 10
k = 6
p = 0.6

prob = binomial_probability(n, k, p)
print("P(X = 6) =", prob)
```

###  Using SciPy (Recommended)

```python
import random
import numpy as np
from scipy.stats import binom

# -----------------------------
# Example 1: Binomial (Coin Flip)
# -----------------------------

n = 10          # number of trials
p = 0.6         # probability of success (heads)

# Simulate one experiment (10 flips)
coin_flips = np.random.binomial(n, p)
print("Number of heads in 10 flips:", coin_flips)

# Probability of exactly k heads (example: k = 6)
k = 6
prob_k = binom.pmf(k, n, p)
print(f"Probability of exactly {k} heads:", prob_k)


# -----------------------------
# Example 2: Non-Binomial (Without Replacement)
# -----------------------------

def draw_two_cards_without_replacement():
    deck = ['K'] * 4 + ['N'] * 48  # K = king, N = non-king
    first = random.choice(deck)
    deck.remove(first)
    second = random.choice(deck)
    return (first == 'K') + (second == 'K')

result = draw_two_cards_without_replacement()
print("Number of kings drawn (without replacement):", result)


# -----------------------------
# Example 3: With Replacement (Binomial Again)
# -----------------------------

def draw_two_cards_with_replacement():
    kings = 0
    for _ in range(2):
        if random.random() < 4/52:
            kings += 1
    return kings

result_with_replacement = draw_two_cards_with_replacement()
print("Number of kings drawn (with replacement):", result_with_replacement)
```

```python
from scipy.stats import binom

# Parameters
n = 10   # number of trials
p = 0.6  # probability of success
k = 6    # number of successes

# Probability of exactly 6 successes
probability = binom.pmf(k, n, p)

print("P(X = 6) =", probability)
```

###  Simulating Binomial Trials

```python
import numpy as np

# Simulate number of heads in 10 flips, repeated 1000 times
n = 10
p = 0.6
samples = np.random.binomial(n, p, 1000)

print("First 10 simulated results:", samples[:10])
print("Average number of successes:", samples.mean())
```
