# Generalizing to successes in attempts

# General Binomial Probability --- Free Throw Example (Summary)

## Review of Previous Example

-   Probability of making a free throw:

    $ p = 0.7 $

-   Probability of missing:

    $ 1 - p = 0.3 $

-   For 6 attempts:

    -   Probability of exactly 2 scores:

    $ \binom{6}{2} \times 0.7^2 \times 0.3^4 $

------------------------------------------------------------------------

## Generalizing the Idea

Instead of specific numbers, define:

-   $f$ = probability of making (scoring) a free throw
-   $1 - f$ = probability of missing
-   $n$ = total number of attempts
-   $k$ = number of successful scores

------------------------------------------------------------------------

## Number of Ways to Get $k$ Scores

To choose $k$ successes out of $n$ attempts:

$ \binom{n}{k} $

This counts all possible arrangements of: - $k$ scores - $n - k$ misses

------------------------------------------------------------------------

## Probability of One Specific Arrangement

If a specific sequence has: - $k$ scores - $n - k$ misses

Its probability is:

$ f^k (1 - f)^{n-k} $

------------------------------------------------------------------------

## General Binomial Formula

The probability of exactly $k$ scores in $n$ attempts is:

$ P(X = k) = \binom{n}{k} f^k (1 - f)^{n-k} $

------------------------------------------------------------------------

## Key Insights

-   This formula applies to:
    -   Free throws
    -   Any repeated independent trials
-   Requirements:
    -   Fixed number of trials ($n$)
    -   Two outcomes (success/failure)
    -   Constant probability ($f$)
    -   Independent trials

------------------------------------------------------------------------

## Big Picture

-   The previous example (2 scores in 6 attempts) is a special case.
-   The formula above defines the **Binomial Distribution**.
-   It describes the probability distribution of the number of successes
    in $n$ independent trials.


### Python Code Example (General Binomial Distribution)

```python
import math

# Parameters
n = 6          # total attempts
f = 0.7        # probability of making (scoring)
k = 2          # number of scores we want

# Combination function
def combination(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))

# Function to compute binomial probability
def binomial_probability(n, k, f):
    return combination(n, k) * (f ** k) * ((1 - f) ** (n - k))

# Compute probability of exactly k successes
prob_k = binomial_probability(n, k, f)

print(f"Probability of exactly {k} scores in {n} attempts:")
print(f"P(X = {k}) = {prob_k:.6f}")
print(f"Approximately {prob_k * 100:.2f}%\n")

# Compute full distribution
print("Full Probability Distribution:")
print("k   P(X = k)")
print("-" * 20)

for i in range(n + 1):
    prob = binomial_probability(n, i, f)
    print(f"{i}   {prob:.6f}")
```

### What This Code Does

* Defines a reusable **binomial_probability()** function
* Computes:

    - A single probability (e.g., exactly 2 scores)
    - The full distribution for all values from 0 to 𝑛

* Uses only the **math** library (no external dependencies)
