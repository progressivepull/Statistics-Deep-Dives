## Summary: Probabilities from Density Curves

### Example 1: Uniform Density Curve

We are given a **density curve** that represents the probability distribution of a **continuous random variable** $x$.

- The variable $x$ can take on values from **1 to 5**
- All values in this interval are **equally likely**
- Therefore, the density curve is a **rectangle** with constant height

Because this is a valid density curve, the **total area under the curve is 1**.

---

#### Find the Probability That $x < 4$

There is **no probability** that $x$ is less than 1, so we only consider the interval from 1 to 4.

- Width of the interval: $4 - 1 = 3$
- Height of the rectangle: $0.25$

The probability is the **area under the curve** from 1 to 4:

$$
P(x < 4) = 0.25 \times 3 = 0.75
$$

**Final Answer:**
- $P(x < 4) = 0.75$
- This means there is a **75% probability** that $x$ is less than 4.

---

### Example 2: Normal Distribution of Heights

A group of middle school students’ heights is **normally distributed** with:

- Mean: $150$ cm
- Standard deviation: $20$ cm

Let $H$ represent the height of a randomly selected student.

---

#### Find and Interpret $P(H > 170)$

First, note that:
- $170 = 150 + 20$
- So 170 cm is **one standard deviation above the mean**

We are asked to find the **area to the right** of one standard deviation above the mean.

---

#### Using Properties of the Normal Distribution

Key facts about the normal distribution:
- About **68%** of values lie within **one standard deviation** of the mean
- About **34%** lie between the mean and one standard deviation above the mean
- **50%** of values lie below the mean

So, the area **below 170 cm** is approximately:

$$
50\% + 34\% = 84\% = 0.84
$$

Since the total area is 1, the area **above 170 cm** is:

$$
P(H > 170) = 1 - 0.84 = 0.16
$$

Using a Z-table would give a more precise value of about $0.159$, but $0.16$ is a very good approximation.

---

### Final Answers

- **Uniform distribution:**  
  $$P(x < 4) = 0.75$$

- **Normal distribution:**  
  $$P(H > 170) \approx 0.16$$

---

### Key Takeaways

- For **continuous random variables**, probabilities are found using **area under the density curve**
- Exact values have probability 0; only **intervals** matter
- The **entire area under any density curve equals 1**
- For normal distributions, knowing the **68–95–99.7 rule** allows quick probability estimates
- Area interpretation makes probability intuitive and visual


```python
import numpy as np
from scipy.stats import norm

# -----------------------------
# Example 1: Uniform distribution from 1 to 5
# -----------------------------

a, b = 1, 5  # interval
x = 4

# Uniform PDF height
height = 1 / (b - a)

# Probability P(x < 4)
prob_uniform = height * (x - a)

print("Uniform distribution example:")
print(f"P(x < 4) = {prob_uniform:.2f} ({prob_uniform*100:.0f}%)")


# -----------------------------
# Example 2: Normal distribution of heights
# -----------------------------

mean = 150      # cm
std_dev = 20    # cm
value = 170     # cm

# Probability P(H > 170)
prob_normal = 1 - norm.cdf(value, mean, std_dev)

print("\nNormal distribution example:")
print(f"P(H > 170) = {prob_normal:.4f} ({prob_normal*100:.2f}%)")
```
