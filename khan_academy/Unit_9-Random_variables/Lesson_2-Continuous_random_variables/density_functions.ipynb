## Summary: Continuous Random Variables and Probability Density Functions

We began by introducing **random variables**, and then distinguished between the two main types:

- **Discrete random variables**:  
  These take on a **finite (or countable)** number of values.  
  They are often integers, but they do **not have to be**.

- **Continuous random variables**:  
  These can take on **infinitely many values** within an interval.

---

## Example of a Continuous Random Variable

Let the random variable $Y$ be defined as:

> **The exact amount of rain tomorrow**

Because the amount of rain can take on **any real value** (for example, 1.999 inches, 2.00001 inches, etc.), $Y$ is a **continuous random variable**.

---

## Probability Density Function (PDF)

For continuous random variables, we describe probabilities using a **probability density function**, often denoted as $f(x)$.

- The horizontal axis represents the value of the random variable (e.g., inches of rain)
- The vertical axis represents the *density*, not the probability itself

---

## Probability of an Exact Value

A crucial idea for continuous random variables:

> The probability that a continuous random variable takes on **exactly one specific value** is **0**.

For example:

$$
P(Y = 2) = 0
$$

This is because:
- There are infinitely many possible values near 2
- Exact measurements to infinite precision are essentially impossible
- A single point has **no area**

---

## Probabilities Over Intervals

Probabilities for continuous random variables are found over **intervals**, not points.

For example, instead of asking $P(Y = 2)$, we ask:

$$
P(1.9 < Y < 2.1)
$$

This probability corresponds to the **area under the curve** of the PDF between 1.9 and 2.1:

$$
P(1.9 < Y < 2.1) = \int_{1.9}^{2.1} f(x)\,dx
$$

This is why **area is the key concept** for continuous distributions.

---

## Total Probability Must Equal 1

Just as with discrete random variables, the total probability must equal 1.

For a continuous random variable:

$$
\int_{0}^{\infty} f(x)\,dx = 1
$$

This means the **entire area under the probability density curve** is equal to 1.

---

## Comparison with Discrete Distributions

For a discrete random variable:
- Probabilities are assigned to individual outcomes
- The probabilities must sum to 1

Example (coin flip):

$$
P(X=0) + P(X=1) = 1
$$

You cannot assign probabilities that sum to more than 1, since that would exceed 100% likelihood.

---

## Key Takeaways

- Continuous random variables assign probability to **intervals**, not exact values
- Probability is represented by **area under a curve**
- Exact values have probability 0
- The total area under a probability density function must equal 1
- This mirrors the rule that probabilities in discrete distributions must sum to 1

In the next step, these ideas naturally lead into defining **expected value** for continuous random variables.


```python
import numpy as np
from scipy.integrate import quad

# Example probability density function (PDF)
# This is a simple bell-shaped curve (not tied to real rainfall data)
def pdf(x):
    return np.exp(-((x - 2) ** 2)) / np.sqrt(np.pi)

# 1) Probability that Y is exactly 2 (always 0 for continuous variables)
prob_exact_2 = 0.0

# 2) Probability that Y is between 1.9 and 2.1
prob_interval, _ = quad(pdf, 1.9, 2.1)

# 3) Verify total probability integrates to 1
total_prob, _ = quad(pdf, -np.inf, np.inf)

print("Continuous random variable example:")
print(f"P(Y = 2) = {prob_exact_2}")
print(f"P(1.9 < Y < 2.1) = {prob_interval:.4f}")
print(f"Total area under PDF = {total_prob:.4f}")
```

# This code demonstrates that:

* Exact values have probability **0**
* Probabilities come from **areas under the curve**
* The total area under a probability density function equals **1**
