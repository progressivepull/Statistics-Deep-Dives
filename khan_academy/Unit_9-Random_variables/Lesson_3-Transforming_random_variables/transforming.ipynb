## Summary: Mean and Standard Deviation of a Net Gain Random Variable

Anush is playing a carnival game involving **two free throws**.  
Let the random variable $X$ represent the **number of shots Anush makes** in two attempts.

This is a **discrete random variable** because it takes on only a finite (countable) set of values:

$$
X \in \{0, 1, 2\}
$$

The probability distribution of $X$ is given, along with:
- Mean: $$\mu_X = 1.2$$
- Standard deviation: $$\sigma_X = 0.69$$

---

## Defining the Net Gain Random Variable

The game costs **$15** to play, and Anush wins **$10 per shot made**.

We define a new random variable $N$ as Anush’s **net gain**:

$$
N = 10X - 15
$$

---

## Probability Distribution of $N$

Each value of $N$ corresponds directly to a value of $X$:

| $X$ | $N = 10X - 15$ | Probability |
|---|---|---|
| 0 | $-15$ | 0.16 |
| 1 | $-5$ | 0.48 |
| 2 | $+5$ | 0.36 |

---

## Mean of the Net Gain

Using the rules for transforming random variables:

- **Scaling** a random variable scales the mean
- **Shifting** a random variable shifts the mean

So the mean of $N$ is:

$$
\mu_N = 10\mu_X - 15 = 10(1.2) - 15 = -3
$$

**Interpretation:**  
On average, Anush **loses $3 per game**.

---

## Standard Deviation of the Net Gain

For standard deviation:
- **Scaling** affects standard deviation
- **Shifting does not**

So the standard deviation of $N$ is:

$$
\sigma_N = 10\sigma_X = 10(0.69) \approx 6.9
$$

---

## Final Results

- **Mean net gain:** $$\mu_N = -3$$
- **Standard deviation of net gain:** $$\sigma_N \approx 6.9$$

---

## Key Takeaways

- Linear transformations of random variables make calculations much easier
- Adding or subtracting a constant shifts the mean but does **not** change the spread
- Multiplying by a constant scales **both** the mean and the standard deviation
- Even with a negative expected value, outcomes can still vary widely

This example neatly combines **probability distributions**, **expected value**, **standard deviation**, and **linear transformations of random variables**.


```python
import math

# Given probability distribution for X = number of shots made
# Format: {value: probability}
X_distribution = {
    0: 0.16,
    1: 0.48,
    2: 0.36
}

# Given summary statistics for X
mean_X = 1.2
std_X = 0.69

# Define net gain random variable N = 10X - 15
def net_gain(x):
    return 10 * x - 15

# Build probability distribution for N
N_distribution = {
    net_gain(x): p for x, p in X_distribution.items()
}

# Compute mean of N directly from the distribution
mean_N = sum(n * p for n, p in N_distribution.items())

# Compute variance and standard deviation of N directly
variance_N = sum(((n - mean_N) ** 2) * p for n, p in N_distribution.items())
std_N = math.sqrt(variance_N)

# Compute mean and standard deviation using transformation rules
mean_N_rule = 10 * mean_X - 15
std_N_rule = 10 * std_X

print("Probability distribution for net gain N:")
for n in sorted(N_distribution):
    print(f"P(N = {n}) = {N_distribution[n]}")

print("\nComputed from distribution:")
print(f"Mean of N: {mean_N}")
print(f"Standard deviation of N: {std_N:.2f}")

print("\nComputed using transformation rules:")
print(f"Mean of N: {mean_N_rule}")
print(f"Standard deviation of N: {std_N_rule:.2f}")
```

# What this demonstrates:

* How to construct a new probability distribution from a transformed random variable
* How to compute mean and standard deviation directly from probabilities
* Verification that transformation rules for mean and standard deviation work correctly
