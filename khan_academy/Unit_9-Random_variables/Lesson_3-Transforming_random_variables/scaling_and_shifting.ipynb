## Summary: How Adding and Scaling Affect Random Variables

Let $X$ be a random variable, such as the **height of a randomly selected person**.  
Assume $X$ has some probability distribution (shown as a bell curve), with:
- a **mean**
- a **standard deviation** (one unit above and below the mean)

We want to understand how the **mean** and **standard deviation** change when we **add a constant** or **scale** the random variable.

---

## Adding a Constant to a Random Variable

Define a new random variable:

$$
Y = X + k
$$

where $k$ is a **constant** (not a random variable).

### What Happens Visually?
- The entire distribution **shifts horizontally** by $k$
- The shape and spread of the distribution **do not change**

If $k > 0$, the distribution shifts to the right.  
If $k < 0$, it shifts to the left.

---

### Effect on the Mean

Adding a constant increases the mean by that constant:

$$
\mu_Y = \mu_X + k
$$

---

### Effect on the Standard Deviation

The standard deviation measures **spread around the mean**, and that spread does not change when we shift everything equally:

$$
\sigma_Y = \sigma_X
$$

---

### Key Result (Addition)

- Mean: **changes by $k$**
- Standard deviation: **unchanged**

---

## Scaling a Random Variable

Now define another random variable:

$$
Z = kX
$$

where $k$ is a constant.

---

### What Happens Visually?
- The distribution is **stretched or compressed** by a factor of $k$
- To keep total probability equal to 1, the curve becomes **wider and shorter** (or narrower and taller)

---

### Effect on the Mean

Scaling multiplies the mean by $k$:

$$
\mu_Z = k\mu_X
$$

---

### Effect on the Standard Deviation

The spread is also scaled by the same factor:

$$
\sigma_Z = k\sigma_X
$$

---

### Key Result (Scaling)

- Mean: **multiplied by $k$**
- Standard deviation: **multiplied by $k$**

---

## Big Takeaways

- **Adding a constant**:
  - Shifts the distribution
  - Changes the mean
  - Does **not** change the standard deviation

- **Multiplying by a constant**:
  - Stretches or compresses the distribution
  - Changes both the mean and the standard deviation

These rules apply to **any random variable**, regardless of its distribution shape, and are fundamental tools in probability and statistics.


```python
import numpy as np

# Simulate a random variable X (e.g., heights)
np.random.seed(0)
X = np.random.normal(loc=170, scale=10, size=10000)  # mean=170, std=10

# Constants
k_add = 10     # adding constant
k_scale = 2    # scaling constant

# New random variables
Y = X + k_add      # add a constant
Z = k_scale * X   # scale the random variable

# Compute statistics
mean_X = np.mean(X)
std_X = np.std(X)

mean_Y = np.mean(Y)
std_Y = np.std(Y)

mean_Z = np.mean(Z)
std_Z = np.std(Z)

print("Original random variable X:")
print(f"Mean: {mean_X:.2f}, Standard Deviation: {std_X:.2f}")

print("\nAfter adding a constant (Y = X + k):")
print(f"Mean: {mean_Y:.2f}, Standard Deviation: {std_Y:.2f}")

print("\nAfter scaling (Z = kX):")
print(f"Mean: {mean_Z:.2f}, Standard Deviation: {std_Z:.2f}")
```

# What this shows:

* Adding a constant shifts the **mean** but leaves the **standard deviation unchanged**
* Scaling multiplies **both** the mean and the standard deviation by the same factor
