**🔹 Bullet Summary**

- We want to model the **number of cars passing in an hour** at a fixed
  point on a road.

- Define a random variable:

$$
X = \text{number of cars passing in one hour}
$$

- After observing traffic for many hours, we estimate:

$$
E\lbrack X\rbrack = \lambda
$$

(e.g., $\lambda = 9.3$cars/hour).

- First attempt: model using a **binomial distribution**:

  - Divide the hour into $n$intervals.

  - Probability of a car in each interval:

$$
p = \frac{\lambda}{n}
$$

- Then:

$$
X \sim \text{Binomial}(n,p)
$$

- Problem: binomial trials allow **only one car per interval**, but real
  traffic can have multiple cars in tiny intervals.

- Solution: let the number of intervals $n \rightarrow \infty$.

- Using limits, factorial identities, and the fact that:

$$
\left( 1+\frac{a}{n} \right)^{n} \rightarrow e^{a},
$$

the binomial distribution converges to:

$$
P(X = k) = \frac{e^{- \lambda}\lambda^{k}}{k!}
$$

- This is the **Poisson distribution**, ideal for modeling:

  - independent arrivals

  - constant average rate

  - rare events in small intervals

- Example:

$$
P(X = 2) = \frac{e^{- 9} \cdot 9^{2}}{2!}
$$

**🔹 Explanation (Intuitive)**

**1. Start with the binomial idea**

If you divide the hour into $n$tiny intervals, each interval has a
small chance of containing a car.

- More intervals → smaller probability per interval.

- Expected number of cars stays fixed at $\lambda$.

**2. Why binomial fails**

Even with seconds or milliseconds, multiple cars can arrive in the same
tiny interval.

**3. The limiting idea**

As $n$becomes extremely large:

- Probability of a car in each interval becomes tiny.

- But the total expected number of cars stays $\lambda$.

- The binomial formula simplifies beautifully into the Poisson formula.

**4. Why Poisson works**

It naturally handles:

- multiple arrivals in tiny intervals

- independence

- constant average rate

This is why Poisson is used for traffic, call centers, radioactive
decay, email arrivals, etc.


# 🔹 Python Code Example

This code:

* Computes the Poisson probability directly.
* Shows how the binomial approximation converges to Poisson as 𝑛 increases.


```python
import numpy as np
from scipy.stats import binom, poisson
import matplotlib.pyplot as plt

# Average number of cars per hour
lam = 9.3

# Values of k to compare
k_values = np.arange(0, 25)

# Binomial approximations with increasing n
ns = [60, 3600, 36000]

plt.figure(figsize=(10,6))

for n in ns:
    p = lam / n
    probs = binom.pmf(k_values, n, p)
    plt.plot(k_values, probs, marker='o', label=f"Binomial n={n}")

# True Poisson distribution
poisson_probs = poisson.pmf(k_values, lam)
plt.plot(k_values, poisson_probs, marker='x', linestyle='--', color='black', label="Poisson")

plt.xlabel("k (number of cars)")
plt.ylabel("Probability")
plt.title("Binomial(n, λ/n) Converging to Poisson(λ)")
plt.legend()
plt.show()
```

**What you’ll see**

- For small $n$, the binomial curve is rough.

- As $`n`$increases, the binomial distribution **smoothly approaches**
  the Poisson curve.

- The Poisson distribution emerges naturally as the limit.
