## Summary: Discrete vs. Continuous Random Variables

Random variables come in **two main types**:  
**discrete** and **continuous**. Understanding the difference is essential for probability and statistics.

---

## Discrete Random Variables

A **discrete random variable** can take on **distinct, separate, and countable values**.

- The values may be **finite or infinite**
- The key requirement is that the values can be **listed or counted**

### Examples

#### 1. Coin Flip
Let $X = 1$ if a coin lands heads, and $X = 0$ if it lands tails.

- Possible values: {0, 1}
- Clearly countable  
✅ **Discrete**

---

#### 2. Year of Birth
Let $Y$ be the year a randomly selected student was born.

- Possible values: 1995, 1996, 1997, ...
- You cannot be born in between years  
✅ **Discrete**

---

#### 3. Number of Ants Born Tomorrow
Let $Z$ be the number of ants born in the universe tomorrow.

- Possible values: 0, 1, 2, 3, …
- Even though the number could be extremely large, it is countable  
✅ **Discrete**

---

## Continuous Random Variables

A **continuous random variable** can take on **any value within an interval**, including infinitely many values between any two points.

- Values are **uncountable**
- Typically represent measurements

### Examples

#### 1. Mass of a Random Animal
Let $Y$ be the **exact mass** of a randomly selected animal at a zoo.

- Can take any real value within a range
- Infinitely many possible values between any two masses  
✅ **Continuous**

---

#### 2. Exact Winning Time in a Race
Let $X$ be the **exact winning time** for the men’s 100-meter dash at the 2016 Olympics.

- Even though official clocks round to hundredths, the true time is infinitely precise
- Could be 9.571, 9.571238, etc.  
✅ **Continuous**

---

## Rounding Can Change the Type

If we redefine the previous variable as:

> The **winning time rounded to the nearest hundredth**

- Possible values: 9.56, 9.57, 9.58, …
- These values are countable  
✅ **Discrete**

---

## Key Distinction

| Feature | Discrete | Continuous |
|------|--------|------------|
| Values | Separate, countable | Any value in an interval |
| Can list values? | Yes | No |
| Infinite values possible? | Yes (countable) | Yes (uncountable) |
| Examples | Coin flips, counts, years | Time, mass, distance |

---

## Key Takeaways
- Discrete random variables count outcomes
- Continuous random variables measure outcomes
- Rounding a continuous measurement can turn it into a discrete variable
- The distinction depends on **how the variable is defined**, not the situation itself

This distinction lays the groundwork for understanding probability distributions in future lessons.


```python
import random

# -----------------------------
# Discrete random variable examples
# -----------------------------

# Example 1: Coin flip random variable
# X = 1 if heads, 0 if tails
def coin_flip():
    outcome = random.choice(["heads", "tails"])
    X = 1 if outcome == "heads" else 0
    return outcome, X


# Example 2: Year of birth (discrete)
def random_birth_year(start=1980, end=2010):
    return random.randint(start, end)


# Example 3: Number of ants born tomorrow (discrete count)
def ants_born():
    return random.randint(0, 10_000_000)


# -----------------------------
# Continuous random variable examples
# -----------------------------

# Example 4: Mass of a random animal (continuous)
def animal_mass():
    # Mass in kilograms (continuous range)
    return random.uniform(0.001, 5000.0)


# Example 5: Exact winning time of a race (continuous)
def exact_race_time():
    # Time in seconds
    return random.uniform(9.0, 12.0)


# Example 6: Rounded race time (discrete)
def rounded_race_time():
    return round(exact_race_time(), 2)


# -----------------------------
# Demonstration
# -----------------------------

print("Discrete random variables:")
print("Coin flip:", coin_flip())
print("Birth year:", random_birth_year())
print("Ants born:", ants_born())

print("\nContinuous random variables:")
print("Animal mass (kg):", animal_mass())
print("Exact race time (s):", exact_race_time())

print("\nDiscrete version after rounding:")
print("Rounded race time (s):", rounded_race_time())
```
