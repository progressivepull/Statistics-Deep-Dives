## Summary: Estimating Probabilities from Observed Data

This example shows how **random variables and probability distributions** can be estimated using **real-world observations**.

---

### Scenario
- You visit a frozen yogurt store **every school day at 4:00 PM**
- You observe the **size of the line** on **50 visits**
- Line size is treated as a **discrete random variable**

---

### Observed Data

| Line Size | Times Observed |
|----------|----------------|
| 0 people | 24 |
| 1 person | 18 |
| 2 people | 8 |
| **Total** | **50** |

These are the only outcomes observed in the study.

---

### Estimating Probabilities (Relative Frequencies)

Since the true probabilities are unknown, we estimate them using **relative frequency**:

$$
\text{Estimated Probability} = \frac{\text{Number of times observed}}{\text{Total observations}}
$$

| Line Size | Probability Estimate |
|----------|---------------------|
| 0 people | $ \frac{24}{50} = 0.48 $ (48%) |
| 1 person | $ \frac{18}{50} = 0.36 $ (36%) |
| 2 people | $ \frac{8}{50} = 0.16 $ (16%) |

These estimates:
- Are between 0 and 1
- Sum to 1
- Form a **valid discrete probability distribution**

---

### Using the Distribution for Prediction

Suppose you plan to visit the store **500 more times**.

To estimate how many times you will see a **2-person line**:

$$
\text{Expected count} = 500 \times 0.16 = 80
$$

So you would **expect about 80 visits** with a two-person line.

---

### Important Caveats
- These probabilities are **estimates**, not exact truths
- Observed data can vary due to randomness
- Even with accurate probabilities, actual outcomes may differ (e.g., 65 or 85 times instead of exactly 80)
- Larger sample sizes generally give better estimates

---

### Key Takeaways
- Probabilities can be estimated using **empirical data**
- Relative frequency is a natural way to build a probability distribution
- Expected values describe long-run averages, not guaranteed outcomes
- This approach connects probability theory to real-world decision-making


```python
# Observed data from the study
line_sizes = [0, 1, 2]
times_observed = [24, 18, 8]
total_observations = sum(times_observed)

# Estimate probabilities using relative frequencies
probabilities = {
    size: count / total_observations
    for size, count in zip(line_sizes, times_observed)
}

print("Estimated probability distribution:")
for size, prob in probabilities.items():
    print(f"P(line size = {size}) = {prob:.2f}")

# Predict expected counts over future visits
future_visits = 500
expected_counts = {
    size: prob * future_visits
    for size, prob in probabilities.items()
}

print("\nExpected number of visits (out of 500):")
for size, expected in expected_counts.items():
    print(f"Line size {size}: {expected:.0f} times")
```
