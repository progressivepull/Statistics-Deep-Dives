## Summary: Expected Value of a Discrete Random Variable

We define the random variable **$X$** as the **number of workouts done in a given week**.  
The probability distribution for $X$ is given by a table with the following possible values:

$$
X \in \{0, 1, 2, 3, 4\}
$$

Because $X$ can take on only a **finite set of values**, it is a **discrete random variable**.

---

### Validity of the Probability Distribution
The distribution is valid because:
- All probabilities are **non-negative**
- The probabilities sum to **1**:
$$
0.10 + 0.15 + 0.40 + 0.25 + 0.10 = 1
$$

---

### Expected Value (Mean) of a Discrete Random Variable

The **expected value** (also called the **mean**) of a discrete random variable is denoted by the Greek letter **$\mu$** and is calculated as a **weighted average**:

$$
\mu = E[X] = \sum x \cdot P(X = x)
$$

---

### Calculating the Expected Value

$$
\begin{aligned}
E[X] &= 0(0.10) + 1(0.15) + 2(0.40) + 3(0.25) + 4(0.10) \\
&= 0 + 0.15 + 0.80 + 0.75 + 0.40 \\
&= 2.10
\end{aligned}
$$

---

### Interpretation

- The **expected number of workouts per week** is **2.1**
- This does **not** mean you will literally do 2.1 workouts in a single week
- Instead, it represents a **long-term average**

For example:
- Over **10 weeks**, you would expect about **21 workouts**
- Over **100 weeks**, you would expect about **210 workouts**

---

### Key Takeaways
- Expected value is a **long-run average**, not a guaranteed outcome
- Discrete random variables can have **non-integer expected values**
- Expected value is extremely useful for understanding overall trends and planning
- Even when outcomes are integers, the average behavior may not be

This concept forms the foundation for more advanced topics such as variance, standard deviation, and distributions.


```python
# Probability distribution for X = number of workouts in a week
# Format: {value: probability}
distribution = {
    0: 0.10,
    1: 0.15,
    2: 0.40,
    3: 0.25,
    4: 0.10
}

# Compute expected value E[X]
expected_value = sum(x * p for x, p in distribution.items())

print("Probability distribution:")
for x in sorted(distribution):
    print(f"P(X = {x}) = {distribution[x]}")

print(f"\nExpected value (mean) of X: {expected_value}")

# Interpretation example
weeks = 10
print(f"Expected workouts over {weeks} weeks: {expected_value * weeks}")
```
