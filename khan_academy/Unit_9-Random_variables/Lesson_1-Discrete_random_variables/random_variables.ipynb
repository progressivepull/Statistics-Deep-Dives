## Summary: Understanding Random Variables

This lesson introduces the concept of a **random variable** and explains how it differs from traditional variables used in algebra.

---

### What Is a Random Variable?
A **random variable** is a way to **assign numbers to the outcomes of a random process**.  
Instead of solving for a fixed unknown, a random variable **maps possible outcomes to numerical values**.

Random variables are typically denoted using **capital letters** (e.g., $X, Y$).

---

### Examples of Random Variables

#### Example 1: Coin Flip
Define a random variable $X$:

- $X = 1$ if the coin lands **heads**
- $X = 0$ if the coin lands **tails**

This converts the random outcome of a coin flip into a numerical value.  
Other numerical assignments would also be valid, but this is a common and intuitive choice.

---

#### Example 2: Rolling Dice
Define a random variable $Y$:

- $Y =$ the **sum of the upward-facing values** after rolling **7 dice**

Here, the random process is rolling dice, and the random variable quantifies the outcome by summing the results.

---

### Why Random Variables Are Useful
Once outcomes are expressed numerically, we can:
- Use **cleaner mathematical notation**
- Apply **probability and statistical tools**
- Perform calculations and comparisons more easily

For example:
- Instead of writing:  
  “the probability that the sum of the upward faces after rolling seven dice is less than or equal to 30,”  
  we can simply write:
  $$
  P(Y \le 30)
  $$

- Instead of asking whether the sum is even, we can write:
  $$
  P(Y \text{ is even})
  $$

---

### Random Variables vs. Algebra Variables

| Algebra Variables | Random Variables |
|------------------|------------------|
| Usually lowercase (e.g., $x, y$) | Usually uppercase (e.g., $X, Y$) |
| Represent unknowns or inputs | Represent outcomes of random processes |
| Can be solved for or assigned | Cannot be fixed to one value |
| Take one value at a time | Take many values with different probabilities |

A random variable is **not solved** like $x + 5 = 6$.  
Instead, we talk about:
- $P(X = x)$
- $P(X \le x)$
- Properties of the distribution of $X$

---

### Key Takeaways
- Random variables map random outcomes to numbers
- They enable concise notation and deeper mathematical analysis
- Unlike algebra variables, they represent **distributions**, not single values
- Understanding random variables is foundational for probability and statistics

The next step is to explore **different types of random variables** and how their probabilities are described.


```python
import random

# Example 1: Random variable for a coin flip
# X = 1 if heads, 0 if tails
def coin_flip_random_variable():
    flip = random.choice(["heads", "tails"])
    X = 1 if flip == "heads" else 0
    return flip, X


# Example 2: Random variable for the sum of 7 dice
def dice_sum_random_variable(num_dice=7):
    rolls = [random.randint(1, 6) for _ in range(num_dice)]
    Y = sum(rolls)
    return rolls, Y


# Demonstration
coin_result, X = coin_flip_random_variable()
print(f"Coin flip result: {coin_result}, Random variable X = {X}")

dice_rolls, Y = dice_sum_random_variable()
print(f"Dice rolls: {dice_rolls}")
print(f"Sum of dice (Random variable Y): {Y}")

# Example probability-style checks
print(f"Is Y <= 30? {'Yes' if Y <= 30 else 'No'}")
print(f"Is Y even? {'Yes' if Y % 2 == 0 else 'No'}")
```
