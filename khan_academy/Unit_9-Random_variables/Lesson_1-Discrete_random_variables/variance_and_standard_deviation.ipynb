## Summary: Variance and Standard Deviation of a Discrete Random Variable

We continue with the discrete random variable **$X$**, defined as the **number of workouts in a week**, with an expected value (mean) of:

$$
\mu = E[X] = 2.1
$$

Now we extend this idea beyond central tendency to **measure spread**, using **variance** and **standard deviation**.

---

## Variance of a Discrete Random Variable

The **variance** measures how spread out the values of a random variable are around the mean.

For a discrete random variable, variance is calculated as:

$$
\text{Var}(X) = \sum (x - \mu)^2 \cdot P(X = x)
$$

That is:
1. Subtract the mean from each outcome  
2. Square the difference  
3. Multiply by the probability of that outcome  
4. Add everything together  

---

### Applying the Formula

With mean $\mu = 2.1$:

$$
\begin{aligned}
\text{Var}(X) &= (0 - 2.1)^2(0.10) \\
&+ (1 - 2.1)^2(0.15) \\
&+ (2 - 2.1)^2(0.40) \\
&+ (3 - 2.1)^2(0.25) \\
&+ (4 - 2.1)^2(0.10)
\end{aligned}
$$

Evaluating each term:
- $2.1^2 \cdot 0.10$
- $1.1^2 \cdot 0.15$
- $0.1^2 \cdot 0.40$
- $0.9^2 \cdot 0.25$
- $1.9^2 \cdot 0.10$

Adding them all together gives:

$$
\text{Var}(X) = 1.19
$$

---

## Standard Deviation

The **standard deviation** is the square root of the variance and is denoted by the Greek letter **$\sigma$**:

$$
\sigma = \sqrt{\text{Var}(X)} = \sqrt{1.19} \approx 1.09
$$

---

## Interpretation

- **Mean:** $2.1$ workouts per week  
- **Standard deviation:** approximately **1.09 workouts**

This means:
- Most outcomes lie within about **1 workout** of the mean
- One standard deviation above the mean is about **3.2**
- One standard deviation below the mean is about **1.0**

Even though the random variable only takes **integer values**, both the **mean** and **standard deviation** can be **non-integer**.

---

## Visual Intuition

- The probability distribution is centered around **2**
- The mean at **2.1** reflects the weighted center
- The standard deviation of **1.09** reasonably captures how spread out the outcomes are

---

## Key Takeaways
- **Variance** measures average squared distance from the mean
- **Standard deviation** gives spread in the same units as the data
- These concepts extend naturally from datasets to **random variables**
- Together, mean and standard deviation describe both **center** and **spread** of a distribution


```python
import math

# Probability distribution for X = number of workouts in a week
# Format: {value: probability}
distribution = {
    0: 0.10,
    1: 0.15,
    2: 0.40,
    3: 0.25,
    4: 0.10
}

# Compute mean (expected value)
mean = sum(x * p for x, p in distribution.items())

# Compute variance
variance = sum(((x - mean) ** 2) * p for x, p in distribution.items())

# Compute standard deviation
std_dev = math.sqrt(variance)

print("Probability distribution:")
for x in sorted(distribution):
    print(f"P(X = {x}) = {distribution[x]}")

print(f"\nMean (μ): {mean}")
print(f"Variance (σ²): {variance}")
print(f"Standard Deviation (σ): {std_dev:.2f}")
```
