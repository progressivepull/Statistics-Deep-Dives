Hugo plans to buy packs of baseball cards until he gets the card of his favorite player, but he only has enough money to buy at most 4 packs. Suppose that each pack has probability 0.2 of containing the card Hugo is hoping for.

Let the random variable X be the number of packs of cards Hugo buys. Here is the probability distribution for X:

|  X = # of Packs |    1   |    2   |    3   |   4   |
| --------------- | ------ | ------ | ------ | ----- |
|  P(X)           |  0.2   | 0.16   | 0.128  |   ?   |

Find the indicated probability.

**P( X $\ge$ 2 ) =** ________________ 



## Summary: Probability Distribution for Buying Baseball Card Packs

Hugo buys packs of baseball cards **until he gets the card of his favorite player**, but he can buy **at most 4 packs**.  
Each pack has a **0.2 probability** of containing the card he wants.

Let the random variable **$X$** be the **number of packs Hugo buys**.

---

### Understanding the Given Probability Distribution

- **$P(X = 1) = 0.2$**  
  Hugo gets the card in the first pack and stops.

- **$P(X = 2) = 0.16$**  
  He does not get the card in the first pack (0.8), then gets it in the second (0.2):  
  $$
  0.8 \times 0.2 = 0.16
  $$

- **$P(X = 3) = 0.128$**  
  He misses twice, then succeeds on the third pack.

- **$P(X = 4)$** is initially unknown, but Hugo must stop at four packs whether or not he gets the card.

---

### Finding the Missing Probability

All probabilities in a valid probability distribution must sum to 1:

$$
P(X = 4) = 1 - (0.2 + 0.16 + 0.128) = 1 - 0.488 = 0.512
$$

So there is a **51.2% probability** that Hugo buys all four packs.

---

### Question: What Is $P(X \ge 2)$?

This means Hugo buys **two or more packs**:

$$
P(X \ge 2) = P(X = 2) + P(X = 3) + P(X = 4)
$$

Instead of adding them directly, we can use the complement:

$$
P(X \ge 2) = 1 - P(X = 1) = 1 - 0.2 = 0.8
$$

---

### Final Answers

- **$P(X \ge 2) = 0.8$**
- **$P(X = 4) = 0.512$**

---

### Key Takeaways
- This is a **discrete probability distribution**
- Probabilities must be **non-negative** and **sum to 1**
- A high probability of buying all 4 packs makes sense because Hugo is forced to stop there even if he never gets the card
- Complementary probabilities are often the fastest way to answer “at least” questions


```python
# Probability distribution for X = number of packs Hugo buys
probabilities = {
    1: 0.2,
    2: 0.16,
    3: 0.128
}

# Compute missing probability for X = 4
probabilities[4] = 1 - sum(probabilities.values())

# Probability that X >= 2
prob_x_ge_2 = sum(probabilities[x] for x in probabilities if x >= 2)

print("Probability distribution for X:")
for x in sorted(probabilities):
    print(f"P(X = {x}) = {probabilities[x]:.3f}")

print(f"\nP(X ≥ 2) = {prob_x_ge_2:.3f}")
```
