## Summary: Valid Probability Models

### Example 1: Free Throw Probability Model

Anthony DeNoon analyzes the outcomes of his **next two free throws** with the following probabilities:

| Outcome | Probability |
|------|------------|
| Miss both free throws | 0.2 |
| Make exactly one free throw | 0.5 |
| Make both free throws | 0.1 |

#### Is this a valid probability model?

A probability model is **valid** if:
1. **All probabilities are non-negative**
2. **The probabilities sum to 1 (100%)**

- All probabilities are non-negative ✔️
- Sum of probabilities:
  $$
  0.2 + 0.5 + 0.1 = 0.8
  $$

Since the probabilities add up to **0.8 (80%)**, not 1, **this is NOT a valid probability model**.

---

### Example 2: Alien Sampling Earth Creatures

A space alien abducts:
- **97 chickens**
- **47 cows**
- **77 humans**

Total creatures:
$$
97 + 47 + 77 = 221
$$

Each creature is equally likely to be selected.

---

### Constructing the Probability Model

The probability of selecting each type is the fraction of that type in the total sample:

| Creature | Probability |
|--------|------------|
| Chicken | $ \frac{97}{221} $ |
| Cow | $ \frac{47}{221} $ |
| Human | $ \frac{77}{221} $ |

#### Is this a valid probability model?

- All probabilities are non-negative ✔️
- Sum of probabilities:
  $$
  \frac{97 + 47 + 77}{221} = \frac{221}{221} = 1
  $$

✅ **This is a valid probability model**

---

### Key Takeaways
- A valid probability model must:
  - Have probabilities between 0 and 1
  - Sum exactly to 1
- Relative frequencies naturally form probability models when outcomes are equally likely
- Checking validity is always the first step when working with probability tables


```python
from fractions import Fraction

# -----------------------------
# Example 1: Check validity of a probability model
# -----------------------------

free_throw_model = {
    "Miss both free throws": 0.2,
    "Make exactly one free throw": 0.5,
    "Make both free throws": 0.1
}

total_probability = sum(free_throw_model.values())

print("Free throw probability model:")
for outcome, prob in free_throw_model.items():
    print(f"{outcome}: {prob}")

print(f"\nTotal probability: {total_probability}")

if total_probability == 1 and all(p >= 0 for p in free_throw_model.values()):
    print("This IS a valid probability model.")
else:
    print("This is NOT a valid probability model.")


# -----------------------------
# Example 2: Construct a valid probability model
# -----------------------------

chickens = 97
cows = 47
humans = 77
total_creatures = chickens + cows + humans

probability_model = {
    "Chicken": Fraction(chickens, total_creatures),
    "Cow": Fraction(cows, total_creatures),
    "Human": Fraction(humans, total_creatures)
}

print("\nAlien probability model (fractions):")
for creature, prob in probability_model.items():
    print(f"{creature}: {prob}")

print("\nAlien probability model (decimals):")
for creature, prob in probability_model.items():
    print(f"{creature}: {float(prob):.2f}")

print(f"\nTotal probability: {sum(probability_model.values())}")
print("This IS a valid probability model.")
```
