## Summary: Probability Distribution of a Discrete Random Variable

We define a **random variable**:

$$
X = \text{number of heads obtained when flipping a fair coin three times}
$$

The goal is to determine the **probability distribution** of $X$, meaning the probability that $X$ takes on each of its possible values.

---

### Step 1: List All Possible Outcomes
Flipping a fair coin three times produces:

- HHH  
- HHT  
- HTH  
- HTT  
- THH  
- THT  
- TTH  
- TTT  

There are **8 equally likely outcomes**.

---

### Step 2: Map Outcomes to Values of $X$

| Outcome | Number of Heads | Value of $X$ |
|------|----------------|---------------|
| TTT | 0 | 0 |
| HTT, THT, TTH | 1 | 1 |
| HHT, HTH, THH | 2 | 2 |
| HHH | 3 | 3 |

---

### Step 3: Compute Probabilities

- $P(X = 0) = \frac{1}{8}$
- $P(X = 1) = \frac{3}{8}$
- $P(X = 2) = \frac{3}{8}$
- $P(X = 3) = \frac{1}{8}$

These probabilities sum to 1, as required.

---

### Step 4: Probability Distribution

This information defines the **probability distribution** of $X$:

| $X$ | 0 | 1 | 2 | 3 |
|---|---|---|---|---|
| $P(X)$ | 1/8 | 3/8 | 3/8 | 1/8 |

When plotted, this distribution appears as a **bar chart**, with the height of each bar representing the probability of each value.

---

### Key Takeaways
- A **probability distribution** assigns probabilities to all possible values of a random variable.
- Because $X$ can only take on whole numbers (0, 1, 2, 3), this is a **discrete probability distribution**.
- Discrete random variables cannot take values between these points (e.g., 1.5 or π).
- This example follows a **binomial distribution** with $n = 3$ and $p = 0.5$.

This forms the foundation for studying probability distributions more generally.


```python
import itertools
from collections import Counter

def probability_distribution_three_flips():
    """
    Computes the probability distribution of
    X = number of heads in 3 fair coin flips.
    """
    outcomes = list(itertools.product(["H", "T"], repeat=3))
    total_outcomes = len(outcomes)

    # Count number of heads in each outcome
    head_counts = [outcome.count("H") for outcome in outcomes]

    # Count frequencies of each value of X
    frequency = Counter(head_counts)

    # Convert frequencies to probabilities
    distribution = {
        x: freq / total_outcomes
        for x, freq in sorted(frequency.items())
    }

    return distribution


# Compute and display the distribution
dist = probability_distribution_three_flips()

print("Probability distribution for X = number of heads in 3 coin flips:")
for x, p in dist.items():
    print(f"P(X = {x}) = {p} ({p:.3f})")
```
