A significance test is going to be performed using a significance level
of α = 0.05. Suppose that the null hypothesis is actually false.

If the significance level was lowered to α = 0.01, which of the
following would be true?

Choose 1 answer:

A. Both the probability of a Type II error and the power of the test
would decrease.

B. Both the probability of a Type II error and the power of the test
would increase.

C. The probability of a Type II error would increase and the power of
the test would decrease.

D. The probability of a Type II error would decrease and the power of
the test would increase.


Here’s a clean **bullet summary**, a clear **explanation**, and a
**Python code example** to make the ideas of significance level,
Type%20II%20error, and power feel intuitive and concrete.

**Bullet Summary**

**Part 1 — Lowering α from 0.05 to 0.01**

- Lowering the significance level makes the rejection region
  **smaller**.

- Smaller rejection region → harder to reject the null hypothesis.

- Therefore:

  - **Probability of Type II error increases.**

  - **Power decreases.**

- Correct choice:

  - **Type II error ↑**

  - **Power ↓**

**Part 2 — Asha’s Coffee Machine Test**

- Hypotheses:

  - $H_{0}:p = 0.30$

  - $H_{a}:p > 0.30$

- Power increases when:

  - Sample size $n$is **larger**.

  - True proportion $p$is **farther** from 0.30.

- Highest power occurs when:

  - **Largest sample size** (n = 200)

  - **True proportion farthest from 0.30** (p = 0.50)

- Correct choice: **n = 200, true p = 0.50**

**Explanation**

Lowering the significance level from 5% to 1% makes the test more
conservative. The rejection region shrinks, so you need stronger
evidence to reject the null hypothesis.

That means:

- You are **less likely** to reject $H_{0}$even when it is false.

- This increases the chance of a Type II error.

- Since power is

$$
1 - P(\text{Type II error}),
$$

power must **decrease**.

For Asha’s scenario, power is highest when the sampling distributions
under the null and alternative are **most separated** and **least spread
out**. That happens when:

- $n$is large → narrower sampling distribution.

- True $p$is far from 0.30 → distributions separate more.

Thus, **n = 200 and true p = 0.50** gives the strongest ability to
detect the effect.

**Python Code Example — Simulating Power at Different α Levels**

This simulation shows how lowering α reduces power for a one‑sided test
of a proportion.




```python
import numpy as np
from scipy.stats import binomtest

# True and null proportions
p_null = 0.30
p_true = 0.40   # true proportion under the alternative
n = 100         # sample size
trials = 20000  # number of simulations

def estimate_power(alpha):
    rejections = 0
    for _ in range(trials):
        sample = np.random.binomial(n, p_true)
        p_value = binomtest(sample, n, p_null, alternative='greater').pvalue
        if p_value < alpha:
            rejections += 1
    return rejections / trials

power_05 = estimate_power(0.05)
power_01 = estimate_power(0.01)

print("Power at alpha = 0.05:", power_05)
print("Power at alpha = 0.01:", power_01)
print("Difference:", power_05 - power_01)
```

**What this code shows**

- Power at α = 0.05 will be **higher**.

- Power at α = 0.01 will be **lower**.

- The difference quantifies how much harder it becomes to detect the
  effect when α is reduced.
