# Type I errors

[Type 1 errors | Inferential statistics | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=EowIec7Y8HM&t=19s)

**Bullet summary**

- A **Type I error** occurs when you **reject the null hypothesis even
  though it is actually true**.

- In significance testing, we **assume the null hypothesis** is true and
  compute how likely our sample statistic would be under that
  assumption.

- The **p‑value** measures the probability of getting a statistic as
  extreme or more extreme than the observed one *if the null is true*.

- If the p‑value is below the **significance level** $\alpha$, we
  reject the null.

- A Type I error happens when the null is true, but our sample falls in
  the “extreme” region anyway.

- The probability of a Type I error is exactly $\alpha$.

- Example: If $\alpha = 0.01$, then even when the null is true, 1% of
  samples will fall in the rejection region.

- Rejecting the null in those rare cases is the Type I error.

**Explanation**

In hypothesis testing, we start by assuming the null hypothesis
$H_{0}$is true. We then compute a test statistic (like a sample mean
or proportion) and ask:

$$
P(\text{statistic as extreme as observed} \mid H_{0}\text{ is true})
$$

This probability is the p‑value. If it is smaller than our chosen
significance level $\alpha$, we reject $H_{0}$.

But even if the null is truly correct, random sampling can still produce
extreme results. If such a rare sample occurs and we reject the null, we
have committed a **Type I error**.

Graphically, imagine the sampling distribution under the null. The
rejection region is the tail area whose probability equals $\alpha$.
If the true mean is actually the null value, then $\alpha$fraction of
samples will fall in that tail by chance. Rejecting the null in those
cases is the Type I error.

So:

- **Null true + reject null = Type I error**

- **Probability of Type I error =** $\alpha$

This is why choosing $\alpha$matters: it directly controls how often
you are willing to risk falsely rejecting a true null.

**Python code example (simulating Type I errors)**

This simulation shows that when the null hypothesis is true, the
proportion of rejections is approximately $\alpha$.




```python
import numpy as np
from statsmodels.stats.weightstats import ztest

np.random.seed(42)

# True population mean under the null
mu_null = 50

# Population standard deviation (assumed known for z-test)
sigma = 10

# Sample size
n = 40

# Significance level
alpha = 0.01

# Number of simulations
num_sims = 5000

type1_count = 0

for _ in range(num_sims):
    # Generate a sample from a population where the null is true
    sample = np.random.normal(mu_null, sigma, n)
    
    # Perform a one-sided z-test: H_a: mean > mu_null
    stat, p_value = ztest(sample, value=mu_null, alternative='larger')
    
    # Count Type I errors (rejecting a true null)
    if p_value < alpha:
        type1_count += 1

type1_rate = type1_count / num_sims

print(f"Estimated Type I error rate: {type1_rate:.4f}")
print(f"Expected Type I error rate (alpha): {alpha}")
```

**What this code demonstrates**

- The simulation assumes the null hypothesis is **actually true**.

- It repeatedly samples data and performs a significance test.

- The fraction of times the test incorrectly rejects the null is the
  **empirical Type I error rate**.

- This rate will be close to $\alpha$, confirming that $\alpha$is
  the probability of a Type I error.
