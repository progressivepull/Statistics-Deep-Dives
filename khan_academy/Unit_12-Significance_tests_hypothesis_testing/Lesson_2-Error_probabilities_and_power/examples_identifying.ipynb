# Examples identifying Type I and Type II errors

[Examples identifying Type I and Type II errors | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=JNOWWJeZBTc&t=16s)

A large nationwide poll recently showed an unemployment rate of 9% in
the US. The mayor of a local town wonders if this national result holds
true for her town, so she plans on taking a sample of her residents to
see if the unemployment rate is significantly different than 9% in her
town.

Let p represent the unemployment rate in her town. Here are the
hypotheses she’ll use:  
$$
{H_{0}:p = 0.09
}{H_{a}:p \neq 0.09}
$$

Under which of the following conditions would the mayor commit a Type I
error?

A She concludes the town’s unemployment rate is not 9% when it actually
is.

B She concludes the town’s unemployment rate is not 9% when it actually
is not.

C She concludes the town’s unemployment rate is 9% when it actually is.

D She concludes the town’s unemployment rate is 9% when it actually is
not.

Here’s a clean, structured **bullet summary**, a clear **explanation**,
and a **Python simulation**—all grounded in the ideas of Type I error
and Type II error.

**Bullet Summary**

**Example 1: Mayor Testing Unemployment Rate**

- The mayor tests whether her town’s unemployment rate $P$differs from
  **9%**.

- **Null hypothesis**: $P = 0.09$.

- **Alternative hypothesis**: $P \neq 0.09$.

- A **Type I error** occurs when she **rejects** the null hypothesis
  even though it is **true**.

- In context:

  - She concludes “the unemployment rate is not 9%” **when it actually
    is 9%**.

**Example 2: University Testing Meal Plan Interest**

- University tests whether more than **40%** of students want a meal
  plan.

- **Null hypothesis**: $P \leq 0.40$.

- **Alternative hypothesis**: $P > 0.40$.

- A **Type II error** occurs when they **fail to reject** the null
  hypothesis even though it is **false**.

- In context:

  - More than 40% actually want a meal plan, but they conclude there is
    **not enough evidence** → they **don’t build a new cafeteria** even
    though they **should**.

**Explanation**

A Type I error is rejecting a true null hypothesis. A Type II error is
failing to reject a false null hypothesis.

**Example 1 (Mayor)**

- Reality: unemployment **is** 9%.

- She concludes: unemployment **is not** 9%.

- She rejected a true null → **Type I error**.

**Example 2 (University)**

- Reality: more than 40% want a meal plan.

- They conclude: not enough evidence → **don’t build cafeteria**.

- They failed to reject a false null → **Type II error**.

**Python Simulation Example**

Below is a simulation showing how often Type I and Type II errors occur
for a two‑sided test of a proportion.




```python
import numpy as np
from scipy.stats import binomtest

# Parameters
p_null = 0.09          # Null hypothesis proportion
p_true = 0.12          # True proportion (for Type II error simulation)
n = 200                # Sample size
alpha = 0.05           # Significance level
trials = 10000

type1_count = 0
type2_count = 0

# Type I error simulation (null is true)
for _ in range(trials):
    sample = np.random.binomial(n, p_null)
    p_value = binomtest(sample, n, p_null, alternative='two-sided').pvalue
    if p_value < alpha:
        type1_count += 1

# Type II error simulation (null is false)
for _ in range(trials):
    sample = np.random.binomial(n, p_true)
    p_value = binomtest(sample, n, p_null, alternative='two-sided').pvalue
    if p_value >= alpha:
        type2_count += 1

print("Estimated Type I error rate:", type1_count / trials)
print("Estimated Type II error rate:", type2_count / trials)
```

**What this code does**

- Simulates many samples under the **null** to estimate the Type I rate.

- Simulates many samples under an **alternative** to estimate the Type
  II rate.

- Uses a two‑sided binomial test, matching the mayor’s hypothesis test.
