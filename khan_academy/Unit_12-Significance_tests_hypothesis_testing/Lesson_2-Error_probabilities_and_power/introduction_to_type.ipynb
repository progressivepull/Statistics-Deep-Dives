# Introduction to Type I and Type II errors

[Introduction to Type I and Type II errors | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=Hdbbx7DIweQ&t=8s)

**Bullet summary**

- **Significance tests** compare data to a **null hypothesis** to decide
  whether to reject it.

- A **p-value** measures how surprising the sample statistic would be
  **if the null were true**.

- A **significance level** ($\alpha$) is chosen ahead of time (e.g.,
  0.05).

- Decision rule:

  - If $p < \alpha$: **reject** $H_{0}$.

  - If $p \geq \alpha$: **fail to reject** $H_{0}$.

- But decisions can be wrong → two kinds of errors:

  - **Type I error**: Rejecting $H_{0}$when it is actually true.

  - **Type II error**: Failing to reject $H_{0}$when it is actually
    false.

- Probability of a Type I error = $\alpha$.

- Probability of a Type II error = $\beta$, related to **test power**.

- A 2×2 grid helps visualize:

  - True $H_{0}$+ reject → Type I error

  - True $H_{0}$+ fail to reject → correct

  - False $H_{0}$+ reject → correct

  - False $H_{0}$+ fail to reject → Type II error

**Explanation**

Significance testing always involves uncertainty because we base
decisions on a sample rather than the entire population. After setting
up a null hypothesis $H_{0}$and an alternative $H_{a}$, we collect a
sample, compute a statistic (like a sample mean or proportion), and
calculate a p-value.

The p-value tells us how likely it would be to observe a statistic at
least as extreme as ours **if the null hypothesis were true**. If this
probability is below our chosen significance level $\alpha$, we reject
the null.

But even when we follow the rules perfectly, we can still be wrong:

**Type I error**

- Occurs when $H_{0}$is actually true, but we reject it.

- This is a “false positive.”

- Its probability is exactly $\alpha$.

  - If $\alpha = 0.05$, then even when the null is true, 5% of samples
    will look extreme enough to cause rejection.

**Type II error**

- Occurs when $H_{0}$is false, but we fail to reject it.

- This is a “false negative.”

- Its probability is $\beta$, and $1 - \beta$is the **power** of the
  test.

The grid of possibilities:

| **Reality** | **Decision: Reject** $H_{0}$ | **Decision: Fail to Reject** $H_{0}$ |
|----|----|----|
| $H_{0}$true | Type I error | Correct |
| $H_{0}$false | Correct | Type II error |

This framework helps us understand the risks involved in hypothesis
testing and why significance levels matter.

**Python code example (simulating Type I and Type II errors)**

Below is a simple simulation showing how Type I and Type II errors arise
in practice.






```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

np.random.seed(42)

# Parameters
alpha = 0.05
n = 40

# Null hypothesis proportion
p_null = 0.5

# Alternative true proportion for Type II error simulation
p_alt = 0.6

# Number of simulations
num_sims = 5000

type1_errors = 0
type2_errors = 0

# Simulate Type I errors (null is true)
for _ in range(num_sims):
    sample = np.random.binomial(n, p_null)
    stat, p = proportions_ztest(sample, n, value=p_null, alternative='larger')
    if p < alpha:
        type1_errors += 1

# Simulate Type II errors (null is false)
for _ in range(num_sims):
    sample = np.random.binomial(n, p_alt)
    stat, p = proportions_ztest(sample, n, value=p_null, alternative='larger')
    if p >= alpha:
        type2_errors += 1

print(f"Estimated Type I error rate: {type1_errors / num_sims:.3f}")
print(f"Estimated Type II error rate: {type2_errors / num_sims:.3f}")
```

**What this code demonstrates**

- When the null is true, the proportion of rejections is close to
  $\alpha$(≈ 0.05).

- When the null is false, the proportion of failures to reject gives an
  estimate of $\beta$.

- This simulation makes the abstract concepts of Type I and Type II
  errors concrete.
