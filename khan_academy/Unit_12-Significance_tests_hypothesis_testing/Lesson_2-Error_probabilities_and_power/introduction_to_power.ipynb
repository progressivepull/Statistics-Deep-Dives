**Bullet Summary**

**What Power Means**

- **Power** is the probability of correctly rejecting the null
  hypothesis when it is **false**.

- Formally:

$$
\text{Power} = P(\text{Reject }H_{0} \mid H_{0}\text{ is false})
$$

- Equivalent view:

$$
\text{Power} = 1 - P(\text{Type II error})
$$

- A Type II error occurs when you **fail to reject** a false null
  hypothesis.

**What Affects Power**

- **Higher significance level** $\alpha$→ higher power, but also
  higher Type I error.

- **Larger sample size** $n$→ narrower sampling distributions → higher
  power.

- **Lower population variability** (smaller $\sigma$) → narrower
  sampling distributions → higher power.

- **True parameter farther from the null value** → distributions
  separate more → higher power.

**Graphical Intuition**

- Two sampling distributions:

  - One centered at the null value $\mu_{1}$

  - One centered at the true value $\mu_{2}$

- **Overlap** between them creates the chance of a Type II error.

- **Power** is the area under the alternative distribution that falls in
  the rejection region.

**Explanation**

Power is all about how likely your test is to detect a real effect. When
the null hypothesis is false, the “correct” action is to reject it. But
randomness in sampling means you might not always get a sample extreme
enough to trigger rejection.

So:

- If the null is false but your sample looks too “ordinary,” you **fail
  to reject** → Type II error.

- Power is the probability that this does **not** happen.

**Why sample size matters**

A larger sample size shrinks the standard error:

$$
\text{SE} = \frac{\sigma}{\sqrt{n}}
$$

Smaller SE → narrower sampling distributions → less overlap → fewer Type
II errors → higher power.

**Why α matters**

A larger significance level widens the rejection region. That makes it
easier to reject the null, increasing power—but also increasing the
chance of a Type I error.

**Python Simulation Example**

This simulation shows how power increases with sample size for a
one‑sided z‑test of a mean.




```python
import numpy as np
from scipy.stats import norm

# Parameters
mu_null = 100        # Null hypothesis mean
mu_true = 103        # True mean (alternative)
sigma = 15           # Population SD
alpha = 0.05         # Significance level
sample_sizes = [20, 50, 100, 200, 500]
trials = 20000

def estimate_power(n):
    # Critical value for right-tailed test
    z_crit = norm.ppf(1 - alpha)
    
    # Simulate sampling distribution under the true mean
    samples = np.random.normal(mu_true, sigma / np.sqrt(n), size=trials)
    
    # Compute z-scores relative to null
    z_scores = (samples - mu_null) / (sigma / np.sqrt(n))
    
    # Reject when z > z_crit
    return np.mean(z_scores > z_crit)

# Estimate power for each sample size
for n in sample_sizes:
    power = estimate_power(n)
    print(f"Sample size {n}: Estimated power = {power:.3f}")
```

**What this code demonstrates**

- As $n$increases, the sampling distribution under the alternative
  becomes narrower.

- The test becomes more sensitive to the difference between
  $\mu_{1} = 100$and $\mu_{2} = 103$.

- Power increases—often dramatically.
