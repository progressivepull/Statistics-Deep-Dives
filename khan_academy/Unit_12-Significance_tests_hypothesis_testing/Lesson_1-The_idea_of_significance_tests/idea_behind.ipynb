**⭐ Bullet Summary of the Scenario**

- A cholesterol test is known (magically) to be **99% accurate** on any
  single use.

- You apply the test to **100 people**.

- The number of accurate results out of 100 is a **binomial random
  variable** with

  - $n = 100$trials

  - success probability $p = 0.99$.

- Possible outcomes include:

  - **100 accurate** (prob ≈ 36.6%)

  - **99 accurate** (prob ≈ 37.0%)

  - **98 accurate** (prob ≈ 18.5%)

  - **97 accurate** (prob ≈ 6.0%)

  - **96 accurate** (prob ≈ 1.5%)

  - **95 accurate** (prob ≈ 0.3%)

- These probabilities come from the binomial distribution.

- Now imagine you invent a **new test** and *hope* it is 99% accurate.

- You test it 100 times and find **95 correct results**.

- Under the hypothesis that the test is truly 99% accurate, the
  probability of observing only 95 correct is **0.3%**.

- Since this is extremely small, you would likely **reject the
  hypothesis** that the test is 99% accurate.

- This mirrors the logic of hypothesis testing.

**🔍 Explanation (Intuition + Math)**

**1. Modeling accuracy as a binomial process**

Each test result is either:

- **Accurate** (success) with probability $p = 0.99$

- **Inaccurate** (failure) with probability $1 - p = 0.01$

Repeating this 100 times gives a binomial distribution:

$$
X \sim \text{Binomial}(n = 100,p = 0.99)
$$

**2. Probability of exactly** $\mathbf{k}$**accurate results**

$$
P(X = k) = \binom{100}{k}(0.99)^{k}(0.01)^{100 - k}
$$

This formula produces the values:

- $P(X = 100) \approx 0.366$

- $P(X = 99) \approx 0.370$

- $P(X = 98) \approx 0.185$

- $P(X = 97) \approx 0.060$

- $P(X = 96) \approx 0.015$

- $P(X = 95) \approx 0.003$

**3. Using this for hypothesis testing**

Your hypothesis is:

“My new test is 99% accurate.”

You test it 100 times and observe **95 correct**.

If the hypothesis were true, the probability of seeing 95 correct is:

$$
P(X = 95) \approx 0.003 = 0.3\%
$$

This is extremely unlikely under the hypothesis, so you would **reject**
the claim that the test is 99% accurate.

This is exactly how statisticians use **p-values** and **significance
thresholds** like 5%.

**🧪 Python Code Example (Reproducing All Probabilities)**




```python
from math import comb

p = 0.99
n = 100

def binomial_prob(k):
    return comb(n, k) * (p**k) * ((1-p)**(n-k))

# Compute probabilities for k = 95 to 100
for k in range(100, 94, -1):
    print(k, round(binomial_prob(k)*100, 3), "%")
```

This matches the values in your scenario exactly.
