# P-values and significance tests

[P-values and significance tests | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=KS6KEWaoOOE)

**Bullet summary**

- **Goal:** Test whether changing the website background to yellow
  increases mean time on site beyond $20$minutes.

- **Null hypothesis** $H_{0}$**:** $\mu = 20$minutes (yellow has no
  effect on mean time).

- **Alternative hypothesis** $H_{a}$**:** $\mu > 20$minutes (yellow
  increases mean time).

- **Significance level:** Choose $\alpha$(e.g. $\alpha = 0.05$)
  *before* seeing the data.

- **Data collection:** Take a sample of size $n$(e.g. $n = 100$),
  compute sample mean $\overset{ˉ}{x}$and sample standard deviation
  $s$.

- **Test statistic:** Use a $t$-statistic (or $z$if $\sigma$known)
  based on $\overset{ˉ}{x},s,n$.

- **p-value:**
  $p = P(\text{getting a sample mean at least as extreme as observed} \mid H_{0}\text{ is true})$.

- **Decision rule:**

  - If $p < \alpha$: reject $H_{0}$, say you have evidence that
    $\mu > 20$.

  - If $p \geq \alpha$: do **not** reject $H_{0}$; you *don’t* claim
    an effect, but you also don’t “prove” no effect.

- **Key clarification:** The p-value is $P(\text{data} \mid H_{0})$,
  **not** $P(H_{0} \mid \text{data})$.

**Step-by-step explanation in this context**

1.  **Set up hypotheses**

    - **Null:** $H_{0}:\mu = 20$minutes.

    - **Alternative:** $H_{a}:\mu > 20$minutes (one-sided test,
      because you only care about an increase).

2.  **Choose significance level**

    - Pick $\alpha = 0.05$.

    - This is your threshold for what you consider “too unlikely under
      $H_{0}$.”

3.  **Collect sample and compute statistics**

    - Suppose you collect $n = 100$users after the yellow change.

    - You compute:

      - Sample mean: $\overset{ˉ}{x} = 25$minutes.

      - Sample standard deviation: $s$(estimated from the data).

4.  **Compute test statistic**

For a one-sample $t$-test:

$$
t = \frac{\overset{ˉ}{x} - \mu_{0}}{s/\sqrt{n}}
$$

where $\mu_{0} = 20$is the null mean.

5.  **Compute p-value**

    - For a one-sided test $H_{a}:\mu > 20$:

$$
p = P(T \geq t_{\text{obs}} \mid H_{0})
$$

where $T$follows a $t$-distribution with $n - 1$degrees of freedom
(approximately normal if $n$is large).

- Intuition: If $H_{0}$were true, how surprising is a sample mean as
  large as $25$(or larger)?

6.  **Decision**

    - If $p < 0.05$: you say “this result would be quite unlikely if
      the true mean were still 20, so we reject $H_{0}$and conclude
      there is evidence that the yellow background increased time on
      site.”

    - If $p \geq 0.05$: you say “this result is not unlikely enough
      under $H_{0}$; we do not reject $H_{0}$. We don’t have strong
      evidence that yellow changed the mean time.”

7.  **Interpretation nuance**

    - You are **not** computing the probability that $H_{0}$is true.

    - You are computing the probability of seeing data at least as
      extreme as yours *assuming* $H_{0}$is true.

    - The decision is a rule: if that probability is below $\alpha$,
      you treat the data as evidence against $H_{0}$.

**Python code example**

Below is a minimal example that:

1.  Simulates a sample of user times after the yellow change.

2.  Performs a one-sample $t$-test against $\mu_{0} = 20$.



```python
import numpy as np
from scipy import stats

# Known/assumed values
mu_null = 20          # H0: mean time = 20 minutes
alpha = 0.05          # significance level
n = 100               # sample size

# --- Step 1: simulate sample data after the yellow change ---
# For illustration, assume the true mean after change is 25, sd = 10
np.random.seed(42)
true_mu_after = 25
true_sigma_after = 10
sample = np.random.normal(loc=true_mu_after, scale=true_sigma_after, size=n)

# --- Step 2: compute sample statistics ---
x_bar = np.mean(sample)
s = np.std(sample, ddof=1)

print(f"Sample mean: {x_bar:.2f}")
print(f"Sample std:  {s:.2f}")

# --- Step 3: compute t-statistic for one-sample test ---
t_stat = (x_bar - mu_null) / (s / np.sqrt(n))
df = n - 1

print(f"t-statistic: {t_stat:.3f}, df = {df}")

# --- Step 4: compute one-sided p-value for H_a: mu > mu_null ---
# p = P(T >= t_stat | H0), where T ~ t(df)
p_value = 1 - stats.t.cdf(t_stat, df=df)

print(f"One-sided p-value: {p_value:.4f}")

# --- Step 5: decision rule ---
if p_value < alpha:
    print(f"Reject H0 at alpha = {alpha}. Evidence that mean time > {mu_null}.")
else:
    print(f"Do NOT reject H0 at alpha = {alpha}. No strong evidence of an increase.")
```

**What this code is doing conceptually**

- **Simulated data:** Pretends the yellow background truly increases
  mean time to $25$minutes with some variability.

- **Test:** Checks whether the observed sample mean is significantly
  greater than $20$under the assumption that the true mean is $20$.

- **p-value:** Quantifies how extreme the observed mean is under
  $H_{0}$.

- **Decision:** Uses the rule $p < \alpha$to decide whether to claim
  evidence that the yellow background increased time on site.

If you’d like, we can tweak this to:

- Compare different $\alpha$values.

- Run many simulations to see how often you correctly detect an effect
  (power).
