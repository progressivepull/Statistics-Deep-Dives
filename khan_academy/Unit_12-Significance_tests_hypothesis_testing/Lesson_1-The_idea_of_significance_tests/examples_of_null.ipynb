**⭐ Bullet Summary**

- **A restaurant owner installs a drink machine designed to dispense 530
  mL on the medium setting.**

- **The owner suspects the machine is dispensing too much liquid.**

- **They take a sample of 30 medium drinks to test whether the average
  amount exceeds 530 mL.**

- **Hypotheses must be about the population mean** $\mathbf{\mu}$**,
  not a proportion.**

- **Null hypothesis: machine is working as intended →**
  $\mathbf{\mu = 530}$**.**

- **Alternative hypothesis: machine dispenses more than intended →**
  $\mathbf{\mu > 530}$**.**

- **This is a right‑tailed test, because the concern is “too much
  liquid.”**

- **Choices involving proportions are irrelevant because the variable is
  continuous volume, not a proportion.**

- **Therefore, the correct hypotheses are:**

  - $\mathbf{H}_{\mathbf{0}}\mathbf{:\mu = 530}$

  - $\mathbf{H}_{\mathbf{a}}\mathbf{:\mu > 530}$

**You can explore related ideas like null hypotheses, alternative
hypotheses, and one‑tailed tests.**

**🔍 Explanation (Intuition + Reasoning)**

**1. Identify the parameter**

**We care about the population mean amount dispensed on the medium
setting:**

$$
\mathbf{\mu =}\text{average amount dispensed}
$$

**2. Form the null hypothesis**

**The null hypothesis represents “no change,” “no problem,” or “machine
works as designed”:**

$$
\mathbf{H}_{\mathbf{0}}\mathbf{:\mu = 530}
$$

**3. Form the alternative hypothesis**

**The owner suspects the machine is dispensing too much, so we test
whether the mean is greater than 530:**

$$
\mathbf{H}_{\mathbf{a}}\mathbf{:\mu > 530}
$$

**This is a directional claim, so the test is right‑tailed.**

**4. Reject or fail to reject**

**After collecting data (sample of 30 drinks), the owner would compute a
test statistic and p‑value. If the p‑value is small (below a threshold
like 0.05), they would reject** $\mathbf{H}_{\mathbf{0}}$**and
conclude the machine dispenses too much.**

**You can explore p‑values or significance levels.**

**🧪 Python Code Example**

**Below is a simple simulation showing how you might test the hypothesis
if you had sample data. (Here I simulate fake data just to demonstrate
the workflow.)**


```python
import numpy as np
from scipy import stats

# Simulate sample data (example: machine dispensing too much)
np.random.seed(0)
sample = np.random.normal(loc=535, scale=10, size=30)

# Known hypothesized mean
mu0 = 530

# One-sample t-test (right-tailed)
t_stat, p_two_tailed = stats.ttest_1samp(sample, mu0)

# Convert to right-tailed p-value
p_right = p_two_tailed / 2 if t_stat > 0 else 1 - p_two_tailed / 2

print("t-statistic:", t_stat)
print("Right-tailed p-value:", p_right)
```

**What this code does:**

- Generates a sample of 30 drink volumes.

- Performs a **one‑sample t‑test** comparing the sample mean to 530.

- Converts the result to a **right‑tailed** p‑value because the
  alternative is $\mu > 530$.

- A small p‑value would support the owner’s suspicion.
