**⭐ Bullet Summary of the Scenario**

- Four siblings randomly draw a name each night to decide who washes
  dishes.

- Each sibling has a $1/4$chance of being picked; Bill has a
  $3/4$chance of **not** being picked.

- **Probability Bill is not picked for 3 nights:**
  $\left( 3/4)^{3} = 27/64 \approx 0.42 \right.\ $.

- **Probability Bill is not picked for 12 nights:**
  $\left( 3/4)^{12}\approx 0.032 \right.\ $(about 3.2%).

- As the streak gets longer, the probability under a fair process
  becomes extremely small.

- This mirrors how statisticians evaluate evidence against a hypothesis
  using thresholds like the **5% significance level**.

**🔍 Explanation (Intuition + Math)**

**1. One night**

Bill has a $1/4$chance of being picked, so

$$
P(\text{Bill not picked}) = \frac{3}{4}.
$$

**2. Three nights in a row**

Because each night is independent, multiply the probabilities:

$$
P(\text{Bill not picked 3 nights}) = \left( \frac{3}{4} \right)^{3} = \frac{27}{64} \approx 0.42.
$$

A 42% chance is not suspicious at all — streaks happen.

**3. Twelve nights in a row**

$$
P(\text{Bill not picked 12 nights}) = \left( \frac{3}{4} \right)^{12} \approx 0.032.
$$

Now we’re down to **3.2%**, which is below the common **5% significance
threshold** used in hypothesis testing.

This is exactly how statisticians decide whether an observed outcome is
unlikely enough to doubt the “fairness” assumption.

**🧪 Python Simulation (Monte Carlo)**

Here’s a simulation to empirically verify the probability of Bill
avoiding dish duty for 12 nights:




```python
import random
import numpy as np

def simulate_streak(nights=12, trials=1000000):
    count = 0
    for _ in range(trials):
        # 0 = Bill, 1/2/3 = other siblings
        picks = [random.randint(0, 3) for _ in range(nights)]
        if all(p != 0 for p in picks):
            count += 1
    return count / trials

prob_estimate = simulate_streak()
prob_estimate
```

**Expected output:**

A value close to

$$
\left( 3/4)^{12}\approx 0.032. \right.\ 
$$

Simulations like this are great for building intuition — especially when
probabilities get tiny.
