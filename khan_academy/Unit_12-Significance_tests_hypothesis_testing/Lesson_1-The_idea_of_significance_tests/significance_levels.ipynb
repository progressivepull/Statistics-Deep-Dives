# Comparing P-values to different significance levels

[Comparing P-values to different significance levels | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=DQCF4kTXf9Q&t=7s)

Rahim heard that spinning—rather than flipping—a penny raises the
probability above 50% that the penny lands showing heads.  
He tested this by spinning 10 different pennies 10 times each.  
His hypotheses were $H_{0}:p = 0.50$ versus $H_{a}:p > 0.50$,
where $p$ is the true proportion of spins that a penny would land
showing “heads”.

In his 100 spins, the penny landed showing “heads” in 59 spins.  
Rahim calculated that the
statistic $\widehat{p} = \frac{59}{100} = 0.59$ had an
associated $p$-value of approximately 0.036.

**Bullet summary of the scenario**

- **Significance level** ($\alpha$) is a threshold chosen *before*
  running an experiment.

- It determines when you will **reject the null hypothesis**.

- Rahim tests whether **spinning a penny** increases the probability of
  landing heads above 50%.

- Hypotheses:

  - $H_{0}:p = 0.50$

  - $H_{a}:p > 0.50$

- He spins pennies 100 times and observes **59 heads**, so
  $\widehat{p} = 0.59$.

- The associated **p-value** is **0.036**.

- If $\alpha = 0.05$:

  - $0.036 < 0.05$→ **Reject** $H_{0}$→ Evidence that spinning
    increases the probability of heads.

- If $\alpha = 0.01$:

  - $0.036 > 0.01$→ **Fail to reject** $H_{0}$→ Not enough evidence
    at this stricter threshold.

- This difference shows why $\alpha$must be chosen **before** seeing
  the data—changing it afterward is unethical.

- Researchers might be tempted to “move the goalposts” to get a desired
  result, which is why pre‑registration matters.

**Explanation (step-by-step)**

**1. Setting up the hypotheses**

Rahim wants to test whether spinning a penny increases the chance of
heads.

- **Null hypothesis:**
  $\lbrack p = 0.50\rbrack(copilot - action://composer - send?text = Why$

- **Alternative hypothesis:**
  $\lbrack p > 0.50\rbrack(copilot - action://composer - send?text = Why$

This is a **one‑tailed test** because he only cares about increases.

**2. Collecting data**

He performs 100 spins and observes:

- Heads = 59

- Sample proportion = $\widehat{p} = 0.59$

**3. Computing the p-value**

The p-value is:

$$
P(\widehat{p} \geq 0.59 \mid p = 0.50)
$$

Rahim computes this to be **0.036**.

**4. Decision at** $\mathbf{\alpha = 0.05}$

Since $0.036 < 0.05$, the result is “unlikely enough” under the null.
→ **Reject** $H_{0}$→ Evidence that spinning increases the probability
of heads.

**5. Decision at** $\mathbf{\alpha = 0.01}$**\*\***

Now the threshold is stricter. Since $0.036 > 0.01$: → **Fail to
reject** $H_{0}$→ Not enough evidence at this level.

**6. Why choosing** $\mathbf{\alpha}$**ahead of time matters**

If researchers choose $\alpha$*after* seeing the p-value, they can
manipulate conclusions. This is unethical because it biases results
toward “interesting” findings.

**Python code example**

Below is a reproducible example showing how to compute the p-value for
Rahim’s test using a **one‑proportion z‑test**.


```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# Observed data
heads = 59
n = 100
p_null = 0.50

# One-sided test: H_a: p > 0.50
count = heads
nobs = n

z_stat, p_value = proportions_ztest(count, nobs, value=p_null, alternative='larger')

print(f"Sample proportion: {count/nobs:.3f}")
print(f"Z-statistic: {z_stat:.3f}")
print(f"One-sided p-value: {p_value:.4f}")

alpha_1 = 0.05
alpha_2 = 0.01

print("\nDecision at alpha = 0.05:")
print("Reject H0" if p_value < alpha_1 else "Fail to reject H0")

print("\nDecision at alpha = 0.01:")
print("Reject H0" if p_value < alpha_2 else "Fail to reject H0")
```

**What this code demonstrates**

- It computes the **z‑statistic** and **p‑value** for Rahim’s sample.

- It evaluates the decision at two different significance levels.

- It shows how the same data can lead to different conclusions depending
  on $\alpha$.
