# Estimating a P-value from a simulation

[Estimating a P-value from a simulation | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=hNDlqAYRDyg&t=2s)

Evie read an article that said 6% of teenagers were vegetarians, but she thinks it’s higher for students at her large school. To test her theory, Evie took a random sample of 25 students at her school, and 20% of them were vegetarians.

To see how likely a sample like this was to happen by random chance alone, Evie performed a simulation. She simulated 40 samples of n = 25 students from a large population where 6% of the students were vegetarian. She recorded the proportion of vegetarians in each sample. Here are the sample proportions from her 40 samples:

![vegetarians.png](./IMAGES/vegetarians.png)

**Bullet summary**

- **Evie’s claim: She believes the proportion of vegetarians at her
  school is higher than the 6% population rate reported for teenagers.**

- **Null hypothesis:**
  $\mathbf{H}_{\mathbf{0}}\mathbf{:p = 0.06}$**(her school’s true
  proportion is the same as the general teen population).**

- **Alternative hypothesis:**
  $\mathbf{H}_{\mathbf{a}}\mathbf{:p > 0.06}$**(her school’s
  proportion is higher).**

- **Sample:** $\mathbf{n = 25}$**students, with**
  $\widehat{\mathbf{p}}\mathbf{= 0.20}$**vegetarians.**

- **p-value meaning: Probability of getting a sample proportion ≥ 0.20
  if the true proportion were 0.06.**

- **Simulation: Evie simulated 40 samples of size 25 under the
  assumption** $\mathbf{p = 0.06}$**.**

- **Observed extreme samples: 3 out of 40 simulated proportions were ≥
  0.20.**

- **Approximate p-value:**
  $\mathbf{3}\mathbf{/}\mathbf{40 \approx 0.075}$**(≈ 7.5%).**

- **Interpretation: If the true proportion were 6%, samples as extreme
  as Evie’s occur about 7.5% of the time.**

**Explanation**

**Evie wants to test whether vegetarianism is more common at her school
than the national teen rate of 6%. She sets up:**

- $\mathbf{H}_{\mathbf{0}}\mathbf{:p = 0.06}$

- $\mathbf{H}_{\mathbf{a}}\mathbf{:p > 0.06}$

**She samples 25 students and finds 20% vegetarians—much higher than 6%.
To evaluate whether this difference could be due to random sampling
variation, she simulates 40 samples of size 25 from a population where
only 6% are vegetarians.**

**Each simulated sample gives a sample proportion. She then counts how
many simulated proportions are ≥ 0.20, because that’s her observed
sample proportion.**

**Out of 40 simulated samples, 3 were at least 0.20. So:**

$$
\text{p-value}\mathbf{\approx}\frac{\mathbf{3}}{\mathbf{40}}\mathbf{= 0.075}
$$

**This means: If the true proportion were 6%, you’d see a sample
proportion of 20% or higher about 7.5% of the time.**

**This p-value is not extremely small, but it does suggest that Evie’s
sample is somewhat unusual under the null.**

**Python code example (simulation approach)**

**Below is a Python simulation that mirrors Evie’s process.**




```python
import numpy as np

# Null hypothesis proportion
p_null = 0.06

# Sample size
n = 25

# Number of simulated samples
num_sims = 40

# Observed sample proportion
observed_p = 0.20

# Run simulation
np.random.seed(42)
simulated_props = np.random.binomial(n, p_null, num_sims) / n

# Count how many simulated proportions are >= observed
extreme_count = np.sum(simulated_props >= observed_p)

p_value_estimate = extreme_count / num_sims

print("Simulated proportions:", simulated_props)
print(f"Extreme samples: {extreme_count} out of {num_sims}")
print(f"Approximate p-value: {p_value_estimate:.4f}")
```

**What this code does**

- Simulates 40 samples of size 25 assuming the rue proportion is 6%.

- Computes the sample proportion for each.

- Counts how many are ≥ 0.20.

- Estimates the p-value exactly as Evie did.
