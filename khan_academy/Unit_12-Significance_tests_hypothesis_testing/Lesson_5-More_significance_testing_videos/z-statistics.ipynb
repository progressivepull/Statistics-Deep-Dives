# Z-statistics vs. T-statistics

[Z-statistics vs. T-statistics | Inferential statistics | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=5ABpqVSx33I&t=5s)

A **Z‑statistic** measures how many **population-standard-deviation
units** a sample mean is from the hypothesized mean, while a
**T‑statistic** measures how many **sample-standard-deviation units** it
is away when the population standard deviation is unknown and the sample
is small. The key difference is that the **Z-statistic uses a normal
distribution**, while the **T-statistic uses a heavier‑tailed
t‑distribution**, especially important when $n < 30$.

**🔹 Bullet summary of the difference**

- **Z‑statistic** — used when the population standard deviation is known
  or when the sample size is large (typically $n \geq 30$).

- **T‑statistic** — used when the population standard deviation is
  unknown and the sample size is small.

- **Sampling distribution** of the mean becomes approximately normal for
  large $n$, making the Z‑statistic reliable.

- **Standard error** estimated using sample standard deviation
  introduces extra uncertainty when $n$is small.

- **T‑distribution** has heavier tails, reflecting that extra
  uncertainty.

- **Rule of thumb**:

  - Use **Z** when $n \geq 30$.

  - Use **T** when $n < 30$.

- Both statistics measure the same structure:

$$
\frac{\overset{ˉ}{x} - \mu_{0}}{\text{standard error}}
$$

but differ in which standard error and which reference distribution they
use.

**🔍 Explanation of the intuition**

When you compute a **Z‑statistic**, you assume the denominator (the
standard deviation of the sampling distribution) is known or
well‑estimated. With large samples, the sample standard deviation is a
stable estimate of the population standard deviation, so the resulting
statistic behaves like a standard normal variable.

When the sample is small, the sample standard deviation is a noisy
estimate. That noise makes the distribution of the statistic **wider**
and **heavier‑tailed** than the normal distribution. The
**T‑distribution** accounts for this by giving more probability to
extreme values, preventing you from being overly confident.

So:

- Large samples → variability is stable → **normal distribution** →
  **Z‑statistic**.

- Small samples → variability is unstable → **t‑distribution** →
  **T‑statistic**.

Both statistics answer the same question: *How extreme is my sample mean
relative to what I expected?* They just use different reference curves.

**🧪 Code example: computing both Z and T statistics**




```python
import numpy as np
from scipy.stats import norm, t

# Example data
xbar = 5.2        # sample mean
mu0 = 5.0         # hypothesized population mean
s = 1.1           # sample standard deviation
n = 12            # sample size (small, so t-test is appropriate)

# Standard error
SE = s / np.sqrt(n)

# Z-statistic (not ideal for small n)
z_stat = (xbar - mu0) / SE
p_value_z = 2 * norm.sf(abs(z_stat))

# T-statistic (correct for small n)
t_stat = (xbar - mu0) / SE
df = n - 1
p_value_t = 2 * t.sf(abs(t_stat), df)

print("Z-statistic:", z_stat)
print("Two-tailed Z p-value:", p_value_z)

print("T-statistic:", t_stat)
print("Degrees of freedom:", df)
print("Two-tailed T p-value:", p_value_t)
```

This example shows that the **numerator is identical**, but the
**reference distribution differs**, producing different p‑values when
$n$is small.
