# Small sample hypothesis test

[Small sample hypothesis test | Inferential statistics | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=D2sMsmL0ScQ&t=2s)

The mean emission of all engines of a new design needs to be below 20 ppm if the design is to meet new emission requirements. Ten engines are manufactured for testing purposes, and the emission level of each is determined. The emission data is:    

15.6   
16.2   
22.5    
20.5     
16.4    
19.4     
16.6     
17.9       
12.7     
13.9      



Does the data supply sufficient evidence to conclude that this type of engine meets the new standard? Assume we are willing to risk a Type I error with probability = 0.01

**Bullet summary**

- **Goal:** Test if the new engine design’s true mean emission is
  **below 20 ppm** at significance level $\alpha = 0.01$.

- **Given:**

  - Sample size: $n = 10$

  - Sample mean: $\overset{ˉ}{x} = 17.17$

  - Sample standard deviation: $s = 2.98$

  - Standard requirement: population mean $\mu < 20$ppm

- **Hypotheses:**

  - $H_{0}:\mu = 20$(does **not** meet standard)

  - $H_{a}:\mu < 20$(meets standard; mean emission is below 20 ppm)

- **Test statistic (one-sample t-test):**

$$
t = \frac{\overset{ˉ}{x} - \mu_{0}}{s/\sqrt{n}} = \frac{17.17 - 20}{2.98/\sqrt{10}} \approx - 3.00
$$

- **Degrees of freedom:** $df = n - 1 = 9$.

- **Critical value (left-tailed,** $\alpha = 0.01$**,**
  $df = 9$**):**

$$
t_{0.01,\text{ }9} \approx - 2.821
$$

- **Decision rule:**

  - Reject $H_{0}$if $t \leq t_{0.01,\text{ }9} = - 2.821$.

- **Comparison:**

  - Observed $t \approx - 3.00 < - 2.821$→ falls in rejection region.

- **Conclusion:**

  - Reject $H_{0}$.

  - There is sufficient evidence at the 1% significance level that the
    engine’s mean emission is **below 20 ppm**, so the design meets the
    new standard.

- **Type I error risk:**

  - At $\alpha = 0.01$, there is at most a 1% chance of incorrectly
    concluding the design meets the standard when it actually does not.

**Step-by-step explanation**

**1. Framing the question as a hypothesis test**

We want to know if the **true mean emission** $\mu$of this engine
design is **less than 20 ppm**. That’s a classic **one-sample,
left-tailed t-test** on the mean.

- **Null hypothesis** $H_{0}$**:**

$$
\mu = 20
$$

This represents the “borderline fail” case—exactly at the limit, not
below it.

- **Alternative hypothesis** $H_{a}$**:**

$$
\mu < 20
$$

This is what we want to show: the design truly meets the standard.

We set the **significance level** to $\alpha = 0.01$, meaning we
accept at most a 1% chance of a **Type I error** (rejecting
$H_{0}$when it is actually true).

**2. Choosing the test and test statistic**

We have:

- Small sample size: $n = 10$

- Unknown population standard deviation (we only have sample $s$)

So we use a **t-test** with the test statistic:

$$
t = \frac{\overset{ˉ}{x} - \mu_{0}}{s/\sqrt{n}}
$$

Where:

- $\overset{ˉ}{x} = 17.17$(sample mean)

- $\mu_{0} = 20$(hypothesized mean under $H_{0}$)

- $s = 2.98$(sample standard deviation)

- $n = 10$

Compute:

$$
t = \frac{17.17 - 20}{2.98/\sqrt{10}} = \frac{- 2.83}{2.98/\sqrt{10}} \approx - 3.00
$$

Degrees of freedom:

$$
df = n - 1 = 9
$$

**3. Critical value and rejection region**

We’re doing a **left-tailed** test at $\alpha = 0.01$with $df = 9$.

From a t-table (or software), the **critical value** is:

$$
t_{0.01,\text{ }9} \approx - 2.821
$$

- **Rejection region:** $t \leq - 2.821$

- **Non-rejection region:** $t > - 2.821$

Our observed statistic:

$$
t_{\text{obs}} \approx - 3.00
$$

Since:

$$
- 3.00 < - 2.821
$$

the observed t-statistic lies in the **rejection region**.

**4. Interpretation and Type I error**

Because our t-statistic is more extreme (more negative) than the
critical value, the probability of seeing such a low sample mean **if
the true mean were actually 20** is **less than 1%**.

So:

- We **reject** $H_{0}$.

- We conclude there is strong evidence that $\mu < 20$ppm.

- In words: the data support that this engine design **meets the
  emission standard**.

- The chance that we’re wrong in this specific way (saying it meets the
  standard when the true mean is actually 20 or higher) is at most
  $\alpha = 0.01$, i.e., a **Type I error probability of 1%**.

**Python code example**

Here’s a compact Python example using scipy.stats to:

- Compute the t-statistic

- Compute the one-sided p-value

- Compare to $\alpha = 0.01$




```python
import numpy as np
from scipy import stats

# Given data
n = 10
xbar = 17.17      # sample mean
s = 2.98          # sample standard deviation
mu0 = 20          # hypothesized mean under H0
alpha = 0.01

# Compute t-statistic
t_stat = (xbar - mu0) / (s / np.sqrt(n))
df = n - 1

# One-sided (left-tailed) p-value: P(T <= t_stat)
p_value = stats.t.cdf(t_stat, df=df)

print(f"t-statistic: {t_stat:.4f}")
print(f"Degrees of freedom: {df}")
print(f"One-sided p-value: {p_value:.6f}")

# Decision
if p_value < alpha:
    print("Reject H0: There is sufficient evidence that the mean emission is below 20 ppm.")
else:
    print("Fail to reject H0: Insufficient evidence that the mean emission is below 20 ppm.")
```

If you’d like, we can also simulate many samples under $\mu = 20$to
empirically see how rare a mean as low as 17.17 is—very in your
wheelhouse.
