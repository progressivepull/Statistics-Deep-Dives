# Large sample proportion hypothesis testing

[Large sample proportion hypothesis testing | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=1JT9oODsClE&t=3s)

We want to test the hypothesis that more than 30% of U.S. households have Internet access (with a significance level of 5%). We collect a sample of 150 households and find that 57 have access.

![chart.png](./IMAGES/chart.png)

**Bullet summary**

- **Goal:** Test whether the true population proportion of U.S.
  households with internet access is **greater than 30%** at a **5%
  significance level**.

- **Sample:** $n = 150$households, with $57$having internet access →
  sample proportion

$$
\widehat{p} = \frac{57}{150} = 0.38
$$

- **Hypotheses:**

  - $H_{0}:p \leq 0.30$

  - $H_{a}:p > 0.30$

- **Assume** $H_{0}$**is true** and compute the sampling distribution
  of $\widehat{p}$.

- **Standard error under** $H_{0}$**:**

$$
SE = \sqrt{\frac{p_{0}(1 - p_{0})}{n}} = \sqrt{\frac{0.3 \cdot 0.7}{150}} \approx 0.037
$$

(Matches the document’s value: “*this value… was 0.037*” )

- **Z‑statistic:**

$$
z = \frac{\widehat{p} - p_{0}}{SE} = \frac{0.38 - 0.30}{0.037} \approx 2.14
$$

- **Critical value (one‑tailed,** $\alpha = 0.05$**):**

$$
z_{0.05} = 1.65
$$

(The document states: “*this critical Z‑value is equal to 1.65*” )

- **Decision:**

  - Since $2.14 > 1.65$, reject $H_{0}$.

- **Conclusion:**

  - There is sufficient evidence at the 5% level that **more than 30%**
    of U.S. households have internet access.

**Explanation**

The question asks whether the true proportion of U.S. households with
internet access exceeds 30%. This is a **one‑sample, one‑tailed Z‑test
for a population proportion**.

**1. Hypotheses**

We formalize the claim as:

$$
H_{0}:p \leq 0.30,H_{a}:p > 0.30
$$

The alternative hypothesis matches the claim that “*more than 30% of
U.S. households have internet access*” .

**2. Sample proportion**

From the sample of 150 households:

$$
\widehat{p} = \frac{57}{150} = 0.38
$$

This is noticeably higher than 0.30, but we must check whether it is
**statistically** higher.

**3. Sampling distribution under the null**

Under $H_{0}$, the sampling distribution of $\widehat{p}$is
approximately normal because:

- $np_{0} = 150(0.3) = 45 > 5$

- $n(1 - p_{0}) = 150(0.7) = 105 > 5$

The document notes this condition explicitly: “*150 times 0.3 is
obviously greater than 5… you can approximate with a normal
distribution*” .

The standard error is:

$$
SE = \sqrt{\frac{0.3 \cdot 0.7}{150}} \approx 0.037
$$

**4. Compute the Z‑statistic**

$$
z = \frac{0.38 - 0.30}{0.037} \approx 2.14
$$

This matches the document’s calculation: “*our Z‑statistic… is 2.14*” .

**5. Compare to critical value**

For a one‑tailed test at $\alpha = 0.05$:

$$
z_{0.05} = 1.65
$$

Since $2.14 > 1.65$, the observed sample proportion is **significantly
higher** than 0.30.

**6. Conclusion**

We reject the null hypothesis. The data provide strong evidence that
**more than 30%** of U.S. households have internet access.

**Python code example**



```python
import numpy as np
from scipy.stats import norm

# Given data
n = 150
x = 57
p0 = 0.30
alpha = 0.05

# Sample proportion
phat = x / n

# Standard error under H0
SE = np.sqrt(p0 * (1 - p0) / n)

# Z-statistic
z_stat = (phat - p0) / SE

# One-tailed p-value (right tail)
p_value = 1 - norm.cdf(z_stat)

print(f"Sample proportion: {phat:.4f}")
print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.6f}")

if p_value < alpha:
    print("Reject H0: Evidence suggests p > 0.30.")
else:
    print("Fail to reject H0: Insufficient evidence that p > 0.30.")
```
