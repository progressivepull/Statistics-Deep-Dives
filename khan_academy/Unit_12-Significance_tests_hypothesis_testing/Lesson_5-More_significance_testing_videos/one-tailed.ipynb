# One-tailed and two-tailed tests

[One-tailed and two-tailed tests | Inferential statistics | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=mvye6X_0upA&t=4s)

A **one‑tailed test** makes the evidence for the drug even stronger
because all of the probability mass is concentrated in the direction
that matters: *lower* response times.

**🧩 Bullet summary of the one‑tailed test**

- **Null hypothesis**: $H_{0}:\mu = 1.2$seconds (drug has no effect).

- **Alternative hypothesis**: $H_{a}:\mu < 1.2$seconds (drug lowers
  response time).

- **Sampling distribution** under $H_{0}$: mean $1.2$, standard
  error $0.05$.

- **Observed sample mean**: $1.05$, which is **3 standard deviations**
  below the null mean.

- **Z‑statistic**:

$$
z = \frac{1.05 - 1.2}{0.05} = - 3
$$

- **One‑tailed p‑value**: about **0.0015** (0.15%).

- **Interpretation**: If the drug truly had no effect, the chance of
  seeing a sample mean this low or lower is only **0.15%**.

- **Conclusion**: Reject $H_{0}$. Strong evidence the drug **reduces**
  response time.

**🧠 Explanation of why the p‑value changes**

A **two‑tailed test** splits the rejection region into both tails
because any extreme deviation (higher or lower) counts as evidence
against $H_{0}$. That’s why the earlier p‑value was about **0.003**.

A **one‑tailed test** focuses only on the direction specified in the
alternative. Since the normal distribution is symmetric, the one‑tailed
area is **half** the two‑tailed area:

$$
0.003/2 = 0.0015
$$

This makes the evidence *stronger* when the observed effect is in the
predicted direction.

**🧪 Python code example (one‑tailed test + simulation)**

```python
import numpy as np
from scipy.stats import norm

# Given values
mu_null = 1.2
xbar = 1.05
s = 0.5
n = 100

# Standard error
SE = s / np.sqrt(n)

# Z-statistic
z = (xbar - mu_null) / SE

# One-tailed p-value (lower tail)
p_value_one_tailed = norm.cdf(z)

print("Z-statistic:", z)
print("One-tailed p-value:", p_value_one_tailed)

# Optional simulation to confirm
sim_means = np.random.normal(mu_null, s, size=(100000, n)).mean(axis=1)
sim_p = np.mean(sim_means <= xbar)

print("Simulated one-tailed p-value:", sim_p)
```

This code computes the theoretical one‑tailed p‑value and verifies it
with a Monte Carlo simulation.
