# Hypothesis testing and p-values

[Hypothesis testing and p-values | Inferential statistics | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=-FtlH4svqx4&t=2s)

A neurologist is testing the effect of a drug on response time by injecting 100 rats with a unit dose of the drug, subjecting each to neurological stimulus, and recording its response time. The neurologist knows that the mean response time for rats not injected with the drug is 1.2 seconds. The mean of the 100 injected rats’ response times is 1.05 seconds with a sample standard deviation of 0.5 seconds. Do you think that the drug has an effect on response time?

A drug that produces a sample mean of **1.05 seconds** when the known
baseline is **1.2 seconds** yields a **z‑statistic of 3**, which
corresponds to a **p‑value ≈ 0.003**, so the data provide **strong
evidence** that the drug affects response time.

**🧩 Bullet summary of the hypothesis test**

- **Null hypothesis**: $H_{0}:\mu = 1.2$seconds (drug has no effect).

- **Alternative hypothesis**: $H_{a}:\mu \neq 1.2$seconds (drug
  changes response time).

- **Sample size**: $n = 100$.

- **Sample mean**: $\overset{ˉ}{x} = 1.05$.

- **Sample standard deviation**: $s = 0.5$.

- **Standard error**: $SE = s/\sqrt{n} = 0.5/10 = 0.05$.

- **Z‑statistic**:

$$
z = \frac{1.05 - 1.2}{0.05} = - 3
$$

- **Two‑tailed p‑value**: about **0.003**.

- **Conclusion**: Reject $H_{0}$. The drug **does affect** response
  time.

**🧠 Explanation of the reasoning**

The neurologist compares the observed sample mean to what would be
expected if the drug had **no effect**. Under the null hypothesis, the
sampling distribution of the mean is centered at **1.2 seconds** with
standard error **0.05**.

A sample mean of **1.05 seconds** is **3 standard deviations** below the
null mean. In a normal distribution, values more than 3 SD from the mean
occur only **0.3%** of the time.

That probability (the **p‑value**) is so small that it is unlikely the
sample came from a population with mean 1.2 seconds. This leads to
rejecting the null hypothesis in favor of the alternative: the drug
changes response time.

🧪 Python code example (simulation + z‑test)




```python
import numpy as np
from scipy.stats import norm

# Given values
mu_null = 1.2
xbar = 1.05
s = 0.5
n = 100

# Standard error
SE = s / np.sqrt(n)

# Z-statistic
z = (xbar - mu_null) / SE

# Two-tailed p-value
p_value = 2 * norm.sf(abs(z))

print("Z-statistic:", z)
print("p-value:", p_value)

# Optional: simulate sampling distribution under H0
sim_means = np.random.normal(mu_null, s, size=(100000, n)).mean(axis=1)
sim_p = np.mean(np.abs(sim_means - mu_null) >= abs(xbar - mu_null))

print("Simulated p-value:", sim_p)
```

This code computes the z‑statistic, the theoretical p‑value, and a
simulation-based p‑value to confirm the result.
