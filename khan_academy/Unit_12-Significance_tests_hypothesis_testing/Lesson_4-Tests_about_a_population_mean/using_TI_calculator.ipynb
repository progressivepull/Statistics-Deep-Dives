# Example calculating t statistic for a test about a mean

[Example calculating t statistic for a test about a mean | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=BJl9WA787-Q&t=2s)

Rory suspects that teachers in his school district have less than 5
years of experience on average. He decided to test H₀ : μ = 5 versus Hₐ
: μ \< 5 using a sample of 25 teachers. His sample mean was 4 years and
his sample standard deviation was 2 years.

Rory wants to use these sample data to conduct a t test on the mean.
Assume that all conditions for inference have been met.

**Calculate the test statistic for Rory's test.**  
You may round your answer to two decimal places.

**Test statistic:** $t = - 2.5$

**🔹 Bullet summary**

- **Null hypothesis** $H_{0}$: $\mu = 5$years

- **Alternative hypothesis** $H_{a}$: $\mu < 5$years

- **Sample size**: $n = 25$

- **Sample mean**: $\overset{ˉ}{x} = 4$

- **Sample standard deviation**: $s = 2$

- **Standard error**: $SE = \frac{s}{\sqrt{n}} = \frac{2}{5} = 0.4$

- **Test statistic formula**:

$$
t = \frac{\overset{ˉ}{x} - \mu_{0}}{s/\sqrt{n}}
$$

- **Plug in values**:

$$
t = \frac{4 - 5}{0.4} = - 2.5
$$

**🔹 Explanation**

Rory wants to test whether teachers have **less than 5 years** of
experience on average. Because the **population standard deviation is
unknown**, he must use a **one-sample t‑test**.

The test statistic measures how many standard errors the sample mean is
below the hypothesized mean:

$$
t = \frac{\overset{ˉ}{x} - \mu_{0}}{s/\sqrt{n}}
$$

Compute the standard error:

$$
SE = \frac{s}{\sqrt{n}} = \frac{2}{\sqrt{25}} = \frac{2}{5} = 0.4
$$

Then compute the t‑value:

$$
t = \frac{4 - 5}{0.4} = \frac{- 1}{0.4} = - 2.5
$$

A t‑value of **–2.5** means the sample mean is **2.5 standard errors
below** the hypothesized mean of 5 years.

**🔹 Python code example**




```python
import scipy.stats as stats
import numpy as np

# Given values
xbar = 4
mu0 = 5
s = 2
n = 25

# Compute t statistic
SE = s / np.sqrt(n)
t_stat = (xbar - mu0) / SE

t_stat
```
