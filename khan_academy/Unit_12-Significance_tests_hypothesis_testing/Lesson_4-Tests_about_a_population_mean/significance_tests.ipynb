# When to use z or t statistics in significance tests

Below is a clean **bullet summary**, a clear **explanation**, and a
**Python code example** showing the difference between using a
**z‑statistic** and a **t‑statistic** in significance testing. I’ll also
embed guided links so you can explore concepts like the z statistic, the
t statistic, and the standard error.

**Bullet Summary**

**When to Use a Z Statistic**

- Used for **proportions**.

- Also used for **means** *only if* the **population standard
  deviation** $\sigma$is known (rare in practice).

- Sampling distribution standard deviation is known:

$$
\sqrt{\frac{p_{0}(1 - p_{0})}{n}}
$$

- Test statistic:

$$
z = \frac{\widehat{p} - p_{0}}{\sqrt{p_{0}(1 - p_{0})/n}}
$$

**When to Use a T Statistic**

- Used for **means** when the **population standard deviation is
  unknown** (almost always).

- Replace $\sigma$with the **sample standard deviation** $s$.

- Extra uncertainty → use the **t‑distribution** instead of the normal
  distribution.

- Test statistic:

$$
t = \frac{\overset{ˉ}{x} - \mu_{0}}{s/\sqrt{n}}
$$

**Why the Difference?**

- For proportions, the standard deviation of the sampling distribution
  is **fully determined** by $p_{0}$.

- For means, the population standard deviation is **unknown**, so we
  estimate it → introduces variability → use **t**.

**Explanation**

A significance test always compares a sample statistic to a hypothesized
population parameter. The question is: **which distribution should we
use to measure how far the sample statistic is from the null value?**

**Proportions → Z**

For a proportion test, the sampling distribution of $\widehat{p}$has a
known standard deviation under the null:

$$
\sqrt{\frac{p_{0}(1 - p_{0})}{n}}
$$

Because this is known exactly, we can standardize using a
**z‑statistic**.

**Means → T**

For a mean test, the sampling distribution of $\overset{ˉ}{x}$has
standard deviation:

$$
\frac{\sigma}{\sqrt{n}}
$$

But we almost never know $\sigma$. So we estimate it using the sample
standard deviation $s$, which adds uncertainty. To account for that
uncertainty, we use the **t‑distribution**, which has heavier tails.

As the sample size grows, the t‑distribution approaches the normal
distribution, so the t‑test and z‑test become nearly identical.

**Python Code Example — Z Test for a Proportion & T Test for a Mean**




```python
import math
from scipy.stats import norm, t

# -----------------------------
# Z TEST FOR A PROPORTION
# -----------------------------
p_null = 0.40
n = 200
x = 98
p_hat = x / n

# Standard error under the null
se_prop = math.sqrt(p_null * (1 - p_null) / n)

# Z-statistic
z_stat = (p_hat - p_null) / se_prop

# One-sided p-value (greater than)
p_value_z = 1 - norm.cdf(z_stat)

print("Z-test statistic:", z_stat)
print("Z-test p-value:", p_value_z)

# -----------------------------
# T TEST FOR A MEAN
# -----------------------------
mu_null = 50
xbar = 54
s = 12
n2 = 15

# Standard error using sample SD
se_mean = s / math.sqrt(n2)

# T-statistic
t_stat = (xbar - mu_null) / se_mean

# Degrees of freedom
df = n2 - 1

# Two-sided p-value
p_value_t = 2 * (1 - t.cdf(abs(t_stat), df))

print("T-test statistic:", t_stat)
print("T-test p-value:", p_value_t)
```

**What this code shows**

- The **z‑test** uses a known standard error from the null proportion.

- The **t‑test** uses the sample standard deviation and adjusts for
  degrees of freedom.

- Both compute test statistics and p‑values, but the underlying logic
  differs because of how uncertainty is handled.
