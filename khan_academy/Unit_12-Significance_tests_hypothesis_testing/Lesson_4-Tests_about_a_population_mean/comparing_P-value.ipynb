# Comparing P-value from t statistic to significance level

[Comparing P-value from t statistic to significance level | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=I5WSC_iRtwE&t=3s)

Jude was curious if the automated machine at his restaurant was filling
drinks with the proper amount. He filled a sample of 20 drinks to test
$H_{0}:\mu = 530\text{ mL}$versus $H_{a}:\mu \neq 530\text{ mL}$,
where $\mu$is the mean filling amount.

The drinks in the sample contained a mean amount of 528 mL with a
standard deviation of 4 mL. These results produced a test statistic of
$t = - 2.236$and a P-value of approximately 0.038.

Assuming the conditions for inference were met, what is an appropriate
conclusion at the $\alpha = 0.05$significance level?

Choose 1 answer:

A Reject $H_{0}$. This is strong evidence that the mean filling amount
is different than 530 mL.

B Reject $H_{0}$. This isn't enough evidence to conclude that the mean
filling amount is different than 530 mL.

C Fail to reject $H_{0}$. This is strong evidence that the mean
filling amount is different than 530 mL.

D Fail to reject $H_{0}$. This isn't enough evidence to conclude that
the mean filling amount is different than 530 mL.

**Conclusion:** Because the p‑value (0.038) is **less** than the
significance level α = 0.05, Jude should **reject the null hypothesis**.
There is statistically significant evidence that the true mean filling
amount is **different from 530 mL**.

**🔹 Bullet Summary**

- **Null hypothesis** $H_{0}:\mu = 530$mL

- **Alternative hypothesis** $H_{a}:\mu \neq 530$mL (two‑sided test)

- Sample size: $n = 20$

- Sample mean: $\overset{ˉ}{x} = 528$mL

- Sample standard deviation: $s = 4$mL

- Test statistic: $t = - 2.236$

- P‑value: $0.038$

- Significance level: $\alpha = 0.05$

- Since **0.038 \< 0.05**, reject $H_{0}$.

- Interpretation: There is **significant evidence** that the mean
  filling amount is **not 530 mL**.

**🔹 Explanation**

This is a **one‑sample t‑test** for a population mean. Jude wants to
know whether the automated machine is filling drinks correctly.

The **p‑value** represents the probability of observing a sample mean as
extreme as 528 mL (or more extreme), *assuming the true mean really is
530 mL*.

- If the p‑value is **less than α**, the observed result is unlikely
  under the null hypothesis.

- Here, $0.038 < 0.05$, so the sample provides enough evidence to
  conclude the machine is **not filling drinks at the target amount**.

Thus, Jude should **reject the null hypothesis** and conclude that the
true mean filling amount is **different from 530 mL**.

This matches the correct choice: **“Reject the null hypothesis. There is
strong evidence that the mean filling amount is different from 530
mL.”**

**🔹 Python Code Example**

This code reproduces the t‑statistic and p‑value for Jude’s test.




```python
import math
from scipy.stats import t

# Given values
mu_null = 530
xbar = 528
s = 4
n = 20

# Standard error
se = s / math.sqrt(n)

# t-statistic
t_stat = (xbar - mu_null) / se

# Degrees of freedom
df = n - 1

# Two-sided p-value
p_value = 2 * (1 - t.cdf(abs(t_stat), df))

print("t-statistic:", t_stat)
print("p-value:", p_value)
```
