# Conditions for a t test about a mean

[Conditions for a t test about a mean | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=GtokpL4f32s&t=3s)

Sunil and his friends have been using a group messaging app for over a year to chat with each other. He suspects that, on average, they send each other more than 100 messages per day.        

Sunil takes a random sample of 7 days from their chat history and records how many messages were sent on those days. The sample data are strongly skewed to the right with a mean of 125 messages and a standard deviation of 44 messages. He wants to use these sample data to conduct a t test about the mean.        
Which conditions for performing this type of significance test have been met?       

**Bullet summary**

- **Goal:** Sunil wants to test whether the **mean** number of messages
  per day is **greater than 100**.

- **Sample:** Random sample of **7 days**, mean
  $\overset{ˉ}{x} = 125$, standard deviation $s = 44$; data are
  **strongly right‑skewed**.

- **Random condition:** **Met** — the problem states he took a **random
  sample** of days.

- **Independence condition:** **Met** — 7 days is **less than 10%** of
  the \>365 days in the population, so sampling without replacement is
  acceptable.

- **Normal condition:** **Not met** — small sample size ($n = 7$) and
  **strongly skewed** data; no guarantee that the sampling distribution
  of $\overset{ˉ}{x}$is approximately normal.

- **Conclusion:** Only **random** and **independence** conditions are
  satisfied; the **normal condition is violated**, so a t‑test is **not
  well justified**.

**Explanation**

Sunil is interested in the **population mean** number of messages per
day, $\mu$. A natural setup would be:

- $H_{0}:\mu = 100$

- $H_{a}:\mu > 100$

He takes a **random sample** of 7 days and computes:

- Sample mean: $\overset{ˉ}{x} = 125$

- Sample standard deviation: $s = 44$

To use a **one‑sample t‑test**, we need three key conditions:

1.  **Random condition** The data should come from a **random sample**.
    The problem explicitly says he took a **random sample of seven
    days**, so this condition is met.

2.  **Independence condition** Observations should be (approximately)
    independent. A common rule: if sampling without replacement, the
    sample size should be **≤ 10%** of the population. Here, $n = 7$,
    population days \> 365 → $7 \leq 0.1 \times 365$. So independence
    is reasonable.

3.  **Normal condition** The sampling distribution of
    $\overset{ˉ}{x}$should be approximately normal. This can be
    justified if:

    - The population is normal, or

    - The sample size is large ($n \geq 30$), or

    - The sample data are roughly symmetric with no strong outliers.

In this case:

- $n = 7$(small)

- The sample data are **strongly skewed to the right** → None of the
  normality justifications hold.

So, **random** and **independence** are satisfied, but the **normal
condition fails**, making a t‑test unreliable here.

**Code example (t‑test anyway, and a note about conditions)**

Even though the normal condition is not met, here’s how a one‑sample
t‑test would be coded in Python. This is more to connect the theory to
practice—you’d still **question the validity** of the result because of
the skew and small $n$.




```python
import math
from scipy.stats import t

# Hypothesized mean
mu_null = 100

# Sample statistics (from the problem)
xbar = 125
s = 44
n = 7

# Standard error
se = s / math.sqrt(n)

# t-statistic for H0: mu = 100 vs Ha: mu > 100
t_stat = (xbar - mu_null) / se

# Degrees of freedom
df = n - 1

# One-sided p-value (greater than)
p_value = 1 - t.cdf(t_stat, df)

print("Sample mean:", xbar)
print("Sample SD:", s)
print("t-statistic:", t_stat)
print("p-value (one-sided):", p_value)
```

**Interpretation (with a big asterisk):**

- This code gives you the t‑statistic and p‑value for the test
  $H_{0}:\mu = 100$vs $H_{a}:\mu > 100$.

- However, because the **normal condition is violated** (small $n$,
  strongly skewed data), the p‑value should be treated with caution—it
  doesn’t rest on the usual t‑test assumptions.
