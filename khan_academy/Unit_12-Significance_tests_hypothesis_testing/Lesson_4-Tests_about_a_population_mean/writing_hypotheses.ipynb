# Writing hypotheses for a significance test about a mean

[Writing hypotheses for a significance test about a mean | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=89Dp9Xq4eaA&t=2s)

A quality control expert at a drink bottling factory took a random sample of bottles from a batch and measured the amount of liquid in each bottle in the sample. The amounts in the sample had a mean of 503 mL and a standard deviation of 5 mL. They want to test if this is convincing evidence that the mean amount for bottles in this batch is different than the target value of 500 mL.                  

Let μ be the mean amount of liquid in each bottle in the batch.                 

Write an appropriate set of hypotheses for their significance test.               

Here’s a clean **bullet summary**, a clear **explanation**, and a
**Python code example** for the quality‑control hypothesis test. I’ll
also embed guided links so you can explore concepts like the null
hypothesis, the alternative hypothesis, and the sampling distribution of
the mean.

**Bullet Summary**

- A quality‑control expert wants to test whether the **true mean**
  amount of liquid in bottles differs from the **target value** of 500
  mL.

- Let $\mu$= true mean amount of liquid in the batch.

- Sample of bottles:

  - Sample mean $\overset{ˉ}{x} = 503$mL

  - Sample standard deviation $s = 5$mL

- Appropriate hypotheses:

  - **Null hypothesis:** $H_{0}:\mu = 500$

  - **Alternative hypothesis:** $H_{a}:\mu \neq 500$

- This is a **two‑sided test** because the expert is checking for *any*
  difference, not specifically higher or lower.

**Explanation**

A significance test always compares two competing claims:

**1. Null hypothesis** $\mathbf{H}_{\mathbf{0}}$

This represents the “no news,” “status‑quo,” or “target” value. Here,
the bottling machine is *supposed* to fill bottles with **500 mL**, so:

$$
H_{0}:\mu = 500
$$

**2. Alternative hypothesis** $\mathbf{H}_{\mathbf{a}}$

This represents the possibility that something has changed. The expert
wants to know whether the mean is **different**, not specifically higher
or lower:

$$
H_{a}:\mu \neq 500
$$

This makes it a **two‑tailed test**, which looks for deviations in
either direction.

**Why the sample mean (503 mL) is *not* in the hypotheses**

Hypotheses are always about the **population parameter** $\mu$, not
the sample statistic $\overset{ˉ}{x}$. The sample mean is used
**afterward** to compute a test statistic and p‑value.

If you want to explore how to choose between one‑sided and two‑sided
tests, you can tap: One‑sided vs two‑sided

**Python Code Example — Compute the t‑statistic and p‑value**

Since the population standard deviation is unknown, this is a
**one‑sample t‑test**.




```python
import math
from scipy.stats import t

# Given values
mu_null = 500
xbar = 503
s = 5
n = 40  # sample size not given in the prompt, but needed for the test

# Standard error
se = s / math.sqrt(n)

# t-statistic
t_stat = (xbar - mu_null) / se

# Two-sided p-value
df = n - 1
p_value = 2 * (1 - t.cdf(abs(t_stat), df))

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

**What this code does**

- Computes the **t‑statistic** for testing $H_{0}:\mu = 500$.

- Computes the **two‑sided p‑value**.

- Uses the sample standard deviation because the population standard
  deviation is unknown.
