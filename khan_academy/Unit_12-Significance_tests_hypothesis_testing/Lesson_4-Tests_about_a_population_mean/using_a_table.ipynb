Miriam was testing H₀ : μ = 18 versus Hₐ : μ < 18 with a sample of 7 observations. Her test statistic was t = −1.9. Assume that the conditions for inference were met.       

What is the approximate P-value for Miriam's test?

The **approximate p‑value is about 0.053**, because with a **t‑statistic
of −1.9** and **df = 6**, the left‑tail probability under the t
distribution is roughly **5.3%**.

**🔹 Bullet Summary**

- Null hypothesis: $\mu = 18$

- Alternative hypothesis: $\mu < 18$(left‑tailed test)

- Sample size: $n = 7$→ degrees of freedom $df = 6$

- Test statistic: $t = - 1.9$

- p‑value = $P(T \leq - 1.9)$for a t distribution with 6 df

- Approximate p‑value: **0.053**

- Interpretation: If α = 0.05, this is *slightly above* α → **fail to
  reject** the null.

**🔹 Explanation**

Because Miriam is testing a **left‑tailed** hypothesis ($\mu < 18$),
the p‑value is the probability of getting a t‑statistic **less than
−1.9** under the assumption that the null hypothesis is true.

We use the t distribution with

$$
df = n - 1 = 6.
$$

So the p‑value is:

$$
p = P(T_{6} \leq - 1.9) \approx 0.053.
$$

This is just slightly above the common α = 0.05 threshold, so the
evidence is *borderline but not quite strong enough* to reject the null.

**🔹 Code Example (Python)**




```python
import scipy.stats as stats

t_stat = -1.9
df = 6

p_value = stats.t.cdf(t_stat, df)
p_value
```
