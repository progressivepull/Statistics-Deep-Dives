# Free response example: Significance test for a mean

[Free response example: Significance test for a mean | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=8nCsnsD8Fnw&t=1s)

Regulations require that product labels on containers of food that are available for sale to the public accurately state the amount of food in those containers. Specifically, if milk containers are labeled to have 128 fluid ounces and the mean number of fluid ounces of milk in the containers is at least 128, the milk processor is considered to be in compliance with the regulations. The filling machines can be set to the labeled amount. Variability in the filling process causes the actual contents of milk containers to be normally distributed. A random sample of 12 containers of milk was drawn from the milk processing line in a plant, and the amount of milk in each container was recorded.      

(a) The sample mean and standard deviation of this sample of 12 containers of milk were 127.2 ounces and 2.1 ounces, respectively. Is there sufficient evidence to conclude that the packaging plant is not in compliance with the regulations? Provide statistical justification for your answer.

**Bullet summary**

- **Parameter:** $\mu$= true mean amount of milk (fl oz) in containers
  at the plant

- **Null hypothesis** $H_{0}$**:** $\mu \geq 128$(plant is in
  compliance)

- **Alternative hypothesis** $H_{a}$**:** $\mu < 128$(plant is not
  in compliance)

- **Sample size:** $n = 12$

- **Sample mean:** $\overset{ˉ}{x} = 127.2$

- **Sample standard deviation:** $s = 2.1$

- **Test type:** One-sample, **left-tailed t-test**

- **Test statistic:**

$$
t = \frac{\overset{ˉ}{x} - 128}{s/\sqrt{n}} \approx \frac{127.2 - 128}{2.1/\sqrt{12}} \approx - 1.32
$$

- **Degrees of freedom:** $df = 11$

- **P-value:** $P(T \leq - 1.32) \approx 0.107$

- **Significance level:** $\alpha = 0.05$

- **Decision:** $0.107 > 0.05 \Rightarrow$**fail to reject** $H_{0}$

- **Conclusion:** There is **not sufficient statistical evidence** to
  conclude the plant is **out of compliance**.

**Explanation**

We want to know if the plant is **not in compliance**, which happens
when the true mean fill is **less than 128 fl oz**. That’s why the
alternative hypothesis is $H_{a}:\mu < 128$, making this a **one-sided
(left-tailed) t-test**.

Because the population standard deviation is unknown and the sample size
is small ($n = 12$), we use a **t-test**:

$$
t = \frac{\overset{ˉ}{x} - \mu_{0}}{s/\sqrt{n}} = \frac{127.2 - 128}{2.1/\sqrt{12}} \approx - 1.32
$$

With $df = 11$, we compute the **left-tail probability**
$P(T \leq - 1.32)$, which is the p-value for this one-sided test. That
p-value is approximately $0.107$.

We compare:

- **p-value = 0.107**

- $\alpha = 0.05$

Since $0.107 > 0.05$, the sample result is **not extreme enough**
(under the assumption $\mu = 128$or higher) to reject the null
hypothesis. Statistically, we **cannot conclude** that the mean fill is
below 128 fl oz.

So, in regulatory language:

There is **not sufficient evidence** to say the plant is **not in
compliance** with the labeling requirement.

**Python code example**

Below is a Python snippet that computes the t-statistic and one-sided
p-value using scipy.stats:




```python
import math
from scipy.stats import t

# Given data
mu0 = 128        # hypothesized minimum mean (compliance threshold)
xbar = 127.2     # sample mean
s = 2.1          # sample standard deviation
n = 12           # sample size

# Standard error
se = s / math.sqrt(n)

# t-statistic
t_stat = (xbar - mu0) / se

# Degrees of freedom
df = n - 1

# One-sided (left-tailed) p-value: P(T <= t_stat)
p_value = t.cdf(t_stat, df)

print("t-statistic:", t_stat)
print("p-value (left-tailed):", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject H0: evidence plant is NOT in compliance.")
else:
    print("Fail to reject H0: NOT enough evidence of non-compliance.")
```

You could easily tweak this to simulate repeated samples and see how
often you’d get a t-statistic as low as this under $\mu = 128$, if you
want to connect theory to empirical behavior.
