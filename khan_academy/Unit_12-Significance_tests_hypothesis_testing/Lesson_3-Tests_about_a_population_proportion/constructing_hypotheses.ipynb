# Constructing hypotheses for a significance test about a proportion

Amanda read a report saying that 49% of teachers in the United States were members of a labor union. She wants to test whether this holds true for teachers in her state, so she is going to take a random sample of these teachers and see what percent of them are members of a union.

Let p represent the proportion of teachers in her state that are members of a union.

Write an appropriate set of hypotheses for her significance test.

Here’s a clean **bullet summary**, a clear **explanation**, and a
**Python code example** to reinforce how to write hypotheses for a
significance test. I’ll also embed guided links so you can explore

**Bullet Summary**

**Example 1 — Amanda Testing Union Membership**

- National report: **49%** of teachers belong to a union.

- Amanda wants to test whether her state’s proportion differs.

- Let $P$= true proportion of teachers in her state who are union
  members.

- Appropriate hypotheses:

  - **Null hypothesis**: $H_{0}:P = 0.49$

  - **Alternative hypothesis**: $H_{a}:P \neq 0.49$

- This is a **two‑sided test** because she is checking whether the state
  proportion is *different*, not specifically higher or lower.

**Example 2 — Internet Access in California**

- Old poll: **90%** of homes had internet access.

- Researchers want to test whether the proportion is now **higher**.

- Sample: 920 out of 1000 homes → sample proportion = 0.92.

- Let $P$= true proportion of California homes with internet access.

- Appropriate hypotheses:

  - **Null hypothesis**: $H_{0}:P = 0.90$

  - **Alternative hypothesis**: $H_{a}:P > 0.90$

- This is a **right‑tailed test** because they are testing for an
  increase.

**Explanation**

A significance test always compares two competing claims:

- The null hypothesis $H_{0}$: The “no change,” “no difference,” or
  “status quo” claim.

- The alternative hypothesis $H_{a}$: The claim that represents a
  meaningful difference or effect.

A crucial rule: **Hypotheses are always written about the true
population parameter**, not the sample statistic.

So even though Amanda observes a sample proportion later, the hypotheses
must be about the true proportion $P$, not the sample proportion
$\widehat{p}$.

Similarly, in the internet example, the 92% sample result is *not* used
in the hypotheses—it is only used to compute a test statistic and
p‑value.

**Python Code Example — Testing a Proportion**

Here’s a simple example showing how to perform a one‑sample proportion
test in Python using the internet‑access scenario.




```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# Sample data
successes = 920       # homes with internet
n = 1000              # total homes sampled
p_null = 0.90         # null hypothesis proportion

# One-sided test: H_a: p > 0.90
z_stat, p_value = proportions_ztest(successes, n, p_null, alternative='larger')

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

**What this code does**

- Computes the z‑statistic for the sample proportion.

- Computes the p‑value under the assumption that $P = 0.90$.

- If the p‑value is below the chosen significance level (e.g., 0.05), we
  reject $H_{0}$and conclude the proportion is likely higher.
