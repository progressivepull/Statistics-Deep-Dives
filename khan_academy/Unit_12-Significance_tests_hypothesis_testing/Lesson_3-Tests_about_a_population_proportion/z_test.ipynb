# Conditions for a z test about a proportion

[Conditions for a z test about a proportion | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=5JlJ4Qfk8g4&t=2s)

Jules works on a small team of 40 employees. Each employee receives an
annual rating, the best of which is “exceeds expectations.” Management
claimed that 10% of employees earn this rating, but Jules suspected it
was actually less common. She obtained an anonymous random sample of 10
ratings for employees on her team. She wants to use the sample data to
test $H_{0}:p = 0.1$versus $H_{a}:p < 0.1$, where $p$is the
proportion of all employees on her team who earned “exceeds
expectations.”

Which conditions for performing this type of test did Jules’ sample
meet?


Below is a clean **bullet summary**, a clear **explanation**, and a
**Python code example** that mirrors Jules’s scenario. I’ll also embed
guided links so you can explore concepts like the random condition, the
normal condition, and the independence condition.

**Bullet Summary**

**Jules’s Scenario**

- Population size: **40 employees**

- Sample size: **10 employees**

- Hypotheses:

  - $H_{0}:p = 0.10$

  - $H_{a}:p < 0.10$

**Condition Checks**

- **Random Condition** → ✔ Met

  - She obtained a **random sample** of 10 ratings.

- **Normal Condition** → ✘ Not met

  - Expected successes: $10 \cdot 0.10 = 1$

  - Expected failures: $10 \cdot 0.90 = 9$

  - Both must be ≥ 10 for a normal approximation → fails.

- **Independence Condition** → ✘ Not met

  - Sample without replacement must satisfy the **10% rule**:
    $10 \leq 0.10 \cdot 40 = 4$→ but 10 \> 4

  - Sample is **25%** of population → too large to assume independence.

**Conclusion**

- Jules meets **only the random condition**.

- She **does not** meet the normal or independence conditions.

- A normal‑approximation z‑test for a proportion is **not valid** here.

**Explanation**

A significance test for a proportion relies on the sampling distribution
of the sample proportion $\widehat{p}$. To justify using a normal
model, three conditions must be satisfied:

**1. Random Condition**

Ensures the sample is representative. Jules used a **random sample**, so
this condition is satisfied.

**2. Normal Condition (Large Counts Condition)**

Requires:

$$
np_{0} \geq 10\text{and}n(1 - p_{0}) \geq 10
$$

Using the null value $p_{0} = 0.10$:

- Expected successes: $10 \cdot 0.10 = 1$

- Expected failures: $10 \cdot 0.90 = 9$

Both are **far below 10**, so the sampling distribution of
$\widehat{p}$is **not approximately normal**.

**3. Independence Condition**

Either:

- Sampling **with replacement**, or

- Sample size ≤ 10% of population.

Here:

$$
\frac{10}{40} = 0.25 = 25\%
$$

This violates the **10% rule**, so independence cannot be assumed.

Because two of the three conditions fail, a normal‑approximation test
(z‑test) is **not appropriate**.

If you want to explore alternatives like an **exact binomial test**, you
can tap: Exact test

**Python Code Example — Exact Binomial Test**

Even though the normal approximation fails, we *can* run an exact
binomial test, which does **not** require the normal condition.




```python
from scipy.stats import binomtest

# Jules's scenario
n = 10               # sample size
successes = 0        # example: suppose 0 employees got "exceeds expectations"
p_null = 0.10        # null hypothesis proportion

# One-sided test: H_a: p < 0.10
result = binomtest(successes, n, p_null, alternative='less')

print("P-value:", result.pvalue)
print("Test statistic (successes):", successes)
```

**What this code does**

- Performs an **exact** test using the binomial distribution.

- Computes the probability of observing the sample result (or something
  more extreme) under $H_{0}$.

- Avoids the invalid normal approximation.
