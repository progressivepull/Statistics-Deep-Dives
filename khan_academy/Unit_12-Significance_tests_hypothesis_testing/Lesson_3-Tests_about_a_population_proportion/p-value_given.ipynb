# Calculating a P-value given a z statistic 

[Calculating a P-value given a z statistic | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=NNpv-n_Hbvc&t=2s)

Fay read an article that said 26% of Americans can speak more than one
language. She was curious if this figure was higher in her city, so she
tested $H_{0}:p = 0.26$vs. $H_{a}:p > 0.26$, where $p$represents
the proportion of people in her city that can speak more than one
language.

She found that 40 of 120 people sampled could speak more than one
language. The test statistic for these results was $z \approx 1.83$.

Assuming that the necessary conditions are met, what is the approximate
P-value for Fay’s test?


Below is a clean **bullet summary**, a clear **explanation**, and a
**Python code example** that walks through Fay’s hypothesis test, the
z‑statistic, and the p‑value. I’ll also embed guided links so you can
explore concepts like the p‑value, the z‑statistic, and the sampling
distribution.

**Bullet Summary**

**Given**

- National bilingual rate: **26%**

- Fay’s sample: **40 out of 120** →

$$
\widehat{p} = \frac{40}{120} = \frac{1}{3} \approx 0.33
$$

- Hypotheses:

  - $H_{0}:p = 0.26$

  - $H_{a}:p > 0.26$

- Test statistic:

$$
z \approx 1.83
$$

**Goal**

Find the **p‑value** for a **right‑tailed** test:

$$
P(Z \geq 1.83)
$$

**Result**

Using the standard normal distribution:

$$
P(Z \geq 1.83) \approx 0.0336
$$

So the approximate p‑value is **0.0336**.

**Explanation**

Fay wants to know whether the proportion of multilingual speakers in her
city is **greater** than the national value of 26%. That makes this a
**one‑sided, right‑tailed** test.

**1. Compute the sample proportion**

$$
\widehat{p} = \frac{40}{120} = 0.3333
$$

**2. Compute the z‑statistic**

$$
z = \frac{\widehat{p} - p_{0}}{\sqrt{p_{0}(1 - p_{0})/n}}
$$

Plugging in:

- $p_{0} = 0.26$

- $n = 120$

$$
z = \frac{0.33 - 0.26}{\sqrt{0.26 \cdot 0.74/120}} \approx 1.83
$$

This means the sample proportion is **1.83 standard deviations above**
the null hypothesis proportion.

**3. Compute the p‑value**

Because this is a right‑tailed test:

$$
p\text{-value} = P(Z \geq 1.83)
$$

Using the symmetry of the normal curve:

$$
P(Z \geq 1.83) = P(Z \leq - 1.83) \approx 0.0336
$$

**4. Interpretation**

- If Fay’s significance level is **5%**, then

$$
0.0336 < 0.05
$$

→ **Reject** $H_{0}$→ Evidence suggests the proportion is higher than
26%.

- If her significance level is **1%**, then

$$
0.0336 > 0.01
$$

→ **Fail to reject** $H_{0}$→ Not enough evidence at that stricter
threshold.

If you want to explore how the significance level affects decisions, you
can tap that link.

**Python Code Example — Compute z‑statistic and p‑value**

Here’s a Python snippet that reproduces Fay’s result:




```python
import math
from scipy.stats import norm

# Given values
p_null = 0.26
n = 120
x = 40
p_hat = x / n

# Standard error under the null
se = math.sqrt(p_null * (1 - p_null) / n)

# Z-test statistic
z = (p_hat - p_null) / se

# One-sided p-value (greater than)
p_value = 1 - norm.cdf(z)

print("Sample proportion:", p_hat)
print("Z-statistic:", z)
print("P-value:", p_value)
```

**What this code does**

- Computes the sample proportion.

- Computes the standard error under the null.

- Computes the z‑statistic.

- Computes the **right‑tailed** p‑value.
