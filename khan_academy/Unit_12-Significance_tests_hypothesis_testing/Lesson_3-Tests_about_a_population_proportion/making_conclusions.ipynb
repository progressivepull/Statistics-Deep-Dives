# Making conclusions in a test about a proportion

[Making conclusions in a test about a proportion | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=RcjZwZJO5qo)

A public opinion survey investigated whether a majority (more than 50%)
of adults supported a tax increase to help fund the local school system.
A random sample of 200 adults showed that 113 of those sampled supported
the tax increase.

Researchers used these results to test H₀ : p = 0.5 versus Hₐ : p \>
0.5, where p is the true proportion of adults that support the tax
increase. They calculated a test statistic of z ≈ 1.84 and a
corresponding P-value of approximately 0.033.

**(A)** At the α = 0.01 significance level, they should conclude that
more than 50% of adults support the tax increase.

**(B)** At the α = 0.01 significance level, they should conclude that
less than 50% of adults support the tax increase.

**(C)** At the α = 0.05 significance level, they should conclude that
more than 50% of adults support the tax increase.

**(D)** At the α = 0.05 significance level, they should conclude that
less than 50% of adults support the tax increase.


**Bullet summary**

- **Context:** Survey of 200 adults; 113 support a tax increase →
  $\widehat{p} = 113/200 = 0.565$.

- **Hypotheses:**

  - **Null:** $H_{0}:p = 0.50$

  - **Alternative:** $H_{a}:p > 0.50$(testing if a majority supports
    the tax).

- **Test statistic:** $z \approx 1.84$.

- **P-value:** $p \approx 0.033$(right‑tailed).

- **Decision at** $\alpha = 0.01$**:** $p > \alpha$→ **fail to
  reject** $H_{0}$; cannot conclude majority support.

- **Decision at** $\alpha = 0.05$**:** $p < \alpha$→ **reject**
  $H_{0}$; conclude **more than 50%** support the tax.

- **Appropriate conclusion (from choices):** At $\alpha = 0.05$, they
  should conclude that **more than 50% of adults support the tax
  increase**.

**Explanation**

We’re testing whether **more than half** of adults support a tax
increase:

- $H_{0}:p = 0.50$(exactly half support it)

- $H_{a}:p > 0.50$(a majority supports it)

From the sample:

$$
\widehat{p} = \frac{113}{200} = 0.565
$$

The z‑statistic measures how many standard deviations $\widehat{p}$is
above the null value $p_{0} = 0.50$. They’ve already computed:

$$
z \approx 1.84
$$

Using the standard normal distribution, the **right‑tailed** p‑value is:

$$
p = P(Z \geq 1.84) \approx 0.033
$$

Now compare this p‑value to different significance levels:

- **At** $\alpha = 0.01$**:** $0.033 > 0.01$→ not small enough →
  **fail to reject** $H_{0}$. We **cannot** confidently claim that
  more than 50% support the tax at this stricter level.

- **At** $\alpha = 0.05$**:** $0.033 < 0.05$→ small enough →
  **reject** $H_{0}$. We have sufficient evidence to conclude that
  **more than 50%** of adults support the tax increase.

So the appropriate conclusion is: **At the 5% significance level, they
should conclude that more than 50% of adults support the tax increase.**

**Code example (compute z, p-value, and decisions)**




```python
import math
from scipy.stats import norm

# Given data
n = 200
x = 113
p_null = 0.50

# Sample proportion
p_hat = x / n

# Standard error under the null
se = math.sqrt(p_null * (1 - p_null) / n)

# Z-statistic
z = (p_hat - p_null) / se

# One-sided p-value (H_a: p > 0.5)
p_value = 1 - norm.cdf(z)

print("Sample proportion:", p_hat)
print("Z-statistic:", z)
print("P-value:", p_value)

# Decisions at alpha = 0.01 and 0.05
for alpha in [0.01, 0.05]:
    decision = "Reject H0" if p_value < alpha else "Fail to reject H0"
    print(f"Alpha = {alpha}: {decision}")
```

This code:

- Recomputes $\widehat{p}$, $z$, and the **right‑tailed** p‑value.

- Shows how the decision changes at $\alpha = 0.01$vs
  $\alpha = 0.05$.
