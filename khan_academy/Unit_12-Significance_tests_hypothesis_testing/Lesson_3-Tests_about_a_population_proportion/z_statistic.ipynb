# Calculating a z statistic in a test about a proportion

[Calculating a z statistic in a test about a proportion | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=DfpFfIdwcIo&t=5s)

The mayor of a town saw an article that claimed the national
unemployment rate is 8%. They wondered if this held true in their town,
so they took a sample of 200 residents to test  
$H_{0}:p = 0.08$versus $H_{a}:p \neq 0.08$, where $p$is the
proportion of residents in the town that are unemployed. The sample
included 22 residents who were unemployed.

Assuming that the conditions for inference have been met, identify the
correct test statistic for this significance test.

Here’s a clean **bullet summary**, a clear **explanation**, and a
**Python code example** that walks through the test statistic for the
mayor’s unemployment‑rate hypothesis test. I’ll also embed guided links
so you can explore concepts like the test statistic, the sampling
distribution, and the z‑test for proportions.

**Bullet Summary**

**Given**

- National unemployment rate: **8%**

- Sample size: **200**

- Sample unemployed: **22**

- Sample proportion:

$$
\widehat{p} = \frac{22}{200} = 0.11
$$

- Hypotheses:

  - $H_{0}:p = 0.08$

  - $H_{a}:p \neq 0.08$

**Goal**

Compute the **z‑test statistic** for a one‑sample test of a proportion.

**Test Statistic Formula**

$$
z = \frac{\widehat{p} - p_{0}}{\sqrt{p_{0}(1 - p_{0})/n}}
$$

**Plugging in values**

- $p_{0} = 0.08$

- $\widehat{p} = 0.11$

- $n = 200$

$$
z = \frac{0.11 - 0.08}{\sqrt{0.08 \cdot 0.92/200}}
$$

This z‑value tells us how many standard deviations the sample proportion
is from the null proportion.

**Explanation**

A significance test for a proportion compares the observed sample
proportion $\widehat{p}$to the hypothesized population proportion
$p_{0}$. To do that, we standardize the difference using the
**standard deviation of the sampling distribution**, which under the
null is:

$$
\sqrt{\frac{p_{0}(1 - p_{0})}{n}}
$$

This gives us a **z‑statistic**, which measures how far the sample
result is from the null in standard‑deviation units. Once we have the
z‑value, we can compute a **p‑value** and compare it to the chosen
significance level.

If you want to explore how the p‑value is computed from the z‑statistic,
you can tap that link.

**Python Code Example — Computing the Test Statistic**

Here’s a simple Python snippet that computes the z‑statistic and p‑value
for this exact scenario.




```python
import math
from scipy.stats import norm

# Given values
p_null = 0.08
n = 200
x = 22
p_hat = x / n

# Standard error under the null
se = math.sqrt(p_null * (1 - p_null) / n)

# Z-test statistic
z = (p_hat - p_null) / se

# Two-sided p-value
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Sample proportion:", p_hat)
print("Z-statistic:", z)
print("P-value:", p_value)
```

**What this code does**

- Computes the sample proportion.

- Computes the standard error under the null.

- Computes the z‑statistic.

- Computes the **two‑sided** p‑value (because $H_{a}:p \neq 0.08$).
