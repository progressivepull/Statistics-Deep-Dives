# Effects of Shifting and Scaling Data on Statistical Parameters

Suppose we have a **population dataset** representing the **ages of students**, stored in a spreadsheet (such as Microsoft Excel or Google Sheets). We want to calculate several **statistical parameters** and examine how those parameters change when the data are **shifted** or **scaled**.

The parameters we will examine are:
- Mean  
- Standard deviation  
- Median  
- Interquartile range (IQR)  

---

## Measures of Central Tendency

### Mean

The **mean** is calculated using the spreadsheet `AVERAGE` function.

#### Effect of Shifting the Data
If we add a constant value (for example, **+5**) to every data point:
- Each value increases by 5
- The **mean also increases by 5**

This happens because the mean is based on the sum of all values divided by the number of values. Adding the same constant to every value increases the total sum proportionally.

#### Effect of Scaling the Data
If we multiply every data point by a constant (for example, **×5**):
- Each value becomes five times larger
- The **mean also becomes five times larger**

**Conclusion:**  
The mean **shifts** when data are shifted and **scales** when data are scaled.

---

### Median

The **median** is the middle value when the data are ordered.

#### Effect of Shifting the Data
- Adding a constant to every value preserves the order
- The middle value increases by the same constant

#### Effect of Scaling the Data
- Multiplying all values preserves order
- The median is multiplied by the same factor

**Conclusion:**  
Like the mean, the median **shifts and scales** in the same way as the data.

---

## Measures of Spread

### Standard Deviation

The **standard deviation** measures how far data values are spread from the mean.

#### Effect of Shifting the Data
- Shifting all values by the same amount does **not change** distances from the mean
- The **standard deviation remains the same**

#### Effect of Scaling the Data
- Scaling increases or decreases distances from the mean
- The **standard deviation scales by the same factor**

**Conclusion:**  
Standard deviation is **unchanged by shifts** but **scales with the data**.

---

### Interquartile Range (IQR)

The **interquartile range** is defined as:

$$
\text{IQR} = Q_3 - Q_1
$$

It measures the spread of the **middle 50%** of the data.

#### Effect of Shifting the Data
- Both $ Q_1 $ and $ Q_3 $ increase by the same amount
- Their difference remains unchanged

#### Effect of Scaling the Data
- Both quartiles are multiplied by the same factor
- The difference between them scales accordingly

**Conclusion:**  
The IQR **does not change when data are shifted**, but **scales when data are scaled**.

---

## Key Takeaways

- **Mean and median (central tendency):**
  - Shift when data are shifted
  - Scale when data are scaled

- **Standard deviation and IQR (spread):**
  - Do **not** change when data are shifted
  - **Scale proportionally** when data are scaled

These properties hold regardless of the specific number used for shifting or scaling (addition, subtraction, multiplication, or division).


# Below is a Python example for a Jupyter Notebook that:

1. Computes **mean, median, standard deviation, and IQR**
2. Creates **shifted (+5)** and **scaled (×5)** versions of the data
3. **Plots the distributions** to visually show the effects of shifting and scaling


# Python Code: Shifting and Scaling Data

```python
import numpy as np
import matplotlib.pyplot as plt

# Original population data (example: student ages)
data = np.array([7, 8, 9, 9, 10, 10, 11, 11, 12, 12, 13, 14])

# Shifted and scaled datasets
data_shifted = data + 5
data_scaled = data * 5

# Function to compute statistics
def summary_stats(x):
    mean = np.mean(x)
    median = np.median(x)
    std = np.std(x)          # population standard deviation
    q1 = np.percentile(x, 25)
    q3 = np.percentile(x, 75)
    iqr = q3 - q1
    return mean, median, std, iqr

# Print statistics
labels = ["Original", "Shifted (+5)", "Scaled (×5)"]
datasets = [data, data_shifted, data_scaled]

for label, d in zip(labels, datasets):
    mean, median, std, iqr = summary_stats(d)
    print(f"{label} Data")
    print(f"  Mean: {mean:.2f}")
    print(f"  Median: {median:.2f}")
    print(f"  Std Dev: {std:.2f}")
    print(f"  IQR: {iqr:.2f}\n")
```

# Plot 1: Original vs Shifted Data

```python
plt.figure()
plt.hist(data, bins=6, alpha=0.7)
plt.hist(data_shifted, bins=6, alpha=0.7)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Original Data vs Shifted Data (+5)")
plt.show()
```

### What this shows:

* The entire distribution moves right
* TShape and spread remain the same
* TStandard deviation and IQR are unchanged


# Plot 2: Original vs Scaled Data

```python
plt.figure()
plt.hist(data, bins=6, alpha=0.7)
plt.hist(data_scaled, bins=6, alpha=0.7)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Original Data vs Scaled Data (×5)")
plt.show()
```

## What this shows:

* The distribution stretches horizontally
* Spread increases
* Standard deviation and IQR increase proportionally


### Key Visual Takeaways

* **Shifting data** moves the distribution but does not change its spread
* **Scaling data** stretches or compresses the distribution
* Measures of **center** move and scale with the data
* Measures of **spread** only change when scaling occurs
