# Comparing LSAT and MCAT Performance Using Z-Scores

Before applying to professional schools in the United States, students must take standardized exams:
- **LSAT** for law school
- **MCAT** for medical school

Because these exams use **different scoring scales and distributions**, raw scores cannot be directly compared. Instead, we compare performance using **z-scores**, which measure how many standard deviations a score is from the mean.

---

## Given Summary Statistics

### LSAT
- **Mean (μ):** 151  
- **Standard Deviation (σ):** 10  
- **Juwan’s Score:** 172  

### MCAT
- **Mean (μ):** 25.1  
- **Standard Deviation (σ):** 6.4  
- **Juwan’s Score:** 37  

---

## Z-Score Formula

$$
z = \frac{x - \mu}{\sigma}
$$

Where:
- $ x $ = individual score  
- $ \mu $ = mean  
- $ \sigma $ = standard deviation  

---

## LSAT Z-Score

$$
z_{LSAT} = \frac{172 - 151}{10} = \frac{21}{10} = 2.1
$$

- Juwan scored **2.1 standard deviations above the mean** on the LSAT.

---

## MCAT Z-Score

$$
z_{MCAT} = \frac{37 - 25.1}{6.4} = \frac{11.9}{6.4} \approx 1.86
$$

- Juwan scored **approximately 1.86 standard deviations above the mean** on the MCAT.

---

## Comparison and Interpretation

- **LSAT z-score:** 2.1  
- **MCAT z-score:** 1.86  

Because Juwan’s LSAT z-score is higher, he performed **relatively better on the LSAT** than on the MCAT.

That said, the scores are **quite close**, and in everyday terms his performance on both exams would be considered **comparable**, with strong results well above average in each case.

---

## Final Conclusion

Juwan did **slightly better on the LSAT** relative to other test takers, though his performance on both exams was similarly strong.


# Below is a Python code example:

1. Computes **z-scores** for the LSAT and MCAT
2. Plots **both normal distributions**
3. Shows **where Juwan’s scores fall** on each distribution

Each exam is plotted separately for clarity, using standard matplotlib defaults.

# Python Code: LSAT and MCAT Comparison

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Exam statistics
lsat_mean, lsat_std, lsat_score = 151, 10, 172
mcat_mean, mcat_std, mcat_score = 25.1, 6.4, 37

# Calculate z-scores
lsat_z = (lsat_score - lsat_mean) / lsat_std
mcat_z = (mcat_score - mcat_mean) / mcat_std

print(f"LSAT z-score: {lsat_z:.2f}")
print(f"MCAT z-score: {mcat_z:.2f}")

# ---------- LSAT Plot ----------
x_lsat = np.linspace(lsat_mean - 4*lsat_std, lsat_mean + 4*lsat_std, 1000)
y_lsat = norm.pdf(x_lsat, lsat_mean, lsat_std)

plt.figure()
plt.plot(x_lsat, y_lsat)
plt.axvline(lsat_score)
plt.xlabel("LSAT Score")
plt.ylabel("Probability Density")
plt.title("LSAT Score Distribution")

plt.show()

# ---------- MCAT Plot ----------
x_mcat = np.linspace(mcat_mean - 4*mcat_std, mcat_mean + 4*mcat_std, 1000)
y_mcat = norm.pdf(x_mcat, mcat_mean, mcat_std)

plt.figure()
plt.plot(x_mcat, y_mcat)
plt.axvline(mcat_score)
plt.xlabel("MCAT Score")
plt.ylabel("Probability Density")
plt.title("MCAT Score Distribution")

plt.show()
```

# What this code demonstrates

* **Z-score calculation** to compare relative performance across exams
* **Two normal distributions**, each centered at its exam’s mean
* **Vertical lines** showing Juwan’s score on each curve
* Clear visual evidence that:
    - LSAT score is farther into the right tail than MCAT
    - LSAT performance is **relatively stronger**


# Interpretation Reminder

* LSAT z-score ≈ 2.10
* MCAT z-score ≈ 1.86

Since the LSAT z-score is higher, Juwan performed **slightly better relative to other test takers on the LSAT**.
