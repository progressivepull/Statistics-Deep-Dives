# Z-Scores and the Normal Distribution  
*CK-12 AP Statistics FlexBook – Problem 2*

## Problem Overview

The grades on a statistics midterm at a high school are **normally distributed** with:

- **Mean (μ):** 81  
- **Standard deviation (σ):** 6.3  

We are asked to **calculate the z-scores** for several exam grades and **interpret their positions** on a normal distribution.

---

## What Is a Z-Score?

A **z-score** measures how many **standard deviations** a value is from the mean.

The formula is:

$$
z = \frac{x - \mu}{\sigma}
$$

Where:
- $ x $ = exam score  
- $ \mu $ = mean  
- $ \sigma $ = standard deviation  

- A **negative z-score** means the score is **below the mean**
- A **positive z-score** means the score is **above the mean**

---

## Sketch of the Distribution

- The distribution is **normal (bell-shaped)**.
- The **mean** is at **81**.
- One standard deviation above the mean is:
  $$
  81 + 6.3 = 87.3
  $$
- One standard deviation below the mean is:
  $$
  81 - 6.3 = 74.7
  $$

Additional standard deviations extend symmetrically outward from the mean.

---

## Z-Score Calculations

### Part A: Score = 65

$$
z = \frac{65 - 81}{6.3} = \frac{-16}{6.3} \approx -2.54
$$

- The score is **2.54 standard deviations below the mean**
- This places it far into the **left tail** of the distribution

---

### Part B: Score = 83

$$
z = \frac{83 - 81}{6.3} = \frac{2}{6.3} \approx 0.32
$$

- The score is **0.32 standard deviations above the mean**
- Slightly above average

---

### Part C: Score = 93

$$
z = \frac{93 - 81}{6.3} = \frac{12}{6.3} \approx 1.90
$$

- The score is **1.9 standard deviations above the mean**
- Well above average

---

### Part D: Score = 100

$$
z = \frac{100 - 81}{6.3} = \frac{19}{6.3} \approx 3.02
$$

- The score is **just over 3 standard deviations above the mean**
- This lies deep in the **right tail** of the distribution
- Such scores are **rare** in a normal distribution

---

## Interpretation

- Z-scores provide a **standardized way** to compare exam scores.
- Scores far from the mean (large absolute z-scores) correspond to **low-probability outcomes**.
- A score near the mean has a z-score close to **0**.
- Scores beyond **±3 standard deviations** are considered **extreme**.

---

## Summary Table

| Score | Z-Score | Interpretation |
|------|--------|---------------|
| 65   | -2.54  | Far below average |
| 83   | 0.32   | Slightly above average |
| 93   | 1.90   | Well above average |
| 100  | 3.02   | Extremely high score |

---

This problem illustrates how z-scores quantify **relative standing** within a normal distribution. In later applications, z-scores can be used to determine **probabilities and percentiles**.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Given parameters
mean = 81
std_dev = 6.3
scores = np.array([65, 83, 93, 100])

# Calculate z-scores
z_scores = (scores - mean) / std_dev

# Print z-scores
for score, z in zip(scores, z_scores):
    print(f"Score: {score}, Z-score: {z:.2f}")

# Generate x values for the normal curve
x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
y = norm.pdf(x, mean, std_dev)

# Plot the normal distribution
plt.figure()
plt.plot(x, y)
plt.xlabel("Exam Score")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Statistics Midterm Scores")

# Plot vertical lines for each score
for score in scores:
    plt.axvline(score)

plt.show()
```
