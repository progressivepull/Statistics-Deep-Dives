# Visualizing Distributions and Density Curves

In this section, we explore how to **visualize distributions of data**, analyze those visualizations, and develop an understanding of **density curves**.

---

## Example: Glasses of Water per Day

Suppose we survey **16 students** and ask each student to:
- Record how many glasses of water they drink per day over the last 30 days
- Report the **average** number of glasses per day

Each data point represents one student:
- A value like **0.5** means the student averaged half a glass per day
- A value like **8.1** means the student averaged just over eight glasses per day

---

## Frequency Histograms

One way to visualize this data is with a **frequency histogram**.

- Data are grouped into **categories (bins)**, such as:
  - $[0, 1)$
  - $[3, 4)$
- The **height of each bar** represents the **number of data points** in that category

For example:
- If **2 out of 16 students** fall into the $[0, 1)$ category, the bar height is **2**
- If **4 out of 16 students** fall into the $[3, 4)$ category, the bar height is **4**

This gives a clear picture of how often values occur.

---

## Relative Frequency Histograms

Sometimes percentages are more informative than raw counts, especially with large datasets.

A **relative frequency histogram**:
- Uses **percentages** instead of counts
- Represents the **same data** as a frequency histogram

For example:
- 2 out of 16 students → $ \frac{2}{16} = 12.5\% $
- 4 out of 16 students → $ \frac{4}{16} = 25\% $

Both histograms show the same distribution, but the relative frequency version emphasizes **proportions** rather than totals.

---

## Increasing Granularity

As datasets grow larger, we can make bins **more granular**:
- From 1 glass wide → 0.5 glasses → 0.25 glasses
- More bins give a clearer picture of the distribution

As the bins become **infinitely thin**, the tops of the bars form a smooth curve.

This leads to the concept of a **density curve**.

---

## Density Curves

A **density curve** is a smooth representation of a distribution where:
- Data values can lie anywhere on a **continuous scale**
- Values are not grouped into coarse bins

### Key Properties

- The **total area under the curve equals 1 (or 100%)**
- The curve **never dips below zero**
- The **area under the curve over an interval** represents the **proportion of data** in that interval

---

## Interpreting a Density Curve

### Example 1: Interval Probabilities

To estimate the percentage of data between **2 and 4 glasses per day**:
- Look at the area under the curve from 2 to 4
- Eyeballing the curve suggests this area is about **40%**

### Example 2: Values Greater Than a Number

To estimate the percentage of data **greater than 3 glasses per day**:
- Look at the area to the right of 3
- This appears to be roughly **50%**

These estimates improve with precise tables or computational tools.

---

## A Common Misconception

**Question:**  
What percentage of students drink **exactly 3.000 glasses** of water per day?

**Answer:**  
**0%**

### Why?

- Probabilities on a density curve come from **area**, not height
- A single exact value has **no width**, so it has **zero area**
- Therefore, the probability at an exact point is zero

---

## Intervals vs. Exact Values

Instead of asking for an exact value, we consider an **interval**, such as:
$$
2.9 \leq x \leq 3.1
$$

This interval has width, so it has area.

### Approximation

- Height ≈ 0.2  
- Width = $ 3.1 - 2.9 = 0.2 $

$$
\text{Area} \approx 0.2 \times 0.2 = 0.04
$$

So approximately **4% of the data** falls in this interval.

---

## Key Takeaways

- Histograms visualize distributions using bins
- Relative frequency histograms emphasize proportions
- Density curves represent continuous distributions
- Probabilities come from **area under the curve**
- Exact values have **zero probability** in continuous distributions
- Intervals are essential for meaningful probability estimates


## Below is a Python code example for a Jupyter Notebook that demonstrates:

1. A frequency histogram

# Python Code: From Histogram to Density Curve

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
```

# Sample Data: Average Glasses of Water per Day

```python
# Example data: average glasses of water per day for 16 students
data = np.array([
    0.5, 0.8, 1.2, 1.9,
    2.3, 2.7, 3.1, 3.4,
    3.8, 4.2, 4.9, 5.4,
    6.1, 6.8, 7.4, 8.1
])
```

# 1. Frequency Histogram

```python
plt.figure()
plt.hist(data, bins=8)
plt.xlabel("Glasses of Water per Day")
plt.ylabel("Frequency")
plt.title("Frequency Histogram of Water Consumption")
plt.show()
```

## What this shows

* Counts of observations per bin
* Useful for small datasets


# 2. Relative Frequency Histogram

```python
plt.figure()
plt.hist(data, bins=8, density=True)
plt.xlabel("Glasses of Water per Day")
plt.ylabel("Relative Frequency")
plt.title("Relative Frequency Histogram of Water Consumption")
plt.show()
```

## What this shows

* Histogram scaled so total area = 1
* Heights represent proportions instead of counts


# 3. Density Curve Approximation

```python
# Estimate a smooth density curve
kde = gaussian_kde(data)

x = np.linspace(0, 9, 500)
y = kde(x)

plt.figure()
plt.plot(x, y)
plt.xlabel("Glasses of Water per Day")
plt.ylabel("Density")
plt.title("Density Curve of Water Consumption")
plt.show()
```

### What this shows

* A smooth curve approximating the distribution
* Total area under the curve equals 1
* Allows probability estimation over intervals


# 4. Estimating Probability Over an Interval

```python
# Approximate probability between 2 and 4 glasses per day
x_interval = np.linspace(2, 4, 300)
y_interval = kde(x_interval)

area = np.trapz(y_interval, x_interval)
print(f"Approximate proportion between 2 and 4 glasses: {area:.2f}")
```

### Key Visual Insights

* **Frequency histograms** show raw counts
* **Relative frequency histograms** how proportions
* **Density curves** allow probability estimation
* Probabilities come from **area**, not curve height
* Exact values (e.g., exactly 3.0) have probability 0
