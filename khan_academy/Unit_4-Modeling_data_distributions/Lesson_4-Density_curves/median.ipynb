# Describing Density Curves: Mean, Median, and Skewness

Density curves provide a **summary of a distribution** of data and help us understand how values are spread across a continuous range. In this discussion, we focus on what we can **infer from density curves**, particularly the **median**, **mean**, and **skewness** of the distributions they represent.

---

## The Median of a Density Curve

Recall that for a set of ordered values, the **median** is the middle value (or the average of the two middle values).  
For a **density curve**, the median is defined as:

> The value where **half of the area** under the curve lies to the left and **half lies to the right**.

### Symmetric Distributions

- In a **symmetric distribution**, the median lies on the **line of symmetry**.
- This applies whether the distribution is:
  - **Unimodal** (one peak), or
  - **Bimodal** (two peaks), as long as it is symmetric.

In both cases, the equal-area property holds, making the point of symmetry the median.

---

## Medians in Non-Symmetric Distributions

For **non-symmetric distributions**, the same principle applies:
- The median is still the value that divides the area into two equal halves.
- However, it may not lie at the highest point of the curve.

By visually estimating where the left and right areas appear equal, we can approximate the median even when the curve is skewed.

---

## The Mean of a Density Curve

The **mean** of a distribution can be thought of as a **weighted average** of all possible values, where each value is weighted by its frequency.

A helpful physical interpretation:
- The mean is the **balance point** of the density curve.
- If the curve were made of a uniform material, the mean is where a **fulcrum** would balance it.

### Symmetric Distributions

- For symmetric distributions:
  - **Mean = Median**
  - The balance point lies at the center of symmetry.

---

## Mean vs. Median in Skewed Distributions

In **skewed distributions**, the mean and median are no longer equal.

### Right-Skewed Distributions

- Characterized by a **long tail to the right**
- The tail pulls the mean **to the right**
- Relationship:
  $$
  \text{Mean} > \text{Median}
  $$

### Left-Skewed Distributions

- Characterized by a **long tail to the left**
- The tail pulls the mean **to the left**
- Relationship:
  $$
  \text{Mean} < \text{Median}
  $$

---

## Skewness

**Skewness** describes the asymmetry of a distribution.

- **Right-skewed (positive skew):**
  - Long right tail
  - Mean to the right of the median

- **Left-skewed (negative skew):**
  - Long left tail
  - Mean to the left of the median

While the technical definition of skewness can be complex, visually it is often identified by:
- The direction of the **long tail**, or
- The direction in which the **mean is pulled away from the median**

---

## Key Takeaways

- The **median** divides the area under a density curve into two equal halves
- In **symmetric distributions**, mean and median are equal
- In **skewed distributions**, the mean is pulled toward the tail
- Long right tail → **right-skewed**
- Long left tail → **left-skewed**

Density curves provide a powerful visual tool for understanding the **center, shape, and skewness** of continuous data.

# Python Code: Mean, Median, and Skewness with Density Curves

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, skewnorm
```

# 1. Symmetric Distribution (Mean = Median)

```python
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)

mean = 0
median = 0

plt.figure()
plt.plot(x, y)
plt.axvline(mean)
plt.axvline(median)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Symmetric Distribution (Mean = Median)")
plt.show()
```

# 2. Right-Skewed Distribution (Mean > Median)

```python
x = np.linspace(-4, 6, 1000)
y = skewnorm.pdf(x, a=6)

mean = skewnorm.mean(a=6)
median = skewnorm.median(a=6)

plt.figure()
plt.plot(x, y)
plt.axvline(mean)
plt.axvline(median)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Right-Skewed Distribution (Mean > Median)")
plt.show()
```

# 3. Left-Skewed Distribution (Mean < Median)

```python
x = np.linspace(-6, 4, 1000)
y = skewnorm.pdf(x, a=-6)

mean = skewnorm.mean(a=-6)
median = skewnorm.median(a=-6)

plt.figure()
plt.plot(x, y)
plt.axvline(mean)
plt.axvline(median)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Left-Skewed Distribution (Mean < Median)")
plt.show()
```

# What These Plots Demonstrate

* **Symmetric distribution**
    - Mean and median occur at the same location
* **Right-skewed distribution**
    - Long right tail pulls the mean to the right
* **Left-skewed distribution**
    - Long left tail pulls the mean to the left

These visualizations reinforce the key idea that:       
The median divides area equally, while the mean acts as the balance point and is pulled toward the tail.
