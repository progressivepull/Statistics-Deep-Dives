# Analyzing a Triangular Density Curve

Consider the **density curve** shown below. Although it looks different from the smooth bell-shaped curves often seen in statistics, this **triangular shape** is still a valid density curve.

We analyze several statements about this curve and determine which are true.

---

## Statement 1  
**The mean of the density curve is less than the median.**

### Analysis
- The **median** is the value that divides the total area under the curve into two equal halves.
- By visual inspection, the median appears to lie slightly to the right of the center of the base.
- The curve extends farther to the **left** than to the right, creating a **longer left tail**.

This indicates a **left-skewed distribution**.

### Conclusion
- In left-skewed distributions, the **mean is pulled toward the tail**.
- Therefore, the **mean lies to the left of the median**.

✅ **This statement is true.**

---

## Statement 2  
**The median of the density curve is 3.**

### Analysis
- If the median were at 3, the area to the left of 3 would have to equal the area to the right.
- Visual inspection shows that the area to the left of 3 is **much smaller** than the area to the right.

### Conclusion
- The median cannot be 3.

❌ **This statement is false.**

---

## Statement 3  
**The area underneath the density curve is 1.**

### Analysis
- By definition, **all density curves have total area equal to 1**.
- This represents 100% of the data.

### Conclusion
✅ **This statement is true.**

---

## Determining the Height of the Density Curve

We can use the fact that the area under the density curve is **1** to determine the height of the triangle.

### Given
- The curve is triangular
- Base extends from **1 to 6**
- Base length = $ 6 - 1 = 5 $
- Area of triangle:
  $$
  \text{Area} = \frac{1}{2} \times \text{base} \times \text{height}
  $$

### Solve for Height

$$
1 = \frac{1}{2} \times 5 \times h
$$

$$
1 = \frac{5}{2}h
$$

$$
h = \frac{2}{5}
$$

---

## Final Results

- **Mean < Median** → True  
- **Median = 3** → False  
- **Total area = 1** → True  
- **Height of the density curve:**  
  $$
  h = \frac{2}{5}
  $$

---

## Key Takeaways

- Skewness affects the relative positions of the mean and median
- The median is determined by **equal area**, not symmetry
- The total area under any density curve is always **1**
- Simple geometric shapes allow direct calculation of density values


# Below is a Python code example for a Jupyter Notebook that:

* Defines **a triangular density curve**
* Verifies the **total area equals 1**
* Plots the curve
* Illustrates **mean and median locations** conceptually


# Python Code: Triangular Density Curve

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Define the Triangular Density Curve

We are given:
* Base from x = 1 to x = 6

* Height  $h = \frac{2}{5}$


	​

```python
# Define x range
x = np.linspace(1, 6, 500)

# Define triangular density function
# Peak at x = 4 (right-skewed triangle)
h = 2 / 5

y = np.where(
    x <= 4,
    h * (x - 1) / (4 - 1),        # rising side
    h * (6 - x) / (6 - 4)        # falling side
)
```

# Verify Total Area = 1

```python
area = np.trapz(y, x)
print(f"Total area under the curve: {area:.3f}")
```

# Plot the Density Curve

```python
plt.figure()
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Triangular Density Curve")
plt.show()
```

# Interpretation Notes

* The **area under the curve equals 1**, confirming it is a valid density curve
* The **longer tail to the left** makes the distribution **left-skewed**
* The** mean lies to the left of the median**
* The **median** is the value that splits the area into two equal halves
* The **mean** acts as the balance point and is pulled toward the tail
