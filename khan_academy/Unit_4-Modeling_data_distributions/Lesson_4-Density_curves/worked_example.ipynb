# Summary: Area Under a Density Curve

## Example 1: Finding the Area Where $ x > 2 $

Consider a density curve that is more angular than typical smooth curves. Although it looks different, it is easier to work with when calculating areas.

We are asked to find the **percent of the area under the density curve where $ x > 2 $**.

### Step 1: Identify the Relevant Area
- The density curve exists only from $ x = 1 $ to $ x = 3 $.
- The region where $ x > 2 $ corresponds to the area from $ x = 2 $ to $ x = 3 $.

### Step 2: Verify the Total Area Under the Density Curve
All density curves must have a **total area of 1**.

The full curve forms a trapezoid from $ x = 1 $ to $ x = 3 $ with heights:
- Left height: 0.25  
- Right height: 0.75  
- Base length: 2  

Using the trapezoid area formula:

$$
\text{Area} = \frac{(0.25 + 0.75)}{2} \times 2 = 1
$$

This confirms the curve is a valid density curve.

### Step 3: Calculate the Area Where $ x > 2 $
The region from $ x = 2 $ to $ x = 3 $ also forms a trapezoid with:
- Left height: 0.5  
- Right height: 0.75  
- Base length: 1  

$$
\text{Area} = \frac{(0.5 + 0.75)}{2} \times 1 = \frac{1.25}{2} = 0.625
$$

### Step 4: Convert to a Percentage
$$
0.625 = 62.5\%
$$

**Answer:**  
The percent of the area under the density curve where $ x > 2 $ is **62.5%**.

## Example 2: Finding the Area Where $ x > 3 $

Consider another angular density curve.

We are asked to find the **percent of the area under the density curve where $ x > 3 $**.

### Key Observation
- The entire density curve lies to the right of $ x = 3 $.

Since the **entire area under any density curve is 1**, the area where $ x > 3 $ includes the whole curve.

**Answer:**  
The percent of the area under the density curve where $ x > 3 $ is **100%**.


```python
# Density Curve Area Calculations
# Example 1: Percent of area where x > 2

# Heights of the trapezoid segment from x = 2 to x = 3
left_height = 0.5
right_height = 0.75
base_length = 1  # from x=2 to x=3

# Trapezoid area formula
area_x_greater_than_2 = (left_height + right_height) / 2 * base_length

# Convert to percentage
percent_x_greater_than_2 = area_x_greater_than_2 * 100

area_x_greater_than_2, percent_x_greater_than_2
```

```python
# Example 2: Percent of area where x > 3

# By definition, the entire density curve lies where x > 3
area_x_greater_than_3 = 1.0
percent_x_greater_than_3 = area_x_greater_than_3 * 100

area_x_greater_than_3, percent_x_greater_than_3
```
