# Normal Distribution: Student Heights

## Problem Summary

A group of middle school students’ heights is **normally distributed** with:

- **Mean (μ):** 150 centimeters  
- **Standard deviation (σ):** 20 centimeters  

Darnell is a middle school student with a height of **161.4 centimeters**.

**Question:**  
What **proportion of students** have heights **lower than Darnell’s height**?

---

## Understanding the Approach

Because the heights follow a **normal distribution**, we can determine the proportion of students shorter than Darnell by:

1. Calculating how many **standard deviations above the mean** Darnell’s height is  
2. Converting this value into a **z-score**  
3. Using a **z-table** to find the proportion of values below that z-score  

---

## Step 1: Compute the Z-Score

The z-score formula is:

$$
z = \frac{x - \mu}{\sigma}
$$

Substitute the given values:

$$
z = \frac{161.4 - 150}{20} = \frac{11.4}{20} = 0.57
$$

So, Darnell’s height is **0.57 standard deviations above the mean**.

---

## Step 2: Use the Z-Table

Looking up a z-score of **0.57** in the standard normal (z) table gives:

$$
P(Z < 0.57) = 0.7157
$$

This value represents the **proportion of students with heights less than Darnell’s height**.

---

## Final Answer

$$
\boxed{0.7157}
$$

**Interpretation:**  
Approximately **71.57%** of middle school students are **shorter than Darnell**.


```python
# Example: Finding the proportion of students shorter than Darnell
# using a normal distribution in Python

from scipy.stats import norm

# Given values
mean_height = 150        # mean (cm)
std_dev = 20             # standard deviation (cm)
darnell_height = 161.4   # Darnell's height (cm)

# Step 1: Calculate the z-score
z_score = (darnell_height - mean_height) / std_dev

# Step 2: Find the proportion below Darnell's height
proportion_below = norm.cdf(z_score)

# Display results
z_score, proportion_below
```

This means approximately **71.57%** of students are shorter than Darnell.
