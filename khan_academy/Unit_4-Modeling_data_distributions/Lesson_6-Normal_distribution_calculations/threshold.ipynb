# Normal Distribution: Drive-Through Wait Times

## Problem Summary

The average wait times in drive-through restaurant lines in a town are **approximately normally distributed** with:

- **Mean (μ):** 185 seconds  
- **Standard deviation (σ):** 11 seconds  

Amelia prefers to use the drive-through **only at restaurants whose average wait times are in the bottom 10%** for the town.

**Question:**  
What is the **maximum average wait time** for restaurants where Amelia likes to use the drive-through?  
Round to the **nearest whole second**.

---

## Understanding the Problem

- The **bottom 10%** corresponds to the **10th percentile** of a normal distribution.
- This cutoff value represents the **largest wait time that is still shorter than 90% of restaurants**.
- Because this cutoff is **below the mean**, the corresponding **z-score will be negative**.

---

## Step 1: Find the Z-Score for the 10th Percentile

Using a standard normal (z) table:

- The z-score closest to a cumulative probability of **0.10** is:

$$
z = -1.29
$$

This means the cutoff is **1.29 standard deviations below the mean**.

---

## Step 2: Convert the Z-Score to a Wait Time

Use the formula:

$$
x = \mu + z\sigma
$$

Substitute the given values:

$$
x = 185 + (-1.29)(11)
$$

$$
x = 185 - 14.19 = 170.81
$$

---

## Step 3: Round to the Nearest Whole Second

$$
170.81 \approx 171
$$

---

## Final Answer

$$
\boxed{171 \text{ seconds}}
$$

**Interpretation:**  
Restaurants with an average drive-through wait time of **about 171 seconds or less** fall within the **bottom 10%**, making them acceptable to Amelia.


```python
# Example: Finding the maximum average wait time in the bottom 10%
# using a normal distribution in Python

from scipy.stats import norm

# Given values
mean_wait = 185       # mean wait time (seconds)
std_dev = 11          # standard deviation (seconds)
bottom_percent = 0.10 # bottom 10%

# Step 1: Find the z-score for the 10th percentile
z_score = norm.ppf(bottom_percent)

# Step 2: Convert the z-score to the wait time
max_wait_time = mean_wait + z_score * std_dev

# Display results
z_score, max_wait_time, round(max_wait_time)
```

This shows that the **maximum average wait time** for Amelia’s preferred restaurants is approximately **171 seconds**.
