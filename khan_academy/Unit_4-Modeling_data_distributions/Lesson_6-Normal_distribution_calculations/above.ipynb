# Normal Distribution: Philosophy Exam Scores

## Problem Summary

A set of philosophy exam scores is **normally distributed** with the following parameters:

- **Mean (μ):** 40 points  
- **Standard deviation (σ):** 3 points  

Ludwig scored **47.5 points** on the exam.

**Question:**  
What **proportion of exam scores are higher than Ludwig’s score**?  
Give the answer **to four decimal places**.

---

## Conceptual Overview

Because the exam scores follow a **normal distribution**, we can determine the desired proportion by:

1. Calculating how many **standard deviations above the mean** Ludwig’s score lies  
2. Converting this value into a **z-score**  
3. Using a **z-table** to find the proportion of scores **below** that z-score  
4. Subtracting that value from 1 to get the proportion **above** Ludwig’s score  

---

## Step 1: Calculate the Z-Score

The z-score formula is:

$$
z = \frac{x - \mu}{\sigma}
$$

Substitute the given values:

$$
z = \frac{47.5 - 40}{3} = \frac{7.5}{3} = 2.5
$$

Ludwig’s score is **2.5 standard deviations above the mean**.

---

## Step 2: Use the Z-Table

From the standard normal (z) table:

$$
P(Z < 2.50) = 0.9938
$$

This value represents the **proportion of exam scores lower than Ludwig’s score**.

---

## Step 3: Find the Proportion Above Ludwig’s Score

Since the total area under the normal curve is 1:

$$
P(Z > 2.50) = 1 - 0.9938 = 0.0062
$$

---

## Final Answer

$$
\boxed{0.0062}
$$

**Interpretation:**  
Approximately **0.62%** of exam scores are **higher than Ludwig’s score**, which is reasonable given that his score is **2.5 standard deviations above the mean**.


```python
# Example: Finding the proportion of exam scores higher than Ludwig's
# using a normal distribution in Python

from scipy.stats import norm

# Given values
mean_score = 40        # mean score
std_dev = 3            # standard deviation
ludwig_score = 47.5    # Ludwig's score

# Step 1: Calculate the z-score
z_score = (ludwig_score - mean_score) / std_dev

# Step 2: Find the proportion above Ludwig's score
proportion_above = 1 - norm.cdf(z_score)

# Display results
z_score, proportion_above
```

This shows that approximately **0.62%** of exam scores are higher than Ludwig’s.
