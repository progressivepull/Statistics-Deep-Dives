# Normal Distribution: Laptop Prices

## Problem Summary

A set of laptop prices is **normally distributed** with the following parameters:

- **Mean (μ):** $750  
- **Standard deviation (σ):** $60  

**Question:**  
What **proportion of laptop prices** are between **$624 and $768**?

---

## Visualizing the Distribution

Because prices follow a **normal distribution**, the curve is symmetric around the mean:

- The center of the distribution is at **$750**
- One standard deviation above the mean is **$810**
- One standard deviation below the mean is **$690**

The interval **$624 to $768** spans values **below and above the mean**, so we must find the **area under the curve between these two prices**.

---

## Step 1: Find the Z-Score for $768

The z-score formula is:

$$
z = \frac{x - \mu}{\sigma}
$$

For the upper bound:

$$
z = \frac{768 - 750}{60} = \frac{18}{60} = 0.30
$$

Using a standard normal (z) table:

$$
P(Z < 0.30) = 0.6179
$$

This is the proportion of laptop prices **less than $768**.

---

## Step 2: Find the Z-Score for $624

For the lower bound:

$$
z = \frac{624 - 750}{60} = \frac{-126}{60} = -2.10
$$

From the z-table:

$$
P(Z < -2.10) = 0.0179
$$

This is the proportion of laptop prices **less than $624**.

---

## Step 3: Find the Proportion Between $624 and $768

To find the proportion between the two prices, subtract:

$$
0.6179 - 0.0179 = 0.6000
$$

---

## Final Answer

$$
\boxed{0.6000}
$$

**Interpretation:**  
Exactly **60% of laptop prices** fall between **$624 and $768**.


```python
# Example: Finding the proportion of laptop prices between $624 and $768
# using a normal distribution in Python

from scipy.stats import norm

# Given values
mean_price = 750     # mean price ($)
std_dev = 60         # standard deviation ($)
lower_price = 624
upper_price = 768

# Step 1: Calculate z-scores (optional, for inspection)
z_lower = (lower_price - mean_price) / std_dev
z_upper = (upper_price - mean_price) / std_dev

# Step 2: Find the proportion between the two prices
proportion_between = norm.cdf(upper_price, mean_price, std_dev) - \
                     norm.cdf(lower_price, mean_price, std_dev)

# Display results
z_lower, z_upper, proportion_between
```

This shows that 60.00% of laptop prices are between $624 and $768.
