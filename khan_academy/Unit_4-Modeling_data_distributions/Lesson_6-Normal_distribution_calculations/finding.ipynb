# Normal Distribution: Resting Pulse Rates

## Problem Summary

The resting pulse rates of students at **Santa Maria High School** are approximately **normally distributed** with:

- **Mean (μ):** 80 beats per minute (bpm)  
- **Standard deviation (σ):** 9 beats per minute  

The school nurse plans to provide **additional screening** to students whose resting pulse rates fall in the **top 30%** of all students tested.

**Question:**  
What is the **minimum resting pulse rate** for students who will receive additional screening?  
Round your answer to the **nearest whole number**.

---

## Understanding the Problem

- Being in the **top 30%** means that **70% of students have pulse rates below** the cutoff value.
- We must find the pulse rate corresponding to the **70th percentile** of a normal distribution.
- This is a **reverse z-score problem**: we are given an area (0.70) and must find the corresponding value.

---

## Step 1: Find the Corresponding Z-Score

Using a standard normal (z) table:

- The closest value to **0.7000** is **0.7019**
- This corresponds to a **z-score of 0.53**

So, the cutoff pulse rate is **0.53 standard deviations above the mean**.

---

## Step 2: Convert the Z-Score to a Pulse Rate

Use the formula:

$$
x = \mu + z\sigma
$$

Substitute the known values:

$$
x = 80 + (0.53)(9) = 80 + 4.77 = 84.77
$$

---

## Step 3: Round to the Nearest Whole Number

$$
84.77 \approx 85
$$

---

## Final Answer

$$
\boxed{85 \text{ beats per minute}}
$$

**Interpretation:**  
Students with a resting pulse rate of **85 bpm or higher** are in the **top 30%** and will receive additional screening.


```python
# Example: Finding the minimum resting pulse rate for the top 30%
# using a normal distribution in Python

from scipy.stats import norm

# Given values
mean_pulse = 80        # mean pulse rate (bpm)
std_dev = 9            # standard deviation (bpm)
top_percentage = 0.30  # top 30%

# Step 1: Find the z-score for the top 30%
# This corresponds to the 70th percentile
z_score = norm.ppf(1 - top_percentage)

# Step 2: Convert the z-score to the pulse rate
cutoff_pulse = mean_pulse + z_score * std_dev

# Display results
z_score, cutoff_pulse, round(cutoff_pulse)
```

This shows that the minimum resting pulse rate for additional screening is approximately 85 beats per minute.
