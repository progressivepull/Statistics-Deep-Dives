# Cumulative Relative Frequency Analysis of Sugar Content in Starbucks Drinks

Nutritionists measured the sugar content (in grams) of **32 Starbucks drinks** and displayed the results using a **cumulative relative frequency graph**.

## Understanding the Graph

- The **horizontal axis** represents the amount of sugar (in grams).
- The **vertical axis** represents the **cumulative relative frequency** (the proportion of drinks with *that amount of sugar or less*).
- The graph starts at 0%, meaning **0% of drinks have 0 grams of sugar**.
- As sugar content increases, the cumulative relative frequency also increases.
- At 50 grams of sugar, the cumulative relative frequency reaches **100%**, indicating that **all drinks have 50 grams of sugar or less**.

For example:
- At **5 grams of sugar**, the cumulative relative frequency is **0.10**, meaning **10% of drinks** have **5 grams of sugar or less**.
- At **15 grams of sugar**, the cumulative relative frequency is **0.20**, meaning **20% of drinks** have **15 grams of sugar or less**.

---

## Percentile Estimation

### Question:  
**An iced coffee has 15 grams of sugar. What is its percentile (nearest whole percent)?**

- From the graph, a sugar content of **15 grams** corresponds to a cumulative relative frequency of **0.20**.
- This means **20% of drinks** have **15 grams of sugar or less**.

**Answer:**  
The iced coffee is in the **20th percentile**.

---

## Median Estimation

### Question:  
**Estimate the median sugar content of the drinks.**

- The median corresponds to the **50th percentile**.
- A cumulative relative frequency of **0.50** occurs at approximately **25 grams of sugar**.
- This indicates that **half of the drinks** have **25 grams of sugar or less**.

**Answer:**  
The median sugar content is approximately **25 grams**.

---

## Interquartile Range (IQR)

### Question:  
**Estimate the interquartile range (IQR) of the distribution.**

The IQR is calculated as:

$$
\text{IQR} = Q_3 - Q_1
$$

- **First Quartile (25th percentile):**
  - Occurs at a cumulative relative frequency of **0.25**
  - Estimated at approximately **18 grams of sugar**

- **Third Quartile (75th percentile):**
  - Occurs at a cumulative relative frequency of **0.75**
  - Estimated at approximately **39 grams of sugar**

$$
\text{IQR} = 39 - 18 = 21 \text{ grams}
$$

**Answer:**  
The estimated interquartile range is approximately **21 grams**, making **20 grams** the closest reasonable choice.

---

## Summary of Key Results

- **20th percentile:** 15 grams of sugar  
- **Median (50th percentile):** 25 grams of sugar  
- **Interquartile range:** Approximately 20–21 grams  

This cumulative relative frequency graph provides a clear way to interpret percentiles, medians, and variability in the sugar content of Starbucks drinks.


```python
import matplotlib.pyplot as plt
import numpy as np

# Approximate sugar content values (grams)
sugar_grams = np.array([0, 5, 10, 15, 20, 25, 30, 35, 40, 45, 50])

# Corresponding cumulative relative frequencies
cum_rel_freq = np.array([0.0, 0.1, 0.15, 0.2, 0.35, 0.5, 0.6, 0.7, 0.75, 0.9, 1.0])
```

```python
# Create the plot
plt.figure()
plt.plot(sugar_grams, cum_rel_freq, marker='o')
plt.xlabel("Sugar Content (grams)")
plt.ylabel("Cumulative Relative Frequency")
plt.title("Cumulative Relative Frequency of Sugar Content in Starbucks Drinks")
plt.ylim(0, 1.05)
plt.grid(True)

plt.show()
```

# What this code does

* Uses **matplotlib** to create a cumulative relative frequency graph
* The **x-axis** shows sugar content in grams
* The **y-axis** shows cumulative relative frequency (0 to 1)
* Each point represents the proportion of drinks **with that amount of sugar or less**
