# Estimating Percentile Rank from a Dot Plot

The dot plot represents the number of hours of daily driving time for **14 school bus drivers**, where each dot corresponds to one driver.

From the plot:
- 1 driver drives **1 hour** per day  
- 2 drivers drive **2 hours** per day  
- 1 driver drives **3 hours** per day  
- 5 drivers drive **7 hours** per day  
- The remaining data points fill in the other values shown on the plot  

The question asks for the **closest estimate of the percentile rank** for a driver who drives **6 hours per day**.

---

## Understanding Percentile Rank

Percentile rank can be interpreted in two common ways:

1. **Percentage of data values strictly below** the value in question  
2. **Percentage of data values at or below** the value in question  

Both methods are valid, depending on the convention used.

---

## Calculations

### Method 1: Below 6 hours
- Number of drivers below 6 hours: **7**
- Total number of drivers: **14**

$$
\frac{7}{14} = 0.50 = 50\%
$$

Using this method, **6 hours is at the 50th percentile**.

---

### Method 2: At or Below 6 hours
- Number of drivers at or below 6 hours: **8**
- Total number of drivers: **14**

$$
\frac{8}{14} = \frac{4}{7} \approx 0.5556 = 55.6\%
$$

Using this method, **6 hours is approximately at the 56th percentile**.

---

## Final Answer

Both interpretations are reasonable:
- **50th percentile** (below-only method)
- **Approximately 55th–56th percentile** (at-or-below method)

Given the answer choices, the **closest estimate** is:

**✅ 55th percentile**

This indicates the question uses the **“at or below”** definition of percentile rank.


```python
import numpy as np
import matplotlib.pyplot as plt

# Example data
driving_hours = [2, 3, 4, 5, 6, 7, 8, 9]
value = 6

# Percentile calculation using NumPy
percentile_numpy = np.mean(np.array(driving_hours) <= value) * 100
print("Percentile:", percentile_numpy)

# --- Dot Plot ---
plt.figure(figsize=(8, 2))

# y-values all zero to create a horizontal dot plot
y = np.zeros_like(driving_hours)

plt.scatter(driving_hours, y, color="blue", s=80)

# Highlight the chosen value
plt.scatter(value, 0, color="red", s=120, label=f"value = {value}")

plt.yticks([])  # Hide y-axis ticks
plt.xlabel("Driving Hours")
plt.title("Dot Plot of Driving Hours")
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.show()
```

# What This Code Does 🎯

The code performs the following tasks:

- Computes the **percentile** of a specific `value` relative to the data in `driving_hours` using the empirical cumulative distribution approach:

  $$
  \text{percentile} = \text{mean}(x \leq \text{value}) \times 100
  $$

  This calculates the percentage of data points in `driving_hours` that are less than or equal to the given `value`.

- Creates a **horizontal dot plot** (also known as a dotplot or strip plot), where:
  - Each individual data point from `driving_hours` is represented as a dot.
  - The specific `value` of interest is highlighted in **red** to distinguish it from the other data points.

This visualization helps to intuitively understand the relative position of the highlighted value within the overall distribution of driving hours.
