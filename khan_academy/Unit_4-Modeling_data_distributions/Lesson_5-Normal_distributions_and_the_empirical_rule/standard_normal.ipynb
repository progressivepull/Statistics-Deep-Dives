# Normal Distribution Practice — Problem 4 (CK-12 AP Statistics)

This problem comes from **Problem #4** in the *Normal Distribution* chapter of the **CK-12 FlexBook for AP Statistics** (open-source and freely available).

The goal is to **order several quantities related to a standard normal distribution from smallest to largest**, using the **Empirical Rule (68–95–99.7)**.

---

## Standard Normal Distribution Review

A **standard normal distribution** has:

- Mean:  
  $
  \mu = 0
  $
- Standard deviation:  
  $
  \sigma = 1
  $

Key reference points:
- $ z = \pm 1 $ → 1 standard deviation from the mean  
- $ z = \pm 2 $ → 2 standard deviations  
- $ z = \pm 3 $ → 3 standard deviations  

The distribution is **perfectly symmetric**.

---

## Using the Empirical Rule

- **68%** of data lie within $ \pm 1\sigma $
- **95%** of data lie within $ \pm 2\sigma $
- **99.7%** of data lie within $ \pm 3\sigma $

From this:
- Area in both tails beyond $ \pm 1\sigma $: 32% → **16% per tail**
- Area in both tails beyond $ \pm 2\sigma $: 5% → **2.5% per tail**

---

## Evaluating Each Quantity

### (a) Percentage of data **below 1**
- Includes:
  - 50% to the left of the mean
  - plus half of the 68% within $ \pm 1\sigma $
- Equivalent to:
  $
  68\% + 16\% = 84\%
  $
- Decimal form: **0.84**

---

### (b) Percentage of data **below −1**
- This is the **left tail beyond −1σ**
- Area = **16%**
- Decimal form: **0.16**

---

### (c) The **mean**
- By definition for a standard normal distribution:
  $
  0
  $

---

### (d) The **standard deviation**
- By definition:
  $
  1
  $

---

### (e) Percentage of data **above 2**
- Right tail beyond $ +2\sigma $
- Area = **2.5%**
- Decimal form: **0.025**

---

## Ordering from Smallest to Largest

Interpreting percentages as **decimals**, we have:

| Label | Value |
|------|-------|
| (c) Mean | 0 |
| (e) Above 2 | 0.025 |
| (b) Below −1 | 0.16 |
| (a) Below 1 | 0.84 |
| (d) Standard deviation | 1 |

### ✅ Final Order (smallest → largest):
$
\boxed{c,\ e,\ b,\ a,\ d}
$

---

## Important Note on Ambiguity

There is **potential ambiguity** in the problem:

- If percentages are interpreted as **whole numbers** (e.g., 84 instead of 0.84), the ordering could change.
- On an exam, this should be **clarified with the instructor**.

---

## Key Takeaway

This problem reinforces:
- Understanding the **standard normal distribution**
- Applying the **Empirical Rule without calculators**
- Recognizing symmetry and relative magnitudes when ordering values

A strong conceptual grasp makes these problems much easier—even without precise calculations.


```python
"""
Standard Normal Distribution — Empirical Rule Ordering Example
CK-12 AP Statistics, Normal Distribution Problem #4

This example:
1. Defines a standard normal distribution
2. Encodes Empirical Rule probabilities
3. Orders quantities from smallest to largest
"""

# Standard normal parameters
mean = 0
std_dev = 1

# Empirical Rule probabilities (as decimals)
percent_below_1 = 0.84      # P(Z < 1)
percent_below_neg1 = 0.16   # P(Z < -1)
percent_above_2 = 0.025     # P(Z > 2)

# Store quantities with labels
quantities = {
    "c) Mean": mean,
    "e) % above 2": percent_above_2,
    "b) % below -1": percent_below_neg1,
    "a) % below 1": percent_below_1,
    "d) Std deviation": std_dev,
}

# Sort from smallest to largest
sorted_quantities = sorted(quantities.items(), key=lambda item: item[1])

# Display results
print("Ordered from smallest to largest:\n")
for label, value in sorted_quantities:
    print(f"{label:<20} = {value}")
```
