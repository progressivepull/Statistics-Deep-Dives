# Summary: Identifying Data Likely to Follow a Normal Distribution

The normal distribution is one of the most important building blocks in statistics, appearing frequently in both theory and real-world applications. Because of this, it is valuable to practice recognizing when data are likely—or unlikely—to follow a normal distribution.

The following examples are adapted from the **CK-12 Open Source AP Statistics FlexBook**, specifically from the chapter on normal distributions. The question asks:

> **Which of the following data sets is most likely to be normally distributed?**  
> For the other choices, explain why they would not follow a normal distribution.

This type of question is somewhat subjective and relies on statistical reasoning rather than exact calculation.

---

## Choice A: Hand Span of High School Seniors

**Description:**  
Hand span is measured from the tip of the thumb to the tip of the extended fifth finger (pinky) for a random sample of high school seniors.

**Reasoning:**  
- Hand span is influenced by many small random factors, such as genetics, nutrition, and growth patterns.
- When a variable is the result of many independent influences, it often approximates a normal distribution.
- While hand span cannot be negative and has a physical upper limit, within realistic bounds it should cluster symmetrically around a mean.

**Caveats:**  
- The distribution cannot be perfectly normal due to physical constraints.
- It may be **bimodal** if males and females have noticeably different average hand spans.

**Conclusion:**  
Despite limitations, hand span is a strong candidate for being approximately normally distributed.

---

## Choice B: Annual Salaries of All Employees in a Large Shipping Company

**Reasoning:**  
- There is a minimum wage or minimum salary, so values cannot extend indefinitely to the left.
- Many employees earn lower wages, while a small number of executives earn very high salaries.
- This produces a **long right tail**, making the distribution **right-skewed**.
- There may also be multiple peaks (e.g., hourly workers vs. management), making it **bimodal** or **multimodal**.

**Conclusion:**  
This data set is not normally distributed due to skewness and multiple modes.

---

## Choice C: Annual Salaries of 50 CEOs (25 Women and 25 Men)

**Reasoning:**  
- CEO salaries often have a lower bound but can extend very far to the right.
- A small number of “mega-CEOs” may earn millions, creating a strong right skew.
- If there is a gender pay gap, the distribution may be **bimodal**, with separate peaks for men and women.
- In some cases, the distribution could even be **trimodal** due to extreme outliers.

**Conclusion:**  
CEO salaries are unlikely to be normally distributed because of skewness and possible multiple modes.

---

## Choice D: Dates of 100 Pennies from a Convenience Store Cash Drawer

**Reasoning:**  
- Most pennies in circulation are relatively recent.
- Older pennies are lost, removed, or worn out over time.
- This leads to many observations clustered at recent years, with a tail extending into the past.
- The result is a **left-skewed (negatively skewed)** distribution.

**Statistical Insight:**  
- In left-skewed distributions, the **mean lies to the left of the median**.
- In right-skewed distributions, the **mean lies to the right of the median**.

**Conclusion:**  
The distribution of penny dates is not normal due to strong skewness.

---

## Final Answer

Among all the choices, **Choice A (hand span of high school seniors)** is the most likely to be approximately normally distributed.

While none of the data sets will follow a perfect normal distribution, hand span best fits the assumptions of symmetry and natural variability that characterize normal behavior in real-world data.


```python
import numpy as np
import matplotlib.pyplot as plt
```

# Example 1: Approximately Normal Distribution (Hand Span)

```python
# Simulate hand span data (approximately normal)
np.random.seed(42)

hand_span = np.random.normal(loc=8.5, scale=0.7, size=1000)

plt.hist(hand_span, bins=30, density=True)
plt.title("Simulated Hand Span Distribution (Approximately Normal)")
plt.xlabel("Hand Span (inches)")
plt.ylabel("Density")
plt.show()
```

### Interpretation:
This resembles a bell-shaped curve, making it a good approximation of a normal distribution.

# Example 2: Right-Skewed Distribution (Employee Salaries)

```python
# Simulate employee salaries using a log-normal distribution
salaries = np.random.lognormal(mean=10.5, sigma=0.8, size=1000)

plt.hist(salaries, bins=50, density=True)
plt.title("Simulated Employee Salaries (Right-Skewed)")
plt.xlabel("Salary")
plt.ylabel("Density")
plt.show()
```

### Interpretation:
Most salaries cluster at lower values, with a long tail extending to the right—typical of income data.

# Example 3: Bimodal Distribution (CEO Salaries with Gender Gap)

```python
# Simulate bimodal CEO salaries
female_ceos = np.random.normal(loc=700_000, scale=100_000, size=500)
male_ceos = np.random.normal(loc=900_000, scale=120_000, size=500)

ceo_salaries = np.concatenate([female_ceos, male_ceos])

plt.hist(ceo_salaries, bins=40, density=True)
plt.title("Simulated CEO Salaries (Bimodal Distribution)")
plt.xlabel("Salary")
plt.ylabel("Density")
plt.show()
```

### Interpretation:
Two visible peaks indicate a bimodal distribution, which violates normality.

# Example 4: Left-Skewed Distribution (Dates of Pennies)

```python
# Simulate penny dates (more recent years are more common)
current_year = 2025
penny_years = current_year - np.random.exponential(scale=15, size=1000)

plt.hist(penny_years, bins=30, density=True)
plt.title("Simulated Penny Dates (Left-Skewed)")
plt.xlabel("Year")
plt.ylabel("Density")
plt.show()
```

### Interpretation:
Most pennies are recent, with a long tail toward older years—characteristic of a left-skewed distribution.

# Key Takeaways (Demonstrated in Code)

* **Normal distributions** arise from many small random effects.
* **Right-skewed distributions** are common with income and salary data.
* **Bimodal distributions** indicate distinct subpopulations.
* **Left-skewed distributions** often occur with aging or replacement processes.
