# Normal Distribution Practice — Problems 5 & 6 (CK-12 AP Statistics)

This summary covers **Problems #5 and #6** from the *Normal Distribution* chapter of the **CK-12 FlexBook for AP Statistics**. These problems reinforce two key ideas:

1. **Computing z-scores**
2. **Estimating probabilities using the Empirical Rule (68–95–99.7 rule)**

---

## Problem 5: Z-Score Calculation (Non-Normal Distribution)

### Given
- Mean exam score:  
  $$
  \mu = 2.8
  $$
- Standard deviation:  
  $$
  \sigma = 1.34
  $$
- Exam score of interest:  
  $$
  x = 5
  $$

> Note: The distribution is **not normal**, but z-scores can still be calculated.

---

### Z-Score Formula
$$
z = \frac{x - \mu}{\sigma}
$$

---

### Calculation
$$
z = \frac{5 - 2.8}{1.34} = \frac{2.2}{1.34} \approx 1.64
$$

---

### ✅ Final Answer
- **z-score ≈ 1.64** (Choice **C**)

---

### Key Concept Clarification
A **z-score does not require a normal distribution**.  
It simply measures how many standard deviations a value is from the mean.

---

## Problem 6: Probability Using the Empirical Rule

### Given
- Heights of fifth-grade boys are **approximately normally distributed**
- Mean height:  
  $$
  \mu = 143.5 \text{ cm}
  $$
- Standard deviation:  
  $$
  \sigma = 7.1 \text{ cm}
  $$
- Height of interest:  
  $$
  x = 157.7 \text{ cm}
  $$

---

### Step 1: Determine Standard Deviations Above the Mean

- One standard deviation above the mean:
  $$
  143.5 + 7.1 = 150.6
  $$
- Two standard deviations above the mean:
  $$
  150.6 + 7.1 = 157.7
  $$

So, **157.7 cm is exactly 2 standard deviations above the mean**.

---

### Step 2: Apply the Empirical Rule

- **95%** of values lie within ±2σ
- Remaining **5%** lies outside this range
- Because the distribution is symmetric:
  - **2.5%** lies above +2σ
  - **2.5%** lies below −2σ

---

### ✅ Final Answer
$$
P(X > 157.7) = 2.5\%
$$

---

## Final Takeaways

- **Z-scores** measure distance from the mean in standard deviations and work for *any* distribution.
- When data are **approximately normal**, the **Empirical Rule** provides fast, reliable probability estimates.
- Drawing a labeled bell curve helps clarify where values fall and which area represents the probability of interest.

These problems are excellent practice for building intuition without relying heavily on calculators or tables.


```python
"""
Normal Distribution Practice — Problems 5 & 6 (CK-12 AP Statistics)

This notebook-style example demonstrates:
1. Computing a z-score (even when the distribution is not normal)
2. Estimating probabilities using the Empirical Rule (68–95–99.7)
"""

# -----------------------------
# Problem 5: Z-score calculation
# -----------------------------

# Given values
mean_exam = 2.8
std_exam = 1.34
score = 5

# Z-score formula
z_score = (score - mean_exam) / std_exam

print("Problem 5: Z-score Calculation")
print(f"Mean = {mean_exam}")
print(f"Standard deviation = {std_exam}")
print(f"Score = {score}")
print(f"Z-score ≈ {z_score:.2f}")

# -----------------------------
# Problem 6: Empirical Rule
# -----------------------------

# Given values
mean_height = 143.5      # cm
std_height = 7.1         # cm
height = 157.7           # cm

# Number of standard deviations above the mean
num_std = (height - mean_height) / std_height

print("\nProblem 6: Height Probability Estimate")
print(f"Mean height = {mean_height} cm")
print(f"Standard deviation = {std_height} cm")
print(f"Height of interest = {height} cm")
print(f"Standard deviations above mean = {num_std:.1f}")

# Empirical Rule probabilities
# Within ±2σ → 95%
# Above +2σ → 2.5%
prob_taller_than_height = 0.025

print(f"Probability(height > {height} cm) ≈ {prob_taller_than_height * 100:.1f}%")
```
