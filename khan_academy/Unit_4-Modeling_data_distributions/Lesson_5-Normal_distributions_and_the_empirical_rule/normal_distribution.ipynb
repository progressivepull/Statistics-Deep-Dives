# Normal Distribution Practice — Empirical Rule (68–95–99.7)

This problem comes from the **normal distribution section of CK-12’s open-source AP Statistics book**. Because the problem explicitly says *“without using a calculator”*, it signals that we should use the **Empirical Rule**, also known as the **68–95–99.7 rule**.

---

## Given Information

- The weights of one-year-old girls in the U.S. are **normally distributed**.
- Mean weight:  
  $$
  \mu = 9.5 \text{ kg}
  $$
- Standard deviation:  
  $$
  \sigma = 1.1 \text{ kg}
  $$

> Note: The problem text states grams, but context makes it clear the correct unit is **kilograms**.

---

## Review: The Empirical Rule

For a normal distribution:

- **68%** of values lie within **±1 standard deviation** of the mean  
- **95%** of values lie within **±2 standard deviations**  
- **99.7%** of values lie within **±3 standard deviations**

The distribution is **symmetric**, so areas in the left and right tails are equal.

---

## Key Values on the Distribution

| Standard Deviations | Weight (kg) |
|--------------------|-------------|
| −3σ | 6.2 |
| −2σ | 7.3 |
| −1σ | 8.4 |
| Mean | 9.5 |
| +1σ | 10.6 |
| +2σ | 11.7 |
| +3σ | 12.8 |

---

## Part A  
**What percentage of one-year-old girls weigh less than 8.4 kg?**

- 8.4 kg is **1 standard deviation below the mean**
- Area within ±1σ is **68%**
- Remaining area outside this range is **32%**
- Because the distribution is symmetric:
  $$
  \frac{32\%}{2} = 16\%
  $$

### ✅ Answer:
$$
\boxed{16\%}
$$

---

## Part B  
**What percentage weigh between 7.3 kg and 11.7 kg?**

- 7.3 kg = −2σ  
- 11.7 kg = +2σ  
- The Empirical Rule tells us that **95%** of values lie within ±2σ

### ✅ Answer:
$$
\boxed{95\%}
$$

---

## Part C  
**What percentage weigh more than 12.8 kg?**

- 12.8 kg is **+3 standard deviations**
- Area within ±3σ is **99.7%**
- Remaining area in both tails is:
  $$
  100\% - 99.7\% = 0.3\%
  $$
- Split evenly between two tails:
  $$
  \frac{0.3\%}{2} = 0.15\%
  $$

### ✅ Answer:
$$
\boxed{0.15\%}
$$

---

## Final Answers Summary

| Question | Answer |
|--------|--------|
| Less than 8.4 kg | **16%** |
| Between 7.3 kg and 11.7 kg | **95%** |
| More than 12.8 kg | **0.15%** |

---

## Key Takeaway

When a problem specifies *no calculator* and the data are normally distributed, the **Empirical Rule** provides quick, reliable estimates without needing z-tables or computations.


```python
"""
Normal Distribution — Empirical Rule Example
AP Statistics / CK-12 style problem

This example shows how to:
1. Define a normal distribution
2. Compute key values using the mean and standard deviation
3. Estimate probabilities using the Empirical Rule (no calculator logic)
"""

# Given parameters
mean = 9.5        # kg
std_dev = 1.1     # kg

# Key values at standard deviations
values = {
    "-3σ": mean - 3 * std_dev,
    "-2σ": mean - 2 * std_dev,
    "-1σ": mean - 1 * std_dev,
    "mean": mean,
    "+1σ": mean + 1 * std_dev,
    "+2σ": mean + 2 * std_dev,
    "+3σ": mean + 3 * std_dev,
}

print("Key values on the normal distribution:")
for k, v in values.items():
    print(f"{k:>4}: {v:.1f} kg")

print("\nEmpirical Rule Estimates:")

# Part A: Less than 1 standard deviation below the mean
prob_less_than_1sd_below = 0.16
print(f"Probability(weight < {values['-1σ']:.1f} kg): {prob_less_than_1sd_below * 100:.1f}%")

# Part B: Within 2 standard deviations of the mean
prob_within_2sd = 0.95
print(f"Probability({values['-2σ']:.1f} kg < weight < {values['+2σ']:.1f} kg): {prob_within_2sd * 100:.1f}%")

# Part C: Greater than 3 standard deviations above the mean
prob_more_than_3sd_above = 0.0015
print(f"Probability(weight > {values['+3σ']:.1f} kg): {prob_more_than_3sd_above * 100:.2f}%")
```
