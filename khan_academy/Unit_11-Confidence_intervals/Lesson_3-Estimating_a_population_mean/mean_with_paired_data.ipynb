# Confidence interval for a mean with paired data

[Confidence interval for a mean with paired data | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=gWaA8mVBZQk&t=2s)

A group of friends wondered how much faster they could snap their fingers on one hand versus the other hand. Each person snapped their fingers with their dominant hand for 10 seconds and their non dominant hand for 10 seconds. Each participant flipped a coin to determine which hand they would use first. Here are the data for how many snaps they performed with each hand, the difference for each participant, and summary statistics.       



               

| Participant | Jeff | David | Kim | Charlotte | Jake |
|------------|------|-------|-----|-----------|------|
| Dominant   | 44   | 42    | 40  | 37        | 42   |
| Non‑dominant | 35 | 37    | 32  | 31        | 36   |
| Difference (dom − non) | 9 | 5 | 8 | 6 | 6 |



            


| Summary      | Mean              | Standard deviation     | Sample size |
|--------------|-------------------|-------------------------|-------------|
| Dominant     | $\bar{x}_{\text{dom}}$ = 41.0     | $s_{\text{dom}}$ ≈ 2.65           | 5           |
| Non dominant | $\bar{x}_{\text{non}}$= 34.2     | $s_{\text{non}}$ ≈ 2.59           | 5           |




Here’s a clean, structured breakdown of the **matched‑pairs
t‑interval**, why it works, and a Python simulation that reproduces the
confidence interval for the finger‑snapping experiment.

**⭐ Bullet Summary**

- This is a **matched pairs** design: each participant provides two
  measurements (dominant vs non‑dominant), and we analyze their
  **differences**.

- The sample of differences is: $9,5,8,6,6$.

- Summary statistics:

  - Mean difference: $\overset{ˉ}{d} = 6.8$

  - Standard deviation of differences: $s_{d} = 1.64$

  - Sample size: $n = 5$

- Degrees of freedom:

$$
\text{df} = n - 1 = 4
$$

- For a **95% confidence interval**, the critical value from the
  t‑distribution with 4 df is:

$$
t^{\text{*}} = 2.776
$$

- Margin of error:

$$
2.776 \cdot \frac{1.64}{\sqrt{5}} \approx 2.036
$$

- Final 95% confidence interval:

$$
6.8 \pm 2.036 \Rightarrow (4.764,\text{ 						}8.836)
$$

- Interpretation:

  - We are 95% confident the **true mean difference** in snaps (dominant
    − non‑dominant) for this group lies between **4.764 and 8.836**.

  - Since the entire interval is **above zero**, the dominant hand
    appears faster for this group.

**📘 Explanation (Intuition + Conditions)**

**1. Why matched pairs?**

Each participant is their own control. We analyze the **difference**
$d_{i} = \text{dom}_{i} - \text{non}_{i}$. This reduces variability
and focuses on the effect of handedness.

**2. Conditions for a valid t‑interval**

- **Randomness**: Not random for all humans, but acceptable for
  inference about this group.

- **Normality**: With $n = 5$, we rely on the differences being
  roughly symmetric. The dot plot shows no extreme skew.

- **Independence**: Each participant’s snapping ability doesn’t affect
  others.

Given these, a t‑interval is reasonable.

**3. Building the interval**

The formula for a matched‑pairs t‑interval is:

$$
\overset{ˉ}{d} \pm t^{\text{*}}\frac{s_{d}}{\sqrt{n}}
$$

Plugging in the values:

$$
6.8 \pm 2.776 \cdot \frac{1.64}{\sqrt{5}}
$$

Margin of error ≈ 2.036 Interval ≈ (4.764, 8.836)

**🧪 Python Code Example**

This code reproduces the confidence interval calculation.




```python
import numpy as np
from scipy.stats import t

# Differences from the study
diffs = np.array([9, 5, 8, 6, 6])

# Sample statistics
d_bar = diffs.mean()
s_d = diffs.std(ddof=1)
n = len(diffs)
df = n - 1

# Critical value for 95% CI
t_star = t.ppf(0.975, df)

# Margin of error
ME = t_star * s_d / np.sqrt(n)

# Confidence interval
lower = d_bar - ME
upper = d_bar + ME

print("Mean difference:", d_bar)
print("Standard deviation:", s_d)
print("t* =", t_star)
print("Margin of error:", ME)
print("95% CI =", (lower, upper))
```

**Expected output**

- Mean difference: 6.8

- Standard deviation: 1.64

- $t^{\text{*}} = 2.776$

- Margin of error ≈ 2.036

- CI ≈ (4.764, 8.836)
