What is the critical value t* for constructing a 98% confidence interval for a mean from a sample size of n = 15 observations?

Here’s a clean, structured breakdown of how to find the **critical
value** $t^*$ for a **98% confidence interval**
with **sample size** $n = 15$— plus an intuitive explanation and a
Python example that reproduces the lookup.

**⭐ Bullet Summary**

- We want a **t‑interval** for a population mean because the population
  standard deviation is unknown.

- Sample size: $n = 15$.

- Degrees of freedom:

$$
\text{df} = n - 1 = 14
$$

- Confidence level: **98%**, so we use the **98% column** of a t‑table.

- The critical value $t^*$ is the value such
  that **98% of the t‑distribution** with 14 df lies between
  $- t^*$ and $t^*$.

- Looking up df = 14 and 98% confidence gives:

$$
t^* = 2.624
$$

- This is the number used in the margin of error:

$$
t^* \cdot \frac{s}{\sqrt{n}}
$$

**📘 Explanation (Intuition + Steps)**

**1. Why we use the t‑distribution**

When estimating a mean, we rarely know the population standard
deviation. Using the **t‑distribution** prevents underestimating the
margin of error.

**2. Degrees of freedom**

For a one‑sample t‑interval,

$$
\text{df} = n - 1
$$

So here:

$$
\text{df} = 15 - 1 = 14
$$

**3. Finding the correct column**

A **98% confidence interval** leaves **1% in each tail**. On a t‑table,
this corresponds to the **98% confidence** column (or tail probability
0.01).

**4. Reading the table**

Row: df = 14 Column: 98% confidence Value: **2.624**

That’s the critical value.

**🧪 Python Code Example**

This code computes the same $t^*$value using
SciPy.


```python
from scipy.stats import t

confidence = 0.98
df = 14

# t* is the value where the middle 98% is captured
t_star = t.ppf(0.5 + confidence/2, df)

print("Critical t value (t*):", t_star)
```

Which rounds to **2.624**, matching the t‑table.
