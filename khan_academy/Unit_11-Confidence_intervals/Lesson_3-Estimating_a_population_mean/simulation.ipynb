Below is a clean, structured breakdown of **why z‑intervals fail when σ
is unknown**, **why t‑intervals fix the problem**, and a **Python

**⭐ Bullet Summary**

- A **population mean** is usually unknown, so we estimate it with a
  sample mean.

- A **confidence interval** has the form:

$$
\overset{ˉ}{x} \pm (\text{critical value}) \cdot (\text{standard error})
$$

- If the **population standard deviation** $\sigma$is known, we use a
  z‑interval.

- In practice, $\sigma$is almost never known.

- Replacing $\sigma$with the **sample standard deviation** $s$and
  still using z‑critical values **underestimates the true variability**.

- This produces **confidence intervals that are too narrow**, meaning
  they miss the true mean more often than they should.

- The fix is the t‑distribution, which has heavier tails and adjusts for
  uncertainty in estimating σ.

- The correct interval when σ is unknown is:

$$
\overset{ˉ}{x} \pm t^{\text{*}}\frac{s}{\sqrt{n}}
$$

- Simulation shows:

  - **z + σ** → correct coverage (~95%)

  - **z + s** → under‑coverage (~92% in the example)

  - **t + s** → coverage close to 95% again

- This is why **t‑intervals** are the standard for estimating a
  population mean.

**📘 Explanation (Intuition + Simulation Logic)**

**1. The “ideal world”**

If you magically know the population standard deviation $\sigma$, then
the sampling distribution of $\overset{ˉ}{x}$is normal with standard
error

$$
\frac{\sigma}{\sqrt{n}}
$$

Using z‑critical values gives exactly the right confidence level.

**2. The “real world”**

You almost never know $\sigma$. So you estimate it with the sample
standard deviation $s$. But $s$varies from sample to
sample—sometimes too small, sometimes too large.

Using **z‑critical values** assumes perfect knowledge of σ, so the
interval becomes **too optimistic** (too narrow).

**3. Why the t‑distribution fixes it**

The t‑distribution has heavier tails, meaning it allows for more
uncertainty. Its critical values are larger than z‑values for small
samples.

As sample size grows, t approaches the normal distribution.

**🧪 Python Code Example (Simulation of the Three Scenarios)**

This simulation mirrors the Khan Academy scratchpad:

- True mean = 2.0

- True σ = 0.5

- Sample size = 12

- Confidence level = 95%

- Compare:

  1.  z + σ

  2.  z + s

  3.  t + s



```python
import numpy as np
from scipy.stats import norm, t

# True population parameters
mu = 2.0
sigma = 0.5

# Simulation settings
n = 12
num_samples = 5000
confidence = 0.95

# Critical values
z_star = norm.ppf(0.5 + confidence/2)

# Counters
z_sigma_hits = 0
z_s_hits = 0
t_s_hits = 0

for _ in range(num_samples):
    sample = np.random.normal(mu, sigma, n)
    xbar = np.mean(sample)
    s = np.std(sample, ddof=1)

    # 1. z + sigma (ideal world)
    se_sigma = sigma / np.sqrt(n)
    lower = xbar - z_star * se_sigma
    upper = xbar + z_star * se_sigma
    z_sigma_hits += (lower <= mu <= upper)

    # 2. z + s (incorrect)
    se_s = s / np.sqrt(n)
    lower = xbar - z_star * se_s
    upper = xbar + z_star * se_s
    z_s_hits += (lower <= mu <= upper)

    # 3. t + s (correct)
    t_star = t.ppf(0.5 + confidence/2, df=n-1)
    lower = xbar - t_star * se_s
    upper = xbar + t_star * se_s
    t_s_hits += (lower <= mu <= upper)

print("Coverage using z + sigma:", z_sigma_hits / num_samples)
print("Coverage using z + s    :", z_s_hits / num_samples)
print("Coverage using t + s    :", t_s_hits / num_samples)
```

**Expected output pattern**

- **z + σ** → ~0.95

- **z + s** → ~0.92 (under‑coverage)

- **t + s** → ~0.95 (correct coverage)

This exactly matches the behavior shown in the Khan Academy
demonstration.
