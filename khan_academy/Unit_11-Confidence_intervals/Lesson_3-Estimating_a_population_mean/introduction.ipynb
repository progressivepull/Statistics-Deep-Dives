# Introduction to t statistics

[Introduction to t statistics | Confidence intervals | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=a2rd4Qy8yNI&t=7s)

Here’s a clean, structured breakdown of the idea behind **confidence
intervals**, **unknown population parameters**, and why we switch from
the **z‑distribution** to the **t‑distribution** when estimating a
population mean.

**⭐ Bullet Summary**

- A **population parameter** (like a population mean or population
  proportion) is usually unknown.

- We estimate it using a **sample statistic** (like $\overset{ˉ}{x}$or
  $\widehat{p}$).

- A **confidence interval** has the form:

$$
\text{statistic} \pm \text{critical value} \cdot \text{standard error}
$$

- For a **proportion**, we use the sample proportion $\widehat{p}$and
  estimate the standard error with

$$
\sqrt{\frac{\widehat{p}(1 - \widehat{p})}{n}}
$$

- For a **mean**, if the population standard deviation $\sigma$is
  known (rare), we use a **z‑interval**.

- When $\sigma$is **unknown** (almost always), replacing it with the
  sample standard deviation $s$introduces extra uncertainty.

- That extra uncertainty means the **z‑distribution is too narrow** → it
  **underestimates** the margin of error.

- The fix: use the **t‑distribution**, which has heavier tails and
  depends on **degrees of freedom**.

- The correct confidence interval for a mean with unknown $\sigma$is:

$$
\overset{ˉ}{x} \pm t^{\text{*}}\frac{s}{\sqrt{n}}
$$

- The value $t^{\text{*}}$comes from a t‑table.

**📘 Explanation (Intuition First)**

**1. Why we need confidence intervals**

A **sample statistic** (like $\overset{ˉ}{x}$) varies from sample to
sample. A confidence interval gives a **range of plausible values** for
the true parameter.

**2. Why we estimate the standard error**

The **standard deviation of the sampling distribution** depends on the
population. But we rarely know population values, so we estimate using
sample data.

**3. Why z‑intervals fail when** $\mathbf{\sigma}$**is unknown**

If you plug in the sample standard deviation $s$instead of $\sigma$,
the variability of the statistic increases. The **z‑distribution does
not account for this extra uncertainty**, so the interval becomes too
narrow.

**4. Why the t‑distribution fixes it**

The t‑distribution has **heavier tails**, meaning it allows for more
variability. As sample size increases, the t‑distribution approaches the
normal distribution.

**🧪 Python Code Example**

Below is a simulation showing the difference between **z‑intervals** and
**t‑intervals** when $\sigma$is unknown.

```python
import numpy as np
from scipy.stats import t, norm

# True population parameters (unknown in practice)
mu = 50
sigma = 10

# Simulation settings
n = 15
num_samples = 5000
confidence = 0.95

z_star = norm.ppf(0.5 + confidence/2)

t_star = t.ppf(0.5 + confidence/2, df=n-1)

z_contains = 0
t_contains = 0

for _ in range(num_samples):
    sample = np.random.normal(mu, sigma, n)
    xbar = np.mean(sample)
    s = np.std(sample, ddof=1)

    # z-interval (incorrect when sigma unknown)
    z_lower = xbar - z_star * s / np.sqrt(n)
    z_upper = xbar + z_star * s / np.sqrt(n)

    # t-interval (correct)
    t_lower = xbar - t_star * s / np.sqrt(n)
    t_upper = xbar + t_star * s / np.sqrt(n)

    if z_lower <= mu <= z_upper:
        z_contains += 1
    if t_lower <= mu <= t_upper:
        t_contains += 1

print("Coverage probability (z-interval):", z_contains / num_samples)
print("Coverage probability (t-interval):", t_contains / num_samples)
```

**Expected outcome**

- The **z‑interval** will cover the true mean **less than 95%** of the
  time.

- The **t‑interval** will be **very close to 95%**, as intended.
