# Sample size for a given margin of error for a mean

[Sample size for a given margin of error for a mean | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=FSVDJoHZ3rM&t=2s)

Nadia wants to create a confidence interval to estimate the mean driving range for her company’s new electric vehicle. She wants the margin of error to be no more than 10 kilometers at a 90% level of confidence. A pilot study suggests that the driving ranges for this type of vehicle have a standard deviation of 15 kilometers.         

Which of these is the smallest approximate sample size required to obtain the desired margin of error?     

A 5     

B 7    

C 10     

D 15     


Here’s a crisp, structured breakdown of how to determine the **minimum
sample size** needed so that the **margin of error is no more than 10
km** for a **90% confidence interval** — plus an intuitive explanation
and a Python code example that reproduces the calculation.

**⭐ Bullet Summary**

- Nadia wants a **90% confidence interval** for the **mean driving
  range** of a new EV.

- Desired **margin of error (ME)**: ≤ 10 km.

- Pilot study suggests population standard deviation ≈ 15 km.

- To avoid the complication of unknown degrees of freedom in a
  t‑distribution, we use a **z‑interval approximation**.

- Margin of error formula for means using z:

$$
\text{ME} = z^* \cdot \frac{\sigma}{\sqrt{n}}
$$

- For a **90% confidence level**, the critical value is

$$
z^* = 1.645
$$

- Solve the inequality

$$
1.645 \cdot \frac{15}{\sqrt{n}} \leq 10
$$

- Algebra gives

$$
n \geq 6.09
$$

- Since sample size must be a whole number, the smallest valid $n$is

$$
\boxed{7}
$$

**📘 Explanation (Intuition + Steps)**

**1. Why use a z‑interval instead of a t‑interval?**

A **t‑interval** requires knowing the degrees of freedom, which depends
on $n$, the very thing we’re solving for. A **z‑interval** avoids this
circular dependency and is standard practice when planning sample sizes.

**2. Margin of error requirement**

We want:

$$
\text{ME} = z^* \cdot \frac{\sigma}{\sqrt{n}} \leq 10
$$

Plug in:

- $z^* = 1.645$for 90% confidence

- $\sigma = 15$

So:

$$
1.645 \cdot \frac{15}{\sqrt{n}} \leq 10
$$

**3. Solve for** $\mathbf{n}$

$$
\frac{1.645 \cdot 15}{\sqrt{n}} \leq 10
$$

Divide both sides:

$$
\frac{1}{\sqrt{n}} \leq \frac{10}{1.645 \cdot 15}
$$

Take reciprocals:

$$
\sqrt{n} \geq \frac{1.645 \cdot 15}{10}
$$

Square both sides:

$$
n \geq (1.645 \cdot 1.5)^{2} \approx 6.09
$$

Thus:

$$
\boxed{n = 7}
$$

**🧪 Python Code Example**

This code reproduces the sample‑size calculation.




```python
import math
from scipy.stats import norm

# Desired values
ME = 10
sigma = 15
confidence = 0.90

# Critical z-value
z_star = norm.ppf(0.5 + confidence/2)

# Solve for n
n = (z_star * sigma / ME)**2

print("z* =", z_star)
print("Required n (not rounded):", n)
print("Smallest whole number n:", math.ceil(n))
```

**Expected output**

- z* ≈ 1.645

- Required n ≈ 6.09

- Smallest whole number n = **7**
