# Conditions for valid t intervals

[Conditions for valid t intervals | Confidence intervals | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=WPxnNjD_yoU&t=6s)

Flavia wanted to estimate the mean age of the faculty members at her large university. She took an SRS of 20 of the approximately 700 faculty members, and each faculty member in the sample provided Flavia with their age. The data were skewed to the right with a sample mean of x̄ = 38.75 . She's considering using her data to make a confidence interval to estimate the mean age of faculty members at her university.

Which conditions for constructing a t interval have been met?                             

Choose all answers that apply:                       

A. The data is a random sample from the population of interest.       
B. The sampling distribution of x̄ is approximately normal.       
C. Individual observations can be considered independent.        


Here’s a clean, structured breakdown of the **t‑interval conditions**,
why they matter, and a **Python simulation** that mirrors the scenario.

**⭐ Bullet Summary**

- Flavia took an **SRS** of 20 faculty members from a population of
  ~700.

- The goal is to estimate the **population mean** age using a
  **t‑interval**.

- Three conditions must be checked:

  - **Randomness condition** → ✔ Met (sample is an SRS).

  - **Independence condition** → ✔ Met (20 is \< 10% of 700).

  - **Normality condition** → ✘ Not met (sample size \< 30 and data is
    right‑skewed).

- Because the normality condition fails, the t‑interval is **not fully
  reliable** here.

- The main issue: the **sampling distribution of** $\overset{ˉ}{x}$may
  not be approximately normal.

**📘 Explanation (Clear + Intuitive)**

**1. Randomness Condition**

The sample must come from a **random process**. Flavia used an **SRS**,
so this condition is satisfied. This corresponds to:

- **Choice A**: “The data is a random sample from the population of
  interest.”

**2. Independence Condition**

Observations must be independent. Two ways to satisfy this:

- Sampling **with replacement**, or

- Sampling **without replacement** but sample size ≤ 10% of population.

Here:

- Population = 700

- Sample = 20

- $20 \leq 0.10 \times 700 = 70$

So independence holds. This corresponds to:

- **Choice C**: “Individual observations can be considered independent.”

**3. Normality Condition**

We need the **sampling distribution of** $\overset{ˉ}{x}$to be
approximately normal.

This can happen if:

- Sample size $n \geq 30$, OR

- The population distribution is normal or roughly symmetric.

But here:

- $n = 20$(too small for CLT guarantee)

- Data is **right‑skewed**, not symmetric

- Population distribution likely skewed as well

So the normality condition **fails**. This eliminates:

- **Choice B**.

**✔ Final Answer**

The conditions that **have been met** are:

- **A. The data is a random sample from the population of interest.**

- **C. Individual observations can be considered independent.**

The **normality condition is NOT met**, so a t‑interval is questionable
here.

**🧪 Python Code Example**

Below is a simulation showing how skewed data with $n = 20$ can cause
t‑intervals to misbehave.




```python
import numpy as np
from scipy.stats import t

# Simulate a right-skewed population (e.g., lognormal)
np.random.seed(0)
population = np.random.lognormal(mean=3.5, sigma=0.4, size=100000)

true_mean = population.mean()
n = 20
confidence = 0.95
num_samples = 5000

hits = 0

for _ in range(num_samples):
    sample = np.random.choice(population, n, replace=False)
    xbar = sample.mean()
    s = sample.std(ddof=1)
    
    t_star = t.ppf(0.5 + confidence/2, df=n-1)
    lower = xbar - t_star * s / np.sqrt(n)
    upper = xbar + t_star * s / np.sqrt(n)
    
    hits += (lower <= true_mean <= upper)

print("Coverage rate of t-interval with skewed data:", hits / num_samples)
```

**Expected outcome**

Coverage will be **below 95%**, showing why the normality condition
matters when $n < 30$.
