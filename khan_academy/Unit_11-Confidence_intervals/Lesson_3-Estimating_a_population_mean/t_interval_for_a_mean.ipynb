# Example constructing a t interval for a mean

[Example constructing a t interval for a mean | Confidence intervals | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=6G3jMjAqNqM&t=1s)

A nutritionist wants to estimate the average caloric content of the burritos at a popular restaurant. They obtain a random sample of 14 burritos and measure their caloric content. Their sample data are roughly symmetric with a mean of 700 calories and a standard deviation of 50 calories.        

Based on this sample, which of the following is a 95% confidence interval for the mean caloric content of these burritos?           

A 700 ± 23.7                  
B 700 ± 26.2                 
C 700 ± 28.7                
D 700 ± 28.9                 
E 700 ± 100                      


Below is a clean, structured breakdown of how to build the **95%
t‑interval** for the burrito‑calorie problem — with a crisp bullet
summary, an intuitive explanation, and a Python code example that
reproduces the calculation.

**⭐ Bullet Summary**

- We want a **t‑interval** because the population standard deviation is
  unknown.

- Sample size: $n = 14$.

- Sample mean: $\overset{ˉ}{x} = 700$.

- Sample standard deviation: $s = 50$.

- Degrees of freedom:

$$
\text{df} = n - 1 = 13
$$

- For a **95% confidence interval**, the critical value from the
  **t‑distribution** with 13 df is:

$$
t^{\text{*}} = 2.160
$$

- Margin of error:

$$
2.160 \cdot \frac{50}{\sqrt{14}} \approx 28.86
$$

- Final 95% confidence interval:

$$
700 \pm 28.86 \Rightarrow (671.14,\text{ 						}728.86)
$$

**📘 Explanation (Intuition + Steps)**

**1. Why we use the t‑distribution**

We don’t know the population standard deviation, so we use the **sample
standard deviation**. This introduces extra uncertainty, and the
**t‑distribution** accounts for it with heavier tails.

**2. Degrees of freedom**

For a one‑sample t‑interval:

$$
\text{df} = n - 1 = 13
$$

**3. Finding the critical value**

A **95% confidence interval** leaves **2.5% in each tail**. On a
t‑table, look up:

- df = 13

- tail probability = 0.025

This gives:

$$
t^{\text{*}} = 2.160
$$

**4. Margin of error**

$$
\text{ME} = t^{\text{*}} \cdot \frac{s}{\sqrt{n}}
$$

$$
= 2.160 \cdot \frac{50}{\sqrt{14}} \approx 28.86
$$

**5. Final interval**

$$
700 - 28.86 = 671.14
$$

$$
700 + 28.86 = 728.86
$$

So the 95% confidence interval is approximately:

$$
( 671.1, 728.9 )
$$

**🧪 Python Code Example**

This code reproduces the t‑interval calculation.




```python
import numpy as np
from scipy.stats import t

# Given sample statistics
xbar = 700
s = 50
n = 14
df = n - 1

# Critical value for 95% CI
t_star = t.ppf(0.975, df)

# Margin of error
ME = t_star * s / np.sqrt(n)

# Confidence interval
lower = xbar - ME
upper = xbar + ME

print("t* =", t_star)
print("Margin of error =", ME)
print("95% CI =", (lower, upper))
```

**Expected output**

- $t^{\text{*}} \approx 2.160$

- Margin of error ≈ 28.86

- CI ≈ (671.14, 728.86)
