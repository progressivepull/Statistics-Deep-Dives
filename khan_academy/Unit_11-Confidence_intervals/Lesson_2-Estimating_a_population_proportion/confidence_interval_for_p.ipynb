# Example constructing and interpreting a confidence interval for p

[ Example constructing and interpreting a confidence interval for p | AP Statistics | Khan Academy| YouTube](https://www.youtube.com/watch?v=sjG9Ns_8x6c&t=3s)

**Della has over 500 songs on her mobile phone, and she wants to
estimate what proportion of the songs are by a female artist. She takes
an SRS of 50 songs on her phone and finds that 20 of the songs sampled
are by a female artist.**

**Based on this sample, which of the following is a 99% confidence
interval for the proportion of songs on her phone that are by a female
artist?”**

A\)

$$
20 \pm 1.96\sqrt{\frac{20(30)}{50}}
$$

B\)

$$
20 \pm 2.576\sqrt{\frac{20(30)}{50}}
$$

C\)

$$
0.40 \pm 1.96\sqrt{\frac{0.40(0.60)}{50}}
$$

D\)

$$
0.40 \pm 2.576\sqrt{\frac{0.40(0.60)}{50}}
$$


Here you go, John — a clean **bullet summary**, a clear **explanation**,
and a **Python code example** for Della’s 99% confidence‑interval
problem. This one is a classic: identify the correct formula structure
and the correct **critical value** $z^{\text{*}}$.

**📌 Bullet Summary**

**Given**

- Population: 500+ songs

- Sample size: $n = 50$

- Songs by female artists: $20$

- Sample proportion:

$$
\widehat{p} = \frac{20}{50} = 0.40
$$

- Confidence level: **99%**

**Conditions Check**

- **Random sample**: Yes — SRS given.

- **Normal condition** (success–failure):

$$
20 \geq 10,30 \geq 10
$$

✔️ Condition met.

- **Independence condition** (10% rule):

$$
50 < 0.10 \times 500
$$

✔️ Condition met.

**Confidence Interval Formula**

$$
\widehat{p} \pm z^{\text{*}}\sqrt{\frac{\widehat{p}(1 - \widehat{p})}{n}}
$$

**Critical Value for 99% CI**

- Middle 99% → 0.5% in each tail

- Look up cumulative area $0.995$

- $z^{\text{*}} = 2.576$

**Correct Answer Choice**

$$
0.40 \pm 2.576\sqrt{\frac{0.40(0.60)}{50}}
$$

This corresponds to **choice D**.

**🧠 Explanation**

Della wants a **99% confidence interval** for the proportion of songs by
female artists. She takes a simple random sample of 50 songs and finds
that 20 are by female artists, giving a sample proportion of
$\widehat{p} = 0.40$.

To build a confidence interval for a proportion, we use:

$$
\widehat{p} \pm z^{\text{*}}\sqrt{\frac{\widehat{p}(1 - \widehat{p})}{n}}
$$

The only question is: **which** $z^{\text{*}}$**matches a 99%
confidence level?**

A 99% CI leaves 1% outside the middle — 0.5% in each tail. A standard
normal table shows that the z‑value with 0.995 cumulative area is
**2.576**.

So the correct interval is:

$$
0.40 \pm 2.576\sqrt{\frac{0.40(0.60)}{50}}
$$

That’s exactly **choice D**.

**💻 Python Code Example**

Here’s a quick Python snippet that computes the 99% confidence interval:


```python
import math

# Sample data
n = 50
p_hat = 20 / 50

# Standard error
SE = math.sqrt(p_hat * (1 - p_hat) / n)

# Critical value for 99% CI
z_star = 2.576

# Margin of error
ME = z_star * SE

# Confidence interval
lower = p_hat - ME
upper = p_hat + ME

print(f"Sample proportion: {p_hat:.3f}")
print(f"Standard Error: {SE:.4f}")
print(f"Margin of Error: {ME:.4f}")
print(f"99% Confidence Interval: [{lower:.4f}, {upper:.4f}]")
```
