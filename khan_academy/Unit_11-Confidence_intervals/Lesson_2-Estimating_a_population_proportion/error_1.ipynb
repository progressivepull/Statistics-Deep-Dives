**📌 Bullet Summary: Sampling, Proportions, and Confidence Intervals**

**Scenario**

- Population: 100 million voters

- Two candidates: A and B

- True proportion voting for B = $p$(unknown)

- Sample: 100 randomly selected voters

- Sample results:

  - 57 vote A → 0

  - 43 vote B → 1

- Sample proportion:

$$
\widehat{p} = \frac{43}{100} = 0.43
$$

**Sample Variance and Standard Deviation**

- Sample variance formula:

$$
s^{2} = \frac{\sum(x_{i} - \widehat{p})^{2}}{n - 1}
$$

- Computed sample variance: **0.2475**

- Sample standard deviation:

$$
s = \sqrt{0.2475} \approx 0.50
$$

**Sampling Distribution of the Sample Mean**

- Mean of sampling distribution:

$$
\mu_{\widehat{p}} = p
$$

- Standard deviation (standard error):

$$
SE = \frac{s}{\sqrt{n}} = \frac{0.50}{10} = 0.05
$$

**Goal**

Construct a **95% confidence interval** for the true proportion $p$.

**Confidence Interval Logic**

- For 95% confidence, use **±2 standard errors** (approximation).

- Margin of error:

$$
ME = 2 \times 0.05 = 0.10
$$

- Confidence interval:

$$
0.43 \pm 0.10 = \lbrack 0.33,\text{ 						}0.53\rbrack
$$

**Interpretation**

- We are **95% confident** that the true proportion of voters choosing
  candidate B lies between **33% and 53%**.

- Meaning: If we repeatedly sampled 100 voters and built intervals the
  same way, **about 95%** of those intervals would contain the true
  proportion $p$.

**🧠 Explanation in Plain English**

You can’t ask all 100 million people who they’ll vote for, so you take a
sample of 100. In that sample, 43% say they’ll vote for candidate B. But
that number will bounce around from sample to sample — sometimes 40%,
sometimes 47%, sometimes 52%.

To understand how much it bounces, you estimate the **standard error**,
which tells you how far your sample proportion is likely to be from the
true proportion.

Once you know the standard error, you can build a **confidence
interval**. A 95% confidence interval means:

If you repeated this sampling process over and over, 95% of the
intervals you build would contain the true proportion of voters
supporting candidate B.

It does *not* mean there is a 95% chance that the true proportion is in
this specific interval — the true proportion is fixed. The randomness is
in the sampling process.

**💻 Python Code Example: Compute the 95% Confidence Interval**

Here’s a clean Python snippet that reproduces the calculations:




```python
import math

# Sample data
n = 100
successes = 43

# Sample proportion
p_hat = successes / n

# Sample variance (Bernoulli)
# s^2 = p_hat * (1 - p_hat) * n / (n - 1)
s2 = (p_hat * (1 - p_hat)) * n / (n - 1)
s = math.sqrt(s2)

# Standard error
SE = s / math.sqrt(n)

# 95% confidence uses approx z = 2
z = 2
ME = z * SE

# Confidence interval
lower = p_hat - ME
upper = p_hat + ME

print(f"Sample proportion: {p_hat:.3f}")
print(f"Sample variance: {s2:.3f}")
print(f"Sample SD: {s:.3f}")
print(f"Standard Error: {SE:.3f}")
print(f"95% Confidence Interval: [{lower:.3f}, {upper:.3f}]")
```
