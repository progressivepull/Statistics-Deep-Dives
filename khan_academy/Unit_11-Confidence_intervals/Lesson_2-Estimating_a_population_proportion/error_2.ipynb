**📌 Bullet Summary: Building a 95% Confidence Interval for a
Proportion**

**Given**

- Sample size: $n = 100$

- Sample results: 43 voters for candidate B → $\widehat{p} = 0.43$

- Sample variance (computed earlier): $s^{2} \approx 0.2475$

- Sample standard deviation: $s \approx 0.50$

**Sampling Distribution**

- Mean of sampling distribution:

$$
\mu_{\widehat{p}} = p
$$

- Standard error (estimated):

$$
SE = \frac{s}{\sqrt{n}} = \frac{0.50}{10} = 0.05
$$

**Normal Approximation**

- For a normal distribution:

  - About **95%** of values lie within **±2 standard deviations** of the
    mean.

- So:

$$
\widehat{p} \pm 2 \cdot SE = 0.43 \pm 0.10
$$

**95% Confidence Interval**

$$
\left\lbrack 0.33,\text{ }\text{	}\text{	}\text{	}\text{	}\text{	}\text{	}\text{	}\text{	}0.53 \right\rbrack
$$

**Interpretation**

- We are **95% confident** that the true proportion of voters supporting
  candidate B lies between **33% and 53%**.

- Meaning: If we repeatedly sampled 100 voters and built intervals the
  same way, **about 95%** of those intervals would contain the true
  proportion $p$.

**Margin of Error**

- Margin of error = $2 \times SE = 0.10$

- This is why polls often say: “Candidate B: 43% ± 10% (95%
  confidence).”

**How to Narrow the Interval**

- Increase sample size $n$. Larger $n$→ smaller SE → smaller margin
  of error → tighter interval.

**🧠 Explanation in Plain English**

You took a sample of 100 voters and found that 43% support candidate B.
But that number will bounce around from sample to sample. To understand
how much it bounces, you estimate the **standard error**, which tells
you how far your sample proportion is likely to be from the true
proportion.

Once you know the standard error, you can build a **95% confidence
interval** by going **two standard errors** above and below your sample
proportion.

This gives you a range — 33% to 53% — where you believe the true
proportion lies.

The key idea is this:

If you repeated this sampling process many times, 95% of the intervals
you build would contain the true proportion.

It’s not that the true proportion moves around — it’s fixed. It’s the
**intervals** that move around from sample to sample.

**💻 Python Code Example: Compute the 95% Confidence Interval**


```python
import math

# Sample data
n = 100
successes = 43

# Sample proportion
p_hat = successes / n

# Sample variance (Bernoulli estimate)
s2 = (p_hat * (1 - p_hat)) * n / (n - 1)
s = math.sqrt(s2)

# Standard error
SE = s / math.sqrt(n)

# 95% confidence uses approx z = 2
z = 2
ME = z * SE

# Confidence interval
lower = p_hat - ME
upper = p_hat + ME

print(f"Sample proportion: {p_hat:.3f}")
print(f"Standard Error: {SE:.3f}")
print(f"Margin of Error: {ME:.3f}")
print(f"95% Confidence Interval: [{lower:.3f}, {upper:.3f}]")
```
