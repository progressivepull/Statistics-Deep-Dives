# Conditions for valid confidence intervals for a proportion

[Conditions for valid confidence intervals | Confidence intervals | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=8E8YQY5qE3s&t=5s)

**📌 Bullet Summary: When Confidence Intervals Work — and When They
Don’t**

**1. What Confidence Intervals Assume**

To trust a confidence interval for a proportion, three conditions must
hold:

**A. Random Sample**

- The sample must be representative of the population.

- If you only survey seniors, college students, or one neighborhood, the
  interval is biased before you even compute it.

**B. Normal Condition (a.k.a. Success–Failure Condition)**

- The sampling distribution of the sample proportion must be
  approximately normal.

- Rule of thumb:


Number of successes ≥ 10 and number of failures ≥ 10     

- If you have fewer than 10 successes or failures, the normal
  approximation breaks down → confidence interval accuracy drops.

**C. Independence Condition (10% Rule)**

- When sampling **without replacement**, the sample size must be less
  than 10% of the population:

$$
n < 0.10 \cdot N
$$

- Ensures that each observation is “almost independent.”

**2. What Happens When Conditions Break**

**If independence fails (sample too large relative to population):**

- Standard error is underestimated.

- Confidence intervals become too narrow.

- Actual coverage drops below the advertised level (e.g., 95% becomes
  ~93%).

**If normal condition fails (too few successes/failures):**

- Sampling distribution is skewed.

- Confidence intervals miss the true parameter more often than expected.

- Example from the simulation: only ~94% coverage instead of 95%.

**3. Big Takeaway**

Confidence intervals are only as good as the assumptions behind them.
If:

- the sample is random,

- the success–failure condition is met, and

- the independence condition holds,

…then a 95% confidence interval will contain the true parameter about
**95% of the time** across repeated sampling.

Break the assumptions → break the guarantee.

**🧠 Explanation in Plain English**

A confidence interval is a promise about long‑run performance: “If we
repeated this sampling process over and over, 95% of the intervals we
build will contain the true proportion.”

But that promise only holds if the math behind the interval is valid.

The math assumes:

- your sample behaves like independent draws,

- the sampling distribution is roughly normal,

- and the sample actually represents the population.

If you violate these assumptions — say you sample 200 out of a
population of 1,000 (too big), or you only get 3 successes (too few), or
your sample isn’t random — then the confidence interval no longer
performs as advertised.

You might think you’re building a 95% interval, but in reality it might
only capture the truth 93% of the time. That’s the whole point of the
simulations in the video: to show that the method *fails quietly* when
assumptions fail.

**💻 Python Code Example: Simulating Coverage When Assumptions Break**

Below is a simulation showing how confidence interval coverage changes
when the **normal condition** fails (too few successes/failures).


```python
import math

# Sample data
n = 50
successes = 12

# Sample proportion
p_hat = successes / n

# Check success–failure condition
if successes >= 10 and (n - successes) >= 10:
    print("Success–failure condition met.")
    
    # Standard error
    SE = math.sqrt(p_hat * (1 - p_hat) / n)
    
    # 95% confidence interval (z = 1.96)
    z = 1.96
    ME = z * SE
    
    lower = p_hat - ME
    upper = p_hat + ME
    
    print(f"95% CI: [{lower:.3f}, {upper:.3f}]")
else:
    print("Success–failure condition NOT met.")
    print("Normal-based confidence interval may be inaccurate.")
```
