# Critical value (z*) for a given confidence level

[Critical value (z*) for a given confidence level | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=VILsYH84HJE&t=2s)

Elena wants to build a one-sample z interval to estimate what proportion of computers produced at a factory have a certain defect. She chooses a confidence level of 94%. A random sample of 200 computers shows that 12 computers have the defect.

What critical value z* should Elena use to construct this confidence interval?

Here you go, John — a clean **bullet summary**, a clear **explanation**,
and a **Python code example** for Elena’s one‑sample z‑interval problem.
This one is all about finding the **critical value** $z^{\text{*}}$for
a **94% confidence interval**.

**📌 Bullet Summary: Finding the Critical Value**
$\mathbf{z}^{\text{*}}$**for 94% Confidence**

**Given**

- Confidence level: **94%**

- Sample size: $n = 200$

- Defective computers: 12 (This is not needed to find $z^{\text{*}}$,
  but part of the context.)

**Goal**

Find the **critical value** $z^{\text{*}}$such that the middle **94%**
of a standard normal distribution is captured.

**Key Steps**

1.  A 94% confidence interval leaves **6%** outside the middle.

2.  Because the normal distribution is symmetric:

$$
\frac{6\%}{2} = 3\%
$$

is in each tail.

3.  So we want the z‑value where **97%** of the area is to the left:

$$
P(Z \leq z^{\text{*}}) = 0.97
$$

4.  Look up 0.97 in a z‑table → closest value is **1.88**.

**Final Answer**

$$
z^{\text{*}} = 1.88
$$

**🧠 Explanation**

A confidence interval is built by taking your statistic (in this case,
the sample proportion) and going **plus or minus** a certain number of
standard deviations. That number of standard deviations is the
**critical value** $z^{\text{*}}$.

For a 94% confidence interval, you want the middle 94% of the normal
curve. That leaves 6% outside the interval — 3% in each tail.

Most z‑tables give you the **cumulative area from the far left up to
z**. So instead of looking for 94%, you look for:

$$
0.94 + 0.03 = 0.97
$$

The z‑value with cumulative area 0.97 is **1.88**. That’s the number of
standard deviations you go above and below your sample proportion to
build the interval.

**💻 Python Code Example: Compute** $\mathbf{z}^{\text{*}}$**for 94%
Confidence**

Here’s a quick Python snippet using scipy.stats to compute the critical
value:




```python
from scipy.stats import norm

# Confidence level
confidence = 0.94

# Tail probability
alpha = 1 - confidence
tail = alpha / 2

# Critical value z*
z_star = norm.ppf(1 - tail)

print(f"Critical value z*: {z_star:.3f}")
```

Which rounds to **1.88**, matching the z‑table result.
