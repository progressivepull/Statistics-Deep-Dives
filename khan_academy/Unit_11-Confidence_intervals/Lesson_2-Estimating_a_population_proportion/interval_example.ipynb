# Confidence interval example

[Confidence interval example | Inferential statistics | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=SeQeYVJZ2gE&t=6s)

In a local teaching district a technology grant is available to teachers in order to install a cluster of four computers in their classrooms. From the 6250 teachers in the district, 250 were randomly selected and asked if they felt that computers were an essential teaching tool for their classroom. Of those selected, 142 teachers felt that computers were an essential teaching tool.

1. Calculate a 99% confidence interval for the proportion of teachers who felt that computers are an essential teaching tool.

2. How could the survey be changed to narrow the confidence interval but to maintain the 99% confidence interval?

**📌 Bullet Summary: 99% Confidence Interval for Teacher Proportion**

**Given**

- Population: 6,250 teachers

- Sample: 250 teachers

- Number who said computers are essential: 142

- Sample proportion:

$$
\widehat{p} = \frac{142}{250} = 0.568
$$

**Standard Error**

- Standard error for a proportion:

$$
SE = \sqrt{\frac{\widehat{p}(1 - \widehat{p})}{n}}
$$

- Plug in values:

$$
SE \approx \sqrt{\frac{0.568 \cdot 0.432}{250}} \approx 0.031
$$

**Critical Value for 99% CI**

- For 99% confidence, the z‑score is approximately **2.58**.

**Margin of Error**

$$
ME = z \cdot SE = 2.58 \times 0.031 \approx 0.08
$$

**Confidence Interval**

$$
\widehat{p} \pm ME = 0.568 \pm 0.08
$$

$$
\text{CI} = \lbrack 0.488,\text{ 						}0.648\rbrack
$$

**Interpretation**

- We are **99% confident** that the true proportion of teachers who
  believe computers are essential lies between **48.8% and 64.8%**.

- This means: If we repeatedly took samples of 250 teachers and built
  99% confidence intervals, **about 99% of those intervals would contain
  the true population proportion**.

**How to Narrow the Interval**

- Increase the sample size. Larger $n$→ smaller SE → smaller margin of
  error → narrower CI.

**🧠 Explanation in Plain English**

You surveyed 250 teachers and found that 142 of them think computers are
essential. That gives a sample proportion of 0.568. But because this is
only a sample, it naturally contains sampling variation.

A **99% confidence interval** gives a range of plausible values for the
*true* proportion of all 6,250 teachers who feel this way.

To build the interval, you estimate how much your sample proportion
might vary (the standard error), then expand outward by a multiplier
(2.58) that corresponds to 99% confidence.

The result is a range from **48.8% to 64.8%**.

This does *not* mean there is a 99% chance the true proportion is in
this specific interval. Instead, it means the **method** used to create
the interval succeeds 99% of the time in the long run.

**💻 Python Code Example**

Here’s a clean Python snippet that computes the 99% confidence interval:


```python
import math

# Sample data
n = 250
successes = 142

# Sample proportion
p_hat = successes / n

# Standard error
SE = math.sqrt(p_hat * (1 - p_hat) / n)

# z-score for 99% confidence
z = 2.58

# Margin of error
ME = z * SE

# Confidence interval
lower = p_hat - ME
upper = p_hat + ME

print(f"Sample proportion: {p_hat:.3f}")
print(f"Standard Error: {SE:.3f}")
print(f"Margin of Error: {ME:.3f}")
print(f"99% Confidence Interval: [{lower:.3f}, {upper:.3f}]")
```
