# Determining sample size based on confidence and margin of error

[Determining sample size based on confidence and margin of error | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=VyFs7fsWE6w&t=3s)

Della wants to make a one-sample z interval to estimate what proportion of her community members favor a tax increase for more local school funding. She wants her margin of error to be no more than ±2% at the 95% confidence level.

What is the smallest sample size required to obtain the desired margin of error?

A. 267     
B. 601     
C. 1068      
D. 2401        

Here you go, John — a clean **bullet summary**, a clear **explanation**,
and a **Python code example** for Della’s *sample‑size for a desired
margin of error* problem. This is one of the most important mechanics in
confidence‑interval work: solving for **n** when you want a specific
margin of error.

**📌 Bullet Summary**

**Goal**

Della wants:

- A **one‑sample z interval** for a proportion

- **95% confidence level**

- **Margin of error ≤ 2%** (i.e., $\leq 0.02$)

**General Margin‑of‑Error Formula**

For a proportion:

$$
ME = z^{\text{*}}\sqrt{\frac{\widehat{p}(1 - \widehat{p})}{n}}
$$

**Critical Value for 95% CI**

$$
z^{\text{*}} = 1.96
$$

**Worst‑Case Proportion**

Because Della has **not** taken a sample yet, we use the value of
$\widehat{p}$that **maximizes**

$$
\widehat{p}(1 - \widehat{p})
$$

That maximum occurs at:

$$
\widehat{p} = 0.5
$$

**Set Up the Inequality**

$$
1.96\sqrt{\frac{0.5(0.5)}{n}} \leq 0.02
$$

**Solve for** $\mathbf{n}$

This leads to:

$$
n \geq 2401
$$

**✔️ Correct smallest sample size: 2401**

**🧠 Explanation**

Della wants her margin of error to be no more than ±2%. The margin of
error for a proportion depends on three things:

1.  The **critical value** $z^{\text{*}}$

2.  The **sample size** $n$

3.  The **sample proportion** $\widehat{p}$

But since she hasn’t taken the sample yet, she doesn’t know
$\widehat{p}$. To be safe, statisticians use the value that makes the
margin of error **largest**, which is $\widehat{p} = 0.5$. This
ensures the sample size is large enough no matter what the actual
proportion turns out to be.

Plugging into the inequality and solving gives:

$$
n \geq 2401
$$

So she needs to sample **at least 2401 people** to guarantee a margin of
error of 2% at the 95% confidence level.

**💻 Python Code Example**

Here’s a Python snippet that computes the required sample size:


```python
import math

# Desired margin of error
ME = 0.02

# Critical value for 95% CI
z = 1.96

# Worst-case p-hat
p_hat = 0.5

# Solve for n
n = (z**2 * p_hat * (1 - p_hat)) / (ME**2)

print(f"Required sample size (rounded up): {math.ceil(n)}")
```
