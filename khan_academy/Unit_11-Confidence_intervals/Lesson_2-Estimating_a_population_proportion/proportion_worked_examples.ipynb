Ali is in charge of the dinner menu for his senior prom, and he wants to use a one-sample z interval to estimate what proportion of seniors would order a vegetarian option. He randomly selects 30 of the 150 total seniors and finds that 7 of those sampled would order the vegetarian option.

Which conditions for constructing this confidence interval did Ali's sample meet?

Choose all answers that apply:

A The data is a random sample from the population of interest.

B n p̂ ≥ 10 and n(1 − p̂) ≥ 10

C  Individual observation can be considered independent. 

Here you go, John — a clean **bullet summary**, a clear **explanation**,
and a **Python code example** that captures exactly what’s going on in
Ali’s prom‑dinner sampling problem. This one is all about checking
whether the **conditions for a one‑sample Z interval** are satisfied.

**📌 Bullet Summary: Which Conditions Did Ali’s Sample Meet?**

**Given**

- Population size: 150 seniors

- Sample size: $n = 30$

- Vegetarians in sample: 7

- Sample proportion:

$$
\widehat{p} = \frac{7}{30} \approx 0.233
$$

**Conditions for a One‑Sample Z Interval**

**1. Random Condition — ✔️ Met**

- Ali *randomly selected* 30 seniors.

- The problem explicitly states this, so we accept it as true.

**2. Normal Condition (Success–Failure Condition) — ❌ Not Met**

- Need:

$$
n\widehat{p} \geq 10\text{and}n(1 - \widehat{p}) \geq 10
$$

- Compute:

$$
n\widehat{p} = 30 \cdot \frac{7}{30} = 7 < 10
$$

$$
n(1 - \widehat{p}) = 30 \cdot \frac{23}{30} = 23 \geq 10
$$

- Failures ≥ 10, but **successes = 7**, which is **less than 10** →
  condition violated.

**3. Independence Condition (10% Rule) — ❌ Not Met**

- Need:

$$
n < 0.10 \cdot N
$$

- Compute:

$$
30 < 0.10 \cdot 150 = 15
$$

- But $30 > 15$→ sample is **20%** of population → independence
  condition violated.

**Final Verdict**

- **Only the random condition is satisfied.**

- **Normal condition fails.**

- **Independence condition fails.**

- A one‑sample Z interval is **not appropriate** for this data.

**🧠 Explanation**

Ali wants to estimate the proportion of seniors who would choose a
vegetarian dinner option. He takes a random sample of 30 students and
finds that 7 would choose vegetarian.

But before building a confidence interval, he must check whether the
**assumptions behind the Z interval** are valid.

The random condition is fine — he sampled randomly.

But the **normal condition** fails because he only has 7 “successes,”
and the rule of thumb requires at least 10 successes and 10 failures to
ensure the sampling distribution of $\widehat{p}$is approximately
normal.

The **independence condition** also fails because he sampled 30 out of
150 students — that’s 20% of the population, exceeding the 10% rule for
sampling without replacement.

Because two of the three conditions fail, Ali’s Z interval would not be
reliable.

**💻 Python Code Example: Automatically Check the Conditions**




```python
import math

# Sample data
N = 150        # population size
n = 30         # sample size
successes = 7  # vegetarians

# Sample proportion
p_hat = successes / n

# Conditions
random_condition = True  # given in the problem

normal_condition = (successes >= 10) and ((n - successes) >= 10)
independence_condition = n < 0.10 * N

print(f"Sample proportion: {p_hat:.3f}")
print(f"Random condition met: {random_condition}")
print(f"Normal condition met: {normal_condition}")
print(f"Independence condition met: {independence_condition}")

if all([random_condition, normal_condition, independence_condition]):
    print("A one-sample Z interval is appropriate.")
else:
    print("A one-sample Z interval is NOT appropriate.")
```
