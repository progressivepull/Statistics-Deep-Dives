# Small sample size confidence intervals

[Small sample size confidence intervals | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=K4KDLWENXm0&t=2s)

7 patients’ blood pressures have been measured after having been given a new drug for 3 months. They had blood pressure increases of 1.5, 2.9, 0.9, 3.9, 3.2, 2.1 and 1.9. Construct a 95% confidence interval for the true expected blood pressure increase for all patients in a population.

![table_data.png](./IMAGES/table_data.png)

**Bullet summary**

- **Goal:** Build a 95% confidence interval for the true mean
  blood‑pressure increase after taking a new drug.

- **Sample size:** $n = 7$

- **Sample mean:** $\overset{ˉ}{x} = 2.34$

- **Sample standard deviation:** $s = 1.04$

- **Reason for using a t‑distribution:**

  - Population SD is unknown

  - Small sample size ($n < 30$)

  - Extra uncertainty → heavier tails

- **Degrees of freedom:**

$$
df = n - 1 = 6
$$

- **Critical t‑value** for a 95% two‑sided interval with 6 df:

$$
t_{0.975,6} = 2.447
$$

- **Standard error:**

$$
SE = \frac{s}{\sqrt{n}} = \frac{1.04}{\sqrt{7}} \approx 0.39
$$

- **Margin of error:**

$$
ME = t_{0.975,6} \cdot SE \approx 2.447 \cdot 0.39 \approx 0.96
$$

- **95% confidence interval:**

$$
\overset{ˉ}{x} \pm ME = 2.34 \pm 0.96 = (1.38,\text{ 						}3.30)
$$

- **Interpretation:** We are 95% confident the true mean blood‑pressure
  increase for all patients lies between **1.38** and **3.30**.

**Explanation (clear and intuitive)**

The sample of 7 patients gives us a mean increase of **2.34** units.
Because the sample is small and we don’t know the population standard
deviation, we use the **t‑distribution** instead of the normal
distribution.

The t‑distribution accounts for the fact that the sample standard
deviation $s = 1.04$is a noisy estimate when $n$is small. This extra
uncertainty shows up as **fatter tails**, meaning we need a larger
critical value (2.447 instead of 1.96).

We compute the **standard error** using

$$
SE = \frac{s}{\sqrt{n}} = 0.39
$$

Then multiply by the critical t‑value to get the **margin of error**:

$$
ME = 0.96
$$

Finally, we add and subtract this from the sample mean to get the
confidence interval:

$$
( 1.38,3.30)
$$

This interval estimates where the true mean blood‑pressure increase lies
for the entire population of patients who might take this drug.

**Python code example**

Here’s a clean, reproducible Python snippet that computes the confidence
interval:


```python
import numpy as np
from scipy import stats

# Sample data (replace with actual values if needed)
data = np.array([2.1, 3.0, 1.8, 2.7, 1.9, 3.4, 1.6])

n = len(data)
xbar = np.mean(data)
s = np.std(data, ddof=1)

df = n - 1
alpha = 0.05

# t critical value for 95% CI
t_crit = stats.t.ppf(1 - alpha/2, df)

# standard error
SE = s / np.sqrt(n)

# margin of error
ME = t_crit * SE

# confidence interval
lower = xbar - ME
upper = xbar + ME

print(f"Sample mean: {xbar:.2f}")
print(f"Sample SD: {s:.2f}")
print(f"t critical: {t_crit:.3f}")
print(f"95% CI: ({lower:.2f}, {upper:.2f})")
```
