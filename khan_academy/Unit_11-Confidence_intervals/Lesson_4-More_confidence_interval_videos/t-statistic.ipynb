# T-statistic confidence interval

[T-statistic confidence interval | Inferential statistics | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=hV4pdjHCKuA&t=1s)

The mean emission of all engines of a new design needs to be below 20 ppm if the design is to meet new emission requirements. Ten engines are manufactured for testing purposes, and the emission level of each is determined. The emission data is:                           

15.6 16.2 22.5 20.5 16.4 19.4 16.6 17.9 12.7 13.9                             

Does the data supply sufficient evidence to conclude that this type of engine meets the new standard? Assume we are willing to risk a Type I error with probability = 0.01                                       

**Bullet summary**

- **Goal:** Construct a 95% confidence interval for the true mean
  emission of a new engine design.

- **Given data:**

  - Sample size $n = 10$

  - Sample mean $\overset{ˉ}{x} = 17.17$

  - Sample standard deviation $s = 2.98$

- **Distribution choice:** Use the **t-distribution** because $n$is
  small and $\sigma$(population SD) is unknown.

- **Degrees of freedom:**

$$
df = n - 1 = 9
$$

- **Critical t-value for 95% CI (two-sided, df = 9):**

$$
t_{0.975,9} = 2.262
$$

- **Standard error of the mean:**

$$
SE = \frac{s}{\sqrt{n}} = \frac{2.98}{\sqrt{10}} \approx 0.942
$$

- **Margin of error:**

$$
ME = t_{0.975,9} \cdot SE \approx 2.262 \cdot 0.942 \approx 2.13
$$

- **95% confidence interval for** $\mu$**:**

$$
\overset{ˉ}{x} \pm ME = 17.17 \pm 2.13 \Rightarrow (15.04,\text{ 						}19.30)
$$

- **Interpretation:** We are 95% confident that the true mean emission
  for this engine design lies between **15.04** and **19.30** (in the
  same units as the data).

**Step-by-step explanation**

1.  **Why t-distribution?**

    - The sample size is small ($n = 10$).

    - The population standard deviation $\sigma$is unknown.

    - Under these conditions, the sampling distribution of the
      standardized mean follows a t-distribution with $n - 1$degrees
      of freedom.

2.  **Set up the t-statistic inequality** The generic t-statistic for
    the sample mean is:

$$
T = \frac{\overset{ˉ}{X} - \mu}{s/\sqrt{n}}
$$

For a 95% confidence interval with df = 9:

$$
- 2.262 \leq T \leq 2.262
$$

Substitute the expression for $T$:

$$
- 2.262 \leq \frac{\overset{ˉ}{x} - \mu}{s/\sqrt{n}} \leq 2.262
$$

3.  **Plug in the sample values**

    - $\overset{ˉ}{x} = 17.17$

    - $s = 2.98$

    - $n = 10$First compute the standard error:

$$
SE = \frac{2.98}{\sqrt{10}} \approx 0.942
$$

So:

$$
- 2.262 \leq \frac{17.17 - \mu}{0.942} \leq 2.262
$$

4.  **Multiply through by the standard error**

$$
- 2.262 \cdot 0.942 \leq 17.17 - \mu \leq 2.262 \cdot 0.942
$$

$$
- 2.13 \leq 17.17 - \mu \leq 2.13
$$

5.  **Solve for** $\mu$Subtract $17.17$from all parts, or
    equivalently rearrange to isolate $\mu$:

$$
- 2.13 \leq 17.17 - \mu \leq 2.13
$$

Multiply by $- 1$(and flip inequalities):

$$
2.13 \geq \mu - 17.17 \geq - 2.13
$$

Add $17.17$to all parts:

$$
19.30 \geq \mu \geq 15.04
$$

Or written in interval form:

$$
\mu \in (15.04,\text{ 						}19.30)
$$

6.  **Interpretation in words** If we repeated this sampling process
    many times and built a 95% t-interval each time, about 95% of those
    intervals would contain the true mean emission. For this particular
    sample, our interval estimate for the true mean is from 15.04 to
    19.30.

**Python code example**

Here’s a small, reproducible Python example that:

- Computes the sample mean and standard deviation,

- Uses the t-distribution to get the critical value,

- Constructs the 95% confidence interval.


```python
import numpy as np
from scipy import stats

# Sample data (example: 10 emission measurements)
data = np.array([16.5, 18.2, 19.0, 15.8, 17.3, 20.1, 14.9, 18.0, 16.7, 18.4])

n = len(data)
xbar = np.mean(data)
s = np.std(data, ddof=1)  # sample standard deviation
alpha = 0.05
df = n - 1

# t critical value for two-sided 95% CI
t_crit = stats.t.ppf(1 - alpha/2, df)

# standard error
SE = s / np.sqrt(n)

# margin of error
ME = t_crit * SE

# confidence interval
lower = xbar - ME
upper = xbar + ME

print(f"Sample mean: {xbar:.2f}")
print(f"Sample SD: {s:.2f}")
print(f"t critical (df={df}): {t_crit:.3f}")
print(f"95% CI: ({lower:.2f}, {upper:.2f})")
```
