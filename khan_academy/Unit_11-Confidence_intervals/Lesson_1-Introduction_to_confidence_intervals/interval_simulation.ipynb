**📌 Bullet Summary: Confidence Intervals via Simulation**

**Setup**

- A gumball machine has a true (but unknown to the sampler) proportion
  of green gumballs.

- Example: true proportion $p = 0.60$.

- A sampler draws repeated samples of size $n = 50$(or any chosen
  size).

- For each sample, they compute the sample proportion $\widehat{p}$.

**Confidence Interval Construction**

- Using a 95% confidence level → use **±2 standard errors** around
  $\widehat{p}$.

- Standard error estimated from sample:

$$
SE = \sqrt{\frac{\widehat{p}(1 - \widehat{p})}{n}}
$$

- Confidence interval:

$$
\widehat{p} \pm 2SE
$$

**Key Observations from the Simulation**

- Each sample gives a *different* $\widehat{p}$, so each confidence
  interval is different.

- Some intervals are wide, some narrow—depending on the sample.

- But across many samples:

  - **About 95% of the intervals contain the true proportion**.

  - **About 5% miss it**, which is expected.

**Effect of Sample Size**

- Increasing sample size (e.g., from 50 to 200):

  - Standard error decreases.

  - Confidence intervals become **narrower**.

  - But still, **about 95%** of intervals contain the true proportion.

**Core Insight**

Confidence intervals don’t guarantee that *your* interval contains the
truth. They guarantee that the **method** you used will succeed about
**95% of the time** in the long run.

**🧠 Explanation in Plain English**

Imagine someone doesn’t know what fraction of gumballs are green. They
take a sample of 50, count the green ones, and compute a sample
proportion. That number jumps around from sample to sample—sometimes
0.60, sometimes 0.52, sometimes 0.68.

For each sample, they build a **95% confidence interval** using:

sample proportion ± 2 × standard error

Because each sample is different, each interval is different.

Now here’s the magic: If you repeat this process hundreds or thousands
of times, **about 95% of those intervals will contain the true
proportion** (which is 0.60 in the simulation). The other 5% will miss
it—just by random chance.

This is what “95% confidence” *really* means. It’s not about one
interval—it’s about the long‑run performance of the method.

And when you increase the sample size, the intervals shrink (because the
standard error shrinks), but the success rate—95%—stays the same.

**💻 Python Code Example: Simulating Confidence Intervals**

This code simulates exactly what the Khan Academy scratchpad
demonstrates.




```python
import numpy as np

# True population proportion
p_true = 0.60

# Simulation settings
n = 50          # sample size
num_samples = 10000

def confidence_interval(p_hat, n):
    SE = np.sqrt(p_hat * (1 - p_hat) / n)
    ME = 2 * SE
    return p_hat - ME, p_hat + ME

contains_true = 0

for _ in range(num_samples):
    # Draw a sample of size n from a Bernoulli(p_true)
    sample = np.random.binomial(1, p_true, n)
    p_hat = sample.mean()
    
    lower, upper = confidence_interval(p_hat, n)
    
    # Check if the interval contains the true proportion
    if lower <= p_true <= upper:
        contains_true += 1

coverage_rate = contains_true / num_samples

print(f"Estimated coverage rate: {coverage_rate:.4f}")
```

Which is beautifully close to the theoretical 95%.
