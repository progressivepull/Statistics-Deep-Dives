**📌 Bullet Summary: Confidence Intervals for a Proportion**

**The Scenario**

- Population of 100,000 likely voters; we want to estimate the
  proportion who support **Candidate A**.

- We cannot survey everyone, so we take a **sample**.

**Sampling**

- Sample size: $n = 100$

- Number supporting A: 54 → Sample proportion: $\widehat{p} = 0.54$

**Sampling Distribution**

- If we repeatedly sampled 100 voters, the sample proportion would vary.

- For large enough samples and proportions not near 0 or 1, the sampling
  distribution of $\widehat{p}$is **approximately normal**.

- Mean of sampling distribution: the true population proportion $p$

- Standard deviation (true):

$$
\sigma_{\widehat{p}} = \sqrt{\frac{p(1 - p)}{n}}
$$

**Confidence Interval Logic**

- 95% of sample proportions fall within **±2 standard deviations** of
  the true proportion.

- This implies: There is a 95% chance the **true proportion** is within
  **±2 standard deviations** of the **sample proportion**.

**Standard Error (SE)**

- Because we don’t know $p$, we estimate using $\widehat{p}$:

$$
SE = \sqrt{\frac{\widehat{p}(1 - \widehat{p})}{n}}
$$

- With $\widehat{p} = 0.54$and $n = 100$: $SE \approx 0.05$

**95% Confidence Interval**

- Margin of error (ME):

$$
ME = 2 \times SE = 0.10
$$

- CI:

$$
0.54 \pm 0.10 \Rightarrow \lbrack 0.44,\text{ 						}0.64\rbrack
$$

- Interpretation: We are 95% confident the true proportion of voters
  supporting A is between **44% and 64%**.

**Key Insights**

- Larger sample size → smaller SE → smaller margin of error → tighter
  interval.

- The CI varies from sample to sample.

- The *method* captures the true proportion 95% of the time.

**🧠 Explanation in Plain English**

You take a sample because you can’t survey everyone. Your sample gives
you a **sample proportion**, but that number naturally varies from
sample to sample.

Statistics tells us that if we repeatedly sampled the same number of
people, the sample proportions would form a **bell-shaped curve**
centered on the true population proportion.

We don’t know the true proportion, but we *do* know the shape of the
sampling distribution well enough to estimate how far off our sample
might be.

A **95% confidence interval** is built by taking:

sample proportion ± 2 × standard error

This gives a range of plausible values for the true proportion.

**💻 Code Example (Python)**

Here’s a simple Python snippet that computes the 95% confidence interval
for a sample proportion:

```python
import math

# Sample data
n = 100
p_hat = 0.54

# Standard error
SE = math.sqrt(p_hat * (1 - p_hat) / n)

# Margin of error for ~95% CI (z = 2)
ME = 2 * SE

# Confidence interval
lower = p_hat - ME
upper = p_hat + ME

print(f"Standard Error: {SE:.4f}")
print(f"Margin of Error: {ME:.4f}")
print(f"95% Confidence Interval: [{lower:.4f}, {upper:.4f}]")
```
