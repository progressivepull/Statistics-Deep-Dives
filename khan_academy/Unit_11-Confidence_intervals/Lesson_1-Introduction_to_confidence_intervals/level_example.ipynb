# Interpreting confidence level example

[Interpreting confidence level example | Confidence intervals | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=XZAGtQt-lQo&t=3s)

A zookeeper took a random sample of 30 days and observed how much food an elephant ate on each of those days. The sample mean was 350 kg and the sample standard deviation was 25 kg. The resulting 90% confidence interval for the mean amount of food was (341, 359) kg.

Which of the following statements is a correct interpretation of the 90% confidence level?

Choose 1 answer:

A. The elephant ate between 341 kg and 359 kg on 90% of all the days.     

B. There is a 0.90 probability that the true mean amount of food is between 341 kg and 359 kg.     

C. In repeated sampling, this method produces intervals that capture the population mean in about 90% of samples.    

D. In repeated sampling, this method produces a sample mean between 341 kg and 359 kg in about 90% of samples.   

Here you go — a clean **bullet summary**, a clear **explanation**, and a
**Python code example** that mirrors the logic behind
confidence‑interval interpretation questions like this one. This is one
of the most commonly misunderstood topics in intro statistics, so
getting the intuition right is huge.

**📌 Bullet Summary: Correct Interpretation of a 90% Confidence Level**

**Scenario**

- Sample of 30 days of elephant feeding.

- Sample mean: **350 kg**

- Sample standard deviation: **25 kg**

- 90% confidence interval: **\[341, 359\] kg**

**What a 90% Confidence Level *Means***

- If we repeatedly took samples of 30 days…

- …and built a 90% confidence interval from each sample…

- …then **about 90% of those intervals would contain the true population
  mean**.

- The *method* succeeds 90% of the time — not any single interval.

**What It Does *NOT* Mean**

- NOT: “The elephant eats between 341 and 359 kg on 90% of days.”

- NOT: “There is a 90% probability the true mean is in this specific
  interval.”

- NOT: “90% of sample means fall between 341 and 359.”

**Correct Interpretation**

- **Choice C**: *In repeated sampling, this method produces intervals
  that capture the population mean in about 90% of samples.*

**🧠 Explanation in Plain English**

A confidence interval is not about the elephant’s daily eating habits.
It’s not about the probability of the true mean “moving around.” The
true mean is fixed — we just don’t know it.

Here’s the real idea:

If the zookeeper kept taking new random samples of 30 days, each sample
would give a different sample mean and a different confidence interval.
Some intervals would be too far left, some too far right, some wide,
some narrow.

But if the zookeeper used the **same method** every time to build a
**90% confidence interval**, then:

About **90%** of those intervals would contain the true mean amount of
food the elephant eats.

This is why the correct interpretation always talks about **repeated
sampling** and **the method**, not the specific interval you got.

**💻 Python Code Example: Simulating 90% Confidence Intervals**

This code simulates the process described in the explanation —
repeatedly sampling, building intervals, and checking how often they
contain the true mean.




```python
import numpy as np
from scipy.stats import t

# True population mean (unknown in real life)
mu_true = 350

# Simulation settings
n = 30
sigma = 25
num_samples = 10000

def confidence_interval(sample):
    xbar = np.mean(sample)
    s = np.std(sample, ddof=1)
    
    # t* for 90% CI with df = n - 1
    t_star = t.ppf(0.95, df=n-1)
    
    margin = t_star * s / np.sqrt(n)
    return xbar - margin, xbar + margin

contains_true = 0

for _ in range(num_samples):
    # Draw a sample from a normal distribution
    sample = np.random.normal(mu_true, sigma, n)
    lower, upper = confidence_interval(sample)
    
    if lower <= mu_true <= upper:
        contains_true += 1

coverage_rate = contains_true / num_samples
print(f"Estimated coverage rate: {coverage_rate:.4f}")
```

Which is beautifully close to the theoretical **90%**.
