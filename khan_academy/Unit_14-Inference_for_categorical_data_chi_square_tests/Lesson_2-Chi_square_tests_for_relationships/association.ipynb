# Chi-square test for association (independence)

**Bullet summary**

- The goal is to test whether **hand length** and **foot length** are
  *associated* or *independent* using a chi‑square test for
  independence.

- **Null hypothesis:** no association; hand length and foot length are
  independent.

- **Alternative hypothesis:** there is an association; the variables are
  not independent.

- A **3×3 contingency table** is constructed from a random sample of 100
  people, categorizing each person by which hand is longer and which
  foot is longer.

- Row and column totals are computed to determine marginal
  probabilities.

- Under the null hypothesis of independence, **expected counts** are
  computed using

$$
E_{ij} = \frac{\left( \text{row total})(\text{column total} \right)}{100}.
$$

- All expected counts are ≥ 5, so the chi‑square test conditions are
  satisfied.

- The chi‑square statistic is computed by summing

$$
\chi^{2} = \sum\frac{\left( O-E)^{2} \right.\ }{E}
$$

across all nine cells, giving

$$
\chi^{2} = 11.942.
$$

- Degrees of freedom for a 3×3 table:

$$
(3 - 1)(3 - 1) = 4.
$$

- Using df = 4, the p‑value is approximately **0.018**.

- With a significance level of 0.05, the p‑value is smaller, so we
  **reject the null hypothesis**.

- Conclusion: there **is** evidence of an association between hand
  length and foot length.

**Explanation**

A chi‑square test for independence evaluates whether two categorical
variables are related. Here, the variables are:

- Hand length category (right longer, left longer, same)

- Foot length category (right longer, left longer, same)

The null hypothesis assumes independence, meaning the distribution of
hand lengths should not depend on foot lengths. Under this assumption,
expected counts are computed using the product of row and column
proportions.

Once expected counts are computed, the chi‑square statistic measures how
far the observed table deviates from what independence would predict.
Larger deviations produce larger chi‑square values.

The degrees of freedom reflect how many cells can vary freely once row
and column totals are fixed. For a 3×3 table, df = 4.

A p‑value of 0.018 means that if hand and foot length were truly
independent, there is only a 1.8% chance of observing a table this
extreme. Since this is below the 5% significance threshold, the data
provide evidence of an association.

**Code example (Python)**

```python
import numpy as np
from scipy.stats import chi2

# Observed counts (3x3 table)
# Rows: Right hand longer, Left hand longer, Same hand length
# Columns: Right foot longer, Left foot longer, Same foot length
observed = np.array([
    [11, 3, 8],
    [2, 9, 14],
    [12, 13, 28]
])

# Totals
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)
total = observed.sum()

# Expected counts under independence
expected = np.outer(row_totals, col_totals) / total

# Chi-square statistic
chi_square_stat = ((observed - expected)**2 / expected).sum()

# Degrees of freedom
df = (observed.shape[0] - 1) * (observed.shape[1] - 1)

# p-value
p_value = 1 - chi2.cdf(chi_square_stat, df)

print(f"Chi-square statistic: {chi_square_stat:.3f}")
print(f"Degrees of freedom: {df}")
print(f"P-value: {p_value:.3f}")
```

This code reproduces the chi‑square statistic of about **11.942** and a
p‑value near **0.018**, matching the reasoning above.
