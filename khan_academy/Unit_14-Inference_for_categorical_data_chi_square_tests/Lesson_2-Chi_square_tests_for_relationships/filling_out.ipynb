# Filling out frequency table for independent events

[Filling out frequency table for independent events | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=TQJQeTpH_lw)

One rainy Saturday morning, Adam woke up to hear his mom complaining about the house being dirty.

“Mom is always grouchy when it rains,” Adam’s brother said to him.

So Adam decided to figure out if this statement was actually true. For the next year, he charted every time it rained and every time his mom was grouchy. What he found was very interesting – rainy days and his mom being grouchy were entirely independent events. Some of his data are shown in the table below.

Fill in the missing values from the frequency table.

|                | Raining	      |Not raining     | Row total      |
| -------------- | -------------- | -------------- | -------------- |
| Grouchy        |			      |                | 73             |
| Not grouchy    |			      |                | 292            |
| Column total   | 35	          | 330	           | 365            |





**Bullet summary**

- Adam tracked **rainy vs. not‑rainy days** and **mom grouchy vs. not
  grouchy** for a full year (365 days).

- Totals given: 35 rainy days, 330 not‑rainy days; 73 grouchy days, 292
  not‑grouchy days.

- Key fact: the events *rain* and *mom grouchy* are **independent**.

- Independence means

$$
P(\text{grouchy} \mid \text{rain}) = P(\text{grouchy}) = \frac{73}{365} = 0.2.
$$

- So **20% of rainy days** should be grouchy → $0.2 \cdot 35 = 7$.

- Fill table using totals:

  - Rain & Grouchy = 7

  - Rain & Not Grouchy = $35 - 7 = 28$

  - Not Rain & Grouchy = $73 - 7 = 66$

  - Not Rain & Not Grouchy = $330 - 28 = 264$

- All row and column totals match, confirming the completed table.

**Explanation**

The problem gives marginal totals but not the internal cell counts.
Normally, this would be impossible to fill uniquely, but the key
statement—*rain and mom’s mood are independent*—provides the missing
structure.

Independence means the probability of mom being grouchy does not depend
on whether it rains:

$$
P(\text{grouchy} \mid \text{rain}) = P(\text{grouchy}).
$$

From the totals:

$$
P(\text{grouchy}) = \frac{73}{365} = 0.2.
$$

So on rainy days, 20% should be grouchy:

$$
0.2 \cdot 35 = 7.
$$

Once that cell is known, the rest follow from row and column totals:

- Rain & Not Grouchy = $35 - 7 = 28$

- Not Rain & Grouchy = $73 - 7 = 66$

- Not Rain & Not Grouchy = $330 - 28 = 264$

This produces a full, consistent contingency table that satisfies
independence.

**Code example (Python)**


```python
import numpy as np

# Totals
rain_total = 35
not_rain_total = 330
grouchy_total = 73
not_grouchy_total = 292

# Probability mom is grouchy
p_grouchy = grouchy_total / (rain_total + not_rain_total)

# Expected counts under independence
rain_grouchy = p_grouchy * rain_total
rain_not_grouchy = rain_total - rain_grouchy
not_rain_grouchy = grouchy_total - rain_grouchy
not_rain_not_grouchy = not_rain_total - rain_not_grouchy

table = np.array([
    [rain_grouchy, rain_not_grouchy],
    [not_rain_grouchy, not_rain_not_grouchy]
])

print(table)
```

If you want to extend this, we can also compute the chi‑square test for
independence to confirm statistically that rain and mood are unrelated.
