# Contingency table chi-square test

[ Contingency table chi-square test | Probability and Statistics | Khan Academy| YouTube](https://www.youtube.com/watch?v=hpWdDmgsIRE&t=8s)

**Bullet summary**

- Three groups were studied during flu season: **Herb 1**, **Herb 2**,
  and **Placebo**, with counts of sick vs. not sick recorded in a
  contingency table.

- Totals: 380 participants, 80 sick, 300 not sick.

- **Null hypothesis:** the herbs do nothing; sickness rates should be
  the same across all groups.

- **Alternative hypothesis:** at least one group differs (herbs have
  some effect).

- Expected counts under the null are computed using overall sickness
  rate

$$
P(\text{sick}) = \frac{80}{380} \approx 0.21,P(\text{not sick}) = 0.79.
$$

- Expected values for each group are found by multiplying group size by
  these probabilities.

- The chi‑square statistic is

$$
\chi^{2} = 2.53.
$$

- Degrees of freedom for a $2 \times 3$table:

$$
(2 - 1)(3 - 1) = 2.
$$

- Critical value at $\alpha = 0.10$, df = 2 is **4.60**.

- Since $2.53 < 4.60$, the result is **not** in the rejection region.

- Conclusion: **fail to reject the null hypothesis**; the data do not
  show evidence that either herb differs from the placebo.

**Explanation**

This setup uses a **chi‑square test of independence** to determine
whether sickness outcomes depend on which group a participant was
assigned to. The null hypothesis assumes the herbs have no effect, so
all groups should share the same sickness rate.

The overall sickness proportion is

$$
\frac{80}{380} \approx 0.21,
$$

so each group’s expected number of sick participants is

$$
0.21 \times \text{group size}.
$$

For example, in Herb 1 (120 people):

- Expected sick: $0.21 \cdot 120 = 25.3$

- Expected not sick: $120 - 25.3 = 94.7$

Repeating this for all groups gives a full expected table. The
chi‑square statistic sums the squared deviations between observed and
expected values, scaled by the expected values:

$$
\chi^{2} = \sum\frac{\left( O-E)^{2} \right.\ }{E} = 2.53.
$$

With 2 degrees of freedom, the critical value at a 10% significance
level is 4.60. Because the observed statistic is smaller, the data are
consistent with random variation under the null hypothesis. There is no
statistical evidence that either herb changes flu outcomes compared to
the placebo.

**Code example**

```python
import numpy as np
from scipy.stats import chi2

# Observed contingency table
# Rows: Sick, Not Sick
# Columns: Herb1, Herb2, Placebo
observed = np.array([
    [20, 30, 30],   # sick
    [100, 110, 90]  # not sick
])

# Totals
total = observed.sum()
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)

# Expected counts under independence
expected = np.outer(row_totals, col_totals) / total

# Chi-square statistic
chi_square_stat = ((observed - expected)**2 / expected).sum()

# Degrees of freedom
df = (observed.shape[0] - 1) * (observed.shape[1] - 1)

# Critical value for alpha = 0.10
critical_value = chi2.ppf(0.90, df)

# p-value
p_value = 1 - chi2.cdf(chi_square_stat, df)

print(f"Chi-square statistic: {chi_square_stat:.2f}")
print(f"Degrees of freedom: {df}")
print(f"Critical value (alpha=0.10): {critical_value:.2f}")
print(f"P-value: {p_value:.3f}")
```
