# Introduction to the chi-square test for homogeneity

[Introduction to the chi-square test for homogeneity | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=t_jfTOE44YQ&t=8s)

**Bullet summary**

- The goal is to test whether **right‑handed and left‑handed people
  share the same distribution of subject preferences** (STEM,
  humanities, neither).

- **Null hypothesis:** both groups have the *same* distribution of
  preferences.

- **Alternative hypothesis:** the distributions are *different*.

- Total preferences across both groups: 40 STEM, 40 humanities, 20
  neither → proportions 0.40, 0.40, 0.20.

- Expected counts for right‑handed group (n = 60): 24 STEM, 24
  humanities, 12 neither.

- Expected counts for left‑handed group (n = 40): 16 STEM, 16
  humanities, 8 neither.

- Chi‑square statistic is computed by summing
  $\left( O-E)^{2}/E \right.\ $across all six cells.

- Degrees of freedom for a $3 \times 2$table: $(3 - 1)(2 - 1) = 2$.

- A p‑value is obtained from the chi‑square distribution with df = 2; if
  it is below the chosen significance level, reject the null.

- This is a **chi‑square test for homogeneity**, used to compare
  distributions across groups.

**Explanation**

A chi‑square test for homogeneity evaluates whether two or more
independent groups share the same distribution across categories. Here,
the categories are subject preferences (STEM, humanities, neither), and
the groups are right‑handed and left‑handed people.

The null hypothesis assumes the distributions are identical. Under that
assumption, the best estimate of the population distribution comes from
the combined totals:

- STEM: $40/100 = 0.40$

- Humanities: $40/100 = 0.40$

- Neither: $20/100 = 0.20$

These proportions are applied to each group to compute expected counts.
The chi‑square statistic then measures how far the observed counts
deviate from these expected values:

$$
\chi^{2} = \sum\frac{\left( O-E)^{2} \right.\ }{E}.
$$

A larger statistic indicates greater deviation from the null hypothesis.
The degrees of freedom reflect how many cells can vary independently
once row and column totals are fixed:

$$
df = (r - 1)(c - 1) = (3 - 1)(2 - 1) = 2.
$$

Using the chi‑square distribution with df = 2, the p‑value gives the
probability of observing a statistic as large as the one computed if the
null hypothesis were true. A small p‑value suggests the distributions
differ; a large p‑value suggests the observed differences could be due
to chance.

**Code example**

```python
import numpy as np
from scipy.stats import chi2

# Observed counts
# Rows: STEM, Humanities, Neither
# Columns: Right-handed, Left-handed
observed = np.array([
    [30, 10],   # STEM
    [15, 25],   # Humanities
    [15, 5]     # Neither
])

# Totals
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)
total = observed.sum()

# Expected counts under homogeneity
expected = np.outer(row_totals, col_totals) / total

# Chi-square statistic
chi_square_stat = ((observed - expected)**2 / expected).sum()

# Degrees of freedom
df = (observed.shape[0] - 1) * (observed.shape[1] - 1)

# p-value
p_value = 1 - chi2.cdf(chi_square_stat, df)

print(f"Chi-square statistic: {chi_square_stat:.2f}")
print(f"Degrees of freedom: {df}")
print(f"P-value: {p_value:.3f}")
```
