# Chi-square statistic for hypothesis testing

[Chi-square statistic for hypothesis testing | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=jABsbNBPXIk&t=12s)

**Bullet summary**

- The test makers claim each answer choice (A, B, C, D) is correct **25%
  of the time**.

- You take a **sample of 100 questions** and record how often each
  letter is correct.

- **Null hypothesis:** the distribution of correct answers is equal
  across A, B, C, D (each 25%).

- **Alternative hypothesis:** the distribution is *not* equal.

- Expected counts under the null: each letter should appear **25
  times**.

- Observed counts: A = 20, B = 20, C = 25, D = 35.

- Compute the chi‑square statistic:

$$
\chi^{2} = \sum\frac{\left( O_{i}-E_{i})^{2} \right.\ }{E_{i}} = 6.
$$

“So our chi-squared statistic… came out nice and clean… at six.”

- Degrees of freedom = number of categories − 1 = **3**.

- From chi‑square tables, with df = 3, the probability of getting a
  value ≥ 6 is **greater than 10%**.

- Since this p‑value is **greater than the 5% significance level**, you
  **fail to reject the null hypothesis**.

- Interpretation: the sample does not provide strong evidence that the
  answer distribution is unequal.

**Explanation**

This is a **chi‑square goodness‑of‑fit test**, used to compare observed
frequencies with expected frequencies under a hypothesized distribution.

**1. Hypotheses**

- **Null:** each answer choice is correct 25% of the time.

- **Alternative:** at least one answer choice has a different
  probability.

**2. Expected vs. observed**

With 100 questions and equal probabilities, the expected count for each
letter is:

$$
E_{A} = E_{B} = E_{C} = E_{D} = 25.
$$

The observed counts differ slightly: A = 20, B = 20, C = 25, D = 35.

**3. Chi‑square statistic**

The chi‑square statistic measures how far the observed counts deviate
from the expected counts:

$$
\chi^{2} = \sum\frac{\left( O_{i}-E_{i})^{2} \right.\ }{E_{i}}.
$$

Plugging in the values:

- A: $\left( 20 - 25)^{2}/25 = 25/25 = 1 \right.\ $

- B: same as A → 1

- C: $\left( 25 - 25)^{2}/25 = 0 \right.\ $

- D: $\left( 35 - 25)^{2}/25 = 100/25 = 4 \right.\ $

Total:

$$
\chi^{2} = 1 + 1 + 0 + 4 = 6.
$$

**4. Degrees of freedom**

With 4 categories:

$$
df = 4 - 1 = 3.
$$

**5. p‑value and decision**

Using chi‑square tables for df = 3:

- A chi‑square value of **6.25** corresponds to a right‑tail probability
  of **10%**.

- Since 6 is slightly *less* extreme than 6.25,

$$
p > 0.10.
$$

Because the p‑value is **greater than 0.05**, the result is **not
statistically significant**.

**6. Interpretation**

The observed differences (especially the high number of D’s) could
easily occur by random chance. There is no strong evidence that the test
makers’ claim of equal distribution is false.

**Code example (Python)**




```python
import numpy as np
from scipy.stats import chi2

# Observed counts from the sample
observed = np.array([20, 20, 25, 35])

# Expected counts under equal distribution
expected = np.array([25, 25, 25, 25])

# Chi-square statistic
chi_square_stat = ((observed - expected)**2 / expected).sum()

# Degrees of freedom
df = len(observed) - 1

# p-value (right-tail probability)
p_value = 1 - chi2.cdf(chi_square_stat, df)

print(f"Chi-square statistic: {chi_square_stat:.2f}")
print(f"Degrees of freedom: {df}")
print(f"P-value: {p_value:.3f}")

if p_value < 0.05:
    print("Reject H0: Evidence suggests the distribution is not equal.")
else:
    print("Fail to reject H0: No strong evidence against equal distribution.")
```
