# Chi-square distribution introduction

[Chi-square distribution introduction | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=dXB3cUGnaxQ&t=5s)

**Bullet summary**

- A chi‑square distribution arises when you **square independent
  standard normal variables and add them**.

- If $X \sim N(0,1)$, then $X^{2}$follows a **chi‑square
  distribution with 1 degree of freedom**.

- If you take several independent standard normals
  $X_{1},X_{2},\ldots,X_{k}$and compute

$$
Q = X_{1}^{2} + X_{2}^{2} + \cdots + X_{k}^{2},
$$

then $Q$follows a **chi‑square distribution with** $k$**degrees of
freedom**.

- More degrees of freedom shift the distribution to the right and make
  it more symmetric.

- Chi‑square values are always **non‑negative**, because they are sums
  of squares.

- Chi‑square tables give **right‑tail probabilities**: the probability
  that a chi‑square variable exceeds a given value.

- Example: For 2 degrees of freedom, the probability that
  $Q_{2} > 2.41$is **0.30**.

“If I want the probability of this random variable being greater than
2.41… it is 30%.”

**Explanation**

The chi‑square distribution is built from the standard normal
distribution. A standard normal variable $X$has mean 0 and variance 1.
If you square it, the result is always non‑negative and tends to be
small because most normal values lie near 0. This squared variable
$X^{2}$is the simplest chi‑square distribution, with **1 degree of
freedom**.

If you take multiple independent standard normals and square each one,
then add them, the result is still chi‑square but with more degrees of
freedom. Each squared term contributes a small positive amount, so the
distribution shifts right as the number of terms increases.

The shape changes with degrees of freedom:

- With **1 degree of freedom**, the distribution is sharply peaked near
  0 and heavily right‑skewed.

- With **2 degrees of freedom**, the peak lowers and the distribution
  spreads out.

- With **3 or more**, the distribution becomes smoother and more
  symmetric, though still right‑skewed.

Chi‑square tables are used to find tail probabilities. Because
chi‑square is used in goodness‑of‑fit tests and variance tests, we often
want the probability that the statistic exceeds a certain value. Tables
list values such that the area to the right equals a given probability.

For example, with 2 degrees of freedom, the table shows that the
probability of exceeding 2.41 is 0.30. That means 30% of the
distribution lies to the right of 2.41.

This distribution becomes essential in statistical inference because
many test statistics—especially those involving squared
deviations—naturally follow chi‑square distributions.

**Code example (Python)**

The following code simulates chi‑square distributions and compares them
with theoretical density curves:






```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import chi2

# Degrees of freedom to illustrate
dfs = [1, 2, 3]

x = np.linspace(0, 10, 400)

plt.figure(figsize=(10, 6))

for df in dfs:
    # Theoretical chi-square density
    plt.plot(x, chi2.pdf(x, df), label=f'df = {df}')

plt.title("Chi-square distributions for different degrees of freedom")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

And here is how to compute a right‑tail probability such as
$P(Q_{2} > 2.41)$:

```python
from scipy.stats import chi2

df = 2
value = 2.41

p_right_tail = 1 - chi2.cdf(value, df)
print(f"P(Q2 > 2.41) = {p_right_tail:.2f}")
```
