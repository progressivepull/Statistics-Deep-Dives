# Chi-square goodness-of-fit example

[Chi-square goodness-of-fit example | AP Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=VjO9hWVlOBo&t=2s)

In the game rock-paper-scissors, Kenny expects to win, tie, and lose with equal frequency. Kenny plays rock-paper-scissors often, but he suspected his own games were not following that pattern, so he took a random sample of 24 games and recorded their outcomes. Here are his results:

| Outcome | Win | Loss | Tie |
| ------- | --- | ---- | --- |
| Games   | 4   | 13   | 7   |

He wants to use these results to carry out a χ² goodness-of-fit test to determine if the distribution of his outcomes disagrees with an even distribution.

What are the values of the test statistic and P-value for Kenny’s test?

(A)        
χ² = 5.25;        
0.05 < P-value < 0.10          

(B)        
χ² = 5.25;            
0.15 < P-value < 0.2           

(C)            
χ² = 21.875;           
P-value < 0.0005             

(D)             
χ² = 21.875;             
0.0005 < P-value < 0.001                     

**Bullet summary**

- Kenny expects **win, tie, lose** to occur with **equal probability**
  in rock‑paper‑scissors.

- He takes a **random sample of 24 games**: wins = 4, losses = 13, ties
  = 7.

- **Null hypothesis:** outcomes are equally likely (each with
  probability $1/3$).

- **Alternative hypothesis:** outcomes are **not** equally likely.

- Expected counts under $H_{0}$:

$$
E_{\text{win}} = E_{\text{loss}} = E_{\text{tie}} = \frac{1}{3} \cdot 24 = 8.
$$

- Chi‑square statistic:

$$
\chi^{2} = \sum\frac{\left( O_{i}-E_{i})^{2} \right.\ }{E_{i}} = \frac{\left( 4-8)^{2} \right.\ }{8} + \frac{\left( 13-8)^{2} \right.\ }{8} + \frac{\left( 7-8)^{2} \right.\ }{8} = 5.25.
$$

- Degrees of freedom: $df = 3 - 1 = 2$.

- Using a chi‑square table, $P(\chi^{2} \geq 5.25)$with $df = 2$lies
  between **0.05 and 0.10**.

- So the **test statistic** is $\chi^{2} = 5.25$, and the **p‑value**
  is between 0.05 and 0.10 (greater than 0.05 → fail to reject
  $H_{0}$at 5%).

**Explanation**

Kenny wants to know if his game outcomes (win, lose, tie) deviate from
an **even distribution**. This is a classic **chi‑square
goodness‑of‑fit** setup.

1.  **Hypotheses**

    - $H_{0}$:
      $P(\text{win}) = P(\text{loss}) = P(\text{tie}) = 1/3$.

    - $H_{a}$: The probabilities are not all equal.

2.  **Expected counts** With 24 games and equal probabilities:

$$
E_{\text{win}} = E_{\text{loss}} = E_{\text{tie}} = \frac{1}{3} \cdot 24 = 8.
$$

3.  **Observed counts**

    - Wins: 4

    - Losses: 13

    - Ties: 7

4.  **Chi‑square statistic** For each category:

$$
\chi^{2} = \sum\frac{\left( O_{i}-E_{i})^{2} \right.\ }{E_{i}}.
$$

Plug in:

$$
\chi^{2} = \frac{\left( 4-8)^{2} \right.\ }{8} + \frac{\left( 13-8)^{2} \right.\ }{8} + \frac{\left( 7-8)^{2} \right.\ }{8} = \frac{16}{8} + \frac{25}{8} + \frac{1}{8} = 2 + 3.125 + 0.125 = 5.25.
$$

5.  **Degrees of freedom** df = \text{#categories} - 1 = 3 - 1 = 2.

6.  **p‑value** We want:

$$
p = P(\chi^{2} \geq 5.25 \mid df = 2).
$$

From chi‑square tables, 5.25 lies between the critical values for 0.10
and 0.05, so:

$$
0.05 < p < 0.10.
$$

If Kenny had chosen $\alpha = 0.05$, he would **fail to reject** the
null hypothesis, since the p‑value is larger than 0.05.

**Code example (Python)**

```python
import numpy as np
from scipy.stats import chi2

# Observed counts
observed = np.array([4, 13, 7])  # wins, losses, ties

# Expected counts under equal probabilities
n = observed.sum()
expected = np.array([n/3, n/3, n/3])

# Chi-square statistic
chi_square_stat = ((observed - expected)**2 / expected).sum()

# Degrees of freedom
df = len(observed) - 1

# p-value (right-tail)
p_value = 1 - chi2.cdf(chi_square_stat, df)

print(f"Chi-square statistic: {chi_square_stat:.2f}")
print(f"Degrees of freedom: {df}")
print(f"P-value: {p_value:.3f}")
```

This will give a chi‑square statistic of about **5.25** and a p‑value
between **0.05 and 0.10**, matching the reasoning above.
