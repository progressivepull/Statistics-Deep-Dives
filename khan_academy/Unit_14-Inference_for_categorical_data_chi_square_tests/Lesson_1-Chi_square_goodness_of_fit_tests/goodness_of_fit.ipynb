# Pearson’s chi square test (goodness of fit)

[Pearson's chi square test (goodness of fit) | Probability and Statistics | Khan Academy | YouTube](https://www.youtube.com/watch?v=2QeDRsxSF9M&t=6s)

**Bullet summary**

- The restaurant owner provides a **claimed distribution** of customer
  traffic across the week (percentages for Mon–Sat).

- You collect **observed counts** for each day and want to test whether
  the owner’s distribution is accurate.

- **Null hypothesis:** the owner’s distribution is correct.

- **Alternative hypothesis:** the owner’s distribution is not correct.

- Convert percentages into **expected counts** using the total observed
  customers (200).

- Compute the **chi‑square statistic** using

$$
\chi^{2} = \sum\frac{\left( O_{i}-E_{i})^{2} \right.\ }{E_{i}}.
$$

- The computed statistic is **11.44**.

“This right here is going to be 11.44. This is my chi-square statistic.”

- Degrees of freedom = number of categories minus 1 = **5**.

- Critical chi‑square value at α = 0.05 and df = 5 is **11.07**.

- Since **11.44 \> 11.07**, the result is more extreme than the 5%
  cutoff.

- Conclusion: **reject the owner’s distribution**; the observed data
  does not fit his claimed distribution.

**Explanation**

This is a classic **chi‑square goodness‑of‑fit test**, used to compare
observed frequencies with expected frequencies under a proposed
distribution.

**1. Hypotheses**

- **Null:** The owner’s distribution is correct.

- **Alternative:** The owner’s distribution is incorrect.

**2. Expected counts**

You observed 200 customers total. Using the owner’s percentages:

- Monday: $0.10 \cdot 200 = 20$

- Tuesday: $20$

- Wednesday: $30$

- Thursday: $40$

- Friday: $60$

- Saturday: $30$

These are the counts you *should* see if the owner’s distribution is
true.

**3. Chi‑square statistic**

For each day, compute

$$
\frac{\left( O_{i}-E_{i})^{2} \right.\ }{E_{i}}
$$

and sum across all six days. The video computes:

$$
\chi^{2} = 11.44.
$$

This measures how far the observed pattern deviates from the expected
pattern.

**4. Degrees of freedom**

With 6 categories, df = $6 - 1 = 5$. One category is determined
automatically once the others are known.

**5. Critical value**

At α = 0.05 and df = 5, the chi‑square critical value is:

$$
\chi_{0.05,5}^{2} = 11.07.
$$

This is the threshold beyond which results are considered too unlikely
under the null.

**6. Compare statistic to critical value**

Since

$$
11.44 > 11.07,
$$

the observed deviation is **larger** than what we would expect 95% of
the time if the owner’s distribution were true.

**7. Conclusion**

The observed data is unlikely under the owner’s distribution. You
**reject the null hypothesis** and conclude the distribution is not a
good fit.

**Code example (Python)**




```python
import numpy as np
from scipy.stats import chi2

# Observed counts
observed = np.array([30, 14, 34, 45, 57, 20])

# Owner's claimed percentages
percentages = np.array([0.10, 0.10, 0.15, 0.20, 0.30, 0.15])

# Total customers observed
total = observed.sum()

# Expected counts
expected = percentages * total

# Chi-square statistic
chi_square_stat = ((observed - expected)**2 / expected).sum()

# Degrees of freedom
df = len(observed) - 1

# Critical value for alpha = 0.05
critical_value = chi2.ppf(0.95, df)

print(f"Chi-square statistic: {chi_square_stat:.2f}")
print(f"Critical value (df={df}): {critical_value:.2f}")

if chi_square_stat > critical_value:
    print("Reject H0: The owner's distribution does not fit the observed data.")
else:
    print("Fail to reject H0: No evidence against the owner's distribution.")
```
