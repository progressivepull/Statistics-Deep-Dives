# Summary: Sampling Bias in Podcast Listener Polls

David hosts a podcast and wants to know how much his listeners like his show. He conducts two different informal surveys. In each case, the key question is: **what is the most concerning source of bias?**

---

## Scenario 1: Online Poll on His Website

- **Method**: David asks listeners to visit his website and voluntarily participate in an online poll.
- **Result**: About 200 respondents; 89% say they “love” the show.
- **Population**: Presumably much larger than 200 (e.g., thousands of listeners).

### Most Concerning Bias: **Voluntary Response Bias**

- The sample is **not random**; listeners choose whether to participate.
- People who already like David or are more engaged with his show are **more likely to respond**.
- Listeners who are indifferent or dislike the show are less likely to visit the website and complete the poll.
- As a result, the **89% is likely an overestimate** of how many listeners truly love the show.

### Why Not the Other Biases?
- **Response bias**: The question (“Do you love the show?”) is not sensitive or misleading.
- **Undercoverage**: While some listeners are missed, the main issue is that participation is self-selected, which points more directly to voluntary response bias.

---

## Scenario 2: Polling Listeners Who Sent Fan Emails

- **Method**: David polls the next 100 listeners who sent him fan emails.
- **Responses**: 97 respond; 94 say they “love” the show.

### Most Concerning Bias: **Convenience Sampling**

- The sample is chosen because it is **easy to access** (his inbox).
- People who send fan emails are **already highly engaged and positive** about the show.
- This sample is **not representative** of the full listener population.
- Again, the result likely **overestimates** the true proportion of listeners who love the show.

### Other Biases Present but Less Concerning
- **Nonresponse bias**: 3 out of 100 did not respond, but this is minor.
- **Voluntary response bias**: Some effect exists, but most people responded, making it less significant than convenience sampling.

---

## Key Takeaway

- **Scenario 1**: The biggest issue is **voluntary response bias**.
- **Scenario 2**: The biggest issue is **convenience sampling**.
- In both cases, the sampling method leads to an **overestimate** of how much the overall audience likes David’s podcast.


```python
import random
import numpy as np

# -----------------------------
# Step 1: Define the population
# -----------------------------

# Total listeners
population_size = 10_000

# True proportion who love the show (unknown to David)
true_love_rate = 0.60

# Create population: 1 = loves show, 0 = does not
population = np.random.choice(
    [1, 0],
    size=population_size,
    p=[true_love_rate, 1 - true_love_rate]
)

print(f"True population love rate: {population.mean():.2%}")

# ----------------------------------
# Step 2: Voluntary response sampling
# ----------------------------------

# Fans are more likely to respond
def voluntary_response(population, sample_size=200):
    responses = []
    for person in population:
        if person == 1:
            # fans respond with higher probability
            if random.random() < 0.08:
                responses.append(person)
        else:
            # non-fans rarely respond
            if random.random() < 0.01:
                responses.append(person)

        if len(responses) >= sample_size:
            break

    return np.mean(responses)

voluntary_result = voluntary_response(population)
print(f"Voluntary response estimate: {voluntary_result:.2%}")

# --------------------------------
# Step 3: Convenience sample
# --------------------------------

# Only people who sent fan emails (mostly fans)
fan_email_group = population[population == 1]

convenience_sample = np.random.choice(
    fan_email_group,
    size=100,
    replace=False
)

print(f"Convenience sample estimate: {convenience_sample.mean():.2%}")

# -----------------------------
# Step 4: Random sample (ideal)
# -----------------------------

random_sample = np.random.choice(
    population,
    size=200,
    replace=False
)

print(f"Random sample estimate: {random_sample.mean():.2%}")
```

# What this program shows

* **True population rate** is around 60%
* **Voluntary response** produces a much higher estimate (similar to the 89%)
* **Convenience sampling** produces an even more inflated result
* **Random sampling** stays close to the true value
