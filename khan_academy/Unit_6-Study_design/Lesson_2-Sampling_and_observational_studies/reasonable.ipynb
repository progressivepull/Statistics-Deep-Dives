# City Councilwoman 

City Councilwoman Kelly wants to know how the residents of her district feel about a proposed school redistricting plan.

Which of the following survey methods will allow Councilwoman Kelly to make a valid conclusion about how residents of her district feel about the proposed plan?

# Valid Survey Methods and Representative Samples

## Core Idea
To make a **valid conclusion about a population**, a survey must use a **representative sample**. This usually means:
- Avoiding samples that overrepresent a specific group (geographic, age, social group, etc.)
- Using **random selection** or surveying the **entire population**

---

## Scenario 1: City Councilwoman Kelly and School Redistricting

### Goal
City Councilwoman Kelly wants to understand how **all residents of her district** feel about a proposed school redistricting plan.

To do this, she needs a sample that fairly represents the entire district.

### Evaluating Survey Methods

- ❌ **Ask her neighbors**  
  This oversamples one geographic area and may reflect unusually strong opinions.

- ❌ **Ask residents of Whispering Pines Retirement Community**  
  This oversamples one age group (older residents), who may not reflect district-wide views.

- ✅ **Ask 200 residents of her district chosen at random**  
  This is the best option. Random selection reduces bias and increases representativeness.

- ❌ **Ask a group of parents at a local playground**  
  This oversamples a specific demographic and a single location.

### Conclusion
✔️ The **random sample of 200 residents** is the most valid method for understanding district-wide opinion.

---

## Scenario 2: Mimi and the Class President Election

### Goal
Mimi wants to determine which candidate—**Napoleon Dynamite** or **Blair Waldorf**—is leading among her **300 classmates**.

### Evaluating Survey Methods

- ❌ **Ask students at Blair’s lunch table**  
  Biased toward Blair.

- ❌ **Ask members of Napoleon’s soccer team**  
  Likely biased toward (or against) Napoleon.

- ✅ **Put all names in a hat and draw 50 at random**  
  A strong random sample that can represent the whole class.

- ❌ **Ask students whose names start with N or B**  
  Not truly random, potentially very small and biased.

- ✅ **Ask every student in the class**  
  This is a full census and gives the most accurate result.

- ✅ **Assign numbers 1–300 and randomly select 50 using a computer**  
  Equivalent to drawing names from a hat; a valid random sample.

### Conclusion
✔️ Valid methods include:
- Surveying **every student**
- Using a **randomly selected sample**

---

## Final Takeaway
Valid survey conclusions depend on **how participants are chosen**, not just how many are surveyed.  
- **Random samples** and **complete populations** produce the most reliable results  
- Convenience or targeted samples often introduce bias

Understanding this distinction is critical in statistics, polling, and real-world decision-making.



```python
# Valid Survey Methods Demonstration
# ---------------------------------
# This program demonstrates how random sampling leads to valid conclusions
# using the two scenarios:
# 1) City Council district opinion
# 2) Class president election

import random
import pandas as pd

# ============================================================
# Scenario 1: City Councilwoman Kelly's District Survey
# ============================================================

# Create population of district residents
district_size = 5000

district = pd.DataFrame({
    "ResidentID": range(1, district_size + 1),
    # Opinion on redistricting: True = support, False = oppose
    "SupportsPlan": [random.choice([True, False]) for _ in range(district_size)]
})

# Random sample of 200 residents
sample_200 = district.sample(n=200, random_state=1)

# Compare opinions
district_support_rate = district["SupportsPlan"].mean()
sample_support_rate = sample_200["SupportsPlan"].mean()

print("=== City Council Survey ===")
print("Population support rate:", round(district_support_rate, 3))
print("Sample support rate:", round(sample_support_rate, 3))


# ============================================================
# Scenario 2: Mimi's Class President Survey
# ============================================================

# Create class population
class_size = 300

classmates = pd.DataFrame({
    "StudentID": range(1, class_size + 1),
    # Vote choice: 'Napoleon' or 'Blair'
    "Vote": [random.choice(["Napoleon", "Blair"]) for _ in range(class_size)]
})

# Method 1: Census (ask everyone)
census_results = classmates["Vote"].value_counts(normalize=True)

# Method 2: Random sample of 50 students
random_sample = classmates.sample(n=50, random_state=42)
sample_results = random_sample["Vote"].value_counts(normalize=True)

print("\n=== Class President Election ===")
print("\nCensus results (entire class):")
print(census_results)

print("\nRandom sample results (50 students):")
print(sample_results)


# ============================================================
# Key Insight Output
# ============================================================

print("\nConclusion:")
print("- Random samples closely approximate population results")
print("- Biased or convenience samples would distort conclusions")
print("- Census data provides the most accurate outcome")
```

This program:

* Builds **populations**
* Draws **random samples**
* Compares sample results to true population values
* Demonstrates why **random sampling** leads to valid conclusions
