# Survey Bias: Internet Privacy Poll

## Scenario
A senator wants to understand how people in her state feel about **internet privacy issues**.  
To do this, she conducts a poll by:

- Randomly selecting **100 names from the phone book**
- Calling those numbers until **all 100 people respond**
- Finding that **42% of respondents are very concerned about internet privacy**

**Important detail:**  
Phone books do **not** include:
- Mobile-only phone users  
- Unlisted numbers  

The question is: **What is the most concerning source of bias in this poll?**

---

## Evaluating Possible Sources of Bias

### 1. Nonresponse Bias ❌
Nonresponse bias occurs when selected individuals do not respond, and their lack of response affects results.

- In this case, the senator’s office **continued calling until all 100 people responded**.
- Therefore, **nonresponse bias is not present**.

---

### 2. Undercoverage Bias ✅
Undercoverage occurs when **part of the population is excluded** from the sampling frame.

- The sample was drawn **only from the phone book**
- This excludes:
  - People with mobile phones only
  - People with unlisted landlines
- These excluded groups may be **more concerned about privacy**, since:
  - They chose not to list their number
  - They avoid traditional directories

This means the survey **misses an important segment of the population**.

✔️ **Undercoverage is the most concerning source of bias in this scenario.**

---

### 3. Voluntary Response Bias ❌
Voluntary response bias occurs when participants **choose themselves** to be part of a survey.

- Example: posting an online poll and asking people to respond
- In this case, respondents were **randomly selected and contacted**
- Participation was not voluntary in the sampling stage

So, **voluntary response bias does not apply here**.

---

## Direction of the Bias

Because people excluded from the phone book may be **more privacy-conscious**, the result:

- **42% very concerned** is likely an **underestimate**
- The true percentage of constituents concerned about internet privacy is probably **higher**

---

## Conclusion

- **Most concerning bias:** Undercoverage  
- **Why:** The phone book excludes key groups likely to care more about privacy  
- **Effect:** The reported 42% likely underestimates public concern  

This example highlights why a **sampling frame** must closely match the full population to produce reliable results.



```python
# Undercoverage Bias Demonstration
# --------------------------------
# This program simulates the senator's internet privacy poll
# and shows how undercoverage can bias results.

import random
import pandas as pd

# ------------------------------------------------
# Create the full population of constituents
# ------------------------------------------------
population_size = 5000

# Each person has:
# - phone_type: "landline_listed", "landline_unlisted", or "mobile_only"
# - concern: True = very concerned about internet privacy
population = []

for i in range(population_size):
    phone_type = random.choices(
        ["landline_listed", "landline_unlisted", "mobile_only"],
        weights=[0.4, 0.2, 0.4]  # phone book only covers 40%
    )[0]

    # People without listed numbers are more privacy-conscious
    if phone_type == "landline_listed":
        concern = random.random() < 0.35
    else:
        concern = random.random() < 0.55

    population.append({
        "PersonID": i + 1,
        "PhoneType": phone_type,
        "VeryConcerned": concern
    })

df = pd.DataFrame(population)

# ------------------------------------------------
# True population value
# ------------------------------------------------
true_concern_rate = df["VeryConcerned"].mean()

# ------------------------------------------------
# Senator's sample (phone book only)
# ------------------------------------------------
phone_book_population = df[df["PhoneType"] == "landline_listed"]

sample = phone_book_population.sample(n=100, random_state=42)
sample_concern_rate = sample["VeryConcerned"].mean()

# ------------------------------------------------
# Output results
# ------------------------------------------------
print("=== Internet Privacy Poll Simulation ===")
print("True population concern rate:", round(true_concern_rate, 3))
print("Phone-book sample concern rate:", round(sample_concern_rate, 3))

print("\nBreakdown by phone type:")
print(df.groupby("PhoneType")["VeryConcerned"].mean())

# ------------------------------------------------
# Interpretation
# ------------------------------------------------
print("\nConclusion:")
print("- Phone book sampling excludes mobile-only and unlisted numbers")
print("- Excluded groups are more privacy-conscious")
print("- Sample result underestimates true concern in the population")
```

# What This Program Demonstrates

* **Undercoverage bias** occurs when part of the population cannot be sampled
* The senator’s phone-book-only survey misses privacy-conscious individuals
* The observed **42% concern** is likely an **underestimate**
* Even with random sampling and no nonresponse, **bias still exists**
