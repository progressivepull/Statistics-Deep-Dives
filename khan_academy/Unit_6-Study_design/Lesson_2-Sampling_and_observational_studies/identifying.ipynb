# Population vs. Sample: Riverview High School Survey

## Scenario
Administrators at **Riverview High School** conducted a survey to understand how seniors feel about the lunch offerings in the school cafeteria. Instead of surveying every senior, they selected a **random sample of 100 seniors**.

## Key Statistical Concepts

### Population
The **population** is the entire group the researchers want information about.

- In this case, the population is **all seniors at Riverview High School**.

The goal of the survey is to understand how *this entire group* feels about the cafeteria food.

### Sample
The **sample** is the subset of the population that is actually surveyed.

- Here, the sample is **the 100 seniors who were surveyed**.

Surveying every senior would be impractical, so a random sample is used to estimate the opinions of the full population.

## Evaluating Possible Choices

- ❌ **Population: all high school seniors in the world**  
  This is incorrect because the survey is only about Riverview High School.

- ❌ **Population: all students at Riverview High; Sample: all seniors at Riverview High**  
  This is incorrect because not all seniors were surveyed—only 100 of them.

- ✅ **Population: all seniors at Riverview High; Sample: the 100 seniors surveyed**  
  This correctly matches the scenario.

## Conclusion
- **Population:** All seniors at Riverview High School  
- **Sample:** The 100 seniors who were surveyed  

This distinction is essential in statistics, as conclusions about the population are drawn based on data collected from the sample.



```python
# Population vs Sample Demonstration
# ---------------------------------
# This program models the Riverview High School survey example.
# It clearly distinguishes between a population and a sample.

import random
import pandas as pd

# -----------------------------
# Create the population
# -----------------------------
# Assume Riverview High has 400 seniors
population_size = 400

# Create a list representing all seniors (the population)
population = pd.DataFrame({
    "StudentID": range(1, population_size + 1),
    "LikesLunch": [random.choice([True, False]) for _ in range(population_size)]
})

# -----------------------------
# Draw a random sample
# -----------------------------
# Randomly sample 100 seniors
sample_size = 100
sample = population.sample(n=sample_size, random_state=42)

# -----------------------------
# Analyze results
# -----------------------------
population_lunch_opinion = population["LikesLunch"].mean()
sample_lunch_opinion = sample["LikesLunch"].mean()

# -----------------------------
# Output results
# -----------------------------
print("Population size:", population_size)
print("Sample size:", sample_size)
print("\nProportion who like lunch (Population):", round(population_lunch_opinion, 2))
print("Proportion who like lunch (Sample):", round(sample_lunch_opinion, 2))

# Show first few sampled students
sample.head()
```

# What This Program Demonstrates

* **Population:** All seniors at Riverview High School (simulated as 400 students)
* **Sample:** A random sample of 100 seniors
* The sample proportion is used to estimate the population proportion
* Small differences between sample and population values illustrate sampling variability

This mirrors exactly what administrators are doing in the real survey—using a sample to infer information about the entire population.
