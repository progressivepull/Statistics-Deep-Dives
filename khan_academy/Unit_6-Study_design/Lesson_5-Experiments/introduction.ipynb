# Designing an Experiment: Testing a Diabetes Medication

## Scenario
A pharmaceutical company has developed a new pill intended to help people with diabetes by **lowering hemoglobin A1c levels**.

- **Hemoglobin A1c** reflects a person’s **average blood sugar over about three months**
- Lower A1c levels indicate better blood sugar control
- The goal is to determine whether the new pill lowers A1c levels **more than would be expected by chance or other factors**

---

## Variables in the Experiment

### Explanatory Variable
- **Whether a person takes the new pill**
- This is the variable the researchers control or manipulate

### Response Variable
- **Hemoglobin A1c level**
- This measures the outcome and indicates whether the treatment is effective

---

## Experimental Design

### Participants
- **100 individuals** who need to control their diabetes

### Groups
- **Treatment group:** receives the new diabetes medication
- **Control group:** receives a placebo (a pill with no active medication)

---

## Why Use a Placebo?

If only the treatment group received pills, differences in behavior or psychology could affect results:
- Participants might change diet or habits because they believe the pill will help
- Psychological effects alone can influence health outcomes

To control for this:
- **Both groups receive pills**
- The pills look identical

---

## Blinding

### Blind Experiment
- Participants do **not know** whether they are receiving the placebo or the real medication
- This prevents psychological bias (placebo effect)

### Double-Blind Experiment
- **Neither participants nor administrators** know which pill is given
- Prevents subtle cues or differences in treatment by researchers

(Some studies even use **triple-blind designs**, where data analysts are also unaware of group assignments.)

---

## Random Assignment

Participants are **randomly assigned** to the control or treatment group to:
- Balance lurking variables (age, sex, lifestyle, diet, genetics)
- Reduce systematic bias

A simple method:
- Assign numbers to participants
- Use a random number generator to select group membership

---

## Block Design (Stratified Random Assignment)

To further reduce bias, researchers may use a **block design**:
- Divide participants into groups (blocks) based on a key variable (e.g., sex)
- Randomly assign equal numbers from each block to control and treatment groups

Example:
- Randomly assign equal numbers of men and women to each group

This helps ensure important characteristics are evenly distributed.

---

## Measuring Outcomes

1. Measure A1c levels **before** the experiment
2. Administer pills for a fixed period (e.g., 3 months)
3. Measure A1c levels **after** treatment
4. Compare changes between groups

---

## Interpreting Results

- **No difference between groups:** suggests the pill may not be effective
- **Significant improvement in treatment group:** suggests effectiveness

However:
- Results are always probabilistic
- Improvements could still occur due to random chance
- Larger sample sizes reduce this risk but do not eliminate it

---

## Replication and Scientific Validity

Even a well-designed experiment does not prove results with 100% certainty.

Key principles:
- Experiments should be **well-documented**
- Other researchers should be able to **replicate the study**
- Consistent results across replications strengthen confidence

---

## Key Takeaways

- Experiments are used to establish **causality**
- Control groups, placebos, blinding, and random assignment reduce bias
- Block designs help control known confounding variables
- Random chance can still produce false positives or negatives
- Replication is essential for scientific credibility

---


```python
# Designing and Analyzing a Diabetes Drug Experiment
# --------------------------------------------------
# This program simulates:
# - Random assignment
# - Placebo vs treatment
# - Block design (by sex)
# - A1c measurements before and after treatment
# - How causality is evaluated probabilistically

import numpy as np
import pandas as pd

np.random.seed(42)

# --------------------------------------------------
# Study parameters
# --------------------------------------------------
N = 100  # total participants
BASELINE_A1C_MEAN = 8.0
BASELINE_A1C_SD = 0.6

# True treatment effect (unknown in practice)
TRUE_TREATMENT_EFFECT = -0.8   # average reduction in A1c
PLACEBO_EFFECT = -0.2          # psychological / behavior effect
NOISE_SD = 0.3

# --------------------------------------------------
# Create participants
# --------------------------------------------------
participants = pd.DataFrame({
    "ParticipantID": range(1, N + 1),
    "Sex": np.random.choice(["Male", "Female"], size=N)
})

# Baseline A1c
participants["A1c_Before"] = np.random.normal(
    BASELINE_A1C_MEAN, BASELINE_A1C_SD, size=N
)

# --------------------------------------------------
# Block design: random assignment within sex
# --------------------------------------------------
participants["Group"] = ""

for sex in ["Male", "Female"]:
    idx = participants[participants["Sex"] == sex].index
    treatment_idx = np.random.choice(idx, size=len(idx) // 2, replace=False)
    participants.loc[treatment_idx, "Group"] = "Treatment"
    participants.loc[idx.difference(treatment_idx), "Group"] = "Control"

# --------------------------------------------------
# Apply effects
# --------------------------------------------------
def a1c_after(row):
    if row["Group"] == "Treatment":
        return (
            row["A1c_Before"]
            + PLACEBO_EFFECT
            + TRUE_TREATMENT_EFFECT
            + np.random.normal(0, NOISE_SD)
        )
    else:
        return (
            row["A1c_Before"]
            + PLACEBO_EFFECT
            + np.random.normal(0, NOISE_SD)
        )

participants["A1c_After"] = participants.apply(a1c_after, axis=1)
participants["A1c_Change"] = participants["A1c_After"] - participants["A1c_Before"]

# --------------------------------------------------
# Results
# --------------------------------------------------
summary = participants.groupby("Group").agg(
    Mean_A1c_Before=("A1c_Before", "mean"),
    Mean_A1c_After=("A1c_After", "mean"),
    Mean_Change=("A1c_Change", "mean")
)

print("=== Diabetes Drug Experiment Results ===")
print(summary.round(2))

# --------------------------------------------------
# Interpretation
# --------------------------------------------------
print("\nInterpretation:")
print("- Random assignment balances lurking variables")
print("- Placebo affects both groups")
print("- Difference in mean A1c change estimates the drug effect")
print("- Results are probabilistic, not absolute proof")
print("- Replication strengthens scientific confidence")
```

# What this program demonstrates

* **Explanatory variable**: pill type (treatment vs placebo)
* **Response variable**: A1c change
* **Random assignment & block design**: reduce confounding
* **Placebo effect**: affects both groups
* **Causal inference**: comes from comparing group outcomes

This program mirrors **a properly designed randomized, double-blind clinical experiment** and shows how causality is inferred statistically—not guaranteed with certainty, but supported by probability and replication.
