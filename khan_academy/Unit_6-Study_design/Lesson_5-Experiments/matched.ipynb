# Improving Experimental Design: Matched Pairs Experiments

## Background
In the previous experiment, a drug company tested whether a new pill could help control diabetes by lowering **hemoglobin A1c**, which reflects **average blood sugar over about three months**.

- **Explanatory variable:** Whether a participant takes the pill
- **Response variable:** Change in hemoglobin A1c

The study used:
- A **control group** (placebo)
- A **treatment group** (real pill)
- **Random assignment**
- A **block design** to balance men and women
- **Blind** or **double-blind** procedures to reduce bias

This is a strong and common experimental design, but it is **not perfect**.

---

## Remaining Limitations of the Design
Even with random assignment and blocking by sex, **lurking variables** may still affect results, such as:
- Age
- Geographic region
- Lifestyle or diet
- Behavioral responses to medication

For example:
- Younger participants may respond better to treatment
- Older participants may change behavior differently when taking a pill

Random chance can still create imbalances that influence outcomes.

---

## Matched Pairs Design

### What Is a Matched Pairs Design?
A **matched pairs design** is an experimental technique that further reduces the impact of lurking variables.

Instead of assigning people to only one group, **each participant serves as their own control**.

---

## How It Works (Diabetes Drug Example)

1. Start with 100 participants
2. Randomly divide them into two groups:
   - Group A: treatment
   - Group B: placebo
3. Measure A1c levels after the first 3 months
4. **Switch treatments**:
   - Group A now gets the placebo
   - Group B now gets the treatment
5. Measure A1c levels again after the second 3 months

Throughout the experiment:
- Participants do **not know** which pill they are receiving
- Pills look identical
- The experiment remains blind or double-blind

---

## Why Matched Pairs Are Powerful
- Each participant experiences **both conditions**
- Individual differences (age, genetics, lifestyle) are held constant
- Lurking variables are far less likely to bias results
- Differences in A1c can be more confidently attributed to the pill

---

## Practical Considerations
Matched pairs designs are not always feasible:
- Treatments may have long-lasting effects
- Carryover effects may occur
- Ethical or logistical constraints may exist

When feasible, however, they provide **very strong evidence of causality**.

---

## Key Takeaways
- No experiment is perfect
- Randomization and blocking reduce bias but don’t eliminate it
- **Matched pairs designs** further reduce the impact of lurking variables
- Good experimental design balances rigor with practicality
- Replication by other researchers is essential to confirm findings

---

## Final Summary
Matched pairs experiments strengthen causal conclusions by allowing each participant to act as their own control. When combined with blinding and randomization, they offer one of the most effective ways to reduce bias and uncover true causal relationships in experimental studies.



```python
# Matched Pairs Experimental Design Simulation
# --------------------------------------------
# This program simulates a matched-pairs experiment
# for a diabetes drug affecting hemoglobin A1c.

import numpy as np
import pandas as pd

np.random.seed(42)

# --------------------------------------------------
# Parameters
# --------------------------------------------------
N = 100  # participants
BASELINE_A1C_MEAN = 8.0
BASELINE_A1C_SD = 0.6

# True effects
PLACEBO_EFFECT = -0.2      # psychological / behavior effect
DRUG_EFFECT = -0.8         # additional effect of the drug
NOISE_SD = 0.25            # biological randomness

# --------------------------------------------------
# Create participants
# --------------------------------------------------
participants = pd.DataFrame({
    "ParticipantID": range(1, N + 1),
    "A1c_Baseline": np.random.normal(BASELINE_A1C_MEAN, BASELINE_A1C_SD, N)
})

# --------------------------------------------------
# Period 1: Random assignment
# --------------------------------------------------
participants["Period1_Group"] = np.random.choice(
    ["Treatment", "Placebo"], size=N
)

def apply_effect(baseline, group):
    if group == "Treatment":
        return (
            baseline
            + PLACEBO_EFFECT
            + DRUG_EFFECT
            + np.random.normal(0, NOISE_SD)
        )
    else:
        return (
            baseline
            + PLACEBO_EFFECT
            + np.random.normal(0, NOISE_SD)
        )

participants["A1c_Period1"] = participants.apply(
    lambda row: apply_effect(row["A1c_Baseline"], row["Period1_Group"]),
    axis=1
)

# --------------------------------------------------
# Period 2: Switch treatments (matched pairs)
# --------------------------------------------------
participants["Period2_Group"] = participants["Period1_Group"].map(
    {"Treatment": "Placebo", "Placebo": "Treatment"}
)

participants["A1c_Period2"] = participants.apply(
    lambda row: apply_effect(row["A1c_Period1"], row["Period2_Group"]),
    axis=1
)

# --------------------------------------------------
# Calculate within-person differences
# --------------------------------------------------
participants["Treatment_A1c"] = np.where(
    participants["Period1_Group"] == "Treatment",
    participants["A1c_Period1"],
    participants["A1c_Period2"]
)

participants["Placebo_A1c"] = np.where(
    participants["Period1_Group"] == "Placebo",
    participants["A1c_Period1"],
    participants["A1c_Period2"]
)

participants["Difference"] = (
    participants["Treatment_A1c"] - participants["Placebo_A1c"]
)

# --------------------------------------------------
# Results
# --------------------------------------------------
results = {
    "Mean A1c under Treatment": participants["Treatment_A1c"].mean(),
    "Mean A1c under Placebo": participants["Placebo_A1c"].mean(),
    "Mean Within-Person Difference (Treatment - Placebo)":
        participants["Difference"].mean()
}

print("=== Matched Pairs Experiment Results ===")
for k, v in results.items():
    print(f"{k}: {v:.2f}")

# --------------------------------------------------
# Interpretation
# --------------------------------------------------
print("\nInterpretation:")
print("- Each participant serves as their own control")
print("- Individual differences are canceled out")
print("- Mean within-person difference estimates drug effect")
print("- Matched pairs reduce bias from lurking variables")
```

# What this program demonstrates

* **Matched pairs design:** every participant receives both placebo and treatment
* **Within-person comparison:** removes confounding from age, lifestyle, genetics
* **Stronger causal inference** than parallel-group designs
* **Still probabilistic:** random noise remains, but bias is greatly reduced

This mirrors the diabetes A1c matched-pairs experiment and shows why it is one of the **strongest experimental designs** when feasible.
