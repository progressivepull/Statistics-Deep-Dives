# Simple Random Sampling: Estimating Average Height

## Scenario
Suppose a school (or grade level) has a **population of 80 students**.  
The goal is to estimate the **average height** of all 80 students.

Measuring everyone is impractical, so a **simple random sample of 30 students** is selected.  
The average height of these 30 students is then used as an estimate of the population average.

---

## Population and Sample
- **Population:** All 80 students
- **Sample:** 30 randomly selected students

A *simple random sample* means that **every student has an equal chance of being selected**.

---

## Methods for Selecting a Simple Random Sample

### 1. Drawing Names from a Bowl
- Write each student’s name on identical slips of paper (80 total)
- Place all slips into a bowl
- Mix thoroughly
- Blindly draw **30 names without replacement**
- The selected students form the sample

✔️ This is a valid and effective way to create a simple random sample.

---

### 2. Using a Random Number Generator (Computer or Calculator)
- Assign each student a number from **01 to 80**
- Use a random number generator to produce numbers between 1 and 80
- Select students whose numbers are generated
- Ignore repeated numbers until **30 unique students** are chosen

✔️ This method is efficient and commonly used with technology.

---

### 3. Using a Random Digit Table
- Assign numbers **01–80** to students
- Use a table of randomly generated digits
- Read digits in pairs (two at a time)
- Keep numbers that:
  - Are between 01 and 80
  - Have not already been selected
- Skip numbers outside the range (e.g., 00, 81, 91)
- Continue until **30 valid, unique numbers** are obtained

✔️ This is a traditional but still valid approach.

---

## What *Not* to Do
- ❌ Making up numbers “randomly” in your head  
  Humans are **not good at producing true randomness**, and this often introduces bias.

---

## Key Takeaways
- A **simple random sample** ensures fairness and representativeness
- Sampling must be done **without replacement**
- Using objective random methods (bowls, generators, digit tables) is best practice
- Proper random sampling allows the sample average to be a reliable estimate of the population average

---


```python
# Simple Random Sampling: Estimating Average Height
# ------------------------------------------------
# This program demonstrates how to take a simple random sample
# from a population of students and estimate the population mean.

import random
import pandas as pd

# ------------------------------------------------
# Create the population
# ------------------------------------------------
# Population of 80 students with heights (in inches)
population_size = 80

population = pd.DataFrame({
    "StudentID": range(1, population_size + 1),
    # Simulated heights with realistic variation
    "Height": [round(random.gauss(66, 3), 1) for _ in range(population_size)]
})

# ------------------------------------------------
# True population average (unknown in real life)
# ------------------------------------------------
true_population_mean = population["Height"].mean()

# ------------------------------------------------
# Take a simple random sample of 30 students
# ------------------------------------------------
sample_size = 30
sample = population.sample(n=sample_size, random_state=42)

sample_mean = sample["Height"].mean()

# ------------------------------------------------
# Output results
# ------------------------------------------------
print("=== Simple Random Sampling Example ===")
print("Population size:", population_size)
print("Sample size:", sample_size)

print("\nTrue population average height:", round(true_population_mean, 2))
print("Sample average height:", round(sample_mean, 2))

print("\nFirst 5 students in the sample:")
print(sample.head())

# ------------------------------------------------
# Interpretation
# ------------------------------------------------
print("\nConclusion:")
print("- The sample mean is used to estimate the population mean")
print("- Random sampling gives each student an equal chance")
print("- Small differences are due to sampling variability")
```

# What This Program Demonstrates

* **Population**: All 80 students
* **Sample**: 30 randomly selected students (no repeats)
* **Parameter**: True population average height
* **Statistic**: Sample average height

This mirrors exactly how a simple random sample is used in practice to estimate a population parameter.
