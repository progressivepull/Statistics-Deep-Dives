# Magician Performing at a Birthday Party

A magician performing at a birthday party stands inside a circle of 15 kids. He's going to choose a volunteer, and he wants each kid to have the same chance of getting chosen.

### Method 1: 
The magician starts with the birthday boy and moves clockwise, passing out 100 pieces of paper numbered 1 through 100. He cycles around the circle until all the pieces are distributed. He then uses a random number generator to pick an integer 1 through 100, and chooses the volunteer with that number.

### Method 2: 
The magician starts with the birthday boy and moves counter-clockwise, passing out 75 pieces of paper numbered 1 through 75. He cycles around the circle until all the pieces are distributed. He then uses a random number generator to pick an integer 1 through 75, and chooses the volunteer with that number.

### Method 3: 
The magician starts with the birthday boy and moves clockwise, passing out 30 pieces of paper numbered 1 through 30. He cycles around the circle until all the pieces are distributed. He gives #1 to the birthday boy, #2 to the next kid, and so on. He then counts the number of windows in the room and chooses the volunteer with that number.


# Fairness in Random Selection: Choosing a Volunteer

## Scenario
A magician is performing at a birthday party and stands in a circle of **15 kids**.  
He wants to choose **one volunteer**, ensuring that **each child has an equal chance** of being selected.

Three different selection methods are proposed. We evaluate whether each method is **fair**, meaning every child has the same probability of being chosen.

---

## Method 1: 100 Numbered Papers

### Description
- The magician hands out **100 pieces of paper**, numbered 1–100
- He cycles around the circle, starting with the birthday child
- A random number from 1–100 is generated
- The child holding that number is selected

### Analysis
- 100 is **not divisible by 15**
- Each child gets **6 papers**, but **10 children get an extra (7th) paper**
- Children with 7 papers have a **higher chance** of being chosen

### Conclusion
❌ **Not fair**  
Not all children receive the same number of chances.

---

## Method 2: 75 Numbered Papers

### Description
- The magician hands out **75 pieces of paper**, numbered 1–75
- Papers are distributed evenly around the circle
- A random number from 1–75 is generated

### Analysis
- 75 **is divisible by 15**
- Each child receives **exactly 5 papers**
- Each child has the **same number of chances**

### Conclusion
✅ **Fair**  
Every child has an equal probability of being selected.

---

## Method 3: 30 Numbered Papers + Number of Windows

### Description
- The magician hands out **30 pieces of paper**, numbered 1–30
- Each child receives **2 papers**
- The magician counts the **number of windows in the room**
- The child holding that number is chosen

### Analysis
- While paper distribution is even, the **number of windows is not a uniform random variable**
- Some window counts are more likely than others
- The selection mechanism is **not evenly distributed**

### Conclusion
❌ **Not fair**  
The number of windows is not a reliable random number generator.

---

## Final Summary

| Method | Fair? | Reason |
|------|------|--------|
| Method 1 | ❌ No | Unequal number of papers |
| Method 2 | ✅ Yes | Equal distribution and uniform randomness |
| Method 3 | ❌ No | Non-uniform random mechanism |

---

## Key Takeaway
For a selection to be fair:
- Each participant must have **the same number of chances**
- The random process must be **uniform**, meaning every outcome is equally likely

Only **Method 2** satisfies both conditions.



```python
# Fairness in Random Selection: Volunteer Simulation
# --------------------------------------------------
# This program simulates the three magician methods
# and shows whether each child has an equal chance
# of being selected.

import random
import pandas as pd

NUM_KIDS = 15
TRIALS = 100_000

kids = list(range(1, NUM_KIDS + 1))

# --------------------------------------------------
# Method 1: 100 papers (NOT FAIR)
# --------------------------------------------------
papers_100 = []
for i in range(100):
    papers_100.append(kids[i % NUM_KIDS])

counts_1 = {k: 0 for k in kids}

for _ in range(TRIALS):
    chosen_paper = random.randint(0, 99)
    counts_1[papers_100[chosen_paper]] += 1


# --------------------------------------------------
# Method 2: 75 papers (FAIR)
# --------------------------------------------------
papers_75 = []
for i in range(75):
    papers_75.append(kids[i % NUM_KIDS])

counts_2 = {k: 0 for k in kids}

for _ in range(TRIALS):
    chosen_paper = random.randint(0, 74)
    counts_2[papers_75[chosen_paper]] += 1


# --------------------------------------------------
# Method 3: 30 papers + windows (NOT FAIR)
# --------------------------------------------------
# Assume window counts are NOT uniformly distributed
# (more likely to be around 8–12)
papers_30 = []
for i in range(30):
    papers_30.append(kids[i % NUM_KIDS])

counts_3 = {k: 0 for k in kids}

for _ in range(TRIALS):
    # Simulated window count (skewed distribution)
    windows = int(random.gauss(10, 3))
    if 1 <= windows <= 30:
        counts_3[papers_30[windows - 1]] += 1


# --------------------------------------------------
# Results Table
# --------------------------------------------------
results = pd.DataFrame({
    "Kid": kids,
    "Method_1_Selections": [counts_1[k] for k in kids],
    "Method_2_Selections": [counts_2[k] for k in kids],
    "Method_3_Selections": [counts_3[k] for k in kids],
})

print(results)
print("\nRelative selection rates:")
print(results.set_index("Kid") / TRIALS)


# --------------------------------------------------
# Interpretation
# --------------------------------------------------
print("\nInterpretation:")
print("- Method 1 favors some kids due to extra papers")
print("- Method 2 gives nearly identical probabilities")
print("- Method 3 is biased due to non-uniform window counts")
```

# What This Program Shows

* **Method 1:** Some kids are chosen more often because 100 ÷ 15 is not equal
* **Method 2:** All kids are chosen at nearly identical rates
* **Method 3:** Selection is skewed because the number of windows is not uniformly random

This directly demonstrates **why only Method 2 is fair**.
