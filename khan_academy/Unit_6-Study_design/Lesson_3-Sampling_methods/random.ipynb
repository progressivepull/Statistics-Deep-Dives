# Sampling Methods and Bias in Surveys

## Scenario
A school wants to understand how students feel about the **quality of math instruction**.  
The **population** consists of all students at the school. Because the school is large (e.g., a college with 10,000 students), surveying everyone is impractical, so a **sample** is used.

To ensure the results represent the entire population, the sample should be **random** and designed to minimize bias.

---

## Random Sampling Methods

### 1. Simple Random Sample
- Every student is assigned an identifier (e.g., student ID)
- A random number generator selects students
- Example: randomly choosing 100 students from the entire population

**Strengths**
- Each student has an equal chance of being selected
- Easy to understand and implement

**Limitation**
- By chance, the sample may overrepresent certain groups (e.g., too many freshmen, too many males, too many engineering majors)

---

### 2. Stratified Random Sample
- The population is divided into **strata** (subgroups)
- Examples of strata:
  - Freshmen, sophomores, juniors, seniors
  - Male and female students
- A random sample is taken **from each stratum**
- Example: selecting 25 students from each class year

**Strengths**
- Ensures representation from key subgroups
- Reduces the risk of imbalance across important categories

---

### 3. Cluster Random Sample
- The population is divided into **clusters** that are each representative of the whole
- Example clusters:
  - Classrooms with balanced male/female ratios
- Entire clusters are randomly selected
- Every individual within selected clusters is surveyed

**Strengths**
- Efficient when individuals are naturally grouped
- Still random if clusters are chosen randomly

---

## Non-Random Sampling Methods (Sources of Bias)

### 1. Voluntary Response Sample
- Students choose whether to participate (e.g., online survey link)
- Likely to attract students with strong opinions or more free time

**Problem**
- High risk of bias because participants self-select

---

### 2. Convenience Sample
- Surveying those who are easiest to reach
- Examples:
  - First 100 students who arrive on campus
  - Students currently in front of the surveyor

**Problem**
- May systematically exclude parts of the population
- Results may not represent the school as a whole

---

## Other Sources of Bias

### Question Wording Bias
- Leading or emotionally charged questions can influence responses
- Example:
  - “Do you feel lucky to receive a math education few others have?” (positive bias)
  - “Do you dislike that many students fail algebra here?” (negative bias)

---

### Response Bias
- Respondents may not answer truthfully
- Causes include:
  - Fear of repercussions
  - Desire to appear favorable
  - Lack of trust in anonymity

---

## Key Takeaways
- **Random sampling** reduces the risk of bias
- Simple random samples are good but not perfect
- Stratified and cluster samples can improve representativeness
- Voluntary and convenience samples often introduce bias
- Survey design and wording are just as important as sampling method

Careful sampling and thoughtful survey construction are essential for drawing reliable conclusions about a population.


```python
# Sampling Methods and Bias Demonstration
# --------------------------------------
# This program simulates different sampling techniques
# to estimate student satisfaction with math instruction.

import random
import pandas as pd

import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

random.seed(42)

# --------------------------------------------------
# Create population
# --------------------------------------------------
POPULATION_SIZE = 10000

population = pd.DataFrame({
    "StudentID": range(1, POPULATION_SIZE + 1),
    # Demographics
    "Year": random.choices(
        ["Freshman", "Sophomore", "Junior", "Senior"],
        weights=[0.27, 0.25, 0.24, 0.24],
        k=POPULATION_SIZE
    ),
    "Gender": random.choices(
        ["Male", "Female"],
        weights=[0.48, 0.52],
        k=POPULATION_SIZE
    ),
    # Opinion: True = satisfied with math instruction
    "Satisfied": [random.random() < 0.65 for _ in range(POPULATION_SIZE)]
})

# True population satisfaction rate
true_rate = population["Satisfied"].mean()

# --------------------------------------------------
# 1. Simple Random Sample
# --------------------------------------------------
simple_sample = population.sample(n=100, random_state=1)
simple_rate = simple_sample["Satisfied"].mean()

# --------------------------------------------------
# 2. Stratified Sample (by class year)
# --------------------------------------------------
stratified_sample = (
    population
    .groupby("Year", group_keys=False)
    .apply(lambda x: x.sample(n=25, random_state=1))
)
stratified_rate = stratified_sample["Satisfied"].mean()

# --------------------------------------------------
# 3. Cluster Sample (by classroom)
# --------------------------------------------------
# Simulate classrooms (clusters)
population["Classroom"] = random.choices(
    range(1, 401),  # 400 classrooms
    k=POPULATION_SIZE
)

cluster_ids = random.sample(list(population["Classroom"].unique()), 4)
cluster_sample = population[population["Classroom"].isin(cluster_ids)]
cluster_rate = cluster_sample["Satisfied"].mean()

# --------------------------------------------------
# 4. Voluntary Response Sample (biased)
# --------------------------------------------------
voluntary_sample = population[
    population["Satisfied"] | (random.random() < 0.1)
].sample(n=100, random_state=1)
voluntary_rate = voluntary_sample["Satisfied"].mean()

# --------------------------------------------------
# Results
# --------------------------------------------------
results = pd.DataFrame({
    "Method": [
        "True Population",
        "Simple Random Sample",
        "Stratified Sample",
        "Cluster Sample",
        "Voluntary Response Sample"
    ],
    "Satisfaction Rate": [
        true_rate,
        simple_rate,
        stratified_rate,
        cluster_rate,
        voluntary_rate
    ]
})

print(results)

# --------------------------------------------------
# Interpretation
# --------------------------------------------------
print("\nInterpretation:")
print("- Simple random samples are generally accurate but can vary")
print("- Stratified samples reduce imbalance across subgroups")
print("- Cluster samples depend on how representative clusters are")
print("- Voluntary response samples tend to be biased")
```
