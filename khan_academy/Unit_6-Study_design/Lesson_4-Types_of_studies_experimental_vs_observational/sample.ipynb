# Duckies Factory
Roy's Toys received a shipment of 100,000 rubber duckies from the factory. The factory couldn't promise that all rubber duckies are in perfect form, but they promised that the percentage of defective toys won't exceed 5%.

Roy wanted to get an estimation of the percentage of defective toys, and since he couldn't go over the entire 100,000 duckies, he took a random sample of 10 duckies. He found that 10% of them were defective.


# Identifying a Sample Study: Defective Rubber Duckies

## Scenario
Roy’s Toys received a shipment of **100,000 rubber duckies** from a factory.  
The factory promised that **no more than 5%** of the toys would be defective.

Because inspecting every ducky is impractical, Roy:
- Took a **random sample of 10 duckies**
- Found that **1 out of 10** was defective
- Concluded that **10%** of the sampled duckies were defective

---

## Type of Statistical Study

### ✔ Sample Study
This is a **sample study** because:
- Roy selected a **random sample** from a large population
- He used the sample statistic (10% defective) to estimate a **population parameter**
- The parameter of interest is the **true percentage of defective duckies** in the shipment

This is not:
- An **experiment** (no treatment or manipulation)
- An **observational study** focused on relationships between variables

---

## What Can Roy Conclude?

### ❌ What He *Cannot* Conclude
Roy **cannot reasonably conclude** that the factory broke its promise based on this result alone.

Why?
- The sample size is **extremely small**
- With only 10 duckies, results are highly sensitive to chance
- Getting 1 defective ducky out of 10 can happen even if the true defect rate is below 5%

For example:
- If Roy sampled 5 duckies and found 1 defective, he might incorrectly think 20% are defective
- Small samples produce **unstable and unreliable estimates**

---

## Why Sample Size Matters
- Larger samples are **more likely** to reflect the true population parameter
- As sample size increases:
  - Random variation decreases
  - Estimates become more precise
- A sample of **10 out of 100,000** is not sufficient to challenge the factory’s claim

If Roy had sampled **100 or more duckies** and still observed a 10% defect rate, that result would be far more convincing.

---

## Key Takeaways
- This is a **sample study** used to estimate a population parameter
- Small samples can produce misleading results
- One sample result does not prove a guarantee was violated
- Larger samples provide more reliable estimates

---

## Final Summary
Roy correctly used a sample study to estimate the defect rate, but his sample size was too small to draw strong conclusions. To meaningfully assess whether the factory violated its promise, he would need a **much larger random sample**.



```python
# Sample Size and Variability Demonstration
# ----------------------------------------
# This program shows why Roy's sample of 10 duckies
# is too small to reliably estimate the defect rate.

import random
import pandas as pd

random.seed(42)

# --------------------------------------------------
# True population parameters
# --------------------------------------------------
POPULATION_SIZE = 100_000
TRUE_DEFECT_RATE = 0.05  # 5% defective as promised

# Create population (True = defective)
population = [
    random.random() < TRUE_DEFECT_RATE
    for _ in range(POPULATION_SIZE)
]

# --------------------------------------------------
# Function to take a sample and compute defect rate
# --------------------------------------------------
def sample_defect_rate(sample_size):
    sample = random.sample(population, sample_size)
    return sum(sample) / sample_size

# --------------------------------------------------
# Compare different sample sizes
# --------------------------------------------------
sample_sizes = [10, 30, 50, 100, 500]
results = []

for size in sample_sizes:
    estimates = [sample_defect_rate(size) for _ in range(1000)]
    results.append({
        "Sample Size": size,
        "Average Estimate": sum(estimates) / len(estimates),
        "Min Estimate": min(estimates),
        "Max Estimate": max(estimates)
    })

df = pd.DataFrame(results)

# --------------------------------------------------
# Output
# --------------------------------------------------
print("=== Defect Rate Estimation by Sample Size ===")
print(df)

# --------------------------------------------------
# Interpretation
# --------------------------------------------------
print("\nInterpretation:")
print("- Small samples produce highly variable estimates")
print("- A sample of 10 can easily suggest 10% or more defects by chance")
print("- Larger samples cluster closer to the true 5% defect rate")
print("- Reliable conclusions require sufficiently large samples")
```

# What This Program Shows

* A **true defect rate** of 5%
* **Repeated random samples** of different sizes
* How **sampling variability shrinks** as sample size increases
* Why Roy’s conclusion based on 10 duckies is unreliable

This directly reinforces the statistical lesson from the rubber ducky example.
