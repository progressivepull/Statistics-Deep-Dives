# Farmers' approach to Cows' Milk Yield

British researchers were interested in the relationship between farmers' approach to their cows and the cows' milk yield. They prepared a survey questionnaire regarding the farmers' perception of the cows' mental capacity, the treatment they give to the cows, and the cows' yield. The survey was filled by all the farms in Great Britain.

After analyzing their results, they found that on farms where cows were called by name, milk yield was 258 liters higher on average than on farms where this was not the case.

# Observational Study: Cow Naming and Milk Yield

## Scenario
British researchers investigated the relationship between **farmers’ interactions with their cows** and **milk yield**.

They conducted a **survey** of all farms in Great Britain, asking about:
- Whether cows were **called by name**
- Farmers’ perceptions of cows’ mental capacity
- How cows were treated
- The cows’ **milk yield**

After analyzing the survey data, researchers found that:
> Farms where cows were called by name produced **258 liters more milk on average** than farms where cows were not named.

---

## Variables Involved

- **Naming cows**  
  - Categorical variable (Yes / No)

- **Milk yield**  
  - Quantitative variable (measured in liters)

The researchers identified a relationship between these two variables.

---

## Type of Statistical Study

### ✔ Observational Study

This study is an **observational study** because:
- Researchers **did not assign treatments**
- They did **not manipulate** whether cows were named
- They simply **observed existing practices** and outcomes through a survey

### Why It Is *Not*:
- **Not a sample study**: the goal was not to estimate a single population parameter
- **Not an experiment**: farmers were not randomly assigned to name or not name cows

An experiment would have required:
- Randomly assigning farms to:
  - A *naming* group
  - A *non-naming* (control) group
- Holding all other conditions constant
- Measuring milk yield before and after treatment

---

## What Can Be Concluded?

### Valid Conclusion
- There is a **positive correlation** between **naming cows** and **milk yield** among farms in Great Britain.

This means:
- Farms that name cows tend to have higher milk production
- The two variables move together

---

## What Cannot Be Concluded?

### ❌ Causation
The study **does not prove** that naming cows *causes* higher milk yield.

Possible alternative explanations include:
- Farmers with higher-producing cows may be more likely to name them
- A **confounding variable**, such as farmer behavior or care quality

---

## Example of a Confounding Variable

**Farmer temperament (“nice farmer”)**
- More likely to:
  - Treat cows gently
  - Provide better care
  - Name cows
- Better care may independently lead to higher milk yield

If farmer care were controlled for, naming alone might not have any effect.

---

## Key Takeaways

- Observational studies can identify **correlations**
- They **cannot establish causation**
- Confounding variables may explain observed relationships
- Experiments are required to determine cause-and-effect

---

## Final Summary
This study shows a **correlation** between naming cows and higher milk yield, but it does **not** demonstrate that naming cows causes increased production. Any causal claim would require a properly designed experiment.



```python
# Observational Study vs Experiment Simulation
# -------------------------------------------
# This program demonstrates why the cow-naming study
# can show correlation but NOT causation due to confounding.

import random
import pandas as pd
import numpy as np

random.seed(42)
np.random.seed(42)

# --------------------------------------------------
# Create population of farms
# --------------------------------------------------
NUM_FARMS = 1000

farms = []

for i in range(NUM_FARMS):
    # Confounding variable: farmer care quality
    # 0 = low care, 1 = high care
    care_quality = np.random.binomial(1, 0.5)

    # Naming cows is more likely for high-care farmers
    names_cows = np.random.binomial(1, 0.8 if care_quality else 0.3)

    # Milk yield depends mostly on care quality
    base_yield = 6000
    care_bonus = 300 if care_quality else 0
    random_noise = np.random.normal(0, 100)

    milk_yield = base_yield + care_bonus + random_noise

    farms.append({
        "CareQuality": care_quality,
        "NamesCows": names_cows,
        "MilkYield": milk_yield
    })

df = pd.DataFrame(farms)

# --------------------------------------------------
# Observational result
# --------------------------------------------------
observational_means = df.groupby("NamesCows")["MilkYield"].mean()

# --------------------------------------------------
# Control for confounding variable
# --------------------------------------------------
controlled_means = (
    df.groupby(["CareQuality", "NamesCows"])["MilkYield"]
    .mean()
    .reset_index()
)

# --------------------------------------------------
# Results
# --------------------------------------------------
print("=== Observational Study Result ===")
print("Average milk yield (not naming cows):",
      round(observational_means[0], 1))
print("Average milk yield (naming cows):",
      round(observational_means[1], 1))

print("\n=== After Controlling for Farmer Care ===")
print(controlled_means)

# --------------------------------------------------
# Interpretation
# --------------------------------------------------
print("\nInterpretation:")
print("- Naming cows appears correlated with higher milk yield")
print("- After controlling for care quality, the difference shrinks")
print("- Care quality is the true driver (confounding variable)")
print("- Observational studies cannot establish causation")
```

# What This Program Demonstrates

* **Observational correlation**: Farms that name cows appear to produce more milk
* **Confounding variable**: Farmer care quality influences both naming and yield
* **Controlled analysis**: Once care quality is accounted for, naming alone explains little
* **Key lesson**: Correlation ≠ causation

This mirrors the British cow study and clearly shows **why experiments are needed to prove causal effects**.
