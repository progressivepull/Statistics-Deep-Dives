# Sample Study vs. Experiment: Testing a New Antibiotic

## Scenario
Alma has developed a new antibiotic that must kill **at least 90% of harmful bacteria** to be considered effective.

To evaluate this, she:
- Applied the antibiotic to a petri dish containing harmful bacteria
- Took a **random sample of 300 bacteria**
- Found that **94% of the sampled bacteria were dead**
- Calculated a **margin of error**, concluding that the true percentage of dead bacteria is **most likely above 90%**

Based on this, Alma feels reasonably confident that more than 90% of the bacteria in the dish died.

---

## What Type of Statistical Study Is This?

Alma:
- Took a **random sample** from a larger population (all bacteria in the dish)
- Calculated a **statistic** (94% dead)
- Used it to estimate a **population parameter** (true percentage of dead bacteria)

### ✔ Type of Study: **Sample Study**
A sample study uses data from a random sample to estimate a population parameter.

---

## Is This Study Appropriate for the Question?

### What Question Is Alma Really Trying to Answer?
At a deeper level, Alma wants to know:
> *Did the antibiotic cause the bacteria to die?*

While her sample provides evidence that **many bacteria died**, it does **not** prove that:
- The antibiotic caused the deaths

Other possible explanations include:
- The petri dish material
- Temperature or air conditions
- Lighting
- A faulty or unstable bacteria culture

Even if **95% or more** of the bacteria truly died, this alone does **not establish causation**.

---

## What’s Missing? A Controlled Experiment

To establish causality, Alma would need:
- A **treatment group** (bacteria exposed to the antibiotic)
- A **control group** (bacteria under identical conditions *without* the antibiotic)
- All other factors held constant

If bacteria die in the treatment group but not in the control group, then Alma could reasonably conclude that the antibiotic caused the deaths.

---

## Final Conclusion

- **Type of study used:** Sample study  
- **Is it appropriate for proving effectiveness?** ❌ No  

### Correct reasoning:
> The study is not appropriate because it lacks a **control group and a treatment group**, so causation cannot be established.

Alma’s results may be **suggestive**, but a **controlled experiment** is required to confidently conclude that the antibiotic is effective.

---

## Key Takeaways
- Sample studies estimate population parameters
- Experiments are required to establish causation
- High percentages alone do not prove cause-and-effect
- Control and treatment groups are essential in scientific testing



```python
# Sample Study vs Controlled Experiment Simulation
# ------------------------------------------------
# This program demonstrates why Alma's study estimates a parameter
# but does NOT establish causation without a control group.

import random
import pandas as pd

random.seed(42)

# ------------------------------------------------
# Population of bacteria
# ------------------------------------------------
POPULATION_SIZE = 10000

# True (unknown) baseline death rate without antibiotic
baseline_death_rate = 0.05

# True effect of antibiotic
antibiotic_effect = 0.90  # kills 90% when applied

# ------------------------------------------------
# Sample Study (what Alma actually did)
# ------------------------------------------------
# Antibiotic applied, but no control group
population_with_antibiotic = [
    random.random() < antibiotic_effect
    for _ in range(POPULATION_SIZE)
]

sample_size = 300
sample = random.sample(population_with_antibiotic, sample_size)

sample_dead_rate = sum(sample) / sample_size

# Margin of error (approximate 95% CI for proportion)
import math
moe = 1.96 * math.sqrt(
    (sample_dead_rate * (1 - sample_dead_rate)) / sample_size
)

# ------------------------------------------------
# Controlled Experiment (what Alma SHOULD do)
# ------------------------------------------------
# Control group: no antibiotic
control_group = [
    random.random() < baseline_death_rate
    for _ in range(POPULATION_SIZE)
]

# Treatment group: antibiotic applied
treatment_group = [
    random.random() < antibiotic_effect
    for _ in range(POPULATION_SIZE)
]

control_sample = random.sample(control_group, sample_size)
treatment_sample = random.sample(treatment_group, sample_size)

control_rate = sum(control_sample) / sample_size
treatment_rate = sum(treatment_sample) / sample_size

# ------------------------------------------------
# Results
# ------------------------------------------------
results = pd.DataFrame({
    "Study Type": [
        "Sample Study (No Control)",
        "Controlled Experiment - Control Group",
        "Controlled Experiment - Treatment Group"
    ],
    "Observed Death Rate": [
        sample_dead_rate,
        control_rate,
        treatment_rate
    ]
})

print(results)

print("\nMargin of Error (Sample Study): ±", round(moe, 3))
print("Lower bound estimate:", round(sample_dead_rate - moe, 3))

# ------------------------------------------------
# Interpretation
# ------------------------------------------------
print("\nInterpretation:")
print("- The sample study estimates the percentage of dead bacteria")
print("- It cannot prove the antibiotic CAUSED the deaths")
print("- The controlled experiment isolates the antibiotic effect")
print("- Only the experiment establishes causation")
```

# What This Program Demonstrates

* **Sample study** → estimates a population parameter
* **Margin of error** → gives confidence in the estimate
* **Controlled experiment** → compares control vs treatment
* **Causation** → only established when groups differ due to treatment

This mirrors Alma’s situation exactly and clearly shows why a control group is required to prove the antibiotic works.
