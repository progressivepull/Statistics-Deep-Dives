# Main Types of Statistical Studies

## Overview
There are **three main types of statistical studies** used to analyze data and answer questions about populations:

1. **Sample Studies**
2. **Observational Studies**
3. **Experiments**

Each type serves a different purpose and supports different kinds of conclusions.

---

## 1. Sample Study

### Purpose
A **sample study** is used to **estimate a population parameter** when it is impractical to measure the entire population.

### Example
- **Population:** All people in a city
- **Parameter of interest:** Average daily time spent on a computer
- Surveying every resident is impractical

### Method
- Randomly select a **sample** from the population
- Calculate a **statistic** (e.g., sample mean)
- Use that statistic to estimate the population parameter

### Key Point
- Sample studies estimate **what a population value is**
- They do **not** study relationships between variables

---

## 2. Observational Study

### Purpose
An **observational study** examines the **relationship between two variables** in a population.

### Example
- **Variables:**
  - Daily computer time
  - Blood pressure
- Data collected by surveying or observing individuals

### Method
- Measure both variables for each individual
- Look for patterns or trends
- Determine whether the variables are **correlated**

### Possible Result
- A **positive correlation** (as one increases, the other tends to increase)
- A **negative correlation** (as one increases, the other tends to decrease)

### Important Limitation
- **Correlation does not imply causation**
- A **confounding (lurking) variable** may influence both variables

#### Example Confounding Variable
- **Physical activity**
  - Less activity → more computer time
  - Less activity → higher blood pressure

The observed relationship may be due to the confounder, not a direct causal link.

---

## 3. Experiment

### Purpose
An **experiment** is designed to establish **causality**.

### Method
1. Select participants
2. **Randomly assign** them to groups:
   - **Control group**
   - **Treatment group**
3. Apply different conditions to each group
4. Measure outcomes and compare results

### Example
- Control group: limited computer time
- Treatment group: extended computer time
- Measure blood pressure before and after

### Why Random Assignment Matters
- Helps evenly distribute confounding variables
- Reduces bias
- Strengthens causal conclusions

### Conclusion from an Experiment
- If groups differ significantly after treatment, the difference may be **caused by the treatment**

---

## Comparison Summary

| Study Type | Main Goal | Can Show Correlation? | Can Show Causation? |
|----------|----------|---------------------|-------------------|
| Sample Study | Estimate a parameter | ❌ No | ❌ No |
| Observational Study | Study relationships | ✅ Yes | ❌ No |
| Experiment | Test cause-and-effect | ✅ Yes | ✅ Yes |

---

## Key Takeaways
- **Sample studies** estimate population values
- **Observational studies** identify correlations
- **Experiments** establish causality
- Confounding variables limit conclusions in non-experimental studies
- Random assignment is critical for causal claims

---


```python
# Main Types of Statistical Studies: Demonstration
# -----------------------------------------------
# This program illustrates:
# 1) Sample Study (estimating a population parameter)
# 2) Observational Study (correlation with confounding)
# 3) Experiment (establishing causation via random assignment)

import random
import numpy as np
import pandas as pd

random.seed(42)
np.random.seed(42)

# ============================================================
# 1) SAMPLE STUDY: Estimate a population parameter
# ============================================================
# Population: city residents' daily computer time (hours)
POP_SIZE = 10000
true_mean_hours = 3.5

population_hours = np.random.normal(loc=true_mean_hours, scale=1.2, size=POP_SIZE)

# Take a random sample
sample_size = 200
sample_hours = np.random.choice(population_hours, size=sample_size, replace=False)

sample_study_results = {
    "True Population Mean (unknown in practice)": population_hours.mean(),
    "Sample Mean Estimate": sample_hours.mean()
}

# ============================================================
# 2) OBSERVATIONAL STUDY: Correlation with a confounder
# ============================================================
# Variables:
# - ComputerTime (hours)
# - BloodPressure (mmHg)
# Confounder:
# - PhysicalActivity (0=low, 1=high)

N = 2000
physical_activity = np.random.binomial(1, 0.5, N)

# Computer time depends on activity
computer_time = np.where(
    physical_activity == 1,
    np.random.normal(2.5, 0.8, N),
    np.random.normal(4.5, 1.0, N)
)

# Blood pressure depends on activity (not computer time directly)
blood_pressure = np.where(
    physical_activity == 1,
    np.random.normal(120, 8, N),
    np.random.normal(135, 10, N)
)

observational_df = pd.DataFrame({
    "ComputerTime": computer_time,
    "BloodPressure": blood_pressure,
    "PhysicalActivity": physical_activity
})

obs_correlation = observational_df["ComputerTime"].corr(observational_df["BloodPressure"])

# ============================================================
# 3) EXPERIMENT: Random assignment to test causation
# ============================================================
# Randomly assign computer-time exposure

EXP_N = 200
assignment = np.random.binomial(1, 0.5, EXP_N)  # 0=control, 1=treatment

# Baseline BP
baseline_bp = np.random.normal(125, 10, EXP_N)

# Treatment increases BP slightly
bp_after = baseline_bp + np.where(assignment == 1, 6, 0) + np.random.normal(0, 3, EXP_N)

experiment_df = pd.DataFrame({
    "Group": np.where(assignment == 1, "Treatment", "Control"),
    "BP_Before": baseline_bp,
    "BP_After": bp_after
})

experiment_results = experiment_df.groupby("Group")["BP_After"].mean()

# ============================================================
# OUTPUT
# ============================================================
print("=== 1) SAMPLE STUDY ===")
for k, v in sample_study_results.items():
    print(f"{k}: {v:.2f}")

print("\n=== 2) OBSERVATIONAL STUDY ===")
print(f"Correlation (Computer Time vs Blood Pressure): {obs_correlation:.3f}")
print("Note: Correlation exists due to a confounding variable (physical activity).")

print("\n=== 3) EXPERIMENT ===")
print(experiment_results)

print("\nInterpretation:")
print("- Sample study estimates a population parameter.")
print("- Observational study finds correlation but not causation.")
print("- Experiment, with random assignment, supports causal conclusions.")
```

# What this program shows

* **Sample study**: How a random sample estimates a population mean.
* **Observational study**: A correlation that arises from a confounder.
* **Experiment**: Random assignment isolates the treatment effect, enabling causal inference.
