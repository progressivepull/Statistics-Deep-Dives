# Introduction to Statistics and Statistical Questions

## What Is Statistics?

**Statistics** is a broad field focused on working with data. It includes:

- **Collecting data**
- **Presenting data** (tables, charts, lists, etc.)
- **Analyzing data**

Altogether, statistics encompasses the tools and methods we use to answer questions, understand patterns, and learn about the world through data.

---

## The Idea of Variability

A central concept in statistics is **variability**.

- **Variability** refers to how much data points differ from one another.
- In everyday terms: how much something *changes* or *varies*.
- In statistics: the degree to which data values are different.

### Example: Bricks vs. Grapes

- **Bricks eaten yesterday**:  
  Data = `[0, 0, 0, 0, 0]`  
  → No variability

- **Grapes eaten yesterday**:  
  Data = `[0, 235, 17, 5, 318]`  
  → High variability

Even without calculating anything, we can see that the grape data varies much more than the brick data.

A major goal of statistics is to **measure variability** and assign numbers that describe how spread out or variable a dataset is.

---

## What Is a Statistical Question?

A **statistical question** is one that:

> **Requires collecting data with variability in order to answer it.**

Key ideas:
- There must be **more than one data point**
- The data must be **able to vary**

If a question can be answered with a single fixed value, it is **not statistical**.

---

## Examples: Statistical vs. Non-Statistical Questions

### 1. How much does my pet grapefruit weigh?
- One measurement → one data point
- No variability
- ❌ **Not statistical**

---

### 2. What is the average number of cars in a parking lot on Monday mornings?
- Requires counting cars on many Mondays
- Data will vary from day to day
- ✅ **Statistical**

---

### 3. Am I hungry?
- Yes/no answer
- No data collection with variability
- ❌ **Not statistical**

---

### 4. How many teeth does my mother have?
- One person → one count
- No variability
- ❌ **Not statistical**

*(If rephrased as “How many teeth do mothers have on average?”, it would become statistical.)*

---

### 5. How much time do the members of my family spend eating per year?
- Multiple people → multiple data points
- Likely variation between individuals
- ✅ **Statistical**

---

### 6. How many times have I watched Star Wars?
- One person → one number
- No variability
- ❌ **Not statistical**

*(If rephrased as “How many times have my coworkers watched Star Wars?”, it would be statistical.)*

---

## Key Takeaways

- **Statistics** is about working with data to answer questions.
- **Variability** is fundamental—statistics exists because data vary.
- A **statistical question** requires:
  - Multiple data points
  - Potential for variability

When a question requires collecting and analyzing variable data, it’s time to bring out your **statistical toolkit**.


```python
# Example: Exploring variability and identifying a statistical question
# This code is suitable for a Jupyter Notebook

import numpy as np

# -----------------------------
# Example 1: No variability
# -----------------------------
# Number of bricks eaten yesterday by 5 people
bricks = np.array([0, 0, 0, 0, 0])

print("Bricks data:", bricks)
print("Mean:", bricks.mean())
print("Variance:", bricks.var())
print("Standard deviation:", bricks.std())

print("\nInterpretation:")
print("All values are the same, so there is no variability.")
print("This dataset would not support a statistical question.")

# -----------------------------
# Example 2: High variability
# -----------------------------
# Number of grapes eaten yesterday by 5 people
grapes = np.array([0, 235, 17, 5, 318])

print("\nGrapes data:", grapes)
print("Mean:", grapes.mean())
print("Variance:", grapes.var())
print("Standard deviation:", grapes.std())

print("\nInterpretation:")
print("Values differ substantially from one another.")
print("This dataset has high variability and supports statistical analysis.")

# -----------------------------
# Example 3: Statistical question example
# -----------------------------
# Number of cars in a parking lot on different Monday mornings
cars = np.array([50, 49, 50, 63, 52, 48])

print("\nCars in parking lot on Monday mornings:", cars)
print("Average number of cars:", cars.mean())
print("Standard deviation:", cars.std())

print("\nInterpretation:")
print("We collected multiple data points with variability.")
print("This is a statistical question: 'What is the average number of cars on Monday mornings?'")

# -----------------------------
# Example 4: Non-statistical question
# -----------------------------
# How many times one person watched Star Wars
star_wars_views = np.array([7])

print("\nStar Wars views:", star_wars_views)
print("Mean:", star_wars_views.mean())
print("Variance:", star_wars_views.var())

print("\nInterpretation:")
print("Only one data point exists, so variability cannot be measured.")
print("This is not a statistical question.")
```
