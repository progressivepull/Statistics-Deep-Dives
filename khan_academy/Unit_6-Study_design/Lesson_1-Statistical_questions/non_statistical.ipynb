# Statistical Questions: What Makes a Question Statistical?
A statistical question is one that anticipates variability in the data and typically requires data collection from multiple sources or individuals to answer.

### ✅ Examples of Statistical Questions

* How old are the people who have watched this video in 2013
* Do dogs run faster than cats
* Do wolves weigh more than dogs
* Does it rain more in Seattle than Singapore
* What was the difference in rainfall between Singapore and Seattle in 2013
* In general, will I use less gas driving at 55 mph than 70 mph
* Do English professors get paid less than math professors
* These questions require data from multiple observations and involve variability.

### ❌ Examples of Non-Statistical Questions

* How old are you
* Does your dog weigh more than that wolf
* Does the most highly paid English professor at Harvard get paid more than the most highly paid math professor at MIT


These are specific to individuals or single comparisons and do not involve data variability.


# Statistical Questions Summary Notebook

## Overview

This notebook summarizes a discussion on **statistical questions**—questions that require data with variability to answer—versus **non-statistical questions**, which can be answered using a single known value.

---

## 🎯 Objective

The goals of this notebook are to:

- Understand what types of questions require statistics  
- Identify questions that do *not* require statistical tools  
- Learn how variability and data needs define a statistical question  

---

## 🧪 What Is a Statistical Question?

### A **statistical question**:

- Involves **variability** in the data  
- Refers to a **group or population**  
- Requires **summary measures** such as the mean, median, or mode  

### A **non-statistical question**:

- Refers to a **specific individual or single instance**  
- Has one definite, fixed answer  
- Does **not** require statistical analysis  

---

## ✅ Examples of Statistical Questions

These questions involve groups and variability, so statistical tools are needed:

- **How old are the people who have watched this video in 2013?**  
  Multiple viewers with different ages → use averages or medians.

- **Do dogs run faster than cats?**  
  Many dogs and cats with varying speeds → compare central tendencies.

- **Do wolves weigh more than dogs?**  
  Weights vary within each group → compare means or medians.

- **Does it rain more in Seattle than in Singapore?**  
  Rainfall varies by year → compare average rainfall.

- **In general, will I use less gas driving at 55 mph than at 70 mph?**  
  Fuel efficiency varies → compare average gas usage.

- **Do English professors get paid less than math professors?**  
  Salaries vary → compare average salaries across groups.

---

## ❌ Examples of Non-Statistical Questions

These questions can be answered with a single known value:

- **How old are you?**  
  One person → one exact value.

- **Does your dog weigh more than that wolf?**  
  Specific dog and wolf → direct comparison.

- **What was the difference in rainfall between Singapore and Seattle in 2013?**  
  Known historical values → simple subtraction.

- **Does the most highly paid English professor at Harvard earn more than the most highly paid math professor at MIT in 2013?**  
  Two individuals → direct comparison.

---

## 🧠 Key Patterns to Remember

- Questions about **groups with variation** → **statistical**
- Questions about **individuals or single events** → **non-statistical**

---

## 📌 Tip for Learners

When deciding whether a question is statistical, ask yourself:

- Is there **variability** in the data?
- Am I asking about a **group or population**?
- Do I need to compute an **average, spread, or distribution**?

If the answer is *yes* to these, you’re likely dealing with a **statistical question**.


# Python example: labeling statistical vs non-statistical questions
Below is a self‑contained example you could drop into a Jupyter notebook.
It:

* Stores the questions from your text
* Tags them as statistical or not (based on the reasoning you described)
* Prints them nicely
* Shows one simple heuristic function for deciding if a question is likely “statistical”


```python
# List of questions from the narration
questions = [
    "How old are you?",
    "How old are the people who have watched this video in 2013?",
    "Do dogs run faster than cats?",
    "Does dog A run faster than cat B?",
    "Do wolves weigh more than dogs?",
    "Does your dog weigh more than that wolf?",
    "Does it rain more in Seattle than Singapore?",
    "What was the difference in rainfall between Singapore and Seattle in 2013?",
    "In general, will I use less gas driving at 55 miles an hour than 70 miles per hour?",
    "Do English professors get paid less than math professors?",
    "Does the most highly paid English professor at Harvard get paid more than the most highly paid math professor at MIT in 2013?"
]

# Ground-truth labels based on the explanation in your text
# True  -> statistical question
# False -> not a statistical question
labels = {
    "How old are you?": False,
    "How old are the people who have watched this video in 2013?": True,
    "Do dogs run faster than cats?": True,
    "Does dog A run faster than cat B?": False,
    "Do wolves weigh more than dogs?": True,
    "Does your dog weigh more than that wolf?": False,
    "Does it rain more in Seattle than Singapore?": True,
    "What was the difference in rainfall between Singapore and Seattle in 2013?": False,
    "In general, will I use less gas driving at 55 miles an hour than 70 miles per hour?": True,
    "Do English professors get paid less than math professors?": True,
    "Does the most highly paid English professor at Harvard get paid more than the most highly paid math professor at MIT in 2013?": False
}

def is_statistical(question: str) -> bool:
    """
    Very rough heuristic:
    - Statistical questions usually involve groups/populations,
      variability, and phrases like 'in general', 'on average',
      'people who', plural nouns, etc.
    - Non-statistical questions here tend to be about specific
      individuals or single events.
    This is NOT a real classifier, just a demo aligned with the video.
    """
    q = question.lower()

    # Clues that we might be talking about a group or repeated measurements
    group_clues = [
        "people", "professors", "dogs", "cats", "wolves",
        "seattle", "singapore", "in general", "on average",
        "less gas", "more gas", "get paid", "rain more"
    ]

    # Clues that we might be talking about specific individuals or
    # specific single measurements (making it non-statistical)
    specific_clues = [
        "how old are you",                 # single person
        "dog a", "cat b",                  # specific animals
        "your dog", "that wolf",           # specific individuals
        "difference in rainfall",          # specific year measured exactly
        "most highly paid",                # specific top individuals
        "in 2013"                          # tying it to single measurement
    ]

    # If question matches a specific clue, call it non-statistical
    for clue in specific_clues:
        if clue in q:
            return False

    # If question matches any group clue, call it statistical
    for clue in group_clues:
        if clue in q:
            return True

    # Fallback: default to non-statistical
    return False

# Display results
print("Question classification:\n")
for q in questions:
    true_label = labels[q]
    predicted = is_statistical(q)

    label_str = "Statistical" if true_label else "Not statistical"
    pred_str = "Statistical" if predicted else "Not statistical"

    print(f"Q: {q}")
    print(f"  Ground truth: {label_str}")
    print(f"  Heuristic   : {pred_str}")
    print()
```
